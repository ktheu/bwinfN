## 1. Eine Größe, die sich nie ändert: Invarianten

Eine Reihe von Münzen liegt auf dem Tisch, jede mit **Kopf** (`1`) oder **Zahl** (`0`) oben. In einem **Zug** drehst du `laenge` **benachbarte** Münzen um (zunächst `laenge = 2`).
Gefragt ist: Kann man mit solchen Zügen jedes Muster auf **lauter Zahl** bringen? Das Muster `(1, 1, 0, 1, 1)` bedeutet: Kopf, Kopf, Zahl, Kopf, Kopf.

Das ist die Art Frage, die in der 2. Runde oft versteckt ist: „Geht es überhaupt, und wenn ja, wie?“ Ein Programm, das alle Möglichkeiten durchsucht, beantwortet sie, aber bei `n` Münzen gibt es `2^n` Muster.
Wir beginnen mit dem **Schiedsrichter**: der Suche über alle Muster, die man durch Züge erreichen kann (die Technik aus [Notebook 05](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/05_breitensuche.ipynb) und [06](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/06_zustaende.ipynb)).

1. Die lösbaren Muster haben eine **gerade** Zahl von Köpfen (`(1, 1, 0, 1, 1)` hat vier), die nicht lösbaren eine **ungerade** (`(1, 0, 0, 0, 0)` hat einen).

   Damit gilt: **lösbar genau dann, wenn die Zahl der Köpfe gerade ist.** Das haben wir jetzt bewiesen, nicht nur für `n ≤ 7`.

</details>

## 2. XOR: Gleiches hebt sich auf

Das **exklusive Oder** `^` verknüpft Zahlen Bit für Bit: Das Ergebnisbit ist `1`, wenn genau eines der beiden Bits `1` ist. Wichtig sind drei Eigenschaften, die du aus [Notebook 18](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/18_mathematik.ipynb) kennst:

- `a ^ a = 0` (gleiche Zahlen heben sich auf),
- `a ^ 0 = a` (die 0 ändert nichts),
- die **Reihenfolge ist egal**: `a ^ b = b ^ a`, `(a ^ b) ^ c = a ^ (b ^ c)`.

Daraus folgt: Kommt in einer langen XOR-Kette jede Zahl **gerade oft** vor, ist das Ergebnis `0`, und Zahlen, die zweimal vorkommen, **verschwinden** einfach, egal wo sie in der Kette stehen.

#### Aufgabe 2.1: Was wird ausgegeben?

Überlege zuerst, was der Code ausgibt, und schreibe es auf. Führe die Zelle erst danach aus.

## 3. Mengen: Möglichkeiten ausschließen

Oft kennst du nicht die Lösung, aber du kannst **Möglichkeiten streichen**. Das ist die Technik von [Notebook 04](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/04_logisch_schliessen.ipynb) (zwingende Folgerungen) mit Mengen in Python:
`a & b` ist der **Schnitt** (nur Elemente, die in beiden liegen), `a - b` die **Differenz** (Elemente von `a`, die nicht in `b` liegen), `a | b` die **Vereinigung**.

Ein Rätsel: Die gesuchte **Hausnummer** liegt zwischen 1 und 60. Du bekommst Hinweise der Form „sie ist durch `t` teilbar“ (`ja = True`) oder „sie ist nicht durch `t` teilbar“ (`ja = False`).

## 4. Eine erzwungene Reihenfolge

Matrjoschka-Puppen: Jede Puppe hat eine Höhe und eine Breite. Eine Puppe passt in eine andere, wenn sie in **beiden** Maßen **kleiner** ist (strikt). Gegeben sind `n` Puppen. Können alle **ineinander** gesteckt werden,
also in eine Reihenfolge `p1, p2, ..., pn`, in der `p1` in `p2` passt, `p2` in `p3` und so weiter?

Das Brute Force probiert alle `n!` Reihenfolgen durch (siehe [Notebook 08](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/08_reihenfolgen.ipynb)). Aber schon ein bisschen Nachdenken zeigt, dass die Reihenfolge **erzwungen** ist.

````python
from itertools import permutations


def passt(klein, gross):
    return klein[0] < gross[0] and klein[1] < gross[1]


def kette_probieren(puppen):
    """Brute Force: Gibt (Reihenfolge oder None, Zahl der geprüften Reihenfolgen) zurück."""
    versuche = 0
    for reihe in permutations(puppen):
        versuche += 1
        if all(passt(reihe[i], reihe[i + 1]) for i in range(len(reihe) - 1)):
            return list(reihe), versuche
    return None, versuche

print(kette_probieren([(4, 5), (1, 2), (3, 4), (2, 3)]))
print(kette_probieren([(4, 5), (1, 2), (3, 3), (2, 4)]))
````

## Wettbewerbs-Ecke 1: Alles Käse

*Angelehnt an Aufgabe 2 „Alles Käse“ der 2. Runde des 41. Bundeswettbewerbs Informatik (2022/23).*

Antje hat einen Käsequader Scheibe für Scheibe abgeschnitten, jede Scheibe **1 mm dick**. Zwischen den Schnitten hat sie den Käse gedreht, deshalb hat jede Scheibe zwei Seitenlängen `(a, b)`
(die kurze zuerst, ob gedreht oder nicht, spielt keine Rolle). Kann man aus den Scheiben wieder einen **vollständigen Quader** zusammensetzen, und in welcher **Reihenfolge**?

In `daten/` liegen drei Beispiele (Format: erste Zeile die Zahl der Scheiben, dann je Scheibe `a b`): `kaese_mini1.txt` (12 Scheiben), `kaese_mini2.txt` (5 Scheiben) und `kaese_mini3.txt` (23 Scheiben).
Das sind die ersten drei Beispiele des Wettbewerbs, die größeren sind mit dem Verfahren aus dem [Lösungsnotebook](https://ktheu.github.io/bwinfN/notebooks/index.html?path=bwinf2/41_AllesKaese/kaese_loesung.ipynb) auch schnell fertig.

Das Brute Force probiert alle Reihenfolgen. Bei den 12 Scheiben von `kaese_mini1` sind das `12! = 479 001 600`, bei den 23 Scheiben von `kaese_mini3` etwa `2,6 · 10^22`. **Das geht nicht.**
Du suchst die **Eigenschaft**, die die Reihenfolge erzwingt.

````python
from collections import Counter


def lies_scheiben(dateiname):
    """Liest die Scheiben als Liste von Paaren (kurze Seite, lange Seite)."""
    with open('daten/' + dateiname, encoding='utf-8') as datei:
        zeilen = datei.read().split('\n')
    scheiben = []
    for zeile in zeilen[1:]:                     # Zeile 0 enthält nur die Anzahl
        if zeile.strip() != '':
            a, b = zeile.split()
            a, b = int(a), int(b)
            scheiben.append((min(a, b), max(a, b)))
    return scheiben

def spiele(reihenfolge):
    """Legt die Scheiben in dieser Reihenfolge an (Schiedsrichter). Gibt die sortierten Maße des Käses zurück oder None.
    Eine Scheibe (p, q) passt, wenn zwei Seiten des Käses genau p und q sind. Die dritte Seite wird dann um 1 länger."""
    a, b = reihenfolge[0]
    masse = [0, a, b]                            # vor der ersten Scheibe hat der Käse keine Dicke
    for scheibe in reihenfolge:
        for i in range(3):
            uebrige = sorted(masse[j] for j in range(3) if j != i)
            if uebrige == list(scheibe):
                masse[i] += 1
                break
        else:
            return None                          # keine Seite passt
    return sorted(masse)

from itertools import permutations


def kaese_probieren(scheiben):
    """Brute Force: Gibt (Maße oder None, Zahl der geprüften Reihenfolgen) zurück."""
    versuche = 0
    for reihe in permutations(scheiben):
        versuche += 1
        masse = spiele(reihe)
        if masse is not None:
            return masse, versuche
    return None, versuche

scheiben2 = lies_scheiben('kaese_mini2.txt')
print(scheiben2)
print(spiele(scheiben2))
print(spiele([(1000, 1000), (2, 1000), (2, 1000), (2, 1000), (2, 1000)]))
print(kaese_probieren(scheiben2))
````

Die Zelle oben ist der **Schiedsrichter** (`spiele`) und das Brute Force (`kaese_probieren`). Das Beispiel `kaese_mini2` hat nur 5 Scheiben, da geht es noch (höchstens `5! = 120` Reihenfolgen).
Der Schiedsrichter legt die Scheiben der Reihe nach an: Eine Scheibe `(p, q)` passt, wenn **zwei** Seiten des Käses genau `p` und `q` sind. Dann wächst die dritte Seite um 1.

Eine Scheibe `(x, y)` passt dann **nie wieder** an den Käse, und die Reihenfolge ist gescheitert (außer die Scheibe lag nie im Vorrat). Also darf man nichts anderes zuerst legen.

**`(x, y)` fehlt, `(x, z)` ist da:** Dann muss es `(x, z)` sein, aus demselben Grund: Legte man `(y, z)`, würde `x` wachsen, und `(x, z)` passte nie mehr.

**Nur `(y, z)` ist da:** Dann ist diese Scheibe die einzige Möglichkeit.

Die Reihenfolge ist also **erzwungen**. Zu jeder **Startscheibe** gibt es höchstens **einen** Aufbau, und `n!` Reihenfolgen schrumpfen auf `n` Schritte pro Startscheibe.

</details>

### Finde den Fehler: zwei Zeilen vertauscht

Jemand prüft die drei Möglichkeiten in einer anderen Reihenfolge: **erst** `(x, z)`, dann `(x, y)`, dann `(y, z)`. Das sieht harmlos aus. Wir zählen, bei wie vielen der verschiedenen Startscheiben der Aufbau gelingt:

````python
def gelungene_starts(scheiben, falsch):
    """Zahl der Startgrößen, von denen aus alle Scheiben angelegt werden. falsch=True: die Zeilen sind vertauscht."""
    vorrat = Counter(scheiben)
    gelungen = 0
    for start in list(vorrat):
        vorrat[start] -= 1
        masse = [1, start[0], start[1]]
        angelegt = 1
        while True:
            x, y, z = masse
            moeglich = [(x, y, 2), (x, z, 1), (y, z, 0)]
            if falsch:
                moeglich = [(x, z, 1), (x, y, 2), (y, z, 0)]        # vertauscht
            wahl = None
            for p, q, i in moeglich:
                if vorrat[(p, q)] > 0:
                    wahl = (p, q, i)
                    break
            if wahl is None:
                break
            vorrat[(wahl[0], wahl[1])] -= 1
            masse[wahl[2]] += 1
            masse.sort()
            angelegt += 1
        vorrat = Counter(scheiben)                                  # Vorrat neu füllen
        if angelegt == len(scheiben):
            gelungen += 1
    return gelungen


for name in ['kaese_mini1.txt', 'kaese_mini2.txt', 'kaese_mini3.txt']:
    scheiben = lies_scheiben(name)
    print(name, 'richtig:', gelungene_starts(scheiben, False), 'vertauscht:', gelungene_starts(scheiben, True), 'von', len(set(scheiben)), 'Größen')
````

Donald hat **Spieße beobachtet**: Zu jedem Spieß kennt er die **Menge der Sorten** auf dem Spieß und die **Menge der Schüsseln**, an denen sich der Gast bedient hat (nicht, welche Sorte aus welcher Schüssel kam).

`spiesse_mini0.txt` (die Beobachtungen aus dem Text der Aufgabe, `n = 6`, Wunsch: Weintraube, Brombeere, Apfel), `spiesse_mini1.txt` (`n = 10`) und `spiesse_mini3.txt` (`n = 15`).

Das Brute Force probiert alle Zuordnungen von Sorten zu Schüsseln: `n!`, bei `n = 6` sind es 720, bei `n = 15` etwa `1,3 · 10^12`. Das geht nur bei ganz kleinen `n`.

````python
def lies_spiesse(dateiname):
    """Gibt (n, Wunschsorten als Liste, Spieße als Liste von (Sortenmenge, Schüsselmenge)) zurück."""
    with open('daten/' + dateiname, encoding='utf-8') as datei:
        zeilen = [zeile.split() for zeile in datei.read().splitlines() if zeile.strip()]
    n = int(zeilen[0][0])
    wunsch = zeilen[1]
    spiesse = []
    for i in range(int(zeilen[2][0])):
        schuesseln = set(int(x) for x in zeilen[3 + 2 * i])
        sorten = set(zeilen[4 + 2 * i])
        spiesse.append((sorten, schuesseln))
    return n, wunsch, spiesse


def zeige(menge):
    """Menge sortiert als Text, z. B. {1, 4}."""
    return '{' + ', '.join(str(x) for x in sorted(menge)) + '}'

n, wunsch, spiesse = lies_spiesse('spiesse_mini0.txt')
print(n, wunsch)
for sorten, schuesseln in spiesse:
    print(sorted(sorten), zeige(schuesseln))
````

**Der Schiedsrichter.** Für ganz kleine `n` können wir **alle** Zuordnungen von Sorten zu Schüsseln durchgehen und die behalten, die zu **allen** Spießen passen. Die Wunschsorten sind „eindeutig“, wenn alle passenden Zuordnungen
ihnen dieselbe Schüsselmenge geben. Sorten, deren Namen in keiner Beobachtung vorkommen, bekommen Ersatznamen `?1`, `?2`, ...

````python
from itertools import permutations


def alle_sorten(n, wunsch, spiesse):
    """Alle bekannten Sortennamen (alphabetisch), dann Ersatznamen für Sorten, die nirgends vorkommen."""
    namen = set(wunsch)
    for sorten, schuesseln in spiesse:
        namen = namen | sorten
    namen = sorted(namen)
    return namen + ['?%d' % i for i in range(1, n - len(namen) + 1)]


def durchprobieren(n, wunsch, spiesse):
    """Gibt (Zahl der passenden Zuordnungen, Menge der möglichen Schüsselmengen der Wunschsorten) zurück."""
    sorten = alle_sorten(n, wunsch, spiesse)
    passend = 0
    mengen = set()
    for reihe in permutations(range(1, n + 1)):
        schuessel = dict(zip(sorten, reihe))              # Sorte -> Schüssel
        if all(set(schuessel[s] for s in auf_spiess) == benutzt for auf_spiess, benutzt in spiesse):
            passend += 1
            mengen.add(frozenset(schuessel[w] for w in wunsch))
    return passend, mengen

import random


def zufallsfall(startwert):
    """Eine geheime Zuordnung, daraus passende Spieße und zufällige Wunschsorten (n von 2 bis 6)."""
    zufall = random.Random(startwert)
    n = zufall.randint(2, 6)
    namen = 'ABCDEFG'[:n]
    nummern = list(range(1, n + 1))
    for i in range(n - 1, 0, -1):                          # Nummern vertauschen (geheime Zuordnung)
        j = zufall.randint(0, i)
        nummern[i], nummern[j] = nummern[j], nummern[i]
    geheim = dict(zip(namen, nummern))
    spiesse = []
    for _ in range(zufall.randint(0, 4)):
        sorten = set(s for s in namen if zufall.random() < 0.5)
        if len(sorten) == 0:
            sorten = {zufall.choice(namen)}
        spiesse.append((sorten, set(geheim[s] for s in sorten)))
    wunsch = [s for s in namen if zufall.random() < 0.5]
    if len(wunsch) == 0:
        wunsch = [zufall.choice(namen)]
    return n, wunsch, spiesse


eindeutige = 0
for startwert in range(300):
    n, wunsch, spiesse = zufallsfall(startwert)
    eindeutig, alle = ist_eindeutig(n, wunsch, spiesse)
    passend, mengen = durchprobieren(n, wunsch, spiesse)
    assert eindeutig == (len(mengen) == 1), (startwert, n, wunsch, spiesse)
    if eindeutig:
        assert mengen == {frozenset(alle)}
        eindeutige += 1
print('300 Fälle verglichen, davon', eindeutige, 'eindeutig; Mengen-Verfahren und Durchprobieren stimmen immer überein')
n, wunsch, spiesse = lies_spiesse('spiesse_mini0.txt')
print('spiesse_mini0:', durchprobieren(n, wunsch, spiesse)[0], 'von', 720, 'Zuordnungen passen zu den Spießen')
````

Liegt die Sorte `f` in der Schüssel `b`, haben beide **denselben Fingerabdruck**: Die Sorte ist genau dann auf dem Spieß, wenn ihre Schüssel benutzt wurde. Die beiden Regeln lassen darum genau die Schüsseln mit demselben Fingerabdruck übrig.
Sorten und Schüsseln mit gleichem Fingerabdruck bilden eine **Gruppe**. In jeder Gruppe gibt es gleich viele Sorten wie Schüsseln (jede Sorte liegt in einer Schüssel ihrer Gruppe, in jeder Schüssel liegt eine Sorte).

**Nicht eindeutig ist ehrlich:** Vertauscht man in einer Gruppe zwei Sorten, passt die Zuordnung weiter zu **allen** Spießen (jeder Spieß enthält entweder beide Sorten oder keine, und er hat beide Schüsseln benutzt oder keine).
Enthält eine Gruppe eine Wunschsorte **und** eine andere Sorte, kann die Wunschsorte also in jeder Schüssel der Gruppe liegen, es gibt keinen Spieß, der es entscheidet. Die Antwort kann gar nicht eindeutig sein.

**Nie kleiner:** Die Vereinigung besteht aus allen Schüsseln der Gruppen, in denen eine Wunschsorte liegt. Wegen „gleich viele Sorten wie Schüsseln“ hat sie so viele Schüsseln, wie diese Gruppen Sorten haben, also mindestens so viele wie Wunschsorten.
Genau gleich viele sind es, wenn in diesen Gruppen **nur** Wunschsorten liegen. Dann ist die Schüsselmenge eindeutig.

### Zum Schluss der Ecke: Die Gegenprobe mit MiniZinc

Ein **Solver** braucht die Einsicht mit den Mengen nicht: Man beschreibt, was gelten soll, und er sucht eine Zuordnung. Das ist eine **Gegenprobe**, die unabhängig vom Python-Verfahren ist.
Du kennst von den MiniZinc-Arbeitsblättern `alldifferent`, 2D-Arrays (`array2d`), `forall` und **Variablen als Index**. Jede Sorte `f` bekommt eine Schüssel `x[f]`, alle verschieden. Sorte `f` liegt genau dann auf Spieß `s`,
wenn ihre Schüssel `x[f]` für `s` benutzt wurde, also `auf[s, f] = benutzt[s, x[f]]`. Die Sorten sind durchnummeriert (bekannte Sorten alphabetisch, dann die ohne Namen).
Neu ist nur `set of int`, eine Menge von Zahlen (in den Daten als `{1, 3, 4}` geschrieben). **MiniZinc läuft nicht in JupyterLite:** Kopiere das Modell in die MiniZinc IDE oder in den [MiniZinc Playground](https://play.minizinc.org/)
und füge die Daten als `.dzn` dazu (die Python-Zelle darunter erzeugt sie).

```minizinc
include "alldifferent.mzn";
int: n;                                   % Zahl der Sorten und Schüsseln
int: L;                                   % Zahl der Spieße
array[1..L, 1..n] of 0..1: auf;           % auf[s, f] = 1: Sorte f liegt auf Spieß s
array[1..L, 1..n] of 0..1: benutzt;       % benutzt[s, b] = 1: Schüssel b wurde für Spieß s benutzt
set of int: wunsch;                       % Nummern der Wunschsorten
array[1..n] of var 1..n: x;               % x[f]: die Schüssel der Sorte f

constraint alldifferent(x);               % jede Sorte in einer eigenen Schüssel
constraint forall(s in 1..L, f in 1..n)(auf[s, f] = benutzt[s, x[f]]);

solve satisfy;
output ["x = \(x)\n", "Wunschsorten: \([x[f] | f in wunsch])\n"];
```

Der Solver findet **eine** passende Zuordnung, aber ob die Schüsselmenge der Wunschsorten eindeutig ist, sagt sie noch nicht. Dafür ein **zweiter Lauf**: Wir geben die Schüsselmenge `W` der ersten Lösung vor und verlangen,
dass **mindestens eine** Wunschsorte außerhalb von `W` liegt (`exists` ist das Gegenstück zu `forall`: „für mindestens ein `f`“). Antwortet der Solver `UNSATISFIABLE`, gibt es keine solche Zuordnung, und `W` ist eindeutig.
Findet er eine, hat er gezeigt, dass es **nicht** eindeutig ist. Dazu kommen diese Zeilen vor `solve`:
```minizinc
set of int: W;                            % die Schüsselmenge aus der ersten Lösung
constraint exists(f in wunsch)(not (x[f] in W));
```
Die Datei mit den Daten erzeugt die Python-Zelle `als_dzn` (für den zweiten Lauf mit `W`).

````python
def als_dzn(n, wunsch, spiesse, W=None):
    sorten = alle_sorten(n, wunsch, spiesse)
    auf = [1 if sorten[f] in auf_spiess else 0 for auf_spiess, benutzt in spiesse for f in range(n)]
    benutzt = [1 if b in schuesseln else 0 for auf_spiess, schuesseln in spiesse for b in range(1, n + 1)]
    nummern = sorted(sorten.index(w) + 1 for w in wunsch)
    text = 'n = %d;\nL = %d;\nauf = array2d(1..L, 1..n, %s);\nbenutzt = array2d(1..L, 1..n, %s);\nwunsch = %s;' % (
        n, len(spiesse), auf, benutzt, zeige(nummern))
    if W is not None:
        text = text + '\nW = %s;' % zeige(W)
    return text


n, wunsch, spiesse = lies_spiesse('spiesse_mini0.txt')
print(alle_sorten(n, wunsch, spiesse))
print(als_dzn(n, wunsch, spiesse))
````

Zara hat zehn Ferienhäuser. Jedes hat ein Schloss mit einem eigenen **Codewort**, einer Folge von Bits (die Lochkarte). Als Sicherung hat sie eine **elfte Karte** angefertigt: das **XOR aller zehn** Karten.
Ihre Freunde haben unter die elf Karten 100 fremde Karten gemischt. Zara soll herausfinden, welche elf Karten die echten sind.

**Die Einsicht:** XOR der elf echten Karten ist `0`. Denn die Sicherungskarte ist `w1 ^ w2 ^ ... ^ w10`, und `w1 ^ ... ^ w10 ^ (w1 ^ ... ^ w10)` ist `0`, weil sich jede Karte mit sich selbst aufhebt.
Gesucht sind also **`K + 1` Karten, deren XOR `0` ist**. Hier ist `K + 1` die Zahl der echten Karten (im Original 11).

In `daten/` liegen zwei **kleine** Stapel (erste Zeile: `N K M` = Zahl der Karten, `K`, Länge der Karten in Bit; dann je Karte eine Zeile aus `0` und `1`):
`stapel_mini0.txt` (20 Karten, 5 echte) und `stapel_mini1.txt` (20 Karten, 9 echte). Die Karten des Originals mit 111 Karten à 128 Bit gibt es auch, dafür ist aber die Einsicht nötig.

#### Frage Z.2

Eine Auswahl mit XOR `0` ist dasselbe wie: Das XOR der ersten Hälfte der Auswahl ist **gleich** dem XOR der zweiten Hälfte. Warum? Wie kann man das ausnutzen, wenn man `K + 1 = 5` Karten sucht und `N = 200` Karten hat?
Wie viele Zahlen musst du dann ansehen, statt `C(200, 5)`?

<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

`a ^ b ^ c = d ^ e` gilt genau dann, wenn `a ^ b ^ c ^ d ^ e = 0` ist (beide Seiten mit `d ^ e` XORen, `d ^ e ^ d ^ e = 0`). Also teilt man die fünf Karten in **2 + 3**:
Berechne das XOR **aller Paare** und merke sie in einem Dictionary (`XOR -> Liste der Paare`, das sind `C(200, 2) = 19 900` Einträge). Gehe dann alle Dreierauswahlen durch (`C(200, 3) = 1 313 400`)
und schaue nach, ob ihr XOR im Dictionary steht, mit einem Paar ohne gemeinsame Karte. Das sind etwa `1,3 Millionen + 20 000` statt `C(200, 5) ≈ 2,5 · 10^9` Auswahlen. Das ist das Verfahren **Meet in the Middle** („Teile und herrsche“ in den Lösungshinweisen).

</details>

import math
import itertools


def lies_stapel(dateiname):
    """Gibt (N, K, M, Karten) zurück. Jede Karte ist eine Zahl; ihre Bits sind die Löcher."""
    with open('daten/' + dateiname, encoding='utf-8') as datei:
        zeilen = datei.read().split('\n')
    n, k, m = [int(x) for x in zeilen[0].split()]
    karten = [int(z, 2) for z in zeilen[1:n + 1]]
    return n, k, m, karten


def xor_von(karten, indizes):
    ergebnis = 0
    for i in indizes:
        ergebnis = ergebnis ^ karten[i]
    return ergebnis

def suche_teile(karten, k1):
    """Meet in the Middle: Gibt (Auswahl mit XOR 0 als Liste, Zahl der geprüften Auswahlen der zweiten Hälfte) zurück."""
    a = k1 // 2
    b = k1 - a
    tabelle = {}                                 # XOR -> Liste der Auswahlen der Größe a
    for auswahl in itertools.combinations(range(len(karten)), a):
        tabelle.setdefault(xor_von(karten, auswahl), []).append(auswahl)
    geprueft = 0
    for auswahl in itertools.combinations(range(len(karten)), b):
        geprueft += 1
        for partner in tabelle.get(xor_von(karten, auswahl), []):
            if not set(partner) & set(auswahl):
                return sorted(partner + auswahl), geprueft
    return None, geprueft

for name in ['stapel_mini0.txt', 'stapel_mini1.txt']:
    n, k, m, karten = lies_stapel(name)
    auswahl, geprueft = suche_teile(karten, k + 1)
    if auswahl is not None:
        assert xor_von(karten, auswahl) == 0 and len(auswahl) == k + 1
    print(f'{name}: {geprueft} Auswahlen der zweiten Hälfte geprüft -> Karten {auswahl}')
```

Beide Stapel liefern dieselben Karten wie das Durchprobieren, `stapel_mini0` mit 2 + 3 Karten, `stapel_mini1` mit 4 + 5. Weil es nur **eine** Auswahl mit XOR 0 gibt, muss das so sein
(auch das ist ein Vergleich mit dem Schiedsrichter `suche_naiv`). Statt 15 504 und 167 960 Auswahlen sind es nur 326 und 4 011 geprüfte Auswahlen der zweiten Hälfte (dazu kommen die Einträge der Tabelle, `C(20, 2) = 190` und `C(20, 4) = 4 845`). Bei einem Stapel mit 200 Karten und nur 5 echten Karten ist das die Rettung
(so löst es das Lösungsnotebook für `stapel5`).

**Wo es weitergeht.** Bei 11 echten Karten aus 111 (der Stapel der Aufgabe) hilft „2 + 3“ nicht mehr, denn die Tabelle hätte `C(111, 5) = 128 Millionen` Einträge. Dort steckt die **nächste Einsicht**:

# Ablauf bei jeder Aufgabe, bei der die Suche zu groß ist
# 1. vermuten:    kleine Fälle von Hand
# 2. prüfen:      alle kleinen Fälle mit Brute Force (Schiedsrichter)
# 3. begründen:   warum gilt die Eigenschaft immer?
# 4. ausnutzen:   schnelles Verfahren, mit dem Brute Force vergleichen
# 5. zählen:      Zahl der Möglichkeiten vorher und nachher (keine Sekunden)

# Invariante (Münzen): jeder Zug ändert die Zahl der Köpfe um gerade
def loesbar(muster):
    return sum(muster) % 2 == 0

# XOR: Gleiches hebt sich auf
ergebnis = 0
for zahl in alle_nummern + alle_karten:
    ergebnis = ergebnis ^ zahl

# Mengen: Schnitt bei „ja“, Differenz bei „nein“
if sorte in sorten:
    moeglich = moeglich & schuesseln
else:
    moeglich = moeglich - schuesseln
```

- Eine **Invariante** (eine Größe, die bei jedem Zug gleich bleibt) beantwortet „geht das überhaupt?“ ohne Suche. Sie muss **bewiesen** werden, und nach jeder Änderung der Regeln **neu geprüft** (Länge 3 statt 2 machte sie falsch).
- **XOR** hebt gleiche Zahlen auf (`a ^ a = 0`), ist in der Reihenfolge egal, und löst Aufgaben wie „welche Karte fehlt“ in einem Durchgang.
- **Mengen** (`&`, `-`, `|`) streichen ganze Gruppen von Möglichkeiten auf einmal, die Reihenfolge der Hinweise spielt keine Rolle. Die Antwort ist **eindeutig**, wenn genau so viele Möglichkeiten übrig sind wie gesucht werden; sonst meldet man das.
- Eine **erzwungene Reihenfolge** (Matrjoschka nach der Höhe, Käse nach der Regel `(x, y)` vor `(x, z)` vor `(y, z)`) macht aus `n!` Reihenfolgen eine. Die Begründung ist ein „was wächst, passt nie wieder“-Argument.
- Das **Brute Force** ist der Schiedsrichter für kleine Fälle (hier: 100 bis 300 Zufallsfälle), nie der Beweis. Eine **falsche Vermutung** entlarvt ein Gegenbeispiel.
- Ein **Solver** (MiniZinc) kann die Einsicht als unabhängige **Gegenprobe** prüfen, ersetzt sie aber meist nicht.
