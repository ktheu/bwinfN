## 2. Theoretische Analyse

„Theoretische Analyse“ klingt nach Beweisen. In einer Dokumentation für den Bundeswettbewerb meint es etwas Bescheideneres, aber Wichtiges: Du sagst **begründet**, wie sich dein Verfahren bei
größeren Eingaben verhält, **wie schwer** die Aufgabe vermutlich ist und **wie gut** dein Ergebnis ist. Drei Teile:

1. **Laufzeit:** Welche Größen gibt es (`n`, `k`, …)? Wie wächst die Zahl der Schritte? Belegt mit **Zählern** (nicht mit Sekunden, die hängen vom Rechner ab).
2. **Schwierigkeit:** Gibt es ein schnelles exaktes Verfahren, oder ist die Aufgabe mit einem bekannten schweren Problem verwandt? Dann ist eine **Heuristik** gerechtfertigt.
3. **Güte:** Wie weit liegt das Ergebnis höchstens vom Optimum entfernt? Dafür brauchst du **Schranken** ([Notebook 01](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf2/01_pruefen_und_schranken.ipynb)) und **Vergleiche** ([16](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf2/16_experimente.ipynb)).

Unser neutrales Beispiel: **Gewichte in zwei Gruppen teilen.** Gegeben sind `n` Gewichte. Man soll sie so auf zwei Gruppen A und B verteilen, dass der Unterschied der beiden Summen
möglichst klein ist. Zum Beispiel `[3, 3, 2, 2, 2]`: Gruppe A mit 3 + 3 = 6 und Gruppe B mit 2 + 2 + 2 = 6, Unterschied 0.

**Laufzeit des Durchprobierens.** Jedes Gewicht kommt nach A oder B. Das sind `2ⁿ` Möglichkeiten, und wenn man bedenkt, dass A und B vertauscht werden dürfen, noch `2ⁿ⁻¹`.

### Greedy und seine Laufzeit

Ein schnelleres Verfahren ist **Greedy** ([Notebook 03](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf2/03_heuristiken_bauen.ipynb)): Man nimmt die Gewichte der Größe nach, das schwerste zuerst, und legt jedes in die **gerade leichteste Gruppe**.
Das geht für beliebig viele Gruppen `k`. Dazu müssen die Gewichte einmal **sortiert** werden (`O(n log n)`), und für jedes Gewicht sucht man die leichteste Gruppe (`k − 1` Vergleiche), also `O(n · k)`.

#### Aufgabe 2.3

Die Funktion `verteile` ist fertig bis auf einen **Zähler** für die Vergleiche zwischen Gruppenlasten. Ergänze `vergleiche` und gib `(lasten, vergleiche)` zurück.
Teste mit `k = 3` und `n = 100`, `200`, `400` Gewichten (`fall(1, n)`). Wie viele Vergleiche sind es, und was heißt das für die Laufzeit?

**Erwartete Ausgabe:**
```
n = 100 : 200 Vergleiche
n = 200 : 400 Vergleiche
n = 400 : 800 Vergleiche
```

````python
import random

def fall(startwert, n):
    zufall = random.Random(startwert)
    return [zufall.randint(1, 50) for _ in range(n)]

def verteile(gewichte, k):
    """Greedy: das schwerste Gewicht zuerst, immer in die leichteste Gruppe. Gibt die Lasten zurück."""
    lasten = [0] * k
    for w in sorted(gewichte, reverse=True):
        j = 0
        for i in range(1, k):
            if lasten[i] < lasten[j]:
                j = i
        lasten[j] += w
    return lasten

for n in [100, 200, 400]:
    pass     # dein Code
````

### „Wie schwer ist die Aufgabe?“

Bei Optimierungsaufgaben fragt man sich: Gibt es ein **schnelles exaktes** Verfahren, oder lohnt sich eine **Heuristik**? Die Informatik hat dafür einen Begriff:

- Ein Problem liegt in der Klasse **NP**, wenn man eine **vorgeschlagene Lösung schnell prüfen** kann (zum Beispiel: „Gruppe A = diese Gewichte, ist der Unterschied höchstens 0?“ lässt sich sofort nachrechnen).
- Ein Problem heißt **NP-schwer**, wenn es **mindestens so schwer ist wie jedes Problem aus NP**. Gäbe es für ein NP-schweres Problem ein Verfahren mit **polynomieller** Laufzeit (`n²`, `n³`, … im
  Gegensatz zu `2ⁿ`), dann gäbe es eines für alle Probleme in NP. Bis heute hat niemand eines gefunden, und die meisten Fachleute glauben, dass es keines gibt. Bewiesen ist das **nicht** (die Frage heißt „P gegen NP“).
- Bekannte NP-schwere Probleme sind unter anderem das **Rundreiseproblem** (kürzeste Tour durch alle Orte), der **Rucksack**, das **Färben** von Graphen und das **Aufteilen von Gewichten in zwei gleich schwere Gruppen**
  (bei uns oben). Ein **Beweis** gehört nicht in die Dokumentation einer Schülereinsendung, aber der **Verweis** „unsere Aufgabe enthält … als Spezialfall“ schon, wenn du ihn begründen kannst.

Eine Einschränkung, damit du korrekt bleibst: „NP-schwer“ bedeutet **nicht**, dass jede Eingabe schwer ist, und es sagt nichts über Eingaben mit **kleinen Zahlen**. Das Aufteilen von Gewichten
lässt sich zum Beispiel mit einer Tabelle über die möglichen Summen (dynamische Programmierung) schnell lösen, solange die Summe der Gewichte nicht riesig ist.

### B. Modell und Idee

**Modell.** Eine Verteilung `V` ist eine Menge von drei Adressen. Für ein Haus `h` sei `d_V(h)` sein Weg zur nächsten Bude von `V` (Kreisabstand: das Kürzere von `|a − b|` und `L − |a − b|`).
Ein Vorschlag `W` gewinnt gegen `V`, wenn mindestens `N // 2 + 1` Häuser `h` mit `d_W(h) < d_V(h)` existieren. `V` heißt **stabil**, wenn **kein** `W` gegen `V` gewinnt.

**Idee 1: Alles durchprobieren (vollständige Suche).** Es gibt `C(L, 3)` Verteilungen `V`, und für jede muss man alle `W` testen: `C(L, 3)²`. Für `L = 20` sind das 1,3 Millionen, für
`L = 247` etwa `6 · 10¹²` Paare. Das ist für die großen Beispiele zu langsam.

**Idee 2: Gebote.** Für feste `V` und jede **einzelne** Adresse `a` berechnen wir, **welche Häuser** für eine Bude bei `a` stimmen würden (ihr **Gebot**). Ein Vorschlag `W = (a, b, c)` bekommt die Vereinigung
der drei Gebote (ein Haus zählt nur einmal). `V` ist instabil genau dann, wenn es drei Gebote gibt, deren Vereinigung mindestens `N // 2 + 1` Häuser enthält. So muss man `W` nicht mehr vollständig durchgehen: Man sortiert
die Gebote nach der Größe und bricht ab, wenn selbst die größten übrigen nicht mehr reichen.

**Idee 3: Gegenvorschläge merken.** Fast jede Verteilung ist instabil, und derselbe Gegenvorschlag schlägt oft sehr viele davon. Wir merken uns jeden gefundenen Gegenvorschlag (als Liste der Wege aller Häuser).
Für eine neue Verteilung prüfen wir zuerst, ob ein **gemerkter** Gegenvorschlag sie schlägt (ein Vergleich über `N` Häuser). Nur wenn keiner sie schlägt, rechnen wir die Gebote aus („gründlich“).
Eine nützliche Beobachtung: Ein Haus, das direkt an einer Bude wohnt (Weg 0), kann nie mit Ja stimmen.

### C. Die Funktionen

Falls du Bitmasken noch nicht kennst: Eine Menge von Häusern wird als **eine Zahl** gespeichert, `1 << i` ist die Menge `{Haus i}`, `x | y` ist die **Vereinigung** zweier Mengen, und `bit_count()` zählt die Elemente.

````python
def lies_eis(dateiname):
    with open(dateiname) as f:
        zahlen = [int(x) for x in f.read().split()]
    return zahlen[0], zahlen[2:2 + zahlen[1]]          # Umfang, Adressen der Häuser


def abstand(a, b, umfang):
    d = abs(a - b)
    return min(d, umfang - d)
````

Gegen `(0, 4, 5)` haben Haus 0 und Haus 4 Weg 0, sie können nie mit Ja stimmen. Haus 8 hat zur nächsten alten Bude Weg 3 (zu 5). Bei `(2, 6, 10)` ist der Weg 2 (zu 6 oder 10), also Ja, und bei `(0, 4, 8)` ist er 0, ebenfalls Ja.
In beiden Fällen nur **1** Ja-Stimme, die Mehrheit (2) wird nicht erreicht.

</details>

````python
def gebote(alt, umfang, haeuser):
    """Für jede Adresse a die Menge der Häuser, die für eine Bude bei a stimmen würden (als Bitmaske).
    Ergebnis: Liste (Zahl der Häuser, Adresse, Maske), die größte Zahl zuerst."""
    nah = [min(abstand(h, b, umfang) for b in alt) for h in haeuser]
    liste = []
    for a in range(umfang):
        maske = 0
        for i, h in enumerate(haeuser):
            if abstand(h, a, umfang) < nah[i]:
                maske |= 1 << i
        if maske:
            liste.append((maske.bit_count(), a, maske))
    liste.sort(reverse=True)
    return liste


def gegenvorschlag3(alt, umfang, haeuser):
    """Drei Adressen, die zusammen eine Mehrheit der Häuser gewinnen, oder None."""
    mehrheit = len(haeuser) // 2 + 1
    g = gebote(alt, umfang, haeuser)
    n = len(g)
    for x in range(n):
        sx, ax, mx = g[x]
        if 3 * sx < mehrheit:                       # die übrigen Gebote sind nicht größer
            break
        for y in range(x + 1, n):
            sy, ay, my = g[y]
            if sx + 2 * sy < mehrheit:
                break
            mxy = mx | my                           # ein Haus, das für zwei Adressen stimmt, zählt nur einmal
            for z in range(y + 1, n):
                sz, az, mz = g[z]
                if sx + sy + sz < mehrheit:
                    break
                if (mxy | mz).bit_count() >= mehrheit:
                    return (ax, ay, az)
    return None
````

**Was sind „Gebote“?** `gebote(alt, umfang, haeuser)` gibt für jede Adresse `a`, bei der **mindestens ein** Haus mit Ja stimmen würde, die Menge dieser Häuser (als Bitmaske) zurück, die größte Menge zuerst.
`gegenvorschlag3` sucht drei Gebote, deren Vereinigung eine Mehrheit hat (mit **Abbruch**, wenn die übrigen Gebote nicht mehr reichen). Dann gibt es einen Gegenvorschlag.
Sonst ist `alt` **stabil**. Die dritte Funktion, `stabile3`, geht alle Verteilungen durch und **merkt** die Gegenvorschläge (Idee 3).

````python
from itertools import combinations


def stabile3(umfang, haeuser):
    """Alle stabilen Verteilungen von 3 Buden. Gegenvorschläge werden gemerkt (als Liste der Wege aller Häuser)."""
    n = len(haeuser)
    mehrheit = n // 2 + 1
    D = [[abstand(a, h, umfang) for h in haeuser] for a in range(umfang)]     # D[a][i]: Abstand Adresse a zu Haus i
    gemerkt = []
    stabile = []
    gruendlich = 0
    for alt in combinations(range(umfang), 3):
        wege = [min(D[b][i] for b in alt) for i in range(n)]
        geschlagen = False
        for pos, ref in enumerate(gemerkt):
            ja = 0
            for i in range(n):
                if ref[i] < wege[i]:
                    ja += 1
            if ja >= mehrheit:                      # ein gemerkter Gegenvorschlag schlägt alt
                geschlagen = True
                gemerkt.insert(0, gemerkt.pop(pos)) # wer erfolgreich war, kommt nach vorn
                break
        if geschlagen:
            continue
        gruendlich += 1
        w = gegenvorschlag3(alt, umfang, haeuser)
        if w is None:
            stabile.append(alt)
        else:
            gemerkt.append([min(D[b][i] for b in w) for i in range(n)])
    return stabile, gruendlich
````

### F. Theoretische Analyse

**Laufzeit.** Es gibt `C(L, 3) ≈ L³/6` Verteilungen. Die **einfache** Prüfung einer Verteilung (`gegenvorschlag3`) berechnet zuerst die Gebote: für jede der `L` Adressen den Abstand zu allen `N` Häusern, `O(L · N)`.
Dann kommt die Suche nach drei Geboten, die im ungünstigsten Fall `O(L³)` kostet. Zusammen kostet das **Durchprobieren ohne Merken** etwa `O(L³ · (L · N + L³))`. Bei `N ≈ L/2` ist die erste Summe
`O(L⁵)`. Ob sie oder die Suche überwiegt, zeigt der Zähler.

verdoppeln `L` von 8 über 16 auf 32.

**Mit Merken.** Die Gegenvorschläge zu merken senkt die Zahl der gründlichen Prüfungen (D.1: 86 von 161 700). Die **Zahl** hängt davon ab, **in welcher Reihenfolge** die Verteilungen kommen und welche Gegenvorschläge zuerst
gefunden werden, wir haben dafür **keine Garantie**, nur Messungen. Der Zähler gehört deshalb in die Dokumentation, nicht eine Formel.

**Wie schwer ist die Aufgabe?** Für **drei** Buden ist die Zahl der Verteilungen `C(L, 3)`, also ein Polynom in `L`: Die vollständige Suche ist **polynomiell**, nur mit hohem Exponenten (`L⁵`, grob `L⁶`).
Es gibt also keinen Grund, eine Heuristik zu verwenden. Aber der Exponent wächst mit der Zahl der Buden (Erweiterung unten): bei `k` Buden gibt es `C(L, k)` Verteilungen. Ob es für beliebiges `k` ein schnelles Verfahren gibt,
wissen wir nicht, und wir behaupten es nicht. Das Verfahren ist **exakt** (kein Näherungsverfahren): Es findet **alle** stabilen Verteilungen, eine Güteangabe ist nicht nötig, **aber** eine Prüfung auf Richtigkeit: dazu D.2, E und die Gegenprobe in H.

### G. Die Erweiterung: `k` Buden statt drei

**Was.** Das Programm kann eine beliebige Zahl `k` von Buden (1, 2, 3, 4, …), und wir untersuchen, **wie die Zahl der stabilen Verteilungen von `k` abhängt**.

**Warum sinnvoll.** Die Aufgabe fragt nach drei Buden. Die Aufgabe nennt drei Buden. Mit `k` kann man fragen: Gibt es bei **einer** Bude schon stabile Standorte (ein „Median“)? Ändert sich etwas
mit der Zahl der Buden, und was bedeutet es, wenn die Häuserzahl gerade oder ungerade ist? Das ist eine inhaltliche Erweiterung der Fragestellung, keine Verschönerung.

**Eigene Schwierigkeit.**
- Für `k = 3` stehen drei verschachtelte Schleifen über die Gebote. Für beliebiges `k` geht das nicht: Wir brauchen eine **rekursive** Suche (`suche`), die `k` Gebote wählt.
- Der **Abbruch** muss neu überlegt werden: Hat man `noch` Gebote zu wählen und das nächste hat `stimmen` Häuser, dann kann es mit lauter Geboten dieser Größe höchstens `maske.bit_count() + noch · stimmen` werden. Wird selbst das nicht reichen, bricht man ab.
- Die Zahl der Verteilungen wächst mit `C(L, k)`, die Prüfung bleibt nur mit dem **Merken** brauchbar.

**Prüfung.** Drei Dinge:
1. Für `k = 3` müssen dieselben stabilen Verteilungen herauskommen wie mit `stabile3` (alle Beispiele).
2. Für kleine Siedlungen und `k = 1` bis `4` vergleichen wir mit einer **rohen Gegenprobe** (`stabile_roh`: jede Verteilung gegen jede andere abstimmen lassen, ohne jeden Trick).
3. Eine **Messung** der Zahl der stabilen Verteilungen für `k = 1` bis `4` (Tabelle unten).

````python
def gegenvorschlag_k(alt, umfang, haeuser, k):
    """Höchstens k Adressen, die zusammen eine Mehrheit gewinnen, oder None."""
    mehrheit = len(haeuser) // 2 + 1
    g = gebote(alt, umfang, haeuser)

    def suche(start, noch, maske, gewaehlt):
        if maske.bit_count() >= mehrheit:
            return gewaehlt
        if noch == 0:
            return None
        for x in range(start, len(g)):
            stimmen, adresse, m = g[x]
            if maske.bit_count() + noch * stimmen < mehrheit:    # selbst mit lauter gleich großen Geboten zu wenig
                break
            ergebnis = suche(x + 1, noch - 1, maske | m, gewaehlt + [adresse])
            if ergebnis is not None:
                return ergebnis
        return None

    return suche(0, k, 0, [])


def stabile_k(umfang, haeuser, k):
    """Wie stabile3, aber für k Buden. Gibt (stabile Verteilungen, Zahl der gründlichen Prüfungen) zurück."""
    n = len(haeuser)
    mehrheit = n // 2 + 1
    D = [[abstand(a, h, umfang) for h in haeuser] for a in range(umfang)]
    gemerkt = []
    stabile = []
    gruendlich = 0
    for alt in combinations(range(umfang), k):
        wege = [min(D[b][i] for b in alt) for i in range(n)]
        geschlagen = False
        for pos, ref in enumerate(gemerkt):
            ja = 0
            for i in range(n):
                if ref[i] < wege[i]:
                    ja += 1
            if ja >= mehrheit:
                geschlagen = True
                gemerkt.insert(0, gemerkt.pop(pos))
                break
        if geschlagen:
            continue
        gruendlich += 1
        w = gegenvorschlag_k(alt, umfang, haeuser, k)
        if w is None:
            stabile.append(alt)
        else:
            gemerkt.append([min(D[b][i] for b in w) for i in range(n)])
    return stabile, gruendlich
````

````python
def stabile_roh(umfang, haeuser, k):
    """Gegenprobe ohne jeden Trick: jede Verteilung gegen jede andere abstimmen lassen."""
    mehrheit = len(haeuser) // 2 + 1
    stabile = []
    for alt in combinations(range(umfang), k):
        if all(ja_stimmen(alt, neu, umfang, haeuser) < mehrheit for neu in combinations(range(umfang), k)):
            stabile.append(alt)
    return stabile


def siedlung(startwert, umfang, n):
    zufall = random.Random(startwert)
    haeuser = set()
    while len(haeuser) < n:
        haeuser.add(zufall.randint(0, umfang - 1))
    return sorted(haeuser)
````

````python
import random

def lies_eis(dateiname):
    with open(dateiname) as f:
        zahlen = [int(x) for x in f.read().split()]
    return zahlen[0], zahlen[2:2 + zahlen[1]]          # Umfang, Adressen der Häuser


def abstand(a, b, umfang):
    d = abs(a - b)
    return min(d, umfang - d)

def ja_stimmen(alt, neu, umfang, haeuser):
    """Wie viele Häuser stimmen für die neuen Buden `neu` gegen die alten `alt`? (Nur wer näher kommt, stimmt mit Ja.)"""
    anzahl = 0
    for h in haeuser:
        weg_alt = min(abstand(h, b, umfang) for b in alt)
        weg_neu = min(abstand(h, b, umfang) for b in neu)
        if weg_neu < weg_alt:
            anzahl += 1
    return anzahl

def gebote(alt, umfang, haeuser):
    """Für jede Adresse a die Menge der Häuser, die für eine Bude bei a stimmen würden (als Bitmaske).
    Ergebnis: Liste (Zahl der Häuser, Adresse, Maske), die größte Zahl zuerst."""
    nah = [min(abstand(h, b, umfang) for b in alt) for h in haeuser]
    liste = []
    for a in range(umfang):
        maske = 0
        for i, h in enumerate(haeuser):
            if abstand(h, a, umfang) < nah[i]:
                maske |= 1 << i
        if maske:
            liste.append((maske.bit_count(), a, maske))
    liste.sort(reverse=True)
    return liste


def gegenvorschlag3(alt, umfang, haeuser):
    """Drei Adressen, die zusammen eine Mehrheit der Häuser gewinnen, oder None."""
    mehrheit = len(haeuser) // 2 + 1
    g = gebote(alt, umfang, haeuser)
    n = len(g)
    for x in range(n):
        sx, ax, mx = g[x]
        if 3 * sx < mehrheit:                       # die übrigen Gebote sind nicht größer
            break
        for y in range(x + 1, n):
            sy, ay, my = g[y]
            if sx + 2 * sy < mehrheit:
                break
            mxy = mx | my                           # ein Haus, das für zwei Adressen stimmt, zählt nur einmal
            for z in range(y + 1, n):
                sz, az, mz = g[z]
                if sx + sy + sz < mehrheit:
                    break
                if (mxy | mz).bit_count() >= mehrheit:
                    return (ax, ay, az)
    return None

from itertools import combinations


def stabile3(umfang, haeuser):
    """Alle stabilen Verteilungen von 3 Buden. Gegenvorschläge werden gemerkt (als Liste der Wege aller Häuser)."""
    n = len(haeuser)
    mehrheit = n // 2 + 1
    D = [[abstand(a, h, umfang) for h in haeuser] for a in range(umfang)]     # D[a][i]: Abstand Adresse a zu Haus i
    gemerkt = []
    stabile = []
    gruendlich = 0
    for alt in combinations(range(umfang), 3):
        wege = [min(D[b][i] for b in alt) for i in range(n)]
        geschlagen = False
        for pos, ref in enumerate(gemerkt):
            ja = 0
            for i in range(n):
                if ref[i] < wege[i]:
                    ja += 1
            if ja >= mehrheit:                      # ein gemerkter Gegenvorschlag schlägt alt
                geschlagen = True
                gemerkt.insert(0, gemerkt.pop(pos)) # wer erfolgreich war, kommt nach vorn
                break
        if geschlagen:
            continue
        gruendlich += 1
        w = gegenvorschlag3(alt, umfang, haeuser)
        if w is None:
            stabile.append(alt)
        else:
            gemerkt.append([min(D[b][i] for b in w) for i in range(n)])
    return stabile, gruendlich

def gegenvorschlag_k(alt, umfang, haeuser, k):
    """Höchstens k Adressen, die zusammen eine Mehrheit gewinnen, oder None."""
    mehrheit = len(haeuser) // 2 + 1
    g = gebote(alt, umfang, haeuser)

    def suche(start, noch, maske, gewaehlt):
        if maske.bit_count() >= mehrheit:
            return gewaehlt
        if noch == 0:
            return None
        for x in range(start, len(g)):
            stimmen, adresse, m = g[x]
            if maske.bit_count() + noch * stimmen < mehrheit:    # selbst mit lauter gleich großen Geboten zu wenig
                break
            ergebnis = suche(x + 1, noch - 1, maske | m, gewaehlt + [adresse])
            if ergebnis is not None:
                return ergebnis
        return None

    return suche(0, k, 0, [])


def stabile_k(umfang, haeuser, k):
    """Wie stabile3, aber für k Buden. Gibt (stabile Verteilungen, Zahl der gründlichen Prüfungen) zurück."""
    n = len(haeuser)
    mehrheit = n // 2 + 1
    D = [[abstand(a, h, umfang) for h in haeuser] for a in range(umfang)]
    gemerkt = []
    stabile = []
    gruendlich = 0
    for alt in combinations(range(umfang), k):
        wege = [min(D[b][i] for b in alt) for i in range(n)]
        geschlagen = False
        for pos, ref in enumerate(gemerkt):
            ja = 0
            for i in range(n):
                if ref[i] < wege[i]:
                    ja += 1
            if ja >= mehrheit:
                geschlagen = True
                gemerkt.insert(0, gemerkt.pop(pos))
                break
        if geschlagen:
            continue
        gruendlich += 1
        w = gegenvorschlag_k(alt, umfang, haeuser, k)
        if w is None:
            stabile.append(alt)
        else:
            gemerkt.append([min(D[b][i] for b in w) for i in range(n)])
    return stabile, gruendlich

def stabile_roh(umfang, haeuser, k):
    """Gegenprobe ohne jeden Trick: jede Verteilung gegen jede andere abstimmen lassen."""
    mehrheit = len(haeuser) // 2 + 1
    stabile = []
    for alt in combinations(range(umfang), k):
        if all(ja_stimmen(alt, neu, umfang, haeuser) < mehrheit for neu in combinations(range(umfang), k)):
            stabile.append(alt)
    return stabile


def siedlung(startwert, umfang, n):
    zufall = random.Random(startwert)
    haeuser = set()
    while len(haeuser) < n:
        haeuser.add(zufall.randint(0, umfang - 1))
    return sorted(haeuser)

def tabelle(anzahl_haeuser, umfang, startwerte):
    pass     # dein Code
````

## Eisbudendilemma: Dokumentation

### 1. Aufgabe und Lesarten

Gesucht sind drei Standorte am Ufer eines kreisförmigen Sees (Umfang `L`), sodass kein neuer Vorschlag bei der Abstimmung der `N` Häuser eine Mehrheit bekommt (**stabile Verteilung**). Ein Haus stimmt genau dann mit Ja, wenn der neue Vorschlag seinen Weg zur nächsten Bude verkürzt.

- **Standorte:** ganzzahlige Adressen `0 … L − 1` (auch an Hausadressen), drei **verschiedene** Adressen pro Verteilung. (Zwei Buden an einer Adresse wären praktisch nur zwei Buden; ob das die stabilen Verteilungen ändert, haben wir nicht untersucht.)
- **Mehrheit:** „mehr Ja als Nein“, also mindestens `N // 2 + 1` Ja-Stimmen. Bei gerader `N` genügt ein Gleichstand **nicht**.
- Ein Vorschlag darf alle drei Buden gleichzeitig verschieben. Die Buden sind gleich, eine Verteilung ist eine **Menge** von drei Adressen.

### 2. Lösungsidee

*Modell.* `d_V(h)` ist der Weg des Hauses `h` zur nächsten Bude von `V`. Ein Vorschlag `W` gewinnt gegen `V`, wenn `|{ h : d_W(h) < d_V(h) }| ≥ N // 2 + 1`. `V` ist stabil, wenn kein `W` gewinnt.

*Gebote.* Für feste `V` und jede Adresse `a` sei `G(a) = { h : dist(h, a) < d_V(h) }` das **Gebot** von `a`. Dann gewinnt `W = {a, b, c}` genau dann, wenn `|G(a) ∪ G(b) ∪ G(c)| ≥ N // 2 + 1`. Die Gebote werden der Größe nach sortiert; die Suche bricht ab, wenn
selbst die größten übrigen Gebote die Mehrheit nicht mehr erreichen.

*Merken.* Wir merken jeden gefundenen Gegenvorschlag `W` als Liste der Wege `d_W`. Eine neue Verteilung `V` ist **instabil**, wenn ein gemerktes `W` sie schlägt, das kostet `N` Vergleiche. Nur wenn keines schlägt, berechnen wir die Gebote („gründlich“).

*Begründung der Richtigkeit.* Das Verfahren erklärt `V` nur dann für stabil, wenn die Gebote-Suche **keinen** Gegenvorschlag findet; die Gebote-Suche ist vollständig (sie probiert alle Dreiergruppen von Geboten, bis auf die Abbrüche, die nur Dreiergruppen mit zu wenig Stimmen überspringen).
Als Gegenprobe dienen (a) von Hand begründete eigene Beispiele, (b) eine **rohe** Prüfung ohne Gebote und Merken (`stabile_roh`, keine Abweichung in 20 Fällen: 5 kleine Siedlungen, `k = 1` bis `4`) und (c) ein **MiniZinc**-Modell (gleiche stabile Verteilungen bei 5 Dateien).

### 3. Theoretische Analyse

- *Laufzeit.* Es gibt `C(L, 3)` Verteilungen. Die einfache Prüfung einer Verteilung berechnet die Gebote in `O(L · N)`; ohne Merken ist das Verfahren mit `N ≈ L/2` `O(L⁵)` (zusätzlich die Suche nach drei Geboten, im ungünstigsten Fall `O(L³)` pro Verteilung).
  *Beleg:* Zählt man die Abstandsberechnungen bei `L = 8, 16, 32`, sind es 2 464, 85 120 und 2 777 600, der Faktor beim Verdoppeln ist 34,5 und 32,6, also etwa `2⁵`.
- *Mit Merken* prüfen wir nur wenige Verteilungen gründlich (Tabelle). Eine **Schranke** dafür haben wir nicht: Die Zahl hängt von der Reihenfolge und den zuerst gefundenen Gegenvorschlägen ab.
- *Schwierigkeit.* Bei drei Buden ist die Aufgabe polynomiell (`C(L, 3)` Verteilungen), eine Heuristik ist nicht nötig. Bei `k` Buden wächst der Exponent mit `k`; ob es für beliebiges `k` ein schnelles Verfahren gibt, wissen wir nicht.
- *Güte.* Das Verfahren ist **exakt**: Es liefert **alle** stabilen Verteilungen oder stellt fest, dass es keine gibt. Eine Näherung ist es nicht, deshalb keine Güteangabe, aber die Gegenproben oben.
