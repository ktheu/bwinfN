- **Reihenfolge der Seiten:** links, oben, rechts, unten (also im Uhrzeigersinn). Teil 0 `acBd` liegt in der Lösung von Beispiel 1

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Backtracking (Rückverfolgung):** Wir legen Feld für Feld ein passendes Teil. Passt auf ein Feld nichts mehr, nehmen wir das
  zuletzt gelegte Teil wieder weg und probieren dort das nächste. Das schreibt sich am einfachsten als **rekursive Funktion**. Damit
  findet die Suche *alle* Lösungen, und nur so können wir sagen, ob ein Puzzle eindeutig lösbar ist.
- **Nachschlagen statt Durchprobieren:** Eine Tabelle (Dictionary), die zu den gewünschten Seiten gleich die passenden Teile liefert.
- **Reihenfolge der Felder:** Felder mit vielen schon belegten Nachbarn zuerst, dort passen nur wenige Teile.
- **Symmetrie ausnutzen:** Jede Lösung kann man als Ganzes drehen und bekommt wieder eine Lösung. Wir suchen nur eine davon.
- **Bergsteigen (für Teil 2):** Ein zufälliges Puzzle schrittweise verändern und eine Änderung nur behalten, wenn sie nicht schadet.
- Zur Laufzeit: Wie viele Möglichkeiten gibt es? Fakultät und Potenzen, mit denen wir abschätzen, warum Durchprobieren nicht geht.

**Nützliche Funktionen und Bibliotheken**
- Zeichenketten: `teil[k:] + teil[:k]` verschiebt die Buchstaben um `k` Stellen (damit drehen wir ein Teil). `'a'.swapcase()` ist
  `'A'` und `'A'.swapcase()` ist `'a'`: So bekommen wir zu jeder Seite die passende Gegenseite.
- Dictionarys mit Tupeln als Schlüssel, z. B. `feld[(zeile, spalte)]`.
- `math.factorial(n)` berechnet n! = 1 · 2 · … · n.
- `random.Random(zahl)` ist ein Zufallsgenerator mit festem Startwert: Er liefert bei jedem Lauf dieselben „Zufallszahlen“, damit die
  Ausgaben im Notebook gleich bleiben. Davon brauchen wir `choice(folge)`, `randrange(n)` und `shuffle(liste)`.

## Die Idee in Worten

**Darstellung.** Ein Teil ist eine Zeichenkette aus vier Buchstaben: links, oben, rechts, unten. Drehen wir das Teil um eine
Vierteldrehung gegen den Uhrzeigersinn, rutscht die obere Seite nach links, die rechte nach oben usw. Die Zeichenkette wird also
um eine Stelle nach links verschoben: aus `acBd` wird `cBda`. Jedes Teil hat so höchstens vier **Lagen**. Zwei Seiten passen
aneinander, wenn die eine der Großbuchstabe der anderen ist.

**Wie viele Möglichkeiten gibt es?** Beim 5 × 5-Puzzle können 25 Teile auf das erste Feld, 24 auf das zweite usw., und jedes in vier
Lagen. Das sind 25! · 4^25 Anordnungen, eine Zahl mit 40 Stellen (siehe die Hinweise). Alle durchzuprobieren ist hoffnungslos.

**Backtracking.** Wir füllen die Felder nacheinander und legen auf jedes nur Teile, die zu den schon belegten Nachbarn passen. Passt
nichts, ist dieser Weg eine Sackgasse, und wir gehen zurück. Die meisten der riesig vielen Anordnungen sehen wir so nie an: Schon
nach wenigen Teilen passt meist nichts mehr.

**Schneller werden.** Wir verbessern die Suche in drei Schritten, wie die Lösungshinweise es vorschlagen:
1. **Kandidatenlisten:** Vorher eine Tabelle anlegen, welche Lagen zu welchen Nachbarseiten passen, statt jedes Mal alle Teile
   durchzuprobieren.
2. **Gute Reihenfolge der Felder:** möglichst bald Felder belegen, die schon zwei belegte Nachbarn haben. Dort passen im Mittel nur
   etwa 100 / 64 ≈ 1,6 der 100 Lagen eines 5 × 5-Puzzles, bei nur einem Nachbarn etwa 100 / 8 = 12,5.
3. **Drehung des ganzen Puzzles:** Mit jeder Lösung sind auch ihre Drehungen um 90°, 180° und 270° Lösungen (bei einem nicht
   quadratischen Puzzle nur die um 180°). Wir legen ein Teil ungedreht fest und suchen nur noch die Lösungen, in denen es so liegt.
   Das spart viel Arbeit und zählt jede Lösung „bis auf Drehung“ genau einmal.

**Teil 2.** Ein lösbares Puzzle zu erfinden ist leicht: Wir füllen das Feld mit zufälligen Teilen, die aneinander passen. Eindeutig
lösbar ist es aber selten. Dann ändern wir immer wieder eine zufällige Seite (und die passende Seite des Nachbarn) und behalten die
Änderung, wenn das Puzzle dadurch nicht mehr Lösungen bekommt. Die Lösung aus Teil 1 zählt die Lösungen.

## Schritt 1: Puzzles einlesen, Teile drehen, Lösungen anzeigen

Ein Puzzle besteht aus `m`, `n` und der Liste der Teile. `drehe(teil, k)` verschiebt um `k` Stellen. `lagen(teil)` liefert die
verschiedenen Lagen eines Teils: meist vier, aber ein Teil wie `abab` sieht nach einer halben Drehung wieder gleich aus und hat nur
zwei, `aaaa` sogar nur eine. Damit wir solche Lagen nicht doppelt zählen, nehmen wir jede nur einmal.

Zwei Teile sind „genau gleich“, wenn das eine eine Lage des anderen ist (man kann sie ja drehen). Als **Normalform** eines Teils
nehmen wir seine im Alphabet kleinste Lage. Gleiche Teile haben dieselbe Normalform.

````python
def puzzle_aus_text(text):
    """Text im Eingabeformat -> (m, n, Liste der Teile)."""
    woerter = text.split()
    m = int(woerter[0])
    n = int(woerter[1])
    teile = woerter[2:]
    return m, n, teile


def lies_puzzle(dateiname):
    with open(dateiname) as datei:
        return puzzle_aus_text(datei.read())


def drehe(teil, k):
    """Vierteldrehungen gegen den Uhrzeigersinn: Die Seiten rutschen um k Stellen nach links."""
    return teil[k:] + teil[:k]


def lagen(teil):
    """Die verschiedenen Lagen eines Teils (ohne Wiederholungen)."""
    ergebnis = []
    for k in range(4):
        if drehe(teil, k) not in ergebnis:
            ergebnis.append(drehe(teil, k))
    return ergebnis


def normalform(teil):
    return min(lagen(teil))


def alle_verschieden(teile):
    formen = set()
    for teil in teile:
        formen.add(normalform(teil))
    return len(formen) == len(teile)


m1, n1, teile1 = lies_puzzle('beispieldaten/beispiel1.txt')
print(m1, n1, teile1)
print('alle verschieden:', alle_verschieden(teile1))
print('Lagen von acBd:', lagen('acBd'))
print('Lagen von abab:', lagen('abab'), ' Normalform von cBda:', normalform('cBda'))
````

Eine **Lösung** speichern wir als Dictionary: Zu jedem Feld `(zeile, spalte)` gehört ein Paar `(nummer, lage)`, also welches Teil
dort in welcher Lage liegt. `zeige` gibt sie so aus wie die Lösungshinweise: Jedes Teil steht auf drei Zeilen, die obere Seite
oben, die linke und rechte in der Mitte, die untere unten. Rechts daneben stehen die Nummern der Teile.

`ist_loesung` prüft eine Lösung: Jedes Teil kommt genau einmal vor, jede Lage gehört zu ihrem Teil, und an jeder inneren Kante
passen die Seiten. Die Richtungen nummerieren wir wie die Seiten: 0 links, 1 oben, 2 rechts, 3 unten. Die Seite des Nachbarn, die
uns gegenüberliegt, hat die Nummer `(richtung + 2) % 4`: Der linke Nachbar zeigt uns seine rechte Seite.

````python
NACHBARN = [(0, -1), (-1, 0), (0, 1), (1, 0)]     # links, oben, rechts, unten als (dz, ds)


def passt(seite1, seite2):
    return seite1 == seite2.swapcase()


def zeige(m, n, loesung):
    for z in range(m):
        oben = ''
        mitte = ''
        unten = ''
        nummern = ''
        for s in range(n):
            nr, t = loesung[(z, s)]
            oben = oben + ' ' + t[1] + '  '
            mitte = mitte + t[0] + ' ' + t[2] + ' '
            unten = unten + ' ' + t[3] + '  '
            nummern = nummern + f'{nr:3}'
        print(oben)
        print(mitte + '   ' + nummern)
        print(unten)


def ist_loesung(m, n, teile, loesung):
    nummern = []
    for z in range(m):
        for s in range(n):
            nr, t = loesung[(z, s)]
            nummern.append(nr)
            if t not in lagen(teile[nr]):
                return False
            for richtung in range(4):
                dz, ds = NACHBARN[richtung]
                if (z + dz, s + ds) in loesung:
                    t2 = loesung[(z + dz, s + ds)][1]
                    if not passt(t[richtung], t2[(richtung + 2) % 4]):
                        return False
    return sorted(nummern) == list(range(len(teile)))


# die Lösung von Beispiel 1 aus den Lösungshinweisen
hinweis1 = {(0, 0): (0, 'acBd'), (0, 1): (2, 'bDda'), (1, 0): (1, 'aDac'), (1, 1): (3, 'AAab')}
zeige(m1, n1, hinweis1)
print('Lösung?', ist_loesung(m1, n1, teile1, hinweis1))
````

Teil 2 liegt in der Lage `bDda`, also `drehe('abDd', 1)`, und Teil 3 in der Lage `AAab` = `drehe('abAA', 2)`. An allen vier inneren
Kanten passen die Seiten: `B` an `b`, `d` an `D`, `a` an `A` und `a` an `A`. Die Seiten am Rand spielen keine Rolle.

## Schritt 2: Die einfachste Suche

Wir belegen die Felder **zeilenweise** von links oben nach rechts unten. Auf das Feld mit der Nummer `pos` (Zeile `pos // n`, Spalte
`pos % n`) probieren wir jedes noch freie Teil in jeder Lage. Passt es zum linken und zum oberen Nachbarn (die sind schon belegt),
legen wir es hin und machen mit dem nächsten Feld weiter. Danach nehmen wir es wieder weg und probieren die nächste Möglichkeit.
Sind alle Felder belegt, haben wir eine Lösung gefunden und merken sie uns.

Alle Angaben, die die rekursive Funktion braucht, stehen in einem Dictionary `stand`. Dort zählen wir auch mit: `geprueft` zählt die
geprüften Lagen, `gelegt` die gelegten Teile. Diese Zähler nehmen wir statt Rechenzeiten, um die Verfahren zu vergleichen.

````python
def lege_einfach(pos, stand):
    m = stand['m']
    n = stand['n']
    teile = stand['teile']
    feld = stand['feld']
    if pos == m * n:                                   # alle Felder belegt: eine Lösung
        stand['loesungen'].append(dict(feld))
        return
    z = pos // n
    s = pos % n
    for nr in range(len(teile)):
        if stand['frei'][nr]:
            for t in lagen(teile[nr]):
                stand['geprueft'] += 1
                if s > 0 and not passt(feld[(z, s - 1)][1][2], t[0]):    # rechte Seite des linken Nachbarn
                    continue
                if z > 0 and not passt(feld[(z - 1, s)][1][3], t[1]):    # untere Seite des oberen Nachbarn
                    continue
                stand['gelegt'] += 1
                feld[(z, s)] = (nr, t)                 # legen ...
                stand['frei'][nr] = False
                lege_einfach(pos + 1, stand)
                stand['frei'][nr] = True               # ... und wieder wegnehmen
                del feld[(z, s)]


def suche_einfach(m, n, teile):
    stand = {'m': m, 'n': n, 'teile': teile, 'feld': {}, 'frei': [True] * len(teile),
             'loesungen': [], 'geprueft': 0, 'gelegt': 0}
    lege_einfach(0, stand)
    return stand


for nummer in [1, 2]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    stand = suche_einfach(m, n, teile)
    print(f'Beispiel {nummer}: {len(stand["loesungen"])} Lösungen, {stand["geprueft"]} Lagen geprüft, '
          f'{stand["gelegt"]} Teile gelegt')
zeige(m, n, stand['loesungen'][0])
````

Beide Puzzles haben **4 Lösungen**: Es ist immer dieselbe Lösung, einmal so, wie die Suche sie zuerst findet, und dann um 90°, 180°
und 270° gedreht. Bis auf Drehung sind beide Puzzles also eindeutig lösbar. Die gezeigte Lösung von Beispiel 2 ist genau die aus den
Lösungshinweisen.

Für Beispiel 2 prüft die Suche 43 280 Lagen. Bei Beispiel 3 (4 × 4) sind es schon etwa 16 Millionen; das rechnen wir hier
nicht vor. Fast alle Prüfungen sind vergeblich: Von den 64 Lagen der 16 Teile passen auf ein Feld mit zwei belegten Nachbarn im
Mittel nur etwa 64 / 64 = 1, denn jede der beiden Seiten passt mit einer Wahrscheinlichkeit von 1/8.

## Schritt 3: Kandidatenlisten

Statt alle Teile durchzuprobieren, legen wir vorher eine Tabelle an. Der Schlüssel ist ein **Wunsch**: ein Tupel aus vier Zeichen
für die Seiten links, oben, rechts, unten, wobei `'?'` „egal“ heißt. Unter dem Schlüssel stehen alle Lagen, die den Wunsch
erfüllen, als Paare `(nummer, lage)`. Jede Lage tragen wir unter 2 · 2 · 2 · 2 = 16 Wünschen ein: Für jede ihrer Seiten entweder
den Buchstaben oder `'?'`.

`wunsch(feld, z, s)` berechnet, was ein Teil auf Feld `(z, s)` braucht: an jeder Seite die Gegenseite des Nachbarn, falls dort schon
ein Teil liegt, sonst `'?'`. Am Rand gibt es keinen Nachbarn, dort ist es also immer egal.

Die Suche bekommt die Reihenfolge der Felder als Liste `folge`. Damit können wir in Schritt 4 verschiedene Reihenfolgen vergleichen.
Mit `stand['hoechstens']` kann man die Suche anhalten lassen, sobald so viele Lösungen gefunden sind (das brauchen wir in Teil 2).

````python
def kandidatentabelle(teile):
    tabelle = {}
    for nr in range(len(teile)):
        for t in lagen(teile[nr]):
            for links in [t[0], '?']:
                for oben in [t[1], '?']:
                    for rechts in [t[2], '?']:
                        for unten in [t[3], '?']:
                            schluessel = (links, oben, rechts, unten)
                            if schluessel not in tabelle:
                                tabelle[schluessel] = []
                            tabelle[schluessel].append((nr, t))
    return tabelle


def wunsch(feld, z, s):
    """Welche Seiten braucht ein Teil auf Feld (z, s)? '?' heißt: egal."""
    ergebnis = []
    for richtung in range(4):
        dz, ds = NACHBARN[richtung]
        if (z + dz, s + ds) in feld:
            nachbar = feld[(z + dz, s + ds)][1]
            ergebnis.append(nachbar[(richtung + 2) % 4].swapcase())
        else:
            ergebnis.append('?')
    return tuple(ergebnis)


def lege(folge, i, stand):
    feld = stand['feld']
    if i == len(folge):
        stand['loesungen'].append(dict(feld))
        return
    z, s = folge[i]
    w = wunsch(feld, z, s)
    if w in stand['tabelle']:
        for nr, t in stand['tabelle'][w]:
            if stand['frei'][nr] and len(stand['loesungen']) < stand['hoechstens']:
                stand['gelegt'] += 1
                feld[(z, s)] = (nr, t)
                stand['frei'][nr] = False
                lege(folge, i + 1, stand)
                stand['frei'][nr] = True
                del feld[(z, s)]


def neuer_stand(teile, hoechstens):
    return {'tabelle': kandidatentabelle(teile), 'feld': {}, 'frei': [True] * len(teile),
            'loesungen': [], 'gelegt': 0, 'hoechstens': hoechstens}


def suche_mit_folge(teile, folge, hoechstens=10 ** 9):
    stand = neuer_stand(teile, hoechstens)
    lege(folge, 0, stand)
    return stand


def zeilenweise(m, n):
    folge = []
    for z in range(m):
        for s in range(n):
            folge.append((z, s))
    return folge


tabelle1 = kandidatentabelle(teile1)
print('Wunsch (?, ?, ?, ?):', len(tabelle1[('?', '?', '?', '?')]), 'Lagen')
print('Wunsch (b, ?, ?, ?):', tabelle1[('b', '?', '?', '?')])
print('Wunsch (b, D, ?, ?):', tabelle1[('b', 'D', '?', '?')])
print()
for nummer in [1, 2, 3]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    stand = suche_mit_folge(teile, zeilenweise(m, n))
    print(f'Beispiel {nummer}: {len(stand["loesungen"])} Lösungen, {stand["gelegt"]} Teile gelegt')
````

Die Suche legt genau dieselben Teile wie in Schritt 2 (bei Beispiel 2 sind es 2 153 Teile), sie spart sich nur das
vergebliche Prüfen. Deshalb schafft sie jetzt auch Beispiel 3 und findet die Lösung samt ihrer drei Drehungen. Dafür legt sie
444 380 Teile. Wie viele Teile gelegt werden, hängt jetzt nur noch von der **Reihenfolge der Felder** ab.

## Schritt 4: Eine bessere Reihenfolge der Felder

Bei der zeilenweisen Reihenfolge hat jedes Feld der ersten Zeile nur einen belegten Nachbarn (links), auf jedes passen also etwa
ein Achtel aller Lagen. Erst ab der zweiten Zeile gibt es zwei Nachbarn. Die Lösungshinweise schlagen deshalb vor, das Quadrat **entlang der
Diagonalen** wachsen zu lassen (Abbildung 1b der Hinweise): erst das 1 × 1-Quadrat links oben, dann rechts eine Spalte und unten eine
Zeile dazu, sodass es 2 × 2 groß ist, dann 3 × 3 usw. Für ein 5 × 5-Puzzle ergibt das

```
 1  2  5 10 17
 3  4  6 11 18
 7  8  9 12 19
13 14 15 16 20
21 22 23 24 25
```

Auch hier haben 16 Felder zwei belegte Nachbarn, sie kommen aber früher dran. Für nicht quadratische Puzzles ist nicht klar, wie
man so eine Reihenfolge fortsetzt. Allgemeiner ist die Regel **„meiste Nachbarn zuerst“**: Als nächstes Feld nehmen wir eines mit
möglichst vielen schon eingeplanten Nachbarn, bei Gleichstand das nächstgelegene zum Startfeld und dann das erste in Lesereihenfolge.
Das klappt für jedes Rechteck und jedes Startfeld; das brauchen wir in Schritt 5.

````python
def diagonal(m, n):
    """Reihenfolge wie Abbildung 1b der Hinweise (das Quadrat links oben wächst)."""
    folge = []
    for k in range(max(m, n)):
        for z in range(k):                 # neue Spalte k, von oben bis über die Diagonale
            if z < m and k < n:
                folge.append((z, k))
        for s in range(k + 1):             # neue Zeile k, von links bis zur Diagonale
            if k < m and s < n:
                folge.append((k, s))
    return folge


def meiste_nachbarn(m, n, start):
    folge = [start]
    eingeplant = {start}
    while len(folge) < m * n:
        bestes = None
        for z in range(m):
            for s in range(n):
                if (z, s) not in eingeplant:
                    anzahl = 0
                    for dz, ds in NACHBARN:
                        if (z + dz, s + ds) in eingeplant:
                            anzahl += 1
                    abstand = abs(z - start[0]) + abs(s - start[1])
                    wert = (-anzahl, abstand, z, s)        # kleinster Wert gewinnt
                    if anzahl > 0 and (bestes is None or wert < bestes):
                        bestes = wert
        feld = (bestes[2], bestes[3])
        folge.append(feld)
        eingeplant.add(feld)
    return folge


def als_tabelle(m, n, folge):
    """Zeigt, als wievieltes jedes Feld belegt wird."""
    platz = {}
    for i in range(len(folge)):
        platz[folge[i]] = i + 1
    for z in range(m):
        zeile = ''
        for s in range(n):
            zeile = zeile + f'{platz[(z, s)]:3}'
        print(zeile)


als_tabelle(5, 5, diagonal(5, 5))
print()
als_tabelle(5, 5, meiste_nachbarn(5, 5, (0, 0)))
print()
als_tabelle(3, 5, meiste_nachbarn(3, 5, (1, 2)))
````

Von der Ecke aus ergibt „meiste Nachbarn zuerst“ genau die diagonale Reihenfolge der Hinweise. Vom mittleren Feld eines
3 × 5-Rechtecks aus wächst ein Block nach außen. Wie viele Teile legt die Suche mit den Reihenfolgen?

````python
for nummer in [1, 2, 3]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    a = suche_mit_folge(teile, zeilenweise(m, n))['gelegt']
    b = suche_mit_folge(teile, diagonal(m, n))['gelegt']
    c = suche_mit_folge(teile, meiste_nachbarn(m, n, (0, 0)))['gelegt']
    print(f'Beispiel {nummer}: zeilenweise {a:8}   diagonal {b:8}   meiste Nachbarn {c:8}')
````

Bei Beispiel 3 legt die diagonale Reihenfolge 169 071 statt 444 380 Teile, weniger als die Hälfte.
„Meiste Nachbarn zuerst“ legt von der Ecke aus gleich viele Teile, es ist ja dieselbe Reihenfolge.

## Schritt 5: Die Drehung des ganzen Puzzles ausnutzen

Jede Lösung kommt viermal vor (bei einem nicht quadratischen Puzzle zweimal, denn nach einer Vierteldrehung hätte es ja `n` Zeilen
und `m` Spalten). Das ist Arbeit, die wir uns sparen können. Wir wählen ein **Ankerteil** und erlauben ihm nur die Lage `drehe(teil,
0)`, also ungedreht. Dreht man eine Lösung als Ganzes um 90°, wird auch jedes einzelne Teil um 90° gedreht. In den vier Drehungen
einer Lösung liegt der Anker also in seinen vier verschiedenen Lagen, und genau eine davon ist die ungedrehte. Bei einem nicht
quadratischen Puzzle erlauben wir dem Anker die Lagen 0 und 1, denn eine halbe Drehung macht aus Lage 0 die Lage 2 und aus Lage 1
die Lage 3.

Das klappt nur mit einem Anker, der in jeder Lage anders aussieht, also nicht mit `abab` oder `aaaa`. Wir nehmen das erste Teil, das
nach einer halben Drehung anders aussieht. (Dann sind auch alle vier Lagen verschieden, denn ein Teil, das nach einer
Vierteldrehung gleich aussieht, sieht auch nach einer halben gleich aus.) Gibt es keines, zählen wir alle Lösungen und teilen
durch 4 bzw. 2.

Wo der Anker liegt, wissen wir nicht. Deshalb probieren wir jedes Feld als **Startfeld** für den Anker und füllen von dort aus mit
„meiste Nachbarn zuerst“. Ein Puzzle ist genau dann eindeutig lösbar, wenn `loese` genau eine Lösung findet.

````python
def ankerteil(teile):
    for nr in range(len(teile)):
        if drehe(teile[nr], 2) != teile[nr]:
            return nr
    return None


def loese(m, n, teile, hoechstens=10 ** 9, startfelder=None):
    anker = ankerteil(teile)
    faktor = 1                                       # so oft kommt jede Lösung vor
    if anker is None:                                # Sonderfall: alle Teile wie abab oder aaaa
        anker = 0
        ankerlagen = lagen(teile[0])
        if m == n:
            faktor = 4
        else:
            faktor = 2
    elif m == n:
        ankerlagen = [teile[anker]]
    else:
        ankerlagen = [teile[anker], drehe(teile[anker], 1)]
    if startfelder is None:
        startfelder = zeilenweise(m, n)
    stand = neuer_stand(teile, hoechstens * faktor)
    for start in startfelder:
        folge = meiste_nachbarn(m, n, start)
        for t in ankerlagen:
            if len(stand['loesungen']) < stand['hoechstens']:
                stand['gelegt'] += 1
                stand['feld'][start] = (anker, t)
                stand['frei'][anker] = False
                lege(folge, 1, stand)                # Feld 0 der Folge ist schon belegt
                stand['frei'][anker] = True
                del stand['feld'][start]
    stand['anzahl'] = len(stand['loesungen']) // faktor
    return stand


for nummer in [1, 2, 3]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    stand = loese(m, n, teile)
    print(f'Beispiel {nummer}: {stand["anzahl"]} Lösung(en) bis auf Drehung, {stand["gelegt"]} Teile gelegt')
````

Die Beispiele 1 bis 3 sind **eindeutig lösbar**, so wie es die Aufgabe verspricht. Für Beispiel 3 legt die Suche jetzt 75 544
Teile, statt 169 071 mit der diagonalen Reihenfolge und 444 380 zeilenweise. Ganz auf ein Viertel kommen wir gegenüber
der diagonalen Reihenfolge nicht: Die Suche beginnt jetzt auch auf Feldern am Rand und in der Mitte, und von dort aus wächst sie
anders als von der Ecke.

Die Lösung, die `loese` findet, ist gegenüber den Hinweisen womöglich gedreht (dort liegt der Anker nicht unbedingt ungedreht). Zum
Vergleich drehen wir sie als Ganzes: Bei einer Vierteldrehung gegen den Uhrzeigersinn kommt das Teil von Feld `(z, s)` auf Feld
`(n - 1 - s, z)` und wird selbst mit `drehe(t, 1)` gedreht. Aus einem m × n-Feld wird dabei ein n × m-Feld.

````python
def drehe_ganzes(m, n, loesung):
    """Vierteldrehung des ganzen Puzzles gegen den Uhrzeigersinn; danach hat es n Zeilen und m Spalten."""
    neu = {}
    for (z, s), (nr, t) in loesung.items():
        neu[(n - 1 - s, z)] = (nr, drehe(t, 1))
    return neu


def nummern(m, n, loesung):
    ergebnis = []
    for z in range(m):
        zeile = []
        for s in range(n):
            zeile.append(loesung[(z, s)][0])
        ergebnis.append(zeile)
    return ergebnis


# Nummern der Lösung von Beispiel 3 aus den Hinweisen
hinweis3 = [[0, 14, 9, 4], [8, 13, 12, 15], [11, 7, 1, 2], [5, 6, 3, 10]]
m, n, teile = lies_puzzle('beispieldaten/beispiel3.txt')
loesung = loese(m, n, teile)['loesungen'][0]
for k in range(4):
    if nummern(m, n, loesung) == hinweis3:
        print(f'nach {k} Vierteldrehungen wie in den Hinweisen:')
        zeige(m, n, loesung)
    loesung = drehe_ganzes(m, n, loesung)
````

## Wie lange dauert die Suche?

Die Hinweise erwarten eine Abschätzung mit kombinatorischen Überlegungen. Ohne jede Prüfung gäbe es (m·n)! · 4^(m·n) Anordnungen.
Das Backtracking sieht davon nur einen winzigen Teil an:

````python
import math

gelegt_5x5 = 41536511            # außerhalb des Notebooks mit loese gezählt (alle 25 Startfelder)
for nummer in [1, 2, 3, 4]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    anordnungen = math.factorial(m * n) * 4 ** (m * n)
    if nummer < 4:
        gelegt = loese(m, n, teile)['gelegt']
    else:
        gelegt = gelegt_5x5
    print(f'{m} x {n}: {anordnungen:.1e} Anordnungen, gelegt {gelegt:10}')
````

Warum wächst die Suche trotzdem so schnell? Sehen wir uns an, wie viele **Teillösungen** es gibt, also Belegungen eines Teils der
Felder, bei denen alle Nachbarn passen. Für ein Quadrat aus k × k Feldern, das wir mit Teilen aus einem 5 × 5-Puzzle belegen,
gibt es 25 · 24 · … · (25 − k² + 1) · 4^(k²) Belegungen. Jede der 2 · k · (k − 1) inneren Kanten passt nur mit einer
Wahrscheinlichkeit von etwa 1/8, wenn die Seiten zufällig sind. Erwartet werden also ungefähr

$$\frac{25!}{(25-k^2)!} \cdot 4^{k^2} \cdot 8^{-2k(k-1)}$$

passende Belegungen. Eine Suche, die von der Ecke aus wächst, muss jede davon ansehen, denn erst die Teile, die danach kommen,
entscheiden, ob es weitergeht. Gegen Ende fallen die Teillösungen schnell weg, weil dann jedes Feld zwei belegte Nachbarn hat:

````python
for k in range(1, 6):
    belegungen = math.factorial(25) // math.factorial(25 - k * k) * 4 ** (k * k)
    erwartet = belegungen / 8 ** (2 * k * (k - 1))
    print(f'{k} x {k}-Quadrat: etwa {erwartet:12.0f} passende Belegungen')
````

Für das 4 × 4-Quadrat links oben sind es etwa 39 Millionen, dieselbe Größenordnung wie die 41,5 Millionen gelegten Teile bei
Beispiel 4. Die meiste Arbeit steckt also in der Mitte der Suche. Für das ganze 5 × 5-Quadrat kommt die Formel auf etwa 13 000
Lösungen, obwohl Beispiel 4 nur eine hat (mit ihren Drehungen vier). Die Formel ist eben nur eine grobe Schätzung: In einem echten
Puzzle sind die Seiten nicht unabhängig voneinander zufällig. Bei einem 6 × 6-Puzzle mit 36 Teilen käme sie in der Mitte der Suche
schon auf Billionen von Teillösungen.

## Teil 2: Eindeutige Puzzles erfinden

**Ein lösbares Puzzle** bauen wir zeilenweise: Jedes Feld bekommt vier zufällige Seiten aus `abcdABCD`, nur die linke Seite ist die
Gegenseite der rechten Seite des linken Nachbarn, die obere die Gegenseite der unteren Seite des oberen Nachbarn. Sind dabei zwei
gleiche Teile entstanden, fangen wir neu an. (Die Teile am Rand haben zufällige Seiten, die keine Rolle spielen.) Das Feld
speichern wir als Dictionary `(z, s) -> Liste der vier Seiten`, damit wir Seiten später ändern können.

Wie oft ist so ein Zufallspuzzle eindeutig lösbar? Wir erfinden von jeder Größe einige und zählen ihre Lösungen.

````python
import random

SEITEN = 'abcdABCD'


def teile_aus(m, n, feld):
    teile = []
    for z in range(m):
        for s in range(n):
            teile.append(''.join(feld[(z, s)]))
    return teile


def zufallsfeld(m, n, zufall):
    while True:
        feld = {}
        for z in range(m):
            for s in range(n):
                seiten = []
                for k in range(4):
                    seiten.append(zufall.choice(SEITEN))
                if s > 0:
                    seiten[0] = feld[(z, s - 1)][2].swapcase()
                if z > 0:
                    seiten[1] = feld[(z - 1, s)][3].swapcase()
                feld[(z, s)] = seiten
        if alle_verschieden(teile_aus(m, n, feld)):
            return feld


zufall = random.Random(1)
for m, n, wie_oft in [(2, 2, 100), (2, 3, 100), (3, 3, 100), (3, 4, 50), (4, 4, 10)]:
    eindeutig = 0
    meiste = 0
    for i in range(wie_oft):
        anzahl = loese(m, n, teile_aus(m, n, zufallsfeld(m, n, zufall)))['anzahl']
        if anzahl == 1:
            eindeutig += 1
        meiste = max(meiste, anzahl)
    print(f'{m} x {n}: {eindeutig:3} von {wie_oft} eindeutig lösbar, bis zu {meiste} Lösungen')
````

Schon bei 3 × 4 ist kaum noch ein Zufallspuzzle eindeutig lösbar, bei 4 × 4 haben wir keines gefunden (auch außerhalb des
Notebooks unter 100 Zufallspuzzles keines). Einfach so lange neue
Puzzles zu würfeln, bis eines eindeutig ist, wird also schnell aussichtslos. Die Hinweise schlagen vor, ein eindeutiges kleineres
Puzzle zu einem größeren zu erweitern. Wir machen es anders, mit **Bergsteigen**:

1. Wir beginnen mit einem zufälligen lösbaren Feld und zählen die Lösungen.
2. Solange es mehr als eine gibt: Wir ändern eine zufällige Seite eines zufälligen Teils (und die passende Seite des Nachbarn,
   falls es einen gibt). Unser Feld bleibt dabei eine Lösung. Sind danach alle Teile verschieden und hat das neue Puzzle **nicht
   mehr Lösungen** als das alte, behalten wir die Änderung.
3. Zum Schluss mischen wir die Teile und drehen jedes zufällig, damit man die Lösung nicht an der Reihenfolge ablesen kann.

Für Schritt 2 müssen wir nicht alle Lösungen des neuen Puzzles zählen: Sobald `loese` eine mehr gefunden hat als vorher, wissen wir,
dass die Änderung schlecht ist. Dafür gibt es `hoechstens`. Weil jedes Puzzle mindestens eine Lösung hat (unser Feld), endet
`erfinde` nur mit einem eindeutig lösbaren Puzzle: Das ist die Begründung, die die Hinweise verlangen. Dass es überhaupt endet, ist
nicht bewiesen, in unseren Versuchen hat es aber immer geklappt.

````python
def aendere_seite(m, n, feld, zufall):
    neu = {}
    for f in feld:
        neu[f] = list(feld[f])                     # Kopie, das alte Feld bleibt, wie es ist
    z = zufall.randrange(m)
    s = zufall.randrange(n)
    richtung = zufall.randrange(4)
    buchstabe = zufall.choice(SEITEN)
    neu[(z, s)][richtung] = buchstabe
    dz, ds = NACHBARN[richtung]
    if (z + dz, s + ds) in neu:                    # innere Kante: die Gegenseite ändert sich mit
        neu[(z + dz, s + ds)][(richtung + 2) % 4] = buchstabe.swapcase()
    return neu


def erfinde(m, n, zufall):
    feld = zufallsfeld(m, n, zufall)
    stand = loese(m, n, teile_aus(m, n, feld))
    anzahl = stand['anzahl']
    gelegt = stand['gelegt']
    aenderungen = 0
    while anzahl > 1:
        neu = aendere_seite(m, n, feld, zufall)
        teile = teile_aus(m, n, neu)
        if alle_verschieden(teile):
            aenderungen += 1
            stand = loese(m, n, teile, hoechstens=anzahl + 1)
            gelegt += stand['gelegt']
            if stand['anzahl'] <= anzahl:
                feld = neu
                anzahl = stand['anzahl']
    teile = teile_aus(m, n, feld)
    zufall.shuffle(teile)
    gemischt = []
    for t in teile:
        gemischt.append(drehe(t, zufall.randrange(4)))
    return gemischt, aenderungen, gelegt


def als_text(m, n, teile):
    return f'{m} {n}\n' + '\n'.join(teile) + '\n'


zufall = random.Random(2013)
for m, n in [(2, 2), (3, 3), (2, 5), (3, 4), (4, 3)]:
    teile, aenderungen, gelegt = erfinde(m, n, zufall)
    print(f'{m} x {n}: {aenderungen:3} Änderungen ausprobiert, {gelegt:7} Teile gelegt, Teile: {" ".join(teile)}')
````

Ein 4 × 4-Puzzle braucht oft einige Dutzend Änderungen und Millionen gelegte Teile (bei den Startwerten 1 bis 25 im
Mittel 43 Änderungen). Für das Notebook haben wir einen Startwert gewählt, bei dem es schnell geht.
Die Ausgabe hat das Eingabeformat aus Teil 1, und zur Probe lösen wir das neue Puzzle gleich wieder:

````python
zufall = random.Random(2)
teile, aenderungen, gelegt = erfinde(4, 4, zufall)
text = als_text(4, 4, teile)
print(text)
print(aenderungen, 'Änderungen ausprobiert,', gelegt, 'Teile gelegt')
print()
m, n, teile = puzzle_aus_text(text)
stand = loese(m, n, teile)
print(stand['anzahl'], 'Lösung:')
zeige(m, n, stand['loesungen'][0])
````

Unsere eigenen Puzzles `rechteck2x3`, `streifen1x6` und `rechteck3x5` sind auf dieselbe Weise entstanden.

**Grenzen.** Für 5 × 5 dauert schon das Zählen der Lösungen eines einzigen Puzzles sehr lange (bei Beispiel 4 sind es 41,5 Millionen
gelegte Teile), und das Bergsteigen braucht viele solche Zählungen. Mit unserem Python-Programm sind deshalb nur Puzzles bis etwa
4 × 4 oder 3 × 5 in vernünftiger Zeit zu erfinden.

### Dynamische Reihenfolge

Die Hinweise finden **dynamische Reihenfolgen** interessanter als feste: Nach jedem gelegten Teil wird das nächste Feld neu
gewählt, und zwar das, auf das am wenigsten Lagen passen. Findet sich dabei ein Feld, auf das nichts mehr passt, ist die Suche
an dieser Stelle sofort zu Ende. Dafür muss die Suche bei jedem Schritt alle leeren Felder mit belegten Nachbarn durchsehen.

````python
def lege_dynamisch(m, n, stand):
    feld = stand['feld']
    if len(feld) == m * n:
        stand['loesungen'].append(dict(feld))
        return
    bestes = None                                   # (Feld, passende Lagen) mit den wenigsten Lagen
    for z in range(m):
        for s in range(n):
            w = wunsch(feld, z, s)
            if (z, s) not in feld and w != ('?', '?', '?', '?'):
                passende = []
                if w in stand['tabelle']:
                    for nr, t in stand['tabelle'][w]:
                        if stand['frei'][nr]:
                            passende.append((nr, t))
                if bestes is None or len(passende) < len(bestes[1]):
                    bestes = ((z, s), passende)
    (z, s), passende = bestes
    for nr, t in passende:
        stand['gelegt'] += 1
        feld[(z, s)] = (nr, t)
        stand['frei'][nr] = False
        lege_dynamisch(m, n, stand)
        stand['frei'][nr] = True
        del feld[(z, s)]


def loese_dynamisch(m, n, teile):
    """Wie loese, aber mit dynamischer Reihenfolge (ohne den Sonderfall ohne Anker)."""
    anker = ankerteil(teile)
    if m == n:
        ankerlagen = [teile[anker]]
    else:
        ankerlagen = [teile[anker], drehe(teile[anker], 1)]
    stand = neuer_stand(teile, 10 ** 9)
    for start in zeilenweise(m, n):
        for t in ankerlagen:
            stand['gelegt'] += 1
            stand['feld'][start] = (anker, t)
            stand['frei'][anker] = False
            lege_dynamisch(m, n, stand)
            stand['frei'][anker] = True
            del stand['feld'][start]
    return stand


for nummer in [1, 2, 3]:
    m, n, teile = lies_puzzle(f'beispieldaten/beispiel{nummer}.txt')
    a = loese(m, n, teile)['gelegt']
    stand = loese_dynamisch(m, n, teile)
    print(f'Beispiel {nummer}: {len(stand["loesungen"])} Lösung, fest {a:6} Teile gelegt, dynamisch {stand["gelegt"]:6}')
````

Die dynamische Reihenfolge legt bei Beispiel 3 nur 35 092 statt 75 544 Teile, bei Beispiel 4 (außerhalb des Notebooks
gerechnet) 13,8 statt 41,5 Millionen. Jeder Schritt kostet aber mehr, weil alle Felder neu untersucht werden. In unserem Programm war
die dynamische Suche deshalb insgesamt langsamer als die feste Reihenfolge.

### Probe mit symmetrischen Teilen

Der Anker-Trick und der Sonderfall sind leicht falsch zu programmieren. Deshalb vergleichen wir `loese` mit der einfachen Suche aus
Schritt 2, die einfach alle Lösungen zählt (bei ihr kommt jede Lösung viermal bzw. zweimal vor). Damit es viele Lösungen und viele
symmetrische Teile wie `abab` gibt, nehmen wir zufällige Teile aus nur vier Buchstaben `aAbB`. Jedes dritte Puzzle besteht nur
aus symmetrischen Teilen, dort gibt es keinen Anker.

````python
zufall = random.Random(5)
unterschiede = 0
verglichen = 0
sonderfall = 0
mit_loesung = 0
for versuch in range(300):
    m, n = zufall.choice([(1, 2), (2, 2), (1, 3), (2, 3)])
    teile = []
    while len(teile) < m * n:
        if versuch % 3 == 0:                         # jedes dritte Puzzle nur aus symmetrischen Teilen wie abab
            x = zufall.choice('aAbB')
            y = zufall.choice('aAbB')
            t = x + y + x + y
        else:
            t = ''
            for k in range(4):
                t = t + zufall.choice('aAbB')
        if alle_verschieden(teile + [t]):
            teile.append(t)
    if m == n:
        faktor = 4
    else:
        faktor = 2
    einfach = len(suche_einfach(m, n, teile)['loesungen']) // faktor
    if loese(m, n, teile)['anzahl'] != einfach:
        print('Unterschied:', m, n, teile)
        unterschiede += 1
    verglichen += 1
    if ankerteil(teile) is None:
        sonderfall += 1
    if einfach > 0:
        mit_loesung += 1
print(verglichen, 'Puzzles verglichen,', mit_loesung, 'davon lösbar,', sonderfall, 'ohne Ankerteil,', unterschiede, 'Unterschiede')
````

Beide Verfahren zählen immer gleich viele Lösungen, auch im Sonderfall ohne Anker.

## Zum Schluss: Komplementär mit MiniZinc

Das Puzzle lässt sich gut als MiniZinc-Modell schreiben. Für jedes Feld gibt es drei Arten von Variablen: welches Teil dort liegt
(`teil`), in welcher Drehung (`dreh`) und welche vier Seiten es dann hat (`seite`). Die Buchstaben schreiben wir als Zahlen:
`a` bis `d` als 1 bis 4 und `A` bis `D` als −1 bis −4. Dann passen zwei Seiten genau dann, wenn ihre **Summe 0** ist.

Neu gegenüber den Arbeitsblättern ist `table`: `table([x1, …, x6], lage)` verlangt, dass die Werte der sechs Variablen zusammen eine
Zeile der Tabelle `lage` bilden. In `lage` stehen alle Lagen aller Teile, wie in unserer Kandidatentabelle aus Schritt 3. So
verbindet `table` Teil, Drehung und Seiten eines Feldes. Außerdem neu ist `->` („wenn …, dann …“): Liegt auf dem Feld der Anker,
dann nur in einer erlaubten Drehung. Mit `alldifferent` kommt jedes Teil genau einmal vor.

```minizinc
include "globals.mzn";

int: m;                                   % Zeilen
int: n;                                   % Spalten
int: L;                                   % Anzahl der Lagen aller Teile
array[1..L, 1..6] of int: lage;           % Teil, Drehung, links, oben, rechts, unten
int: anker;                               % das Ankerteil ...
set of int: ankerdrehungen;               % ... darf nur so gedreht werden

array[1..m, 1..n] of var 0..m*n-1: teil;  % welches Teil liegt auf dem Feld?
array[1..m, 1..n] of var 0..3: dreh;      % in welcher Drehung?
array[1..m, 1..n, 1..4] of var -4..4: seite;  % seine vier Seiten

constraint alldifferent(teil);
constraint forall(i in 1..m, j in 1..n)(
    table([teil[i, j], dreh[i, j], seite[i, j, 1], seite[i, j, 2], seite[i, j, 3], seite[i, j, 4]], lage));
constraint forall(i in 1..m, j in 1..n-1)(seite[i, j, 3] + seite[i, j+1, 1] = 0);   % rechts an links
constraint forall(i in 1..m-1, j in 1..n)(seite[i, j, 4] + seite[i+1, j, 2] = 0);   % unten an oben
constraint forall(i in 1..m, j in 1..n)(teil[i, j] = anker -> dreh[i, j] in ankerdrehungen);

solve satisfy;

output [show_int(3, teil[i, j]) ++ if j == n then "\n" else "" endif | i in 1..m, j in 1..n];
```

Die Datendatei schreibt `als_dzn`, mit Anker und erlaubten Drehungen wie in `loese`. (Den Sonderfall ohne Anker lassen wir hier
weg.)

````python
ZAHL = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'A': -1, 'B': -2, 'C': -3, 'D': -4}


def als_dzn(m, n, teile):
    zeilen = []
    for nr in range(len(teile)):
        gesehen = []
        for k in range(4):
            t = drehe(teile[nr], k)
            if t not in gesehen:                   # jede Lage nur einmal (wie in lagen)
                gesehen.append(t)
                werte = [nr, k, ZAHL[t[0]], ZAHL[t[1]], ZAHL[t[2]], ZAHL[t[3]]]
                zeilen.append('| ' + ', '.join(str(w) for w in werte))
    anker = ankerteil(teile)
    if m == n:
        erlaubt = '{0}'
    else:
        erlaubt = '{0, 1}'
    return (f'm = {m};\nn = {n};\nL = {len(zeilen)};\n'
            f'lage = [' + '\n        '.join(zeilen) + ' |];\n'
            f'anker = {anker};\nankerdrehungen = {erlaubt};\n')


m, n, teile = lies_puzzle('beispieldaten/beispiel1.txt')
print(als_dzn(m, n, teile))
````
