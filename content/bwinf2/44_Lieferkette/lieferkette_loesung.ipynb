## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Graph:** Werke, S und T sind **Knoten**, die Straßen **Kanten** mit einer Fahrzeit als Gewicht. Die Straßen gelten in beiden
  Richtungen, also trägt man jede Straße bei beiden Knoten ein (*Adjazenzliste*).
- **Kürzeste Wege mit Dijkstra:** Der Algorithmus von Dijkstra findet von einem Startknoten aus die kürzesten Wege zu allen anderen
  Knoten. Er nimmt immer den Knoten mit der kleinsten bisher bekannten Entfernung als Nächsten. Das geht, weil es keine negativen
  Fahrzeiten gibt.
- **Prioritätswarteschlange:** Damit man schnell an den Knoten mit der kleinsten Entfernung kommt, benutzt man den Heap aus dem
  Modul `heapq`. `heappush(liste, (entfernung, knoten))` legt ein Tupel ab, `heappop(liste)` holt das kleinste heraus.
- **Dijkstra mit mehreren Startknoten:** Jeder Startknoten bekommt einen **Startwert** (nicht unbedingt 0). Dann ist die Entfernung
  eines Knotens die kleinste Summe „Startwert + Weg“ über alle Startknoten.
- **Dynamische Programmierung (DP):** Man zerlegt das Problem in kleinere Probleme, löst sie nacheinander und merkt sich die
  Ergebnisse. Hier: „Wie lange dauert der Rest der Produktion, wenn ich in Werk u einen Schritt erledigt habe?“ Man rechnet **von
  hinten nach vorn**, vom letzten Arbeitsschritt bis zum ersten.
- **Im schlimmsten Fall planen:** Für jeden möglichen Streik gibt es einen eigenen Verlauf. Die Dauer des Plans ist das **Maximum**
  über alle diese Verläufe. Gesucht ist der Plan, bei dem dieses Maximum möglichst klein ist.
- **Wege rekonstruieren:** Zu jedem Knoten merkt man sich seinen Vorgänger auf dem kürzesten Weg und läuft ihn rückwärts ab.

**Nützliche Funktionen und Bibliotheken**
- `heapq.heappush`, `heapq.heappop`
- `float('inf')` als „unendlich“ (größer als jede Zahl)
- Listen von Listen, Tupel, `dict`

## Die Idee

1. Die Karte einlesen. **S** bekommt die Nummer 0 und **T** die Nummer `w + 1`. Wir tun so, als wären S und T Werke, die den Schritt 0
   und den Schritt `m + 1` erledigen. Sie streiken nie.
2. Rückwärts über die Arbeitsschritte `i = m, m − 1, …, 0` für jedes Werk `u` im Schritt `i` zwei Zahlen berechnen:
   - `ohne[u]`: die kürzeste Dauer **ohne Streik** von hier bis T, wenn `u` seinen Schritt erledigt hat.
   - `mit[u]`: die kürzeste Dauer **im schlimmsten Fall** von hier bis T, wenn noch ein Streik auftreten kann.
3. Ist `mit[S]` höchstens so groß wie die Zeitfrist, ist die Lieferung **möglich**.
4. Aus den gemerkten Entscheidungen den **normalen Plan** und die **Alternativrouten** zusammensetzen und ausgeben.

## Schritt 1: Die Karte einlesen

Wir nummerieren die Knoten: **S = 0**, die Werke 1 bis `w`, **T = `w + 1`**. In `schritt_von[k]` steht der Arbeitsschritt von
Knoten `k` (für S ist es 0, für T ist es `m + 1`). `stufen[i]` ist die Liste aller Knoten, die den Schritt `i` erledigen können.

Die Straßen speichern wir in `graph`: `graph[k]` ist die Liste der Paare `(nachbar, fahrzeit)`. Weil man jede Straße in beide Richtungen
befahren darf, tragen wir sie bei beiden Knoten ein. Außerdem merken wir uns die Fahrzeiten in einem `dict` `zeit`, in dem man
`zeit[(a, b)]` nachschlagen kann. Das brauchen wir später zum Nachrechnen der Routen.

````python
def lies_lieferung(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = []
        for zeile in f.read().splitlines():
            if zeile.split():                              # leere Zeilen überspringen
                zeilen.append(zeile.split())
    frist = int(zeilen[0][0])
    m = int(zeilen[1][0])
    w = int(zeilen[3][0])
    schritt_von = [0] * (w + 2)
    schritt_von[w + 1] = m + 1                             # T erledigt den Schritt m + 1
    for i in range(w):
        schritt_von[int(zeilen[4 + i][0])] = int(zeilen[4 + i][1])
    stufen = []
    for i in range(m + 2):
        stufen.append([])
    for k in range(w + 2):
        stufen[schritt_von[k]].append(k)
    anzahl_strassen = int(zeilen[4 + w][0])
    graph = []
    for k in range(w + 2):
        graph.append([])
    zeit = {}
    for i in range(anzahl_strassen):
        a, b, z = zeilen[5 + w + i]
        a = knoten_nummer(a, w)
        b = knoten_nummer(b, w)
        z = int(z)
        graph[a].append((b, z))
        graph[b].append((a, z))
        zeit[(a, b)] = z
        zeit[(b, a)] = z
    return {'frist': frist, 'm': m, 'w': w, 'schritt_von': schritt_von, 'stufen': stufen, 'graph': graph, 'zeit': zeit}


def knoten_nummer(text, w):
    if text == 'S':
        return 0
    if text == 'T':
        return w + 1
    return int(text)


def knoten_name(k, w):
    if k == 0:
        return 'S'
    if k == w + 1:
        return 'T'
    return str(k)


karte = lies_lieferung('beispieldaten/lieferung00.txt')
print('Frist', karte['frist'], ', Schritte', karte['m'], ', Werke', karte['w'])
print('Schritt von jedem Knoten (S = 0, Werke 1 bis 5, T = 6):', karte['schritt_von'])
print('Werke je Schritt:', karte['stufen'])
for k in range(karte['w'] + 2):
    print('Knoten', knoten_name(k, karte['w']), ': Nachbarn und Fahrzeiten', karte['graph'][k])
````

## Schritt 2: Kürzeste Wege mit Dijkstra

`dijkstra(graph, starts)` bekommt ein `dict` `starts` mit den Startknoten und ihren Startwerten. Es gibt für jeden Knoten drei
Dinge zurück:

- `dist[k]`: die kleinste Entfernung (Startwert plus Weg), `float('inf')`, wenn der Knoten nicht erreichbar ist,
- `ursprung[k]`: von welchem Startknoten aus diese Entfernung erreicht wird,
- `vor[k]`: der Knoten davor auf diesem Weg (−1 bei einem Startknoten, der keinen besseren Weg von einem anderen Startknoten hat).

Ablauf: Alle Startknoten kommen mit ihrem Startwert in die Warteschlange. Dann wiederholen wir: Den Knoten `k` mit der kleinsten
Entfernung `d` aus der Warteschlange nehmen. Ist `d` größer als `dist[k]`, war das ein veralteter Eintrag, und wir überspringen ihn.
Sonst schauen wir uns alle Nachbarn an: Ist der Weg über `k` kürzer als die bisher bekannte Entfernung des Nachbarn, aktualisieren wir
sie, merken `k` als Vorgänger und legen den Nachbarn mit der neuen Entfernung in die Warteschlange.

Mit **einem** Startknoten mit Startwert 0 ist das der gewöhnliche Dijkstra. Hier zum Ausprobieren mit S in `lieferung00.txt`.

````python
from heapq import heappush, heappop

UNENDLICH = float('inf')


def dijkstra(graph, starts):
    n = len(graph)
    dist = [UNENDLICH] * n
    ursprung = [-1] * n
    vor = [-1] * n
    warteschlange = []
    for k in starts:
        if starts[k] < dist[k]:
            dist[k] = starts[k]
            ursprung[k] = k
            heappush(warteschlange, (starts[k], k))
    while len(warteschlange) > 0:
        d, k = heappop(warteschlange)
        if d > dist[k]:
            continue                                       # veralteter Eintrag
        for nachbar, z in graph[k]:
            if d + z < dist[nachbar]:
                dist[nachbar] = d + z
                ursprung[nachbar] = ursprung[k]
                vor[nachbar] = k
                heappush(warteschlange, (d + z, nachbar))
    return dist, ursprung, vor


def pfad(vor, k):
    """Der Weg von k zurück bis zu einem Startknoten, als Liste (beginnt bei k)."""
    weg = [k]
    while vor[k] != -1:
        k = vor[k]
        weg.append(k)
    return weg


dist, ursprung, vor = dijkstra(karte['graph'], {0: 0})
print('Entfernungen von S:', dist)
print('Weg von S nach T (rückwärts gelesen):', pfad(vor, karte['w'] + 1))
````

Die kürzeste Fahrt von S nach T dauert 9 und führt über die Werke 3 und 5 (`S → 3 → 5 → T`, also 3 + 5 + 1). `pfad` gibt den Weg
rückwärts von T zu S an, denn man folgt den Vorgängern.

## Schritt 3: Ohne Streik, rückwärts über die Arbeitsschritte

Wir betrachten zuerst den Fall ohne Streik. Sei `u` ein Werk, das den Schritt `i` erledigt. `ohne[u]` ist die kürzeste Dauer **ab `u`
bis T**, wenn der Schritt `i` in `u` erledigt ist.

Nach `u` muss irgendein Werk `v` den Schritt `i + 1` erledigen. Die Dauer ist dann die Fahrzeit von `u` nach `v` plus `ohne[v]`, und
die beiden Teile hängen nicht voneinander ab. Wir minimieren beide einzeln:

`ohne[u] = min über alle Werke v des nächsten Schritts von (entfernung(u, v) + ohne[v])`

Für T ist `ohne[T] = 0`, denn dort ist man fertig. Man berechnet die Werte also von hinten: erst für den Schritt `m`, dann `m − 1` und
so weiter bis zum Schritt 0, also bis S.

Genau das ist ein **Dijkstra mit mehreren Startknoten**: Die Werke des nächsten Schritts sind die Startknoten mit dem Startwert
`ohne[v]`. Die berechnete Entfernung eines Knotens `u` ist dann das Minimum über alle `v`, und `ursprung[u]` ist das beste `v`. So
müssen wir nicht für jedes Werk einen eigenen Dijkstra starten, sondern nur **einen pro Arbeitsschritt**.

Das Ergebnis speichern wir:

- `naechst_ohne[u]`: das beste Werk `v` für den nächsten Schritt,
- `weg_ohne[u]`: der kürzeste Weg von `u` zu diesem `v` (als Liste der Knoten, von `u` bis `v`).

````python
def rechne_ohne_streik(karte):
    n = len(karte['graph'])
    ohne = [UNENDLICH] * n
    naechst_ohne = [None] * n
    weg_ohne = [None] * n
    ohne[karte['w'] + 1] = 0                               # in T ist man fertig
    for i in range(karte['m'], -1, -1):                    # rückwärts: Schritt m, m-1, ..., 0
        starts = {}
        for v in karte['stufen'][i + 1]:
            starts[v] = ohne[v]
        dist, ursprung, vor = dijkstra(karte['graph'], starts)
        for u in karte['stufen'][i]:
            ohne[u] = dist[u]
            naechst_ohne[u] = ursprung[u]
            weg_ohne[u] = pfad(vor, u)
    return ohne, naechst_ohne, weg_ohne


ohne, naechst_ohne, weg_ohne = rechne_ohne_streik(karte)
for u in range(karte['w'] + 1):
    print('Werk', knoten_name(u, karte['w']), '(Schritt', karte['schritt_von'][u], '): ohne =', ohne[u],
          ', nächstes Werk', knoten_name(naechst_ohne[u], karte['w']), ', Weg', weg_ohne[u])
````

Zur Probe `lieferung00.txt`: Der Schritt 2 wird in Werk 4 oder 5 erledigt. Von Werk 4 sind es 3 Zeiteinheiten bis T, von Werk 5 nur 1.
Für das Werk 3 (Schritt 1) ist `ohne` gleich 6: Über Werk 5 (Fahrzeit 5) und weiter nach T (Fahrzeit 1) braucht man 6. Für S ist
`ohne` gleich 9, das ist die kürzeste Fahrt `S → 3 → 5 → T` vom Anfang.

## Schritt 4: Mit höchstens einem Streik

Jetzt kann in **einem** Werk ein Streik auftreten. Sei wieder `u` ein Werk im Schritt `i`. `mit[u]` ist die kürzeste Dauer **im
schlimmsten Fall** von `u` bis T, wenn `u` im Schritt `i` noch bestreikt werden kann. Es gibt zwei mögliche Verläufe, und der
schlimmste zählt, also das **Maximum**:

1. **Kein Streik in `u`.** Der Schritt gelingt, es geht zu einem Werk `v` des nächsten Schritts, und ab dort kann der Streik noch
   kommen: `x1 = min über v im Schritt i + 1 von (entfernung(u, v) + mit[v])`. Das ist wieder ein Mehrquellen-Dijkstra, diesmal mit
   den Startwerten `mit[v]`.
2. **Streik in `u`.** Der Schritt scheitert, die Fahrerin weicht auf ein **anderes** Werk `u'` desselben Schritts aus. Da höchstens ein
   Werk streikt, geht ab dort alles glatt, es gilt `ohne[u']`:
   `x2 = min über andere Werke u' im Schritt i von (entfernung(u, u') + ohne[u'])`.

`mit[u] = max(x1, x2)`.

**Warum ist das richtig?** Die beiden Verläufe sind unabhängig voneinander planbar: Wohin die Fahrerin ohne Streik weiterfährt, und
welches Werk sie bei einem Streik als Alternative anfährt, sind zwei getrennte Entscheidungen. Jede wählt man so, dass ihre Dauer
möglichst klein ist. Der Plan ist im schlimmsten Fall so schnell wie der langsamere der beiden Verläufe. Auch die Dauern ab `v` und
`u'` hängen nicht davon ab, wie man nach `u` gekommen ist. Daher ist die Zerlegung in kleinere Probleme erlaubt.

Für S (Schritt 0) gibt es keinen Streik, dort ist nur `x1` zu berechnen. Und für T ist `mit[T] = 0`. Hat ein Schritt nur ein einziges
Werk, gibt es dort keine Alternative: `x2` ist unendlich, und die Lieferung ist nicht planbar.

**Die Alternative `u'` mit dem zweiten Dijkstra:** Für `x2` suchen wir von jedem Werk `u` die kleinste Summe
`entfernung(u, u') + ohne[u']` über alle **anderen** Werke `u'` des gleichen Schritts. Wir starten einen Dijkstra bei **allen**
Werken des Schritts mit den Startwerten `ohne[u']`. Das Problem: Für `u` selbst wäre der beste Startknoten `u` (Entfernung 0), und
den dürfen wir nicht nehmen. Wir brauchen also zu jedem Knoten die beste Entfernung von einem Startknoten und die **zweitbeste von
einem anderen Startknoten**.

Dafür bekommt jeder Knoten bis zu **zwei Marken** `(ursprung, entfernung, vorgaenger)`, aber nie zwei vom selben Ursprung. In der
Warteschlange liegen Einträge `(entfernung, knoten, ursprung, vorgaenger)`. Wir nehmen den kleinsten Eintrag. Hat der Knoten schon
zwei Marken oder schon eine mit diesem Ursprung, werfen wir den Eintrag weg. Sonst wird er zur Marke, und wir legen die Nachbarn mit
derselben Herkunft in die Warteschlange.

````python
def dijkstra_zwei(graph, starts):
    """Zu jedem Knoten bis zu zwei Marken (ursprung, entfernung, vorgaenger) von verschiedenen Startknoten."""
    marken = []
    for k in range(len(graph)):
        marken.append([])
    warteschlange = []
    for k in starts:
        if starts[k] < UNENDLICH:
            heappush(warteschlange, (starts[k], k, k, -1))
    while len(warteschlange) > 0:
        d, k, ursprung, vorgaenger = heappop(warteschlange)
        if len(marken[k]) >= 2:
            continue                                       # schon zwei Marken
        schon_da = False
        for marke in marken[k]:
            if marke[0] == ursprung:
                schon_da = True
        if schon_da:
            continue                                       # von diesem Ursprung gibt es schon eine Marke
        marken[k].append((ursprung, d, vorgaenger))
        for nachbar, z in graph[k]:
            if len(marken[nachbar]) < 2:
                heappush(warteschlange, (d + z, nachbar, ursprung, k))
    return marken


def pfad_von_ursprung(marken, k, ursprung):
    """Der Weg von k zurück zum Startknoten ursprung, anhand der Marken dieses Ursprungs."""
    weg = [k]
    while True:
        for marke in marken[k]:
            if marke[0] == ursprung:
                vorgaenger = marke[2]
        if vorgaenger == -1:
            return weg
        k = vorgaenger
        weg.append(k)


def rechne_mit_streik(karte, ohne):
    n = len(karte['graph'])
    mit = [UNENDLICH] * n
    naechst_mit = [None] * n
    weg_mit = [None] * n
    alternative = [None] * n                               # alternative[u]: das Ausweichwerk bei einem Streik in u
    weg_alternative = [None] * n
    mit[karte['w'] + 1] = 0
    for i in range(karte['m'], -1, -1):
        starts = {}
        for v in karte['stufen'][i + 1]:
            starts[v] = mit[v]
        dist, ursprung, vor = dijkstra(karte['graph'], starts)                 # x1 für alle Werke
        marken = None
        if i >= 1:
            starts2 = {}
            for u in karte['stufen'][i]:
                starts2[u] = ohne[u]
            marken = dijkstra_zwei(karte['graph'], starts2)                    # für x2
        for u in karte['stufen'][i]:
            x1 = dist[u]
            naechst_mit[u] = ursprung[u]
            weg_mit[u] = pfad(vor, u)
            x2 = UNENDLICH
            if i >= 1:
                for ursp, d, vorgaenger in marken[u]:
                    if ursp != u and d < x2:               # nur Marken von einem anderen Werk
                        x2 = d
                        alternative[u] = ursp
                        weg_alternative[u] = pfad_von_ursprung(marken, u, ursp)
            if i == 0:
                mit[u] = x1                                # in S kann nicht gestreikt werden
            else:
                mit[u] = max(x1, x2)
    return mit, naechst_mit, weg_mit, alternative, weg_alternative


mit, naechst_mit, weg_mit, alternative, weg_alternative = rechne_mit_streik(karte, ohne)
for u in range(karte['w'] + 1):
    text = 'Werk ' + knoten_name(u, karte['w']) + ' (Schritt ' + str(karte['schritt_von'][u]) + '): ohne = ' + str(ohne[u]) + ', mit = ' + str(mit[u])
    if alternative[u] is not None:
        text = text + ', Ausweichwerk bei Streik: ' + str(alternative[u])
    print(text)
````

`mit[S]` ist 15, also genau die Zeitfrist von `lieferung00.txt`: Im schlimmsten Fall dauert es 15, es **geht also gerade noch**.
Ohne Streik ginge es in 9. Ein Beispiel für den Streikfall ist das Werk 5 (Schritt 2): Dort ist `ohne` gleich 1, aber `mit` gleich 7.
Wird Werk 5 bestreikt, fährt man über T zum Werk 4 (Fahrzeit 1 + 3 = 4), und von dort sind es noch 3 bis T. Das sind 4 + 3 = 7.

## Schritt 5: Ist es möglich?

Die kleinste Dauer im schlimmsten Fall ist `mit[S]`. Ist sie höchstens so groß wie die Zeitfrist, gibt es einen Plan, der die Frist
immer einhält. Wenn nicht, gibt es keinen. Denn kein anderer Plan hätte im schlimmsten Fall eine kleinere Dauer.

````python
def ist_moeglich(karte, mit):
    return mit[0] <= karte['frist']


print('lieferung00: Frist', karte['frist'], ', kleinste Dauer im schlimmsten Fall', mit[0], ', möglich:', ist_moeglich(karte, mit))
````

## Schritt 6: Die Pläne zusammensetzen

Aus den gemerkten Entscheidungen setzen wir die Routen zusammen. Eine Route ist eine Liste von Paaren `(knoten, schritt)`:
`schritt` ist 0, wenn in dem Knoten kein Arbeitsschritt erledigt wird (man fährt nur durch, oder es ist S oder T), sonst die Nummer
des Schritts.

**Die normale Route:** Wir beginnen bei S. Das erste Werk ist `naechst_mit[S]`, das zweite `naechst_mit` des ersten und so weiter bis
T. Zwischen zwei Werken schieben wir den kürzesten Weg `weg_mit` ein, der auch durch Werke ohne Arbeitsschritt führen kann. Warum
`mit` und nicht `ohne`? Mit `ohne` würde man vielleicht eine kürzere Route wählen, für die es im Streikfall keine gute Alternative gibt.

**Die Alternativroute für das Werk `u` im Schritt `i`:** Sie beginnt bei `u` (dort ist der Streik). Dann geht es auf dem Weg
`weg_alternative[u]` zum Ausweichwerk `u'`, das den Schritt `i` erledigt. Ab dort gibt es keinen Streik mehr: Von `u'` aus folgen wir
`naechst_ohne` und den Wegen `weg_ohne` bis T. Die **Gesamtdauer** ist die Zeit, die man auf der normalen Route bis `u` gebraucht
hat, plus die Dauer der Alternativroute.

Die Dauer einer Route rechnen wir nach, indem wir die Fahrzeiten ihrer Straßen aus dem `dict` `zeit` addieren. Das ist zugleich
eine Kontrolle, dass die Route nur über vorhandene Straßen führt.

````python
def haenge_weg_an(route, weg, schritt_am_ende):
    """Hängt den Weg (Liste von Knoten, der erste ist schon die letzte Stelle der Route) an. Der letzte Knoten erledigt den Schritt."""
    for k in weg[1:]:
        route.append((k, 0))
    if schritt_am_ende > 0:
        route[-1] = (route[-1][0], schritt_am_ende)


def dauer(karte, route):
    summe = 0
    for j in range(len(route) - 1):
        summe = summe + karte['zeit'][(route[j][0], route[j + 1][0])]
    return summe


def baue_plaene(karte, naechst_mit, weg_mit, alternative, weg_alternative, naechst_ohne, weg_ohne):
    """Ergebnis: die normale Route und eine Liste von Alternativrouten (mit dem bestreikten Werk und dem Schritt)."""
    schritt_von = karte['schritt_von']
    T = karte['w'] + 1
    normal = [(0, 0)]
    werke = [0]                                            # die Werke der Schritte 0, 1, ..., m
    u = 0
    while u != T:
        v = naechst_mit[u]
        haenge_weg_an(normal, weg_mit[u], schritt_von[v] if v != T else 0)
        werke.append(v)
        u = v
    alternativen = []
    for i in range(1, karte['m'] + 1):
        u = werke[i]
        # bis zum Streik in u: die Dauer der normalen Route bis dahin
        stelle = 0
        for j in range(len(normal)):
            if normal[j][0] == u and normal[j][1] == i:
                stelle = j
        vorlauf = dauer(karte, normal[:stelle + 1])
        route = [(u, 0)]                                   # der Streik: in u wird nichts erledigt
        u2 = alternative[u]
        haenge_weg_an(route, weg_alternative[u], i)
        while u2 != T:
            v = naechst_ohne[u2]
            haenge_weg_an(route, weg_ohne[u2], schritt_von[v] if v != T else 0)
            u2 = v
        alternativen.append((i, vorlauf, route))
    return normal, alternativen


karte = lies_lieferung('beispieldaten/lieferung00.txt')
ohne, naechst_ohne, weg_ohne = rechne_ohne_streik(karte)
mit, naechst_mit, weg_mit, alternative, weg_alternative = rechne_mit_streik(karte, ohne)
normal, alternativen = baue_plaene(karte, naechst_mit, weg_mit, alternative, weg_alternative, naechst_ohne, weg_ohne)
print('normale Route:', normal, ', Dauer', dauer(karte, normal))
for i, vorlauf, route in alternativen:
    print('Streik im Schritt', i, ':', route, ', Dauer von S an', vorlauf + dauer(karte, route))
````

Zwei Dinge sind bei der Alternativroute wichtig: Das Werk mit dem Streik steht ohne Schritt am Anfang (im Beispiel `(3, 0)`), denn dort wird nichts
erledigt. Und bei der Gesamtdauer kommt der **Vorlauf** dazu, die Fahrzeit von S bis zum bestreikten Werk.

## Schritt 7: Die Ausgabe

Jetzt noch die Routen im vorgeschriebenen Format: Ein Werk, in dem ein Arbeitsschritt stattfindet, wird als `[werk schritt]`
geschrieben, alle anderen nur als Nummer, S und T als Buchstaben.

````python
def route_text(karte, route):
    teile = []
    for knoten, schritt in route:
        name = knoten_name(knoten, karte['w'])
        if schritt > 0:
            name = '[' + name + ' ' + str(schritt) + ']'
        teile.append(name)
    return ' '.join(teile)


def ausgabe_text(karte, mit, normal, alternativen):
    if not ist_moeglich(karte, mit):
        return 'UNMOEGLICH\n' + str(mit[0]) + '\n'
    zeilen = ['MOEGLICH', str(mit[0]), '', str(dauer(karte, normal)), route_text(karte, normal)]
    for i, vorlauf, route in alternativen:
        zeilen.append(str(vorlauf + dauer(karte, route)))
        zeilen.append(route_text(karte, route))
    return '\n'.join(zeilen) + '\n'


print(ausgabe_text(karte, mit, normal, alternativen))
````

## Schritt 8: Die Pläne selbst prüfen

Wir verlassen uns nicht auf die Rechnung, sondern prüfen die fertigen Pläne **unabhängig** davon. Eine Prüffunktion untersucht:

- Die normale Route beginnt bei S und endet bei T, und je zwei aufeinanderfolgende Knoten sind durch eine Straße verbunden.
- Die Arbeitsschritte kommen der Reihe nach vor (1, 2, …, m), und das Werk kann den Schritt wirklich erledigen.
- Jede Alternativroute beginnt in einem Werk der normalen Route, in dem ein Schritt `i` erledigt wird. Sie erledigt die Schritte
  `i, i + 1, …, m` der Reihe nach, und den Schritt `i` in einem **anderen** Werk.
- Die Gesamtdauer jeder Route, auch der Alternativen mit dem Vorlauf, ist höchstens die angegebene maximale Dauer.
- Die größte Dauer aller Routen ist genau diese maximale Dauer.

````python
def schritte_der_route(route):
    ergebnis = []
    for knoten, schritt in route:
        if schritt > 0:
            ergebnis.append((knoten, schritt))
    return ergebnis


def vorlauf_der_normalen_route(karte, normal, werk, schritt):
    """Dauer der normalen Route von S bis zum Werk, in dem der Schritt erledigt wird."""
    for j in range(len(normal)):
        if normal[j] == (werk, schritt):
            return dauer(karte, normal[:j + 1])


def pruefe_plan(karte, mit, normal, alternativen):
    T = karte['w'] + 1
    m = karte['m']
    # normale Route
    if normal[0][0] != 0 or normal[-1][0] != T:
        return False
    schritte = schritte_der_route(normal)
    if len(schritte) != m:
        return False
    for i in range(m):
        if schritte[i][1] != i + 1 or karte['schritt_von'][schritte[i][0]] != i + 1:
            return False
    groesste = dauer(karte, normal)                         # die Dauer prüft auch, ob es die Straßen gibt (KeyError sonst)
    if len(alternativen) != m:
        return False
    # Alternativrouten
    for i, vorlauf, route in alternativen:
        gestreikt = schritte[i - 1][0]
        if route[0][0] != gestreikt or route[-1][0] != T:
            return False
        weiter = schritte_der_route(route)
        if len(weiter) != m - i + 1:
            return False
        for j in range(len(weiter)):
            if weiter[j][1] != i + j or karte['schritt_von'][weiter[j][0]] != i + j:
                return False
        if weiter[0][0] == gestreikt:
            return False                                    # das Ausweichwerk muss ein anderes Werk sein
        gesamt = vorlauf_der_normalen_route(karte, normal, gestreikt, i) + dauer(karte, route)
        if gesamt != vorlauf + dauer(karte, route):
            return False
        groesste = max(groesste, gesamt)
    return groesste == mit[0]


print('lieferung00:', pruefe_plan(karte, mit, normal, alternativen))
````

````python
def loese(dateiname):
    karte = lies_lieferung(dateiname)
    ohne, naechst_ohne, weg_ohne = rechne_ohne_streik(karte)
    mit, naechst_mit, weg_mit, alternative, weg_alternative = rechne_mit_streik(karte, ohne)
    if mit[0] == UNENDLICH or not ist_moeglich(karte, mit):
        return karte, mit, None, None
    normal, alternativen = baue_plaene(karte, naechst_mit, weg_mit, alternative, weg_alternative, naechst_ohne, weg_ohne)
    return karte, mit, normal, alternativen


ergebnisse = {}
for nr in range(11):
    dateiname = 'beispieldaten/lieferung' + str(nr).rjust(2, '0') + '.txt'
    karte, mit, normal, alternativen = loese(dateiname)
    ergebnisse[nr] = (karte, mit, normal, alternativen)
    if normal is None:
        print('lieferung' + str(nr).rjust(2, '0') + ': UNMOEGLICH, kleinste Dauer im schlimmsten Fall', mit[0], ', Frist', karte['frist'])
    else:
        print('lieferung' + str(nr).rjust(2, '0') + ': MOEGLICH, höchstens', mit[0], ', Frist', karte['frist'],
              ', Pläne in Ordnung:', pruefe_plan(karte, mit, normal, alternativen))
````

## Für Neugierige

**Warum nicht alle Entfernungen im Voraus?** Man könnte mit dem Algorithmus von Dijkstra oder dem von **Floyd und Warshall** die
Entfernungen zwischen allen Werken ausrechnen und dann die DP-Formeln direkt benutzen. Bei `lieferung10.txt` mit 3932 Werken wären
das fast 16 Millionen Entfernungen, und Floyd-Warshall bräuchte etwa 60 Milliarden Schritte. Mit einem Dijkstra je Arbeitsschritt
brauchen wir nur 3 · m Läufe (m = 200).

**Laufzeit:** Jeder Dijkstra kostet etwa `(V + E) · log V`, und wir brauchen ungefähr drei pro Arbeitsschritt, insgesamt also
`O(m · (V + E) · log V)`. Das Zusammensetzen der Routen braucht `O(m² · V)`, denn jede der `m` Alternativrouten kann `m` Werke
durchlaufen. Der Dijkstra für `x2` mit zwei Marken pro Knoten ist deutlich teurer als die anderen. Er braucht bei den großen
Beispielen den größten Teil der Zeit.

**Speicher:** Neben dem Graphen braucht das Verfahren pro Knoten nur wenige Werte. Die Wege speichern wir pro Werk nur einmal, nicht pro
Alternativroute. Nur die Ausgabe selbst kann lang werden: Sie enthält `m` Alternativrouten mit jeweils bis zu `m` Werken.

**Warum ein einziger normaler Plan?** Die Fahrerin bekommt genau eine normale Route und für jedes Werk darauf eine Alternative. Man
muss nicht bei jedem Streik alle Wege neu berechnen, denn die Alternativrouten ergeben sich direkt aus den DP-Werten `ohne`.

**Mehr als ein Streik:** Würden mehrere Werke streiken können, genügt `ohne` nicht mehr: Man bräuchte für jede mögliche Zahl von
Streiks einen eigenen DP-Wert. Das Verfahren wächst entsprechend.

**MiniZinc:** Für diese Aufgabe haben wir kein Modell versucht. Kürzeste Wege und DP sind Aufgaben für Graphalgorithmen. Ein Solver
müsste dafür erst mühsam die Wege als Variablen beschreiben.
