## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zwangsreihenfolge statt Ausprobieren:** Wer alle Reihenfolgen der Scheiben durchprobiert, braucht `n!` Versuche. Hier ist die
  Reihenfolge aber (fast) erzwungen. Wenn man diese Zwangsregel erkennt, bleibt für jede Startscheibe genau **ein** Aufbau übrig.
- **Eindeutige Vertreter:** Gleich große Scheiben sind austauschbar. Es genügt, jede **Sorte** von Scheiben einmal als Start zu
  probieren, nicht jede einzelne Scheibe.
- **Filter vor dem Probieren:** Ein billiger Test (genug Exemplare der Startscheibe?) spart viele Aufbauversuche.
- **Backtracking:** Muss man zwischen mehreren Möglichkeiten wählen (fehlende Scheiben, zwei Quader), probiert man eine aus,
  macht sie bei Misserfolg **rückgängig** und probiert die nächste.

**Nützliche Funktionen und Bibliotheken**
- `collections.Counter`: zählt, wie oft jede Scheibengröße vorkommt. Für eine Größe, die es nicht gibt, liefert `vorrat[(3, 4)]`
  einfach `0` (ein normales `dict` würde einen Fehler melden).
- Tupel als Schlüssel: Eine Scheibe ist das Paar `(kurze Seite, lange Seite)`, zum Beispiel `(2, 4)`.
- `list.sort()`, `min`, `sum` für die Seitenlängen des Käses.

## Die Idee

**Von innen nach außen.** Wir legen eine Scheibe `(a, b)` hin. Das ist schon ein Quader mit den Maßen `1 × a × b`. Eine weitere Scheibe
passt an den Quader, wenn **zwei** seiner drei Seiten genau die beiden Seitenlängen der Scheibe sind. Die dritte Seite des Quaders
wird dadurch einen Millimeter länger. Sind alle Scheiben angelegt, ist der Käse wieder zusammengesetzt.

**Die Reihenfolge ist erzwungen.** Der Quader habe die sortierten Seiten `x ≤ y ≤ z`. Es können nur Scheiben `(x, y)`, `(x, z)`
oder `(y, z)` angelegt werden.

- Gibt es eine Scheibe `(x, y)`, **muss** sie jetzt angelegt werden. Legt man stattdessen eine andere an, wird `x` oder `y` einen
  Millimeter länger, und die Scheibe `(x, y)` passt nie wieder an den Käse: `x` und `y` sind die zwei kleinsten Seiten und
  können nur wachsen.
- Gibt es sie nicht, muss es die Scheibe `(x, z)` sein (sonst würde `x` wachsen, und `(x, z)` passte nie mehr).
- Gibt es auch die nicht, bleibt nur `(y, z)`.
- Passt keine der drei, ist der Aufbau von dieser Startscheibe aus gescheitert.

Also gibt es zu jeder Startscheibe **höchstens einen** Aufbau. Wir probieren alle Startscheiben (jede Größe einmal) und nehmen die
erste, bei der alle Scheiben angelegt werden. Das sind höchstens `n*` Versuche mit je `n` Schritten (`n*` = Zahl verschiedener
Scheibengrößen), also etwa `n · n*` Schritte statt `n!`.

**Ein Filter für die Startscheibe.** Liegt nur die Startscheibe `(a, b)` im Käse und ist `m` die kürzeste Seite aller Scheiben, dann
hat der Käse die Maße `t × a × b` mit `t < m`. Keine Scheibe hat eine Seite der Länge `t`, also kann nur `(a, b)` selbst angelegt
werden, bis die Dicke `m` erreicht ist. Eine Startscheibe mit weniger als `m` Exemplaren scheidet sofort aus (ist die Gesamtzahl
`n` kleiner als `m`, genügt `n`).

**Teil b.** *Fehlende Scheiben:* Findet man keine passende Scheibe, darf man eine **fiktive** Scheibe erfinden (höchstens so viele,
wie fehlen dürfen). *Zwei Quader:* Man baut den ersten Quader in beliebiger Reihenfolge auf und prüft nach jedem Schritt mit dem
Verfahren aus Teil a, ob die übrigen Scheiben einen Quader ergeben.

## Schritt 1: Scheiben einlesen und zählen

Eine Scheibe speichern wir als Paar `(kurze Seite, lange Seite)`. Ob Antje die Scheibe gedreht hat, spielt keine Rolle. Ein
`Counter` zählt, wie viele Scheiben jeder Größe es gibt, das ist unser **Vorrat**.

````python
from collections import Counter


def lies_scheiben(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = f.read().split('\n')
    scheiben = []
    for zeile in zeilen[1:]:                     # Zeile 0 enthält nur die Anzahl
        if zeile.strip() != '':
            a, b = zeile.split()
            a, b = int(a), int(b)
            scheiben.append((min(a, b), max(a, b)))
    return scheiben


def name(nr):
    return 'beispieldaten/kaese' + str(nr) + '.txt'


daten = {}
for nr in range(1, 7):
    daten[nr] = lies_scheiben(name(nr))
    vorrat_nr = Counter(daten[nr])
    print('kaese%d: %5d Scheiben, %5d verschiedene Größen, kürzeste Seite %d, längste Seite %d' % (
        nr, len(daten[nr]), len(vorrat_nr), min(a for a, b in daten[nr]), max(b for a, b in daten[nr])))
````

## Schritt 2: Welche Scheibe kommt als Nächstes?

`masse` ist die Liste `[x, y, z]` der Seiten des Käses, **sortiert**. Die Funktion prüft in der erzwungenen Reihenfolge, welche
Scheibe im Vorrat noch da ist, und liefert ihre Seiten `p, q` sowie den Index `i` der Käseseite, die dadurch wächst.

````python
def naechste_scheibe(masse, vorrat):
    x, y, z = masse
    # (Seiten der Scheibe, Index der Käseseite, die um 1 wächst), in der erzwungenen Reihenfolge
    for p, q, i in [(x, y, 2), (x, z, 1), (y, z, 0)]:
        if vorrat[(p, q)] > 0:
            return p, q, i
    return None


kleiner_vorrat = Counter([(2, 4), (3, 4), (4, 6)])
print(naechste_scheibe([1, 2, 4], kleiner_vorrat))     # Käse 1 x 2 x 4: Scheibe (2,4) passt, die Dicke 1 wächst
print(naechste_scheibe([2, 3, 4], kleiner_vorrat))     # (2,3) fehlt, (2,4) wäre die Wahl: die Seite 3 wächst
print(naechste_scheibe([3, 5, 7], kleiner_vorrat))     # nichts passt
````

## Schritt 3: Einen Käse von einer Startscheibe aus aufbauen

Wir legen die Startscheibe an, dann so lange `naechste_scheibe`, bis nichts mehr passt. Danach stellen wir den Vorrat wieder her.
Waren am Ende **alle** Scheiben angelegt, ist der Käse vollständig.

````python
def baue_kaese(start, vorrat, anzahl):
    """Baut den Käse von innen nach außen. Ergebnis: (Maße, Reihenfolge) oder None. Der Vorrat ist danach unverändert."""
    vorrat[start] -= 1
    masse = [1, start[0], start[1]]
    reihenfolge = [start]
    while True:
        schritt = naechste_scheibe(masse, vorrat)
        if schritt is None:
            break
        p, q, i = schritt
        vorrat[(p, q)] -= 1
        masse[i] += 1
        masse.sort()
        reihenfolge.append((p, q))
    for scheibe in reihenfolge:                  # Vorrat wieder auffüllen
        vorrat[scheibe] += 1
    if len(reihenfolge) == anzahl:
        return masse, reihenfolge
    return None


vorrat1 = Counter(daten[1])
for start in vorrat1:
    erg = baue_kaese(start, vorrat1, len(daten[1]))
    print('Start', start, '->', 'gescheitert' if erg is None else 'Käse %s' % erg[0])
````

## Schritt 4: Alle Startscheiben ausprobieren (Teil a)

`loese_a` probiert jede Größe einmal als Start, aber nur, wenn der Filter sie nicht ausschließt. Neben dem Ergebnis geben wir
zurück, wie viele Startscheiben wirklich aufgebaut wurden.

````python
def loese_vorrat(vorrat):
    """Teil a für einen Vorrat (Counter). Ergebnis: (Maße, Reihenfolge, Zahl der probierten Starts) oder (None, None, Zahl)."""
    anzahl = sum(vorrat.values())
    if anzahl == 0:
        return None, None, 0
    kuerzeste = min(a for (a, b), k in vorrat.items() if k > 0)
    noetig = min(kuerzeste, anzahl)              # Filter: so viele Exemplare braucht die Startscheibe
    probiert = 0
    for start in list(vorrat):
        if vorrat[start] < noetig:
            continue
        probiert += 1
        erg = baue_kaese(start, vorrat, anzahl)
        if erg is not None:
            return erg[0], erg[1], probiert
    return None, None, probiert


def loese_a(scheiben):
    return loese_vorrat(Counter(scheiben))
````

Zur **Kontrolle** schreiben wir eine zweite, unabhängige Funktion, die eine Reihenfolge nachspielt: Bei jeder Scheibe müssen zwei
Seiten des Käses genau die Seiten der Scheibe sein. Am Ende muss das Volumen des Käses die Summe der Scheibenflächen sein.

````python
def pruefe_aufbau(reihenfolge):
    """Spielt die Reihenfolge nach. Ergebnis: Maße des Käses (sortiert), sonst ein Fehler (assert)."""
    a, b = reihenfolge[0][:2]
    masse = [0, a, b]                            # vor der ersten Scheibe ist der Käse nur eine Fläche
    for scheibe in reihenfolge:
        p, q = scheibe[:2]
        for i in range(3):
            uebrige = sorted(masse[j] for j in range(3) if j != i)
            if uebrige == [p, q]:
                masse[i] += 1
                break
        else:
            assert False, 'Scheibe %s passt nicht an den Käse %s' % (scheibe, masse)
        masse.sort()
    assert masse[0] * masse[1] * masse[2] == sum(s[0] * s[1] for s in reihenfolge)
    return masse


def zeige(scheibe):
    return '%d×%d' % scheibe[:2]


def beschreibe(reihenfolge, anfang=4, ende=4):
    erste = ', '.join(zeige(s) for s in reihenfolge[:anfang])
    letzte = ', '.join(zeige(s) for s in reihenfolge[-ende:])
    return erste + ', ..., ' + letzte
````

### Alle Beispiele

Für jedes Beispiel: Zahl der Scheiben, Zahl der probierten Startscheiben, Maße des Käses (kürzeste Seite zuerst) und die ersten
und letzten vier angelegten Scheiben (von innen nach außen).

````python
ergebnis_a = {}
for nr in range(1, 7):
    masse, reihenfolge, probiert = loese_a(daten[nr])
    assert pruefe_aufbau(reihenfolge) == masse
    assert sorted(reihenfolge) == sorted(daten[nr])          # genau die gegebenen Scheiben, jede einmal
    ergebnis_a[nr] = (masse, reihenfolge, probiert)
    print('kaese%d: %d Scheiben, %d Startscheiben probiert' % (nr, len(daten[nr]), probiert))
    print('   Käse %d × %d × %d' % tuple(masse))
    print('   ' + beschreibe(reihenfolge))
````

Das sind dieselben Käse wie in der Tabelle der Lösungshinweise (`6×6×6`, `2×1000×1000`, `10×1000×1000`, `210×210×210`,
`2310×2730×3570`, `30030×39270×510510`), und auch die ersten und letzten Scheiben stimmen dort. Auch das größte Beispiel,
`kaese6` mit 90 300 Scheiben, ist schnell fertig, denn es werden nur wenige Startscheiben aufgebaut, und die meisten scheitern nach wenigen Schritten.

**Wie viel bringt der Filter?** Er spart bei den größeren Beispielen etwa die Hälfte oder mehr der Versuche. Wir zählen, wie viele Startscheiben ohne und mit Filter aufgebaut werden, bis der Käse gefunden ist:

````python
def probierte_starts_ohne_filter(scheiben):
    vorrat = Counter(scheiben)
    probiert = 0
    for start in list(vorrat):
        probiert += 1
        if baue_kaese(start, vorrat, len(scheiben)) is not None:
            return probiert


for nr in range(1, 7):
    print('kaese%d: %5d verschiedene Größen, ohne Filter %5d Starts, mit Filter %5d Starts' % (
        nr, len(set(daten[nr])), probierte_starts_ohne_filter(daten[nr]), ergebnis_a[nr][2]))
````

### Wenn es nicht geht

Ersetzen wir bei `kaese1` eine `6×6`-Scheibe durch eine `6×7`-Scheibe, lässt sich kein Quader mehr bilden:

````python
kaputt = list(daten[1])
kaputt[kaputt.index((6, 6))] = (6, 7)
masse, reihenfolge, probiert = loese_a(kaputt)
print('Ergebnis:', 'kein Quader möglich' if masse is None else masse, '(%d Startscheiben probiert)' % probiert)
````

## Teil b, Erweiterung 1: Es fehlen Scheiben

Antje hat vielleicht Scheiben aufgegessen. Wie viele mindestens fehlen, wissen wir nicht, also probieren wir `0, 1, 2, ...` fehlende
Scheiben. Passt an den Käse keine echte Scheibe, darf eine **fiktive** Scheibe (eine fehlende) angelegt werden, solange das Budget
reicht.

Die erzwungene Reihenfolge gilt weiter, nur dass die Scheibe mit der höchsten Priorität auch eine **fehlende** sein kann. Fehlt
`(x, y)` im Vorrat, gibt es also zwei Möglichkeiten: `(x, y)` ist fehlend (fiktiv anlegen) oder `(x, y)` gab es nie, dann ist die
nächste echte Scheibe dran. Beides probieren wir; bei Misserfolg machen wir die Schritte wieder rückgängig (**Backtracking**).
Solange das Budget aufgebraucht ist, verhält sich die Funktion genau wie `baue_kaese`.

````python
def fuelle_auf(masse, vorrat, rest, reihenfolge, budget):
    """Setzt den Käse fort. masse: Tupel; rest: Liste mit einer Zahl (noch nicht angelegte echte Scheiben);
    reihenfolge: Liste der angelegten Scheiben (fiktive als (p, q, 'fehlt')); budget: wie viele fiktive noch erlaubt sind.
    Ergebnis: Maße oder None. Bei None sind alle Schritte dieses Aufrufs rückgängig gemacht."""
    anfang = len(reihenfolge)
    masse = list(masse)
    while rest[0] > 0:
        x, y, z = masse
        kandidaten = [(x, y, 2), (x, z, 1), (y, z, 0)]
        echt = None                              # Position der ersten echten Scheibe in der Priorität
        for k in range(3):
            p, q, i = kandidaten[k]
            if vorrat[(p, q)] > 0:
                echt = k
                break
        if budget > 0:
            # jede Scheibe mit höherer Priorität als die echte könnte die fehlende sein
            bis = 3 if echt is None else echt
            gesehen = []
            for p, q, i in kandidaten[:bis]:
                if (p, q) in gesehen:
                    continue
                gesehen.append((p, q))
                neu = list(masse)
                neu[i] += 1
                neu.sort()
                reihenfolge.append((p, q, 'fehlt'))
                erg = fuelle_auf(neu, vorrat, rest, reihenfolge, budget - 1)
                if erg is not None:
                    return erg
                reihenfolge.pop()
        if echt is None:
            break
        p, q, i = kandidaten[echt]
        vorrat[(p, q)] -= 1
        rest[0] -= 1
        masse[i] += 1
        masse.sort()
        reihenfolge.append((p, q))
    else:
        return tuple(masse)                      # alle echten Scheiben angelegt
    while len(reihenfolge) > anfang:             # Misserfolg: rückgängig machen
        s = reihenfolge.pop()
        if len(s) == 2:
            vorrat[s] += 1
            rest[0] += 1
    return None


def loese_fehlend(scheiben, hoechstens):
    """Kleinste Zahl fehlender Scheiben (bis hoechstens), mit der sich ein Quader bauen lässt.
    Ergebnis: (Zahl, Maße, Reihenfolge) oder None."""
    vorrat = Counter(scheiben)
    n = len(scheiben)
    kuerzeste = min(a for a, b in scheiben)
    for fehlt in range(hoechstens + 1):
        for start in list(vorrat):
            if fehlt == 0 and vorrat[start] < min(kuerzeste, n):
                continue                         # Filter nur ohne fehlende Scheiben
            # Start 1: die Startscheibe ist die erste Scheibe überhaupt
            vorrat[start] -= 1
            reihenfolge = [start]
            erg = fuelle_auf((1, start[0], start[1]), vorrat, [n - 1], reihenfolge, fehlt)
            if erg is not None:
                return fehlt, erg, reihenfolge
            vorrat[start] += 1
            if fehlt > 0:
                # Start 2: die erste Scheibe (gleiche Größe) fehlt, die Startscheibe ist die zweite
                vorrat[start] -= 1
                reihenfolge = [start + ('fehlt',), start]
                erg = fuelle_auf((2, start[0], start[1]), vorrat, [n - 1], reihenfolge, fehlt - 1)
                if erg is not None:
                    return fehlt, erg, reihenfolge
                vorrat[start] += 1
    return None
````

**Test.** Wir nehmen aus jedem Beispiel `k` Scheiben heraus (an festen Stellen der Liste, also reproduzierbar) und suchen die kleinste
Zahl fehlender Scheiben. Angegeben sind die entfernten Scheiben, die gefundene Zahl, die Maße und die erfundenen Scheiben. Die
Kontrolle `pruefe_aufbau` spielt den Aufbau samt erfundener Scheiben nach.

````python
def entferne(scheiben, k):
    n = len(scheiben)
    nummern = [(j + 1) * n // (k + 2) for j in range(k)]
    entfernt = [scheiben[i] for i in nummern]
    rest = [s for i, s in enumerate(scheiben) if i not in nummern]
    return rest, entfernt


ergebnis_fehlt = {}
for nr in range(1, 7):
    for k in (1, 2, 3):
        if nr == 6 and k > 1:
            continue                             # dauert länger, siehe Text
        rest, entfernt = entferne(daten[nr], k)
        gefunden = loese_fehlend(rest, k)
        fehlt, masse, reihenfolge = gefunden
        assert pruefe_aufbau(reihenfolge) == list(masse)
        erfunden = [zeige(s) for s in reihenfolge if len(s) == 3]
        ergebnis_fehlt[(nr, k)] = (fehlt, masse, entfernt, erfunden)
        print('kaese%d, %d entfernt (%s): kleinste Zahl fehlender Scheiben %d, Käse %d × %d × %d, erfunden: %s' % (
            nr, k, ', '.join(zeige(s) for s in entfernt), fehlt, masse[0], masse[1], masse[2],
            ', '.join(erfunden) if erfunden else '-'))
````

**Was fällt auf?**

- Wenn nicht die letzte Scheibe fehlt, findet das Verfahren hier genau so viele fehlende Scheiben, wie wir entfernt haben, den
  ursprünglichen Käse und sogar genau die entfernten Scheiben (`kaese3` bis `kaese6`, auch `kaese1` und `kaese2` mit 1 bis 2
  entfernten Scheiben, wenn man von den folgenden Fällen absieht). Gefragt ist nur der **kleinste** Wert; dass die erfundenen
  Scheiben die entfernten sind, ist nicht garantiert, denn man kann eine fehlende Scheibe oft an anderer Stelle durch eine
  andere ersetzen.
- Die **äußerste** Scheibe kann man nicht als fehlend erkennen: Fehlt sie, ist der Käse trotzdem schon ein vollständiger
  (kleinerer) Quader. Bei `kaese2` mit einer entfernten Scheibe (`2×1000`) findet das Verfahren deshalb gleich 0 fehlende Scheiben
  und einen Käse `2×999×1000`.
- Dasselbe passiert bei mehreren entfernten Scheiben: Aus `kaese1` entfernen wir `6×6`, `4×6` und `2×4`, und es genügen zwei
  erfundene Scheiben für einen kleineren Quader `5×6×6`. Bei `kaese3` ergibt sich mit drei entfernten Scheiben `9×1000×1000`,
  bei `kaese2` mit zwei oder drei entfernten Scheiben `2×999×1000` bzw. `2×998×1000`, jeweils mit einer erfundenen Scheibe.
- Die Rechenzeit wächst schnell mit der Zahl der fehlenden Scheiben, denn bei jedem Schritt gibt es bis zu drei Möglichkeiten für
  eine fiktive Scheibe. Bei `kaese5` mit 3 fehlenden Scheiben ist sie noch kurz; bei `kaese6` mit 90 300 Scheiben dauert schon
  der Fall mit 2 fehlenden Scheiben deutlich länger (etwa zehn Sekunden auf einem PC, im Notebook nicht gezeigt). In den Lösungshinweisen heißt es
  ähnlich, dass mit mehr fehlenden Scheiben die großen Beispiele zu schwer werden.

## Teil b, Erweiterung 2: Zwei Quader sind vermischt

Jetzt kommen die Scheiben aus **zwei** Quadern. Die Zwangsregel gilt nicht mehr, denn die Scheibe `(x, y)` kann zum anderen Quader
gehören. Dafür sind es aber höchstens drei Kandidaten pro Schritt (`(x, y)`, `(x, z)`, `(y, z)`, soweit im Vorrat vorhanden).

Wir bauen den **ersten Quader** von innen nach außen auf, indem wir bei jedem Schritt alle vorhandenen Kandidaten ausprobieren
(Rekursion mit Rückgängigmachen). Nach jedem Schritt prüfen wir mit `loese_vorrat` aus Teil a, ob die **übrigen** Scheiben einen
Quader ergeben. Wenn ja, sind wir fertig. Um nichts doppelt zu probieren, merken wir uns die schon besuchten Zustände
(Maße des Käses und welche Scheiben angelegt sind).

Warum ist das vollständig? Gibt es eine Aufteilung in zwei Quader, dann ist der innerste Teil des ersten Quaders eine der
Startscheiben, und die nächste Scheibe des ersten Quaders ist immer eine der höchstens drei Kandidaten.

````python
def mische_zwei(scheiben):
    """Zerlegt die Scheiben in zwei Quader. Ergebnis: ((Maße 1, Reihenfolge 1), (Maße 2, Reihenfolge 2), Zahl der Zustände)
    oder None."""
    vorrat = Counter(scheiben)
    besucht = set()
    reihenfolge = []

    def suche(masse):
        schluessel = (tuple(masse), tuple(sorted(reihenfolge)))
        if schluessel in besucht:
            return None
        besucht.add(schluessel)
        if sum(vorrat.values()) > 0:
            masse2, reihe2, _ = loese_vorrat(vorrat)          # Rest ein Quader?
            if masse2 is not None:
                return (list(masse), list(reihenfolge)), (masse2, reihe2)
        x, y, z = masse
        for p, q, i in [(x, y, 2), (x, z, 1), (y, z, 0)]:
            if vorrat[(p, q)] > 0:
                vorrat[(p, q)] -= 1
                reihenfolge.append((p, q))
                neu = list(masse)
                neu[i] += 1
                neu.sort()
                erg = suche(neu)
                if erg is not None:
                    return erg
                reihenfolge.pop()
                vorrat[(p, q)] += 1
        return None

    for start in list(vorrat):
        vorrat[start] -= 1
        reihenfolge.append(start)
        erg = suche([1, start[0], start[1]])
        if erg is not None:
            return erg[0], erg[1], len(besucht)
        reihenfolge.pop()
        vorrat[start] += 1
    return None
````

**Test.** Wir mischen je zwei der ersten fünf Beispiele (alle zehn Paare; `kaese6` ist mit seinen 90 300 Scheiben für diese Suche zu
groß). Die Kontrolle spielt beide Aufbauten nach und prüft, dass alle Scheiben genau einmal vorkommen.

````python
from itertools import combinations

ergebnis_mix = {}
for a, b in combinations(range(1, 6), 2):
    gemischt = daten[a] + daten[b]
    erg = mische_zwei(gemischt)
    (m1, r1), (m2, r2), zustaende = erg
    assert pruefe_aufbau(r1) == m1 and pruefe_aufbau(r2) == m2
    assert sorted(r1 + r2) == sorted(gemischt)
    ergebnis_mix[(a, b)] = (m1, m2, zustaende)
    print('kaese%d + kaese%d (%4d Scheiben): %d×%d×%d und %d×%d×%d, %d Zustände besucht' % (
        a, b, len(gemischt), m1[0], m1[1], m1[2], m2[0], m2[1], m2[2], zustaende))
````

In jedem Fall werden genau die beiden ursprünglichen Käse gefunden. Auch das Gegenteil funktioniert: Ersetzen wir in der Mischung
`kaese1 + kaese2` eine Scheibe durch eine, die nirgends passt, meldet das Verfahren, dass sich die Scheiben nicht in zwei Quader
zerlegen lassen.

````python
gemischt_kaputt = daten[1] + daten[2]
gemischt_kaputt[gemischt_kaputt.index((6, 6))] = (6, 7)
print('Zerlegung in zwei Quader:', 'unmöglich' if mische_zwei(gemischt_kaputt) is None else 'gefunden')
````

**Grenzen.** Für drei oder mehr Quader müsste man die Suche entsprechend verschachteln (alle Quader gleichzeitig aufbauen,
Quader sperren, die nichts mehr anlegen können). Das wird schnell sehr teuer; wir haben es nicht umgesetzt. Auch fehlende Scheiben
**und** zwei Quader zugleich haben wir nicht kombiniert.

## Zum Schluss: Käse mit MiniZinc

Geht das auch mit einem Solver? Die Reihenfolge der Scheiben ist eine Permutation, und Permutationen sind für Solver oft schwer
(auch bei `44_ChoreoGraph` war das so). Wir versuchen es trotzdem mit einer Sicht, die **ohne** Permutation auskommt: Jeder
Schritt legt eine Scheibe **senkrecht zu einer der drei Achsen** an. Dann ist der Aufbau ein Weg im Gitter: `achse[k] ∈ {1, 2, 3}`
sagt, welches Maß im Schritt `k` um 1 wächst. Mit den Zählern `u[k], v[k], w[k]` (wie oft Achse 1, 2, 3 bis dahin gewählt wurde) und
den Maßen `X0, Y0, Z0` vor der ersten Scheibe ist das Käsemaß in Richtung 1 gerade `X0 + u[k-1]`, usw. Die Scheibe im Schritt
`k` hat dann die beiden **anderen** Maße als Seiten. Wir kodieren sie als eine Zahl `min · (S + 1) + max` und verlangen mit
`global_cardinality`, dass jede Scheibengröße genau so oft vorkommt, wie es sie gibt. Vor der ersten Scheibe hat der Käse keine
Dicke (`X0 · Y0 · Z0 = 0`). Als Hilfe für den Solver verlangen wir, dass das Endvolumen die Summe der Flächen ist.
Neu gegenüber den Arbeitsblättern ist nur, dass Variablen als Index **und** in `let`-Ausdrücken vorkommen.

```minizinc
include "globals.mzn";

int: n;                                   % Zahl der Scheiben
int: m;                                   % Zahl der verschiedenen Scheibengrößen
int: S;                                   % längste Seite einer Scheibe
int: V;                                   % Gesamtvolumen = Summe der Flächen
array[1..m] of int: code;                 % Größe (a, b) als Zahl: a * (S + 1) + b
array[1..m] of int: anz;                  % so viele Scheiben dieser Größe gibt es

array[1..n] of var 1..3: achse;           % Schritt k legt eine Scheibe senkrecht zur Achse 1, 2 oder 3 an
array[0..n] of var 0..n: u;               % u[k], v[k], w[k]: Scheiben bisher je Achse
array[0..n] of var 0..n: v;
array[0..n] of var 0..n: w;
var 0..S: X0;                             % Maße des "Käses" vor der ersten Scheibe
var 0..S: Y0;
var 0..S: Z0;

constraint u[0] = 0 /\ v[0] = 0 /\ w[0] = 0;
constraint forall(k in 1..n)(
  u[k] = u[k-1] + bool2int(achse[k] = 1) /\
  v[k] = v[k-1] + bool2int(achse[k] = 2) /\
  w[k] = w[k-1] + bool2int(achse[k] = 3));

% vor der ersten Scheibe hat der Käse keine Dicke: mindestens ein Maß ist 0
constraint X0 * Y0 * Z0 = 0;

% Schritt k: Der Käse hat die Maße (x, y, z); die Scheibe hat die beiden Seiten senkrecht zur Achse
array[1..n] of var int: c;
constraint forall(k in 1..n)(
  let { var int: x = X0 + u[k-1];
        var int: y = Y0 + v[k-1];
        var int: z = Z0 + w[k-1];
        var int: p = if achse[k] = 1 then y else x endif;
        var int: q = if achse[k] = 3 then y else z endif }
  in c[k] = min(p, q) * (S + 1) + max(p, q));

% genau die gegebenen Scheiben werden verbraucht
constraint global_cardinality(c, code, anz);

% überflüssige, aber beschleunigende Bedingungen
constraint (X0 + u[n]) * (Y0 + v[n]) * (Z0 + w[n]) = V;
constraint u[n] + v[n] + w[n] = n;

solve satisfy;
output ["\(sort([X0 + u[n], Y0 + v[n], Z0 + w[n]]))\n"];
```

Die Datendatei erzeugt eine Python-Zelle aus den Beispieldaten:

````python
def als_dzn(scheiben):
    vorrat = Counter(scheiben)
    groessen = sorted(vorrat)
    laengste = max(b for a, b in scheiben)
    codes = [a * (laengste + 1) + b for a, b in groessen]
    return 'n = %d;\nm = %d;\nS = %d;\nV = %d;\ncode = %s;\nanz = %s;' % (
        len(scheiben), len(groessen), laengste, sum(a * b for a, b in scheiben), codes, [vorrat[g] for g in groessen])


print(als_dzn(daten[1]))
````

## Für Neugierige

- **Volumentrick (Lösungshinweise):** Das Gesamtvolumen `V` der Scheiben ist `x · y · z` des fertigen Käses, und die *letzte*
  angelegte Scheibe hat die Fläche `x · y`. Also muss die Fläche der letzten Scheibe ein **Teiler von `V`** sein. Daran erkennt man
  schnell, welche Scheiben als letzte (äußerste) Scheibe überhaupt in Frage kommen. Baut man den Käse dann von außen nach innen
  ab, spart man viele Startscheiben. Das ist ein zweiter Filter neben unserem.
- **Rekursion und Stapelüberlauf:** In Python ist eine rekursive Lösung mit 90 000 Schritten schwierig (die Rekursionstiefe ist
  begrenzt). Darum bauen wir den Käse in einer Schleife auf; nur die Erweiterungen rekursieren (so tief wie die Zahl der fiktiven
  Scheiben bzw. wie die Scheibenzahl eines kleinen Quaders).
- **Graph der Scheiben:** Verbindet man Scheiben, die mindestens eine Seitenlänge gemeinsam haben, sind für das Anlegen nur
  Nachbarn der letzten Scheibe interessant. Daraus folgt eine weitere Filterregel für die Startscheibe.
- **Mehrere Lösungen:** Unsere Funktion liefert die **erste** Startscheibe, die gelingt. Meist ist der Käse eindeutig; bei
  Mischungen kann es mehrere Zerlegungen geben.
- **Laufzeit:** Teil a braucht höchstens `n · n*` Schritte (Startscheiben mal Aufbauschritte), der Speicher ist linear in `n`.
