## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Graphen:** Ein Graph besteht aus Knoten und Kanten. Wir speichern ihn als Dictionary: Zu jedem Knoten gehört die **Menge seiner
  Nachbarn**. `graph[v]` sind also die Nachbarn von `v`, und `w in graph[v]` fragt, ob es die Kante zwischen `v` und `w` gibt.
- **Clique:** eine Menge von Knoten, die alle paarweise durch Kanten verbunden sind. Eine Clique ist **maximal**, wenn kein weiterer
  Knoten mehr dazupasst.
- **Vollständige Suche mit Rekursion (Backtracking):** Wir legen die Cliquen eine nach der anderen in eine Reihe. Passt nichts mehr,
  gehen wir einen Schritt zurück und probieren die nächste Möglichkeit. Gescheiterte Zwischenstände merken wir uns in einer Menge,
  damit die Suche sie nicht doppelt untersucht.
- **Streichen:** Um eine kleine „schuldige“ Knotenmenge zu finden, lassen wir Knoten für Knoten weg, solange das Problem bestehen
  bleibt.
- Für Neugierige: **Tiefensuche** für Zusammenhangskomponenten.
- In der Fachliteratur heißt ein PLG **Intervallgraph**. Unter diesem Namen findest du viel darüber.

**Nützliche Funktionen und Bibliotheken**
- Mengen: `a & b` (Schnitt), `a | b` (Vereinigung), `a - b` (Differenz), `a <= b` (ist `a` Teilmenge von `b`?), `a < b` (echte
  Teilmenge). `frozenset(a)` ist eine unveränderliche Menge; anders als `set` darf sie in einer anderen Menge oder als Schlüssel eines
  Dictionarys stehen.
- Tupel werden Eintrag für Eintrag verglichen: `(1992, 1, 1) < (1992, 12, 19)`, weil die Jahre gleich sind und `1 < 12`. Deshalb
  speichern wir ein Datum als Tupel `(jahr, monat, tag)`. Eine Liste von Tupeln sortiert `sort()` genauso: nach dem ersten Eintrag,
  bei Gleichstand nach dem zweiten.
- `itertools.combinations(liste, k)` liefert alle Auswahlen von `k` Elementen der Liste (für Neugierige).

## Die Idee in Worten

**Lebensspannen als Intervalle.** Eine Lebensspanne ist ein Zeitabschnitt `[von, bis]`. Zwei Spannen `[a, b]` und `[c, d]`
überlappen sich genau dann, wenn `a <= d` und `c <= b` gilt: Keine der beiden ist zu Ende, bevor die andere anfängt. Den
Lebensgraphen zu gegebenen Spannen auszurechnen ist also leicht (Schritt 1). Die Aufgabe fragt aber umgekehrt: Zu einem Graphen
ohne Jahreszahlen sollen wir Spannen **finden** oder zeigen, dass es keine gibt. (Die Hinweise warnen: Wer nur prüft, ob gegebene
Spannen zum Graphen passen, hat die Aufgabe missverstanden.)

**Zeitpunkte und Cliquen.** Wer zu einem bestimmten Zeitpunkt lebt, ist paarweise gleichzeitig am Leben, also eine Clique im
Graphen. Geht man die Zeitachse entlang, ändert sich diese Clique nur, wenn jemand geboren wird oder stirbt. Die **maximalen**
Cliquen eines Lebensgraphen kommen so in einer Reihenfolge von links nach rechts vor, und jede Person steckt in einem
**zusammenhängenden Stück** dieser Reihe: von ihrer Geburt bis zu ihrem Tod. Wir nennen so eine Reihe eine **Cliquenkette**.

Umgekehrt: Lassen sich die maximalen Cliquen eines Graphen so in eine Reihe legen, dass jeder Knoten in aufeinanderfolgenden
Cliquen vorkommt, dann ist der Graph ein PLG. Wir nummerieren die Plätze 1, 2, 3, … und geben jedem Knoten die Spanne von seinem
ersten bis zu seinem letzten Platz. Haben zwei Knoten eine Kante, liegen sie gemeinsam in einer maximalen Clique, ihre Spannen
teilen also diesen Platz. Überlappen sich zwei Spannen, gibt es einen Platz, an dem beide vorkommen; dort liegt eine Clique, die
beide enthält, also haben sie eine Kante. Dieser Zusammenhang ist als Satz von Gilmore und Hoffman (1964) bekannt.

**Plan.**
1. **Wer stirbt zuerst?** Die Person, deren Leben als erstes endet, hat nur Nachbarn, die bei ihrem Tod noch leben, also
   untereinander eine Clique bilden. Solche Knoten heißen **simplizial**. Wir streichen immer wieder einen simplizialen Knoten.
   Klappt das bis zum Ende, kennen wir daraus die maximalen Cliquen. Bleibt ein Rest ohne simplizialen Knoten, ist der Graph kein
   PLG. (Der Rest enthält dann ein **Loch**, einen Kreis aus mindestens vier Knoten ohne Abkürzungen, wie in den Hinweisen.)
2. **Cliquenkette suchen:** die maximalen Cliquen von links nach rechts legen, mit Regeln, die Sackgassen früh erkennen.
3. **Lebensspannen ablesen** und zur Probe den Lebensgraphen daraus berechnen.
4. **Teil 2:** Knoten für Knoten streichen, solange der Rest kein PLG ist.

## Schritt 1: Der Lebensgraph zu gegebenen Lebensspannen

Zuerst die leichte Richtung, damit klar ist, was ein Lebensgraph ist. Ein Datum wird zum Tupel `(jahr, monat, tag)`. Steht nur ein
Jahr da, nehmen wir bei der Geburt den 1. Januar und beim Tod den 31. Dezember. Wer noch lebt, bekommt als Ende `(9999, 12, 31)`.

````python
def datum(text, ist_ende):
    """'1906' oder '01.01.1992' als Tupel (jahr, monat, tag); leer heißt: lebt noch."""
    text = text.strip()
    if text == '':
        return (9999, 12, 31)
    teile = text.split('.')
    if len(teile) == 3:
        return (int(teile[2]), int(teile[1]), int(teile[0]))
    if ist_ende:
        return (int(text), 12, 31)
    return (int(text), 1, 1)


def lies_lebensspannen(dateiname):
    """Dictionary Name -> (von, bis). Zeilen mit # sind Kommentare."""
    spannen = {}
    with open(dateiname, encoding='utf-8') as datei:
        for zeile in datei:
            if zeile.strip() == '' or zeile.startswith('#'):
                continue
            name, geburt, tod = zeile.split(';')
            spannen[name.strip()] = (datum(geburt, False), datum(tod, True))
    return spannen


def lebensgraph(spannen):
    """Graph als Dictionary Knoten -> Menge der Nachbarn: Kante, wenn sich die Spannen überlappen."""
    graph = {}
    for u in spannen:
        graph[u] = set()
    for u in spannen:
        for v in spannen:
            von_u, bis_u = spannen[u]
            von_v, bis_v = spannen[v]
            if u != v and von_u <= bis_v and von_v <= bis_u:
                graph[u].add(v)
    return graph


spannen_blatt = lies_lebensspannen('beispieldaten/aufgabenblatt.txt')
blatt = lebensgraph(spannen_blatt)
for person in blatt:
    print(f'{person:17}', sorted(blatt[person]))
````

Grace Hopper und Meike Grewing sind beide im Jahr 1992 eingetragen, haben aber keine Kante: Grace Hopper starb am 1. Januar, Meike
Grewing wurde am 19. Dezember geboren. Ada Lovelace hat gar keine Nachbarn. Das ist der Lebensgraph aus dem Aufgabenblatt mit 7
Knoten und 12 Kanten.

Für die Graphen ohne Jahreszahlen lesen wir die Adjazenzmatrix. Damit wir auch mit diesem Format den Graphen aus dem Aufgabenblatt
haben, steht er (nach den Lösungshinweisen) in `beispiel0.txt`.

````python
def lies_graph(dateiname):
    """Liest n und die Adjazenzmatrix. Ergebnis: Dictionary Knoten (0 bis n-1) -> Menge der Nachbarn."""
    with open(dateiname) as datei:
        zeilen = datei.read().split('\n')
    n = int(zeilen[0])
    graph = {}
    for i in range(n):
        werte = zeilen[1 + i].split()
        graph[i] = set()
        for j in range(n):
            if werte[j] == '1':
                graph[i].add(j)
    return graph


def anzahl_kanten(graph):
    summe = 0
    for v in graph:
        summe += len(graph[v])
    return summe // 2          # jede Kante wurde von beiden Enden aus gezählt


bsp0 = lies_graph('beispieldaten/beispiel0.txt')
for v in bsp0:
    print(v, sorted(bsp0[v]))
print(len(bsp0), 'Knoten,', anzahl_kanten(bsp0), 'Kanten')
````

Es ist derselbe Graph, nur mit Nummern statt Namen: Knoten 6 ist Ada Lovelace (ohne Nachbarn), 3 Adele Goldstine, 2 Grace Hopper,
0 Marissa Mayer und 5 Meike Grewing; 1 und 4 sind Frances E. Allen und Anita Borg, die beide mit allen außer Ada Lovelace
zusammen gelebt haben.

## Schritt 2: Wer stirbt zuerst? Simpliziale Knoten

Ein Knoten ist **simplizial**, wenn seine Nachbarn eine Clique bilden. In einem Lebensgraphen ist die Person, die als erste stirbt,
immer simplizial: Jeder ihrer Nachbarn lebt bei ihrem Tod noch (sonst wäre er vorher gestorben) und hat schon gelebt (sonst
würden sich die Spannen nicht überlappen). Alle Nachbarn leben also zu diesem Zeitpunkt gleichzeitig und sind paarweise verbunden.

Lassen wir eine Person weg, bleibt ein Lebensgraph übrig, und darin gibt es wieder einen simplizialen Knoten. Wir streichen deshalb
immer wieder einen simplizialen Knoten aus dem **Rest**. Die Reihenfolge des Streichens heißt **Eliminationsreihenfolge**.
Findet sich im Rest kein simplizialer Knoten, kann der Graph kein PLG sein.

Wir nehmen jeweils den kleinsten simplizialen Knoten (`sorted`), damit das Ergebnis bei jedem Lauf gleich ist.

````python
def ist_clique(graph, menge):
    """Sind alle Knoten der Menge paarweise verbunden?"""
    for u in menge:
        for v in menge:
            if u != v and v not in graph[u]:
                return False
    return True


def eliminationsreihenfolge(graph):
    """Streicht immer wieder einen simplizialen Knoten. Ergebnis: Liste der Knoten in Streich-Reihenfolge,
    oder None, wenn ein Rest ohne simplizialen Knoten übrig bleibt (dann ist der Graph kein PLG)."""
    rest = set(graph)
    reihe = []
    while len(rest) > 0:
        gefunden = None
        for v in sorted(rest):
            if ist_clique(graph, graph[v] & rest):     # Nachbarn von v, die noch im Rest sind
                gefunden = v
                break
        if gefunden is None:
            return None
        reihe.append(gefunden)
        rest.remove(gefunden)
    return reihe


print(eliminationsreihenfolge(blatt))
print(eliminationsreihenfolge(bsp0))
bsp1 = lies_graph('beispieldaten/beispiel1.txt')
print(eliminationsreihenfolge(bsp1))
````

`beispiel1.txt` ist ein Kreis aus vier Knoten 0 – 1 – 2 – 3 – 0 ohne Querverbindung. Jeder Knoten hat zwei Nachbarn, die nicht
verbunden sind; keiner ist simplizial. So ein **Loch** kann kein Lebensgraph sein: Wer von den vier zuerst stirbt, müsste mit zwei
Personen gleichzeitig gelebt haben, die sich nie begegnet sind.

## Schritt 3: Die maximalen Cliquen

Beim Streichen eines Knotens `v` bilden `v` und seine Nachbarn im Rest eine Clique (sie sind ja eine Clique, und `v` ist mit allen
verbunden). So bekommen wir eine Clique pro Knoten. Jede **maximale** Clique ist darunter: Nimm ihren Knoten `v`, der zuerst
gestrichen wird. In diesem Moment sind alle anderen Knoten der Clique noch im Rest und Nachbarn von `v`. Die Clique von `v` enthält
also die maximale Clique, und weil diese maximal ist, sind beide gleich.

Wir behalten von den gefundenen Cliquen nur die, die nicht in einer anderen enthalten sind (`c < d`: echte Teilmenge).

````python
def maximale_cliquen(graph, reihe):
    """Die maximalen Cliquen (Liste von frozensets) aus der Eliminationsreihenfolge."""
    position = {}
    for i in range(len(reihe)):
        position[reihe[i]] = i
    kandidaten = []
    for v in reihe:
        clique = {v}
        for w in graph[v]:
            if position[w] > position[v]:         # w wird erst nach v gestrichen
                clique.add(w)
        kandidaten.append(frozenset(clique))
    cliquen = []
    for c in kandidaten:
        maximal = True
        for d in kandidaten:
            if c < d:
                maximal = False
        if maximal and c not in cliquen:
            cliquen.append(c)
    return cliquen


cliquen_blatt = maximale_cliquen(blatt, eliminationsreihenfolge(blatt))
for nr in range(len(cliquen_blatt)):
    print(nr, sorted(cliquen_blatt[nr]))
````

Vier maximale Cliquen: Ada Lovelace allein, und dreimal vier Personen, die gleichzeitig gelebt haben. Clique 1 ist die Zeit von
1949 bis 1964 (Goldstine, Hopper, Allen, Borg), Clique 2 die Zeit von 1975 bis 1991 (Hopper, Allen, Borg, Mayer), Clique 3 die
Zeit von Ende 1992 bis 2003 (Allen, Borg, Mayer, Grewing).

## Schritt 4: Die Cliquenkette suchen

Jetzt legen wir die maximalen Cliquen von links nach rechts in eine Reihe. Zu jedem Zeitpunkt der Suche gibt es drei Sorten Knoten:

- **vorbei:** kam schon in einer gelegten Clique vor, lebt aber nicht weiter. Er darf nie wieder vorkommen, sonst hätte seine
  Spanne eine Lücke.
- **offen:** steckt in der zuletzt gelegten Clique und kommt auch in einer noch nicht gelegten Clique vor. Er muss in der
  **nächsten** Clique vorkommen, sonst hätte seine Spanne ebenfalls eine Lücke.
- alle anderen: noch nicht geboren.

Eine Clique darf also als nächste gelegt werden, wenn sie **alle offenen Knoten enthält** und **keinen Knoten, der vorbei ist**.
Kommen mehrere Cliquen in Frage, probieren wir sie der Reihe nach und gehen bei einer Sackgasse zurück. Zwei Ideen sparen Arbeit:

- Wir probieren zuerst die Cliquen, die am wenigsten **neue Knoten mitbringen, die später noch gebraucht werden**. Eine Clique, die
  gar keine solchen Knoten mitbringt (alle neuen Knoten leben nur in ihr), legen wir sofort und probieren keine andere: Sie kann
  nicht schaden. Gäbe es eine Kette, in der sie erst später kommt, könnten wir sie nach vorne ziehen: Alle offenen Knoten leben bis
  zu ihr ohnehin weiter, und ihre neuen Knoten stören niemanden.
- Ob die Suche von einem Zwischenstand aus zum Ziel kommt, hängt nur davon ab, **welche** Cliquen schon liegen und welche Knoten
  offen sind, nicht von der Reihenfolge. Gescheiterte Zwischenstände merken wir uns in der Menge `gescheitert`.

Am Anfang ist nichts offen und nichts vorbei; dann darf jede Clique die erste sein, und auch dort probieren wir die Cliquen mit
wenigen geteilten Knoten zuerst (ein guter Kandidat für das linke Ende). Bei einem Graphen aus mehreren Teilen ist zwischendurch
wieder nichts offen; dann beginnt der nächste Teil.

`lege_weiter` arbeitet mit den Nummern der Cliquen. Der Zähler `zaehler['schritte']` zählt die Aufrufe, damit wir sehen, wie viel
die Suche zu tun hat.

````python
zaehler = {'schritte': 0, 'tests': 0}


def lege_weiter(cliquen, kette, offen, gescheitert):
    """kette: Nummern der schon gelegten Cliquen (von links), offen: Knoten, die in der nächsten Clique vorkommen müssen.
    Ergebnis: die vollständige Kette (Liste von Nummern) oder None, wenn es von hier aus nicht weitergeht."""
    zaehler['schritte'] += 1
    rest = []
    for nr in range(len(cliquen)):
        if nr not in kette:
            rest.append(nr)
    if len(rest) == 0:
        return kette
    zustand = (frozenset(kette), offen)
    if zustand in gescheitert:
        return None

    vorbei = set()                    # Knoten, die schon vorkamen und nicht weiterleben
    for nr in kette:
        vorbei |= cliquen[nr]
    vorbei -= offen

    kandidaten = []                   # Paare (Anzahl neuer Knoten, die später noch vorkommen, Nummer)
    neu_offen = {}                    # Nummer -> Knoten, die nach dieser Clique offen sind
    for nr in rest:
        c = cliquen[nr]
        if offen <= c and len(c & vorbei) == 0:
            spaeter = set()           # Knoten der anderen noch freien Cliquen
            for anderes in rest:
                if anderes != nr:
                    spaeter |= cliquen[anderes]
            neu_offen[nr] = frozenset(c & spaeter)
            kandidaten.append((len(neu_offen[nr] - offen), nr))
    kandidaten.sort()
    if len(kandidaten) > 0 and kandidaten[0][0] == 0:
        kandidaten = kandidaten[:1]   # bringt nichts mit, was später gebraucht wird: sofort legen

    for anzahl, nr in kandidaten:
        ergebnis = lege_weiter(cliquen, kette + [nr], neu_offen[nr], gescheitert)
        if ergebnis is not None:
            return ergebnis
    gescheitert.add(zustand)
    return None


def cliquenkette(cliquen):
    return lege_weiter(cliquen, [], frozenset(), set())


zaehler['schritte'] = 0
kette_blatt = cliquenkette(cliquen_blatt)
print('Kette:', kette_blatt, ' Schritte:', zaehler['schritte'])
````

Die Suche beginnt mit Clique 0 (Ada Lovelace; sie teilt mit niemandem einen Knoten) und legt dann 1, 2, 3, ohne einmal
zurückzugehen: 5 Aufrufe für 4 Cliquen.

## Schritt 5: Lebensspannen ablesen – der PLG-Tester

Die Spanne eines Knotens reicht vom ersten bis zum letzten Platz der Kette, an dem er vorkommt (Plätze ab 1 gezählt). `ist_plg`
fasst alles zusammen und gibt die Spannen zurück oder `None`, wenn der Graph kein PLG ist. Zur Probe berechnen wir aus den Spannen
mit `lebensgraph` aus Schritt 1 wieder einen Graphen. Er muss genau der Eingabegraph sein.

````python
def spannen_aus_kette(cliquen, kette):
    spannen = {}
    for platz in range(1, len(kette) + 1):
        for v in cliquen[kette[platz - 1]]:
            if v in spannen:
                spannen[v] = (spannen[v][0], platz)   # neues Ende
            else:
                spannen[v] = (platz, platz)           # erster Platz
    return spannen


def ist_plg(graph):
    """Lebensspannen (Dictionary Knoten -> (von, bis)) oder None, wenn der Graph kein PLG ist."""
    zaehler['tests'] += 1
    reihe = eliminationsreihenfolge(graph)
    if reihe is None:
        return None
    cliquen = maximale_cliquen(graph, reihe)
    kette = cliquenkette(cliquen)
    if kette is None:
        return None
    return spannen_aus_kette(cliquen, kette)


def zeige_spannen(spannen):
    """Jede Spanne als Balken, ein Platz der Kette = drei Zeichen. Sortiert nach Anfang, Ende und Name."""
    zeilen = []
    for v in spannen:
        von, bis = spannen[v]
        zeilen.append((von, bis, str(v)))
    zeilen.sort()
    breite = 0
    for von, bis, name in zeilen:
        breite = max(breite, len(name))
    for von, bis, name in zeilen:
        print(f'{name:{breite}}  {von:2} bis {bis:2}  ' + '   ' * (von - 1) + '===' * (bis - von + 1))


s_blatt = ist_plg(blatt)
zeige_spannen(s_blatt)
print('Probe:', lebensgraph(s_blatt) == blatt)
````

Die Spannen sind keine Jahreszahlen mehr, sondern Plätze in der Kette, aber sie überlappen sich genau dort, wo sich die echten
Lebensspannen überlappen. Wer will, kann die Plätze in Jahre umrechnen (die Hinweise geben zum Beispiel Daten ab 1900 aus); für die
Frage „PLG oder nicht“ spielt das keine Rolle.

Mit Nummern statt Namen (Beispiel 0 der Hinweise):

````python
s0 = ist_plg(bsp0)
zeige_spannen(s0)
print('Probe:', lebensgraph(s0) == bsp0)
````

## Schritt 6: Teil 2 – eine minimale Teilmenge

Ist ein Graph kein PLG, suchen wir eine Knotenmenge, die schon allein (mit den Kanten zwischen ihren Knoten) kein PLG ist, und die
**minimal** ist: Lässt man irgendeinen ihrer Knoten weg, ist der Rest ein PLG.

Dabei hilft eine einfache Beobachtung: Ist ein Graph ein PLG, dann auch jeder Teil davon. (Wir lassen einfach die Spannen der
fehlenden Knoten weg.) Daraus ergibt sich das **Streichverfahren**: Wir gehen alle Knoten der Reihe nach durch. Ist der Rest ohne
diesen Knoten immer noch kein PLG, streichen wir ihn endgültig, sonst behalten wir ihn.

Warum ist das Ergebnis minimal? Ein Knoten `v` blieb drin, weil der Rest ohne `v` damals ein PLG war. Das Ergebnis ohne `v` ist ein
Teil dieses Rests, also auch ein PLG. Das gilt für jeden Knoten des Ergebnisses. Das Verfahren braucht genau einen PLG-Test pro
Knoten (plus einen für den ganzen Graphen).

````python
def teilgraph(graph, knoten):
    """Der Graph, der nur aus den angegebenen Knoten und den Kanten zwischen ihnen besteht."""
    menge = set(knoten)
    teil = {}
    for v in knoten:
        teil[v] = graph[v] & menge
    return teil


def minimale_teilmenge(graph):
    """Liste von Knoten, die schon allein kein PLG bilden, minimal; None, wenn der Graph ein PLG ist."""
    if ist_plg(graph) is not None:
        return None
    knoten = sorted(graph)
    for v in sorted(graph):
        ohne_v = []
        for w in knoten:
            if w != v:
                ohne_v.append(w)
        if ist_plg(teilgraph(graph, ohne_v)) is None:
            knoten = ohne_v          # auch ohne v kein PLG: v endgültig streichen
    return knoten


bsp4 = lies_graph('beispieldaten/beispiel4.txt')
for v in bsp4:
    print(v, sorted(bsp4[v]))
zaehler['tests'] = 0
m4 = minimale_teilmenge(bsp4)
print('minimale Teilmenge:', m4, ' PLG-Tests:', zaehler['tests'])
print('Probe:', ist_plg(teilgraph(bsp4, m4)) is None, [ist_plg(teilgraph(bsp4, [w for w in m4 if w != v])) is not None for v in m4])
````

In Beispiel 4 bilden die Knoten 0 – 1 – 2 – 5 – 0 ein Loch; an Knoten 2 hängen noch 3, 4 und 6. Das Streichverfahren findet

## Für Neugierige

### Minimal ist nicht dasselbe wie am kleinsten

Die Hinweise suchen den **kleinsten** Teilgraphen, der kein PLG ist: Sie probieren alle Teilmengen der Größe nach durch. Das ist
etwas anderes als unsere **minimale** Teilmenge, bei der nur kein einzelner Knoten mehr wegfallen darf. Bei `loch_und_netz` ist
der Kreis mit acht Knoten minimal, das Netz mit sechs Knoten aber kleiner.

Die vollständige Suche nach der kleinsten Teilmenge geht mit `itertools.combinations`. Sie braucht bei `n` Knoten bis zu `2ⁿ`
PLG-Tests und ist nur für kleine Graphen machbar.

````python
from itertools import combinations


def kleinste_teilmenge(graph):
    """Eine kleinste Knotenmenge, die kein PLG bildet (alle Teilmengen der Größe nach), oder None."""
    knoten = sorted(graph)
    for groesse in range(1, len(knoten) + 1):
        for auswahl in combinations(knoten, groesse):
            if ist_plg(teilgraph(graph, auswahl)) is None:
                return list(auswahl)
    return None


kleinste = {}
for name in graphen:
    if name in minimal and len(graphen[name]) <= 14:
        zaehler['tests'] = 0
        kleinste[name] = kleinste_teilmenge(graphen[name])
        print(f'{name:14} minimal {str(minimal[name]):24} kleinste {str(kleinste[name]):24} Tests {zaehler["tests"]}')
````

Bei den Beispielen der Hinweise stimmen minimale und kleinste Teilmenge überein, bei `loch_und_netz` nicht. Für
`zufall_kein_plg` mit 30 Knoten haben wir die Suche außerhalb des Notebooks laufen lassen (im Browser dauert sie zu lange):
Nach 395 852 Tests findet sie die Knoten [2, 3, 6, 7, 10, 26], also
sechs Knoten statt der sieben unseres Sterns.

````python
teil6 = teilgraph(graphen['zufall_kein_plg'], [2, 3, 6, 7, 10, 26])
for v in teil6:
    print(v, sorted(teil6[v]))
print('PLG?', ist_plg(teil6) is not None)
````

### Wann die Cliquensuche viel zu tun hat

Unsere Suche geht zurück, wenn sie in eine Sackgasse läuft. Das kann teuer werden, wenn es viele gleichwertige Möglichkeiten gibt
und die Sackgasse erst ganz am Ende sichtbar wird. Ein Beispiel ist die **Blüte**: Ein Mittelknoten 0 hat `k` Blütenblätter; jedes
Blatt besteht aus drei Knoten `x`, `y`, `z` mit den Kanten 0–x, 0–y, 0–z, x–y, y–z, also aus zwei Dreiecken {0, x, y} und
{0, y, z}. Dazu kommt ein Weg 0 – p – q, und an q hängt ein Stern. Der ganze Graph ist kein PLG (wegen des Sterns), aber die Suche
merkt das erst, wenn sie alle Blätter gelegt hat, und probiert vorher sehr viele Auswahlen von Blättern durch.

````python
def bluete(k, mit_stern=True):
    kanten = []
    neu = 1                                   # nächste freie Knotennummer, 0 ist die Mitte
    for blatt_nr in range(k):
        x, y, z = neu, neu + 1, neu + 2
        neu += 3
        kanten += [(0, x), (0, y), (0, z), (x, y), (y, z)]
    p, q = neu, neu + 1
    neu += 2
    kanten += [(0, p), (p, q)]
    if mit_stern:                             # q ist die Mitte eines Sterns: Wege q-p-0, q-a-b, q-c-d
        a, b, c, d = neu, neu + 1, neu + 2, neu + 3
        neu += 4
        kanten += [(q, a), (a, b), (q, c), (c, d)]
    graph = {}
    for v in range(neu):
        graph[v] = set()
    for u, v in kanten:
        graph[u].add(v)
        graph[v].add(u)
    return graph


schritte_bluete = []
for k in range(1, 11):
    zaehler['schritte'] = 0
    mit = ist_plg(bluete(k)) is not None
    s_mit = zaehler['schritte']
    zaehler['schritte'] = 0
    ohne = ist_plg(bluete(k, False)) is not None
    schritte_bluete.append((k, s_mit, zaehler['schritte']))
    print(f'k = {k:2}: mit Stern PLG {mit!s:5} {s_mit:6} Schritte, ohne Stern PLG {ohne!s:5} {zaehler["schritte"]:3} Schritte')
````

Ohne Stern ist die Blüte ein PLG, und die Suche ist schnell. Mit Stern wächst die Zahl der Schritte mit jedem Blatt auf mehr als das
Doppelte: Die Suche probiert im Wesentlichen alle Teilmengen der Blätter aus, die sie vor dem Weg zum Stern legen könnte. Unsere
Suche ist also im schlechtesten Fall **exponentiell**. Die Hinweise halten so etwas für akzeptabel, wenn man es wie hier mit Regeln
und gemerkten Zuständen eindämmt.

### Ein Schnelltest ohne Suche: Loch und asteroidales Tripel

Die Hinweise nennen einen Satz von Lekkerkerker und Boland (1962): Ein Graph ist genau dann ein PLG, wenn er **kein Loch** und **kein
asteroidales Tripel** hat. Drei Knoten a, b, c bilden ein asteroidales Tripel, wenn man von jedem der drei zu jedem anderen auf einem
Weg kommt, der den dritten und dessen Nachbarn meidet. Beim Stern sind das die drei äußeren Knoten: Von einem zum anderen geht man
über die Mitte, und die ist kein Nachbar des dritten. In einem PLG geht das nicht: Von den drei Spannen liegt eine zwischen den
anderen beiden, und jeder Weg von links nach rechts muss eine Spanne benutzen, die diese mittlere überlappt.

Den Test bauen wir mit einer **Tiefensuche**: Für jeden Knoten c zerlegen wir den Graphen ohne c und seine Nachbarn in
**Zusammenhangskomponenten** (Teile, in denen man von jedem Knoten zu jedem anderen kommt). a und b sind „verbunden ohne c“, wenn sie
in derselben Komponente liegen. Das braucht keine Suche mit Zurückgehen: höchstens `n` Tiefensuchen und `n³` Tripel.

````python
def komponenten_ohne(graph, c):
    """Zusammenhangskomponenten des Graphen ohne c und ohne die Nachbarn von c: Dictionary Knoten -> Nummer."""
    weg = graph[c] | {c}
    nummer = {}
    for start in sorted(graph):
        if start in weg or start in nummer:
            continue
        nummer[start] = start                 # die Komponente bekommt die Nummer ihres ersten Knotens
        stapel = [start]
        while len(stapel) > 0:
            x = stapel.pop()
            for y in graph[x]:
                if y not in weg and y not in nummer:
                    nummer[y] = start
                    stapel.append(y)
    return nummer


def verbunden_ohne(komp, c, a, b):
    """Kommt man von a nach b, ohne c oder einen Nachbarn von c zu benutzen?"""
    return a in komp[c] and b in komp[c] and komp[c][a] == komp[c][b]


def asteroidales_tripel(graph):
    komp = {}
    for c in graph:
        komp[c] = komponenten_ohne(graph, c)
    for a, b, c in combinations(sorted(graph), 3):
        if verbunden_ohne(komp, c, a, b) and verbunden_ohne(komp, a, b, c) and verbunden_ohne(komp, b, a, c):
            return (a, b, c)
    return None


def ist_plg_schnell(graph):
    """Ja/Nein ohne Spannen: kein Loch (Eliminationsreihenfolge klappt) und kein asteroidales Tripel."""
    return eliminationsreihenfolge(graph) is not None and asteroidales_tripel(graph) is None


for name in ['stern', 'netz', 'zelt', 'zufall_kein_plg']:
    print(name, asteroidales_tripel(graphen[name]))
print('Blüte mit 10 Blättern und Stern:', asteroidales_tripel(bluete(10)), ist_plg_schnell(bluete(10)))
````

Beim Stern sind 0, 3 und 5 die äußeren Knoten (A, D und F aus Abbildung 6 der Hinweise), beim Netz die drei angehängten Knoten
3, 4, 5, beim Zelt die drei äußeren Knoten. Die Blüte mit Stern wird ohne jede Suche als „kein PLG“ erkannt.

Stimmen beide Tests immer überein? Wir probieren **alle** Graphen mit bis zu sechs Knoten 0 bis 5 aus: Bei sechs Knoten gibt es 15
mögliche Kanten, also `2¹⁵ = 32 768` Graphen. Nebenbei zählen wir, wie viele kein Loch haben und trotzdem kein PLG sind.

````python
gleich = 0
ohne_loch_kein_plg = {}
for n in range(1, 7):
    paare = list(combinations(range(n), 2))
    ohne_loch_kein_plg[n] = 0
    for maske in range(2 ** len(paare)):          # jedes Bit der Zahl maske sagt, ob eine Kante da ist
        g = {}
        for v in range(n):
            g[v] = set()
        for i in range(len(paare)):
            if (maske >> i) % 2 == 1:
                u, v = paare[i]
                g[u].add(v)
                g[v].add(u)
        plg = ist_plg(g) is not None
        if plg == ist_plg_schnell(g):
            gleich += 1
        if not plg and eliminationsreihenfolge(g) is not None:
            ohne_loch_kein_plg[n] += 1
print('übereinstimmend:', gleich)
print('ohne Loch, aber kein PLG:', ohne_loch_kein_plg)
````

Alle 1 + 2 + 8 + 64 + 1024 + 32 768 = 33 867 Graphen bekommen von beiden Tests dieselbe Antwort. Bis fünf Knoten ist jeder Graph
ohne Loch ein PLG. Bei sechs Knoten gibt es 240 Ausnahmen: genau die Netze und Zelte, jeweils 120 Nummerierungen. Deshalb können
die Netze und Zelte nicht kleiner sein, und deshalb hätte es bei `zufall_kein_plg` (ohne Loch) gereicht, die Suche nach der
kleinsten Teilmenge bei sechs Knoten zu beginnen.

Der Schnelltest sagt nur Ja oder Nein. Für die Spannen brauchen wir weiterhin die Cliquenkette. Man kann aber beides verbinden:
erst `ist_plg_schnell` (verhindert die langen Sackgassen der Blüte), und nur bei Ja die Cliquensuche. Ob unsere Suche auch bei
einem PLG in lange Sackgassen laufen kann, haben wir nicht bewiesen; bei allen Beispielen und Zufallsgraphen hier lief sie ohne
großes Zurückgehen.

### Das Verfahren der Lösungshinweise

Die Hinweise beschreiben ein Verfahren aus der Fachliteratur (Habib, McConnell, Paul und Viennot, 2000), das bei geschickter
Umsetzung **lineare Laufzeit** hat, also nur proportional zur Zahl der Knoten und Kanten wächst:

1. Eine **lexikografische Breitensuche** (Lex-BFS) liefert eine Reihenfolge der Knoten. Rückwärts gelesen ist sie bei Graphen ohne
   Loch eine Eliminationsreihenfolge wie unsere (perfekte Eliminationsanordnung); geprüft wird das mit Teilmengen-Tests.
2. Daraus entstehen die maximalen Cliquen und ein **Cliquenbaum**.
3. Durch fortgesetztes Aufteilen einer Liste von Cliquenmengen (*Partitionsverfeinerung*, mit Knoten als „Pivots“) wird eine
   Cliquenkette gebaut, ganz ohne Zurückgehen.

Für Teil 2 probieren die Hinweise alle Teilmengen der Größe nach, wie unsere `kleinste_teilmenge`, und nennen dafür `O(2^|V|)`.
Unser Streichverfahren liefert dagegen mit nur `n + 1` PLG-Tests eine minimale (nicht unbedingt kleinste) Teilmenge.

## Zum Schluss: Lebenslinien mit MiniZinc

Teil 1 lässt sich fast wörtlich als MiniZinc-Modell schreiben: Für jeden Knoten sind Anfang `von[v]` und Ende `bis[v]` seiner
Spanne Variablen, und für jedes Paar von Knoten gilt „Kante genau dann, wenn Überlappung“. Neu gegenüber den Arbeitsblättern ist nur
`<->`, „genau dann, wenn“: Beide Seiten sind entweder wahr oder beide falsch.

Welche Werte brauchen die Variablen? Die Plätze unserer Cliquenkette reichen: Ein PLG hat höchstens `n` maximale Cliquen (nach Schritt 3
ist jede die Clique eines Knotens). Deshalb genügen die Zahlen 1 bis `n`.

```minizinc
int: n;                                        % Knoten 1..n (im Python-Programm 0..n-1)
array[1..n, 1..n] of 0..1: kante;              % Adjazenzmatrix

array[1..n] of var 1..n: von;                  % Lebensspanne von Knoten v: [von[v], bis[v]]
array[1..n] of var 1..n: bis;

constraint forall(v in 1..n)(von[v] <= bis[v]);

constraint forall(u, v in 1..n where u < v)(   % Kante genau dann, wenn Überlappung
    (kante[u, v] = 1) <-> (von[u] <= bis[v] /\ von[v] <= bis[u]));

solve satisfy;

output ["Knoten \(v - 1): \(von[v]) bis \(bis[v])\n" | v in 1..n];
```

Die Datendatei schreibt `als_dzn`:

````python
def als_dzn(graph):
    n = len(graph)
    zeilen = []
    for i in range(n):
        werte = []
        for j in range(n):
            if j in graph[i]:
                werte.append('1')
            else:
                werte.append('0')
        zeilen.append('| ' + ', '.join(werte))
    return f'n = {n};\nkante = [' + '\n         '.join(zeilen) + ' |];\n'


print(als_dzn(graphen['beispiel5']))
````

`=====UNSATISFIABLE=====` heißt: Es gibt keine Lösung, der Graph ist kein PLG.

Beim Erstellen dieses Notebooks haben wir alle zwölf Graphen und die Blüte mit zehn Blättern (mit Stern) so gerechnet. MiniZinc
antwortet jedes Mal wie unser Python-Programm, und die Spannen von MiniZinc ergeben mit `lebensgraph` genau den Eingabegraphen.
Die Rechenzeiten (Gecode, Chuffed und CP-SAT):

- **Alle zwölf Graphen** schafft jeder der drei Solver in unter einer Sekunde, ebenso einen Kreis aus 30 Knoten.
- **Größere Graphen:** 100 zufällige Spannen (ein PLG) dauern 1 bis 4 Sekunden. 60 zufällige Spannen mit einer zusätzlichen Kante
  (kein PLG) beweisen Chuffed und CP-SAT in etwa einer Sekunde als unlösbar, Gecode braucht etwa 13 Sekunden.
- **Die Blüte mit Stern** ist auch für MiniZinc schwer, aber nicht für jeden Solver. Chuffed beweist selbst bei 10 und 12
  Blättern in unter einer Sekunde, dass es keine Lösung gibt. Gecode braucht bei 6, 7 und 8 Blättern etwa 0,4, 1,6 und 23 Sekunden,
  CP-SAT bei 6 und 7 Blättern etwa 3 und 42 Sekunden; ab 9 Blättern (Gecode) bzw. 8 Blättern (CP-SAT) sind sie nach einer Minute
  nicht fertig. Gecode und CP-SAT probieren hier wie unsere Suche viele Anordnungen der Blätter durch. Chuffed lernt aus
  Widersprüchen: Er merkt sich, *warum* ein Versuch gescheitert ist, und schließt damit gleich viele ähnliche Versuche aus.

Teil 2 passt dagegen nicht gut zu MiniZinc. Eine Teilmenge, die *kein* PLG ist, bedeutet: Für **alle** möglichen Spannen gibt es
einen Widerspruch. „Für alle Belegungen“ kann man in einem Modell nicht als Bedingung schreiben; der Solver findet Belegungen, die
passen, oder beweist, dass es keine gibt. Man könnte MiniZinc nur Teilmengen einzeln prüfen lassen, wie in unserem Streichverfahren.

**Fazit:** Für die Frage „PLG oder nicht?“ ist das MiniZinc-Modell kurz und bei unseren Graphen schnell, auch bei denen, an denen
unsere Suche lange arbeitet. Die eigentliche Idee der Aufgabe, warum etwas kein PLG ist (Löcher, Sterne, Netze), liefert es aber
nicht; dafür brauchen wir Teil 2 in Python.
