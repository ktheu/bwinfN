### Laubmaschen – Lösung Schritt für Schritt

*42. Bundeswettbewerb Informatik (2023/2024), 2. Runde, Aufgabe 1*

Der Herbst ist da, und der Hausmeister der Schule hat seinen getreuen Laubbläser hervorgeholt. In den vergangenen Jahren hatte sein
Gebläse ungefähr denselben Gesamteffekt auf das Laub wie das Fußballspiel auf den Ball; aber damit soll jetzt Schluss sein: Er will
ganz wissenschaftlich vorgehen und hat den quadratischen Schulhof in Planquadrate eingeteilt. Er möchte sich immer in der Mitte eines
Planquadrats aufstellen und in Richtung des Nachbarquadrats A nördlich, südlich, östlich oder westlich davon blasen. Am liebsten hätte
er, wenn das gesamte Laub von Quadrat A auf das dahinter liegende Quadrat B geblasen würde; aber in Wirklichkeit landen jeweils
ungefähr 10 % davon auf den Quadraten rechts und links von B. Und nicht genug damit: Ungefähr 10 % des Laubes, das vorher auf B lag,
werden auf das Quadrat noch hinter B weitergeblasen.

In Wirklichkeit besteht Laub aus einzelnen Blättern, und „ungefähr 10 %“ ist so zu verstehen, dass jedes Blatt mit Wahrscheinlichkeit
10 % „beschließt“, das entsprechende Feld aufzusuchen.

Das Ziel des Hausmeisters ist, das am Anfang noch überall gleichmäßig liegende Laub vom gesamten Schulhof auf ein einziges
Planquadrat Q zu konzentrieren, damit es von dort abgeholt werden kann. Q darf kein Randquadrat sein, weil dort das
Abholungsfahrzeug nicht genug Platz hätte. Wegen der „Seitenverluste“ des Laubbläsers ist sich der Hausmeister nicht sicher, ob es
ihm gelingen wird, das gesamte Laub auf ein Planquadrat zu sammeln; aber dann will er zumindest soviel Laub wie möglich mit dem
Laubbläser auf ein Quadrat bringen, bevor er den Rest mit dem Besen erledigt.

(Auf dem Aufgabenblatt steht dazu ein Bild: Der Hausmeister steht auf einem Quadrat und bläst über A nach B. Von den Blättern auf A
landen 80 % auf B und je 10 % links und rechts von B. Von den Blättern auf B landen 10 % auf dem Quadrat hinter B.)

#### Aufgabe

Hilf dem Hausmeister: Schreibe zuerst ein Programm, mit dem man den Prozess des Laubblasens simulieren kann. Beachte dabei, dass die
oben genannten Regeln für die Wirkung des Blasevorganges am Rand und in den Ecken des Schulhofes nicht unverändert anwendbar sind.
Überlege dir für diese Fälle sinnvolle Ergänzungen der Regeln und verwende diese in deiner Simulation.

Entwickle eine Strategie für den Hausmeister: Die Strategie entscheidet vor jedem Blasevorgang, auf welches Feld er sich stellen und in
welche Richtung er von dort jeweils blasen soll, damit schließlich auf einem gegebenen Zielquadrat, das kein Rand- oder Eckquadrat
ist, möglichst viel Laub versammelt ist.

Lasse die Simulation mit deiner Strategie unter anderem für einen Schulhof aus 5 × 5 Planquadraten laufen, mit anfangs 100 Blättern
auf jedem Planquadrat.

#### Beispieldaten

Es gibt keine Beispieldateien. Der Schulhof ist einfach eine Tabelle mit der Zahl der Blätter pro Quadrat.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**

- **Zufällige Simulation:** Jedes Blatt würfelt für sich, wohin es fliegt. Wiederholt man den Versuch mit anderen Zufallszahlen,
  schwanken die Ergebnisse ein wenig. Man mittelt deshalb über viele Läufe.
- **Erwartungswert:** Statt zu würfeln kann man mit den Wahrscheinlichkeiten rechnen: 100 Blätter mit 10 % Wahrscheinlichkeit
  ergeben im Mittel 10 Blätter. Das ist deterministisch und schnell, man bekommt aber Kommazahlen. Zum Entwickeln einer Strategie
  ist das ideal, zum Prüfen nimmt man dann die richtige Simulation.
- **Übergangstabelle:** Ein Blasvorgang ist nur eine Liste „Blätter von diesem Feld gehen mit dieser Wahrscheinlichkeit auf jenes
  Feld“. Simulation und Erwartungswert benutzen dieselbe Tabelle, so bleiben beide gleich.
- **Gieriges Verfahren und seine Grenzen:** Man wählt immer den Blasvorgang, der eine Bewertung am meisten verbessert. Bleibt man
  stecken, muss man die Reihenfolge der Blasvorgänge **planen** (Phasen).
- **Abbruchbedingung:** Wann hört der Hausmeister auf? Wenn ein weiterer Durchgang nur noch wenig bringt.

**Nützliche Bibliotheken und Funktionen**

- `random.Random()` mit `random()` und `seed(zahl)` für Zufallszahlen, die bei jedem Lauf gleich sind
- Listen von Listen (Tabelle) für den Schulhof, Tupel `(zeile, spalte)` für Felder

## Idee in Worten

1. **Regeln.** Ein Blasvorgang wird durch Stellung `S`, Richtung und die Felder `A` (davor), `B` (dahinter), `C` (hinter `B`) und die
   Seitenfelder von `B` beschrieben. Am Rand ragen manche dieser Felder über den Schulhof hinaus. Wir legen fest, wohin die Blätter
   dann fliegen, und zwar so, dass **kein Blatt den Schulhof verlässt**.
2. **Simulation** und **Erwartungswert** aus derselben Tabelle.
3. **Einfacher Versuch:** Immer den Blasvorgang wählen, der die Blätter im Mittel am nächsten ans Ziel bringt. Das bleibt stecken.
4. **Besser mit Plan.** Drei Beobachtungen:
   - Blätter am **Rand** kommen nur ins Innere, wenn man **am Rand entlang** bläst (dann fliegen 10 % nach innen). Man umrundet den
     Rand deshalb mehrmals.
   - Im Inneren kann man **ganze Reihen** nach unten oder oben zum Feld `V` unter dem Ziel blasen.
   - Liegt viel Laub auf `Q`, **verliert** ein direkter Blasvorgang auf `Q` 10 % von `Q`. Dann hilft **Blasen mit Abstand**: Man
     bläst über ein leeres Feld hinweg auf `V`, dann wandern 10 % von `V` ohne Seitenverlust auf `Q`.

## Schritt 1: Die Regeln für einen Blasvorgang

Felder schreiben wir als `(zeile, spalte)`, oben links ist `(0, 0)`. Die Richtung ist ein Schritt, zum Beispiel `UNTEN = (1, 0)`.
Der Hausmeister steht auf `S`, `A = S + Richtung`, `B = S + 2·Richtung`, `C = S + 3·Richtung`. Die beiden Seitenfelder von `B`
sind `B ± q` mit `q` senkrecht zur Richtung.

**Ohne Rand** (laut Aufgabe): Blätter auf `A` gehen zu 80 % auf `B` und zu je 10 % auf die Seitenfelder von `B`. Blätter auf `B`
gehen zu 10 % auf `C` und bleiben zu 90 % liegen.

**Am Rand** legen wir fest (dieselben Festlegungen wie in den offiziellen Lösungshinweisen, die Blätter bleiben im Schulhof):

- Der Hausmeister steht im Schulhof, und `A` liegt im Schulhof. Sonst ist der Blasvorgang **ungültig**.
- **Seite:** Liegt nur ein Seitenfeld von `B` außerhalb, prallen die Blätter an der Wand ab und landen auf `B`.
- **Wand:** Liegt `B` außerhalb, werden die Blätter von `A` zur Seite geweht, je zur Hälfte nach links und rechts von `A`.
- **Ecke:** Liegt `B` außerhalb und auch ein Seitenfeld von `A`, fliegen 90 % der Blätter von `A` zum anderen Seitenfeld, 10 %
  bleiben in der Ecke hängen.
- Liegt `C` außerhalb, bleiben die Blätter von `B`, die weitergeblasen würden, auf `B`.

Wir schreiben eine Funktion `uebergaenge`, die für einen Blasvorgang eine Liste liefert: pro Quellfeld die Liste der Zielfelder
mit Wahrscheinlichkeit.

```python
OBEN, UNTEN, LINKS, RECHTS = (-1, 0), (1, 0), (0, -1), (0, 1)
RICHTUNGEN = [OBEN, UNTEN, LINKS, RECHTS]


def innen(n, feld):
    return 0 <= feld[0] < n and 0 <= feld[1] < n


def plus(feld, richtung, mal=1):
    return (feld[0] + mal * richtung[0], feld[1] + mal * richtung[1])


def uebergaenge(n, stand, richtung):
    """Liste von (Quellfeld, [(Zielfeld, Wahrscheinlichkeit), ...]); None, wenn der Blasvorgang ungültig ist."""
    a = plus(stand, richtung)
    if not innen(n, stand) or not innen(n, a):
        return None
    b = plus(stand, richtung, 2)
    c = plus(stand, richtung, 3)
    quer = (richtung[1], richtung[0])
    ergebnis = []
    if innen(n, b):
        ziele = [(b, 0.8)]
        for vorzeichen in (1, -1):
            seite = plus(b, quer, vorzeichen)
            if innen(n, seite):
                ziele.append((seite, 0.1))
            else:
                ziele.append((b, 0.1))      # Seite: prallt ab und landet auf B
        ergebnis.append((a, ziele))
        if innen(n, c):
            ergebnis.append((b, [(c, 0.1), (b, 0.9)]))
        else:
            ergebnis.append((b, [(b, 1.0)]))
    else:
        links = plus(a, quer, 1)
        rechts = plus(a, quer, -1)
        if innen(n, links) and innen(n, rechts):
            ergebnis.append((a, [(links, 0.5), (rechts, 0.5)]))                 # Wand
        elif innen(n, links):
            ergebnis.append((a, [(links, 0.9), (a, 0.1)]))                      # Ecke
        else:
            ergebnis.append((a, [(rechts, 0.9), (a, 0.1)]))                     # Ecke
    return ergebnis


def zeige_uebergaenge(n, stand, richtung):
    print('Stand', stand, 'Richtung', richtung)
    for quelle, ziele in uebergaenge(n, stand, richtung):
        text = ', '.join(str(ziel) + ': ' + str(round(w * 100)) + ' %' for ziel, w in ziele)
        print('  Blätter auf', quelle, '->', text)


print('Mitten im Hof (7x7):')
zeige_uebergaenge(7, (2, 3), UNTEN)
print('Seite: entlang der oberen Wand (5x5):')
zeige_uebergaenge(5, (0, 1), RECHTS)
print('Wand: gegen die untere Wand (5x5):')
zeige_uebergaenge(5, (3, 2), UNTEN)
print('Ecke: in die linke obere Ecke (5x5):')
zeige_uebergaenge(5, (1, 0), OBEN)
print('Außerhalb ist ungültig:', uebergaenge(5, (4, 2), UNTEN))
```

## Schritt 2: Simulation und Erwartungswert

Der Schulhof `h` ist eine Tabelle: `h[zeile][spalte]` ist die Zahl der Blätter. `blase` führt einen Blasvorgang aus. Zuerst werden
die Blätter aller Quellfelder gezählt und die Quellfelder geleert (sonst würden Blätter, die gerade auf `B` gelandet sind, gleich
noch einmal weiterfliegen). Dann werden die Blätter verteilt:

- bei `simulation = True` würfelt **jedes Blatt einzeln** mit einer Zufallszahl zwischen 0 und 1; je nach Zufallszahl landet es
  auf dem ersten, zweiten oder dritten Zielfeld,
- bei `simulation = False` bekommt jedes Zielfeld den Anteil der Blätter, der der Wahrscheinlichkeit entspricht (**Erwartungswert**,
  das sind Kommazahlen).

Die Zufallszahlen kommen aus dem Zufallsgenerator `zufall`, den wir mit `zufall.seed(zahl)` auf einen festen Anfang setzen können.
Die Funktion gibt `True` zurück, wenn der Blasvorgang gültig war.

```python
import random

zufall = random.Random(1)


def neuer_hof(n, blaetter=100):
    return [[blaetter] * n for _ in range(n)]


def blase(h, stand, richtung, simulation):
    tabelle = uebergaenge(len(h), stand, richtung)
    if tabelle is None:
        return False
    mengen = []
    for quelle, ziele in tabelle:
        mengen.append(h[quelle[0]][quelle[1]])
        h[quelle[0]][quelle[1]] = 0
    for i in range(len(tabelle)):
        ziele = tabelle[i][1]
        if simulation:
            for blatt in range(mengen[i]):
                wurf = zufall.random()
                summe = 0.0
                for ziel, w in ziele:
                    summe += w
                    if wurf < summe:
                        break
                h[ziel[0]][ziel[1]] += 1
        else:
            for ziel, w in ziele:
                h[ziel[0]][ziel[1]] += mengen[i] * w
    return True


def zeige(h):
    for zeile in h:
        print(' '.join(str(round(x)).rjust(5) for x in zeile))


def gesamt(h):
    return sum(sum(zeile) for zeile in h)


# Beispiel von Abbildung 1.8 der Lösungshinweise: 10 Blätter auf A, 100 auf B (Hof 7x7, Stand (0,3), blasen nach unten)
h = [[0] * 7 for _ in range(7)]
h[1][3] = 10
h[2][3] = 100
blase(h, (0, 3), UNTEN, False)
print('Erwartungswert:')
zeige(h[1:4])
```

Auf `B` bleiben 90 der alten Blätter, dazu kommen 8 von `A`: 98 Blätter, links und rechts je 1, und 10 Blätter wandern nach `C`.
Genau so sieht es auch in den Lösungshinweisen aus. Der direkte Blasvorgang **verliert** hier Blätter auf `B` (100 → 98), obwohl er
10 Blätter heranbringt. Das merken wir uns für später.

Zur Kontrolle der Simulation: 1000-mal dasselbe würfeln, der Mittelwert muss nahe am Erwartungswert liegen, und die Gesamtzahl der
Blätter darf sich nie ändern.

```python
summen = [[0.0] * 7 for _ in range(7)]
laeufe = 1000
for versuch in range(laeufe):
    hs = [[0] * 7 for _ in range(7)]
    hs[1][3] = 10
    hs[2][3] = 100
    blase(hs, (0, 3), UNTEN, True)
    assert gesamt(hs) == 110
    for i in range(7):
        for j in range(7):
            summen[i][j] += hs[i][j] / laeufe
print('Mittelwert über', laeufe, 'Würfe:')
zeige(summen[1:4])
abweichung = max(abs(summen[i][j] - h[i][j]) for i in range(7) for j in range(7))
print('größte Abweichung zum Erwartungswert:', round(abweichung, 2), 'Blätter')
assert abweichung < 1.0
```

## Schritt 3: Ein einfacher Versuch, der steckenbleibt

Wir bewerten einen Schulhof mit einer Zahl `phi`: Blätter nahe am Ziel zählen viel, Blätter weit weg wenig, nämlich
`Blätter · g^Abstand` (Abstand = Manhattan-Abstand, also Zeilenabstand plus Spaltenabstand, mit `g = 0,5`). Das gierige Verfahren
probiert **alle** gültigen Blasvorgänge im Erwartungswert, nimmt den mit dem größten Gewinn für `phi` und macht weiter, bis kein
Blasvorgang `phi` mehr erhöht.

Wir testen das auf dem 5 × 5-Hof mit dem Ziel `(1, 2)` (das Feld oberhalb der Mitte, warum, sehen wir gleich).

```python
def phi(h, ziel, g):
    summe = 0.0
    for i in range(len(h)):
        for j in range(len(h)):
            summe += h[i][j] * g ** (abs(i - ziel[0]) + abs(j - ziel[1]))
    return summe


def gierig(h, ziel, g=0.5):
    """Wählt immer den besten Blasvorgang (im Erwartungswert), bis keiner mehr hilft. Gibt die Zahl der Blasvorgänge zurück."""
    n = len(h)
    anzahl = 0
    while anzahl < 2000:
        bestes = phi(h, ziel, g)
        beste_wahl = None
        for i in range(n):
            for j in range(n):
                for richtung in RICHTUNGEN:
                    probe = [zeile[:] for zeile in h]
                    if blase(probe, (i, j), richtung, False):
                        wert = phi(probe, ziel, g)
                        if wert > bestes + 1e-9:
                            bestes = wert
                            beste_wahl = ((i, j), richtung)
        if beste_wahl is None:
            return anzahl
        blase(h, beste_wahl[0], beste_wahl[1], False)
        anzahl += 1
    return anzahl


ziel = (1, 2)
h_gierig = neuer_hof(5)
anzahl_gierig = gierig(h_gierig, ziel)
print(anzahl_gierig, 'Blasvorgänge, dann hilft keiner mehr:')
zeige(h_gierig)
anteil_gierig = h_gierig[1][2] / gesamt(h_gierig)
print('Auf dem Ziel liegen', round(100 * anteil_gierig), '% aller Blätter')
```

Das gierige Verfahren sammelt schlecht. Die Blätter stauen sich in der Mitte jeder Seite am Rand. Das Feld `(0, 2)` liegt direkt
über dem Ziel, aber von dort kann man nicht nach unten blasen, weil der Hausmeister dafür außerhalb des Hofes stehen müsste. Am
Rand entlang zu blasen führt vom Ziel weg und senkt `phi`. Wir brauchen einen **Plan** mit mehreren Phasen, bei dem einzelne
Blasvorgänge auch einmal vorübergehend schlechter aussehen dürfen (zum Beispiel erst am Rand entlang blasen, um 10 % der Blätter
nach innen zu schicken).

## Schritt 4: Die Bausteine der Strategie

Das Ziel `Q = (r, c)` liegt im Inneren, und das Feld `V = (r+1, c)` unter dem Ziel ist ebenfalls noch nicht am Rand. Zum Ziel
gehören vier Bausteine, die jeweils die Zahl ihrer Blasvorgänge zurückgeben.

**Baustein 1: Rand umrunden.** Der Hausmeister geht den Rand im Uhrzeigersinn ab und bläst auf jedem Randfeld zum nächsten Randfeld.
Dabei wandern die Blätter am Rand entlang, und auf jedem Feld fliegen 10 % nach innen (Fall „Seite“). Mehrere Runden leeren den
Rand immer weiter.

**Baustein 2: In Reihen sammeln.** Von oben her wird Reihe für Reihe (außer dem Ziel selbst) nach unten geblasen, so dass alle Blätter
zur Reihe `r+1` wandern. Von unten her wird genauso nach oben geblasen. In Reihe `r+1` bläst man schließlich von links und rechts
zur Spalte `c`. Fast das ganze Laub aus dem Inneren liegt dann (im Mittel) auf `V` und `Q`.

**Baustein 3: Direkt aufs Ziel blasen,** wenn es sich lohnt: Vom Feld unter `V` aus nach oben, dann ist `V` das Feld `A` und `Q` das
Feld `B`. Es kommen 80 % von `V` hinzu, aber 10 % von `Q` gehen verloren. Wir rechnen es nur dann, wenn
`0,8 · V > 0,2 · Q` ist (die 10 % Verlust fliegen auf das Feld über dem Ziel und müssen von dort wieder geholt werden, das berücksichtigt
der Faktor 0,2 statt 0,1).

**Baustein 4: Mit Abstand blasen.** Der Hausmeister steht auf dem Feld drei unter `Q`, bläst nach oben: `A` ist das Feld
unter `V` (dort liegen nach dem Sammeln nur wenige Blätter), `B = V`, `C = Q`. Von `V` fliegen 10 % direkt auf `Q`, ohne Seitenverlust und ohne dass `Q` etwas verliert, denn `Q` ist
nur `C`. Das wird wiederholt, bis auf `V` höchstens 10 % der Anfangsmenge liegen. Das geht nur, wenn dieses Feld im Schulhof liegt,
also bei `r + 3 ≤ n − 1`. Beim 5 × 5-Hof geht das deshalb mit `(1, 2)`, aber nicht mit dem Mittelfeld `(2, 2)`.

```python
def rand_felder(n):
    """Alle Randfelder im Uhrzeigersinn, beginnend oben links."""
    felder = []
    for j in range(n):
        felder.append((0, j))
    for i in range(1, n):
        felder.append((i, n - 1))
    for j in range(n - 2, -1, -1):
        felder.append((n - 1, j))
    for i in range(n - 2, 0, -1):
        felder.append((i, 0))
    return felder


def blatt_am_rand(h):
    n = len(h)
    return sum(h[i][j] for (i, j) in rand_felder(n))


def rand_umrunden(h, simulation):
    felder = rand_felder(len(h))
    anzahl = 0
    for k in range(len(felder)):
        von = felder[k]
        nach = felder[(k + 1) % len(felder)]
        blase(h, von, (nach[0] - von[0], nach[1] - von[1]), simulation)
        anzahl += 1
    return anzahl


def in_reihen_sammeln(h, ziel, simulation):
    n = len(h)
    zeile, spalte = ziel
    anzahl = 0
    # von oben bis zur Reihe des Ziels nach unten blasen (das Ziel selbst nicht)
    for i in range(1, zeile + 1):
        for j in range(n):
            if (i, j) != ziel:
                blase(h, (i - 1, j), UNTEN, simulation)
                anzahl += 1
    # von unten bis zur Reihe unter dem Ziel nach oben blasen
    for i in range(n - 2, zeile + 1, -1):
        for j in range(n):
            blase(h, (i + 1, j), OBEN, simulation)
            anzahl += 1
    # in der Reihe unter dem Ziel von links und rechts zur Spalte des Ziels
    for j in range(1, spalte):
        blase(h, (zeile + 1, j - 1), RECHTS, simulation)
        anzahl += 1
    for j in range(n - 2, spalte, -1):
        blase(h, (zeile + 1, j + 1), LINKS, simulation)
        anzahl += 1
    return anzahl


def direkt_aufs_ziel(h, ziel, simulation):
    blase(h, (ziel[0] + 2, ziel[1]), OBEN, simulation)
    return 1


def mit_abstand(h, ziel, simulation):
    """Bläst mit Abstand von V auf das Ziel, bis auf V höchstens 10 % der Anfangsmenge liegen."""
    zeile, spalte = ziel
    if zeile + 3 > len(h) - 1:
        return 0
    start = h[zeile + 1][spalte]
    anzahl = 0
    while h[zeile + 1][spalte] > 0.1 * start and anzahl < 100:
        blase(h, (zeile + 3, spalte), OBEN, simulation)
        anzahl += 1
    return anzahl


# Die Bausteine nacheinander im Erwartungswert auf dem 5x5-Hof mit Ziel (1, 2)
ziel = (1, 2)
h = neuer_hof(5)
print('Start:')
zeige(h)
runden = 0
while blatt_am_rand(h) > 0.05 * gesamt(h) and runden < 100:
    rand_umrunden(h, False)
    runden += 1
print('Nach', runden, 'Randrunden:')
zeige(h)
in_reihen_sammeln(h, ziel, False)
print('Nach dem Sammeln in Reihen:')
zeige(h)
direkt_aufs_ziel(h, ziel, False)
print('Nach einem direkten Blasvorgang aufs Ziel:')
zeige(h)
mit_abstand(h, ziel, False)
print('Nach dem Blasen mit Abstand:')
zeige(h)
```

Eine einzige Folge der Bausteine bringt knapp 65 % aufs Ziel (1610 von 2500). Das Blasen mit Abstand tut hier noch nichts, denn
`V` ist nach dem direkten Blasvorgang leer. Auf den Seitenfeldern von `Q` und `V` sowie am Rand bleibt noch viel liegen, weil
beim Sammeln immer wieder Blätter zur Seite fliegen. Wir wiederholen deshalb **Rand, Sammeln, Direkt, Abstand** so lange, bis eine Runde
dem Ziel weniger als 0,2 % aller Blätter mehr bringt. Der Rand wird in jeder Runde auf höchstens 5 % der Blätter geleert.

## Schritt 5: Die ganze Strategie

Die Funktion `strategie` führt die Bausteine aus und gibt die Zahl der Blasvorgänge und die Zahl der Runden zurück. Die
Parameter sind `rand_anteil` (wie leer der Rand werden soll) und `min_gewinn` (Abbruch, wenn eine Runde weniger bringt). Die
Entscheidungen (zum Beispiel, ob direkt geblasen wird) richten sich nach den **tatsächlichen** Blättern, nicht nach dem Plan.

```python
def strategie(h, ziel, simulation, rand_anteil=0.05, min_gewinn=0.002):
    """Wendet die Strategie auf den Hof h an. Gibt (Zahl der Blasvorgänge, Zahl der Runden) zurück."""
    n = len(h)
    zeile, spalte = ziel
    alle = gesamt(h)
    anzahl = 0
    runden = 0
    while runden < 40:
        runden += 1
        vorher = h[zeile][spalte]
        laps = 0
        while blatt_am_rand(h) > rand_anteil * alle and laps < 100:
            anzahl += rand_umrunden(h, simulation)
            laps += 1
        anzahl += in_reihen_sammeln(h, ziel, simulation)
        if 0.8 * h[zeile + 1][spalte] > 0.2 * h[zeile][spalte] + 1:
            anzahl += direkt_aufs_ziel(h, ziel, simulation)
        anzahl += mit_abstand(h, ziel, simulation)
        if h[zeile][spalte] - vorher < min_gewinn * alle:
            break
    return anzahl, runden


# Erwartungswert (kein Zufall) für den 5x5-Hof, Ziel (1, 2)
h = neuer_hof(5)
anzahl, runden = strategie(h, (1, 2), False)
print('Erwartungswert:', anzahl, 'Blasvorgänge,', runden, 'Runden')
zeige(h)
erwartung_5 = h[1][2] / gesamt(h)
print('Anteil auf dem Ziel:', round(100 * erwartung_5, 1), '%')
```

Jetzt die richtige Simulation mit einzelnen Blättern: ein Lauf mit festem Startwert und danach der Mittelwert über 100 Läufe.
Bei den größeren Höfen nehmen wir weniger Läufe, damit das Notebook im Browser nicht zu lange rechnet. Als Ziel nehmen wir
immer ein Feld in der Mitte, bei dem das Blasen mit Abstand noch möglich ist.

```python
zufall.seed(2024)
h = neuer_hof(5)
anzahl, runden = strategie(h, (1, 2), True)
print('Ein Lauf auf dem 5x5-Hof:', anzahl, 'Blasvorgänge,', runden, 'Runden')
zeige(h)
assert gesamt(h) == 2500
print('Auf dem Ziel liegen', h[1][2], 'von', gesamt(h), 'Blättern =', round(100 * h[1][2] / gesamt(h), 1), '%')
```

```python
def mittelwert(n, ziel, laeufe, blaetter=100):
    summe_anteil = 0.0
    summe_blasen = 0
    for seed in range(laeufe):
        zufall.seed(seed)
        hof = neuer_hof(n, blaetter)
        blasen, runden = strategie(hof, ziel, True)
        assert gesamt(hof) == n * n * blaetter
        summe_anteil += hof[ziel[0]][ziel[1]] / gesamt(hof)
        summe_blasen += blasen
    return summe_anteil / laeufe, summe_blasen / laeufe


tabelle = {}
for n, ziel, laeufe in [(5, (1, 2), 100), (6, (2, 3), 100), (10, (4, 5), 30)]:
    anteil, blasen = mittelwert(n, ziel, laeufe)
    tabelle[n] = anteil
    print(str(n) + 'x' + str(n), 'Ziel', ziel, ':', round(100 * anteil, 1), '% auf dem Ziel, im Mittel',
          round(blasen), 'Blasvorgänge (' + str(laeufe), 'Läufe)')
```

Die Lösungshinweise geben für ihre phasenbasierte Strategie (Mittel über 100 Läufe) an: 95,6 % beim 5 × 5-Hof, 93,8 % beim
6 × 6-Hof und 95,1 % beim 10 × 10-Hof. Unsere Zahlen liegen beim 5 × 5- und 6 × 6-Hof etwas darüber, beim 10 × 10-Hof (nur 30 Läufe) knapp darunter; die Unterschiede
sind kleiner als ein Prozentpunkt. Die Strategie ist mit der phasenbasierten Strategie der Hinweise verwandt (Rand umrunden, Reihen
sammeln, Blasen mit Abstand), nur wiederholen wir sie nach einem Gewinnkriterium.

**Was bleibt liegen?** Im Lauf oben lagen alle übrigen 122 Blätter am Rand, davon 69 auf dem Feld neben der Ecke `(0, 1)`. Der Rand
wird nur bis auf 5 % geleert. Mehr Blasvorgänge bringen mehr (siehe „Für Neugierige“).

## Andere Ziele auf dem 5 × 5-Hof

Beim 5 × 5-Hof ist `(2, 2)` das mittlere Feld. Dort geht das Blasen mit Abstand nicht, denn der Hausmeister müsste drei Felder
unterhalb des Ziels stehen, also außerhalb des Hofes. Wir lassen die Strategie trotzdem laufen (`mit_abstand` tut dann einfach
nichts) und vergleichen mit den Feldern der Zeile 1, die zum Rand nur ein Feld Abstand haben, aber ein Blasen mit Abstand erlauben.
Unsere Funktionen setzen voraus, dass das Feld unter dem Ziel noch im Inneren liegt (`1 ≤ Zeile ≤ n − 3`); andere Ziele erreicht man
durch Spiegeln oder Drehen des Hofes.

```python
for ziel in [(1, 1), (1, 2), (1, 3), (2, 1), (2, 2), (2, 3)]:
    anteil_z, blasen_z = mittelwert(5, ziel, 30)
    print('5x5 mit Ziel', ziel, ':', round(100 * anteil_z, 1), '% auf dem Ziel, im Mittel', round(blasen_z), 'Blasvorgänge')
    if ziel == (2, 2):
        anteil_mitte = anteil_z
```

Mit Zielen in der mittleren Zeile bleiben 16 bis 19 Prozentpunkte weniger auf dem Ziel liegen als mit denen in Zeile 1. Deshalb
nehmen auch die Hinweise bei 5 × 5 das Feld über der Mitte. (Die Ziele `(2, 1)` und `(2, 3)` wären durch Drehen des Hofes
dasselbe wie `(1, 2)`; unsere Funktionen drehen aber nicht, sondern nehmen das Feld *unter* dem Ziel, deshalb sind sie hier
nicht besser.)

## Zum Schluss: MiniZinc?

Wir haben MiniZinc für diese Aufgabe nicht benutzt. Es gibt hier keinen Eingabewert, den man lösen müsste, sondern einen Zufallsprozess
und eine Folge von mehreren hundert Entscheidungen. Man könnte im Erwartungswert eine feste Folge von Blasvorgängen suchen, die den
Anteil auf `Q` maximiert. Dann wären die Unbekannten aber die Blasvorgänge (je Schritt ein Feld und eine Richtung, rund 100
Möglichkeiten) und die Blattzahlen aller 25 Felder nach jedem Schritt sind **Kommazahlen**, die mit den Wahrscheinlichkeiten
multipliziert werden. Die Solver aus den Arbeitsblättern (Gecode, Chuffed, CP-SAT) rechnen mit ganzen Zahlen. Rundet man, ist das
Ergebnis keine echte Erwartung mehr. Dazu kommt, dass die Strategie sich am Zufall ausrichten soll (ob direkt geblasen wird, hängt
von den tatsächlichen Blättern ab). Eine feste Folge, die vorher berechnet wird, kann das nicht.

## Für Neugierige

- **Wie leer soll der Rand werden?** Je kleiner `rand_anteil`, desto besser das Ergebnis, aber desto mehr Blasvorgänge. Im Erwartungswert
  auf dem 5 × 5-Hof (Tabelle unten) steigt der Anteil auf dem Ziel von 92,6 % bei `rand_anteil = 0,1` auf 99,0 % bei 0,01, die Zahl der

- **Der Besen:** Mit dem Laubbläser sind 95 % schnell geschafft; den Rest mit dem Besen auf einen Haufen zu kehren ist meist
  einfacher als noch Hunderte Blasvorgänge.

```python
print('rand_anteil  Anteil auf dem Ziel  Blasvorgänge  (Erwartungswert, 5x5, Ziel (1, 2))')
for rand_anteil in [0.1, 0.05, 0.02, 0.01]:
    hof = neuer_hof(5)
    blasen, runden = strategie(hof, (1, 2), False, rand_anteil=rand_anteil)
    print(str(rand_anteil).rjust(11), str(round(100 * hof[1][2] / gesamt(hof), 1)).rjust(10), '%', str(blasen).rjust(15))
```
