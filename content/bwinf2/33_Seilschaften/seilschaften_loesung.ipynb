### Seilschaften – Lösung Schritt für Schritt

*33. Bundeswettbewerb Informatik (2014/2015), 2. Runde, Aufgabe 1*

Lewis Carrol, bekannt als Autor von „Alice in Wonderland“, beschäftigte sich auch mit Rätseln. Eines davon ist besonders bekannt
geworden:

Eine Königin war mit Sohn und Tochter in einem Turmzimmer gefangen. Die Tür war verschlossen, aber vor dem offenen Fenster gab es
einen Seilzug, an dessen Enden je ein Korb befestigt war. Diesen Mechanismus konnten die drei zur Flucht nutzen. Wer in einem der
gleich schweren Körbe abwärts fahren wollte, musste diesen Korb schwerer beladen als den anderen Korb, aber weniger als 15 Pfd.
schwerer, sonst wäre die Fahrt zu rasant gewesen. Die Königin wog 195 Pfd., die Tochter 105 Pfd. und der Sohn 90 Pfd. Im Turmzimmer
lag außerdem ein Stein, der 75 Pfd. wog. Wie konnten die drei sicher entkommen?

Wir wollen eine Verallgemeinerung dieses Rätsels betrachten:

Es gibt n Personen mit Gewichten p1, …, pn und m Steine mit Gewichten s1, …, sm. Jede Person und jeder Stein befindet sich entweder
oben oder unten. Steine können nur durch anwesende Personen in die Körbe geladen werden. Die Körbe können dann fahren, wenn die Ladung
des oberen Korbs schwerer ist als die des unteren Korbs. Personen fahren sicher, wenn die Gewichte der Ladungen sich höchstens um d
unterscheiden.

Mithilfe des Seilzugs und der Körbe sollen alle Personen sicher entkommen.

#### Aufgabe

Schreibe ein Programm, welches die Anzahl der Personen n, die Anzahl der Steine m, alle Gewichte p1, …, pn, s1, …, sm, die Positionen
der Personen und Steine sowie die Schranke d einlesen kann. Anschließend soll dein Programm feststellen, ob es eine Folge von Fahrten
gibt, mit der alle Personen sicher entkommen können. Ist dies der Fall, soll dein Programm eine kürzeste solche Folge in übersichtlicher
Form ausgeben.

Wende dein Programm auf die unter bundeswettbewerb-informatik.de abgelegten Beispieleingaben an und dokumentiere die Ergebnisse. Unter
den Beispieleingaben befinden sich auch die Daten des originalen Rätsels.

#### Beispieldaten

In der ersten Zeile jeder Datei steht d, danach je Zeile ein Objekt: `P` für eine Person oder `S` für einen Stein, das Gewicht und die
Position, `^` für oben und `_` für unten (das Format haben wir aus den Dateien abgelesen). `seilschaften0.txt` ist das originale Rätsel.
Die Datei `seilschaften6.txt` ist auf der BWINF-Webseite **leer** (0 Bytes). Die Lösungshinweise zeigen aber eine Lösung zu Beispiel 6;
daraus haben wir die Objekte abgelesen und in `eigene_testdaten/seilschaften6.txt` gespeichert. Es sind dieselben wie in Beispiel 5,
nur wiegt die letzte Person 85 statt 98. Das d steht nicht in den Hinweisen; wir nehmen d = 20 wie in Beispiel 5. Das passt: Die
Folge der Hinweise braucht an einer Stelle genau die Differenz 20.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zustandsgraph:** Ein *Zustand* beschreibt, wo gerade jede Person und jeder Stein ist. Eine Fahrt führt von einem Zustand zu einem
  anderen. Die Zustände sind die Knoten eines Graphen, die Fahrten seine Kanten.
- **Breitensuche** mit einer **Warteschlange**: Sie besucht erst alle Zustände, die man mit 1 Fahrt erreicht, dann alle mit 2 Fahrten
  usw. Der erste Zielzustand, den sie findet, hat deshalb eine **kürzeste** Folge von Fahrten.
- **Vorgänger merken:** Zu jedem gefundenen Zustand speichern wir, aus welchem Zustand wir ihn erreicht haben. Vom Ziel aus kann man
  so den Weg rückwärts bis zum Start ablesen.
- Für Neugierige: eine **Invariante** (eine Größe, die sich bei jeder Fahrt auf vorhersehbare Weise ändert) als Beweis, dass es nicht
  kürzer geht.

**Nützliche Funktionen und Bibliotheken**
- Tupel wie `('^', '_', '^')` können Schlüssel eines Dictionarys sein, Listen nicht. Deshalb speichern wir Zustände als Tupel.
- `collections.deque` ist eine Warteschlange: `append(x)` hängt hinten an, `popleft()` nimmt vorne weg (schneller als `pop(0)` bei
  einer Liste).
- `itertools.product([0, 1, 2], repeat=n)` liefert alle Tupel aus n Zahlen 0, 1 oder 2, also alle 3ⁿ Möglichkeiten, n Objekten je eine
  von drei Möglichkeiten zuzuordnen.

## Die Idee in Worten

1. **Was genau ist erlaubt?** Wir lesen die Regeln so wie die Lösungshinweise (Abschnitt 1.1):
   - Personen steigen selbst in die Körbe und wieder heraus. Steine legt eine Person hinein oder nimmt sie heraus, die an derselben
     Stelle ist (oben oder unten).
   - Die Körbe fahren, wenn der obere Korb schwerer ist als der untere. Sitzt in einem der Körbe eine Person, darf der Unterschied
     höchstens d sein. Fahren nur Steine, spielt d keine Rolle.
   - Ein Stein, der allein nach unten gefahren ist, bleibt dort im Korb liegen, solange unten keine Person ist, und fährt bei der
     nächsten Fahrt wieder mit nach oben.
2. **Zustand:** Für jedes Objekt merken wir die Position `^` (oben) oder `_` (unten). Für Steine brauchen wir eine dritte Position
   `k`: unten **im Korb**. Ist unten eine Person, kann sie alles umräumen; dann unterscheiden wir `k` und `_` nicht.
3. **Fahrten:** Eine Fahrt ist ein Paar aus dem Inhalt des oberen und des unteren Korbs. Alle erlaubten Paare berechnen wir einmal vorab.
   In einem Zustand ist eine Fahrt möglich, wenn ihr oberer Inhalt oben und ihr unterer Inhalt unten ist; ist unten keine Person, muss
   der untere Inhalt genau das sein, was dort im Korb liegt.
4. **Breitensuche** vom Startzustand aus, bis keine Person mehr oben ist. Findet sie kein solches Ziel, gibt es keine Lösung.
5. **Ausgabe** als Tabelle, **Probe** mit einer unabhängigen Prüffunktion und Vergleich mit den Lösungshinweisen.

## Schritt 1: Einlesen

Ein Objekt speichern wir als Tupel `(art, gewicht, position)`, zum Beispiel `('P', 195, '^')`. `split()` ohne Argument trennt an
beliebig vielen Leerzeichen und entfernt auch Zeilenenden; manche Dateien haben Windows-Zeilenenden oder ein Leerzeichen am Ende.

````python
NAMEN = ['seilschaften0', 'seilschaften1', 'seilschaften2', 'seilschaften3', 'seilschaften4', 'seilschaften5', 'seilschaften6']


def dateiname(name):
    if name == 'seilschaften6':             # auf der BWINF-Seite leer, aus den Lösungshinweisen abgelesen
        return 'eigene_testdaten/seilschaften6.txt'
    return 'beispieldaten/' + name + '.txt'


def lies_eingabe(dateiname):
    zeilen = []
    with open(dateiname, encoding='utf-8') as datei:
        for zeile in datei:
            teile = zeile.split()
            if len(teile) > 0:              # leere Zeilen überspringen
                zeilen.append(teile)
    d = int(zeilen[0][0])
    objekte = []
    for art, gewicht, position in zeilen[1:]:
        objekte.append((art, int(gewicht), position))
    return d, objekte


def objekt_name(objekt):
    return objekt[0] + str(objekt[1])       # z. B. 'P195'


eingaben = {}
for name in NAMEN:
    d, objekte = lies_eingabe(dateiname(name))
    eingaben[name] = (d, objekte)
    texte = []
    for objekt in objekte:
        texte.append(objekt_name(objekt) + objekt[2])
    print(f'{name}: d = {d:2}, ' + ' '.join(texte))
````

Personen und Steine stehen bunt gemischt. In den Beispielen 3 bis 6 ist am Anfang schon jemand unten, in 3, 5 und 6 liegt dort auch
ein Stein.

## Schritt 2: Zustände

Ein Zustand ist ein Tupel mit einer Position je Objekt, in derselben Reihenfolge wie `objekte`:

| Zeichen | Bedeutung |
|---|---|
| `^` | oben (im Korb oder daneben, oben ist ja immer eine Person, die alles umräumen kann) |
| `_` | unten |
| `k` | unten im Korb; nur für Steine und nur, wenn unten **keine** Person ist |

Warum drei Positionen? Im Original fährt zuerst der Stein allein nach unten. Unten ist niemand, also liegt er dort im Korb und **muss**
bei der nächsten Fahrt wieder nach oben fahren. Ein Stein, der von Anfang an unten neben dem Korb liegt, kann dagegen gar nicht fahren,
solange niemand unten ist. Mit nur `^` und `_` könnte man die beiden Fälle nicht auseinanderhalten.

Ist unten eine Person, ersetzt `normalform` jedes `k` durch `_`. So bekommen zwei Zustände, die sich nur durch Umräumen unten
unterscheiden, dasselbe Tupel. Ist oben keine Person mehr, sind alle entkommen: Das ist ein **Zielzustand**.

````python
def person_unten(zustand, objekte):
    for i in range(len(objekte)):
        if objekte[i][0] == 'P' and zustand[i] != '^':
            return True
    return False


def ist_ziel(zustand, objekte):
    for i in range(len(objekte)):
        if objekte[i][0] == 'P' and zustand[i] == '^':
            return False                    # noch eine Person oben
    return True


def normalform(positionen, objekte):
    """Gibt den Zustand als Tupel zurück; ist unten eine Person, wird aus 'k' ein '_'."""
    if person_unten(positionen, objekte):
        ergebnis = []
        for p in positionen:
            if p == 'k':
                ergebnis.append('_')
            else:
                ergebnis.append(p)
        return tuple(ergebnis)
    return tuple(positionen)


def start_zustand(objekte):
    positionen = []
    for objekt in objekte:
        positionen.append(objekt[2])        # am Anfang sind beide Körbe leer: nur '^' oder '_'
    return tuple(positionen)


d0, objekte0 = eingaben['seilschaften0']
print(start_zustand(objekte0), ist_ziel(start_zustand(objekte0), objekte0))
print(normalform(['^', '^', '^', 'k'], objekte0))     # Stein allein unten: bleibt im Korb
print(normalform(['^', '^', 'k', 'k'], objekte0))     # der Sohn ist mit unten: er räumt den Korb aus
print(ist_ziel(('_', '_', '_', '^'), objekte0))
````

## Schritt 3: Alle erlaubten Fahrten

Eine Fahrt ist ein Paar `(oben, unten)`: die Liste der Objekte (als Nummern) im oberen Korb und die im unteren Korb. Jedes Objekt ist
entweder nicht dabei (0), im oberen Korb (1) oder im unteren Korb (2). `product([0, 1, 2], repeat=n)` geht alle 3ⁿ Möglichkeiten durch;
bei 9 Objekten sind das 19.683. Erlaubt ist eine Fahrt, wenn
- der obere Korb schwerer ist als der untere (dann ist er auch nicht leer) und
- der Unterschied höchstens d ist, falls eine Person mitfährt.

Ob eine Fahrt in einem bestimmten Zustand möglich ist, prüfen wir erst im nächsten Schritt.

````python
from itertools import product


def gewicht(nummern, objekte):
    summe = 0
    for i in nummern:
        summe = summe + objekte[i][1]
    return summe


def mit_person(nummern, objekte):
    for i in nummern:
        if objekte[i][0] == 'P':
            return True
    return False


def alle_fahrten(objekte, d):
    fahrten = []
    for wahl in product([0, 1, 2], repeat=len(objekte)):
        oben = []
        unten = []
        for i in range(len(objekte)):
            if wahl[i] == 1:
                oben.append(i)
            elif wahl[i] == 2:
                unten.append(i)
        differenz = gewicht(oben, objekte) - gewicht(unten, objekte)
        if differenz <= 0:
            continue                        # der obere Korb muss schwerer sein
        if mit_person(oben + unten, objekte) and differenz > d:
            continue                        # zu rasant für Personen
        fahrten.append((oben, unten))
    return fahrten


def namen(nummern, objekte):
    texte = []
    for i in nummern:
        texte.append(objekt_name(objekte[i]))
    if len(texte) == 0:
        return '–'
    return ' '.join(texte)


for oben, unten in alle_fahrten(objekte0, d0):
    print(f'oben: {namen(oben, objekte0):10} unten: {namen(unten, objekte0):6} Differenz {gewicht(oben, objekte0) - gewicht(unten, objekte0)}')
print()
for name in NAMEN:
    d, objekte = eingaben[name]
    print(f'{name}: {len(objekte)} Objekte, 3 hoch {len(objekte)} = {3 ** len(objekte):6} Möglichkeiten, davon erlaubt: {len(alle_fahrten(objekte, d)):3}')
````

Im Original gibt es nur vier erlaubte Fahrten: Der Stein fährt allein, Sohn gegen Stein, Tochter gegen Sohn und Königin gegen Tochter und
Stein. Alle drei Fahrten mit Personen haben genau den Unterschied 15. Bei Beispiel 2 mit d = 1 und den Gewichten 1, 2, 4, …, 256 sind
nur 9 Fahrten erlaubt. Bei den Beispielen 3 und 4 mit 9 Objekten bleiben von 19.683 Möglichkeiten 288 bzw. 343 übrig.

## Schritt 4: Eine Fahrt ausführen

`moeglich` prüft, ob eine Fahrt im Zustand möglich ist:
- Alles, was im oberen Korb fahren soll, ist oben. Oben ist immer eine Person, die die Steine einladen kann (sonst wären wir schon am
  Ziel).
- Alles, was im unteren Korb fahren soll, ist unten.
- Ist unten keine Person, kann dort niemand etwas einladen oder ausladen. Der untere Korb fährt dann genau mit den Steinen, die schon
  drin liegen (Position `k`).

`fahre` vertauscht die Positionen: Was oben eingestiegen ist, liegt danach unten im Korb (`k`), was unten eingestiegen ist, ist oben.
Am Ende bringt `normalform` den neuen Zustand in die einheitliche Form.

````python
def moeglich(zustand, fahrt, objekte):
    oben, unten = fahrt
    for i in oben:
        if zustand[i] != '^':
            return False
    for i in unten:
        if zustand[i] == '^':
            return False
    if not person_unten(zustand, objekte):
        im_korb = []
        for i in range(len(zustand)):
            if zustand[i] == 'k':
                im_korb.append(i)
        if im_korb != unten:                # unten kann niemand umräumen
            return False
    return True


def fahre(zustand, fahrt, objekte):
    oben, unten = fahrt
    neu = list(zustand)
    for i in oben:
        neu[i] = 'k'                        # ist jetzt unten im Korb
    for i in unten:
        neu[i] = '^'
    return normalform(neu, objekte)


fahrten0 = alle_fahrten(objekte0, d0)
z = start_zustand(objekte0)
print('Start:      ', z)
for fahrt in fahrten0:
    if moeglich(z, fahrt, objekte0):
        print('möglich:', namen(fahrt[0], objekte0), 'gegen', namen(fahrt[1], objekte0), '->', fahre(z, fahrt, objekte0))
z = ('^', '^', '^', 'k')
print('Stein unten:', z)
for fahrt in fahrten0:
    if moeglich(z, fahrt, objekte0):
        print('möglich:', namen(fahrt[0], objekte0), 'gegen', namen(fahrt[1], objekte0), '->', fahre(z, fahrt, objekte0))
````

Am Anfang kann nur der Stein allein fahren (Sohn gegen Stein geht nicht, weil der Stein nicht unten ist). Liegt der Stein unten im Korb,
geht nur Sohn gegen Stein: Der Stein fährt wieder hoch, der Sohn ist unten. Die Fahrt Stein allein wäre jetzt nicht möglich, denn der
Stein ist nicht mehr oben, und unten kann niemand ihn aus dem Korb nehmen.

## Schritt 5: Breitensuche

Die Warteschlange enthält Zustände, deren Nachfolger wir noch suchen müssen. Am Anfang steht nur der Startzustand darin. Das Dictionary
`vorgaenger` speichert zu jedem schon gefundenen Zustand, aus welchem Zustand wir ihn erreicht haben; es dient zugleich als Menge der
bekannten Zustände. Wir nehmen immer den vordersten Zustand aus der Warteschlange:
- Ist er ein Ziel, lesen wir den Weg über die Vorgänger rückwärts ab.
- Sonst probieren wir alle Fahrten aus. Jeder **neue** Zustand bekommt seinen Vorgänger und kommt hinten in die Warteschlange.

Weil die Warteschlange die Zustände in der Reihenfolge ihrer Entdeckung liefert, kommen erst alle Zustände mit 1 Fahrt, dann alle mit 2
Fahrten usw. dran. Das erste Ziel hat also eine kürzeste Folge. Ist die Warteschlange leer, ohne dass ein Ziel kam, gibt es keine Lösung.

Neben der Folge gibt `kuerzeste_folge` zwei Zähler zurück: wie viele Zustände gefunden wurden und wie oft `moeglich` aufgerufen wurde.

````python
from collections import deque


def kuerzeste_folge(objekte, d):
    fahrten = alle_fahrten(objekte, d)
    start = start_zustand(objekte)
    vorgaenger = {start: None}              # bekannter Zustand -> Zustand, aus dem wir ihn erreicht haben
    warteschlange = deque([start])
    geprueft = 0
    while len(warteschlange) > 0:
        zustand = warteschlange.popleft()
        if ist_ziel(zustand, objekte):
            folge = []
            while zustand is not None:      # rückwärts bis zum Start
                folge.append(zustand)
                zustand = vorgaenger[zustand]
            folge.reverse()
            return folge, len(vorgaenger), geprueft
        for fahrt in fahrten:
            geprueft = geprueft + 1
            if moeglich(zustand, fahrt, objekte):
                neu = fahre(zustand, fahrt, objekte)
                if neu not in vorgaenger:
                    vorgaenger[neu] = zustand
                    warteschlange.append(neu)
    return None, len(vorgaenger), geprueft  # kein Ziel erreichbar


folge0, anzahl0, geprueft0 = kuerzeste_folge(objekte0, d0)
print(len(folge0) - 1, 'Fahrten,', anzahl0, 'Zustände gefunden,', geprueft0, 'Fahrten geprüft')
for zustand in folge0:
    print(''.join(zustand))
````

Das originale Rätsel braucht 10 Fahrten. Die Breitensuche findet dabei nur 15 Zustände.

## Schritt 6: Übersichtliche Ausgabe

Die Lösungshinweise wünschen sich zuerst die Anzahl der Fahrten und dann eine Folge, bei der man die Änderungen leicht erkennt. Wir
schreiben eine Tabelle: eine Spalte je Objekt mit `^` oder `_` (ein `k` zeigen wir als `_`), dazu, wer abwärts und wer aufwärts fährt
und mit welchem Gewichtsunterschied. Wer abwärts fährt, war vorher oben und ist danach unten; wer aufwärts fährt, umgekehrt. Bei langen
Folgen zeigt `zeige_folge` nur den Anfang und die letzten beiden Zeilen.

````python
def abwaerts_aufwaerts(vorher, nachher):
    abwaerts = []
    aufwaerts = []
    for i in range(len(vorher)):
        if vorher[i] == '^' and nachher[i] != '^':
            abwaerts.append(i)
        if vorher[i] != '^' and nachher[i] == '^':
            aufwaerts.append(i)
    return abwaerts, aufwaerts


def zeige_folge(objekte, d, folge, hoechstens=40):
    if folge is None:
        print(f'd = {d}: Es gibt keine Folge von Fahrten, mit der alle Personen entkommen.')
        return
    print(f'd = {d}: {len(folge) - 1} Fahrten')
    breiten = []
    kopf = 'Nr. '
    for objekt in objekte:
        breiten.append(len(objekt_name(objekt)))
        kopf = kopf + ' ' + objekt_name(objekt)
    breite_ab = len('abwärts')              # Spaltenbreiten: so breit wie der längste Eintrag
    breite_auf = len('aufwärts')
    for t in range(1, len(folge)):
        ab, auf = abwaerts_aufwaerts(folge[t - 1], folge[t])
        breite_ab = max(breite_ab, len(namen(ab, objekte)))
        breite_auf = max(breite_auf, len(namen(auf, objekte)))
    zeilen = []
    for t in range(len(folge)):
        zeile = f'{t:3} '
        for i in range(len(objekte)):
            zeichen = '^' if folge[t][i] == '^' else '_'
            zeile = zeile + ' ' + zeichen.rjust(breiten[i])
        if t > 0:
            ab, auf = abwaerts_aufwaerts(folge[t - 1], folge[t])
            differenz = gewicht(ab, objekte) - gewicht(auf, objekte)
            zeile = zeile + '   ' + namen(ab, objekte).ljust(breite_ab) + '  ' + namen(auf, objekte).ljust(breite_auf)
            zeile = zeile + f'  {differenz:5}'
            if not mit_person(ab + auf, objekte):
                zeile = zeile + ' (nur Steine)'
        zeilen.append(zeile)
    print(kopf + '   ' + 'abwärts'.ljust(breite_ab) + '  ' + 'aufwärts'.ljust(breite_auf) + '  Diff.')
    if len(zeilen) <= hoechstens:
        print('\n'.join(zeilen))
    else:
        print('\n'.join(zeilen[:hoechstens - 3]))
        print('...')
        print('\n'.join(zeilen[-2:]))


zeige_folge(objekte0, d0, folge0)
````

So geht das Original: Der Stein fährt allein hinunter (1), dann der Sohn hinunter und der Stein hinauf (2). Der Stein fährt wieder
hinunter, und der Sohn nimmt ihn aus dem Korb (3). Die Tochter fährt hinunter, der Sohn hinauf (4), dann der Sohn hinunter und der Stein
hinauf (5), der Stein wieder hinunter (6). Jetzt sind Tochter, Sohn und Stein unten, und die Königin fährt gegen Tochter und Stein
hinunter (7). Die Fahrten 8 bis 10 bringen wie 3 bis 5 Tochter und Sohn hinunter. Die Lösungshinweise geben eine andere Folge an, die
auch 10 Fahrten lang ist; oft gibt es mehrere kürzeste Folgen, und welche die Breitensuche findet, hängt von der Reihenfolge der Fahrten
in der Liste ab.

## Schritt 7: Probe mit einer unabhängigen Prüffunktion

Stimmt die Folge wirklich? `pruefe_folge` bekommt nur Zeilen aus `^` und `_` (so wie die Lösungshinweise ihre Folgen angeben) und prüft
jede Fahrt direkt an den Regeln der Aufgabe, ohne die Funktionen der Breitensuche zu benutzen. Was unten im Korb liegt, merkt sie sich
selbst: Nach einer Fahrt sind das die Objekte, die gerade abwärts gefahren sind.

````python
def als_zeile(zustand):
    zeichen = []
    for p in zustand:
        zeichen.append('^' if p == '^' else '_')
    return ''.join(zeichen)


def pruefe_folge(objekte, d, zeilen):
    personen = []
    for i in range(len(objekte)):
        if objekte[i][0] == 'P':
            personen.append(i)
    if zeilen[0] != als_zeile(start_zustand(objekte)):
        return 'Die Folge beginnt nicht mit dem Startzustand.'
    korb_unten = []                         # was unten im Korb liegt (nur wichtig, wenn unten niemand ist)
    for t in range(1, len(zeilen)):
        vorher = zeilen[t - 1]
        ab, auf = abwaerts_aufwaerts(vorher, zeilen[t])
        oben_jemand = False
        unten_jemand = False
        for i in personen:
            if vorher[i] == '^':
                oben_jemand = True
            else:
                unten_jemand = True
        differenz = gewicht(ab, objekte) - gewicht(auf, objekte)
        if not oben_jemand:
            return f'Fahrt {t}: Oben ist niemand mehr, der den Korb beladen kann.'
        if differenz <= 0:
            return f'Fahrt {t}: Der obere Korb ist nicht schwerer als der untere.'
        if mit_person(ab + auf, objekte) and differenz > d:
            return f'Fahrt {t}: Unterschied {differenz} ist größer als d = {d}.'
        if not unten_jemand and auf != korb_unten:
            return f'Fahrt {t}: Unten ist niemand, der den Korb beladen oder ausladen kann.'
        korb_unten = ab
    for i in personen:
        if zeilen[-1][i] == '^':
            return 'Am Ende ist noch eine Person oben.'
    return f'in Ordnung ({len(zeilen) - 1} Fahrten)'


zeilen0 = []
for zustand in folge0:
    zeilen0.append(als_zeile(zustand))
print('unsere Folge:          ', pruefe_folge(objekte0, d0, zeilen0))

# die Folge aus den Lösungshinweisen (Abschnitt 1.4), Spalten P195 P105 P90 S75
hinweise0 = ['^^^^', '^^^_', '^^_^', '^_^^', '^_^_', '_^^^', '_^^_', '_^_^', '__^^', '__^_', '___^']
print('Folge der Hinweise:    ', pruefe_folge(objekte0, d0, hinweise0))

# absichtlich falsche Folgen
print('mit d = 14:            ', pruefe_folge(objekte0, 14, zeilen0))
print('Sohn allein hinunter:  ', pruefe_folge(objekte0, d0, ['^^^^', '^^_^']))
print('letzte Fahrt fehlt:    ', pruefe_folge(objekte0, d0, zeilen0[:-1]))
d_klein, objekte_klein = lies_eingabe('eigene_testdaten/stein_unten.txt')
print('stein_unten:', objekte_klein, 'd =', d_klein)
print('Person gegen Stein:    ', pruefe_folge(objekte_klein, d_klein, ['^_', '_^']))
````

Unsere Folge und die der Lösungshinweise sind beide in Ordnung, die absichtlich falschen Folgen fallen auf. Für die letzte Probe haben wir
ein eigenes kleines Beispiel `eigene_testdaten/stein_unten.txt`: Eine Person (10) ist oben, ein Stein (7) liegt unten neben dem Korb,
d = 5. Die Fahrt „Person gegen Stein“ hätte den Unterschied 3, aber unten ist niemand, der den Stein in den Korb legen könnte.

### Typische Fehler

Die Lösungshinweise nennen zwei Stellen, an denen man leicht falsch modelliert:

1. **d auch für Steine:** Wer verlangt, dass auch reine Steinfahrten höchstens den Unterschied d haben, findet beim Original keine
   Lösung, denn der Stein (75) muss allein hinunterfahren.
2. **Nur zwei Positionen** (`^` und `_`): Dann vergisst man, dass unten niemand die Steine ein- oder ausladen kann. Bei den
   Beispielen der BWINF-Seite ändert das zufällig nichts an den Anzahlen. Unser eigenes kleines Beispiel `stein_unten.txt` zeigt den
   Unterschied: Eine Person (10) ist oben, ein Stein (7) liegt unten neben dem Korb, d = 5. Mit zwei Positionen fährt die Person gegen
   den Stein hinunter (Unterschied 3). In Wirklichkeit kann niemand den Stein einladen, und allein ist die Person zu schwer.

Wir bauen die Fehler nach, indem wir `alle_fahrten` und `moeglich` vorübergehend durch falsche Versionen ersetzen (so wie wir sonst
Versionen umschalten) und danach die richtigen wiederherstellen.

````python
richtig_alle_fahrten = alle_fahrten
richtig_moeglich = moeglich


def alle_fahrten_d_fuer_steine(objekte, d):     # Fehler 1: d auch für reine Steinfahrten
    fahrten = []
    for oben, unten in richtig_alle_fahrten(objekte, d):
        if gewicht(oben, objekte) - gewicht(unten, objekte) <= d:
            fahrten.append((oben, unten))
    return fahrten


def moeglich_zwei_positionen(zustand, fahrt, objekte):   # Fehler 2: unten darf immer umgeräumt werden
    oben, unten = fahrt
    for i in oben:
        if zustand[i] != '^':
            return False
    for i in unten:
        if zustand[i] == '^':
            return False
    return True


def anzahl_fahrten(name_oder_datei):
    if name_oder_datei in eingaben:
        d, objekte = eingaben[name_oder_datei]
    else:
        d, objekte = lies_eingabe(name_oder_datei)
    folge = kuerzeste_folge(objekte, d)[0]
    if folge is None:
        return 'keine'
    return len(folge) - 1


BEISPIELE = NAMEN + ['eigene_testdaten/stein_unten.txt']
for version in ['richtig', 'd für Steine', 'zwei Positionen']:
    if version == 'd für Steine':
        alle_fahrten = alle_fahrten_d_fuer_steine
    if version == 'zwei Positionen':
        moeglich = moeglich_zwei_positionen
    werte = []
    for beispiel in BEISPIELE:
        werte.append(str(anzahl_fahrten(beispiel)))
    alle_fahrten = richtig_alle_fahrten     # wieder richtig
    moeglich = richtig_moeglich
    print(f'{version:16}', ' '.join(werte))
print('(Beispiele 0 bis 6, dann stein_unten)')
````

Mit „d für Steine“ bleibt nur Beispiel 2 richtig (dort gibt es keine Steine), bei Beispiel 6 findet man eine zu lange Folge. „Zwei
Positionen“ fällt nur bei `stein_unten` auf: 1 Fahrt statt keiner Lösung.

### Warum braucht Beispiel 2 genau 511 Fahrten?

In Beispiel 2 gibt es nur Personen mit den Gewichten 1, 2, 4, …, 256 und d = 1. Bei jeder Fahrt ist der obere Korb schwerer, aber
höchstens um 1, also um **genau 1**. Das Gewicht unten wächst deshalb bei jeder Fahrt um genau 1: Es kommt der obere Korb dazu, der
untere geht weg. Am Anfang ist unten das Gewicht 0, am Ende 1 + 2 + … + 256 = 511. Also braucht **jede** Lösung genau 511 Fahrten,
nicht nur die kürzeste. Allgemein sind es bei n solchen Personen 2ⁿ − 1 Fahrten. Allein das Ausgeben der Lösung kostet also Zeit, die
mit 2ⁿ wächst; schneller als exponentiell kann kein Verfahren sein (so argumentieren auch die Lösungshinweise).

Nach Fahrt k ist unten genau das Gewicht k. Welche Personen unten sind, ergibt sich aus der Binärdarstellung von k:

## Zum Schluss: Seilschaften mit MiniZinc

Kann MiniZinc einen Fahrplan finden? Wir legen eine Höchstzahl T von Fahrten fest und lassen MiniZinc für jede Fahrt t entscheiden, was
abwärts (`ab[t, i]`) und was aufwärts (`auf[t, i]`) fährt. `oben[t, i]` und `korb[t, i]` beschreiben den Zustand nach Fahrt t, genau
wie unsere Positionen `^` und `k`. Die Bedingungen sind die Regeln aus `moeglich` und `fahre`:
- Was abwärts fährt, war oben; was aufwärts fährt, war unten. Oben muss eine Person sein.
- Der Unterschied ist größer als 0 und höchstens d, falls eine Person mitfährt.
- Ist unten keine Person, fährt genau das aufwärts, was im Korb liegt.
- Nach der Fahrt ist oben, was oben war und nicht abwärts fuhr, oder was aufwärts fuhr. Im Korb liegt, was abwärts fuhr.

Damit „höchstens T Fahrten“ heißt und nicht „genau T“, darf eine Fahrt auch ausfallen (`faehrt[t]` falsch); dann ändert sich nichts.
Ausfallen dürfen nur Fahrten am Ende. Neu gegenüber den Arbeitsblättern:
- `let { … } in` gibt Teilausdrücken einen Namen, z. B. `differenz`.
- `exists(i in 1..n where …)(…)` ist wahr, wenn die Bedingung für mindestens ein i gilt (das Gegenstück zu `forall`).
- `->` heißt „wenn …, dann …“, `/\` ist „und“, `\/` ist „oder“. Ein `bool` zählt in einer Summe als 0 oder 1.

Mit T = Anzahl unserer Fahrten muss MiniZinc eine Lösung finden; mit T − 1 muss es beweisen, dass es keine gibt (`UNSATISFIABLE`). Dann
ist die Anzahl wirklich die kleinste.
```minizinc
int: n;                                       % Anzahl der Objekte
int: d;
int: T;                                       % höchstens T Fahrten
array[1..n] of bool: person;
array[1..n] of int: gewicht;
array[1..n] of bool: start_oben;

array[0..T, 1..n] of var bool: oben;          % Objekt i ist nach Fahrt t oben
array[0..T, 1..n] of var bool: korb;          % Objekt i liegt nach Fahrt t unten im Korb
array[1..T, 1..n] of var bool: ab;            % Objekt i fährt bei Fahrt t abwärts (oberer Korb)
array[1..T, 1..n] of var bool: auf;           % Objekt i fährt bei Fahrt t aufwärts (unterer Korb)
array[1..T] of var bool: faehrt;              % findet Fahrt t überhaupt statt?

constraint forall(i in 1..n)(oben[0, i] = start_oben[i] /\ not korb[0, i]);
constraint forall(t in 1..T-1)(faehrt[t + 1] -> faehrt[t]);    % Leerfahrten nur am Ende

constraint forall(t in 1..T)(
  let { var bool: person_oben = exists(i in 1..n where person[i])(oben[t - 1, i]);
        var bool: person_unten = exists(i in 1..n where person[i])(not oben[t - 1, i]);
        var bool: mit_person = exists(i in 1..n where person[i])(ab[t, i] \/ auf[t, i]);
        var int: differenz = sum(i in 1..n)(gewicht[i] * ab[t, i]) - sum(i in 1..n)(gewicht[i] * auf[t, i]) } in
  forall(i in 1..n)((ab[t, i] -> oben[t - 1, i]) /\ (auf[t, i] -> not oben[t - 1, i])) /\
  if faehrt[t] then
     person_oben /\ differenz > 0 /\ (mit_person -> differenz <= d) /\
     (not person_unten -> forall(i in 1..n)(auf[t, i] = korb[t - 1, i])) /\
     forall(i in 1..n)(oben[t, i] = ((oben[t - 1, i] /\ not ab[t, i]) \/ auf[t, i]) /\ korb[t, i] = ab[t, i])
  else
     forall(i in 1..n)(not ab[t, i] /\ not auf[t, i] /\ oben[t, i] = oben[t - 1, i] /\ korb[t, i] = korb[t - 1, i])
  endif);

constraint forall(i in 1..n where person[i])(not oben[T, i]);  % am Ende ist keine Person oben

solve satisfy;

output ["\(sum(t in 1..T)(fix(faehrt[t]))) Fahrten\n"] ++
       [if fix(oben[t, i]) then "^" else "_" endif ++ if i = n then "\n" else "" endif
        | t in 0..T where t = 0 \/ fix(faehrt[t]), i in 1..n];
```

Die Datendatei schreibt die Funktion `als_dzn`.

````python
def als_dzn(objekte, d, T):
    person = []
    gewichte = []
    start_oben = []
    for art, g, position in objekte:
        person.append('true' if art == 'P' else 'false')
        gewichte.append(str(g))
        start_oben.append('true' if position == '^' else 'false')
    return (f'n = {len(objekte)};\nd = {d};\nT = {T};\n'
            f'person = [{", ".join(person)}];\ngewicht = [{", ".join(gewichte)}];\nstart_oben = [{", ".join(start_oben)}];')


print(als_dzn(objekte0, d0, 10))
````

**Laufzeit und Solver.** CP-SAT schafft jedes Beispiel (T und T − 1) in höchstens etwa 35 Sekunden; am längsten dauert Beispiel 2 mit
