### Rosinen picken – Lösung Schritt für Schritt

*35. Bundeswettbewerb Informatik (2016/2017), 2. Runde, Aufgabe 1*

Ein großes Firmenkonglomerat wird aufgelöst, und du hast die Aufgabe, aus den vielen Bruchstücken eine möglichst wertvolle Teilmenge
auszusuchen. Die Werte der Einzelunternehmen sind unterschiedlich; einige sind sogar Verlustgeschäfte und haben daher einen negativen
Wert. Leider ist es nicht möglich, einfach nur die Unternehmen mit positivem Wert auszuwählen, weil es noch etliche Nebenbedingungen der
Form „Nimmst du A, musst du auch B nehmen“ gibt. In einem Beispielszenario könntest du nur dann die Gummibärenfabrik wählen, wenn du
ebenfalls die Zuckerraffinerie und die Altreifenverwertung nimmst – auch wenn diese für sich genommen wenig attraktiv sein sollten.
Juristische oder wirtschaftliche Notwendigkeiten, die du nicht beeinflussen kannst, sind für diese Nebenbedingungen verantwortlich.
Graphisch können wir eine solche Situation so darstellen:

*(Bild: 36 Kreise mit Zahlen darin, verbunden durch Pfeile. Laut Lösungshinweisen ist es das Beispiel `Quadrat-6`.)*

Die Kreise repräsentieren die Einzelunternehmen und sind mit deren Wert beschriftet. Ein Pfeil von A nach B bedeutet: Wenn man A
auswählt, dann muss auch B gewählt werden.

#### Aufgabe

1. Überlege dir und beschreibe ein Verfahren, mit dem man eine wertvolle Teilmenge finden kann.
2. Erstelle ein entsprechendes Programm und wende es auf die Beispiele an, die du auf der BWINF-Webseite zu dieser Runde findest.
3. Findet dein Verfahren garantiert immer ein bestmögliches Ergebnis oder kommt es diesem Ziel nur nahe? Ist die Laufzeit deines
   Programms erträglich? Natürlich ist eine lange Laufzeit bei einer so wichtigen Entscheidung akzeptabel, wenn die Zahlen in den Kreisen
   für Millionen Euros stehen.

#### Beispieldaten

Im Ordner `beispieldaten` stehen sieben Beispiele: `Kreis-8.txt`, `Quadrat-6.txt`, `Quadrat-8.txt`, `Quadrat-13.txt`, `Zufall-7.txt`,
`Zufall-40.txt` und `Zufall-100.txt`, zu jedem eine Zeichnung als PDF. Zeilen, die mit `#` beginnen, sind Kommentare. Danach kommen die
Anzahl `n` der Unternehmen, `n` Zeilen „Nummer Wert“ (die Unternehmen heißen 0 bis n − 1) und dann die Pfeile, je Zeile „A B“ für
„Nimmst du A, musst du auch B nehmen“. Die Datei `ausgabeformat.txt` schreibt die Ausgabe vor: in der ersten Zeile die Anzahl der
gewählten Unternehmen, in der zweiten ihr Gesamtwert, danach je Zeile die Nummer eines gewählten Unternehmens.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Gerichteter Graph:** Die Unternehmen sind Knoten, die Bedingungen Pfeile. Wir speichern zu jedem Knoten die Liste seiner
  **Nachfolger** (wohin seine Pfeile zeigen) und seiner **Vorgänger** (woher Pfeile zu ihm kommen).
- **Tiefensuche mit einem Stapel:** Von einem Knoten aus findet man alle Knoten, die man über Pfeile erreichen kann.
- **Brute Force:** alle Teilmengen ausprobieren. Bei n Unternehmen sind das 2ⁿ, also nur für ganz kleine Beispiele.
- **Gieriges Verfahren (Greedy):** immer den Schritt machen, der gerade am meisten bringt. Schnell, aber nicht immer das Beste.
- **Backtracking mit Schranke** (auch *Branch and Bound*): Unternehmen für Unternehmen entscheiden, „nehmen“ oder „nicht nehmen“, und
  eine Entscheidung zurücknehmen, wenn sie in eine Sackgasse führt. Eine **Schranke** sagt, wie viel von hier aus höchstens noch
  erreichbar ist; ist das nicht mehr als das bisher Beste, braucht man nicht weiterzusuchen.
- Für Neugierige: **maximaler Fluss und minimaler Schnitt** (Algorithmus von Edmonds und Karp), das Verfahren der Lösungshinweise.

**Nützliche Funktionen und Bibliotheken**
- Mengen (`set`): `a | b` ist die Vereinigung, `a - b` die Elemente von `a`, die nicht in `b` sind, `a & b` die gemeinsamen Elemente.
  `len(a & b) == 0` heißt: `a` und `b` haben nichts gemeinsam.
- `sorted(liste)` sortiert auch Tupel: zuerst nach dem ersten Eintrag, bei Gleichstand nach dem zweiten.
- `random.Random(zahl)`: ein Zufallsgenerator, der bei gleicher Zahl immer dieselben Zufallszahlen liefert.
- `collections.deque` mit `append` und `popleft` als Warteschlange für die Breitensuche (nur für Neugierige).

## Die Idee in Worten

1. **Die Hülle.** Wer A nimmt, muss alle Nachfolger von A nehmen, dann deren Nachfolger und so weiter: also alle Unternehmen, die man
   von A aus über Pfeile erreicht. Diese Menge nennen wir die **Hülle** von A (A selbst gehört dazu).
2. **Erlaubte Mengen.** Eine Auswahl ist erlaubt, wenn kein Pfeil aus der Auswahl herauszeigt. Auch die **leere Auswahl** ist erlaubt
   und hat den Wert 0. Wie die Lösungshinweise sagen: Kein Geschäft ist besser als ein Verlustgeschäft.
3. **Nur die positiven Unternehmen entscheiden.** Ein Unternehmen mit negativem Wert (oder Wert 0) nimmt man nie freiwillig, sondern nur,
   weil es in der Hülle eines gewählten positiven Unternehmens liegt. Eine beste Auswahl ist also immer die Vereinigung der Hüllen
   einiger positiver Unternehmen. Wir müssen nur noch entscheiden, **welche positiven Unternehmen** wir nehmen.
4. **Schiedsrichter:** Eine kleine Funktion prüft jede Auswahl, bevor wir ihr glauben.
5. **Version 1 (Brute Force):** alle Teilmengen prüfen, nur für die zwei kleinsten Beispiele.
6. **Version 2 (gierig):** immer die Hülle dazunehmen, die gerade den größten Gewinn bringt, solange es Gewinn gibt.
7. **Version 3 und 4 (Backtracking mit Schranke):** findet garantiert die beste Auswahl; die Reihenfolge, in der wir die Unternehmen
   entscheiden, macht einen großen Unterschied.
8. **Für Neugierige** und **zum Schluss:** Grenzen des Backtrackings, der minimale Schnitt der Lösungshinweise und ein MiniZinc-Modell.

## Schritt 1: Einlesen

`lies_datei` überspringt die Kommentarzeilen und gibt die Liste der Werte (`wert[v]` ist der Wert von Unternehmen `v`) und die Liste der
Pfeile als Tupel `(a, b)` zurück. Alle Beispiele speichern wir im Dictionary `daten`.

````python
BEISPIELE = ['Kreis-8', 'Quadrat-6', 'Quadrat-8', 'Quadrat-13', 'Zufall-7', 'Zufall-40', 'Zufall-100']


def lies_datei(name):
    zeilen = []
    with open('beispieldaten/' + name + '.txt', encoding='utf-8') as datei:
        for zeile in datei:
            zeile = zeile.strip()
            if zeile != '' and not zeile.startswith('#'):      # Kommentare und Leerzeilen überspringen
                zeilen.append(zeile.split())
    n = int(zeilen[0][0])
    wert = [0.0] * n
    for teile in zeilen[1:n + 1]:                               # n Zeilen „Nummer Wert“
        wert[int(teile[0])] = float(teile[1])
    pfeile = []
    for teile in zeilen[n + 1:]:                                # der Rest: Pfeile „A B“
        pfeile.append((int(teile[0]), int(teile[1])))
    return wert, pfeile


daten = {}
for name in BEISPIELE:
    daten[name] = lies_datei(name)
    wert, pfeile = daten[name]
    positiv = [w for w in wert if w > 0]
    print(f'{name:11} {len(wert):4} Unternehmen, {len(pfeile):4} Pfeile, davon {len(positiv):3} positiv',
          f'(zusammen {sum(positiv):5.0f})')
````

Alle Werte in den Beispielen sind ganze Zahlen (in der Datei mit `.0` geschrieben). Wenn wir einfach alle positiven Unternehmen nehmen
dürften, bekämen wir die Summe in Klammern. Wegen der Bedingungen wird es weniger sein.

## Schritt 2: Nachfolger, Vorgänger und Hülle

`nachfolger[v]` ist die Liste der Unternehmen, auf die ein Pfeil von `v` zeigt, `vorgaenger[v]` die Liste der Unternehmen, deren Pfeil
auf `v` zeigt. `huelle(start, nachfolger)` sammelt mit einer Tiefensuche alles, was man von `start` aus erreicht. Mit der Liste der
Vorgänger statt der Nachfolger findet dieselbe Funktion die **Vorfahren**: alle Unternehmen, die `start` mitbringen würden.

````python
def nachfolger_liste(n, pfeile):
    nachfolger = [[] for v in range(n)]
    for a, b in pfeile:
        nachfolger[a].append(b)
    return nachfolger


def vorgaenger_liste(n, pfeile):
    vorgaenger = [[] for v in range(n)]
    for a, b in pfeile:
        vorgaenger[b].append(a)
    return vorgaenger


def huelle(start, nachfolger):
    """Alle Knoten, die man von start aus über Pfeile erreicht (start eingeschlossen)."""
    gefunden = {start}
    stapel = [start]
    while len(stapel) > 0:
        v = stapel.pop()
        for w in nachfolger[v]:
            if w not in gefunden:
                gefunden.add(w)
                stapel.append(w)
    return gefunden


def gesamtwert(menge, wert):
    summe = 0.0
    for v in menge:
        summe = summe + wert[v]
    return summe


wert, pfeile = daten['Zufall-7']
nachfolger = nachfolger_liste(len(wert), pfeile)
for v in range(len(wert)):
    h = huelle(v, nachfolger)
    print(f'Unternehmen {v} (Wert {wert[v]:5}): Hülle {sorted(h)}, Wert der Hülle {gesamtwert(h, wert)}')
````

In `Zufall-7` gibt es drei positive Unternehmen (0, 1 und 5), aber jedes bringt in seiner Hülle die Unternehmen 2 und 6 mit, die zusammen
−94 wert sind. Jede Hülle mit einem positiven Unternehmen ist also ein Verlustgeschäft, und das Beste ist, **nichts** zu nehmen (Wert 0,
wie in den Lösungshinweisen).

## Schritt 3: Schiedsrichter und Ausgabe

`pruefe` geht alle Pfeile durch: Zeigt ein Pfeil von einem gewählten Unternehmen auf ein nicht gewähltes, ist die Auswahl verboten.
`ausgabe` schreibt eine Auswahl im vorgegebenen Format.

````python
def pruefe(menge, pfeile):
    for a, b in pfeile:
        if a in menge and b not in menge:
            return f'Fehler: {a} gewählt, aber {b} nicht'
    return 'ok'


def ausgabe(menge, wert):
    zeilen = [str(len(menge)), str(float(gesamtwert(menge, wert)))]
    for v in sorted(menge):
        zeilen.append(str(v))
    return '\n'.join(zeilen)


wert, pfeile = daten['Kreis-8']
print(pruefe({0, 2, 3}, pfeile))
alle = set(range(len(wert)))
print(pruefe(alle, pfeile))
print(ausgabe(alle, wert))
````

In `Kreis-8` bilden die Pfeile einen einzigen Kreis 1 → 5 → 6 → 4 → 2 → 0 → 3 → 7 → 1. Wer ein Unternehmen nimmt, muss alle nehmen.
Erlaubt sind also nur „nichts“ (Wert 0) und „alles“ (Wert 4). Das Beste ist „alles“, genau wie in `ausgabeformat.txt`.

## Schritt 4: Version 1 – Brute Force

Wir gehen alle 2ⁿ Teilmengen durch. Jede Teilmenge entspricht einer Zahl `maske` von 0 bis 2ⁿ − 1: Unternehmen `v` ist dabei, wenn die
Stelle `v` der Zahl im Zweiersystem eine 1 ist, also wenn `maske // 2**v % 2 == 1`. Jede Teilmenge prüft der Schiedsrichter.

````python
def brute_force(wert, pfeile):
    n = len(wert)
    beste_menge = set()             # die leere Auswahl (Wert 0) ist immer erlaubt
    bester_wert = 0
    erlaubt = 0
    for maske in range(2 ** n):
        menge = set()
        for v in range(n):
            if maske // 2 ** v % 2 == 1:
                menge.add(v)
        if pruefe(menge, pfeile) == 'ok':
            erlaubt = erlaubt + 1
            w = gesamtwert(menge, wert)
            if w > bester_wert:
                bester_wert = w
                beste_menge = menge
    return beste_menge, erlaubt


for name in ['Kreis-8', 'Zufall-7']:
    wert, pfeile = daten[name]
    menge, erlaubt = brute_force(wert, pfeile)
    print(f'{name}: {2 ** len(wert)} Teilmengen, davon {erlaubt} erlaubt; beste Auswahl {sorted(menge)}',
          f'mit Wert {gesamtwert(menge, wert)}')
````

Schon `Quadrat-6` hat 36 Unternehmen, also 2³⁶ ≈ 69 Milliarden Teilmengen. Selbst bei einer Million Teilmengen pro Sekunde wären das fast
20 Stunden; bei `Zufall-100` würde es länger dauern, als das Weltall alt ist. Brute Force taugt nur zum Nachprüfen kleiner Beispiele.

## Schritt 5: Version 2 – gierig Hüllen dazunehmen

Wir beginnen mit der leeren Auswahl. In jeder Runde rechnen wir für jedes noch nicht gewählte Unternehmen `v` aus, was es bringt, seine
Hülle dazuzunehmen: Das sind die Werte der Unternehmen der Hülle, die **noch nicht** gewählt sind (`huellen[v] - menge`). Wir nehmen die
Hülle mit dem größten Gewinn dazu und hören auf, wenn keine Hülle mehr Gewinn bringt.

Die Auswahl bleibt dabei immer erlaubt: Die Hülle hat keinen Pfeil nach außen, die bisherige Auswahl auch nicht, also auch ihre
Vereinigung nicht.

````python
def greedy(wert, pfeile):
    n = len(wert)
    nachfolger = nachfolger_liste(n, pfeile)
    huellen = []
    for v in range(n):
        huellen.append(huelle(v, nachfolger))
    menge = set()
    while True:
        bester_gewinn = 0
        beste_neue = None
        for v in range(n):
            if v not in menge:
                neu = huellen[v] - menge            # was beim Dazunehmen von v neu hinzukommt
                gewinn = gesamtwert(neu, wert)
                if gewinn > bester_gewinn:
                    bester_gewinn = gewinn
                    beste_neue = neu
        if beste_neue is None:                      # keine Hülle bringt noch Gewinn
            return menge
        menge = menge | beste_neue


ergebnis_greedy = {}
for name in BEISPIELE:
    wert, pfeile = daten[name]
    menge = greedy(wert, pfeile)
    ergebnis_greedy[name] = gesamtwert(menge, wert)
    print(f'{name:11} Wert {ergebnis_greedy[name]:6}, {len(menge):3} Unternehmen, Schiedsrichter: {pruefe(menge, pfeile)}')
````

Das sieht gut aus, ist aber nicht immer das Beste. Ein kleines Gegenbeispiel: Die Unternehmen 0 und 1 sind je 3 wert, brauchen aber
beide das Unternehmen 2 mit dem Wert −5. Jede Hülle für sich ist ein Verlust (3 − 5 = −2), das gierige Verfahren nimmt also nichts.
Zusammen sind sie aber 3 + 3 − 5 = 1 wert.

````python
wert_klein = [3, 3, -5]
pfeile_klein = [(0, 2), (1, 2)]
print('gierig:     ', sorted(greedy(wert_klein, pfeile_klein)))
print('Brute Force:', sorted(brute_force(wert_klein, pfeile_klein)[0]))
````

Gemeinsam genutzte Verlustgeschäfte sieht das gierige Verfahren nicht. Für eine Garantie müssen wir systematisch suchen.

## Schritt 6: Version 3 – Backtracking mit Schranke

Wir legen die positiven Unternehmen in eine Liste `positive` und entscheiden sie der Reihe nach. Dabei merken wir uns
- `menge`: die bisher gewählten Unternehmen (immer eine erlaubte Auswahl),
- `verboten`: die Unternehmen, die wir nicht mehr nehmen dürfen,
- `summe`: den Wert von `menge`.

Für das nächste positive Unternehmen `v` gibt es zwei Möglichkeiten:
- **Nehmen:** Die ganze Hülle von `v` kommt dazu. Das geht nur, wenn kein Unternehmen der Hülle verboten ist.
- **Nicht nehmen:** Dann ist `v` verboten und mit ihm alle seine **Vorfahren**, denn jeder Vorfahre würde `v` mitbringen.

Positive Unternehmen, die schon gewählt (als Teil einer Hülle) oder schon verboten sind, überspringen wir.

**Die Schranke:** Mehr als `summe` plus die Werte aller positiven Unternehmen, die noch weder gewählt noch verboten sind, können wir von
hier aus nicht mehr erreichen (dazukommende negative Unternehmen machen es nur weniger). Ist das nicht mehr als das bisher Beste,
brechen wir diesen Zweig ab.

Die festen Angaben und das bisher Beste stehen im Dictionary `g`, damit `suche` sie nicht alle einzeln als Parameter braucht.
`g['aufrufe']` zählt die Aufrufe von `suche`; bei mehr als `GRENZE` Aufrufen geben wir auf. Die Reihenfolge der positiven Unternehmen
legt die Funktion `reihenfolge` fest. In Version 3 ist es einfach die Reihenfolge der Nummern.

````python
GRENZE = 200000


def reihenfolge_nummer(positive, wert):
    return positive


def reihenfolge(positive, wert):
    return reihenfolge_nummer(positive, wert)


def suche(i, menge, verboten, summe, g):
    g['aufrufe'] = g['aufrufe'] + 1
    if g['aufrufe'] > GRENZE:                       # zu viele Aufrufe: aufgeben
        return
    if summe > g['bester_wert']:                    # neue beste Auswahl
        g['bester_wert'] = summe
        g['beste_menge'] = menge
    positive = g['positive']
    while i < len(positive) and (positive[i] in menge or positive[i] in verboten):
        i = i + 1                                   # schon entschieden: überspringen
    if i == len(positive):
        return
    moeglich = 0                                    # Schranke: alle noch möglichen positiven Werte
    for v in positive[i:]:
        if v not in menge and v not in verboten:
            moeglich = moeglich + g['wert'][v]
    if summe + moeglich <= g['bester_wert']:
        return
    v = positive[i]
    neu = g['huellen'][v] - menge
    if len(neu & verboten) == 0:                    # 1. Möglichkeit: v mit seiner Hülle nehmen
        suche(i + 1, menge | neu, verboten, summe + gesamtwert(neu, g['wert']), g)
    suche(i + 1, menge, verboten | g['vorfahren'][v], summe, g)   # 2. Möglichkeit: v und seine Vorfahren verbieten


def backtracking(wert, pfeile):
    n = len(wert)
    nachfolger = nachfolger_liste(n, pfeile)
    vorgaenger = vorgaenger_liste(n, pfeile)
    g = {'wert': wert, 'huellen': [], 'vorfahren': [], 'aufrufe': 0, 'bester_wert': 0, 'beste_menge': set()}
    for v in range(n):
        g['huellen'].append(huelle(v, nachfolger))
        g['vorfahren'].append(huelle(v, vorgaenger))
    positive = []
    for v in range(n):
        if wert[v] > 0:
            positive.append(v)
    g['positive'] = reihenfolge(positive, wert)
    suche(0, set(), set(), 0, g)
    fertig = g['aufrufe'] <= GRENZE
    return g['beste_menge'], g['aufrufe'], fertig


def teste_backtracking():
    ergebnis = {}
    for name in BEISPIELE:
        wert, pfeile = daten[name]
        menge, aufrufe, fertig = backtracking(wert, pfeile)
        ergebnis[name] = (gesamtwert(menge, wert), aufrufe, fertig)
        if fertig:
            text = 'fertig'
        else:
            text = 'ABGEBROCHEN, bisher bester Wert'
        print(f'{name:11} {aufrufe:7} Aufrufe, {text} {gesamtwert(menge, wert):6}, Schiedsrichter: {pruefe(menge, pfeile)}')
    return ergebnis


ergebnis_v3 = teste_backtracking()
````

Bei allen Beispielen außer `Quadrat-13` ist die Suche schnell fertig. Wenn sie fertig ist, haben wir **garantiert** die beste Auswahl:
Abgeschnitten werden nur Zweige, in denen sicher nichts Besseres liegt. Bei `Quadrat-13` mit 85 positiven Unternehmen reichen aber
200.000 Aufrufe nicht.

Hinweis: `menge` und `verboten` werden in `suche` nie verändert, sondern mit `|` immer neue Mengen gebaut. Deshalb dürfen wir
`g['beste_menge'] = menge` ohne Kopie schreiben, und nach der Rückkehr aus dem ersten Aufruf sind `menge` und `verboten` noch dieselben.

## Schritt 7: Version 4 – die wertvollsten Unternehmen zuerst

Die Schranke wirkt besser, wenn früh eine gute Auswahl gefunden wird. Dann ist `g['bester_wert']` schnell groß und viele Zweige fallen weg.
Wir entscheiden deshalb die **wertvollsten Unternehmen zuerst**. Zum Sortieren bauen wir Tupel `(-wert, nummer)`: `sorted` sortiert
nach dem ersten Eintrag aufsteigend, durch das Minuszeichen also nach dem Wert absteigend.

````python
def reihenfolge_wert(positive, wert):
    paare = []
    for v in positive:
        paare.append((-wert[v], v))
    ergebnis = []
    for minus_wert, v in sorted(paare):
        ergebnis.append(v)
    return ergebnis


def reihenfolge(positive, wert):
    return reihenfolge_wert(positive, wert)


ergebnis_v4 = teste_backtracking()
````

Jetzt sind alle Beispiele fertig, auch `Quadrat-13` (nach 146.641 Aufrufen). Bei `Quadrat-8` sinkt die Zahl der Aufrufe von 3.649 auf 635,
bei `Zufall-100` von 2.761 auf 1.217.

## Schritt 8: Alle Beispiele

Wir vergleichen das gierige Verfahren und das Backtracking (Version 4) mit den Lösungshinweisen. Diese geben für jedes Beispiel den
optimalen Wert an und zusätzlich den Wert, den eine Heuristik der Hinweise gefunden hat (welche, steht dort nicht).

````python
HINWEISE = {'Kreis-8': (4, 4), 'Quadrat-6': (333, 311), 'Quadrat-8': (446, 434), 'Quadrat-13': (961, 952),
            'Zufall-7': (0, 0), 'Zufall-40': (504, 499), 'Zufall-100': (613, 524)}      # (optimal, Heuristik)

print('Beispiel    | gierig | Backtracking | Hinweise optimal | Hinweise Heuristik')
for name in BEISPIELE:
    optimal, heuristik = HINWEISE[name]
    print(f'{name:11} | {ergebnis_greedy[name]:6.0f} | {ergebnis_v4[name][0]:12.0f} | {optimal:16} | {heuristik:18}')
````

**Das Backtracking findet bei allen sieben Beispielen den optimalen Wert der Lösungshinweise.** Das gierige Verfahren ist bei `Kreis-8`,
`Zufall-7` und `Zufall-100` optimal, sonst etwas schlechter (bei `Quadrat-13` um 64). Mit der Heuristik der Hinweise verglichen ist es bei
`Quadrat-6` und `Zufall-100` besser, bei `Kreis-8`, `Zufall-7` und `Zufall-40` gleich und bei `Quadrat-8` und `Quadrat-13` schlechter.

Hier die besten Auswahlen im vorgegebenen Format. Damit es übersichtlich bleibt, stehen die Nummern hier in einer Zeile; in der
Ausgabedatei steht jede in einer eigenen Zeile (Funktion `ausgabe`).

````python
for name in BEISPIELE:
    wert, pfeile = daten[name]
    menge, aufrufe, fertig = backtracking(wert, pfeile)
    zeilen = ausgabe(menge, wert).split('\n')
    print(f'{name}: {zeilen[0]} Unternehmen, Wert {zeilen[1]}, Schiedsrichter: {pruefe(menge, pfeile)}')
    if len(menge) > 0:
        print('   ', ' '.join(zeilen[2:]))
````

## Für Neugierige

### Ist das Backtracking immer schnell genug?

Nein. Im schlimmsten Fall probiert es fast alle 2ᵏ Möglichkeiten für die k positiven Unternehmen aus. Bei den Beispielen hilft die
Schranke sehr, bei größeren Graphen reicht sie aber bald nicht mehr. Wir bauen zufällige Graphen ähnlich wie `Zufall-100`: n
Unternehmen mit ganzzahligen Werten von −50 bis 50 und 1,3 · n zufälligen Pfeilen.

````python
import random


def zufallsgraph(n, saat):
    zufall = random.Random(saat)
    wert = []
    for v in range(n):
        wert.append(float(zufall.randint(-50, 50)))
    pfeile = set()                                  # eine Menge, damit kein Pfeil doppelt vorkommt
    while len(pfeile) < int(1.3 * n):
        a = zufall.randrange(n)
        b = zufall.randrange(n)
        if a != b:
            pfeile.add((a, b))
    return wert, sorted(pfeile)


for n in [150, 200, 300]:
    wert, pfeile = zufallsgraph(n, n)
    menge, aufrufe, fertig = backtracking(wert, pfeile)
    print(f'n = {n}: {aufrufe:6} Aufrufe, fertig: {fertig}, bester gefundener Wert {gesamtwert(menge, wert)}')
````

Aufgabe ist das Backtracking also gut, eine Garantie für eine kurze Laufzeit gibt es aber nicht. Genau das sagen auch die
Lösungshinweise: Backtracking reicht „gerade noch“ für die Beispiele, besser ist ein Verfahren mit polynomieller Laufzeit.

### Der minimale Schnitt (das Verfahren der Lösungshinweise)

Das Problem heißt in der Informatik **Closure-Problem** (closure: abgeschlossene Menge). Es tritt zum Beispiel im Tagebau auf: Um einen
tiefen Block abzubauen, muss man alle Blöcke darüber mit abbauen. Picard hat 1976 gezeigt, dass man es mit einem **Fluss** lösen kann:
- Man fügt eine Quelle `s` und eine Senke `t` hinzu.
- Von `s` führt zu jedem positiven Unternehmen v eine Leitung mit Kapazität `wert[v]`, von jedem negativen Unternehmen v eine Leitung zu
  `t` mit Kapazität `−wert[v]`. Die Pfeile der Aufgabe werden Leitungen mit unendlicher Kapazität.
- Ein **Schnitt** teilt die Knoten in die Seite von `s` (das ist unsere Auswahl) und die Seite von `t`. Er kostet die Summe der Kapazitäten
  aller Leitungen von der `s`-Seite zur `t`-Seite. Das sind die positiven Unternehmen, die wir **nicht** nehmen, und die negativen, die
  wir **doch** nehmen. Ein Pfeil der Aufgabe aus der Auswahl hinaus würde unendlich kosten, kommt also nie vor.
- Kosten des Schnitts = (Summe aller positiven Werte) − (Wert der Auswahl). Der **billigste** Schnitt gehört also zur **wertvollsten**
  erlaubten Auswahl.

Den billigsten Schnitt findet man über den **maximalen Fluss** (Max-Flow-Min-Cut-Satz): Man sucht immer wieder mit einer Breitensuche
einen Weg von `s` nach `t`, auf dem noch etwas fließen kann, und schickt so viel wie möglich hindurch (Algorithmus von Edmonds und
Karp). Fließt etwas von a nach b, kann es später auch wieder „zurückgeschoben“ werden; dafür wächst die Restkapazität von b nach a. Gibt
es keinen Weg mehr, sind die Knoten, die `s` noch erreicht, genau die beste Auswahl.

````python
from collections import deque

UNENDLICH = float('inf')


def leitung(rest, nachbarn, a, b, kapazitaet):
    """Leitung von a nach b; rest[(a, b)] ist die Restkapazität, die Gegenrichtung beginnt mit 0."""
    rest[(a, b)] = rest.get((a, b), 0) + kapazitaet
    rest[(b, a)] = rest.get((b, a), 0)
    nachbarn[a].add(b)
    nachbarn[b].add(a)


def min_schnitt(wert, pfeile):
    n = len(wert)
    s = n                                   # zwei neue Knoten: Quelle und Senke
    t = n + 1
    rest = {}
    nachbarn = [set() for v in range(n + 2)]
    for a, b in pfeile:
        leitung(rest, nachbarn, a, b, UNENDLICH)
    for v in range(n):
        if wert[v] > 0:
            leitung(rest, nachbarn, s, v, wert[v])
        elif wert[v] < 0:
            leitung(rest, nachbarn, v, t, -wert[v])
    fluss = 0
    wege = 0
    while True:
        vorher = {s: None}                  # Breitensuche im Restnetz: vorher[w] = woher wir w erreicht haben
        schlange = deque([s])
        while len(schlange) > 0 and t not in vorher:
            v = schlange.popleft()
            for w in nachbarn[v]:
                if w not in vorher and rest[(v, w)] > 0:
                    vorher[w] = v
                    schlange.append(w)
        if t not in vorher:                 # kein Weg mehr: fertig
            break
        weg = []                            # den Weg von t aus rückwärts ablesen
        w = t
        while vorher[w] is not None:
            weg.append((vorher[w], w))
            w = vorher[w]
        menge_fluss = UNENDLICH             # so viel passt durch den engsten Teil
        for kante in weg:
            menge_fluss = min(menge_fluss, rest[kante])
        for a, b in weg:
            rest[(a, b)] = rest[(a, b)] - menge_fluss
            rest[(b, a)] = rest[(b, a)] + menge_fluss
        fluss = fluss + menge_fluss
        wege = wege + 1
    auswahl = set(vorher)                   # was s im Restnetz noch erreicht, ohne s selbst
    auswahl.remove(s)
    return auswahl, fluss, wege


for name in BEISPIELE:
    wert, pfeile = daten[name]
    auswahl, fluss, wege = min_schnitt(wert, pfeile)
    summe_positiv = gesamtwert([v for v in range(len(wert)) if wert[v] > 0], wert)
    print(f'{name:11} Wert {gesamtwert(auswahl, wert):4.0f} ({len(auswahl):2} Unternehmen), {wege:2} Wege,',
          f'Summe der positiven Werte - Fluss = {summe_positiv:4.0f} - {fluss:3.0f} = {summe_positiv - fluss:3.0f},',
          f'Schiedsrichter: {pruefe(auswahl, pfeile)}')
````

Dieselben Werte wie beim Backtracking, und die Summe der positiven Werte minus dem Fluss ist jeweils genau der beste Wert, wie es die
Überlegung oben verlangt (der Fluss ist so groß wie der billigste Schnitt). Die Zahl der Wege ist klein, und sie wächst nur langsam mit der Größe des Graphen (Edmonds-Karp braucht
höchstens etwa Knoten · Kanten Wege, egal wie groß die Werte sind). Auch große Zufallsgraphen sind schnell gelöst:

````python
for n in [300, 1000]:
    wert, pfeile = zufallsgraph(n, n)
    auswahl, fluss, wege = min_schnitt(wert, pfeile)
    print(f'n = {n:4}: Wert {gesamtwert(auswahl, wert):5.0f} nach {wege} Wegen, Schiedsrichter: {pruefe(auswahl, pfeile)}')
````

Bei n = 300 ist der beste Wert 1218. Das Backtracking hatte nach 200.000 Aufrufen erst eine Auswahl mit dem Wert 691 gefunden.

### Weitere Ideen aus den Lösungshinweisen

- **Vereinfachen:** Unternehmen, die gegenseitig voneinander abhängen (ein Kreis aus Pfeilen, z. B. ganz `Kreis-8`), nimmt man nur
  gemeinsam. Man kann sie zu einem Knoten mit der Summe ihrer Werte zusammenfassen (**starke Zusammenhangskomponenten**, Algorithmus von
  Tarjan); danach ist der Graph kreisfrei. Positive Unternehmen ohne Pfeil nach außen nimmt man sofort, negative, auf die kein Pfeil
  zeigt, streicht man. Solche Regeln verkleinern den Graphen vor einer Suche.
- **Lineare Optimierung:** Mit einer Zahl `nimm[v]` zwischen 0 und 1 für jedes Unternehmen lautet die Aufgabe: Maximiere die Summe von
  `wert[v] · nimm[v]` unter den Bedingungen `nimm[a] ≤ nimm[b]` für jeden Pfeil. Erstaunlicherweise ist die beste Lösung dieses Problems
  automatisch ganzzahlig (die Matrix der Bedingungen ist *total unimodular*). Ein LP-Löser findet sie in polynomieller Zeit. Das nutzen
  wir gleich mit MiniZinc.
- **Erweiterung:** andere Bedingungen wie „Nimmst du A, darfst du B nicht nehmen“. Dann ist das Problem viel schwerer.

## Zum Schluss: Rosinen picken mit MiniZinc

Die Aufgabe ist ein Optimierungsproblem mit einfachen Bedingungen, also gut für MiniZinc geeignet. Für jedes Unternehmen gibt es eine
Variable `nimm[v]` mit dem Wert 0 oder 1. Ein Pfeil von a nach b heißt: Ist `nimm[a] = 1`, dann auch `nimm[b] = 1`, kurz
`nimm[a] <= nimm[b]`. Maximiert wird die Summe der Werte der gewählten Unternehmen.

Neu gegenüber den Arbeitsblättern:
- `array[0..n-1]`: Die Unternehmen sind wie in den Dateien ab 0 nummeriert, deshalb beginnt der Index bei 0. In der Datendatei schreibt
  man ein solches Array mit `array1d(0..n-1, [...])`.
- `fix(nimm[v])` in der Ausgabe gibt den Wert der Variablen in der gefundenen Lösung; damit können wir in der Ausgabe nach ihm auswählen.
```minizinc
int: n;                                   % Anzahl der Unternehmen, nummeriert von 0 bis n - 1
array[0..n-1] of int: wert;
int: m;                                   % Anzahl der Pfeile
array[1..m, 1..2] of int: pfeil;          % Pfeil k zeigt von pfeil[k, 1] nach pfeil[k, 2]

array[0..n-1] of var 0..1: nimm;          % 1 = Unternehmen gewählt

constraint forall(k in 1..m)(nimm[pfeil[k, 1]] <= nimm[pfeil[k, 2]]);

var int: gesamt = sum(v in 0..n-1)(wert[v] * nimm[v]);
solve maximize gesamt;

output [show(sum(nimm)), "\n", show(gesamt), "\n", show([v | v in 0..n-1 where fix(nimm[v]) = 1]), "\n"];
```

Die Datendatei schreibt die Funktion `als_dzn` aus den Beispieldaten (die Werte sind ganze Zahlen):

````python
def als_dzn(wert, pfeile):
    werte_text = ', '.join(str(int(w)) for w in wert)
    pfeile_text = ' | '.join(f'{a}, {b}' for a, b in pfeile)
    return (f'n = {len(wert)};\nwert = array1d(0..{len(wert) - 1}, [{werte_text}]);\n'
            f'm = {len(pfeile)};\npfeil = [| {pfeile_text} |];')


print(als_dzn(*daten['Kreis-8']))
````
