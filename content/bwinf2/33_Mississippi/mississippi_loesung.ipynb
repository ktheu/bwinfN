Die Erbinformation aller Lebewesen befindet sich in der Desoxyribonukleinsäure (DNS, engl. DNA) in Form einer Folge der vier Basen
Adenin (A), Guanin (G), Cytosin (C) und Thymin (T). Lange Folgen von Basen bilden die Chromosomen. Jedes Chromosom ist ein Doppelstrang
aus verdrillten Basenfolgen. Eine derartige „Strickleiter“ als verdrillter Doppelstrang wird auch Doppelhelix genannt.

Das Genom, die Folge der Basen in den Chromosomen, ist für den Menschen inzwischen vollständig bekannt. Für weitergehende Forschungen
bezüglich der Heilung von Krankheiten ist es allerdings notwendig, das Genom genauer zu untersuchen, etwa auf wiederholt vorkommende
Basenfolgen, auch Repetitionen genannt.

#### Aufgabe

Schreibe ein Programm für eine solche Untersuchung, das eine Zeichenkette sowie zwei Zahlen k und l einliest und diejenigen
Teilzeichenketten ausgibt, die
- lang sind (mindestens l Zeichen),
- häufig (mindestens k-mal) in der Zeichenkette auftreten und
- maximal (nicht Teil einer längeren und gleich häufigen Teilzeichenkette) sind.

Wir betrachten als Beispiel die folgende Zeichenkette: CAGGAGGATTA
- Häufige Teilzeichenketten (k = 2): A (kommt 4-mal vor), AG (2), AGG (2), AGGA (2), G (4), GA (2), GG (2), GGA (2), T (2)
- Maximale häufige Teilzeichenketten (k = 2): A, AGGA, G, T
- Ausgabe (k = 2, l = 1): A, AGGA, G, T
- Ausgabe (k = 2, l = 2): AGGA

Wende dein Programm auf die unter bundeswettbewerb-informatik.de abgelegten Beispieleingaben an und dokumentiere die Ergebnisse. Dein
Programm sollte mindestens Eingaben der Länge 10.000 erfolgreich verarbeiten können.

#### Beispieldaten

In `beispieldaten/` liegen zwei Dateien, jede enthält nur die Buchstaben A, C, G und T (ohne Zeilenumbruch):
- `chrM.fa`: die menschliche Mitochondrien-DNA, 16 571 Zeichen. Die Endung `.fa` steht für das FASTA-Format; dort beginnt eine
  Datei meist mit einer Kopfzeile, die mit `>` anfängt. In dieser Datei fehlt sie, wir überspringen solche Zeilen trotzdem.
- `E.coli.100000`: die ersten 100 000 Zeichen aus dem Genom des Darmbakteriums *Escherichia coli*.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zählen mit einem Dictionary:** `anzahl[u] = anzahl.get(u, 0) + 1`. Für kleine Texte kann man so alle Teilzeichenketten zählen.
- **Positionen statt Zeichenketten speichern:** Eine Teilzeichenkette u merken wir uns als Liste der Stellen, an denen sie beginnt,
  zusammen mit ihrer Länge. Das spart sehr viel Speicher.
- **Gruppen verfeinern:** Hängt man an u ein Zeichen an, zerfällt die Liste der Stellen in höchstens vier Gruppen (für A, C, G, T).
  Das ist dieselbe Idee wie bei einem **Trie** (Präfixbaum) und beim **Suffixbaum**, den die Lösungshinweise verwenden.
- **Laufzeit abschätzen:** Wir zählen die Schritte unseres Verfahrens, statt die Zeit zu messen, und überlegen, wie die Anzahl mit der
  Textlänge n wächst.
- Für Neugierige: **Suffixbaum** und **Suffixarray** mit **LCP-Feld** (längstes gemeinsames Präfix benachbarter Suffixe) und einem
  **Stapel** (Stack).

**Nützliche Funktionen und Bibliotheken**
- `''.join(text.split())` entfernt alle Leerzeichen und Zeilenumbrüche aus einem Text.
- `set()` ist eine Menge: Jedes Element kommt nur einmal vor; `len(menge)` sagt, wie viele verschiedene es gibt.
- `text.find(u, start)` liefert die erste Stelle ab `start`, an der `u` in `text` steht, oder `-1`. Damit zählen wir zur Probe nach.
- Tupel werden beim Sortieren erst nach dem ersten Element verglichen, bei Gleichheit nach dem zweiten usw. Mit `(-anzahl, -laenge, u)`
  sortiert `sort()` also nach Häufigkeit absteigend, dann nach Länge absteigend, dann alphabetisch.

## Die Idee in Worten

1. **Was heißt maximal?** u ist *nicht* maximal, wenn es eine längere, gleich häufige Teilzeichenkette w gibt, die u enthält. Dann
   enthält w auch u mit einem Zeichen davor oder dahinter, also cu oder uc. Jedes Vorkommen von w liefert ein Vorkommen von cu (bzw.
   uc) und jedes Vorkommen von cu eines von u; also ist auch cu (bzw. uc) gleich häufig wie u. Es reicht daher, **um ein Zeichen zu
   verlängern**: u ist maximal, wenn weder cu noch uc für ein Zeichen c genauso oft vorkommt wie u.
   - **rechts maximal:** Nicht alle Vorkommen von u haben dasselbe Zeichen dahinter (oder ein Vorkommen endet am Textende).
   - **links maximal:** Nicht alle Vorkommen von u haben dasselbe Zeichen davor (oder ein Vorkommen steht am Textanfang).
2. **Vorkommen zählen wir mit Überlappung**, z. B. kommt AA in AAA zweimal vor. So machen es auch die Lösungshinweise.
3. **Einfach, aber nur für kurze Texte:** alle Teilzeichenketten in einem Dictionary zählen und jede mit ihren Verlängerungen um ein
   Zeichen vergleichen. Bei n Zeichen gibt es n · (n + 1) / 2 Teilzeichenketten, für `chrM.fa` schon über 137 Millionen.
4. **Besser:** Eine Teilzeichenkette ist durch die **Liste ihrer Anfangsstellen** gegeben. Wir beginnen mit der leeren Zeichenkette (sie
   beginnt an jeder Stelle) und hängen Länge für Länge ein Zeichen an: Die Liste zerfällt nach dem nächsten Zeichen in Gruppen. Gruppen
   mit weniger als k Stellen sind nicht häufig, und keine ihrer Verlängerungen ist es; sie fallen weg.
5. Bei jeder Gruppe sehen wir sofort, ob sie **rechts maximal** ist (sie zerfällt in mehrere Gruppen oder ein Vorkommen endet am
   Textende) und ob sie **links maximal** ist (verschiedene Zeichen davor). Ist beides der Fall und die Länge mindestens l, geben wir
   sie aus.
6. **Ausgabe:** zuerst die Anzahl, dann die Teilzeichenketten nach Häufigkeit sortiert (so empfehlen es die Lösungshinweise), bei langen
   Listen gekürzt. **Probe:** Vergleich mit der einfachen Lösung und Nachzählen mit `find`.

## Schritt 1: Einlesen

Wir lesen die ganze Datei, lassen Kopfzeilen mit `>` weg und entfernen alle Zeilenumbrüche und Leerzeichen.

```python
NAMEN = ['chrM.fa', 'E.coli.100000']


def lies_text(dateiname):
    teile = []
    with open(dateiname, encoding='utf-8') as datei:
        for zeile in datei:
            if not zeile.startswith('>'):       # FASTA-Kopfzeile überspringen
                teile.append(zeile.strip())
    return ''.join(''.join(teile).split())     # alle Leerzeichen entfernen


texte = {}
for name in NAMEN:
    texte[name] = lies_text('beispieldaten/' + name)
    print(f'{name}: {len(texte[name])} Zeichen, Zeichen {sorted(set(texte[name]))}, Anfang {texte[name][:40]}')
```

## Schritt 2: Die einfache Lösung für kurze Texte

`zaehle_alle` geht alle Anfänge i und Enden j durch und zählt jede Teilzeichenkette `text[i:j]`. Damit prüfen wir das Beispiel der
Aufgabe.

```python
def zaehle_alle(text):
    anzahl = {}
    for i in range(len(text)):
        for j in range(i + 1, len(text) + 1):
            u = text[i:j]
            anzahl[u] = anzahl.get(u, 0) + 1
    return anzahl


BEISPIEL = 'CAGGAGGATTA'
anzahl_bsp = zaehle_alle(BEISPIEL)
haeufig_bsp = []
for u in anzahl_bsp:
    if anzahl_bsp[u] >= 2:
        haeufig_bsp.append(u)
haeufig_bsp.sort()
print('häufig (k = 2):', ', '.join([f'{u} ({anzahl_bsp[u]})' for u in haeufig_bsp]))
```

Genau die Liste der Aufgabe. Jetzt die Maximalität: Eine häufige Teilzeichenkette u ist maximal, wenn keine Verlängerung `c + u` oder
`u + c` um ein Zeichen genauso oft vorkommt (siehe „Die Idee in Worten“, Punkt 1).

```python
def maximale_naiv(text, k, l):
    anzahl = zaehle_alle(text)
    zeichen = set(text)
    ergebnis = []
    for u in anzahl:
        if anzahl[u] >= k and len(u) >= l:
            maximal = True
            for c in zeichen:
                if anzahl.get(c + u, 0) == anzahl[u] or anzahl.get(u + c, 0) == anzahl[u]:
                    maximal = False
            if maximal:
                ergebnis.append(u)
    ergebnis.sort()
    return ergebnis


print('k = 2, l = 1:', maximale_naiv(BEISPIEL, 2, 1))
print('k = 2, l = 2:', maximale_naiv(BEISPIEL, 2, 2))
print('Teilzeichenketten von chrM.fa (mit Wiederholungen):', len(texte['chrM.fa']) * (len(texte['chrM.fa']) + 1) // 2)
print('Teilzeichenketten von E.coli.100000:               ', len(texte['E.coli.100000']) * (len(texte['E.coli.100000']) + 1) // 2)
```

Das Beispiel stimmt. Für die echten Dateien ist dieser Weg aber hoffnungslos: Das Dictionary müsste über hundert Millionen bzw. Milliarden
Einträge aufnehmen, viele davon Tausende Zeichen lang. Speicher und Zeit wachsen mindestens quadratisch mit n, die Lösungshinweise nennen
quadratischen Speicher ausdrücklich „nicht akzeptabel“. Wir brauchen `maximale_naiv` aber weiter als **Probe** für kurze Texte.

## Schritt 3: Positionen statt Zeichenketten

Eine Teilzeichenkette u der Länge `laenge` speichern wir als Liste ihrer Anfangsstellen. In CAGGAGGATTA (Stellen ab 0 gezählt) beginnt
A an den Stellen 1, 4, 7 und 10. Hängen wir ein Zeichen an, schauen wir an jeder Stelle p auf das Zeichen `text[p + laenge]` dahinter
und sortieren p in die passende Gruppe. Bei p = 10 endet A am Textende; dort gibt es kein Zeichen dahinter.

```python
def verlaengere(text, positionen, laenge):
    """positionen: Anfänge einer Teilzeichenkette u mit der Länge laenge.
    Ergebnis: Dictionary Zeichen c -> Liste der Anfänge von u + c."""
    gruppen = {}
    for p in positionen:
        if p + laenge < len(text):              # sonst endet u hier am Textende
            c = text[p + laenge]
            if c not in gruppen:
                gruppen[c] = []
            gruppen[c].append(p)
    return gruppen


print('leere Zeichenkette:', verlaengere(BEISPIEL, list(range(len(BEISPIEL))), 0))
print('A an 1, 4, 7, 10:  ', verlaengere(BEISPIEL, [1, 4, 7, 10], 1))
print('AG an 1, 4:        ', verlaengere(BEISPIEL, [1, 4], 2))
```

Aus der leeren Zeichenkette entstehen die Gruppen für A, C, G und T. A zerfällt in AG (Stellen 1 und 4) und AT (Stelle 7). AG dagegen
zerfällt nicht: Hinter beiden Vorkommen steht G, also kommt AGG genauso oft vor wie AG, und AG ist **nicht rechts maximal**.

## Schritt 4: Rechts und links maximal

- **rechts maximal:** Keine der Gruppen aus `verlaengere` enthält alle Stellen. (Endet ein Vorkommen am Textende, fehlt diese Stelle in
  jeder Gruppe; dann ist u automatisch rechts maximal.)
- **links maximal:** Wir sammeln die Zeichen vor allen Vorkommen in einer Menge. Sind es mindestens zwei verschiedene, ist u links
  maximal. Steht ein Vorkommen ganz am Anfang (p = 0), kann man dort nicht nach links verlängern; dann ist u ebenfalls links maximal.

```python
def rechts_maximal(positionen, gruppen):
    for gruppe in gruppen.values():
        if len(gruppe) == len(positionen):      # alle Vorkommen haben dasselbe Zeichen dahinter
            return False
    return True


def links_maximal(text, positionen):
    davor = set()
    for p in positionen:
        if p == 0:
            return True                         # Vorkommen am Textanfang
        davor.add(text[p - 1])
    return len(davor) >= 2


for u, pos in [('A', [1, 4, 7, 10]), ('AG', [1, 4]), ('GGA', [2, 5]), ('AGGA', [1, 4])]:
    print(f'{u:5} rechts maximal: {rechts_maximal(pos, verlaengere(BEISPIEL, pos, len(u)))!s:5}  '
          f'links maximal: {links_maximal(BEISPIEL, pos)}')
```

GGA ist rechts maximal (dahinter G bzw. T), aber nicht links maximal: Vor beiden Vorkommen steht A, und AGGA kommt genauso oft vor.
AGGA ist in beide Richtungen maximal (davor C bzw. G, dahinter G bzw. T).

## Schritt 5: Das ganze Verfahren

Wir arbeiten Länge für Länge: `gruppen` enthält die Stellenlisten aller häufigen Teilzeichenketten der aktuellen Länge. Jede Gruppe
wird mit `verlaengere` aufgeteilt; die Teile mit mindestens k Stellen kommen in die Liste für die nächste Länge. Eine Gruppe, die rechts
und links maximal und lang genug ist, kommt ins Ergebnis, und zwar als Tupel `(anzahl, laenge, start)`; die Zeichenkette selbst ist
`text[start:start + laenge]`.

Als Maß für den Aufwand zählen wir in `schritte`, wie viele Stellen `verlaengere` insgesamt angesehen hat. Für k = 1 ist zusätzlich der
ganze Text maximal (er kommt einmal vor, und jede Teilzeichenkette, die nur einmal vorkommt, ist Teil von ihm). Das behandeln wir
gesondert, sonst würde das Verfahren jede einzelne Stelle bis zum Textende verfolgen.

```python
def maximale_teilstrings(text, k, l):
    """Alle maximalen Teilzeichenketten, die mindestens k-mal vorkommen und mindestens l Zeichen haben.
    Ergebnis: (Liste von Tupeln (anzahl, laenge, start), Anzahl der Schritte)."""
    if k < 2:
        ergebnis, schritte = maximale_teilstrings(text, 2, l)
        if len(text) >= l:
            ergebnis.append((1, len(text), 0))  # der ganze Text
        return ergebnis, schritte
    ergebnis = []
    schritte = 0
    gruppen = [list(range(len(text)))]          # Länge 0: die leere Zeichenkette beginnt überall
    laenge = 0
    while len(gruppen) > 0:
        naechste = []
        for positionen in gruppen:
            teile = verlaengere(text, positionen, laenge)
            schritte += len(positionen)
            if (laenge >= 1 and laenge >= l and rechts_maximal(positionen, teile)
                    and links_maximal(text, positionen)):
                ergebnis.append((len(positionen), laenge, positionen[0]))
            for teil in teile.values():
                if len(teil) >= k:              # nur häufige Teilzeichenketten weiter verlängern
                    naechste.append(teil)
        gruppen = naechste
        laenge += 1
    return ergebnis, schritte


def als_text(text, ergebnis):
    """Die Zeichenketten des Ergebnisses, alphabetisch sortiert."""
    liste = []
    for anzahl, laenge, start in ergebnis:
        liste.append(text[start:start + laenge])
    liste.sort()
    return liste


erg_bsp, schritte_bsp = maximale_teilstrings(BEISPIEL, 2, 1)
print('k = 2, l = 1:', als_text(BEISPIEL, erg_bsp), erg_bsp, schritte_bsp, 'Schritte')
print('k = 2, l = 2:', als_text(BEISPIEL, maximale_teilstrings(BEISPIEL, 2, 2)[0]))
print('k = 1, l = 3:', als_text(BEISPIEL, maximale_teilstrings(BEISPIEL, 1, 3)[0]))
```

Beide Ausgaben der Aufgabe stimmen. Mit k = 1 kommt der ganze Text dazu (AGGA kommt mindestens einmal vor und ist weiter maximal).

**Warum ist das richtig?** Jede Teilzeichenkette u, die mindestens k-mal vorkommt, wird erreicht: Alle ihre Anfänge (Präfixe) kommen
mindestens so oft vor wie u, ihre Gruppen werden also nie verworfen. Bei u angekommen, enthält die Gruppe genau die Anfangsstellen von u,
und die beiden Prüfungen entsprechen genau der Bedingung „weder cu noch uc gleich häufig“.

**Speicher:** Auf jeder Länge gehört jede Stelle zu höchstens einer Gruppe, `gruppen` und `naechste` enthalten also zusammen höchstens 2n
Zahlen. Der Speicher wächst nur linear mit n.

**Probe** gegen `maximale_naiv`: 300 Zufallstexte bis 40 Zeichen (auch nur aus A und C, damit es viele Wiederholungen gibt) mit
zufälligem k und l, dazu die ersten 300 Zeichen beider Beispieldateien.

```python
import random

zufall = random.Random(1)
for versuch in range(300):
    alphabet = 'AC' if versuch % 2 == 0 else 'ACGT'
    probe_text = ''
    for i in range(zufall.randint(1, 40)):
        probe_text += zufall.choice(alphabet)
    k_probe = zufall.randint(1, 4)
    l_probe = zufall.randint(1, 4)
    assert als_text(probe_text, maximale_teilstrings(probe_text, k_probe, l_probe)[0]) == maximale_naiv(probe_text, k_probe, l_probe)
for name in NAMEN:
    anfang = texte[name][:300]
    assert als_text(anfang, maximale_teilstrings(anfang, 2, 1)[0]) == maximale_naiv(anfang, 2, 1)
print('Probe in Ordnung: gleiche Ergebnisse wie maximale_naiv')
```

## Schritt 6: Übersichtliche Ausgabe

Zuerst die Anzahl, dann nach Häufigkeit sortiert. Bei langen Listen zeigen wir nur die ersten Zeilen und zusätzlich die längsten
Teilzeichenketten. Sehr lange Zeichenketten kürzen wir auf 60 Zeichen. Dazu geben wir eine Stelle an, an der die Zeichenkette
beginnt (bei unserem Verfahren die erste).

```python
def kurz(u):
    if len(u) <= 60:
        return u
    return u[:57] + '...'


def zeige(text, ergebnis, zeilen=6):
    print(len(ergebnis), 'maximale Teilzeichenketten')
    nach_anzahl = []
    nach_laenge = []
    for anzahl, laenge, start in ergebnis:
        u = text[start:start + laenge]
        nach_anzahl.append((-anzahl, -laenge, u, start))
        nach_laenge.append((-laenge, -anzahl, u, start))
    nach_anzahl.sort()
    nach_laenge.sort()
    if len(ergebnis) > zeilen:
        print('  die häufigsten:')
    for minus_anzahl, minus_laenge, u, start in nach_anzahl[:zeilen]:
        print(f'  {-minus_anzahl:6}-mal  Länge {-minus_laenge:3}  an Stelle {start:6}  {kurz(u)}')
    if len(ergebnis) > zeilen:
        print('  die längsten:')
        for minus_laenge, minus_anzahl, u, start in nach_laenge[:zeilen]:
            print(f'  {-minus_anzahl:6}-mal  Länge {-minus_laenge:3}  an Stelle {start:6}  {kurz(u)}')


zeige(BEISPIEL, erg_bsp)
```

## Schritt 7: Alle Beispiele

Für jede Datei rechnen wir mit verschiedenen Werten für k und l, so wie es die Bewertungskriterien verlangen.

```python
ergebnisse = {}
for k, l in [(2, 1), (2, 12), (3, 8), (20, 5)]:
    ergebnisse[('chrM.fa', k, l)], schritte = maximale_teilstrings(texte['chrM.fa'], k, l)
    print(f'chrM.fa, k = {k}, l = {l}: {schritte} Schritte')
    zeige(texte['chrM.fa'], ergebnisse[('chrM.fa', k, l)])
    print()
```

Die längste Teilzeichenkette, die in der Mitochondrien-DNA zweimal vorkommt, hat 15 Zeichen (CAAACTCAAACTACG). Mit l = 12 bleiben nur
22 Treffer. Fordert man k = 3, ist die längste Wiederholung nur noch 10 Zeichen lang.

```python
for k, l in [(2, 1), (2, 20), (3, 15), (50, 6)]:
    ergebnisse[('E.coli.100000', k, l)], schritte = maximale_teilstrings(texte['E.coli.100000'], k, l)
    print(f'E.coli.100000, k = {k}, l = {l}: {schritte} Schritte')
    zeige(texte['E.coli.100000'], ergebnisse[('E.coli.100000', k, l)])
    print()
```

Bei E. coli gibt es deutlich längere Wiederholungen: Ein Stück aus 84 Zeichen kommt zweimal vor. Schön sieht man hier, was „maximal“
bedeutet: Ein 50 Zeichen langes Stück kommt zweimal vor; seine ersten 42 Zeichen kommen dreimal vor. Für k = 2 sind beide maximal (die
42 Zeichen sind zwar Teil der 50, aber nicht gleich häufig), für k = 3 bleibt nur das kürzere.

**Probe durch Nachzählen.** `vorkommen` zählt mit `find` (auch überlappend), wie oft u im Text steht. Für jedes Ergebnis prüfen wir die
Anzahl, die Bedingungen für k und l und dass keine Verlängerung um ein Zeichen gleich häufig ist. Das prüft, dass alles Ausgegebene
stimmt; dass nichts fehlt, hat schon der Vergleich mit `maximale_naiv` gezeigt (und im Abschnitt „Für Neugierige“ der Vergleich mit einem
ganz anderen Verfahren). Wir prüfen die Ergebnisse mit großem l, da kurze, häufige Zeichenketten mit `find` lange dauern würden.

```python
def vorkommen(text, u):
    anzahl = 0
    p = text.find(u)
    while p != -1:
        anzahl += 1
        p = text.find(u, p + 1)                 # p + 1: auch überlappende Vorkommen
    return anzahl


def probe(text, ergebnis, k, l):
    for anzahl, laenge, start in ergebnis:
        u = text[start:start + laenge]
        if vorkommen(text, u) != anzahl or anzahl < k or laenge < l:
            return 'Fehler bei ' + u
        for c in 'ACGT':
            if vorkommen(text, c + u) == anzahl or vorkommen(text, u + c) == anzahl:
                return 'nicht maximal: ' + u
    return f'in Ordnung ({len(ergebnis)} Ergebnisse geprüft)'


for name, k, l in [('chrM.fa', 2, 12), ('chrM.fa', 3, 8), ('E.coli.100000', 2, 20), ('E.coli.100000', 3, 15)]:
    print(name, k, l, probe(texte[name], ergebnisse[(name, k, l)], k, l))
```

## Schritt 8: Wie viel Arbeit ist das?

Die Lösungshinweise bemängeln Verfahren, deren Laufzeit bei kleinem k stark wächst. Wir zählen die Schritte für verschiedene k (l spielt
für den Aufwand keine Rolle, es filtert nur die Ausgabe).

```python
print(f'{"k":>4} | {"chrM.fa":>20} | {"E.coli.100000":>20}')
print(f'{"":>4} | {"Ergebnisse  Schritte":>20} | {"Ergebnisse  Schritte":>20}')
tabelle = {}
for k in [2, 3, 5, 10, 50]:
    zeile = f'{k:>4}'
    for name in NAMEN:
        erg, schritte = maximale_teilstrings(texte[name], k, 1)
        tabelle[(name, k)] = (len(erg), schritte)
        zeile += f' | {len(erg):>10} {schritte:>9}'
    print(zeile)
```

Von k = 50 bis k = 2 steigt die Zahl der Ergebnisse bei beiden Dateien auf etwa das 37-Fache, die Zahl der Schritte nur auf das
1,6- bis 1,7-Fache. Pro Zeichen sind es bei E. coli knapp 10
Schritte: Jede Stelle p wird so oft angesehen, wie die Teilzeichenketten ab p (Länge 0, 1, 2, …) häufig sind, plus einmal. In einem
Text, dessen Zeichen „zufällig“ aussehen, kommt eine Teilzeichenkette der Länge L aus vier Buchstaben etwa n / 4ᴸ-mal vor; schon ab
L ≈ 9 (4⁹ ≈ 262 000) wird sie meist selten. Für solche Texte ist der Aufwand ungefähr n · log₄(n), also fast linear.

**Der schlechteste Fall.** Besteht der Text nur aus dem Buchstaben A, kommt jedes Stück aus m A's (n − m + 1)-mal vor. Dann wird jede
Stelle bis fast zum Ende verfolgt, und die Zahl der Schritte wächst **quadratisch**:

```python
for n in [500, 1000, 2000]:
    erg, schritte = maximale_teilstrings('A' * n, 2, 1)
    print(f'n = {n:5}: {len(erg):5} Ergebnisse, {schritte:9} Schritte, n * (n + 1) / 2 = {n * (n + 1) // 2}')
```

Jede Verdopplung von n vervierfacht die Arbeit; für einen Text aus 10 000 A wären es 50 Millionen Schritte. Bei DNA kommt so etwas kaum
vor, unser Programm schafft die 100 000 Zeichen von E. coli mit unter einer Million Schritten. Die Hinweise nennen eine quadratische
Laufzeit ausdrücklich „akzeptabel“; ein Verfahren ohne diese Schwäche zeigt der nächste Abschnitt.

### Unser Verfahren baut einen Suffixbaum

Die Lösungshinweise verwenden einen **Suffixbaum**: Alle Endstücke (Suffixe) des Textes, jedes mit einem Schlusszeichen `$`, werden in
einen Baum eingetragen. Von der Wurzel bis zu einem Blatt liest man ein Suffix ab, das Blatt trägt die Nummer der Stelle, an der es
beginnt. Gemeinsame Anfänge teilen sich den Weg, und Wege ohne Verzweigung werden zu einer Kante zusammengefasst. Jeder **innere
Knoten** steht für eine Teilzeichenkette, die so oft vorkommt, wie unter ihm Blätter hängen.

Unsere Gruppen sind genau diese Knoten: Eine Gruppe, die rechts maximal ist, verzweigt sich; das ist ein innerer Knoten. Eine Gruppe,
die nicht rechts maximal ist, liegt mitten auf einer Kante. Mit `verlaengere` und `rechts_maximal` können wir den Baum deshalb direkt
ausgeben. Zu jedem inneren Knoten stehen die Kantenbeschriftung, die ganze Zeichenkette, ihre Häufigkeit und ihre Länge.

```python
def zeige_baum(text, positionen, laenge, einrueckung):
    for p in positionen:
        if p + laenge == len(text):             # dieses Suffix endet hier: Kante nur mit $
            print(einrueckung + '$  Blatt', p)
    teile = verlaengere(text, positionen, laenge)
    for c in sorted(teile):
        teil = teile[c]
        if len(teil) == 1:                      # nur noch ein Vorkommen: Kante bis zum Blatt
            print(einrueckung + text[teil[0] + laenge:] + '$  Blatt', teil[0])
        else:
            neue_laenge = laenge + 1            # ohne Verzweigung weiter: Kante zusammenfassen
            while not rechts_maximal(teil, verlaengere(text, teil, neue_laenge)):
                neue_laenge += 1
            p = teil[0]
            print(f'{einrueckung}{text[p + laenge:p + neue_laenge]}  -> {text[p:p + neue_laenge]} '
                  f'({len(teil)}-mal, Länge {neue_laenge})')
            zeige_baum(text, teil, neue_laenge, einrueckung + '    ')


print('Wurzel')
zeige_baum(BEISPIEL, list(range(len(BEISPIEL))), 0, '    ')
```

Die inneren Knoten sind A, AGGA, G, GA, GGA und T, dieselben wie in Abbildung 6 der Lösungshinweise. Der Weg zu Blatt 5 ist mit G, GA
und TTA$ beschriftet und ergibt das Suffix GGATTA, wie dort beschrieben.

Die Hinweise entfernen danach (Algorithmus 1) alle Knoten, deren Zeichenkette Endstück einer gleich häufigen Knoten-Zeichenkette ist, z. B.
GGA als Endstück von AGGA. Wir prüfen stattdessen `links_maximal`; das ist dasselbe, denn ist u ein Knoten und cu gleich häufig, so ist
auch cu ein Knoten. Ihr Baumaufbau braucht wie unser Verfahren im schlechtesten Fall quadratische Zeit. Es gibt Verfahren, die den
Suffixbaum in **linearer Zeit** bauen (Ukkonen 1995), sie sind aber recht kompliziert; die Hinweise verweisen dafür auf die Literatur,
z. B. Dan Gusfield: *Algorithms on Strings, Trees and Sequences* (1997).

### Suffixarray: auch der schlechteste Fall wird schnell

Ein **Suffixarray** ist die Liste aller Suffixe, alphabetisch sortiert (gespeichert werden nur ihre Anfangsstellen). Das Feld **LCP**
gibt für zwei benachbarte Suffixe an, wie lang ihr gemeinsamer Anfang ist. Für CAGGAGGATTA:

```python
def suffixarray(text):
    """Sortieren durch Verdoppeln: Erst nach dem ersten Zeichen, dann nach den ersten 2, 4, 8, ... Zeichen.
    rang[i] ist die Platznummer des Suffixes ab i, wenn man nur die ersten h Zeichen vergleicht. Die ersten 2h Zeichen
    vergleicht man über das Paar (rang[i], rang[i + h]); Tupel lassen sich direkt sortieren."""
    n = len(text)
    rang = []
    for c in text:
        rang.append(ord(c))                     # Rang nach dem ersten Zeichen: der Zeichencode
    h = 1
    runden = 0
    while True:
        runden += 1
        schluessel = []
        for i in range(n):
            if i + h < n:
                schluessel.append((rang[i], rang[i + h], i))
            else:
                schluessel.append((rang[i], -1, i))     # kürzeres Suffix kommt zuerst
        schluessel.sort()
        neuer_rang = [0] * n
        r = 0
        for x in range(n):
            if x > 0 and schluessel[x][:2] != schluessel[x - 1][:2]:
                r += 1
            neuer_rang[schluessel[x][2]] = r
        rang = neuer_rang
        if r == n - 1 or h >= n:                # alle Ränge verschieden: fertig sortiert
            sa = []
            for eintrag in schluessel:
                sa.append(eintrag[2])
            return sa, runden
        h *= 2


def lcp_feld(text, sa):
    """lcp[x] = Länge des gemeinsamen Anfangs der Suffixe sa[x - 1] und sa[x] (Verfahren von Kasai u. a., 2001).
    Trick: Geht man im Text von Suffix i zu i + 1, wird der gemeinsame Anfang höchstens um 1 kürzer."""
    n = len(text)
    platz = [0] * n
    for x in range(n):
        platz[sa[x]] = x
    lcp = [0] * n
    h = 0
    for i in range(n):
        if platz[i] > 0:
            j = sa[platz[i] - 1]                # das Suffix davor im Suffixarray
            while i + h < n and j + h < n and text[i + h] == text[j + h]:
                h += 1
            lcp[platz[i]] = h
            if h > 0:
                h -= 1
        else:
            h = 0
    return lcp


sa_bsp, runden_bsp = suffixarray(BEISPIEL)
lcp_bsp = lcp_feld(BEISPIEL, sa_bsp)
for x in range(len(BEISPIEL)):
    print(f'{x:2}  Stelle {sa_bsp[x]:2}  LCP {lcp_bsp[x]}  {BEISPIEL[sa_bsp[x]:]}')
```

Alle Vorkommen einer Teilzeichenkette stehen im Suffixarray **hintereinander**, z. B. die vier Suffixe, die mit A beginnen (Zeilen 0
bis 3). Einem inneren Knoten des Suffixbaums entspricht ein Abschnitt, in dem alle LCP-Werte (außer dem ersten) mindestens L sind und
mindestens einer genau L ist; am Rand ist der LCP-Wert kleiner. So ein **LCP-Intervall** findet man mit einem Stapel in einem Durchgang:
Steigt der LCP-Wert, beginnt ein neues Intervall; fällt er, enden Intervalle. Für links maximal merken wir uns zu jedem Intervall das
Zeichen vor seinen Suffixen, oder `*`, sobald es verschiedene sind (oder ein Suffix am Textanfang beginnt). Endet ein Intervall, gibt es
sein Zeichen an das umschließende weiter.

```python
def maximale_mit_suffixarray(text, k, l):
    n = len(text)
    sa, runden = suffixarray(text)
    lcp = lcp_feld(text, sa)

    def davor(x):                               # Zeichen vor dem Suffix auf Platz x
        if sa[x] == 0:
            return '*'                          # Textanfang: zählt wie „verschiedene Zeichen“
        return text[sa[x] - 1]

    def mische(a, b):                           # None = noch kein Zeichen, '*' = verschiedene Zeichen
        if a is None:
            return b
        if b is None or a == b:
            return a
        return '*'

    ergebnis = []
    stapel = [[0, 0, None]]                     # offene Intervalle: [LCP-Wert, erster Platz, Zeichen davor]
    for x in range(1, n + 1):
        wert = lcp[x] if x < n else -1          # am Ende alle Intervalle schließen
        stapel[-1][2] = mische(stapel[-1][2], davor(x - 1))   # Suffix x - 1 gehört zum obersten Intervall
        erster = x - 1
        zeichen = None
        while len(stapel) > 0 and stapel[-1][0] > wert:      # Intervalle mit größerem LCP-Wert enden hier
            L, anfang, ch = stapel.pop()
            anzahl = x - anfang                 # Plätze anfang .. x - 1
            if ch == '*' and anzahl >= k and L >= l and L > 0:
                ergebnis.append((anzahl, L, sa[anfang]))
            erster = anfang
            zeichen = ch
            if len(stapel) > 0 and stapel[-1][0] >= wert:     # das umschließende Intervall ist schon offen
                stapel[-1][2] = mische(stapel[-1][2], ch)
                zeichen = None
        if x < n and stapel[-1][0] < wert:      # neues Intervall beginnt
            if zeichen is None:
                zeichen = davor(x - 1)
            stapel.append([wert, erster, zeichen])
    return ergebnis, runden


def gleich(text, erg1, erg2):
    menge1 = set()
    for anzahl, laenge, start in erg1:
        menge1.add((anzahl, text[start:start + laenge]))
    menge2 = set()
    for anzahl, laenge, start in erg2:
        menge2.add((anzahl, text[start:start + laenge]))
    return menge1 == menge2


print('Beispiel:', als_text(BEISPIEL, maximale_mit_suffixarray(BEISPIEL, 2, 1)[0]))
for name in NAMEN:
    for k, l in [(2, 1), (3, 8)]:
        erg_sa, runden = maximale_mit_suffixarray(texte[name], k, l)
        print(f'{name}, k = {k}, l = {l}: {len(erg_sa)} Ergebnisse, {runden} Runden Verdoppeln, '
              f'gleich wie oben: {gleich(texte[name], erg_sa, maximale_teilstrings(texte[name], k, l)[0])}')
```

Beide Verfahren liefern dieselben Ergebnisse. Das Suffixarray braucht einige Runden, in denen jeweils n Tupel sortiert werden; jede Runde
kostet etwa n · log(n) Schritte, und es gibt höchstens log₂(n) Runden. Der Text aus lauter A ist damit kein Problem mehr, auch nicht
mit 100 000 Zeichen (dort braucht unser erstes Verfahren 5 Milliarden Schritte):

```python
viele_a = 'A' * 100000
erg_a, runden_a = maximale_mit_suffixarray(viele_a, 2, 1)
print(len(erg_a), 'Ergebnisse,', runden_a, 'Runden Verdoppeln')
zeige(viele_a, erg_a, 3)
```

Die Idee, Suffixbäume durch Suffixarrays mit LCP-Feld zu ersetzen, stammt von Abouelhoda, Kurtz und Ohlebusch (2004). Suffixarrays lassen
sich sogar in linearer Zeit sortieren (z. B. mit dem Verfahren SA-IS); dann läuft das ganze Programm in linearer Zeit.

MiniZinc ist stark, wenn man *eine* Lösung sucht, die viele Bedingungen erfüllt, oder eine beste. Hier sollen dagegen *alle* maximalen
Teilzeichenketten aufgezählt werden, oft Zehntausende. Ein Modell müsste für jede Kandidatin Anfang, Länge und alle Vorkommen als
Variablen über 100 000 Textstellen haben und jede Lösung einzeln finden. Das wäre viel langsamer als unsere Gruppen und brächte keine
neue Einsicht; deshalb lassen wir MiniZinc bei dieser Aufgabe weg.
