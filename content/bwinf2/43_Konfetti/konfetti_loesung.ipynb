## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Backtracking (Tiefensuche):** Wir legen die Spalten eine nach der anderen von links nach rechts. Geht es nicht weiter, nehmen
  wir die letzte Spalte zurück und probieren die nächste. So findet man eine Lösung, wenn es eine gibt, und weiß sonst sicher,
  dass es keine gibt.
- **Regeln zum Abschneiden:** Je früher man merkt, dass ein Anfang nicht zu einer Lösung führen kann, desto weniger muss man
  probieren. Die wichtigste Regel ist hier: Ein angefangener Block ✓✓✓ darf erst durch ein × beendet werden, wenn alle ✓ der Zeile
  gelegt sind.
- **Sackgassen merken:** Was nach einem Anfang noch geht, hängt nur davon ab, *welche* Spalten schon liegen, nicht von ihrer
  Reihenfolge. Haben wir mit denselben restlichen Spalten schon einmal vergeblich gesucht, brauchen wir es nicht noch einmal.
- **Gleiche Spalten zusammenfassen:** Zwei gleiche Spalten zu vertauschen ergibt dieselbe Tabelle. Wir zählen deshalb, wie oft
  jede Spalte vorkommt, und probieren jede *verschiedene* Spalte nur einmal.
- **Größte Teilsumme:** Bei Teilaufgabe e) braucht man für eine feste Anordnung die kleinste Zahl von Änderungen in einer Zeile.
  Das ist eine Aufgabe „größte Summe eines zusammenhängenden Stücks“, die man in einem Durchlauf löst.
- **Bergsteigen mit Neustarts:** Eine Anordnung immer wieder ein wenig verändern (eine Spalte an eine andere Stelle schieben) und
  die Änderung behalten, wenn sie besser ist. Das liefert gute, aber nicht sicher die besten Lösungen.
- **Branch and Bound:** Backtracking mit einer **unteren Schranke**: Kann ein Anfang nicht mehr besser werden als die beste
  bekannte Lösung, schneidet man ihn ab. So kann man beweisen, dass eine Lösung optimal ist.

**Nützliche Funktionen und Bibliotheken**
- `set` und `dict`, Tupel als Schlüssel (Tupel sind unveränderlich, Listen nicht)
- `sorted(set(liste))` für die verschiedenen Einträge einer Liste, `liste.count(x)`
- **Sortieren von Tupeln:** `sort()` vergleicht Tupel zuerst nach dem ersten Eintrag, bei Gleichstand nach dem zweiten usw. So
  sortiert man nach mehreren Gesichtspunkten.
- `random.Random(1)` für Zufallszahlen, die bei jedem Lauf gleich sind, `shuffle`, `itertools.permutations` für Tests
- Rekursion: Python erlaubt normalerweise nur 1000 Aufrufe ineinander. Mit `sys.setrecursionlimit` erhöht man die Grenze, denn
  bei 1020 Spalten ruft sich die Suche 1020-mal ineinander auf.

## Die Idee

1. **Zulässig prüfen:** In jeder Zeile dürfen zwischen zwei ✓ keine × stehen. Ein ? kann alles sein, deshalb lassen wir es beim
   Prüfen einfach weg.
2. **Von links nach rechts legen:** Wir merken uns für jede Zeile, wie viele ✓ schon liegen und wie viele noch fehlen. Eine
   Spalte darf als nächste kommen, wenn sie keinen angefangenen Block zu früh unterbricht. Wir probieren die Spalten zuerst,
   die wenige neue Blöcke beginnen und viele laufende fortsetzen.
3. **Zweite Regel:** Zwei angefangene Zeilen können sich gegenseitig blockieren. Das erkennen wir früh und sparen damit bei den
   großen Beispielen fast die ganze Suche.
4. **Annas Zeile (c):** Sie gibt für jede Position vor, ob dort in der ersten Zeile ✓ oder × stehen muss. Das ist eine weitere
   Regel beim Legen.
5. **Eindeutig (b):** Wir suchen eine zweite Lösung, die eine andere Tabelle ergibt. Gibt es keine, ist die erste eindeutig.
6. **Wenige Änderungen (e):** Für eine feste Anordnung rechnen wir die kleinste Zahl von Änderungen aus. Dann suchen wir eine
   Anordnung, bei der sie klein ist, zuerst mit Bergsteigen, dann beweisen wir mit Branch and Bound, dass es nicht besser geht.

**Schreibweise:** Wie die Lösungshinweise schreiben wir in den Ausgaben 1 für ✓ und 0 für ×. Das lässt sich in einer
Tabelle gut lesen.

## Schritt 1: Einlesen und anzeigen

Wir speichern jede Spalte als Text aus `r` Zeichen, zum Beispiel `'nyyy'` für die erste Spalte von `konfetti00.txt`. Eine
**Anordnung** ist dann eine Liste solcher Texte. Der Eintrag in Zeile `z` und Spalte `p` ist `anordnung[p][z]`.

Der ganze Aushang ist ein `dict` mit `r`, `c`, der Liste `spalten` (in der zerschnittenen Reihenfolge) und `anna` (Annas
Zeile als Liste oder `None`, wenn sie unbekannt ist).

````python
import sys
sys.setrecursionlimit(10000)


def lies_aushang(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = f.read().split('\n')
    kopf = zeilen[0].split()
    r = int(kopf[0])
    c = int(kopf[1])
    tabelle = []
    for i in range(r):
        tabelle.append(zeilen[1 + i].split())
    anna = None
    if kopf[2] == 'y':
        anna = zeilen[1 + r].split()          # Annas Zeile in der richtigen Reihenfolge
    spalten = []
    for j in range(c):
        spalte = ''
        for i in range(r):
            spalte = spalte + tabelle[i][j]
        spalten.append(spalte)
    return {'r': r, 'c': c, 'spalten': spalten, 'anna': anna}


ZEICHEN = {'y': '1', 'n': '0', '?': '?'}


def zeige(anordnung):
    for z in range(len(anordnung[0])):
        teile = []
        for spalte in anordnung:
            teile.append(ZEICHEN[spalte[z]])
        print(' '.join(teile))


aushang00 = lies_aushang('beispieldaten/konfetti00.txt')
print(aushang00['spalten'])
zeige(aushang00['spalten'])
````

## Schritt 2: Ist eine Anordnung zulässig?

Eine Zeile ist in Ordnung, wenn zwischen dem ersten und dem letzten ✓ kein × steht. Ein ? kann man immer passend ersetzen: ✓,
wenn es zwischen zwei ✓ steht, sonst ×. Deshalb dürfen wir die ? beim Prüfen weglassen.

`passt_zu_anna` vergleicht die erste Zeile der Anordnung mit Annas Zeile (für Teilaufgabe c). Wo eines von beiden ein ? ist,
passt es immer.

`nach_reihenfolge` baut eine Anordnung aus einer Liste von Spaltennummern. Wie in den Lösungshinweisen zählen wir die Spalten
dabei ab 1.

````python
def zeile_ist_zulaessig(zeile):
    ohne = zeile.replace('?', '')               # die ? weglassen
    if 'y' not in ohne:
        return True                             # gar kein ✓: in Ordnung
    erstes = ohne.index('y')
    letztes = ohne.rindex('y')
    return 'n' not in ohne[erstes:letztes + 1]


def ist_zulaessig(anordnung):
    for z in range(len(anordnung[0])):
        zeile = ''
        for spalte in anordnung:
            zeile = zeile + spalte[z]
        if not zeile_ist_zulaessig(zeile):
            return False
    return True


def passt_zu_anna(anordnung, anna):
    if anna is None:
        return True
    for p in range(len(anordnung)):
        ist = anordnung[p][0]
        soll = anna[p]
        if ist != '?' and soll != '?' and ist != soll:
            return False
    return True


def nach_reihenfolge(aushang, reihenfolge):
    anordnung = []
    for nummer in reihenfolge:
        anordnung.append(aushang['spalten'][nummer - 1])
    return anordnung


print('zerschnittene Reihenfolge:', ist_zulaessig(aushang00['spalten']))
versuch = nach_reihenfolge(aushang00, [2, 3, 1, 4])
print('Reihenfolge 2, 3, 1, 4:  ', ist_zulaessig(versuch))
zeige(versuch)
````

Die Reihenfolge 2, 3, 1, 4 steht so in den Lösungshinweisen. Das ? in der zweiten Zeile steht zwischen zwei 1, es wird also zu ✓.

## Schritt 3: Spalten von links nach rechts legen

Wir legen die Spalten nacheinander. Für jede Zeile `z` merken wir uns

- `gelegt[z]`: wie viele ✓ dieser Zeile schon liegen,
- `rest[z]`: wie viele ✓ dieser Zeile noch fehlen.

Ist `gelegt[z] > 0` und `rest[z] > 0`, dann **läuft** in Zeile `z` gerade ein Block: Er hat angefangen, und es kommen noch ✓.
Eine Spalte mit × in dieser Zeile würde den Block unterbrechen, die ✓ danach lägen dann nicht mehr zusammen. Daraus folgt

**Regel 1:** Eine Spalte darf nur gelegt werden, wenn sie in keiner Zeile mit laufendem Block ein × hat.

Ein ? stört nie: In einer Zeile mit laufendem Block wird es ✓, sonst ×. Wenn alle Spalten liegen und Regel 1 immer
eingehalten wurde, ist die Anordnung zulässig. Umgekehrt hält jede zulässige Anordnung Regel 1 ein. Die Suche ist also
**vollständig**: Sie findet eine Lösung, wenn es eine gibt.

**Annas Zeile:** Ist sie bekannt, muss der erste Eintrag der Spalte an Position `p` zu `anna[p]` passen.

Weitere Bausteine:

- **Gleiche Spalten** fassen wir zusammen: `typen` ist die sortierte Liste der verschiedenen Spalten, `anzahl[t]` sagt, wie viele
  vom Typ `t` noch nicht gelegt sind.
- **Reihenfolge der Kandidaten** (aus der Greedy-Idee der Lösungshinweise): zuerst Spalten, die wenige neue Blöcke beginnen, bei
  Gleichstand die, die viele laufende Blöcke fortsetzen. Dafür sortieren wir Tupel `(neu, -weiter, t)`.
- **Sackgassen:** Was noch geht, hängt nur davon ab, *welche* Spalten schon liegen (davon hängen `gelegt`, `rest` und auch die
  Position für Annas Zeile ab). Scheitert die Suche, speichern wir das Tupel `anzahl` in der Menge `sackgassen`.
- **Knoten** zählen wir, um zu sehen, wie viel die Suche probiert. Ohne einen einzigen Umweg braucht sie `c + 1` Knoten.
  Mit `grenze` kann man sie nach einer Zahl von Knoten abbrechen.
- `paare[a][b]` zählt die noch nicht gelegten Spalten mit ✓ in Zeile `a` und × in Zeile `b`. Wir brauchen das erst in Schritt 4,
  halten es aber schon hier beim Legen und Zurücknehmen aktuell.

````python
def neuer_zustand(aushang):
    spalten = aushang['spalten']
    r = aushang['r']
    typen = sorted(set(spalten))               # die verschiedenen Spalten
    anzahl = []
    for typ in typen:
        anzahl.append(spalten.count(typ))
    rest = [0] * r
    for spalte in spalten:
        for z in range(r):
            if spalte[z] == 'y':
                rest[z] += 1
    # paare[a][b]: Spalten mit ✓ in Zeile a und × in Zeile b (Größe einer Schnittmenge)
    mit_haken = []
    mit_kreuz = []
    for z in range(r):
        haken = set()
        kreuz = set()
        for j in range(len(spalten)):
            if spalten[j][z] == 'y':
                haken.add(j)
            elif spalten[j][z] == 'n':
                kreuz.add(j)
        mit_haken.append(haken)
        mit_kreuz.append(kreuz)
    paare = []
    for a in range(r):
        zeile = []
        for b in range(r):
            zeile.append(len(mit_haken[a] & mit_kreuz[b]))
        paare.append(zeile)
    return {'r': r, 'c': aushang['c'], 'anna': aushang['anna'], 'typen': typen, 'anzahl': anzahl,
            'gelegt': [0] * r, 'rest': rest, 'paare': paare, 'folge': [], 'sackgassen': set(),
            'knoten': 0, 'grenze': 10 ** 9, 'abgebrochen': False}


def lege(s, t):
    veraendere(s, t, 1)
    s['folge'].append(t)


def nimm_zurueck(s, t):
    s['folge'].pop()
    veraendere(s, t, -1)


def veraendere(s, t, d):                       # d = 1: legen, d = -1: zurücknehmen
    spalte = s['typen'][t]
    s['anzahl'][t] -= d
    haken = []
    kreuz = []
    for z in range(s['r']):
        if spalte[z] == 'y':
            s['gelegt'][z] += d
            s['rest'][z] -= d
            haken.append(z)
        elif spalte[z] == 'n':
            kreuz.append(z)
    for a in haken:
        for b in kreuz:
            s['paare'][a][b] -= d


def regel_eins(s, t):
    spalte = s['typen'][t]
    for z in range(s['r']):
        if spalte[z] == 'n' and s['gelegt'][z] > 0 and s['rest'][z] > 0:
            return False                       # × mitten in einem laufenden Block
    return True


def regel_anna(s, t):
    if s['anna'] is None:
        return True
    soll = s['anna'][len(s['folge'])]           # Annas Eintrag an der nächsten Position
    ist = s['typen'][t][0]
    return soll == '?' or ist == '?' or soll == ist


def passt_v1(s, t):
    return regel_eins(s, t) and regel_anna(s, t)


def passt(s, t):
    return passt_v1(s, t)


def kandidaten(s):
    liste = []
    for t in range(len(s['typen'])):
        if s['anzahl'][t] > 0 and passt(s, t):
            spalte = s['typen'][t]
            neu = 0
            weiter = 0
            for z in range(s['r']):
                if spalte[z] == 'y':
                    if s['gelegt'][z] == 0:
                        neu += 1                # beginnt einen neuen Block
                    else:
                        weiter += 1             # setzt einen laufenden Block fort
            liste.append((neu, -weiter, t))
    liste.sort()
    ergebnis = []
    for neu, minus_weiter, t in liste:
        ergebnis.append(t)
    return ergebnis


def suche(s):
    s['knoten'] += 1
    if s['knoten'] > s['grenze']:
        s['abgebrochen'] = True
        return False
    if len(s['folge']) == s['c']:
        return True                            # alle Spalten liegen: Lösung in s['folge']
    schluessel = tuple(s['anzahl'])
    if schluessel in s['sackgassen']:
        return False
    for t in kandidaten(s):
        lege(s, t)
        if suche(s):
            return True
        nimm_zurueck(s, t)
        if s['abgebrochen']:
            return False
    s['sackgassen'].add(schluessel)
    return False


def anordnung_aus(s):
    anordnung = []
    for t in s['folge']:
        anordnung.append(s['typen'][t])
    return anordnung


def loese(aushang, grenze=10 ** 9):
    s = neuer_zustand(aushang)
    s['grenze'] = grenze
    if suche(s):
        return anordnung_aus(s), s
    return None, s


def name(nr):
    return 'konfetti' + str(nr).rjust(2, '0')


for nr in [0, 1, 2, 3, 4, 9, 10]:
    aushang = lies_aushang('beispieldaten/' + name(nr) + '.txt')
    anordnung, s = loese(aushang)
    if anordnung is None:
        print(name(nr), 'nicht lösbar,', s['knoten'], 'Knoten')
    else:
        print(name(nr), 'lösbar, zulässig:', ist_zulaessig(anordnung), ',', s['knoten'], 'Knoten bei', aushang['c'], 'Spalten')
````

`konfetti02` und `konfetti10` sind nicht lösbar, genau wie in den Lösungshinweisen. Bei den lösbaren Beispielen kommt die Suche
mit wenigen Umwegen aus (ohne Umweg wären es `c + 1` Knoten, bei `konfetti09` sind es 54 statt 14). Wie sieht es bei den
großen Beispielen aus? Wir brechen nach 3000 Knoten ab:

````python
for nr in [5, 11, 12]:
    aushang = lies_aushang('beispieldaten/' + name(nr) + '.txt')
    anordnung, s = loese(aushang, grenze=3000)
    if s['abgebrochen']:
        print(name(nr), 'abgebrochen nach', s['grenze'], 'Knoten, schon', len(s['sackgassen']), 'Sackgassen')
    else:
        print(name(nr), 'lösbar, zulässig:', ist_zulaessig(anordnung), ',', s['knoten'], 'Knoten bei', aushang['c'], 'Spalten')
````

`konfetti05` (110 Zeilen, 1020 Spalten) geht ohne Umweg. Bei `konfetti11` und `konfetti12` mit vielen ? verrennt sich die
Suche: Irgendwann früh wurde eine falsche Spalte gelegt, und das merkt sie erst viel später. Dann probiert sie unzählige
Möglichkeiten dahinter durch.

## Schritt 4: Zwei Zeilen, die sich gegenseitig blockieren

Angenommen, in den Zeilen `a` und `b` läuft gerade ein Block. Es gibt noch

- eine Spalte X mit ✓ in Zeile `a` und × in Zeile `b`: Wegen Regel 1 muss sie warten, bis Zeile `b` fertig ist,
- und eine Spalte Y mit ✓ in Zeile `b` und × in Zeile `a`: Sie muss warten, bis Zeile `a` fertig ist.

Zeile `a` wird aber erst fertig, wenn X liegt, und Zeile `b` erst, wenn Y liegt. Keine von beiden kann je gelegt werden: eine
**Blockade**. Daraus folgt

**Regel 2:** Eine Spalte, die in Zeile `a` einen neuen Block beginnt, darf nicht gelegt werden, wenn Zeile `a` danach mit einer
anderen angefangenen Zeile `b` blockiert wäre, also wenn `paare[a][b] > 0` und `paare[b][a] > 0` ist.

Wir müssen nur Zeilen prüfen, in denen die Spalte einen Block **beginnt**: Alle anderen Paare waren vorher schon ohne Blockade,
und durch das Legen werden die Zahlen in `paare` nur kleiner. Ist eine Zeile `b` schon fertig, ist `paare[b][a] = 0`, denn es gibt
dann keine ✓ in `b` mehr. Diese Regel ist eine Verallgemeinerung der zweiten Regel des Greedy-Verfahrens in den Lösungshinweisen.

````python
def regel_blockade(s, t):
    spalte = s['typen'][t]
    neu = []                                   # Zeilen, in denen die Spalte einen Block beginnt
    angefangen = []                            # Zeilen, die danach angefangen sind
    for z in range(s['r']):
        if spalte[z] == 'y' and s['gelegt'][z] == 0:
            neu.append(z)
        if s['gelegt'][z] > 0 or spalte[z] == 'y':
            angefangen.append(z)
    for a in neu:
        for b in angefangen:
            if s['paare'][a][b] > 0 and s['paare'][b][a] > 0:
                return False
    return True


def passt_v2(s, t):
    return regel_eins(s, t) and regel_anna(s, t) and regel_blockade(s, t)


def passt(s, t):
    return passt_v2(s, t)


for nr in [5, 11, 12]:
    aushang = lies_aushang('beispieldaten/' + name(nr) + '.txt')
    anordnung, s = loese(aushang, grenze=3000)
    print(name(nr), 'lösbar, zulässig:', ist_zulaessig(anordnung), 'und passt zu Anna:', passt_zu_anna(anordnung, aushang['anna']),
          ',', s['knoten'], 'Knoten bei', aushang['c'], 'Spalten')
````

Mit Regel 2 kommt die Suche auch bei `konfetti11` und `konfetti12` ohne einen einzigen Umweg durch: `c + 1` Knoten. Die
Regel schneidet die falschen Anfänge ab, bevor sie gelegt werden.

## Schritt 5: Ist die Lösung eindeutig?

Zuerst müssen wir festlegen, wann zwei Anordnungen **verschieden** sind. Wie die Lösungshinweise zählen wir das Vertauschen
gleicher Spalten nicht als neue Lösung: Für Anna entsteht dieselbe Tabelle. Mit ? kommt eine Frage dazu. Zwei Spalten wie
`yy?` und `yyy` sind verschieden, aber wenn das ? mitten in einem Block steht, wird es ohnehin ✓. Deshalb vergleichen wir die
Tabellen **nach dem Ausfüllen**: Ein ? wird ✓, wenn links und rechts davon in seiner Zeile ein ✓ steht, sonst ×.
Zwei Anordnungen sind gleich, wenn sie ausgefüllt dieselbe Tabelle ergeben.

**Ohne Annas Zeile** ist die Lösung fast nie eindeutig: Dreht man eine zulässige Anordnung um (die letzte Spalte nach vorn
usw.), ist sie wieder zulässig. Die Lösungshinweise zeigen, dass man nur dann keine zweite Lösung bekommt, wenn alle Spalten
gleich sind. Mit Annas Zeile ist die Frage spannender.

**Verfahren:** Wir füllen die erste Lösung aus und suchen dann eine zweite, deren ausgefüllte Tabelle **anders** ist. Dafür
merken wir uns beim Legen, ob der Anfang bisher genau mit der ersten Lösung übereinstimmt (`gleich`). Eine Lösung zählt nur, wenn
am Ende `gleich` falsch ist. Wir probieren zuerst die Spalten, die von der ersten Lösung abweichen. Die Sackgassen merken wir uns
jetzt zusammen mit `gleich`.

Beim Legen sieht man schon, was aus einem ? wird: ✓ genau dann, wenn in seiner Zeile gerade ein Block läuft
(`gefuellte_spalte`).

````python
def fuelle(anordnung):
    # die ausgefüllte Anordnung: ? zwischen zwei ✓ wird ✓, sonst ×
    r = len(anordnung[0])
    zeilen = []
    for z in range(r):
        zeile = ''
        for spalte in anordnung:
            zeile = zeile + spalte[z]
        erstes = zeile.find('y')
        letztes = zeile.rfind('y')
        neu = ''
        for p in range(len(zeile)):
            if zeile[p] == '?':
                if erstes != -1 and erstes < p < letztes:
                    neu = neu + 'y'
                else:
                    neu = neu + 'n'
            else:
                neu = neu + zeile[p]
        zeilen.append(neu)
    ergebnis = []
    for p in range(len(anordnung)):
        spalte = ''
        for z in range(r):
            spalte = spalte + zeilen[z][p]
        ergebnis.append(spalte)
    return ergebnis


def gefuellte_spalte(s, t):
    spalte = s['typen'][t]
    neu = ''
    for z in range(s['r']):
        laeuft = s['gelegt'][z] > 0 and s['rest'][z] > 0
        if spalte[z] == 'y' or (spalte[z] == '?' and laeuft):
            neu = neu + 'y'
        else:
            neu = neu + 'n'
    return neu


def suche_andere(s, erste, gleich):
    s['knoten'] += 1
    p = len(s['folge'])
    if p == s['c']:
        return not gleich                      # nur eine andere Tabelle zählt
    schluessel = (tuple(s['anzahl']), gleich)
    if schluessel in s['sackgassen']:
        return False
    abweichend = []
    wie_erste = []
    for t in kandidaten(s):
        if gleich and gefuellte_spalte(s, t) == erste[p]:
            wie_erste.append(t)
        else:
            abweichend.append(t)
    for t in abweichend:                       # zuerst die Spalten, die abweichen
        lege(s, t)
        if suche_andere(s, erste, False):
            return True
        nimm_zurueck(s, t)
    for t in wie_erste:
        lege(s, t)
        if suche_andere(s, erste, True):
            return True
        nimm_zurueck(s, t)
    s['sackgassen'].add(schluessel)
    return False


def zweite_loesung(aushang, anordnung):
    s = neuer_zustand(aushang)
    if suche_andere(s, fuelle(anordnung), True):
        return anordnung_aus(s), s
    return None, s


for nr in [0, 3, 6, 8]:
    aushang = lies_aushang('beispieldaten/' + name(nr) + '.txt')
    anordnung, s = loese(aushang)
    zweite, s2 = zweite_loesung(aushang, anordnung)
    if zweite is None:
        print(name(nr), 'eindeutig,', s2['knoten'], 'Knoten für die Suche nach einer zweiten Lösung')
    else:
        print(name(nr), 'nicht eindeutig, zweite Lösung zulässig:', ist_zulaessig(zweite), 'und passt zu Anna:',
              passt_zu_anna(zweite, aushang['anna']), ', ausgefüllt anders:', fuelle(zweite) != fuelle(anordnung))
````

Bei `konfetti08` gibt es durchaus mehrere Anordnungen der Spalten: Einige Spalten unterscheiden sich nur durch ein ? mitten in
einem Block und können die Plätze tauschen. Ausgefüllt ergeben sie aber alle dieselbe Tabelle, deshalb zählt `konfetti08` als
eindeutig, wie in den Lösungshinweisen.

## Schritt 7: Teilaufgabe e) – wie viele Änderungen braucht eine feste Anordnung?

Nicht lösbar sind `konfetti02`, `07`, `10` und `13`. Jetzt dürfen wir Einträge ändern und wollen möglichst wenige Änderungen.

Zuerst eine einfachere Frage: Die Anordnung ist **fest**. Wie viele Änderungen braucht eine Zeile? Wir wählen für die Zeile einen
Block von Position `a` bis `b` (er darf auch leer sein). Ändern müssen wir jedes ✓ außerhalb und jedes × innerhalb des Blocks.
Ist `E` die Zahl aller ✓ der Zeile, dann braucht der Block

`E − (✓ im Block) + (× im Block)` Änderungen.

Wir suchen also den Block, in dem `(✓ im Block) − (× im Block)` möglichst groß ist. Zählt man ✓ als +1, × als −1 und ? als 0, ist
das die **größte Summe eines zusammenhängenden Stücks**. Die findet man in einem Durchlauf: Man läuft von links nach rechts und
merkt sich die Summe seit dem aktuellen Anfang. Wird sie negativ, lohnt sich dieser Anfang nicht, und man beginnt neu.

In Annas Zeile ist der Block durch Annas Angaben vorgegeben. Dort zählen wir einfach die Einträge, die nicht zu Annas Zeile passen.

````python
def bester_block(zeile):
    # gibt (größte Summe, Anfang, Ende) zurück; Ende < Anfang heißt: leerer Block
    beste = (0, 0, -1)
    summe = 0
    anfang = 0
    for p in range(len(zeile)):
        if summe < 0:
            summe = 0                          # neu anfangen
            anfang = p
        if zeile[p] == 'y':
            summe += 1
        elif zeile[p] == 'n':
            summe -= 1
        if summe > beste[0]:
            beste = (summe, anfang, p)
    return beste


def zeile_von(anordnung, z):
    zeile = ''
    for spalte in anordnung:
        zeile = zeile + spalte[z]
    return zeile


def kosten(anordnung, anna):
    gesamt = 0
    for z in range(len(anordnung[0])):
        zeile = zeile_von(anordnung, z)
        if z == 0 and anna is not None:
            for p in range(len(zeile)):
                if zeile[p] != '?' and anna[p] != '?' and zeile[p] != anna[p]:
                    gesamt += 1
        else:
            gesamt += zeile.count('y') - bester_block(zeile)[0]
    return gesamt


def mit_aenderungen(anordnung, anna):
    # die geänderte Anordnung (die ? bleiben stehen) und die Liste der geänderten Stellen (z, p)
    zeilen = []
    geaendert = []
    for z in range(len(anordnung[0])):
        zeile = zeile_von(anordnung, z)
        if z == 0 and anna is not None:
            soll = anna
        else:
            summe, a, b = bester_block(zeile)
            soll = []
            for p in range(len(zeile)):
                soll.append('y' if a <= p <= b else 'n')
        neu = ''
        for p in range(len(zeile)):
            if zeile[p] != '?' and soll[p] != '?' and zeile[p] != soll[p]:
                neu = neu + soll[p]
                geaendert.append((z, p))
            else:
                neu = neu + zeile[p]
        zeilen.append(neu)
    neue_anordnung = []
    for p in range(len(anordnung)):
        spalte = ''
        for z in range(len(zeilen)):
            spalte = spalte + zeilen[z][p]
        neue_anordnung.append(spalte)
    return neue_anordnung, geaendert


def zeige_mit_aenderungen(anordnung, geaendert):
    # geänderte Einträge mit * davor: *1 heißt, aus × wurde ✓
    for z in range(len(anordnung[0])):
        teile = []
        for p in range(len(anordnung)):
            if (z, p) in geaendert:
                teile.append('*' + ZEICHEN[anordnung[p][z]])
            else:
                teile.append(' ' + ZEICHEN[anordnung[p][z]])
        print(''.join(teile))


print(bester_block('nyynyyn'), bester_block('nnn'), bester_block('y?nny'))
aushang02 = lies_aushang('beispieldaten/konfetti02.txt')
anordnung02 = nach_reihenfolge(aushang02, [9, 5, 3, 7, 8, 1, 2, 4, 6])     # aus den Lösungshinweisen
print('Änderungen bei der Reihenfolge der Lösungshinweise:', kosten(anordnung02, None))
neu02, geaendert02 = mit_aenderungen(anordnung02, None)
zeige_mit_aenderungen(neu02, geaendert02)
print('danach zulässig:', ist_zulaessig(neu02))
````

In `nyynyyn` hat der Block von Position 1 bis 5 die Summe 3 (vier ✓, ein ×). Bei `nnn` ist der leere Block am besten. Mit der
Reihenfolge aus den Lösungshinweisen braucht `konfetti02` zwei Änderungen, so wie dort angegeben.

## Schritt 8: Eine gute Anordnung suchen – Bergsteigen mit Neustarts

Jetzt suchen wir eine Anordnung mit möglichst kleinen Kosten. Ein einfaches Verfahren:

1. Die Spalten zufällig mischen.
2. Jede Spalte probeweise an jede andere Stelle schieben. Werden die Kosten kleiner, behalten wir die neue Anordnung.
3. Schritt 2 wiederholen, bis keine Verschiebung mehr hilft (ein „Gipfel“, von dem kein Schritt mehr bergauf führt).
4. Das Ganze mit neuen Zufallsanordnungen wiederholen (**Neustarts**) und die beste merken.

Wir zählen die Bewertungen (Aufrufe von `kosten`) und wie viele der 20 Neustarts die beste Zahl erreichen.

````python
import random


def bergsteigen(aushang, neustarts, zufall):
    anna = aushang['anna']
    c = aushang['c']
    bewertungen = 0
    beste_kosten = None
    beste = None
    treffer = 0
    for runde in range(neustarts):
        anordnung = aushang['spalten'][:]
        zufall.shuffle(anordnung)
        k = kosten(anordnung, anna)
        bewertungen += 1
        verbessert = True
        while verbessert:
            verbessert = False
            for i in range(c):
                for j in range(c):
                    if i != j:
                        neu = anordnung[:]
                        spalte = neu.pop(i)            # Spalte i herausnehmen ...
                        neu.insert(j, spalte)          # ... und an Stelle j einfügen
                        k_neu = kosten(neu, anna)
                        bewertungen += 1
                        if k_neu < k:
                            anordnung = neu
                            k = k_neu
                            verbessert = True
        if beste is None or k < beste_kosten:
            beste_kosten = k
            beste = anordnung
            treffer = 1
        elif k == beste_kosten:
            treffer += 1
    return beste_kosten, beste, bewertungen, treffer


berg = {}
for nr in [2, 7, 10, 13]:
    aushang = ergebnisse[nr][0]
    k, anordnung, bewertungen, treffer = bergsteigen(aushang, 20, random.Random(1))
    berg[nr] = (k, anordnung)
    print(name(nr), ':', k, 'Änderungen,', bewertungen, 'Bewertungen,', treffer, 'von 20 Neustarts erreichen das')
````

Das sind genau die Zahlen der Lösungshinweise: 2, 2, 3 und 4 Änderungen. Bergsteigen allein kann aber nicht beweisen, dass es
nicht noch besser geht. Die Gipfel, auf denen die anderen Neustarts stehen bleiben, sind nur „lokal“ die besten.

## Schritt 9: Geht es mit weniger? Branch and Bound

Wir legen wieder die Spalten von links nach rechts, jetzt aber **ohne** Regeln, dafür zählen wir die nötigen Änderungen mit.
Für jede Zeile merken wir uns drei Zahlen: die kleinste Zahl von Änderungen bis hierher, wenn die Zeile

- `k0`: noch keinen Block hatte (bisher alles ×),
- `k1`: gerade in ihrem Block ist (✓),
- `k2`: ihren Block schon hinter sich hat (wieder ×).

Kommt ein Eintrag dazu, kostet er als ✓ eine Änderung, wenn er × ist, und als × eine Änderung, wenn er ✓ ist (ein ? nie). Dann ist

- neues `k0` = `k0` + (Kosten als ×),
- neues `k1` = kleineres von `k0` und `k1` + (Kosten als ✓): Der Block beginnt jetzt oder lief schon,
- neues `k2` = kleineres von `k1` und `k2` + (Kosten als ×).

Am Anfang ist `k0 = 0`, die beiden anderen sind „unendlich“ (eine sehr große Zahl).

**Untere Schranke:** Wie viel muss eine Zeile mindestens noch kosten? In den Zuständen „kein Block“ und „im Block“ kann man die
restlichen Einträge dieser Zeile für sich allein immer passend anordnen (erst ×, dann ✓, oder erst ✓, dann ×). Im Zustand „Block
vorbei“ muss jedes restliche ✓ geändert werden. Also kostet die Zeile mindestens das kleinste von `k0`, `k1` und `k2 + (restliche ✓)`.
Die Summe über alle Zeilen ist eine untere Schranke für jede Fortsetzung. Ist sie nicht kleiner als die beste bekannte Lösung,
schneiden wir ab.

**Sackgassen mit Zahlen:** Kommen wir mit denselben restlichen Spalten an eine Stelle, an der wir schon einmal mit in jeder
Zeile höchstens gleich vielen Änderungen waren, kann es diesmal nicht besser werden. Auch dann schneiden wir ab.

Wir starten mit der Zahl vom Bergsteigen als Schranke und suchen eine Anordnung, die **besser** ist. Findet die Suche keine,
ist bewiesen, dass das Bergsteigen das Optimum gefunden hat.

````python
UNENDLICH = 10 ** 9


def bb_zustand(aushang, schranke):
    spalten = aushang['spalten']
    typen = sorted(set(spalten))
    anzahl = []
    for typ in typen:
        anzahl.append(spalten.count(typ))
    rest_haken = []
    for z in range(aushang['r']):
        rest_haken.append(zeile_von(spalten, z).count('y'))
    return {'r': aushang['r'], 'c': aushang['c'], 'anna': aushang['anna'], 'typen': typen, 'anzahl': anzahl,
            'rest_haken': rest_haken, 'beste': schranke, 'beste_folge': None, 'folge': [], 'knoten': 0,
            'schon_da': {}}


def untere_schranke(s, werte, anna_kosten):
    summe = anna_kosten
    for z in range(s['r']):
        k0, k1, k2 = werte[z]
        summe += min(k0, k1, k2 + s['rest_haken'][z])
    return summe


def schlechter_als_frueher(s, werte, anna_kosten):
    schluessel = tuple(s['anzahl'])
    jetzt = [anna_kosten]
    for k0, k1, k2 in werte:
        jetzt.extend([k0, k1, k2])
    if schluessel not in s['schon_da']:
        s['schon_da'][schluessel] = []
    for frueher in s['schon_da'][schluessel]:
        hoechstens = True
        for i in range(len(jetzt)):
            if frueher[i] > jetzt[i]:
                hoechstens = False
                break
        if hoechstens:
            return True                        # früher war es überall höchstens so teuer
    s['schon_da'][schluessel].append(jetzt)
    return False


def bb_suche(s, werte, anna_kosten):
    s['knoten'] += 1
    if untere_schranke(s, werte, anna_kosten) >= s['beste']:
        return                                 # kann nicht besser werden
    p = len(s['folge'])
    if p == s['c']:
        s['beste'] = untere_schranke(s, werte, anna_kosten)
        s['beste_folge'] = s['folge'][:]
        return
    if schlechter_als_frueher(s, werte, anna_kosten):
        return
    for t in range(len(s['typen'])):
        if s['anzahl'][t] == 0:
            continue
        spalte = s['typen'][t]
        neue_werte = []
        for z in range(s['r']):
            k0, k1, k2 = werte[z]
            als_haken = 1 if spalte[z] == 'n' else 0
            als_kreuz = 1 if spalte[z] == 'y' else 0
            neue_werte.append((k0 + als_kreuz, min(k0, k1) + als_haken, min(k1, k2) + als_kreuz))
        neue_anna_kosten = anna_kosten
        if s['anna'] is not None:
            if spalte[0] != '?' and s['anna'][p] != '?' and spalte[0] != s['anna'][p]:
                neue_anna_kosten += 1
        s['anzahl'][t] -= 1
        s['folge'].append(t)
        for z in range(s['r']):
            if spalte[z] == 'y':
                s['rest_haken'][z] -= 1
        bb_suche(s, neue_werte, neue_anna_kosten)
        for z in range(s['r']):
            if spalte[z] == 'y':
                s['rest_haken'][z] += 1
        s['folge'].pop()
        s['anzahl'][t] += 1


def beweise(aushang, schranke):
    s = bb_zustand(aushang, schranke)
    werte = []
    for z in range(aushang['r']):
        if z == 0 and aushang['anna'] is not None:
            werte.append((0, 0, 0))            # Annas Zeile zählen wir extra
        else:
            werte.append((0, UNENDLICH, UNENDLICH))
    bb_suche(s, werte, 0)
    return s


for nr in [2, 7, 10, 13]:
    k = berg[nr][0]
    s = beweise(ergebnisse[nr][0], k)
    if s['beste_folge'] is None:
        print(name(nr), ': keine Anordnung mit weniger als', k, 'Änderungen,', s['knoten'], 'Knoten')
    else:
        print(name(nr), ': besser geht es mit', s['beste'], 'Änderungen')
````

Damit ist bewiesen: 2, 2, 3 und 4 Änderungen sind **optimal**, genau wie in den Lösungshinweisen. `konfetti13` braucht die meisten
Knoten, das dauert im Browser etwas.

Warum braucht die Suche hier so viel mehr Knoten als in Schritt 4? Dort schneiden zwei scharfe Regeln fast alles ab, hier gibt es
nur die Schranke. Solange die Schranke unter dem Optimum liegt, muss die Suche weitermachen, und das sind sehr viele Anfänge.

## Schritt 10: Die Lösungen für Teilaufgabe e)

Für die vier nicht lösbaren Beispiele die Reihenfolge und die Tabelle mit den Änderungen. Ein `*` vor einem Eintrag heißt: Dieser
Eintrag wurde geändert, `*1` ist ein × geworden ✓, `*0` ein ✓ geworden ×. Die ? lassen wir stehen.

## Schritt 11: Selbst prüfen mit Brute Force

Stimmen unsere Verfahren auch bei anderen Tabellen? Wir erzeugen 200 kleine Zufallstabellen (2 bis 4 Zeilen, 2 bis 6 Spalten) mit
gestörten Blöcken, ? und bei der Hälfte mit Annas Zeile. Brute Force probiert alle Reihenfolgen der Spalten
(`itertools.permutations`) und bestimmt

- ob es eine zulässige Anordnung gibt und wie viele verschiedene ausgefüllte Tabellen dabei entstehen,
- die kleinste Zahl von Änderungen.

Das vergleichen wir mit der Suche (Schritt 4), der Suche nach einer zweiten Lösung (Schritt 5), Branch and Bound (Schritt 9) und
dem Bergsteigen (Schritt 8).

````python
import itertools


def zufalls_aushang(zufall):
    r = zufall.randint(2, 4)
    c = zufall.randint(2, 6)
    zeilen = []
    anna = None
    for z in range(r):
        a = zufall.randint(0, c - 1)
        b = zufall.randint(a, c - 1)
        richtig = []
        zeile = []
        for p in range(c):
            wert = 'y' if a <= p <= b else 'n'
            richtig.append(wert)
            x = zufall.random()
            if x < 0.15:
                wert = '?'
            elif x < 0.3:
                wert = 'y' if wert == 'n' else 'n'     # falsch abgelesen
            zeile.append(wert)
        zeilen.append(zeile)
        if z == 0 and zufall.random() < 0.5:
            anna = richtig
    mischung = list(range(c))
    zufall.shuffle(mischung)
    spalten = []
    for j in mischung:
        spalte = ''
        for z in range(r):
            spalte = spalte + zeilen[z][j]
        spalten.append(spalte)
    return {'r': r, 'c': c, 'spalten': spalten, 'anna': anna}


def brute_force(aushang):
    tabellen = set()
    kleinste = None
    for perm in itertools.permutations(aushang['spalten']):
        anordnung = list(perm)
        k = kosten(anordnung, aushang['anna'])
        if kleinste is None or k < kleinste:
            kleinste = k
        if ist_zulaessig(anordnung) and passt_zu_anna(anordnung, aushang['anna']):
            tabellen.add(tuple(fuelle(anordnung)))
    return len(tabellen), kleinste


zufall = random.Random(43)
fehler = 0
zaehler = {'lösbar': 0, 'eindeutig': 0, 'mit Anna': 0, 'Bergsteigen optimal': 0}
for versuch_nr in range(200):
    aushang = zufalls_aushang(zufall)
    zahl_tabellen, kleinste = brute_force(aushang)
    anordnung, s = loese(aushang)
    if (anordnung is not None) != (zahl_tabellen > 0):
        fehler += 1
    if anordnung is not None:
        zaehler['lösbar'] += 1
        zweite, s2 = zweite_loesung(aushang, anordnung)
        if (zweite is None) != (zahl_tabellen == 1):
            fehler += 1
        if zweite is None:
            zaehler['eindeutig'] += 1
    if aushang['anna'] is not None:
        zaehler['mit Anna'] += 1
    s3 = beweise(aushang, UNENDLICH)
    if s3['beste'] != kleinste:
        fehler += 1
    k, beste, bewertungen, treffer = bergsteigen(aushang, 3, zufall)
    if k == kleinste:
        zaehler['Bergsteigen optimal'] += 1
    if k < kleinste:
        fehler += 1
print('Fehler:', fehler)
print(zaehler)
````

Keine Abweichung: Lösbarkeit, Eindeutigkeit und die kleinste Zahl von Änderungen stimmen bei allen 200 Tabellen mit Brute Force
überein. Branch and Bound haben wir hier sogar ohne Startwert laufen lassen (Schranke „unendlich“), es findet das Optimum dann
selbst. Das Bergsteigen mit nur drei Neustarts trifft bei diesen kleinen Tabellen fast immer das Optimum.

## Für Neugierige

**Geht es auch ohne Suche?** Für die Teilaufgaben a) bis c) (ohne ?) gibt es Verfahren mit polynomieller Laufzeit. Die
Lösungshinweise beschreiben **PQ-Bäume**: Ein Baum, dessen Blätter die Spalten sind. Die Kinder eines P-Knotens darf man beliebig
anordnen, die eines Q-Knotens nur in fester Reihenfolge oder genau umgekehrt. Am Anfang hängen alle Spalten an einem P-Knoten
(alles ist erlaubt). Für jede Zeile wird der Baum so umgebaut, dass die Spalten mit ✓ in dieser Zeile nebeneinander liegen
müssen (Operation `REDUCE`). Danach beschreibt der Baum **alle** zulässigen Anordnungen auf einmal, insgesamt in `O(r · c)`
Schritten. Daran sieht man auch die Eindeutigkeit ab. Mit Annas Zeile braucht man dafür noch eine Rechnung wie bei SUBSET SUM.
Die Hinweise nennen auch einen Ansatz mit Graphen (Fulkerson und Gross, 1965) und das Paper von Booth und Lueker (1976).

**Unsere Suche** ist dagegen im schlimmsten Fall exponentiell. Bei den Beispielen kommt sie dank der beiden Regeln fast ohne
Umweg aus, aber es gibt sicher Tabellen, bei denen sie lange probiert. Dafür ist sie kurz und behandelt ?, Annas Zeile und die
Eindeutigkeit mit denselben Bausteinen.

**Mit ? wird es wirklich schwer.** Die Lösungshinweise zeigen, dass Teilaufgabe d) **NP-vollständig** ist. Man kann das bekannte
schwere Problem BETWEENNESS (Zahlen so anordnen, dass für vorgegebene Tripel `(a, b, c)` immer `b` zwischen `a` und `c` steht)
in eine Tabelle übersetzen: pro Tripel zwei Zeilen, die erste mit ✓ bei `a` und `b` und × bei `c`, die zweite mit × bei `a` und ✓ bei
`b` und `c`, sonst überall ?. Teilaufgabe e) ist ebenfalls NP-vollständig, hier übersetzt man die Suche nach einem
**Hamiltonpfad**: eine Spalte pro Knoten und eine Zeile pro Kante mit ✓ bei den beiden Endknoten. Ein schnelles Verfahren für
alle Tabellen wird man also kaum finden. Deshalb sind Suche, Bergsteigen und Branch and Bound hier vernünftige Werkzeuge.

**Greedy ohne Zurücknehmen:** Die Lösungshinweise beschreiben ein Greedy-Verfahren, das immer die nächste Spalte nach den
Regeln wählt und nie etwas zurücknimmt. Es findet bei allen lösbaren Beispielen außer `konfetti00` eine Lösung, ist aber nicht
korrekt. Unsere Suche nimmt eine ähnliche Reihenfolge der Kandidaten und kann zusätzlich zurücknehmen.

**SAT und ILP:** Die Hinweise beschreiben auch, wie man die Aufgabe als SAT-Formel (Variable „Spalte `i` steht vor Spalte `j`“) und
als ganzzahliges lineares Programm (ILP) schreibt. Ein Solver übernimmt dann die Suche. Das probieren wir zum Schluss mit MiniZinc.

## Zum Schluss: dasselbe mit MiniZinc

Die Aufgabe ist ein **Anordnungsproblem**: Jede Spalte bekommt eine Position. Das passt zu MiniZinc. Wir beschreiben, was
gelten soll, und der Solver sucht.

**Modell für a) bis d):** `pos[j]` ist die Position der Spalte `j`, alle Positionen sind verschieden (`alldifferent`). Für jede
Zeile `z` gibt es einen Block von `anfang[z]` bis `ende[z]`. Jeder Haken der Zeile muss im Block liegen, jedes Kreuz außerhalb. Für
die Fragezeichen gibt es keine Bedingung, sie dürfen also im Block oder außerhalb liegen, genau wie bei uns. Ist `ende[z] <
anfang[z]`, ist der Block leer (für Zeilen ganz ohne Haken). Annas Zeile legt fest, welche Positionen im Block der ersten Zeile
liegen. Neu gegenüber den Arbeitsblättern sind höchstens `\/` („oder“), `<->` („genau dann, wenn“) und `->` („wenn …, dann“).

```minizinc
include "alldifferent.mzn";
int: r;                                   % Zeilen (Personen)
int: c;                                   % Spalten (Termine)
array[1..r, 1..c] of 0..2: t;             % 0 = Kreuz, 1 = Haken, 2 = Fragezeichen
bool: mit_anna;                           % ist Annas Zeile bekannt?
array[1..c] of 0..2: anna;                % Annas Zeile in der richtigen Reihenfolge

array[1..c] of var 1..c: pos;             % pos[j] = Position der Spalte j
constraint alldifferent(pos);
array[1..r] of var 1..c: anfang;          % der Block der Zeile z geht von anfang[z] bis ende[z]
array[1..r] of var 1..c: ende;            % (ende < anfang heißt: leerer Block)

constraint forall(z in 1..r, j in 1..c where t[z, j] == 1)(
  anfang[z] <= pos[j] /\ pos[j] <= ende[z]);         % jeder Haken liegt im Block
constraint forall(z in 1..r, j in 1..c where t[z, j] == 0)(
  pos[j] < anfang[z] \/ pos[j] > ende[z]);           % jedes Kreuz liegt außerhalb

constraint mit_anna -> forall(p in 1..c where anna[p] != 2)(
  (anfang[1] <= p /\ p <= ende[1]) <-> anna[p] == 1);

solve satisfy;

output ["pos = ", show(pos), "\n"];
```

Die Daten bekommt MiniZinc als `.dzn`-Datei. Die Python-Zelle `als_dzn` baut sie aus einem Aushang, mit 0 für ×, 1 für ✓ und 2 für ?.

````python
def als_dzn(aushang):
    wert = {'n': 0, 'y': 1, '?': 2}
    zeilen = []
    for z in range(aushang['r']):
        teile = []
        for spalte in aushang['spalten']:
            teile.append(str(wert[spalte[z]]))
        zeilen.append(', '.join(teile))
    if aushang['anna'] is None:
        anna = [2] * aushang['c']
        mit_anna = 'false'
    else:
        anna = []
        for x in aushang['anna']:
            anna.append(wert[x])
        mit_anna = 'true'
    return 'r = %d;\nc = %d;\nt = [| %s |];\nmit_anna = %s;\nanna = %s;' % (
        aushang['r'], aushang['c'], '\n     | '.join(zeilen), mit_anna, anna)


daten13 = als_dzn(ergebnisse[13][0])
print(daten13)
````

**Modell für e):** Jetzt darf jeder Eintrag geändert werden, das kostet aber. `geaendert[z, j]` ist wahr, wenn der Eintrag in
Zeile `z` und Spalte `j` geändert wird. Ein Haken muss im Block liegen **oder** geändert werden, ein Kreuz außerhalb **oder**
geändert werden. Fragezeichen werden nie geändert (das wäre sinnlos). Minimiert wird die Zahl der Änderungen.

```minizinc
include "alldifferent.mzn";
int: r;
int: c;
array[1..r, 1..c] of 0..2: t;             % 0 = Kreuz, 1 = Haken, 2 = Fragezeichen
bool: mit_anna;
array[1..c] of 0..2: anna;

array[1..c] of var 1..c: pos;
constraint alldifferent(pos);
array[1..r] of var 1..c: anfang;
array[1..r] of var 1..c: ende;
array[1..r, 1..c] of var bool: geaendert; % wird der Eintrag in Zeile z, Spalte j geändert?

constraint forall(z in 1..r, j in 1..c where t[z, j] == 1)(
  geaendert[z, j] \/ (anfang[z] <= pos[j] /\ pos[j] <= ende[z]));
constraint forall(z in 1..r, j in 1..c where t[z, j] == 0)(
  geaendert[z, j] \/ pos[j] < anfang[z] \/ pos[j] > ende[z]);
constraint forall(z in 1..r, j in 1..c where t[z, j] == 2)(not geaendert[z, j]);

constraint mit_anna -> forall(p in 1..c where anna[p] != 2)(
  (anfang[1] <= p /\ p <= ende[1]) <-> anna[p] == 1);

var int: aenderungen = sum(z in 1..r, j in 1..c)(bool2int(geaendert[z, j]));
solve minimize aenderungen;

output ["aenderungen = ", show(aenderungen), "\npos = ", show(pos), "\n"];
```

Bei einem Optimierungsproblem gibt MiniZinc jede neue, bessere Lösung aus. Die letzte vor `==========` ist optimal (die Zeile
