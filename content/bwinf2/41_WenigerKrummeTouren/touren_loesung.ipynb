### Weniger krumme Touren – Lösung Schritt für Schritt

*41. Bundeswettbewerb Informatik (2022/2023), 2. Runde, Aufgabe 1*

Ein Anteil von Antons Arbeit ist das Abfliegen aller Außenstellen in Australien. Auf seiner Tour muss er vorgegebene Orte
besuchen, kann sich aber aussuchen, in welcher Reihenfolge dies geschieht.

Anton möchte dies schnell erledigen. Deshalb hat er sich für seine Tour eine Route mit möglichst kurzer Flugstrecke überlegt
(im Bild auf dem Aufgabenblatt 2649 Kilometer lang). Bei dieser Route muss er an manchen Orten sehr scharf abbiegen. Für diese
Orte hat er die Abbiegewinkel in seine Karte eingetragen.

In Zukunft möchte er alle Abbiegewinkel von mehr als 90 Grad vermeiden. Kannst du ihm helfen und eine neue Route für ihn
berechnen? Diese neue Route muss natürlich weiterhin alle Außenstellen enthalten. Sie kann an beliebigen Orten beginnen und
enden. Außerdem muss der Flug von einer Außenstelle zur nächsten geradlinig verlaufen, und Anton muss an keinem Ort mit einem
Winkel von mehr als 90 Grad abbiegen. (Ein Bild auf dem Aufgabenblatt zeigt zwei Abbiegewinkel α: links α > 90° und nicht
gewünscht, rechts α ≤ 90°.)

#### Aufgabe

Schreibe ein Programm, das Koordinaten einlesen kann und eine entsprechende Route berechnet. Kann das Programm immer eine
solche finden? Versuche die Flugstrecke möglichst kurz zu machen, aber es ist nicht verlangt, die allerbeste Route zu finden.

Wende dein Programm mindestens auf alle Beispiele an, die du auf den BWINF-Webseiten findest, und dokumentiere die Ergebnisse.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `wenigerkrumm1.txt` bis `wenigerkrumm7.txt`. In jeder Zeile stehen die Koordinaten `x y` eines
Ortes in Kilometern.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Backtracking (Tiefensuche):** Wir bauen die Route Ort für Ort auf. Kommt man nicht weiter, nimmt man den letzten Ort zurück
  und probiert einen anderen. Ohne Schranke findet man so eine Route, wenn es eine gibt, und weiß sonst sicher, dass es keine gibt.
- **Der Abbiegewinkel braucht keinen Winkel:** Ob man in B höchstens 90° abbiegt, entscheidet das Vorzeichen eines Skalarprodukts
  (siehe Schritt 2). Es kommen nur Plus, Minus und Mal vor, keine Wurzel und kein Arkuskosinus.
- **Gierige Reihenfolge:** Beim Backtracking zuerst den nächsten noch freien Ort probieren. Das liefert kurze Routen und findet
  oft schnell eine Lösung.
- **Schrittgrenze:** Von manchen Startorten aus dauert die Suche sehr lange. Wir geben jedem Start nur eine begrenzte Zahl von
  Schritten und nehmen die Starts, bei denen es schnell klappt.
- **Lokale Suche:** Eine gefundene Route durch kleine Änderungen verkürzen (ein Stück umdrehen, ein kurzes Stück an eine andere
  Stelle setzen) und jede Änderung nur behalten, wenn die Route gültig bleibt und kürzer wird.

**Nützliche Funktionen und Bibliotheken**
- `math.dist(a, b)`: Abstand zweier Punkte (mit Tupeln `(x, y)`)
- **Sortieren von Tupeln:** `sort()` vergleicht Tupel zuerst nach dem ersten Eintrag, bei Gleichstand nach dem zweiten. So
  sortiert man Orte nach dem Abstand, ohne eine eigene Vergleichsfunktion zu brauchen: Liste von Paaren `(abstand, nr)` sortieren.
- Listen schneiden und umkehren: `route[i:j+1][::-1]`
- Rekursion: Python erlaubt normalerweise nur 1000 Aufrufe ineinander. Die Suche geht höchstens so tief wie die Zahl der Orte
  (hier höchstens 120), das reicht.

## Die Idee

1. **Abbiegewinkel prüfen:** Für drei Orte A, B, C nacheinander entscheidet eine kleine Rechnung, ob die Abbiegung in B höchstens
   90° beträgt.
2. **Route aufbauen (Backtracking):** Ausgehend von einem Startort fügen wir immer den nächsten noch nicht besuchten Ort an, der
   die Abbiegebedingung erfüllt. Gibt es keinen, gehen wir einen Schritt zurück und probieren den zweitnächsten.
3. **Es gibt nicht immer eine Route:** Ein spitzwinkliges Dreieck ist ein Gegenbeispiel. Mit vollständigem Backtracking kann man
   das auch vom Programm beweisen lassen.
4. **Alle Startorte, Schrittgrenze:** Wir starten in jedem Ort und brechen jeden Versuch nach einer festen Zahl von Schritten ab.
5. **Verkürzen:** Die kürzesten gefundenen Routen verbessern wir durch Umkehren und Verschieben von Teilstücken. Dann vergleichen
   wir mit den Zahlen der Lösungshinweise (dort auch die tatsächlich kürzesten Routen, mit anderen Verfahren berechnet).

## Schritt 1: Einlesen und Länge

Ein Ort ist ein Tupel `(x, y)`, eine Route eine Liste von Orten. Die Länge einer Route ist die Summe der Abstände aufeinander
folgender Orte.

````python
import math


def lies(pfad):
    """Liest die Orte einer Datei: pro Zeile 'x y'. Gibt eine Liste von Tupeln (x, y) zurück."""
    orte = []
    with open(pfad) as datei:
        for zeile in datei:
            if zeile.strip():
                x, y = zeile.split()
                orte.append((float(x), float(y)))
    return orte


def abstand(a, b):
    return math.dist(a, b)


def laenge(route):
    """Länge der Route: Summe der Abstände aufeinander folgender Orte."""
    summe = 0.0
    for i in range(len(route) - 1):
        summe += abstand(route[i], route[i + 1])
    return summe


def beispiel(nr):
    return lies('beispieldaten/wenigerkrumm' + str(nr) + '.txt')


for nr in range(1, 8):
    print('wenigerkrumm' + str(nr), ':', len(beispiel(nr)), 'Orte')
````

## Schritt 2: Ist die Abbiegung erlaubt?

Wir fliegen von A nach B und wollen weiter nach C. Der **Abbiegewinkel** α in B ist der Winkel zwischen der bisherigen Richtung
(von A nach B) und der neuen Richtung (von B nach C). Im Dreieck ABC liegt bei B der Innenwinkel β, und es gilt α = 180° − β.

Wir brauchen α ≤ 90°, also β ≥ 90°, also `cos β ≤ 0`. Der Kosinus von β steht im **Skalarprodukt** der Vektoren von B nach A
und von B nach C:

`(A − B) · (C − B) = |BA| · |BC| · cos β`

Weil die Längen nicht negativ sind, hat das Skalarprodukt dasselbe Vorzeichen wie `cos β`. Die Abbiegung ist also genau dann
erlaubt, wenn

`(Ax − Bx) · (Cx − Bx) + (Ay − By) · (Cy − By) ≤ 0`

gilt. (Die Lösungshinweise rechnen gleichwertig mit dem Kosinussatz und `|AB|² + |BC|² − |AC|² ≥ 0`.)

**Zum Rechnen mit Kommazahlen:** Liegt ein Ort genau im rechten Winkel, kann das Ergebnis durch Rundungsfehler minimal
von 0 abweichen. Wir erlauben deshalb eine winzige Toleranz von `1e-9`.

````python
def erlaubt(a, b, c):
    """Darf man von a über b nach c fliegen (Abbiegewinkel in b höchstens 90 Grad)?"""
    skalarprodukt = (a[0] - b[0]) * (c[0] - b[0]) + (a[1] - b[1]) * (c[1] - b[1])
    return skalarprodukt <= 1e-9


def abbiegewinkel(a, b, c):
    """Nur zur Kontrolle: der Abbiegewinkel in Grad (180 minus Innenwinkel bei b)."""
    wert = ((a[0] - b[0]) * (c[0] - b[0]) + (a[1] - b[1]) * (c[1] - b[1])) / (abstand(a, b) * abstand(b, c))
    return 180 - math.degrees(math.acos(max(-1.0, min(1.0, wert))))


def ist_gueltig(route):
    """Alle Abbiegungen der Route erlaubt?"""
    for i in range(len(route) - 2):
        if not erlaubt(route[i], route[i + 1], route[i + 2]):
            return False
    return True


A, B = (0, 0), (10, 0)
for c in [(20, 0), (10, 10), (15, 0.001), (0, 5), (3, 1)]:
    print(c, ' Winkel %6.2f' % abbiegewinkel(A, B, c), ' erlaubt:', erlaubt(A, B, c))
````

`(20, 0)` ist geradeaus (0°), `(10, 10)` ist ein rechter Winkel (90°, noch erlaubt), `(15, 0.001)` fast geradeaus. `(0, 5)` und
`(3, 1)` biegen scharf zurück ab und sind verboten.

## Schritt 3: Backtracking von einem Startort aus

Wir legen eine Liste `weg` mit den Nummern der besuchten Orte an. Die Funktion `weiter` hängt einen Ort an, ruft sich selbst auf
und nimmt den Ort wieder zurück (`pop`), wenn es nicht weitergeht. Folgendes ist wichtig:

- **Nächster Ort zuerst:** Für jeden Ort legen wir vorab die Liste aller Orte nach Abstand sortiert an (`nachbarlisten`).
- **Abbiegebedingung:** Ein Ort `j` kommt nur infrage, wenn `erlaubt(vorletzter, letzter, j)` gilt. Beim ersten Flug gibt es keine
  Bedingung.
- **Schrittgrenze:** `schritte[0]` zählt die Aufrufe. Wird die Grenze überschritten, geben wir auf. (Eine Liste mit einem
  Eintrag dient als Zähler, den alle Aufrufe gemeinsam verändern können.)
- Die Funktion gibt die Route (Liste von Orten) und die Zahl der Schritte zurück. Wurde keine Route gefunden, ist die Route `None`.

````python
def nachbarlisten(orte):
    """Für jeden Ort die Nummern aller Orte, nach Abstand aufsteigend sortiert (der Ort selbst zuerst)."""
    liste = []
    for i in range(len(orte)):
        paare = []
        for j in range(len(orte)):
            paare.append((abstand(orte[i], orte[j]), j))
        paare.sort()
        liste.append([j for d, j in paare])
    return liste


def route_ab(orte, nachbarn, start, grenze):
    """Backtracking ab dem Ort start. Gibt (Route, Schritte) zurück; Route ist None, wenn keine gefunden wurde
    (oder die Schrittgrenze erreicht ist)."""
    n = len(orte)
    weg = [start]
    besucht = [False] * n
    besucht[start] = True
    schritte = [0]

    def weiter():
        if len(weg) == n:
            return True
        schritte[0] += 1
        if schritte[0] > grenze:
            return False
        for j in nachbarn[weg[-1]]:
            if besucht[j]:
                continue
            if len(weg) >= 2 and not erlaubt(orte[weg[-2]], orte[weg[-1]], orte[j]):
                continue
            weg.append(j)
            besucht[j] = True
            if weiter():
                return True
            weg.pop()
            besucht[j] = False
            if schritte[0] > grenze:
                return False
        return False

    if weiter():
        return [orte[i] for i in weg], schritte[0]
    return None, schritte[0]


orte4 = beispiel(4)
nachbarn4 = nachbarlisten(orte4)
route, schritte = route_ab(orte4, nachbarn4, 0, 100000)
print('wenigerkrumm4, Start im Ort 0:', len(route), 'Orte, Länge %.1f' % laenge(route), ', Schritte:', schritte, ', gültig:', ist_gueltig(route))
````

## Schritt 4: Kann das Programm immer eine Route finden?

Nein, und das kann man auch beweisen. Bei einem **spitzwinkligen Dreieck** ABC ist jeder Winkel kleiner als 90°. Sobald man in
einer Ecke ist und von dort zu einem anderen Eckpunkt fliegt, kann man in dieser Ecke nur noch mit Abbiegung > 90° zum dritten
weiter. Man kommt also nie zu allen drei Ecken, auch nicht, wenn man noch beliebig viele Orte ins Innere setzt (Abbildung 1.1 der
Lösungshinweise).

Mit einer sehr großen Schrittgrenze läuft das Backtracking für alle Startorte vollständig durch. Findet es keine Route, ist
bewiesen, dass es keine gibt.

````python
def gibt_es_route(orte, grenze=10**7):
    """Probiert alle Startorte mit sehr großer Schrittgrenze. Gibt (Route oder None, Schritte insgesamt) zurück."""
    nachbarn = nachbarlisten(orte)
    insgesamt = 0
    for start in range(len(orte)):
        route, schritte = route_ab(orte, nachbarn, start, grenze)
        insgesamt += schritte
        if route is not None:
            return route, insgesamt
    return None, insgesamt


dreieck = [(0, 0), (10, 0), (5, 8)]
dreieck_mit_punkt = dreieck + [(5, 3), (4, 2)]
for name, orte in [('Dreieck', dreieck), ('Dreieck mit zwei Orten innen', dreieck_mit_punkt),
                   ('Rechtwinkliges Dreieck', [(0, 0), (10, 0), (0, 10)])]:
    route, schritte = gibt_es_route(orte)
    print(name + ':', 'Route', route if route else 'gibt es nicht', '( Schritte:', schritte, ')')
````

Beim rechtwinkligen Dreieck gibt es dagegen eine Route (Abbiegung genau 90° ist erlaubt).

## Schritt 5: Alle Startorte, mit Schrittgrenze

Bei den größeren Beispielen führt das Backtracking von manchen Startorten aus sehr lange in Sackgassen. Wir geben deshalb jedem
Startort nur `GRENZE = 5000` Schritte und sammeln alle Routen, die so gefunden werden.

````python
GRENZE = 5000


def alle_starts(orte, grenze):
    """Backtracking von jedem Ort aus mit Schrittgrenze. Gibt (Liste der gefundenen Routen, Schritte insgesamt) zurück."""
    nachbarn = nachbarlisten(orte)
    routen = []
    insgesamt = 0
    for start in range(len(orte)):
        route, schritte = route_ab(orte, nachbarn, start, grenze)
        insgesamt += schritte
        if route is not None:
            routen.append(route)
    return routen, insgesamt


erste = {}
print('Beispiel  Orte  Starts mit Route  Schritte   kürzeste gefundene Route')
for nr in range(1, 8):
    orte = beispiel(nr)
    routen, schritte = alle_starts(orte, GRENZE)
    erste[nr] = routen
    kuerzeste = min(laenge(r) for r in routen)
    print('%5d %8d %10d %16d %12.1f' % (nr, len(orte), len(routen), schritte, kuerzeste))
````

Für **jedes** Beispiel findet mindestens ein Startort in höchstens 5000 Schritten eine Route (Spalte „Starts mit Route“).

## Schritt 6: Routen verkürzen

Die gefundenen Routen sind brauchbar, aber nicht kurz. Wir verbessern sie durch zwei kleine Änderungen:

1. **Umkehren:** Wir drehen das Stück `route[i..j]` um. Dabei ändern sich nur die beiden Verbindungen an den Enden des Stücks;
   im Stück selbst bleiben alle Abbiegungen erlaubt (Rückwärtsfliegen ändert den Winkel nicht, weil das Skalarprodukt
   symmetrisch in A und C ist). Wir behalten die Änderung, wenn die Route kürzer wird und `ist_gueltig` weiter gilt.
2. **Verschieben:** Ein Stück aus 1, 2 oder 3 aufeinander folgenden Orten nehmen wir heraus und setzen es (vorwärts oder
   rückwärts) an die Stelle, an der es die Route am wenigsten verlängert. Wieder nur, wenn die Route insgesamt kürzer und gültig bleibt.

Beides wiederholen wir, bis sich nichts mehr verbessert. Die aufwendige Gültigkeitsprüfung machen wir erst, wenn eine
Änderung die Route überhaupt verkürzen würde.

````python
def umkehren(route):
    """Kehrt Teilstücke um, wenn die Route dadurch kürzer wird und gültig bleibt. Gibt (Route, geändert?) zurück."""
    n = len(route)
    geaendert = False
    for i in range(n - 1):
        for j in range(i + 1, n):
            alt = 0.0     # die Verbindungen vor i und nach j, so wie sie jetzt sind
            neu = 0.0     # dieselben Verbindungen nach dem Umkehren
            if i > 0:
                alt += abstand(route[i - 1], route[i])
                neu += abstand(route[i - 1], route[j])
            if j < n - 1:
                alt += abstand(route[j], route[j + 1])
                neu += abstand(route[i], route[j + 1])
            if neu < alt - 1e-9:
                neue_route = route[:i] + route[i:j + 1][::-1] + route[j + 1:]
                if ist_gueltig(neue_route):
                    route = neue_route
                    geaendert = True
    return route, geaendert


def verschieben(route):
    """Setzt Stücke aus 1 bis 3 Orten an eine bessere Stelle. Gibt (Route, geändert?) zurück."""
    n = len(route)
    geaendert = False
    for l in (1, 2, 3):
        i = 0
        while i + l <= n:
            stueck = route[i:i + l]
            rest = route[:i] + route[i + l:]
            spart = laenge(route) - laenge(rest)      # so viel Länge spart das Herausnehmen
            beste = None
            for k in range(len(rest) + 1):            # vor rest[k] einfügen
                for s in (stueck, stueck[::-1]):
                    kostet = laenge(s)
                    if k > 0:
                        kostet += abstand(rest[k - 1], s[0])
                    if k < len(rest):
                        kostet += abstand(s[-1], rest[k])
                    if 0 < k < len(rest):
                        kostet -= abstand(rest[k - 1], rest[k])
                    if kostet < spart - 1e-9 and (beste is None or kostet < beste[0]):
                        neue_route = rest[:k] + s + rest[k:]
                        if neue_route != route and ist_gueltig(neue_route):
                            beste = (kostet, neue_route)
            if beste is not None:
                route = beste[1]
                geaendert = True
            i += 1
    return route, geaendert


def verbessere(route):
    """Umkehren und Verschieben, bis sich nichts mehr ändert."""
    while True:
        route, a = umkehren(route)
        route, b = verschieben(route)
        if not (a or b):
            return route


roh = min(erste[5], key=laenge)
besser = verbessere(roh)
print('wenigerkrumm5: erste Route %.1f, nach dem Verbessern %.1f, gültig: %s' % (laenge(roh), laenge(besser), ist_gueltig(besser)))
````

Wir verbessern nun von jedem Beispiel die 10 kürzesten gefundenen Routen und nehmen die beste.

## Schritt 7: Alle Beispiele und Vergleich mit den Lösungshinweisen

Die Lösungshinweise (Abschnitt 1.6) nennen die Routenlängen, die das Programm der Autoren fand („Hinweise“), und die
**optimalen** Längen, die „mit anderen Verfahren“ bestimmt wurden.

````python
ANZAHL = 10                                       # so viele der kürzesten Routen werden verbessert
HINWEISE = {1: 847.4, 2: 2183.7, 3: 1936.0, 4: 1205.1, 5: 3530.2, 6: 4146.8, 7: 5055.0}      # Programm der Lösungshinweise
OPTIMUM = {1: 847.43, 2: 2183.66, 3: 1848.05, 4: 1205.07, 5: 3257.92, 6: 3457.99, 7: 4150.64}  # kürzeste Routen laut Hinweisen

beste_route = {}
print('Beispiel   erste Route   verbessert   Hinweise   Optimum   Abstand zum Optimum')
for nr in range(1, 8):
    routen = sorted([(laenge(r), i) for i, r in enumerate(erste[nr])])
    kandidaten = [verbessere(erste[nr][i]) for d, i in routen[:ANZAHL]]
    beste = min(kandidaten, key=laenge)
    assert ist_gueltig(beste) and sorted(beste) == sorted(beispiel(nr))
    beste_route[nr] = beste
    print('%5d %13.1f %12.1f %10.1f %9.2f %12.1f %%' % (nr, routen[0][0], laenge(beste), HINWEISE[nr], OPTIMUM[nr],
                                                        max(0.0, 100 * (laenge(beste) / OPTIMUM[nr] - 1))))
````

**Auswertung:**
- Die Routen sind **gültig** (alle Abbiegungen höchstens 90°, alle Orte genau einmal); das prüft die Zelle mit `assert`.
- Bei `wenigerkrumm1`, `2`, `4` und `5` erreichen wir die optimalen Längen der Hinweise, bei `3` liegen wir mit 1848,53 statt 1848,05
  Kilometern ganz knapp darüber (+0,03 %). Die Routen des Programms der Hinweise sind bei `3` und `5` deutlich länger.
- Bei `wenigerkrumm6` (3762,8 statt 3458,0 km, +8,8 %) und `7` (4571,3 statt 4150,6 km, +10,1 %) bleiben wir über dem Optimum.
  Die Routen der Hinweise (4146,8 und 5055,0 km) sind dort aber noch länger als unsere.
- Die lokale Suche bleibt in einer Route hängen, in der kein einzelnes Umkehren oder Verschieben mehr hilft. Die optimalen Routen
  sehen oft ganz anders aus.
- Die Schrittgrenze macht das Verfahren zu einem Heuristik-Verfahren: Fände nur eine sehr lange Suche von einem Startort aus eine
  Route, geben wir auf. Bei den sieben Beispielen reichte die Grenze immer. Ohne Grenze würde das Backtracking alle Möglichkeiten
  durchsuchen und nur dann keine Route ausgeben, wenn es wirklich keine gibt.

Zur Anschauung zwei Routen: `wenigerkrumm1` (optimal) und `wenigerkrumm6` (unsere beste, über dem Optimum).

````python
import matplotlib.pyplot as plt

figur, achsen = plt.subplots(1, 2, figsize=(11, 5))
for achse, nr in zip(achsen, (1, 6)):
    route = beste_route[nr]
    achse.plot([p[0] for p in route], [p[1] for p in route], '-', color='tab:blue', linewidth=1)
    achse.plot([p[0] for p in route], [p[1] for p in route], '.', color='black', markersize=4)
    achse.plot(*route[0], 'o', color='green')       # Start
    achse.plot(*route[-1], 's', color='red')        # Ende
    achse.set_title('wenigerkrumm%d: %.1f km' % (nr, laenge(route)))
    achse.set_aspect('equal')
plt.show()
````

## Für Neugierige: Zwangspunkte und das Optimum

- **Zwangspunkte:** Abbildung 1.2 der Hinweise (angelehnt an `wenigerkrumm1`) zeigt zwei Orte A und B, bei denen jede Route
  *durch* sie hindurch einen Abbiegewinkel über 90° hätte. Dann müssen A und B Start und Ziel sein. Solche Orte kann man vorab
  bestimmen und das Backtracking damit abkürzen (nicht nötig für die Beispiele).
- **Optimum:** Die Hinweise nennen die kürzesten Routen, ohne das Verfahren zu beschreiben („mit anderen Verfahren“). Ein Weg dahin
  wäre Branch and Bound mit einer guten unteren Schranke (zum Beispiel über den kürzesten aufspannenden Baum) oder ein ganzzahliges
  Programm mit Nebenbedingungen für jede verbotene Dreierfolge. Mit dem einfachen Backtracking aus diesem Notebook ist das bei 80
  oder 100 Orten nicht zu schaffen.
- **Mehr Aufwand für kürzere Routen:** Mehr Routen verbessern (`ANZAHL`), größere Stücke verschieben oder zufällig gestörte
  Routen neu verbessern bringen oft noch einige Prozent, aber keine Garantie.

## Zum Schluss: Gibt es eine Route? – mit MiniZinc

Die Aufgabe hat zwei Teile: *irgendeine* Route zu finden und eine *kurze*. Der erste Teil lässt sich gut als **Erfüllbarkeitsproblem**
mit MiniZinc beschreiben, und das Modell kann auch beweisen, dass es *keine* Route gibt.

Wir benutzen die Techniken aus den MiniZinc-Arbeitsblättern: Variablen als Index, `alldifferent`, `forall` und ein
mehrdimensionales Array `ok` aus Wahrheitswerten. Die Variable `p[i]` ist der Ort an Stelle `i` der Route. `alldifferent(p)` sorgt
dafür, dass jeder Ort genau einmal vorkommt, und für jede Dreierfolge `p[i], p[i+1], p[i+2]` muss `ok[...]` wahr sein. Mit der
**Variablen als Index** `ok[p[i], p[i+1], p[i+2]]` schlägt MiniZinc in der Tabelle nach.

Die Koordinaten übergeben wir auf Zehntel-Kilometer gerundet als ganze Zahlen, damit die Solver nicht mit Kommazahlen rechnen
müssen. Die Tabelle `ok` hat `n³` Einträge. Statt `solve minimize` benutzen wir `solve satisfy`, weil MiniZinc die Länge nicht in
vernünftiger Zeit minimiert (siehe unten).

```minizinc
include "alldifferent.mzn";

int: n;                              % Zahl der Orte
array[1..n] of int: X;               % Koordinaten in Zehntel-Kilometern (ganzzahlig gerundet)
array[1..n] of int: Y;

% ok[a, b, c] = true, wenn man von a über b nach c fliegen darf (Skalarprodukt höchstens 0)
array[1..n, 1..n, 1..n] of bool: ok =
  array3d(1..n, 1..n, 1..n,
    [ (X[a] - X[b]) * (X[c] - X[b]) + (Y[a] - Y[b]) * (Y[c] - Y[b]) <= 0
      | a in 1..n, b in 1..n, c in 1..n ]);

array[1..n] of var 1..n: p;          % p[i] = der i-te Ort der Route
constraint alldifferent(p);
constraint forall(i in 1..n-2)(ok[p[i], p[i+1], p[i+2]]);

solve satisfy;
output ["p = \(p)\n"];
```

````python
def als_dzn(orte):
    """Datendatei (.dzn) für das MiniZinc-Modell: Koordinaten in Zehntel-Kilometern, auf ganze Zahlen gerundet."""
    xs = [round(x * 10) for x, y in orte]
    ys = [round(y * 10) for x, y in orte]
    return 'n = %d;\nX = %s;\nY = %s;' % (len(orte), xs, ys)


print(als_dzn(dreieck_mit_punkt))
````
