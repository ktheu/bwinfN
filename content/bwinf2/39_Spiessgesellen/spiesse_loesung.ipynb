### Spießgesellen – Lösung Schritt für Schritt

*39. Bundeswettbewerb Informatik (2020/2021), 2. Runde, Aufgabe 2*

Auf Daisys Party gibt es Obstspieße, die die Gäste selbst zusammenstellen. Sechs Obstsorten stehen zur Auswahl: Apfel, Banane,
Brombeere, Erdbeere, Pflaume und Weintraube. Alle Stücke einer Sorte befinden sich in einer eigenen Schüssel. Daisy liebt
Überraschungen, und so hat sie die sechs Schüsseln abgedeckt und nur mit Nummern versehen, damit die Gäste nicht vorher wissen, welche
Obstsorte sie erwartet, wenn sie es bis an die Spitze der Warteschlange vor einer bestimmten Nummer geschafft haben.

Donald gefallen Daisys Überraschungen gar nicht. Er will nur bestimmte Obstsorten essen, gibt dieses aber ungern zu. Deshalb beobachtet
er andere Gäste und notiert sich, in welchen Schlangen sie anstehen, bevor sie ihre Spieße zusammenstellen, und wie die Spieße am Ende
aussehen. So erfährt er allerdings nicht, aus welcher Schüssel welches Obststück gekommen ist. Hier sind seine ersten Beobachtungen:

- Micky hat einen Spieß mit Apfel, Banane und Brombeere, dessen Obststücke aus den Schüsseln 1, 4 und 5 stammen (aber nicht
  notwendigerweise in dieser Reihenfolge).
- Minnies Spieß besteht aus Banane, Pflaume und Weintraube und wurde aus den Schüsseln 3, 5 und 6 zusammengestellt.
- Gustav hat einen Spieß mit Apfel, Brombeere und Erdbeere und hat die Schüsseln 1, 2 und 4 aufgesucht.

Jetzt will Donald sich einen Obstspieß machen, und zwar mit Weintraube, Brombeere und Apfel. Leider kann er aus den vorhandenen
Informationen nicht eindeutig bestimmen, in welchen Schüsseln diese Obstsorten zu finden sind. Daher beobachtet er zum Schluss auch
noch Daisy.

- Daisy hat einen Spieß mit Erdbeere und Pflaume und war an den Schüsseln 2 und 6.

#### Aufgabe

a) Hilf Donald und sage ihm, aus welchen Schüsseln er sich bedienen soll. Skizziere, wie du die Menge der Schüsseln bestimmt hast.

b) Nach der Party ist Daisy begeistert: Das war ein voller Erfolg! Sie ist fest entschlossen, fulminante Obstspieß-Happenings zu
veranstalten, mit zig Obstsorten und vielen Gästen. Donald will natürlich dabei sein. Aber jetzt wird er sich erst recht anstrengen
müssen, um seinen Wunschspieß zusammenzustellen.

Schreibe ein Programm, das Donald helfen kann. Es soll die folgenden Daten einlesen:
- die Menge der verfügbaren Obstsorten;
- Informationen über einige Obstspieße: für jeden Spieß
  - die Menge der Obstsorten auf dem Spieß und
  - die gleich große Menge von Nummern der Schüsseln, aus denen diese Obstsorten stammen;
- eine Menge von Wunschsorten.

Ausgeben soll das Programm die Menge der Schüsseln, in denen die Wunschsorten zu finden sind, sofern die vorliegenden Informationen
ausreichen, um diese Menge eindeutig zu bestimmen. Ansonsten soll das Programm eine möglichst informative Meldung ausgeben.

Wende dein Programm mindestens auf die Beispiele an, die du auf den BWINF-Webseiten findest.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `spiesse1.txt` bis `spiesse7.txt` (ohne Formatbeschreibung; so lesen wir sie): Zeile 1 enthält die
Zahl `n` der Obstsorten (die Schüsseln haben die Nummern 1 bis `n`), Zeile 2 die Wunschsorten, Zeile 3 die Zahl der beobachteten
Spieße. Danach kommen für jeden Spieß zwei Zeilen: die Nummern der benutzten Schüsseln und die Obstsorten auf dem Spieß.
Die Namen aller `n` Sorten stehen nirgends; wir kennen nur die, die in der Datei vorkommen.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Mengen statt Reihenfolgen:** Ein Spieß verrät nur, *welche* Sorten aus *welchen* Schüsseln kamen, nicht, was zu was gehört.
  Darum rechnet man mit Mengen: Schnittmenge, Vereinigung und Differenz.
- **Ausschließen:** Für jede Sorte merkt man sich die Schüsseln, in denen sie noch liegen kann, und streicht mit jeder Beobachtung
  die Schüsseln, die nicht mehr in Frage kommen.
- **Fingerabdruck (Gruppieren):** Zwei Sorten, die immer zusammen auf einem Spieß liegen oder zusammen fehlen, kann man mit den
  Beobachtungen nicht unterscheiden. Mit diesem Gedanken begründet man, dass das Verfahren nichts übersieht.
- **Durchprobieren zur Kontrolle:** Bei 6 Sorten gibt es nur 6! = 720 Zuordnungen von Sorten zu Schüsseln. Die kann man alle prüfen.

**Nützliche Funktionen und Bibliotheken**
- `set`: `a & b` (Schnittmenge), `a | b` (Vereinigung), `a - b` (Differenz), `x in a`, `len(a)`; `set().union(*listen)` vereinigt
  viele Mengen auf einmal.
- Ein Wörterbuch mit `setdefault(schluessel, [])` sammelt Dinge mit gleichem Schlüssel in einer Liste.
- `itertools.permutations` für alle Zuordnungen, `math.factorial` zum Zählen.

## Die Idee

Für jede Wunschsorte führen wir eine Menge von Schüsseln, in denen sie noch liegen **kann**. Am Anfang sind das alle Schüsseln. Dann
gehen wir die Spieße durch:

- Liegt die Sorte **auf** dem Spieß, kam sie aus einer der benutzten Schüsseln: Alle anderen Schüsseln streichen wir.
- Liegt sie **nicht** auf dem Spieß, kann sie in keiner der benutzten Schüsseln liegen (sonst wäre sie auf dem Spieß): Diese
  Schüsseln streichen wir.

Am Ende vereinigen wir die Mengen aller Wunschsorten. Hat die Vereinigung **genauso viele** Schüsseln, wie es Wunschsorten gibt, dann
sind das genau Donalds Schüsseln. Sind es **mehr**, reichen die Beobachtungen nicht aus. Warum das stimmt (und warum es nie weniger
sein können), zeigen wir in Schritt 3.

## Schritt 1: Einlesen

Ein Spieß ist ein Paar `(sorten, schuesseln)` aus zwei Mengen. `lies_spiesse` gibt `n`, die Liste der Wunschsorten und die Liste der
Spieße zurück. `bekannte_sorten` sammelt alle Namen, die in der Datei vorkommen.

````python
def lies_spiesse(dateiname):
    with open(dateiname, encoding='utf-8') as datei:
        zeilen = [zeile.split() for zeile in datei.read().splitlines() if zeile.strip()]
    n = int(zeilen[0][0])
    wunsch = zeilen[1]
    spiesse = []
    for i in range(int(zeilen[2][0])):
        schuesseln = set(int(x) for x in zeilen[3 + 2 * i])
        sorten = set(zeilen[4 + 2 * i])
        assert len(sorten) == len(schuesseln)
        spiesse.append((sorten, schuesseln))
    return n, wunsch, spiesse


def bekannte_sorten(wunsch, spiesse):
    namen = set(wunsch)
    for sorten, schuesseln in spiesse:
        namen = namen | sorten
    return sorted(namen)


def datei(nr):
    return 'beispieldaten/spiesse%d.txt' % nr


for nr in range(1, 8):
    n, wunsch, spiesse = lies_spiesse(datei(nr))
    anzahl_namen = len(bekannte_sorten(wunsch, spiesse))
    print('spiesse%d.txt: n = %2d, bekannte Sorten: %2d, Spieße: %d, Wunschsorten: %2d' % (nr, n, anzahl_namen, len(spiesse), len(wunsch)))
````

In `spiesse3.txt` und `spiesse5.txt` gibt es je eine Sorte, deren Namen nie vorkommt (14 von 15 bzw. 19 von 20). Sie liegt auf keinem
der beobachteten Spieße und ist keine Wunschsorte. Ihre Schüssel gibt es trotzdem; das müssen wir später beachten.

## Schritt 2: Schüsseln streichen (Teil a)

Die Beobachtungen aus Teil a schreiben wir direkt in Python auf (Reihenfolge: Micky, Minnie, Gustav, Daisy).
`moegliche_schuesseln` macht genau das Streichen aus der Idee: Schnittmenge mit den benutzten Schüsseln, wenn die Sorte auf dem
Spieß liegt, sonst Differenz.

````python
TEIL_A = (6, ['Weintraube', 'Brombeere', 'Apfel'],
          [({'Apfel', 'Banane', 'Brombeere'}, {1, 4, 5}),        # Micky
           ({'Banane', 'Pflaume', 'Weintraube'}, {3, 5, 6}),     # Minnie
           ({'Apfel', 'Brombeere', 'Erdbeere'}, {1, 2, 4}),      # Gustav
           ({'Erdbeere', 'Pflaume'}, {2, 6})])                   # Daisy


def zeige(menge):
    """Menge sortiert als Text, z. B. {1, 4}."""
    return '{' + ', '.join(str(x) for x in sorted(menge)) + '}'


def moegliche_schuesseln(sorte, n, spiesse):
    moeglich = set(range(1, n + 1))          # am Anfang alle Schüsseln
    for sorten, schuesseln in spiesse:
        if sorte in sorten:
            moeglich = moeglich & schuesseln  # nur benutzte Schüsseln bleiben
        else:
            moeglich = moeglich - schuesseln  # benutzte Schüsseln fallen weg
    return moeglich


n_a, wunsch_a, spiesse_a = TEIL_A
namen_a = ['Start', 'Micky', 'Minnie', 'Gustav', 'Daisy']
for k in range(len(spiesse_a) + 1):          # nach den ersten k Spießen
    print('%-7s' % namen_a[k], '  '.join('%s %-18s' % (w, zeige(moegliche_schuesseln(w, n_a, spiesse_a[:k]))) for w in wunsch_a))
````

Das ist die Tabelle der Lösungshinweise: Micky trennt Weintraube von Apfel und Brombeere, Minnie streicht für Apfel und Brombeere die
Schüssel 5 und für die Weintraube die Schüssel 2,
Gustav bringt nichts Neues (er hat genau die Schüsseln, die Minnie nicht hatte), und Daisy schließt Schüssel 6 für die Weintraube aus.

Die Vereinigung ist `{1, 3, 4}`, drei Schüsseln für drei Wunschsorten. **Donald soll sich aus den Schüsseln 1, 3 und 4 bedienen.**
In Schüssel 3 liegen die Weintrauben; ob Apfel in 1 und Brombeere in 4 liegt oder umgekehrt, wissen wir nicht. Das ist Donald egal:
Er bekommt beide.

````python
def kandidaten(n, wunsch, spiesse):
    """Für jede Wunschsorte die möglichen Schüsseln und die Vereinigung aller dieser Mengen."""
    moeglich = {w: moegliche_schuesseln(w, n, spiesse) for w in wunsch}
    alle = set().union(*moeglich.values())
    return moeglich, alle


moeglich_a, alle_a = kandidaten(*TEIL_A)
print(zeige(alle_a), 'eindeutig' if len(alle_a) == len(wunsch_a) else 'nicht eindeutig')
````

## Schritt 3: Warum das Verfahren richtig ist

Dass wir nur Schüsseln streichen, die wirklich nicht gehen, ist klar. Schwieriger ist die andere Richtung: Wenn die Vereinigung zu
groß ist, gibt es dann wirklich **keine** eindeutige Antwort? Oder haben wir nur nicht klug genug kombiniert?

**Fingerabdruck.** Für jede Sorte schreiben wir hinter jeden Spieß eine `1`, wenn sie darauf liegt, sonst eine `0`. Für jede Schüssel
genauso: `1`, wenn sie für den Spieß benutzt wurde. Liegt Sorte `f` in Schüssel `b`, dann haben beide denselben Fingerabdruck (die
Sorte ist genau dann auf dem Spieß, wenn ihre Schüssel benutzt wurde). Die Streichregel von oben lässt also genau die Schüsseln übrig,
die denselben Fingerabdruck haben wie die Sorte.

Sorten und Schüsseln mit gleichem Fingerabdruck bilden eine **Gruppe**:

1. In jeder Gruppe gibt es gleich viele Sorten wie Schüsseln, denn jede Sorte liegt in einer Schüssel ihrer Gruppe und in jeder
   Schüssel liegt eine Sorte (auch die, deren Namen wir nicht kennen).
2. Vertauscht man zwei Sorten einer Gruppe, passt die Zuordnung immer noch zu allen Spießen: Jeder Spieß enthält beide oder keine, und
   er benutzt beide Schüsseln oder keine. Innerhalb einer Gruppe kann also **jede** Sorte in **jeder** Schüssel liegen.

Daraus folgt: Die Schüsselmenge der Wunschsorten ist genau dann eindeutig, wenn jede Gruppe mit einer Wunschsorte **nur** Wunschsorten
enthält. Die Vereinigung aus Schritt 2 besteht aus allen Schüsseln dieser Gruppen. Wegen 1. hat sie genau so viele Elemente wie die
Gruppen Sorten haben, also mindestens so viele wie Wunschsorten, und gleich viele genau dann, wenn keine fremde Sorte dabei ist.
Damit ist der Test „Größe der Vereinigung = Zahl der Wunschsorten“ richtig, und er kann auch nie weniger Schüsseln liefern.

`gruppen` bildet die Gruppen; Sorten ohne Namen erkennt man daran, dass eine Gruppe mehr Schüsseln als bekannte Sorten hat.

````python
def fingerabdruck_sorte(sorte, spiesse):
    return ''.join('1' if sorte in sorten else '0' for sorten, schuesseln in spiesse)


def fingerabdruck_schuessel(nummer, spiesse):
    return ''.join('1' if nummer in schuesseln else '0' for sorten, schuesseln in spiesse)


def gruppen(n, wunsch, spiesse):
    """Wörterbuch Fingerabdruck -> (Liste der bekannten Sorten, Liste der Schüsseln)."""
    gruppe = {}
    for sorte in bekannte_sorten(wunsch, spiesse):
        gruppe.setdefault(fingerabdruck_sorte(sorte, spiesse), ([], []))[0].append(sorte)
    for nummer in range(1, n + 1):
        gruppe.setdefault(fingerabdruck_schuessel(nummer, spiesse), ([], []))[1].append(nummer)
    for sorten, schuesseln in gruppe.values():
        assert len(sorten) <= len(schuesseln), 'Beobachtungen widersprechen sich'
    return gruppe


for abdruck, (sorten, schuesseln) in sorted(gruppen(*TEIL_A).items()):
    print(abdruck, sorten, schuesseln)
````

Lesehilfe: `1010` heißt „auf Mickys und Gustavs Spieß, nicht bei Minnie und Daisy“. Apfel und Brombeere haben denselben Fingerabdruck
wie die Schüsseln 1 und 4; sie sind nicht zu unterscheiden. Alle anderen Gruppen haben eine Sorte und eine Schüssel. Daraus sieht man
auch, wie viele Zuordnungen zu den Beobachtungen passen: In einer Gruppe mit `k` Schüsseln gibt es `k!` Möglichkeiten, insgesamt also
das Produkt dieser Fakultäten, hier `2! = 2`.

## Schritt 4: Probe durch Durchprobieren

Bei kleinen `n` können wir **alle** Zuordnungen von Sorten zu Schüsseln durchgehen (`itertools.permutations`) und die behalten, die zu
allen Spießen passen. Eindeutig ist die Antwort, wenn alle passenden Zuordnungen den Wunschsorten dieselbe Schüsselmenge geben.
Sorten ohne Namen bekommen Ersatznamen `?1`, `?2`, … (Teil a hat keine).

````python
from itertools import permutations
from math import factorial


def alle_sorten(n, wunsch, spiesse):
    namen = bekannte_sorten(wunsch, spiesse)
    return namen + ['?%d' % i for i in range(1, n - len(namen) + 1)]


def durchprobieren(n, wunsch, spiesse):
    """Liste der passenden Zuordnungen und Menge der möglichen Schüsselmengen der Wunschsorten."""
    sorten = alle_sorten(n, wunsch, spiesse)
    passend = []
    for reihe in permutations(range(1, n + 1)):
        schuessel = dict(zip(sorten, reihe))          # Sorte -> Schüssel
        if all(set(schuessel[s] for s in auf_spiess) == benutzt for auf_spiess, benutzt in spiesse):
            passend.append(schuessel)
    mengen = set(frozenset(z[w] for w in wunsch) for z in passend)
    return passend, mengen


def anzahl_zuordnungen(n, wunsch, spiesse):
    ergebnis = 1
    for sorten, schuesseln in gruppen(n, wunsch, spiesse).values():
        ergebnis = ergebnis * factorial(len(schuesseln))
    return ergebnis


passend_a, mengen_a = durchprobieren(*TEIL_A)
print(len(passend_a), 'passende Zuordnungen von 720, gezählt mit den Gruppen:', anzahl_zuordnungen(*TEIL_A))
for z in passend_a:
    print(z)
print('Schüsselmengen der Wunschsorten:', [zeige(m) for m in mengen_a])
````

Jetzt dasselbe mit vielen zufälligen kleinen Fällen: Wir legen eine geheime Zuordnung fest, erzeugen daraus zufällige Spieße (dann
passen sie sicher zusammen) und zufällige Wunschsorten. Sorten, die weder auf einem Spieß noch unter den Wünschen vorkommen, bleiben
namenlos, wie in den Beispieldateien. Verglichen wird: Sagt das Verfahren „eindeutig“ genau dann, wenn das Durchprobieren nur eine
Schüsselmenge findet, und ist es dann dieselbe? Und stimmt die gezählte Zahl der Zuordnungen?

````python
import random

zufall = random.Random(1)
NAMEN = 'ABCDEFG'
eindeutige = 0
for versuch in range(400):
    n = zufall.randint(2, 6)
    geheim = dict(zip(NAMEN[:n], zufall.sample(range(1, n + 1), n)))
    spiesse = []
    for _ in range(zufall.randint(0, 4)):
        sorten = set(zufall.sample(NAMEN[:n], zufall.randint(1, n)))
        spiesse.append((sorten, set(geheim[s] for s in sorten)))
    wunsch = zufall.sample(NAMEN[:n], zufall.randint(1, n))
    moeglich, alle = kandidaten(n, wunsch, spiesse)
    passend, mengen = durchprobieren(n, wunsch, spiesse)
    assert (len(alle) == len(wunsch)) == (len(mengen) == 1)
    if len(mengen) == 1:
        assert mengen == {frozenset(alle)}
        eindeutige += 1
    assert len(passend) == anzahl_zuordnungen(n, wunsch, spiesse)
print('400 Fälle verglichen, davon', eindeutige, 'eindeutig; alles gleich')
````

## Schritt 5: Antwort mit informativer Meldung

Die Lösungshinweise verlangen: Erst klar sagen, ob es eine eindeutige Schüsselmenge gibt, dazu für jede Wunschsorte die möglichen
Schüsseln. Ist die Antwort nicht eindeutig, soll Donald erfahren, **welche** Sorten noch Klärungsbedarf haben, damit er gezielt nach
weiteren Spießen Ausschau halten kann. Wir nutzen dafür die Gruppen: Eine Gruppe mit Wunschsorten **und** anderen Sorten ist das
Problem. Weiterhelfen würde ein Spieß, auf dem eine dieser Wunschsorten liegt und eine der anderen Sorten nicht (oder umgekehrt), denn
dann zerfällt die Gruppe. Außerdem sagen wir, welche Schüsseln sicher stimmen.

````python
def antwort(n, wunsch, spiesse):
    moeglich, alle = kandidaten(n, wunsch, spiesse)
    for w in wunsch:
        print('  %-13s %s' % (w, zeige(moeglich[w])))
    if len(alle) == len(wunsch):
        print('Lösung gefunden:', zeige(alle))
        return alle
    print('Keine eindeutige Lösung. In Frage kommen', zeige(alle), '(%d Schüsseln für %d Wunschsorten)' % (len(alle), len(wunsch)))
    sicher = set()
    for sorten, schuesseln in gruppen(n, wunsch, spiesse).values():
        gewuenscht = [s for s in sorten if s in wunsch]
        andere = [s for s in sorten if s not in wunsch]
        namenlos = len(schuesseln) - len(sorten)
        if not gewuenscht:
            continue
        if not andere and namenlos == 0:
            sicher = sicher | set(schuesseln)
            continue
        if namenlos > 0:
            andere.append('%d Sorte(n) ohne Namen' % namenlos)
        print('- %s: Die Schüsseln %s enthalten außerdem %s.' % (', '.join(gewuenscht), zeige(schuesseln), ', '.join(andere)))
        print('  Nimmt Donald alle diese Schüsseln, bekommt er sicher %s, aber auch %s.' % (', '.join(gewuenscht), ', '.join(andere)))
        print('  Weiterhelfen würde ein Spieß, der eine Sorte aus (%s) von einer aus (%s) trennt: die eine darauf, die andere nicht.' %
              (', '.join(gewuenscht), ', '.join(andere)))
    print('Sicher richtig sind die Schüsseln', zeige(sicher))
    return None


antwort(*TEIL_A)
````

## Schritt 6: Alle Beispiele

Zu jedem Beispiel zählen wir zusätzlich, wie viele Zuordnungen von Sorten zu Schüsseln zu den Beobachtungen passen (Produkt der
Fakultäten der Gruppengrößen aus Schritt 3). Durchprobieren könnten wir das bei `n = 26` nicht mehr.

````python
ergebnisse = {}
for nr in range(1, 8):
    n, wunsch, spiesse = lies_spiesse(datei(nr))
    print('spiesse%d.txt (passende Zuordnungen: %d)' % (nr, anzahl_zuordnungen(n, wunsch, spiesse)))
    ergebnisse[nr] = antwort(n, wunsch, spiesse)
    print()
````

**Vergleich mit den Lösungshinweisen:** Alle sieben Beispiele stimmen überein, Sorte für Sorte (geprüft am Ende des Generators).
Eindeutig sind 1, 2, 4, 5 und 6, nicht eindeutig sind 3 und 7, mit denselben Problemfällen wie dort: In `spiesse3.txt` teilt sich
Litschi die Schüsseln 2 und 11 mit Grapefruit, in `spiesse7.txt` teilt sich Ugli die Schüsseln 18 und 25 mit Banane, und Apfel,
Grapefruit und Xenia teilen sich 3, 10, 20 und 26 mit Litschi. Die Hinweise nennen nur die Wunschsorten; dass wir auch die fremde Sorte
nennen, hilft Donald bei der Suche nach einem klärenden Spieß.

Zwei Besonderheiten:

- In `spiesse1.txt` kommt die Wunschsorte Grapefruit auf keinem Spieß vor (die Lösungshinweise verlangen, dass das richtig behandelt
  wird). Trotzdem ist ihre Schüssel bekannt: Alle anderen neun Sorten liegen auf irgendeinem Spieß, nur Schüssel 7 wurde nie benutzt.
- In `spiesse3.txt` gibt es eine Sorte ohne Namen. Sie ist allein in ihrer Gruppe (zusammen mit der nie benutzten Schüssel) und stört
  deshalb nicht. Läge sie mit einer Wunschsorte in einer Gruppe, würde die Meldung „Sorte(n) ohne Namen“ erscheinen.

## Zum Schluss: Mit MiniZinc

Die Aufgabe ist eine Zuordnung, wie du sie von den Arbeitsblättern zu MiniZinc kennst: Jede Sorte `f` bekommt eine Schüssel `x[f]`,
alle verschieden (`alldifferent`). Für die Beobachtungen nutzen wir zwei 0/1-Tabellen (`array2d`) und eine **Variable als Index**:
Sorte `f` liegt genau dann auf Spieß `s`, wenn ihre Schüssel `x[f]` für `s` benutzt wurde, also `auf[s, f] = benutzt[s, x[f]]`.
Die Sorten sind dabei durchnummeriert (bekannte Sorten alphabetisch, dann die ohne Namen). Neu ist `set of int`, eine Menge von Zahlen,
in den Daten als `{1, 3, 4}` geschrieben; `[x[f] | f in wunsch]` in der Ausgabe ist die Liste der Schüsseln der Wunschsorten.
```minizinc
include "alldifferent.mzn";
int: n;                                   % Zahl der Sorten und Schüsseln
int: L;                                   % Zahl der Spieße
array[1..L, 1..n] of 0..1: auf;           % auf[s, f] = 1: Sorte f liegt auf Spieß s
array[1..L, 1..n] of 0..1: benutzt;       % benutzt[s, b] = 1: Schüssel b wurde für Spieß s benutzt
set of int: wunsch;                       % Nummern der Wunschsorten
array[1..n] of var 1..n: x;               % x[f]: die Schüssel der Sorte f

constraint alldifferent(x);               % jede Sorte in einer eigenen Schüssel
constraint forall(s in 1..L, f in 1..n)(auf[s, f] = benutzt[s, x[f]]);

solve satisfy;
output ["x = \(x)\n", "Wunschsorten: \([x[f] | f in wunsch])\n"];
```

Der Solver findet **eine** passende Zuordnung. Ob die Schüsselmenge eindeutig ist, sagt sie noch nicht. Dafür ein zweiter Lauf: Wir
geben die Schüsselmenge `W` der ersten Lösung vor und verlangen, dass **mindestens eine** Wunschsorte außerhalb von `W` liegt
(`exists` ist das Gegenstück zu `forall`: „für mindestens ein `f`“). Antwortet der Solver `UNSATISFIABLE`, gibt es keine solche
Zuordnung, und `W` ist eindeutig. Findet er eine, hat er gezeigt, dass es nicht eindeutig ist. Dazu kommen diese zwei Zeilen vor
`solve`:
```minizinc
set of int: W;                            % die Schüsselmenge aus der ersten Lösung
constraint exists(f in wunsch)(not (x[f] in W));
```

Die Datei mit den Daten erzeugt die Python-Zelle `als_dzn` (für den zweiten Lauf mit `W`).

## Für Neugierige

- **Laufzeit:** Bei `n` Sorten, `k` Wunschsorten und `L` Spießen macht `moegliche_schuesseln` für jede Wunschsorte `L` Mengenoperationen
  mit Mengen aus höchstens `n` Elementen, zusammen `O(L · k · n)`. Die Gruppen kosten `O(L · n)` für die Fingerabdrücke.
- **Bitvektoren:** Die Lösungshinweise speichern Schüsselmengen als Bits einer ganzen Zahl: Schüssel `i` ist Bit `i`. Dann ist die
  Schnittmenge `a & b`, die Differenz `a & ~b`, die Vereinigung `a | b`, und `bin(a).count('1')` zählt die Elemente. Bei wenigen
  Schüsseln ist jede Operation ein einziger Rechenschritt.
- **Bipartites Matching:** Man kann die Aufgabe auch als Graph sehen: links die Sorten, rechts die Schüsseln, eine Kante, wenn die
  Sorte in der Schüssel liegen kann. Gesucht sind die Kanten, die in einer perfekten Zuordnung (Matching) vorkommen können. Weil jeder
  Spieß Sorten **und** Schüsseln gleichermaßen trennt, zerfällt der Graph hier in vollständige Gruppen; darum genügt das einfache
  Streichen, und man braucht keinen Matching-Algorithmus.
- **Wie viele Spieße braucht man?** Jeder Spieß teilt jede Gruppe in zwei Teile (auf dem Spieß, nicht auf dem Spieß). Mit `L` Spießen
  gibt es höchstens `2^L` Fingerabdrücke. Um alle `n` Sorten vollständig zu trennen, braucht Donald also mindestens `log₂ n` Spieße,
  bei 26 Sorten mindestens 5. Die vier Spieße in `spiesse7.txt` können das gar nicht schaffen.
