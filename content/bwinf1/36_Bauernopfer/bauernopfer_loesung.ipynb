### Bauernopfer – Lösung Schritt für Schritt

*36. Bundeswettbewerb Informatik (2017), 1. Runde, Aufgabe 5*

Das Spiel Bauernopfer wird zwischen weißen Bauern und einem schwarzen Turm auf einem Schachbrett gespielt. Zuerst stellen sich die weißen Bauern auf, wie sie wollen, jeder auf einem anderen Feld. Danach stellt sich der schwarze Turm auf, wo er will, aber nicht, wo bereits ein Bauer steht. Dann gibt es abwechselnd einen weißen und einen schwarzen Zug. In einem weißen Zug bewegt sich ein weißer Bauer einen Schritt (um ein Feld) nach oben, nach unten, nach links oder nach rechts. In einem schwarzen Zug bewegt sich der schwarze Turm beliebig viele (evtl. null) Schritte nach oben, nach unten, nach links oder nach rechts, wobei er allerdings weder ein Feld erreichen darf, das bereits von einem weißen Bauern belegt ist, noch durch ein solches Feld „hindurchziehen“ darf.

Das Ziel der weißen Bauern ist, den schwarzen Turm zu fangen, nämlich dadurch, dass einer von ihnen das Feld erreicht, auf dem der Turm steht. Das Ziel des schwarzen Turms ist, nie gefangen zu werden.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Eine Invariante finden:** Eine Eigenschaft der Stellung, die nach jedem Zug des Gegners wieder hergestellt werden kann. Beim Turm gegen sieben Bauern ist das „Ich stehe auf einem Feld, in dessen Reihe und Spalte kein Bauer steht“. Hat man so eine Invariante, ist der Beweis kurz: Der Gegner kann sie nicht zerstören, ohne dass man sie sofort wiederherstellen kann.
- **Schubfachprinzip:** Sieben Bauern belegen höchstens sieben Spalten und höchstens sieben Reihen. Von acht Spalten bleibt also mindestens eine frei, von acht Reihen auch.
- **Eine Strategie als Plan:** Wenn Weiß immer dieselbe Folge von Zügen spielt (egal, was der Turm tut), ist die Partie eine feste Liste von Stellungen. Den besten Turm dagegen findet man durch **Rekursion mit Merkliste**: „Wie lange hält der Turm von hier aus noch durch?“ hängt nur von der Stellung ab.
- **Vollständige Suche im Spielgraph (Gewinnmenge):** Auf kleinen Brettern kann man alle Stellungen aufzählen und ausrechnen, wer gewinnt. Man beginnt mit den Stellungen, in denen Weiß sofort fängt, und ergänzt immer wieder Stellungen, in denen Weiß einen Zug hat, der in eine schon gewonnene Stellung führt (bzw. Schwarz-Stellungen, in denen **alle** Züge des Turms in gewonnene Stellungen führen), bis sich nichts mehr ändert.

**Nützliche Funktionen und Bibliotheken**
- `set` und `frozenset` für Stellungen (die Bauern sind gleich, ihre Reihenfolge ist egal), `in` zum Prüfen, ob ein Feld besetzt ist.
- `itertools.combinations(felder, k)` zählt alle Möglichkeiten auf, k Bauern auf die Felder zu stellen (nur für die vollständige Suche).
- Dictionaries als Merkliste (`merk[(t, feld)] = ergebnis`).

## Die Ideen in Worten

**Teil 1, acht Bauern.** Die Bauern bilden eine Mauer, die das Brett in zwei Hälften teilt. Der Turm steht in einer der Hälften und kann nicht auf die andere Seite. Jetzt schiebt Weiß die Mauer Bauer für Bauer auf den Turm zu, ohne dass je eine Lücke entsteht. Das Gebiet des Turms wird kleiner und kleiner, bis er gefangen ist. Wir spielen zwei Aufstellungen durch:

- *Strategie 1:* Die Bauern stehen auf der Diagonale von a8 nach h1. Weiß zieht von links nach rechts jeden Bauern ein Feld nach unten (wenn der Turm unten links steht) und fängt dann von vorne wieder an. Nach höchstens 7 + 6 + … + 1 = 28 Zügen ist der Turm gefangen, unabhängig davon, was er tut.
- *Strategie 2:* Die Bauern stehen alle in der 5. Reihe. Weiß zieht jeweils den Bauern in der **Spalte des Turms** ein Feld auf den Turm zu. Das reagiert auf den Turm („adaptiv“). In der Mauer entstehen Löcher, aber der Turm kommt nicht durch.

Den Turm lassen wir jeweils so spielen, dass die Partie möglichst lange dauert.

**Teil 2, sieben Bauern.** Hier gewinnt der **Turm**. Er steht auf einem Feld, in dessen Reihe und Spalte kein Bauer steht (so ein Feld gibt es nach dem Schubfachprinzip). Ein Bauernzug kann nur die Reihe *oder* die Spalte des Turms betreten, nie beides. Der Turm zieht dann entlang der anderen, noch freien Linie auf ein Feld, das wieder in einer freien Reihe **und** freien Spalte liegt (auch das gibt es). Gefangen werden kann er so nie.

**Teil 3 und 4.** Für größere l und für die Dame rechnen wir auf kleinen Brettern (3×3 und 4×4) alles durch und vergleichen mit den Aussagen der Lösungshinweise für das 8×8-Brett.

## Schritt 1: Das Brett

Felder sind Paare `(x, y)` mit `x` von 0 (Spalte a) bis 7 (Spalte h) und `y` von 0 (Reihe 1) bis 7 (Reihe 8). Die Bauern stehen in einer Menge. `turm_ziele` liefert alle Felder, auf die der Turm ziehen darf (einschließlich seines eigenen Feldes, denn null Schritte sind erlaubt): in jede Richtung so weit, bis das Brett endet oder ein Bauer im Weg steht.

```python
N = 8


def feldname(feld):
    """(0, 7) -> a8"""
    return "abcdefgh"[feld[0]] + str(feld[1] + 1)


def zeige(bauern, turm):
    """Gibt das Brett als Text aus (B = Bauer, T = Turm), Reihe 8 oben."""
    for y in range(N - 1, -1, -1):
        zeile = str(y + 1) + " "
        for x in range(N):
            if (x, y) in bauern:
                zeile = zeile + "B "
            elif (x, y) == turm:
                zeile = zeile + "T "
            else:
                zeile = zeile + ". "
        print(zeile)
    print("  a b c d e f g h")


def turm_ziele(bauern, turm):
    """Alle Felder, auf die der Turm ziehen darf (auch das eigene Feld)."""
    ziele = [turm]
    for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
        x = turm[0] + dx
        y = turm[1] + dy
        while 0 <= x < N and 0 <= y < N and (x, y) not in bauern:
            ziele.append((x, y))
            x = x + dx
            y = y + dy
    return ziele


beispiel_bauern = {(i, 7 - i) for i in range(N)}          # Diagonale a8 bis h1
zeige(beispiel_bauern, (1, 1))
print(len(turm_ziele(beispiel_bauern, (1, 1))), "Felder erreichbar (mit dem eigenen):", [feldname(f) for f in turm_ziele(beispiel_bauern, (1, 1))])
```

## Teil 1, Strategie 1: die Diagonalmauer

Die Bauern stehen auf `a8, b7, c6, …, h1`, also bei `(i, 7 - i)`. Steht der Turm unterhalb der Diagonale (`x + y < 7`), schiebt Weiß die Mauer nach unten:

- Runde `s = 7`: Die Bauern in den Spalten `0` bis `6` ziehen von links nach rechts je ein Feld nach unten. Danach liegt die Mauer auf `x + y = 6`, und der Bauer in Spalte 7 bleibt auf h1 stehen.
- Runde `s = 6`: Die Bauern in den Spalten `0` bis `5` ziehen wieder ein Feld nach unten, und so weiter bis `s = 1`.

Zu keinem Zeitpunkt gibt es eine Lücke: Zwei benachbarte Bauern der Mauer stehen entweder in derselben Reihe nebeneinander oder diagonal, und durch eine Diagonale kann ein Turm nicht ziehen. Der Plan hängt nicht vom Turm ab, wir speichern ihn deshalb als Liste von Zügen `(altes Feld, neues Feld)`. Steht der Turm oberhalb der Diagonale, spielt Weiß denselben Plan um 180° gedreht (`(x, y)` wird zu `(7 - x, 7 - y)`; die Diagonale geht dabei in sich selbst über).

```python
def drehe(feld):
    """Dreht das Brett um 180 Grad."""
    return (N - 1 - feld[0], N - 1 - feld[1])


def plan_diagonale(oben_rechts):
    """Liste der weißen Züge (altes Feld, neues Feld). oben_rechts=True: Turm steht oberhalb der Diagonale."""
    hoehe = [7 - i for i in range(N)]                      # Reihe des Bauern in Spalte i
    plan = []
    for s in range(7, 0, -1):                               # Mauer von x + y = 7 nach x + y = s - 1
        for i in range(s):
            alt = (i, hoehe[i])
            neu = (i, hoehe[i] - 1)
            hoehe[i] = hoehe[i] - 1
            if oben_rechts:
                alt = drehe(alt)
                neu = drehe(neu)
            plan.append((alt, neu))
    return plan


def stellungen(startbauern, plan):
    """Die Bauern vor dem Zug 0, nach Zug 1, ... (Liste von Mengen)."""
    folge = [set(startbauern)]
    for alt, neu in plan:
        bauern = set(folge[-1])
        bauern.remove(alt)
        bauern.add(neu)
        folge.append(bauern)
    return folge


plan1 = plan_diagonale(False)
print(len(plan1), "Züge, die ersten:", [feldname(a) + "-" + feldname(n) for a, n in plan1[:8]])
folge1 = stellungen(beispiel_bauern, plan1)
```

### Der beste Turm gegen Strategie 1

Gegen einen festen Plan können wir den **bestmöglichen Turm** ausrechnen. `haelt_durch(t, turm)` ist die Nummer des weißen Zuges, mit dem der Turm gefangen wird, wenn Weiß gerade den Zug `t` spielt und der Turm auf `turm` steht und von dort aus so geschickt wie möglich spielt. Zieht der Bauer im Plan genau auf den Turm, ist es `t + 1`. Sonst darf der Turm in der neuen Stellung ein beliebiges Ziel wählen, und wir nehmen das beste. Weil es nur 28 Züge und 64 Felder gibt, ist das schnell fertig, wenn wir uns die Ergebnisse in einem Dictionary merken.

```python
def haelt_durch(t, turm, plan, folge, merk):
    """Nummer des weißen Zuges, mit dem der Turm gefangen wird (der Turm spielt bestmöglich). Gibt auch das nächste Feld zurück."""
    if (t, turm) in merk:
        return merk[(t, turm)]
    alt, neu = plan[t]
    if neu == turm:
        ergebnis = (t + 1, None)
    else:
        ergebnis = (-1, None)
        for ziel in turm_ziele(folge[t + 1], turm):
            wert = haelt_durch(t + 1, ziel, plan, folge, merk)[0]
            if wert > ergebnis[0]:
                ergebnis = (wert, ziel)
    merk[(t, turm)] = ergebnis
    return ergebnis


def bester_start(plan, folge, startfelder):
    """Das Startfeld aus startfelder, mit dem der Turm am längsten durchhält, und die Zahl der weißen Züge."""
    merk = {}
    beste = None
    for feld in startfelder:
        wert = haelt_durch(0, feld, plan, folge, merk)[0]
        if beste is None or wert > beste[1]:
            beste = (feld, wert)
    return beste[0], beste[1], merk


unten_links = [(x, y) for x in range(N) for y in range(N) if x + y < 7]
start1, zuege1, merk1 = bester_start(plan1, folge1, unten_links)
print("bestes Startfeld des Turms:", feldname(start1), "- gefangen mit dem", zuege1, ". weißen Zug (", len(merk1), "Stellungen untersucht)")
```

Der Turm hält höchstens die 28 Züge durch, die auch die Lösungshinweise nennen. Am besten steht er in der Ecke `a1` und bleibt dort stehen, denn kein Zug kann ihn retten. Zur Probe die Treibjagd selbst: Der Turm folgt seinen besten Zügen (`haelt_durch` liefert sie gleich mit). Hier die Stellung vor dem ersten Zug, nach der ersten Runde (7 Züge), nach der dritten Runde (18 Züge) und vor dem letzten Zug.

```python
def spiele_plan(plan, folge, merk, turm):
    """Spielt den Plan durch, der Turm folgt seinen besten Zügen. Gibt die Turmfelder vor jedem weißen Zug zurück."""
    felder = [turm]
    for t in range(len(plan)):
        wert, ziel = merk[(t, felder[-1])]
        if ziel is None:
            return felder, t + 1
        felder.append(ziel)
    return felder, len(plan)


felder1, anzahl1 = spiele_plan(plan1, folge1, merk1, start1)
for t in [0, 7, 18, anzahl1 - 1]:
    print("vor dem weißen Zug", t + 1, "(" + feldname(plan1[t][0]) + "-" + feldname(plan1[t][1]) + "), Turm auf", feldname(felder1[t]))
    zeige(folge1[t], felder1[t])
    print()
print("Der Turm wird mit dem", anzahl1, ". Zug gefangen.")
```

**Turm von oben rechts:** Steht der Turm oberhalb der Diagonale, gilt wegen der Drehung dasselbe. Zur Kontrolle rechnen wir es ebenfalls aus.

```python
plan1_oben = plan_diagonale(True)
folge1_oben = stellungen(beispiel_bauern, plan1_oben)
oben_rechts = [(x, y) for x in range(N) for y in range(N) if x + y > 7]
start1_oben, zuege1_oben, merk1_oben = bester_start(plan1_oben, folge1_oben, oben_rechts)
print("Turm oben rechts: bestes Startfeld", feldname(start1_oben), "- gefangen mit dem", zuege1_oben, ". Zug")
```

## Teil 1, Strategie 2: die Reihe, die dem Turm entgegenkommt

Alle acht Bauern stehen in der 5. Reihe (`y = 4`), in jeder Spalte einer. Der Turm steht irgendwo außerhalb dieser Reihe, also oberhalb oder unterhalb. Weiß spielt **adaptiv**: Er zieht den Bauern in der Spalte des Turms ein Feld in Richtung des Turms. Steht der Bauer schon direkt neben dem Turm, ist das der Fangzug.

Weil jede Spalte einen Bauern hat und der Bauer in der Spalte des Turms ihm den Weg versperrt, kann der Turm die Mauer nie überqueren. Wir speichern nur die Höhe `hoehe[x]` des Bauern in Spalte `x`.

Die Gegenstrategie des Turms steht in den Lösungshinweisen: Er geht in die Ecke (hier `a1`), bleibt stehen und lässt den Bauern bis auf ein Feld herankommen (3 Züge), weicht dann zur nächsten Spalte auf der Grundreihe aus, wo der Bauer noch weit weg ist, und so weiter. Wir übernehmen die Regel „bleibe stehen, solange der Bauer in meiner Spalte noch mehr als ein Feld entfernt ist, sonst weiche in die Spalte aus, in der der Bauer am weitesten entfernt ist (bei Gleichstand die linke)“.

```python
def weisser_zug2(hoehe, turm):
    """Strategie 2: Der Bauer in der Spalte des Turms zieht ein Feld auf den Turm zu. Gibt (Spalte, neue Höhe) zurück."""
    x, y = turm
    if y < hoehe[x]:
        return x, hoehe[x] - 1
    return x, hoehe[x] + 1


def turm_zug2(hoehe, turm):
    """Turm: stehen bleiben, solange der Bauer in meiner Spalte mindestens zwei Felder entfernt ist, sonst in die Spalte mit dem weitesten Bauern."""
    x, y = turm
    if abs(hoehe[x] - y) > 1:
        return turm
    beste_spalte = 0
    for spalte in range(N):
        if abs(hoehe[spalte] - y) > abs(hoehe[beste_spalte] - y):
            beste_spalte = spalte
    return (beste_spalte, y)


def partie2(turm):
    """Spielt die Partie; gibt die Zahl der weißen Züge, die Stellungen (Bauernmengen) und die Turmfelder zurück."""
    hoehe = [4] * N
    stellungen2 = []
    turmfelder2 = []
    zuege = 0
    while True:
        stellungen2.append({(x, hoehe[x]) for x in range(N)})
        turmfelder2.append(turm)
        spalte, neu = weisser_zug2(hoehe, turm)
        zuege = zuege + 1
        if neu == turm[1]:                                  # der Bauer zieht auf den Turm
            return zuege, stellungen2, turmfelder2
        hoehe[spalte] = neu
        turm = turm_zug2(hoehe, turm)


zuege2, stell2, turm2 = partie2((0, 0))
print("Gefangen mit dem", zuege2, ". weißen Zug")
print("Turmfelder vor jedem weißen Zug:", " ".join(feldname(f) for f in turm2))
for t in [0, 3, 12, zuege2 - 1]:
    print("vor dem weißen Zug", t + 1)
    zeige(stell2[t], turm2[t])
    print()
```

Gefangen wird der Turm mit dem 25. Zug, also nach höchstens 3 · 8 + 1 = 25 Zügen wie in den Lösungshinweisen. Der Turm steht dabei jeweils drei weiße Züge lang in einer Spalte (der Bauer rückt von Reihe 5 auf Reihe 2 vor), weicht dann aus und hat im Moment des Ausweichens ein „Loch“ in der Mauer vor sich, durch das er aber nicht entkommen kann, weil in jeder Spalte ein Bauer zwischen ihm und der anderen Seite steht.

Eine vollständige Suche über alle Turmzüge gegen diese Strategie (rund eine Million Stellungen, auf dem Notebook zu langsam) fand keine längere Partie als 25 Züge. Mit dieser Strategie ist also nichts Besseres für den Turm drin.

Zum Vergleich: Strategie 1 braucht im ungünstigsten Fall 28 Züge, Strategie 2 nur 25 und ist dazu noch einfacher zu beschreiben. Dafür muss Weiß in jedem Zug auf den Turm schauen.

## Teil 2: Sieben Bauern können den Turm nicht fangen

**Der Turm gewinnt.** Wir nennen ein Feld *sicher*, wenn in seiner Reihe **und** in seiner Spalte kein Bauer steht.

1. Nach dem Aufstellen gibt es ein sicheres Feld: Sieben Bauern besetzen höchstens sieben Spalten, also bleibt eine Spalte frei, und höchstens sieben Reihen, also bleibt eine Reihe frei. Ihr Kreuzungsfeld ist sicher (und kein Bauer steht darauf).
2. Der Turm steht auf einem sicheren Feld. Weiß zieht einen Bauern um ein Feld. Der Bauer stand weder in der Reihe noch in der Spalte des Turms, nachher ist er in höchstens einer der beiden. Gefangen ist der Turm nicht, denn dafür müsste der Bauer genau auf das Turmfeld ziehen, aber dann wäre er vorher direkt neben dem Turm gewesen und damit in seiner Reihe oder Spalte.
3. Ist das Feld noch sicher, bleibt der Turm stehen. Sonst steht der Bauer jetzt z. B. in der Spalte des Turms, die Reihe ist aber noch frei. Der Turm läuft entlang seiner (leeren) Reihe in eine Spalte, die frei von Bauern ist (die gibt es wieder: sieben Bauern, acht Spalten) und steht wieder auf einem sicheren Feld. Der Weg dorthin ist frei, weil in der Reihe kein Bauer steht. Betritt der Bauer die Reihe, läuft der Turm entsprechend die Spalte entlang.

Das ist die Invariante „Der Turm steht nach seinem Zug auf einem sicheren Feld“. Sie hält für beliebig viele Züge, und das für **jede** Strategie von Weiß.

```python
import random


def ist_sicher(bauern, feld):
    """True, wenn in der Reihe und in der Spalte von feld kein Bauer steht."""
    for b in bauern:
        if b[0] == feld[0] or b[1] == feld[1]:
            return False
    return True


def turm_zug7(bauern, turm):
    """Bleibt stehen, wenn das Feld sicher ist, sonst zieht der Turm auf ein erreichbares sicheres Feld. Gibt None zurück, wenn es keins gibt."""
    if ist_sicher(bauern, turm):
        return turm
    for ziel in turm_ziele(bauern, turm):
        if ist_sicher(bauern, ziel):
            return ziel
    return None


def bauernzuege(bauern):
    """Alle weißen Züge als Liste von (altes Feld, neues Feld)."""
    zuege = []
    for b in bauern:
        for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            z = (b[0] + dx, b[1] + dy)
            if 0 <= z[0] < N and 0 <= z[1] < N and z not in bauern:
                zuege.append((b, z))
    return zuege


# Kontrolle der Invariante: 3000 zufällige Aufstellungen, Turm auf einem sicheren Feld, alle weißen Züge probiert.
random.seed(2)
alle_felder = [(x, y) for x in range(N) for y in range(N)]
geprueft = 0
for versuch in range(3000):
    bauern = set(random.sample(alle_felder, 7))
    turm = random.choice([f for f in alle_felder if ist_sicher(bauern, f)])
    for alt, neu in bauernzuege(bauern):
        assert neu != turm                                   # kein Fang
        danach = (bauern - {alt}) | {neu}
        assert turm_zug7(danach, turm) is not None           # der Turm findet wieder ein sicheres Feld
        geprueft = geprueft + 1
print(geprueft, "weiße Züge aus 3000 Stellungen geprüft: der Turm findet immer ein sicheres Feld")
```

### Partien gegen zwei Strategien von Weiß

Zur Visualisierung lassen wir den Turm gegen zwei Strategien von Weiß spielen, jeweils 200 Züge lang, mit zufälliger Aufstellung der sieben Bauern:

- **zufällig:** ein zufälliger erlaubter Bauernzug;
- **Jäger:** der Bauernzug, nach dem ein Bauer dem Turm (in Schritten, `|dx| + |dy|`) am nächsten ist, bei Gleichstand zufällig.

```python
def weiss_zufall(bauern, turm):
    return random.choice(bauernzuege(bauern))


def weiss_jaeger(bauern, turm):
    zuege = bauernzuege(bauern)
    zufallszahl = [random.random() for z in zuege]
    beste = 0
    for i in range(len(zuege)):
        d = abs(zuege[i][1][0] - turm[0]) + abs(zuege[i][1][1] - turm[1])
        db = abs(zuege[beste][1][0] - turm[0]) + abs(zuege[beste][1][1] - turm[1])
        if (d, zufallszahl[i]) < (db, zufallszahl[beste]):
            beste = i
    return zuege[beste]


def partie7(weiss, nr, zuege=200):
    """Eine Partie. Gibt (gefangen?, Zahl der Turmbewegungen, Stellungen) zurück. nr startet die Zufallszahlen."""
    random.seed(nr)
    bauern = set(random.sample(alle_felder, 7))
    turm = random.choice([f for f in alle_felder if ist_sicher(bauern, f)])
    verlauf = [(set(bauern), turm)]
    bewegungen = 0
    for t in range(zuege):
        alt, neu = weiss(list(bauern), turm)
        if neu == turm:
            return True, bewegungen, verlauf
        bauern.remove(alt)
        bauern.add(neu)
        ziel = turm_zug7(bauern, turm)
        if ziel != turm:
            bewegungen = bewegungen + 1
        turm = ziel
        verlauf.append((set(bauern), turm))
    return False, bewegungen, verlauf


for name, weiss in [("zufällig", weiss_zufall), ("Jäger", weiss_jaeger)]:
    gefangen = 0
    bewegungen = 0
    for nr in range(200):
        g, b, verlauf = partie7(weiss, nr)
        if g:
            gefangen = gefangen + 1
        bewegungen = bewegungen + b
    print("Weiß", name + ":", gefangen, "von 200 Partien mit Fang; der Turm musste im Schnitt", bewegungen / 200, "mal ausweichen")

g, b, verlauf = partie7(weiss_jaeger, 3, zuege=20)
print()
print("Beispielpartie (Jäger), Stellung am Anfang, nach 10 und nach 20 Zügen:")
for t in [0, 10, 20]:
    print("nach", t, "Zügen, Turm auf", feldname(verlauf[t][1]))
    zeige(verlauf[t][0], verlauf[t][1])
    print()
```

Kein Fang, wie es die Invariante verspricht. Der Jäger zwingt den Turm häufiger zum Ausweichen als der Zufall, mehr schafft er nicht.

**Antwort auf Teil 2:** Sieben Bauern fangen den Turm nicht, wenn der Turm wie oben spielt. Die Frage, ob wirklich *alle* Strategien von Weiß scheitern, beantwortet der Beweis oben (er gilt für jeden Weißzug). Den Gegenbeweis für kleinere Bretter rechnen wir unten vollständig durch.

## Teil 3 und 4: Alles durchrechnen auf kleinen Brettern

Auf dem 8×8-Brett sind das viel zu viele Stellungen (`C(64, 7)` = rund 621 Millionen Aufstellungen allein für die Bauern). Auf einem 3×3- oder 4×4-Brett kann man aber **alle** Stellungen aufzählen und ausrechnen, wer gewinnt. Das hilft, Vermutungen zu prüfen, auch wenn Ränder bei so kleinen Brettern stärker ins Gewicht fallen.

**Verallgemeinerung:** `n` ist die Brettgröße, `k` die Zahl der Bauern, `l` die Zahl der Bauernschritte pro weißem Zug (in jedem Zug genau `l` Schritte, auch mit demselben Bauern mehrmals; das ist die Variante, die die Lösungshinweise zuerst behandeln) und `dame` schaltet die diagonalen Turmzüge dazu.

**Vorgehen:**
- Ein *Weiß-Zustand* ist `(Bauernmenge, Turmfeld, m)` mit `m` übrigen Schritten im weißen Zug, ein *Schwarz-Zustand* `(Bauernmenge, Turmfeld)`.
- Wir sammeln die Zustände, in denen Weiß gewinnt, in Mengen: Ein Weiß-Zustand ist gewonnen, wenn ein Bauer auf den Turm ziehen kann oder ein Schritt in einen gewonnenen Zustand führt (bei `m = 1` in einen gewonnenen Schwarz-Zustand, sonst in einen Weiß-Zustand mit `m - 1`). Ein Schwarz-Zustand ist gewonnen, wenn **jeder** Turmzug (auch stehen bleiben) in einen gewonnenen Weiß-Zustand mit `l` Schritten führt.
- Das wiederholen wir, bis sich nichts mehr ändert. Zuletzt gewinnt Weiß, wenn es eine Aufstellung gibt, bei der alle Startfelder des Turms gewonnen sind.

Das ist der Code dazu. Die Bauernmengen sind `frozenset`s, damit man sie in Mengen und als Schlüssel von Dictionaries verwenden kann.

```python
from itertools import combinations


def weiss_gewinnt(n, k, l, dame=False):
    """True, wenn k Bauern auf dem n x n-Brett den Turm (oder die Dame) mit l Schritten pro Zug fangen können."""
    feld_nr = {}
    for y in range(n):
        for x in range(n):
            feld_nr[(x, y)] = y * n + x
    felder = list(feld_nr.keys())
    nachbarn = []
    for (x, y) in felder:
        nachbarn.append([feld_nr[(x + dx, y + dy)] for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)] if (x + dx, y + dy) in feld_nr])
    richtungen = [(1, 0), (-1, 0), (0, 1), (0, -1)]
    if dame:
        richtungen = richtungen + [(1, 1), (1, -1), (-1, 1), (-1, -1)]

    def erreichbar(bauern, r):
        x, y = felder[r]
        ziele = [r]
        for dx, dy in richtungen:
            a = x + dx
            b = y + dy
            while (a, b) in feld_nr and feld_nr[(a, b)] not in bauern:
                ziele.append(feld_nr[(a, b)])
                a = a + dx
                b = b + dy
        return ziele

    aufstellungen = [frozenset(c) for c in combinations(range(n * n), k)]
    schwarz_zuege = {}                                     # (Bauern, Turm) -> erreichbare Turmfelder
    for bauern in aufstellungen:
        for r in range(n * n):
            if r not in bauern:
                schwarz_zuege[(bauern, r)] = erreichbar(bauern, r)
    gewonnen_schwarz = set()                               # Schwarz am Zug, Weiß gewinnt
    gewonnen_weiss = [set() for m in range(l + 1)]         # Weiß am Zug mit m Schritten
    geaendert = True
    while geaendert:
        geaendert = False
        for m in range(1, l + 1):
            for (bauern, r) in schwarz_zuege:
                if (bauern, r) in gewonnen_weiss[m]:
                    continue
                gut = False
                for q in bauern:
                    for z in nachbarn[q]:
                        if z == r:
                            gut = True                     # Fang
                        elif z not in bauern:
                            neue = (bauern - {q}) | {z}
                            if m == 1 and (neue, r) in gewonnen_schwarz:
                                gut = True
                            if m > 1 and (neue, r) in gewonnen_weiss[m - 1]:
                                gut = True
                        if gut:
                            break
                    if gut:
                        break
                if gut:
                    gewonnen_weiss[m].add((bauern, r))
                    geaendert = True
        for (bauern, r), ziele in schwarz_zuege.items():
            if (bauern, r) not in gewonnen_schwarz:
                if all((bauern, z) in gewonnen_weiss[l] for z in ziele):
                    gewonnen_schwarz.add((bauern, r))
                    geaendert = True
    for bauern in aufstellungen:
        if all((bauern, r) in gewonnen_weiss[l] for r in range(n * n) if r not in bauern):
            return True
    return False


for n in [3, 4]:
    print("Brett", n, "x", n, ", ein Schritt pro Zug:")
    for k in range(1, n + 1):
        print("  ", k, "Bauern:", "Weiß gewinnt" if weiss_gewinnt(n, k, 1) else "Turm gewinnt")
```

Auf beiden Brettern gewinnt Weiß mit `n` Bauern und der Turm mit `n - 1` Bauern. Das passt zum Ergebnis von Teil 1 und 2 (8 gewinnen, 7 nicht) und bestätigt, dass das Muster nicht nur für Strategien gilt, die wir uns ausgedacht haben: Hier sind wirklich alle Spielverläufe berücksichtigt. (Bei einem 5×5-Brett braucht dieselbe Rechnung für 4 Bauern etwa 20 Sekunden und liefert ebenfalls „Turm gewinnt“; 5 Bauern haben wir nicht mehr gerechnet.)

### Teil 3: Wie viele Schritte pro Zug (l) braucht Weiß mit k Bauern?

In den Lösungshinweisen steht für das 8×8-Brett: Darf derselbe Bauer mehrmals ziehen, gewinnt Weiß genau dann, wenn `l ≥ ⌈8 / k⌉` (aufgerundet). Die Strategie ist eine Mauer mit höchstens `l` Feldern Abstand zwischen den Bauern: Dann kann Weiß jede Lücke schließen, die der Turm anpeilt, und mit den übrigen Schritten die Mauer vorschieben. Für `k = 7` reichen `l = 2` Schritte. Für `l = 1` gewinnt, wie in Teil 2 gezeigt, der Turm. Die Tabelle für das 8×8-Brett nach dieser Regel (`X` = Weiß gewinnt):

```python
def aufrunden_geteilt(a, b):
    """a / b, aufgerundet, nur mit ganzen Zahlen."""
    return (a + b - 1) // b


print("l \\ k  " + " ".join(str(k) for k in range(1, 9)))
for l in range(1, 9):
    zeile = ""
    for k in range(1, 9):
        zeile = zeile + ("X" if l >= aufrunden_geteilt(8, k) else ".") + " "
    print(" ", l, "   ", zeile)
```

Ob die Regel stimmt, prüfen wir auf dem 4×4-Brett (Regel: `l ≥ ⌈4 / k⌉`) und dem 3×3-Brett (Regel: `l ≥ ⌈3 / k⌉`) mit der vollständigen Suche.

```python
def tabelle(n, kmax, lmax, dame=False):
    """Tabelle mit weiss_gewinnt für k = 1..kmax (Spalten) und l = 1..lmax (Zeilen). X = Weiß gewinnt, . = Turm gewinnt."""
    ergebnis = {}
    print("Brett", n, "x", n, "(Dame)" if dame else "(Turm)", " l \\ k  " + " ".join(str(k) for k in range(1, kmax + 1)))
    for l in range(1, lmax + 1):
        zeile = ""
        for k in range(1, kmax + 1):
            ergebnis[(k, l)] = weiss_gewinnt(n, k, l, dame)
            zeile = zeile + ("X" if ergebnis[(k, l)] else ".") + " "
        print("                     ", l, "  ", zeile)
    return ergebnis


tab4 = tabelle(4, 4, 4)
tab3 = tabelle(3, 3, 3)
for n, tab in [(4, tab4), (3, tab3)]:
    abweichungen = [(k, l) for (k, l) in tab if tab[(k, l)] != (l >= aufrunden_geteilt(n, k))]
    print("Brett", n, ": Abweichungen von der Regel bei (k, l) =", abweichungen)
```

Auf dem 4×4-Brett stimmt die Regel für alle 16 Fälle. Auf dem 3×3-Brett gewinnt Weiß mit nur einem Bauern schon bei `l = 2`, die Regel verlangt 3: Das Brett ist so klein, dass vermutlich die Ränder dem Bauern zu Hilfe kommen. Für das 8×8-Brett bleibt die Regel daher eine plausible, aber hier nicht bewiesene Vermutung. Sie wird gestützt durch die Strategie in den Lösungshinweisen (Mauer, die sich mit `l ≥ ⌈8 / k⌉` Schritten nie öffnet) und durch das 4×4-Ergebnis.

**Die Variante „jeder Bauer darf pro Zug nur einmal ziehen“** (l verschiedene Bauern) haben wir nicht durchgerechnet, dort wäre die Stellung noch um die Information zu ergänzen, welche Bauern in diesem Zug schon gezogen haben. Die Lösungshinweise sagen für 8×8: Dann braucht Weiß mindestens `k ≥ l` Bauern, mit `l = 2` genügen 4 Bauern (die Mauer wird geschickt so aufgebaut, dass der Turm nicht immer denselben Bauern zum Blocken zwingen kann), mit `k = l = 3` dagegen gewinnt vermutlich der Turm. Das haben wir nicht nachgeprüft.

## Teil 4: Die Dame statt des Turms

Die Dame darf zusätzlich diagonal ziehen (nicht durch Bauern hindurch). Zwei einfache Überlegungen vorab:

- **Sieben Bauern verlieren auch gegen die Dame.** Die Dame kann genau die Strategie des Turms spielen (ihre Züge enthalten die des Turms), und gefangen wird sie nur durch einen Bauern, der orthogonal neben ihr steht, also in ihrer Reihe oder Spalte. Auf einem sicheren Feld ist das unmöglich.
- **Acht Bauern gewinnen**, sagen die Lösungshinweise: Weiß stellt die Bauern in eine Reihe und schiebt sie von links nach rechts vor. Die Dame kann nur durch eine Bruchstelle der Mauer schräg entkommen. Das verhindert Weiß, indem er den Bauern neben dem Knick vorzieht; will die Dame über eine künftige Bruchstelle der Diagonale drohen, zieht Weiß den nächsten Bauern zuerst vor, sodass die Bruchstellen höchstens zwei Spalten auseinanderliegen. Auf dem 8×8-Brett kann die Dame nicht zwei Bruchstellen gleichzeitig bedrohen.

Auf den kleinen Brettern rechnen wir wieder alles durch:

```python
dame4 = tabelle(4, 4, 4, dame=True)
dame3 = tabelle(3, 3, 3, dame=True)
print("Dame wie Turm auf 4x4:", dame4 == tab4, ", auf 3x3:", dame3 == tab3)
```

Auf beiden Brettern ist die Tabelle für die Dame **dieselbe** wie für den Turm: Die zusätzlichen Diagonalzüge helfen ihr hier nicht. Das passt zur Antwort der Lösungshinweise („nein, Schwarz kann nicht entkommen“). Auf größeren Brettern könnte das anders sein, das haben wir nicht untersucht.
