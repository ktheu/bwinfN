## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Graph:** Jedes weiße Feld ist ein Knoten, zwei benachbarte weiße Felder sind durch eine Kante verbunden. Ein Weg, der jeden Knoten genau einmal besucht, heißt **Hamilton-Pfad**. Die Aufgabe ist also das *Hamilton-Pfad-Problem* auf einem Gitter.
- **Zusammenhang prüfen (Junioraufgabe 2):** Von Kassiopeias Startfeld aus alle erreichbaren Felder sammeln (Breitensuche mit einer Liste). Sind danach nicht alle weißen Felder gesammelt, gibt es sicher keinen Weg.
- **Backtracking (Zurücknehmen):** Schritt für Schritt einen Weg aufbauen. Sitzt man in einer Sackgasse, nimmt man den letzten Schritt zurück und probiert eine andere Richtung. Das geht am elegantesten **rekursiv**: eine Funktion ruft sich für den nächsten Schritt selbst auf.
- **Brute Force und seine Grenze:** Backtracking probiert im schlimmsten Fall alle Wege durch. Das Hamilton-Pfad-Problem ist **NP-vollständig**, man kennt kein wesentlich besseres Verfahren, das immer schnell ist. Bei kleinen Welten reicht Backtracking, bei großen braucht man zusätzliche Tests, die aussichtslose Wege früh abbrechen (*Pruning*, siehe „Für Neugierige“).
- **Schachbrettfärbung:** Bei jedem Schritt wechselt die Farbe des Feldes. Daraus folgt eine einfache Bedingung, die schon vor der Suche viele unlösbare Welten erkennt.

**Nützliche Funktionen und Bibliotheken**
- `set()` (Menge): schneller Test `(z, s) in besucht`, ob ein Feld schon besucht wurde. Mit `add` und `remove` füllt und leert man sie.
- `list.append` und `list.pop()`: einen Buchstaben an den Weg hängen und wieder entfernen (das Zurücknehmen beim Backtracking).
- `''.join(liste)`: aus einer Liste von Buchstaben ein Wort machen.
- `str.ljust(n, '#')`: eine Zeile auf Länge `n` auffüllen (für Dateien, in denen am Zeilenende Leerzeichen fehlen).

## Die Idee in Worten

**Die Himmelsrichtungen.** Norden ist oben, also eine Zeile **weniger**; Süden ist eine Zeile mehr, Osten eine Spalte mehr, Westen eine Spalte weniger. Für jede Richtung merken wir uns den Buchstaben und die Änderung von Zeile und Spalte.

**Erst Junioraufgabe 2.** Ist ein weißes Feld von Kassiopeia aus gar nicht erreichbar, gibt es keinen Weg über alle weißen Felder. Dieser Test ist billig und spart die Suche.

**Backtracking.** Wir haben die Funktion `suche(z, s)`: Kassiopeia steht auf Feld `(z, s)`, alle Felder auf ihrem bisherigen Weg sind abgefressen (`besucht`).
1. Hat sie alle weißen Felder besucht, ist der Weg fertig.
2. Sonst probiert sie nacheinander die Richtungen N, O, S, W. Führt eine Richtung auf ein weißes, noch nicht besuchtes Feld, geht sie dorthin (Feld in `besucht`, Buchstabe an den Weg) und ruft `suche` für das neue Feld auf.
3. Ist das Ergebnis ein vollständiger Weg, gibt sie „gefunden“ nach oben weiter. Wenn nicht, nimmt sie den Schritt zurück (Feld wieder aus `besucht`, Buchstabe vom Weg) und probiert die nächste Richtung.
4. Geht keine Richtung, gibt die Funktion „nicht gefunden“ zurück, und der Aufrufer probiert seine nächste Richtung.

Weil die Richtungen in der Reihenfolge N, O, S, W probiert werden, ist der **zuerst gefundene** Weg der alphabetisch erste. So können wir ihn direkt mit der ersten Zeile der Tabelle in den Lösungshinweisen vergleichen.

**Warum ist das korrekt?** Jeder mögliche Weg, der nie ein Feld zweimal betritt, wird irgendwann ausprobiert, denn an jeder Stelle werden alle erlaubten Richtungen durchlaufen. Findet das Verfahren nach dem Durchlaufen aller Möglichkeiten keinen Weg, gibt es auch keinen.

## Schritt 1: Die Welt einlesen

`lies_welt(dateiname)` gibt die Welt als Liste von Zeilen (Texten) zurück. Die erste Zeile mit den Maßen lesen wir, die Geländezeilen füllen wir mit `#` auf die richtige Länge auf (es könnten Leerzeichen am Zeilenende fehlen). Bei einer leeren Datei (`kassiopeia7.txt`) ist die Welt eine leere Liste.

````python
def lies_welt(dateiname):
    with open(dateiname, encoding='utf-8') as datei:
        zeilen = datei.read().splitlines()
    if len(zeilen) == 0:
        return []                                   # leere Datei
    anzahl_zeilen, anzahl_spalten = zeilen[0].split()
    anzahl_zeilen = int(anzahl_zeilen)
    anzahl_spalten = int(anzahl_spalten)
    welt = []
    for z in range(anzahl_zeilen):
        zeile = zeilen[1 + z] if 1 + z < len(zeilen) else ''
        welt.append(zeile.ljust(anzahl_spalten, '#'))
    return welt


welt = lies_welt('beispieldaten/kassiopeia0.txt')
for zeile in welt:
    print('|' + zeile + '|')
print(len(lies_welt('beispieldaten/kassiopeia7.txt')), 'Zeilen in kassiopeia7.txt')
````

## Schritt 2: Weiße Felder zählen, Start finden und nach Nachbarn fragen

- `ist_weiss(welt, z, s)`: Das Feld liegt in der Welt und ist kein `#`. (Außerhalb der Welt darf man nie hin, auch wenn der Rand einer Datei einmal kein `#` enthält.)
- `finde_start(welt)`: Position von `K`, oder `None`, wenn es kein `K` gibt (leere Welt).
- `zaehle_weisse(welt)`: Anzahl aller weißen Felder (mit `K`).
- `RICHTUNGEN`: Liste von Tripeln `(Buchstabe, Änderung der Zeile, Änderung der Spalte)`, **in der Reihenfolge N, O, S, W**.

````python
RICHTUNGEN = [('N', -1, 0), ('O', 0, 1), ('S', 1, 0), ('W', 0, -1)]


def ist_weiss(welt, z, s):
    return 0 <= z < len(welt) and 0 <= s < len(welt[z]) and welt[z][s] != '#'


def finde_start(welt):
    for z in range(len(welt)):
        for s in range(len(welt[z])):
            if welt[z][s] == 'K':
                return (z, s)
    return None


def zaehle_weisse(welt):
    anzahl = 0
    for z in range(len(welt)):
        for s in range(len(welt[z])):
            if welt[z][s] != '#':
                anzahl = anzahl + 1
    return anzahl


print('Start:', finde_start(welt), ' weiße Felder:', zaehle_weisse(welt))
````

## Schritt 3: Junioraufgabe 2, sind alle weißen Felder erreichbar?

Breitensuche vom Start: In `offen` stehen Felder, deren Nachbarn wir noch ansehen müssen, in `erreicht` alle bisher gefundenen Felder. Aus `offen` nehmen wir ein Feld, und jeden weißen Nachbarn, den wir noch nicht kennen, merken wir uns in beiden. Wenn `offen` leer ist, vergleichen wir die Zahl der erreichten Felder mit der Zahl aller weißen Felder.

````python
def alles_erreichbar(welt):
    start = finde_start(welt)
    if start is None:
        return False
    erreicht = {start}
    offen = [start]
    while len(offen) > 0:
        z, s = offen.pop()
        for buchstabe, dz, ds in RICHTUNGEN:
            nz = z + dz
            ns = s + ds
            if ist_weiss(welt, nz, ns) and (nz, ns) not in erreicht:
                erreicht.add((nz, ns))
                offen.append((nz, ns))
    return len(erreicht) == zaehle_weisse(welt)


for nr in range(8):
    w = lies_welt('beispieldaten/kassiopeia%d.txt' % nr)
    print('Beispiel', nr, ':', 'Ja' if alles_erreichbar(w) else 'Nein')
````

Bei Beispiel 1 sind die weißen Felder nicht zusammenhängend (links und rechts liegen getrennte Gebiete), dort gibt es also sicher keinen Weg. Bei Beispiel 7 gibt es keine Welt, also auch kein `K`; wir sagen „Nein“.

**Wichtig:** „Alle erreichbar“ heißt nur, dass es *irgendeinen* Weg zu jedem Feld gibt. Einen Weg, der **alle Felder genau einmal** besucht, muss das noch lange nicht geben. Das ist der eigentliche Teil der Aufgabe.

## Schritt 4: Backtracking

`suche(welt, z, s, besucht, weg, anzahl)` arbeitet wie in der Idee beschrieben. `besucht` ist eine Menge von Feldern, `weg` eine Liste von Buchstaben, `anzahl` die Zahl der weißen Felder (damit wir nicht jedes Mal neu zählen). Die Funktion gibt `True` zurück, wenn `weg` zu einem vollständigen Weg geworden ist. Bleibt sie bei `False`, ist `weg` wieder so wie am Anfang des Aufrufs.

Dazu zählt die Variable `aufrufe`, wie oft `suche` aufgerufen wurde. Sie ist ein Maß für den Aufwand (Laufzeiten sind nicht reproduzierbar, Aufrufe schon). `global aufrufe` sagt Python, dass wir die Variable außerhalb der Funktion meinen.

````python
aufrufe = 0


def suche(welt, z, s, besucht, weg, anzahl):
    global aufrufe
    aufrufe = aufrufe + 1
    if len(besucht) == anzahl:                       # alle weißen Felder besucht
        return True
    for buchstabe, dz, ds in RICHTUNGEN:
        nz = z + dz
        ns = s + ds
        if ist_weiss(welt, nz, ns) and (nz, ns) not in besucht:
            besucht.add((nz, ns))                    # Schritt gehen
            weg.append(buchstabe)
            if suche(welt, nz, ns, besucht, weg, anzahl):
                return True
            weg.pop()                                # Schritt zurücknehmen
            besucht.remove((nz, ns))
    return False


def finde_weg(welt):
    """Ein Weg als Text aus N, O, S, W; None, wenn es keinen gibt."""
    global aufrufe
    aufrufe = 0
    start = finde_start(welt)
    if start is None:
        return None
    weg = []
    if suche(welt, start[0], start[1], {start}, weg, zaehle_weisse(welt)):
        return ''.join(weg)
    return None


print(finde_weg(lies_welt('beispieldaten/kassiopeia0.txt')))
print('Aufrufe von suche:', aufrufe)
````

Das ist genau der Weg aus der Aufgabenstellung. Das Programm hat dazu `suche` 199-mal aufgerufen.

## Schritt 5: Alle Beispiele

Wir zeigen für jedes Beispiel erst das Ergebnis der Junioraufgabe (alle erreichbar?), dann den Weg oder „Kein Pfad“ und die Zahl der Aufrufe von `suche`. Wenn die weißen Felder nicht zusammenhängend sind, sparen wir uns die Suche sogar (Zusammenhangstest zuerst).

````python
def bearbeite(dateiname):
    welt = lies_welt(dateiname)
    if not alles_erreichbar(welt):
        return 'Nein', 'Kein Pfad', 0
    weg = finde_weg(welt)
    return 'Ja', weg if weg is not None else 'Kein Pfad', aufrufe


for nr in range(8):
    erreichbar, ergebnis, zahl = bearbeite('beispieldaten/kassiopeia%d.txt' % nr)
    print('Beispiel %d: erreichbar: %-4s  %-26s  Aufrufe: %d' % (nr, erreichbar, ergebnis, zahl))
````

Das stimmt mit der Tabelle der Lösungshinweise überein: Nur bei den Beispielen 0, 3 und 5 gibt es einen Weg. Bei Beispiel 5 ist die Welt ein einziger Gang, den man von seinem Ende her durchläuft. Bei 2, 4 und 6 sind die weißen Felder zusammenhängend, aber es gibt trotzdem keinen Weg, bei 4 zum Beispiel, weil Kassiopeia mitten in einem Gang startet und eine Hälfte liegen lassen müsste. Auch dort musste die Suche alles ausprobieren, bevor sie „Kein Pfad“ sagen konnte, das waren aber nur wenige Aufrufe.

## Schritt 6: Alle Wege zählen

Die Lösungshinweise nennen bei Beispiel 0 und 3 je **vier** Wege. Wir ändern die Suche so, dass sie nach einem gefundenen Weg nicht aufhört, sondern weitersucht und die Wege zählt. Alles andere bleibt gleich: Statt `True` zurückzugeben, erhöhen wir einen Zähler, und nach jedem Zweig wird der Schritt zurückgenommen.

````python
def suche_alle(welt, z, s, besucht, weg, anzahl, gefunden):
    if len(besucht) == anzahl:
        gefunden.append(''.join(weg))
        return
    for buchstabe, dz, ds in RICHTUNGEN:
        nz = z + dz
        ns = s + ds
        if ist_weiss(welt, nz, ns) and (nz, ns) not in besucht:
            besucht.add((nz, ns))
            weg.append(buchstabe)
            suche_alle(welt, nz, ns, besucht, weg, anzahl, gefunden)
            weg.pop()
            besucht.remove((nz, ns))


def alle_wege(welt):
    start = finde_start(welt)
    gefunden = []
    if start is not None:
        suche_alle(welt, start[0], start[1], {start}, [], zaehle_weisse(welt), gefunden)
    return gefunden


for nr in (0, 3, 5):
    wege = alle_wege(lies_welt('beispieldaten/kassiopeia%d.txt' % nr))
    print('Beispiel', nr, ':', len(wege), 'Weg(e)')
    for w in wege:
        print('   ', w)
````

Weil die Richtungen in der Reihenfolge N, O, S, W probiert werden, stehen die Wege schon in alphabetischer Reihenfolge, genau wie in den Lösungshinweisen.

## Für Neugierige: Wenn die Welt größer wird

Die Aufrufe oben waren immer klein. Das Hamilton-Pfad-Problem ist aber **NP-vollständig**; der Aufwand kann mit jedem weiteren Feld auf ein Vielfaches wachsen (jedes Feld hat bis zu drei neue Richtungen). Ein Beispiel ist ein offenes Quadrat von 5 mal 5 Feldern, bei dem Kassiopeia ungünstig startet (`eigene_testdaten/gitter_5x5_unguenstig.txt`):

````python
welt_u = lies_welt('eigene_testdaten/gitter_5x5_unguenstig.txt')
print(finde_weg(welt_u), ' Aufrufe:', aufrufe)
print(finde_weg(lies_welt('eigene_testdaten/gitter_5x5_ecke.txt')), ' Aufrufe:', aufrufe)
print(finde_weg(lies_welt('eigene_testdaten/gitter_6x6_mitte.txt')), ' Aufrufe:', aufrufe)
````

Dort, wo ein Weg existiert (Start in der Ecke, 6 mal 6 mit Start in der Mitte), findet die Suche ihn nach 25 bzw. 36 Aufrufen fast ohne Umwege. Beim ungünstigen Start ist es umgekehrt: Es gibt **keinen** Weg, aber die Suche muss das erst durch Ausprobieren **aller** möglichen Wege beweisen. Sie braucht dafür 124.190 Aufrufe, bei nur 25 Feldern. Mit jeder weiteren Zeile würde es dramatisch mehr. Das ist die Grenze des einfachen Backtrackings. Zwei einfache Tests brechen aussichtslose Wege früh ab (*Pruning*, so nennen es auch die Lösungshinweise):

**1. Zusammenhang.** Nach jedem Schritt müssen alle noch nicht besuchten weißen Felder vom aktuellen Feld aus erreichbar sein, über Felder, die noch nicht besucht sind. Sonst bleibt ein Teil der Welt unerreichbar. Das ist die Breitensuche aus Schritt 3, nur über die unbesuchten Felder.

**2. Farben (Parität).** Wir färben die Welt wie ein Schachbrett: Ein Feld `(z, s)` ist „hell“, wenn `z + s` gerade ist, sonst „dunkel“. Bei **jedem Schritt wechselt die Farbe**. Der Weg von Kassiopeia läuft also hell, dunkel, hell, dunkel, … Von den noch unbesuchten Feldern müssen deshalb abwechselnd Felder der *anderen* Farbe als das aktuelle Feld und Felder der *gleichen* Farbe drankommen, beginnend mit der anderen. Zählt man `anders` und `gleich`, muss also `anders == gleich` oder `anders == gleich + 1` gelten. Sonst gibt es keinen Weg mehr. (Die Lösungshinweise formulieren dasselbe etwas anders, mit dem Endpunkt des Weges.)

````python
def rest_ok(welt, z, s, besucht, anzahl, mit_farben):
    """Kann es von (z, s) aus noch einen Weg über alle unbesuchten Felder geben? (nur notwendige Bedingungen)"""
    # Zusammenhang: Breitensuche über unbesuchte Felder
    erreicht = {(z, s)}
    offen = [(z, s)]
    while len(offen) > 0:
        a, b = offen.pop()
        for buchstabe, dz, ds in RICHTUNGEN:
            c = (a + dz, b + ds)
            if ist_weiss(welt, c[0], c[1]) and c not in besucht and c not in erreicht:
                erreicht.add(c)
                offen.append(c)
    if len(erreicht) - 1 != anzahl - len(besucht):
        return False
    if mit_farben:
        anders = 0                                   # unbesuchte Felder mit anderer Farbe als (z, s)
        gleich = 0                                   # unbesuchte Felder mit gleicher Farbe
        for (a, b) in erreicht:
            if (a, b) != (z, s):
                if (a + b) % 2 == (z + s) % 2:
                    gleich = gleich + 1
                else:
                    anders = anders + 1
        if anders != gleich and anders != gleich + 1:
            return False
    return True


def suche_klug(welt, z, s, besucht, weg, anzahl, mit_farben):
    global aufrufe
    aufrufe = aufrufe + 1
    if len(besucht) == anzahl:
        return True
    if not rest_ok(welt, z, s, besucht, anzahl, mit_farben):
        return False                                 # aussichtslos, gleich zurück
    for buchstabe, dz, ds in RICHTUNGEN:
        nz = z + dz
        ns = s + ds
        if ist_weiss(welt, nz, ns) and (nz, ns) not in besucht:
            besucht.add((nz, ns))
            weg.append(buchstabe)
            if suche_klug(welt, nz, ns, besucht, weg, anzahl, mit_farben):
                return True
            weg.pop()
            besucht.remove((nz, ns))
    return False


def finde_weg_klug(welt, mit_farben):
    global aufrufe
    aufrufe = 0
    start = finde_start(welt)
    if start is None:
        return None
    weg = []
    if suche_klug(welt, start[0], start[1], {start}, weg, zaehle_weisse(welt), mit_farben):
        return ''.join(weg)
    return None


dateien = ['beispieldaten/kassiopeia%d.txt' % nr for nr in range(7)] + ['eigene_testdaten/gitter_5x5_unguenstig.txt']
print('Datei                                           einfach   Zusammenhang   Zusammenhang + Farben')
for name in dateien:
    w = lies_welt(name)
    ergebnisse = []
    zahlen = []
    ergebnisse.append(finde_weg(w))
    zahlen.append(aufrufe)
    for farben in (False, True):
        ergebnisse.append(finde_weg_klug(w, farben))
        zahlen.append(aufrufe)
    assert ergebnisse[0] == ergebnisse[1] == ergebnisse[2]           # alle drei Verfahren finden denselben Weg
    print('%-46s %8d %14d %24d' % (name.split('/')[-1], zahlen[0], zahlen[1], zahlen[2]))
````

Die Tabelle zeigt, wie unterschiedlich die Tests wirken:
- Beim ungünstigen 5 mal 5 Quadrat schrumpft die Zahl der Aufrufe mit dem **Zusammenhangstest** von 124.190 auf 9.181 und mit dem **Farbentest** auf 1 (der Test erkennt schon am Start, dass es unmöglich ist: 25 Felder, aber Kassiopeia startet auf der Farbe mit nur 12 Feldern).
- Beispiel 1 (nicht zusammenhängend) wird vom Zusammenhangstest sofort erledigt (1 Aufruf statt 1.109). Beispiel 2 ist das Beispiel, an dem die Lösungshinweise den Farbentest zeigen: 77 Aufrufe einfach, 45 mit dem Zusammenhangstest und nur 1 mit dem Farbentest.
- Bei Welten, in denen es einen Weg gibt, helfen die Tests weniger, bei Beispiel 0 immerhin von 199 auf 29 Aufrufe, bei 3 und 5 gar nicht. Dabei zählen wir nur die Aufrufe von `suche`; die Arbeit in `rest_ok` (eine Breitensuche pro Aufruf) kommt noch dazu.

Weitere Ideen aus den Lösungshinweisen sind der Test auf **zwingende Endpunkte** (ein weißes Feld mit nur einem unbesuchten Nachbarn muss das Ende des Weges sein, zwei davon gehen nicht), **Gelenkpunkte** und das Ausprobieren von Nachbarn mit **wenigen Nachbarn zuerst**. Alle Beispiele der Aufgabe lassen sich schon mit dem einfachen Backtracking aus Schritt 4 in kurzer Zeit lösen; die vom BWINF genannten größeren eigenen Beispiele („an der Grenze des mit Pruning Machbaren“) nicht. Ganz anders arbeitet die **dynamische Programmierung** über Teilmengen von Feldern (`ist_moeglich(Menge, Endfeld)`): Sie braucht etwa n² · 2ⁿ Schritte, aber auch sehr viel Speicher und ist bei kleinen Welten schon zu langsam.

## Zum Schluss: Geht das mit MiniZinc?

Ja, und bei den Beispielen sogar ohne Mühe. Wir beschreiben nur, **was** ein Weg ist, nicht, wie man ihn findet:

- Die weißen Felder sind von 1 bis `n` durchnummeriert, `nachbar[a, b]` sagt, ob zwei Felder nebeneinander liegen (eine Tabelle der Größe `n` mal `n`, die wir in Python erzeugen).
- `weg[i]` ist das Feld nach `i - 1` Schritten. Die Zeile `alldifferent(weg)` heißt: alle Felder kommen vor, **jedes genau einmal** (es gibt `n` Plätze und `n` Felder).
- `weg[1] = start`, und aufeinanderfolgende Felder müssen Nachbarn sein.

Das ist genau die Definition des Hamilton-Pfades. Findet der Solver keine Belegung, lautet die Antwort „Kein Pfad“ (`=====UNSATISFIABLE=====`).

```minizinc
include "alldifferent.mzn";

int: n;                                  % Zahl der weißen Felder
int: start;                              % Nummer des Startfelds
array[1..n, 1..n] of bool: nachbar;      % nachbar[a, b]: Feld a und Feld b liegen nebeneinander

array[1..n] of var 1..n: weg;            % weg[i]: Feld nach i-1 Schritten

constraint alldifferent(weg);            % jedes Feld genau einmal
constraint weg[1] = start;
constraint forall(i in 1..n-1)(nachbar[weg[i], weg[i+1]]);

solve satisfy;
output ["weg = ", show(weg), "\n"];
```

Die Datendatei erzeugt eine Python-Zelle aus dem Gelände, hier für Beispiel 5 (der einzige Gang aus 13 Feldern; die Tabelle `nachbar` ist ein langer Text, wir zeigen nur den Anfang):
