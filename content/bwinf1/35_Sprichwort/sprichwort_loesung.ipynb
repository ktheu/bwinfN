Zur Weihnachtszeit äußert Christian immer sehr kostspielige Wünsche. Die Standardantwort seiner Mutter ist dann die Redensart: *Das bekommst du, wenn Weihnachten und Ostern auf einen Tag fallen.*

Sein Freund Igor erzählte Christian aber vor Kurzem, dass die orthodoxe Kirche ihr Weihnachtsfest erst am 7. Januar feiert. Der Grund ist, dass diese Kirche einem anderen Kalender folgt. ## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Recherche und Umrechnen:** Die Aufgabe besteht aus zwei Bausteinen, die du getrennt baust und testest: (1) ein Datum vom julianischen in den gregorianischen Kalender umrechnen, (2) das Osterdatum eines Jahres ausrechnen. Danach ist die Suche nur noch eine Schleife.
- **Datum als Tag im Jahr:** Statt mit Monat und Tag zu rechnen, zählt man die Tage seit dem 1. Januar. Dann ist „Datum plus 13 Tage“ eine einfache Addition, und zwei Daten sind gleich, wenn die Zahlen gleich sind. Monat und Tag baut man erst zur Ausgabe wieder zusammen.
- **Lineare Suche:** Alle Jahre der Reihe nach prüfen, beim ersten Treffer aufhören. Wie weit man höchstens suchen muss, schätzt man vorher ab.

**Nützliche Funktionen und Operatoren**
- `//` (ganzzahlige Division) und `%` (Rest): Die Osterformel besteht fast nur daraus.
- Eine Liste mit den Monatslängen, um zwischen „Tag im Jahr“ und „Monat und Tag“ umzurechnen.
- Eine `for`-Schleife über die Jahre.

## Warum Christians Hoffnung berechtigt scheint

Ein Sonnenjahr dauert etwa 365,2422 Tage. Der **julianische Kalender** (Julius Caesar, 45 v. Chr.) rechnet mit 365,25 Tagen: Jedes vierte Jahr ist ein Schaltjahr. Das ist etwas zu lang, die Jahreszeiten wandern im Lauf der Jahrhunderte gegenüber dem Kalender. Der **gregorianische Kalender** (Papst Gregor XIII., 1582) lässt deshalb drei von vier Jahrhundertjahren als Schaltjahr ausfallen (1700, 1800 und 1900 sind keine Schaltjahre, 2000 schon) und kommt so auf 365,2425 Tage. Die katholische und die protestantische Kirche feiern nach dem gregorianischen Kalender, die orthodoxen Kirchen feiern ihre Feste weiterhin nach dem julianischen.

Der gregorianische Kalender hat also **in 400 Jahren drei Schalttage weniger**, und die beiden Kalender laufen immer weiter auseinander. Heute beträgt der Unterschied 13 Tage: Der julianische 25. Dezember 2016 ist der gregorianische 7. Januar 2017. Der Abstand wächst um etwa einen Tag pro 133 Jahre. Das orthodoxe Weihnachten rückt deshalb im gregorianischen Kalender Jahr für Jahr langsam weiter in Richtung Frühling, und irgendwann muss es auf den Tag eines westlichen Ostersonntags treffen (der immer zwischen dem 22. März und dem 25. April liegt). Ebenso rückt das orthodoxe Ostern (julianisch gerechnet, dann umgerechnet) immer weiter in Richtung Jahresende und trifft irgendwann den gregorianischen 25. Dezember. Die Hoffnung ist also berechtigt, die Frage ist nur, **wie lange** man warten muss.

**Grobe Abschätzung:** Vom 7. Januar bis zum frühesten Ostertermin (22. März) fehlen etwa 75 Tage. Pro 400 Jahre kommen 3 Tage dazu, also dauert es rund 75 · 400 / 3 = 10000 Jahre. Für die zweite Frage fehlen etwa 8 Monate, also 240 Tage, das sind rund 32000 Jahre. Wenn der Abstand auf ein ganzes Jahr angewachsen ist (366 Tage, das sind 366 · 400 / 3 = 48800 Jahre), muss es in beiden Fällen einen Treffer gegeben haben. Wir suchen deshalb bis zum Jahr 50000.

## Die Idee

Wir brauchen drei Bausteine:

1. **Abstand der Kalender** in Tagen für ein Jahr (aus den Lösungshinweisen, gilt ab dem Jahr 1583 für Daten ab März): `abstand = jahr // 100 - jahr // 400 - 2`.
2. **Osterdatum** nach der Formel von Gauß, ergänzt von Kinkelin und Lichtenberg (Tabelle unten): Sie liefert, auf den wievielten März Ostersonntag fällt (der 32. März ist der 1. April). Für das **westliche** Ostern ist das Ergebnis ein gregorianisches Datum, für das **östliche** Ostern ein julianisches.
3. **Datum als Tag im Jahr** (1. Januar = Tag 1), um Daten vergleichen und umrechnen zu können.

Dann gilt:

- **Teil 1:** Der julianische 25.12. des Jahres `j - 1` liegt gregorianisch `abstand` Tage später, also `abstand - 6` Tage nach dem 31.12., das ist **Tag `abstand - 6` im Jahr `j`**. Wir suchen das erste Jahr `j`, in dem das westliche Ostern auf genau diesen Tag fällt.
- **Teil 2:** Das östliche Ostern im Jahr `j` ist julianisch der `os`-te März. Gregorianisch ist es `abstand` Tage später. Wir suchen das erste Jahr, in dem das der 25.12. (gregorianisch) ist.

## Schritt 1: Abstand der Kalender

Kontrollen: Für 2016 muss 13 herauskommen. Bei der Einführung des gregorianischen Kalenders 1582 wurden 10 Tage übersprungen (auf den 4. Oktober folgte der 15. Oktober), das passt zum Wert für 1583.

````python
def abstand(jahr):
    """Wie viele Tage der gregorianische Kalender dem julianischen voraus ist (für Daten ab März)."""
    return jahr // 100 - jahr // 400 - 2


for j in [1583, 1700, 1800, 1900, 2016, 2100, 11918]:
    print(j, abstand(j))
````

## Schritt 2: Das Osterdatum

Die Formel aus den Lösungshinweisen (alle Divisionen ganzzahlig). Die Spalten „westlich“ und „östlich“ unterscheiden sich nur in `m` (säkulare Mondschaltung) und `s` (säkulare Sonnenschaltung):

| Wert | westlich | östlich |
|---|---|---|
| Säkularzahl `k` | `jahr // 100` | |
| säkulare Mondschaltung `m` | `15 + (3k + 3) // 4 - (8k + 13) // 25` | `15` |
| säkulare Sonnenschaltung `s` | `2 - (3k + 3) // 4` | `0` |
| Mondparameter `a` | `jahr % 19` | |
| Keim für den ersten Vollmond im Frühling `d` | `(19a + m) % 30` | |
| Korrekturgröße `r` | `(d + a // 11) // 29` | |
| Ostergrenze `og` | `21 + d - r` | |
| erster Sonntag im März `sz` | `7 - (jahr + jahr // 4 + s) % 7` | |
| Osterentfernung in Tagen `oe` | `7 - (og - sz) % 7` | |
| Ostersonntag `os` (März-Tag) | `og + oe` | |

Wir schreiben eine Funktion, die den März-Tag liefert. Ob westlich oder östlich gerechnet wird, steuert ein Parameter `westlich`.

````python
def ostern_maerztag(jahr, westlich):
    """Ostersonntag als Tag im März (Werte über 31 liegen im April).
    westlich: gregorianisches Datum, sonst (östlich): julianisches Datum."""
    k = jahr // 100
    if westlich:
        m = 15 + (3 * k + 3) // 4 - (8 * k + 13) // 25
        s = 2 - (3 * k + 3) // 4
    else:
        m = 15
        s = 0
    a = jahr % 19
    d = (19 * a + m) % 30
    r = (d + a // 11) // 29
    og = 21 + d - r
    sz = 7 - (jahr + jahr // 4 + s) % 7
    oe = 7 - (og - sz) % 7
    return og + oe


# Kontrolle: 2016 war westliches Ostern am 27.3., orthodoxes am 1.5. (gregorianisch) = 18.4. (julianisch) = 49. März
print(ostern_maerztag(2016, True), ostern_maerztag(2016, False))
````

## Schritt 3: Datum als Tag im Jahr

`tag_im_jahr(jahr, monat, tag)` zählt die Tage seit dem 1. Januar (gregorianisch). `datum_aus_tag(jahr, tag)` macht es rückwärts und läuft bei zu großem Tag in das Folgejahr weiter. Zum Prüfen geben wir das westliche und das östliche Ostern 2016 und 2024 als Datum aus (östlich: julianischer März-Tag plus Abstand, dann umgerechnet).

````python
MONATSLAENGEN = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]


def ist_schaltjahr(jahr):
    """Gregorianisches Schaltjahr?"""
    return jahr % 4 == 0 and (jahr % 100 != 0 or jahr % 400 == 0)


def tag_im_jahr(jahr, monat, tag):
    """Tag im Jahr (1. Januar = 1), gregorianisch."""
    summe = tag
    for m in range(1, monat):
        summe = summe + MONATSLAENGEN[m - 1]
        if m == 2 and ist_schaltjahr(jahr):
            summe = summe + 1
    return summe


def datum_aus_tag(jahr, tag):
    """Tag im Jahr -> (jahr, monat, tag), gregorianisch; zu große Tage laufen ins nächste Jahr."""
    while True:
        laenge = 366 if ist_schaltjahr(jahr) else 365
        if tag <= laenge:
            break
        tag = tag - laenge
        jahr = jahr + 1
    monat = 1
    while True:
        ml = MONATSLAENGEN[monat - 1]
        if monat == 2 and ist_schaltjahr(jahr):
            ml = 29
        if tag <= ml:
            return jahr, monat, tag
        tag = tag - ml
        monat = monat + 1


def ostern_westlich(jahr):
    """Westliches Ostern als gregorianischer Tag im Jahr."""
    return tag_im_jahr(jahr, 3, 1) - 1 + ostern_maerztag(jahr, True)


def ostern_oestlich(jahr):
    """Östliches Ostern, umgerechnet in den gregorianischen Kalender, als Tag im Jahr (kann über das Jahresende hinausgehen)."""
    return tag_im_jahr(jahr, 3, 1) - 1 + ostern_maerztag(jahr, False) + abstand(jahr)


def als_text(jahr, monat, tag):
    return '%02d.%02d.%d' % (tag, monat, jahr)


for j in [2016, 2024]:
    print(j, 'westlich:', als_text(*datum_aus_tag(j, ostern_westlich(j))), ' östlich:', als_text(*datum_aus_tag(j, ostern_oestlich(j))))
````

Das stimmt mit den bekannten Terminen überein (2016: 27. März und 1. Mai, 2024: 31. März und 5. Mai).

## Schritt 4: Teil 1, orthodoxes Weihnachten gleich westliches Ostern

Der julianische 25.12. des Jahres `j - 1` liegt gregorianisch bei Tag `abstand(j - 1) - 6` des Jahres `j`. Wir vergleichen diese Zahl mit dem westlichen Ostertag im Jahr `j`. Wir beginnen im Jahr 1584, kurz nach der Einführung des gregorianischen Kalenders, und sammeln alle Treffer bis 50000 (der erste ist die Antwort).

````python
def teil1(ab_jahr, bis_jahr):
    """Alle Jahre j, in denen westliches Ostern auf den julianischen 25.12. des Vorjahres fällt."""
    treffer = []
    for j in range(ab_jahr, bis_jahr + 1):
        weihnachten_orthodox = abstand(j - 1) - 6
        if weihnachten_orthodox == ostern_westlich(j):
            treffer.append(j)
    return treffer


erste = teil1(1584, 50000)[0]
print('Erstes Jahr:', erste)
print('Westliches Ostern:', als_text(*datum_aus_tag(erste, ostern_westlich(erste))))
print('Abstand im Jahr', erste - 1, ':', abstand(erste - 1), 'Tage')
````

**Antwort Teil 1:** Zum ersten Mal fallen orthodoxes Weihnachten (julianischer 25.12.) und westliches Ostern am **23. März 11919** zusammen (ein Sonntag, denn es ist ein Ostersonntag). Der julianische 25.12.11918 ist dann gregorianisch der 23.3.11919.

Probe mit der Abschätzung: Der Abstand beträgt dann 88 Tage, also 75 mehr als heute (13), passend zu den geschätzten 10000 Jahren.

## Schritt 5: Teil 2, westliches Weihnachten gleich orthodoxes Ostern

Hier ist das Jahr für beide Feste dasselbe. Wir rechnen das östliche Ostern in den gregorianischen Kalender um (`ostern_oestlich`) und vergleichen mit dem gregorianischen 25.12. des gleichen Jahres.

````python
def teil2(ab_jahr, bis_jahr):
    """Alle Jahre, in denen das östliche Ostern auf den gregorianischen 25.12. desselben Jahres fällt."""
    treffer = []
    for j in range(ab_jahr, bis_jahr + 1):
        if ostern_oestlich(j) == tag_im_jahr(j, 12, 25):
            treffer.append(j)
    return treffer


erste2 = teil2(1584, 50000)[0]
print('Erstes Jahr:', erste2)
print('Gregorianischer 25.12.:', als_text(erste2, 12, 25))
maerztag = ostern_maerztag(erste2, False)
if maerztag <= 31:
    julianisch = als_text(erste2, 3, maerztag)
else:
    julianisch = als_text(erste2, 4, maerztag - 31)
print('Östliches Ostern julianisch:', julianisch, ', Abstand', abstand(erste2), 'Tage')
````

**Antwort Teil 2:** Zum ersten Mal fallen westliches Weihnachten (1. Weihnachtstag) und orthodoxes Ostern am **25. Dezember 32839** zusammen. Das orthodoxe Ostern ist dann julianisch der 25. April, und der Abstand der Kalender beträgt 244 Tage, das sind 231 mehr als heute, passend zur Schätzung von 240 Tagen. Christian muss auf diese Geschenke also sehr lange warten.

## Alle Treffer bis zum Jahr 50000

Mit denselben Funktionen bekommen wir auch alle weiteren Treffer:

````python
alle1 = teil1(1584, 50000)
alle2 = teil2(1584, 50000)
print('Teil 1:', len(alle1), 'Treffer, die ersten:', alle1[:6], 'letzter:', alle1[-1])
print('Teil 2:', len(alle2), 'Treffer, die ersten:', alle2[:6], 'letzter:', alle2[-1])
````

Die Lösungshinweise listen die Treffer von Teil 1 vollständig auf (135 Jahre von 11919 bis 16500), unsere Liste ist identisch. Bei Teil 2 stimmt die Liste ebenfalls vollständig (139 Jahre von 32839 bis 37312).

**Zur Genauigkeit:** Die Osterformel gilt für die heutigen Regeln der Kirchen. Ob diese in tausenden Jahren noch gelten, weiß niemand (schon heute gibt es Gespräche über ein gemeinsames Osterdatum). Christians Mutter muss sich um den Wunschzettel also keine Sorgen machen.

## Zum Schluss: Mit MiniZinc?

Wir haben beide Teile mit MiniZinc ausprobiert, und es klappt gut und schnell: Das Modell beschreibt direkt die **Bedingung**, und der Solver sucht das **kleinste** Jahr (`solve minimize jahr`). Mit dem Standardsolver **Gecode** dauert jede Rechnung unter einer halben Sekunde. **Chuffed** schafft nur das zweite Modell (beim ersten bricht er mit der Meldung „Cannot handle non-sign-fixed vars“ ab, weil mit `jahr - 1` ein Wert mit unbekanntem Vorzeichen in eine Division geht).

Neu gegenüber den Arbeitsblättern sind nur `function` (eine eigene Funktion im Modell, die mit Variablen rechnet), `div` und `mod` (ganzzahlige Division und Rest, wie `//` und `%` in Python) und `let ... in` (lokale Variablen in einer Funktion). Die Osterformel übernehmen wir fast Zeile für Zeile.

**Das Modell für Teil 1** (`sprichwort1.mzn`). Die Variable `jahr` darf zwischen 1584 und 50000 liegen. Die Funktion `ostern` ist die Tabelle von oben. Die Bedingung sagt: Der gregorianische Tag des julianischen 25.12. (`abstand(jahr - 1) - 6`) ist der westliche Ostertag im Jahr (Tag im Jahr = 59 + Schalttag + März-Tag):

```minizinc
var 1584..50000: jahr;

% Tag im März, auf den Ostern fällt (über 31 läuft in den April); west = 1: gregorianisch, 0: julianisch
function var int: ostern(var int: j, int: west) =
  let {
    var int: k = j div 100;
    var int: m = if west = 1 then 15 + (3*k + 3) div 4 - (8*k + 13) div 25 else 15 endif;
    var int: s = if west = 1 then 2 - (3*k + 3) div 4 else 0 endif;
    var int: a = j mod 19;
    var int: d = (19*a + m) mod 30;
    var int: r = (d + a div 11) div 29;
    var int: og = 21 + d - r;
    var int: sz = 7 - (j + j div 4 + s) mod 7;
    var int: oe = 7 - (og - sz) mod 7;
  } in og + oe;

function var int: abstand(var int: j) = j div 100 - j div 400 - 2;
function var int: schalt(var int: j) = bool2int(j mod 4 = 0 /\ (j mod 100 != 0 \/ j mod 400 = 0));

% Tag im Jahr: 1. März = 60 (+1 in Schaltjahren), also März-Tag + 59 (+ schalt)
constraint abstand(jahr - 1) - 6 = 59 + schalt(jahr) + ostern(jahr, 1);
solve minimize jahr;
output ["\(jahr)\n"];
```

**Ausgabe:**

```
11919
----------
==========
```

**Teil 2** (`sprichwort2.mzn`): Es ändert sich nur die Bedingung, der Rest des Modells bleibt gleich. Das östliche Ostern (julianischer März-Tag plus Abstand der Kalender) soll der 25. Dezember sein, das ist Tag 359 (+1 im Schaltjahr):

constraint 59 + schalt(jahr) + ostern(jahr, 0) + abstand(jahr) = 359 + schalt(jahr);
