### Ameisenfutter – Lösung Schritt für Schritt

*34. Bundeswettbewerb Informatik (2015), 1. Runde, Aufgabe 2*

Auf der Futtersuche krabbeln Ameisen von ihrem Nest aus in eine zufällige, sich ständig ändernde Richtung. Finden sie auf ihrem Weg eine Futterquelle (z. B. ein Zuckerhäufchen), nehmen sie eine Portion des Futters auf und bringen sie schnurstracks zum Nest. Dabei sondern sie entlang dieses Weges einen Duftstoff (ein Pheromon) ab.

Die Pheromonspuren helfen den anderen Ameisen, Futterquellen zu finden: Falls die Ameisen auf ihrer zufälligen Suche den Duftstoff bemerken, folgen sie der Duftspur. Die abgesonderten Pheromone verdunsten allerdings nach einer gewissen Zeit. Wenn die Futterquelle erschöpft ist, verschwindet also nach einiger Zeit der Duftstoff und die Ameisen suchen neue Wege.

Simuliere die Ameisen-Futtersuche im Computer. Nimm dazu an, dass sich 100 Ameisen im Nest in der Mitte eines in 500 × 500 Felder eingeteilten Areals befinden. Die Pheromonspuren werden durch „Duftpunkte“ auf den Feldern modelliert. In dem Areal gibt es fünf zufällig verteilte Futterquellen mit jeweils 50 Portionen Futter.

Jede einzelne Ameise verhält sich dabei wie folgt:

- Wenn die Ameise einen Futterplatz gefunden hat, nimmt sie eine Portion Futter auf.
- Wenn sie sich im Nest befindet und Futter hat, legt sie dieses ab.
- Wenn sie sich auf einem beliebigen Feld befindet und Futter hat, erhöht sie die Pheromonkonzentration dieses Feldes, indem sie einen Duftpunkt hinzufügt. Anschließend bewegt sie sich auf das direkt angrenzende Feld in Richtung zum Nest.
- Wenn sie kein Futter hat, untersucht sie die Pheromonkonzentration aller angrenzenden Felder und läuft in Richtung der stärksten Konzentration weiter weg vom Nest. Falls in den angrenzenden Feldern keine Duftpunkte zu finden sind, bewegt sie sich auf ein zufällig gewähltes dieser Felder.

#### Aufgabe 2

Schreibe ein Programm zur Simulation der Ameisenfuttersuche, das Ameisen, Nest, Futterquellen und Duftspuren graphisch auf dem Bildschirm darstellt. Erweitere dein Programm so, dass es auch für verschiedene Werte von Ameisen, Nestposition, Futterquellen und Verdunstungszeit verwendet werden kann.

Zeigen sich für bestimmte Werte besondere Effekte?

**Zu den Beispieldaten:** Es gibt keine. Die Aufgabe ist eine Simulation mit Zufall, die Ergebnisse hängen vom Startwert des Zufallsgenerators ab.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Simulation in Schritten („Ticks“):** Man beschreibt den **Zustand** (wo ist jede Ameise, wo liegt noch Futter, wie viel Duft liegt auf welchem Feld) und eine Funktion, die aus einem Zustand den nächsten macht. Die Simulation ruft sie immer wieder auf, wie bei einem Film: Bild für Bild.
- **Nur speichern, was da ist:** Das Areal hat 250 000 Felder, aber nur auf wenigen liegt Duft oder Futter. Man speichert deshalb nur diese Felder in einem **Wörterbuch** (Feld → Menge), nicht in einer riesigen Tabelle. Alles, was nicht im Wörterbuch steht, ist leer.
- **Gleichzeitig handeln:** Alle Ameisen bewegen sich im selben Tick. Neue Duftpunkte dürfen deshalb erst gesetzt werden, **nachdem** alle Ameisen entschieden haben. Sonst würde eine Ameise, die später in der Reihe dran ist, schon den Duft der vorherigen riechen.
- **Der Abstand auf einem Gitter:** Wenn man nur nach oben, unten, links und rechts gehen darf, ist der Abstand zweier Felder `|x1 − x2| + |y1 − y2|` (Manhattan-Abstand).
- **Zufall nachvollziehbar machen:** Mit einem festen **Startwert** liefert der Zufallsgenerator jedes Mal dieselbe Folge. Dann kann man Versuche wiederholen und Parameter fair vergleichen.
- **Experimente auswerten:** Ein Lauf sagt wegen des Zufalls wenig. Man wiederholt ihn mit mehreren Startwerten und bildet den Mittelwert.

**Nützliche Funktionen und Bibliotheken**
- `random.Random(startwert)`: ein eigener Zufallsgenerator. `zufall.choice(liste)` wählt ein Element, `zufall.randrange(n)` eine Zahl von 0 bis n − 1.
- `dict.get(schluessel, 0)`: Wert zu einem Schlüssel, oder 0, wenn er fehlt.
- `matplotlib.pyplot` (`scatter`, `plot`): für die graphische Darstellung.

## Festlegungen

Die Aufgabe lässt einiges offen, deshalb legen wir fest (wie die Lösungshinweise):

- Mehrere Ameisen dürfen auf demselben Feld stehen.
- Ein **angrenzendes** Feld ist ein Feld oben, unten, links oder rechts (nicht diagonal). Nur diese vier Felder riecht eine Ameise, und nur dorthin geht sie.
- Eine Ameise **bemerkt** Futter, wenn es auf ihrem Feld oder einem angrenzenden Feld liegt. Sie nimmt dann eine Portion mit und läuft von **dem Feld aus, auf dem sie steht**, heim.
- **Verdunstung:** Jeder Tick verringert die Menge auf jedem Feld um die **Verdunstungsrate** (in Duftpunkten). Bei der Rate 0,005 ist ein Duftpunkt nach 200 Ticks weg. Das Verdunsten geschieht nach und nach, nicht auf einmal.
- Alle Ameisen starten im Nest. Ein **Tick** ist ein Simulationsschritt, die **Laufzeit** ist die Zahl der Ticks, bis alles Futter im Nest ist.
- **Rückweg:** Jeder Schritt führt dichter ans Nest. Wir nehmen von den (höchstens zwei) Feldern, die näher am Nest liegen, das, das der **Geraden** von der Futterstelle zum Nest am nächsten liegt. Dann liegen die Wege verschiedener Ameisen nur in der Nähe des Nests aufeinander. Läuft man dagegen zuerst ganz in x- und dann ganz in y-Richtung, überlagern sich die Spuren auf langen Stücken und der Duft dort wird sehr stark. Beides ist erlaubt, man muss nur wissen, dass es einen Unterschied macht.
- Liegt in den Feldern, die **weiter vom Nest weg** liegen, kein Duft, geht die Ameise auf ein zufällig gewähltes angrenzendes Feld.

## Die Idee in Worten

1. **Zustand:** Feldgröße, Nest, Verdunstungsrate, Zufallsgenerator, das Futter (Wörterbuch Feld → Portionen), pro Ameise der Ort und „trägt sie Futter?“ (wir merken uns dann das Feld, wo sie es genommen hat, für den Rückweg), der Duft (Wörterbuch Feld → Menge) und einige Zähler zur Auswertung.
2. **Ein Tick:** Jede Ameise handelt der Reihe nach.
   - *Sie trägt Futter:* Im Nest legt sie es ab. Sonst merkt sie sich ihr Feld für einen Duftpunkt und geht einen Schritt Richtung Nest.
   - *Sie trägt kein Futter:* Liegt Futter auf ihrem oder einem angrenzenden Feld, nimmt sie eine Portion. Sonst geht sie dorthin, wo der Duft am stärksten ist (nur Felder weiter weg vom Nest), und bei keinem Duft zufällig.

   Danach werden die gemerkten Duftpunkte gesetzt.
3. **Verdunstung ohne Schleife über alle Felder:** Wir speichern pro Feld die Menge **und** den Tick, zu dem sie zuletzt geändert wurde. Wer die Menge wissen will, rechnet `Menge − Rate · (jetzt − damals)` (nicht unter 0). Dann kostet ein Tick nur so viel wie die Ameisen und nicht wie die Zahl der Duftfelder.
4. **Darstellung und Auswertung:** Ein Bild des Zustands (Duft grün, Futter rot, Ameisen schwarz, Nest blauer Stern) und Zähler.

## Schritt 1: Der Anfangszustand

`neue_simulation` legt den Zustand an. Die Futterquellen liegen an zufälligen Feldern (nicht auf dem Nest und nicht zwei auf demselben Feld). Alle Ameisen stehen im Nest. In `start[i]` steht `None`, solange Ameise `i` kein Futter trägt, sonst das Feld, wo sie es genommen hat.

```python
import random


def neue_simulation(groesse=500, ameisen=100, nest=None, quellen=5, portionen=50,
                    rate=0.005, samen=1):
    """Legt den Anfangszustand an. Der Zustand ist ein Wörterbuch."""
    zufall = random.Random(samen)
    if nest is None:
        nest = (groesse // 2, groesse // 2)
    futter = {}
    while len(futter) < quellen:
        feld = (zufall.randrange(groesse), zufall.randrange(groesse))
        if feld != nest:
            futter[feld] = portionen
    return {'groesse': groesse, 'nest': nest, 'rate': rate, 'zufall': zufall,
            'futter': futter,                # Feld -> übrige Portionen
            'ort': [nest] * ameisen,         # Position jeder Ameise
            'start': [None] * ameisen,       # None = trägt nichts, sonst das Feld, wo das Futter genommen wurde
            'duft': {},                      # Feld -> (Menge, Tick der letzten Änderung)
            'tick': 0, 'abgeliefert': 0, 'abliefer_ticks': [], 'gefunden': {}}


z = neue_simulation(groesse=50, ameisen=10, samen=1)
print('Nest:', z['nest'], ' Futterquellen:', z['futter'])
```

## Schritt 2: Duft, Nachbarn, Abstand

`duftmenge` rechnet die Verdunstung erst beim Nachsehen ein (Idee 3). `nachbarn` liefert die bis zu vier angrenzenden Felder innerhalb des Areals, `abstand` den Gitterabstand.

```python
def duftmenge(z, feld):
    """Duftmenge eines Feldes jetzt (die Verdunstung wird erst beim Nachsehen verrechnet)."""
    if feld not in z['duft']:
        return 0
    menge, seit = z['duft'][feld]
    return max(0, menge - z['rate'] * (z['tick'] - seit))


def nachbarn(z, feld):
    """Die angrenzenden Felder (oben, unten, links, rechts), die im Areal liegen."""
    x, y = feld
    liste = [(x + 1, y), (x - 1, y), (x, y + 1), (x, y - 1)]
    return [(a, b) for a, b in liste if 0 <= a < z['groesse'] and 0 <= b < z['groesse']]


def abstand(feld1, feld2):
    """Gitterabstand: so viele Schritte oben/unten/links/rechts braucht man."""
    return abs(feld1[0] - feld2[0]) + abs(feld1[1] - feld2[1])


print(nachbarn(z, (0, 0)), abstand((0, 0), (3, 4)))
```

## Schritt 3: Wohin geht eine Ameise?

**Mit Futter** (`schritt_zum_nest`): Die Felder, die näher am Nest liegen, sind „ein Schritt in x-Richtung zum Nest“ und „ein Schritt in y-Richtung zum Nest“ (falls x bzw. y noch nicht stimmen). Wir wählen das, das der Geraden von `start` (der Futterstelle) zum Nest näher liegt. Der Abstand eines Punktes `(a, b)` zu einer Geraden durch `start` mit Richtung `(dx, dy)` ist proportional zu `|dx · (b − start_y) − dy · (a − start_x)|` (Kreuzprodukt). Mehr müssen wir nicht wissen: Wir vergleichen nur die beiden Kandidaten.

**Ohne Futter** (`schritt_suchen`): Aus den Nachbarfeldern, die **weiter** vom Nest entfernt sind, nehmen wir die mit der größten Duftmenge. Gibt es mehrere gleich starke, wird eines zufällig gewählt. Liegt auf keinem Duft (größte Menge 0), geht die Ameise auf ein zufälliges Nachbarfeld (alle vier, auch zum Nest hin; so kommt sie am Rand nie in eine Sackgasse).

```python
def schritt_zum_nest(z, feld, start):
    """Nächstes Feld auf dem Weg zum Nest, möglichst nah an der Geraden von start zum Nest."""
    nx, ny = z['nest']
    x, y = feld
    kandidaten = []
    if x != nx:
        kandidaten.append((x + (1 if nx > x else -1), y))
    if y != ny:
        kandidaten.append((x, y + (1 if ny > y else -1)))
    dx, dy = nx - start[0], ny - start[1]
    bestes = kandidaten[0]
    for a, b in kandidaten:
        if abs(dx * (b - start[1]) - dy * (a - start[0])) < abs(dx * (bestes[1] - start[1]) - dy * (bestes[0] - start[0])):
            bestes = (a, b)
    return bestes


def schritt_suchen(z, feld):
    """Nächstes Feld einer Ameise ohne Futter: stärkster Duft weiter weg vom Nest, sonst zufällig."""
    nachb = nachbarn(z, feld)
    weiter = [f for f in nachb if abstand(f, z['nest']) > abstand(feld, z['nest'])]
    beste, staerkste = [], 0
    for f in weiter:
        d = duftmenge(z, f)
        if d > staerkste:
            beste, staerkste = [f], d
        elif d == staerkste and d > 0:
            beste.append(f)
    if beste:
        return z['zufall'].choice(beste)
    return z['zufall'].choice(nachb)


# Weg zurück zum Nest von (6, 3) aus, Nest bei (0, 0)
z0 = neue_simulation(groesse=10, ameisen=1, nest=(0, 0))
weg, feld = [(6, 3)], (6, 3)
while feld != z0['nest']:
    feld = schritt_zum_nest(z0, feld, (6, 3))
    weg.append(feld)
print(weg)
```

Der Rückweg von (6, 3) nach (0, 0) ist eine Treppe, die der Geraden folgt (nicht erst 6 Schritte nach links und dann 3 nach unten).

## Schritt 4: Ein Tick und die ganze Simulation

`tick` geht alle Ameisen durch. Wer Futter trägt, legt es im Nest ab oder merkt sich sein Feld für einen Duftpunkt und geht heim. Wer keines trägt, prüft Futter auf dem eigenen und den angrenzenden Feldern, sonst `schritt_suchen`. In `gefunden` merken wir uns, wann eine Quelle zum ersten Mal angezapft wurde (für die Auswertung). Erst am Ende werden die gemerkten Duftpunkte gesetzt; dabei wird die bis dahin verdunstete Menge verrechnet.

`simuliere` ruft `tick` auf, bis alles Futter im Nest ist, aber höchstens `max_ticks` mal.

```python
def tick(z):
    """Ein Simulationsschritt: alle Ameisen handeln, danach werden die neuen Duftpunkte gesetzt."""
    z['tick'] += 1
    neue_duftpunkte = []
    for i in range(len(z['ort'])):
        feld = z['ort'][i]
        if z['start'][i] is not None:                        # trägt Futter
            if feld == z['nest']:
                z['start'][i] = None                         # Futter ablegen
                z['abgeliefert'] += 1
                z['abliefer_ticks'].append(z['tick'])
            else:
                neue_duftpunkte.append(feld)
                z['ort'][i] = schritt_zum_nest(z, feld, z['start'][i])
            continue
        quelle = None                                        # kein Futter: Quelle hier oder daneben?
        for f in [feld] + nachbarn(z, feld):
            if z['futter'].get(f, 0) > 0:
                quelle = f
                break
        if quelle is not None:
            z['futter'][quelle] -= 1
            z['start'][i] = feld
            if quelle not in z['gefunden']:
                z['gefunden'][quelle] = z['tick']
        else:
            z['ort'][i] = schritt_suchen(z, feld)
    for feld in neue_duftpunkte:                             # erst jetzt: Duftpunkte setzen
        z['duft'][feld] = (duftmenge(z, feld) + 1, z['tick'])


def simuliere(z, max_ticks):
    """Läuft, bis alles Futter im Nest ist oder max_ticks erreicht sind. Rückgabe: Zahl der Ticks."""
    gesamt = sum(z['futter'].values()) + z['abgeliefert'] + sum(1 for s in z['start'] if s is not None)
    while z['abgeliefert'] < gesamt and z['tick'] < max_ticks:
        tick(z)
    return z['tick']


z = neue_simulation(groesse=50, ameisen=10, samen=1)
for _ in range(5):
    tick(z)
print('Orte nach 5 Ticks:', z['ort'])
```

## Schritt 5: Das Bild

`zeichne` malt den Zustand in ein Achsenobjekt von `matplotlib`: Duft als grüne Punkte (je dunkler, desto mehr), Futter als rote Quadrate, Ameisen als kleine schwarze Punkte, das Nest als blauer Stern. Wir malen nur die belegten Felder, nicht alle 250 000.

Zuerst das Areal genau wie in der Aufgabe (500 × 500, 100 Ameisen, 5 Quellen mit je 50 Portionen, Nest in der Mitte), 3000 Ticks lang.

```python
import matplotlib.pyplot as plt


def zeichne(z, achse, titel):
    """Malt Duft, Futter, Ameisen und Nest in die Achse."""
    duftfelder = [f for f in z['duft'] if duftmenge(z, f) > 0]
    if duftfelder:
        achse.scatter([f[0] for f in duftfelder], [f[1] for f in duftfelder],
                      c=[duftmenge(z, f) for f in duftfelder], cmap='Greens', vmin=0, s=6)
    rest = [f for f in z['futter'] if z['futter'][f] > 0]
    achse.scatter([f[0] for f in rest], [f[1] for f in rest], marker='s', c='red', s=40)
    achse.scatter([f[0] for f in z['ort']], [f[1] for f in z['ort']], c='black', s=4)
    achse.scatter([z['nest'][0]], [z['nest'][1]], marker='*', c='blue', s=150)
    achse.set_xlim(-1, z['groesse'])
    achse.set_ylim(-1, z['groesse'])
    achse.set_aspect('equal')
    achse.set_title(titel, fontsize=9)
    achse.tick_params(labelsize=7)


z500 = neue_simulation()
simuliere(z500, 3000)
fig, achse = plt.subplots(figsize=(4, 4))
zeichne(z500, achse, 'Tick %d, 500 x 500' % z500['tick'])
plt.show()
print('Futter im Nest:', z500['abgeliefert'], ' Quellen gefunden:', len(z500['gefunden']), 'von 5')
```

Nach 3000 Ticks sind die 100 Ameisen vom Nest in der Mitte zu einer Wolke ausgeschwärmt, aber noch keine Quelle ist gefunden (die roten Quadrate sind bei diesem riesigen Areal kaum zu sehen).

Im Browser braucht jeder Tick etwa eine halbe Millisekunde oder mehr, ein ganzer Lauf dauert also Minuten. Für Experimente mit vielen Parametern ist das zu langsam. Wir machen sie deshalb auf kleineren Arealen (100 × 100 und 50 × 50). Die Parameter `groesse`, `ameisen`, `nest`, `quellen` und `rate` hat `neue_simulation` ja schon, das Verhalten ist qualitativ dasselbe.

## Schritt 6: Ein ganzer Lauf (100 × 100)

Wir nehmen ein Areal von 100 × 100 mit den übrigen Werten wie in der Aufgabe und schauen den Zustand zu vier Zeitpunkten an.

```python
z100 = neue_simulation(groesse=100, samen=1)
fig, achsen = plt.subplots(2, 2, figsize=(8, 8))
zeitpunkte = [300, 1500, 5000, 9000]
for achse, t in zip(achsen.flatten(), zeitpunkte):
    simuliere(z100, t)
    zeichne(z100, achse, 'Tick %d, %d von 250 Portionen im Nest' % (z100['tick'], z100['abgeliefert']))
plt.tight_layout()
plt.show()
ende100 = simuliere(z100, 10 ** 6)
print('Alles Futter im Nest nach', ende100, 'Ticks')
```

Man sieht die typischen Phasen, wie sie auch in den Lösungshinweisen beschrieben sind:

1. **Start:** Die Ameisen schwärmen zufällig aus, die Quellen nahe am Nest werden zuerst gefunden. Von dort laufen Duftspuren zum Nest.
2. **Folgen der Spur:** Wer ohne Futter eine Spur riecht, läuft sie nach außen entlang und kommt direkt zur Quelle. So werden die gefundenen Quellen zügig leer geräumt.
3. **Tote Spur:** Ist eine Quelle leer, endet die Spur im Nichts. Ameisen, die ihr noch folgen, stehen am Ende und suchen wieder zufällig, bis der Duft verdunstet ist.
4. **Reste:** Die letzten, weit entfernten Quellen werden spät gefunden, das kann lange dauern.

Wie schnell das Futter eingesammelt wird, zeigt die Zeit zwischen zwei Lieferungen. Wir mitteln sie über je 50 Portionen.

```python
zeiten = z100['abliefer_ticks']
pausen = [b - a for a, b in zip([0] + zeiten, zeiten)]       # Ticks seit der vorigen Lieferung
for k in range(0, 250, 50):
    print('Portionen %3d bis %3d: im Mittel alle %3d Ticks eine Lieferung' % (k + 1, k + 50, sum(pausen[k:k + 50]) / 50))
plt.figure(figsize=(4.5, 3))
plt.plot(zeiten, range(1, len(zeiten) + 1))
plt.xlabel('Tick')
plt.ylabel('Portionen im Nest')
plt.show()
```

Die Kurve wird flacher: Am Anfang kommt das Futter schnell, am Ende dauert jede einzelne Portion lange. Das ist **Beobachtung 1** der Lösungshinweise: Die Rate, mit der Futter gefunden wird, nimmt mit der Zeit ab. Je weniger Quellen noch übrig sind, desto weniger Ameisen laufen auf Spuren und desto mehr suchen zufällig. Im letzten Block ist die Pause mit 78 Ticks allerdings wieder kürzer als im vierten mit 175: Im Einzelnen entscheidet der Zufall, nur der Gesamttrend (am Anfang schnell, später viel langsamer) ist eindeutig.

## Schritt 7: Der Einfluss der Parameter

Wir ändern jeweils einen Wert und messen die Laufzeit in Ticks auf einem 50 × 50-Areal, gemittelt über drei Startwerte (1, 2, 3; bei gleicher Zahl von Quellen liegen sie in den Versuchen gleich). Die erste Zeile ist die Ausgangseinstellung: 100 Ameisen, Nest in der Mitte, 5 Quellen, Rate 0,005.

```python
def laufzeiten(**einstellung):
    """Laufzeit auf dem 50 x 50-Areal für die Startwerte 1, 2, 3 (und die fertigen Zustände)."""
    zustaende = []
    for samen in (1, 2, 3):
        zz = neue_simulation(groesse=50, samen=samen, **einstellung)
        simuliere(zz, 100000)
        zustaende.append(zz)
    return [zz['tick'] for zz in zustaende], zustaende


versuche = [('Ausgangseinstellung', {}),
            ('10 Ameisen', {'ameisen': 10}),
            ('400 Ameisen', {'ameisen': 400}),
            ('Nest in der Ecke (0, 0)', {'nest': (0, 0)}),
            ('15 Quellen', {'quellen': 15}),
            ('Rate 0,0005 (langsam)', {'rate': 0.0005}),
            ('Rate 0,05 (schnell)', {'rate': 0.05}),
            ('Rate 0,5 (sehr schnell)', {'rate': 0.5})]
ergebnis = {}
print('%-26s %-22s %s' % ('Versuch', 'Ticks (Start 1, 2, 3)', 'Mittel'))
for name, einstellung in versuche:
    ticks, zustaende = laufzeiten(**einstellung)
    ergebnis[name] = (ticks, zustaende)
    print('%-26s %-22s %6d' % (name, ticks, sum(ticks) / 3))
```

Was man daraus ablesen kann (jeder Wert ist nur ein Mittel aus drei Läufen, also mit Vorsicht):

- **Mehr Ameisen, kürzere Laufzeit.** 10 Ameisen brauchen etwa viermal so lange wie 100, 400 Ameisen nur etwa ein Achtel. Mehr Ameisen suchen gleichzeitig und finden schneller etwas.
- **Mehr Quellen, mehr Futter, längere Laufzeit.** Mit 15 statt 5 Quellen ist dreimal so viel Futter einzusammeln, die Laufzeit steigt etwa auf das Dreifache.
- **Nest in der Ecke:** Die Läufe streuen stark (Startwert 1 braucht drei- bis viermal so lange wie 2 und 3). Mit nur drei Läufen sagt das nur, dass die Ecklage die Laufzeit unberechenbarer macht: Die Ameisen schwärmen nur in ein Viertel des Areals, und liegt dort keine Quelle nah, ist der Weg zu jeder weit.
- **Verdunstungsrate:** Eine sehr langsame Verdunstung (0,0005) ändert gegenüber 0,005 wenig. Eine **schnellere** (0,05) verkürzt hier die Laufzeit sogar deutlich auf weniger als die Hälfte, denn tote Spuren verschwinden früher und führen die Ameisen nicht in die Irre. Bei 0,5 hält ein Duftpunkt nur zwei Ticks, die Ameisen können dann kaum noch Spuren folgen, und die Laufzeit liegt wieder höher als bei 0,05.

Dass die beste Rate von der Lage des Nests, der Zahl der Ameisen und der Verteilung der Quellen abhängt, sagen auch die Lösungshinweise. Es gibt nicht „den“ besten Wert.

## Schritt 8: Weitere Beobachtungen

**Beobachtung 2: Nahe Quellen werden eher gefunden.** In den drei Läufen der Ausgangseinstellung schauen wir, in welcher Reihenfolge die Quellen zum ersten Mal angezapft wurden, und geben den Abstand jeder Quelle zum Nest an. (Zum Sortieren bilden wir Paare `(Tick, Quelle)`: Python sortiert Paare nach dem ersten Eintrag.)

```python
for zz in ergebnis['Ausgangseinstellung'][1]:
    paare = sorted([(tick_, quelle) for quelle, tick_ in zz['gefunden'].items()])
    print('Abstände zum Nest in der Reihenfolge des Findens:', [abstand(q, zz['nest']) for t, q in paare])
```

In jedem Lauf gehört die zuletzt gefundene Quelle zu den beiden am weitesten vom Nest entfernten. Das passt zu den Lösungshinweisen: Futterquellen nahe am Nest werden im Schnitt eher gefunden, denn alle Ameisen starten dort. (Bei nur drei Läufen ist das ein Hinweis, kein Beweis.)

**Besondere Effekte.** Die Lösungshinweise beschreiben zwei Effekte, die bei bestimmten Parametern auftreten:

- **Verlorene Spur** (große Verdunstungsrate, wenige Ameisen, Nest am Rand, weit entfernte Quellen): Eine Ameise folgt einer Spur, die unterwegs abreißt, bevor sie die Quelle erreicht, und muss wieder zufällig suchen. Bei dem langen Lauf auf 500 × 500 oben (alle Quellen schon nach 28 639 Ticks gefunden, aber erst nach 356 224 Ticks leer geräumt) dürfte das eine Rolle spielen: Auf 100 bis 250 Feldern Weg reißt eine Spur leicht ab, weil nur wenige Ameisen gleichzeitig auf ihr zurücklaufen und jeder Duftpunkt nach 200 Ticks weg ist. Das haben wir hier nicht gesondert nachgemessen.
- **Tote Spuren** (kleine Verdunstungsrate, viele Ameisen, Nest in der Mitte): Eine Spur bleibt lange liegen, auch wenn die Quelle schon leer ist. Die Ameisen laufen ihr bis zum Ende nach und stehen dort herum, während die anderen Quellen woanders liegen.

Wer das ausprobieren will: In `neue_simulation` die Werte ändern (z. B. `ameisen=1000, rate=0.0005`, `groesse=100`) und mit `zeichne` anschauen.

- **Geschwindigkeit:** Ein Tick kostet bei uns so viel wie die Zahl der Ameisen, weil die Verdunstung erst beim Nachsehen verrechnet wird (Idee 3). Damit sind auch Läufe auf 500 × 500 mit vielen Wiederholungen möglich.
