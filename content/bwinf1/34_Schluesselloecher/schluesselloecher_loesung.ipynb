### Schlüssellöcher – Lösung Schritt für Schritt

*34. Bundeswettbewerb Informatik (2015), 1. Runde, Aufgabe 4*

In einigen älteren Hotels kommen als Schlüssel Plastikkarten zum Einsatz, bei denen im vorderen Bereich 5 × 5 Codepunkte angeordnet sind. An jedem Punkt ist entweder ein Loch oder kein Loch, so dass jede Schlüsselkarte ein Lochmuster hat. Zum Öffnen einer Tür wird die Karte mit dem Lochmuster voran in einen Schlitz gesteckt. Zu jedem Zimmer darf nur eine Schlüsselkarte passen. Da jede Karte um 180 Grad gewendet werden kann, kann nicht jedes Lochmuster genutzt werden. Die beiden ersten der in der Abbildung dargestellten Schlüssel dürfen also nicht gleichzeitig ausgegeben werden.

#### Aufgabe 4

**Teilaufgabe 1:** Denke über unterschiedliche Wege nach, bei der Produktion der Schlüssel die Menge der erzeugten Lochmuster geeignet einzuschränken.

- Betrachte die äußeren Codepunkte in der vordersten Reihe. Die Karten könnten so gestaltet werden, dass an dem rechten Punkt immer ein Loch und an dem linken Punkt nie ein Loch ist (siehe dritter Schlüssel in der Abbildung). Dadurch kann erkannt werden, ob die Karte richtig herum eingesteckt ist. Wie viele verschiedene Schlüssel können so produziert werden?
- Wenn nur spiegelsymmetrische Lochmuster erlaubt sind (wie beim vierten abgebildeten Schlüssel), ist ein verkehrtes Einlegen der Karte nicht möglich. Wie viele verschiedene Schlüssel können dann produziert werden?

**Teilaufgabe 2:** Für ein Hotel sollen N verschiedene Schlüssel produziert werden. Bei der Auswahl der Lochmuster ist es sinnvoll, dass diese sich möglichst stark unterscheiden. Dann macht es nämlich weniger aus, wenn das Lochmuster eines Schlüssels etwa durch Verschmutzung oder Beschädigung verfälscht wird.

Wie kann man die Anforderung, dass sich die Lochmuster möglichst stark unterscheiden, präzise fassen? Entwickle und implementiere ein Verfahren, das für eine geforderte Anzahl an verschiedenen Schlüsseln die Lochmuster so erzeugt, dass sie sich möglichst stark unterscheiden.

Welche Lochmuster erzeugt dein Programm für N = 7 und N = 20? Wie stark unterscheiden sich die Lochmuster für diese Werte sowie für N = 1000?

**Zu den Daten:** Zu dieser Aufgabe gibt es keine Beispieldaten, das Programm bekommt nur die Zahl N.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zählen mit Zweierpotenzen:** Hat jeder von `n` Punkten zwei Zustände, gibt es `2^n` Muster. Legt man einige Punkte fest, bleiben nur die freien Punkte zum Zählen.
- **Ein Muster als Zahl:** 25 Punkte mit „Loch / kein Loch“ sind 25 Bits, also eine Zahl unter `2^25`. Dann sind Vergleiche und Abstände schnell zu rechnen.
- **Hamming-Abstand:** Die Zahl der Stellen, an denen sich zwei Bitfolgen unterscheiden. Das ist die natürliche Antwort auf „wie stark unterscheiden sich zwei Lochmuster?“.
- **Lokale Suche (Hill-Climbing):** Man startet mit einer zufälligen Lösung, ändert sie ein wenig und behält die Änderung, wenn die Lösung nicht schlechter wird. Das ist keine Garantie für das Optimum, findet aber oft sehr gute Lösungen (die Hinweise nennen es „genetischer Algorithmus“).
- **Schranken aus der Kodierungstheorie:** Wie viele Bitfolgen der Länge 25 mit Mindestabstand `d` gibt es höchstens? Die Hinweise nennen eine Tabelle; damit weiß man, wann eine gefundene Lösung nicht mehr zu verbessern ist.
- **Fehlerkorrigierende Codes:** Für sehr viele Schlüssel (hier 1000) baut man die Muster nicht durch Probieren, sondern mit einem bekannten Code (Golay-Code).

**Nützliche Funktionen und Bibliotheken**
- `a ^ b` (XOR) liefert eine Zahl, in der genau die Bits gesetzt sind, in denen sich `a` und `b` unterscheiden. `bin(x).count('1')` zählt die gesetzten Bits.
- `a >> k & 1`: das Bit an Stelle `k` von `a`; `a | 1 << k`: Bit `k` setzen; `a ^ 1 << k`: Bit `k` umdrehen.
- `random.seed`, `random.randrange`, `random.getrandbits(25)`: Zufallszahlen (mit `seed` jedes Mal dieselben).

## Die Idee in Worten

**Teilaufgabe 1.** Jeder der 25 Punkte hat zwei Zustände, also gibt es `2^25` Muster. Wird eine Karte **gewendet**, vertauschen sich links und rechts (die Lochmuster der beiden ersten Schlüssel in der Abbildung sind Spiegelbilder, so verstehen auch die Lösungshinweise das Wenden). Wir rechnen mit Mustern, die als Zahl gespeichert sind: Bit `5 * Zeile + Spalte` ist 1, wenn dort ein Loch ist.

- **Zwei Eckpunkte festlegen:** links oben kein Loch, rechts oben immer ein Loch. Bleiben 23 freie Punkte, also `2^23` Schlüssel. Eine gewendete Karte hat dann links oben ein Loch und fällt sofort auf.
- **Nur spiegelsymmetrische Muster:** Die rechte Hälfte folgt aus der linken. Frei wählbar sind die drei linken Spalten, also `3 · 5 = 15` Punkte und `2^15` Schlüssel.

Das prüfen wir an einer kleineren Karte (3 × 3), bei der wir alle Muster einzeln durchgehen können.

**Teilaufgabe 2.** *Unterschied* zweier Schlüssel ist die Zahl der Punkte, an denen sie sich unterscheiden (Hamming-Abstand). Eine Schlüsselmenge ist gut, wenn **alle** Paare einen großen Abstand haben, also wenn der **kleinste** Abstand (der *Mindestabstand*) möglichst groß ist: Wenn bis zu `m` Punkte verfälscht sind, ist ein Schlüssel bei Mindestabstand `d > 2m` noch eindeutig dem richtigen Zimmer zuzuordnen. Die Hinweise verwenden den Mittelwert mit der Abweichung; den Mindestabstand erwähnen sie als alternatives Maß. Mittelwert und Abweichung rechnen wir zum Vergleich später ebenfalls aus.

Die Schlüssel müssen außerdem **wendesicher** sein: Kein Schlüssel darf sein eigenes Spiegelbild sein, und kein Schlüssel darf das Spiegelbild eines anderen sein. Sonst würde eine falsch herum gesteckte Karte ein anderes Zimmer öffnen.

Das Verfahren: Wir starten mit `N` zufälligen wendesicheren Schlüsseln. In jedem Schritt drehen wir bei einem zufälligen Schlüssel ein oder zwei zufällige Punkte um. Wird der Mindestabstand dadurch nicht schlechter (bei gleichem Mindestabstand zählen wir die Paare, die genau diesen Abstand haben, und wollen möglichst wenige), behalten wir die Änderung.

## Schritt 1: Teilaufgabe 1 nachprüfen

`spiegeln` spiegelt ein Muster der Größe `n × n`, das als Zahl gespeichert ist. Für `n = 3` gibt es nur `2^9 = 512` Muster; wir zählen alle drei Mengen durch.

```python
def spiegeln(muster, n):
    """Spiegelt ein n x n-Muster links <-> rechts. Bit z*n + s ist 1, wenn in Zeile z, Spalte s ein Loch ist."""
    neu = 0
    for z in range(n):
        for s in range(n):
            if muster >> (z * n + s) & 1:
                neu = neu | 1 << (z * n + (n - 1 - s))
    return neu


n = 3
alle = list(range(2 ** (n * n)))
mit_ecken = [x for x in alle
             if x >> 0 & 1 == 0 and x >> (n - 1) & 1 == 1]      # links oben kein Loch, rechts oben Loch
symmetrisch = [x for x in alle if spiegeln(x, n) == x]
print('alle Muster:', len(alle))
print('Ecken festgelegt:', len(mit_ecken), ' = 2 hoch', n * n - 2)
print('spiegelsymmetrisch:', len(symmetrisch), ' = 2 hoch', ((n + 1) // 2) * n)

# Ist eine gewendete Karte mit festgelegten Ecken je wieder ein gültiger Schlüssel?
menge = set(mit_ecken)
print('Spiegelbild wieder gültig:', any(spiegeln(x, n) in menge for x in mit_ecken))
```

Das passt zu den Überlegungen: `2^7` Schlüssel mit festgelegten Ecken (kein Spiegelbild ist wieder gültig), `2^6` spiegelsymmetrische (bei `n = 3` sind zwei Spalten mal drei Punkte frei, `2 · 3 = 6`). Für die 5 × 5-Karte heißt das:

- alle Muster: `2^25 =` **33 554 432**
- Ecken festgelegt: `2^23 =` **8 388 608** Schlüssel
- spiegelsymmetrisch: `2^15 =` **32 768** Schlüssel

Wer jede mögliche Karte nutzen will, nimmt von jedem Paar „Muster und Spiegelbild“ nur eines und lässt die symmetrischen Muster weg. Das sind `(2^25 - 2^15) / 2 =` 16 760 832 Schlüssel, deutlich mehr als bei den Einschränkungen. Auch das prüfen wir für `n = 3`:

```python
paare = {min(x, spiegeln(x, n)) for x in alle if spiegeln(x, n) != x}
print(len(paare), (2 ** 9 - 2 ** 6) // 2)
print((2 ** 25 - 2 ** 15) // 2)
```

## Schritt 2: Abstand, Anzeige und Bewertung

- `abstand(a, b)`: Hamming-Abstand zweier Muster.
- `zeige(...)`: Muster als Raster mit `O` (Loch) und `-` (kein Loch), mehrere nebeneinander.
- `bewertung(S)`: `(kleinster Abstand, Zahl der Paare mit diesem Abstand)` einer Schlüsselliste. Wir geben die Zahl der Paare mit negativem Vorzeichen an, damit `(14, -15)` besser ist als `(14, -20)` und wir Bewertungen einfach mit `>=` vergleichen können.
- `sicher(S, x)`: Darf `x` zu den Schlüsseln `S` hinzukommen, ohne dass etwas gewendet verwechselt wird?

```python
import random


def abstand(a, b):
    return bin(a ^ b).count('1')


def zeige(schluessel, pro_zeile=5):
    """Gibt mehrere 5 x 5-Schlüssel nebeneinander aus (O = Loch, - = kein Loch)."""
    for anfang in range(0, len(schluessel), pro_zeile):
        gruppe = schluessel[anfang:anfang + pro_zeile]
        for z in range(5):
            zeilen = []
            for x in gruppe:
                zeilen.append(''.join('O' if x >> (z * 5 + s) & 1 else '-' for s in range(5)))
            print('  '.join(zeilen))
        print()


def bewertung(S):
    kleinster, anzahl = 99, 0
    for i in range(len(S)):
        for j in range(i):
            d = abstand(S[i], S[j])
            if d < kleinster:
                kleinster, anzahl = d, 1
            elif d == kleinster:
                anzahl = anzahl + 1
    return (kleinster, -anzahl)


def sicher(S, x):
    """True, wenn x weder symmetrisch ist noch ein Schlüssel aus S oder dessen Spiegelbild ist."""
    spiegel_x = spiegeln(x, 5)
    if spiegel_x == x:
        return False
    for y in S:
        if y == x or y == spiegel_x:
            return False
    return True


print(abstand(0b1010, 0b0110), bewertung([0b1010, 0b0110, 0b0000]))
```

## Schritt 3: Zufallsstart und Verbesserung

`suche(N, schritte, startwert)` baut `N` zufällige wendesichere Schlüssel und versucht dann `schritte`-mal, einen davon zu verbessern. Der `startwert` macht den Zufall wiederholbar.

```python
def suche(N, schritte, startwert):
    random.seed(startwert)
    S = []
    while len(S) < N:                                   # zufällige wendesichere Schlüssel
        x = random.getrandbits(25)
        if sicher(S, x):
            S.append(x)
    beste = bewertung(S)
    for _ in range(schritte):
        i = random.randrange(N)
        x = S[i] ^ 1 << random.randrange(25)            # einen Punkt umdrehen
        if random.random() < 0.3:
            x = x ^ 1 << random.randrange(25)           # manchmal noch einen zweiten
        rest = S[:i] + S[i + 1:]
        if not sicher(rest, x):
            continue
        neu = S[:]
        neu[i] = x
        wert = bewertung(neu)
        if wert >= beste:                               # nicht schlechter: behalten
            S, beste = neu, wert
    return S, beste


S7, wert7 = suche(7, 20000, 0)
print('N = 7: kleinster Abstand', wert7[0], ', Paare mit diesem Abstand:', -wert7[1])
zeige(S7, 7)
```

Der kleinste Abstand ist **14**, und genau 15 der 21 Paare haben ihn. Besser geht es nicht: Aus Tabelle 3 der Lösungshinweise (alle 25-stelligen Bitfolgen mit gleichem Mindestabstand) gibt es mit Abstand 14 höchstens 8 Schlüssel, mit Abstand 15 nur noch 4. Für 7 Schlüssel ist 14 also der größtmögliche Mindestabstand.

Diese Schlüssel sind außerdem wendesicher, weil `suche` nur solche Schlüssel zulässt. Prüfen wir das noch einmal und sehen uns an, wie stark sich die Schlüssel im Mittel unterscheiden (Mittelwert und Abweichung wie in den Hinweisen):

```python
def statistik(S):
    """Mittelwert der Abstände und mittlere Abweichung vom Mittelwert (wie U(S) und A(S) in den Hinweisen)."""
    abstaende = []
    for i in range(len(S)):
        for j in range(i):
            abstaende.append(abstand(S[i], S[j]))
    mittel = sum(abstaende) / len(abstaende)
    abweichung = sum(abs(a - mittel) for a in abstaende) / len(abstaende)
    return mittel, abweichung


def wendesicher(S):
    gewendet = [spiegeln(x, 5) for x in S]
    return len(set(S) | set(gewendet)) == 2 * len(S)      # 2N verschiedene Muster: keine Verwechslung


mittel7, abw7 = statistik(S7)
print('Mittelwert', round(mittel7, 3), ', Abweichung', round(abw7, 3), ', wendesicher:', wendesicher(S7))
```

Der Mittelwert der Abstände ist 14,29, die mittlere Abweichung 0,41. Zum Vergleich: Die Hinweise nennen für ihr Verfahren (Maß: Mittelwert und Abweichung) bei 7 Schlüsseln einen Mittelwert von 13,05 mit Abweichung 0,45. Unsere Menge hat einen größeren Mittelwert und einen viel größeren Mindestabstand, weil wir den kleinsten Abstand direkt maximieren.

Jetzt **N = 20**. Hier gibt es mehr Paare (190), deshalb etwas mehr Schritte:

```python
S20, wert20 = suche(20, 20000, 0)
print('N = 20: kleinster Abstand', wert20[0], ', Paare mit diesem Abstand:', -wert20[1], 'von', 20 * 19 // 2)
zeige(S20, 5)
mittel20, abw20 = statistik(S20)
print('Mittelwert', round(mittel20, 3), ', Abweichung', round(abw20, 3), ', wendesicher:', wendesicher(S20))
```

Der kleinste Abstand ist **12**, 95 der 190 Paare haben ihn. Der Mittelwert ist 13,1 mit Abweichung 1,1 (die Hinweise nennen bei 20 Schlüsseln etwa 13,0 mit Abweichung 0,7: im Mittel gleich, dafür ist bei uns der kleinste Abstand garantiert groß). Auch 12 ist optimal: Mit Abstand 13 gibt es nach Tabelle 3 höchstens 14 Schlüssel, wir brauchen aber 20. Die Suche ist also bei 7 und 20 Schlüsseln nicht nur „gut“, sondern der Mindestabstand ist bewiesenermaßen so groß wie möglich (nur die Zahl der Paare mit genau diesem Abstand ließe sich vielleicht noch senken). Der Zufall spielt eine Rolle: Mit anderen Startwerten ergeben sich andere Schlüssel, aber derselbe Mindestabstand.

## Schritt 4: N = 1000

Bei 1000 Schlüsseln hat die lokale Suche zwei Probleme: Ein Schritt kostet jetzt 1000 Abstandsberechnungen statt 20, und die Aufgabe ist viel schwerer. Schon zufällig gewählte 1000 Schlüssel zeigen, wo man startet:

```python
random.seed(5)
zufaellig = []
while len(zufaellig) < 1000:
    x = random.getrandbits(25)
    if sicher(zufaellig, x):
        zufaellig.append(x)
print('zufällige 1000 Schlüssel: kleinster Abstand', bewertung(zufaellig)[0])
```

Zufällig gewählte Schlüssel enthalten also Paare, die sich nur an ganz wenigen Stellen unterscheiden. Aus der Kodierungstheorie kennt man aber einen Code, der genau hier passt: den **erweiterten binären Golay-Code**. Er besteht aus 4096 Bitfolgen der Länge 24, die sich paarweise an mindestens 8 Stellen unterscheiden. Nach Tabelle 3 der Hinweise sind für Abstand 9 höchstens 836 Schlüssel möglich, für 1000 Schlüssel ist 8 also das Optimum, und der Golay-Code erreicht es.

Konstruktion (ohne Beweis): Man multipliziert eine 12-stellige Zahl `m` (die 4096 möglichen Werte) **ohne Übertrag** mit der festen Bitfolge `g = x^11 + x^10 + x^6 + x^5 + x^4 + x^2 + 1`, das heißt, man bildet die XOR-Summe von `g << i` für alle gesetzten Bits `i` von `m`. Das ergibt 23 Bits, dazu kommt ein 24. Bit, so dass die Zahl der Einsen gerade ist. Die 25. Stelle der Karte bleibt immer ohne Loch. Die XOR-Summe zweier Codewörter ist wieder ein Codewort. Deshalb ist der kleinste Abstand zweier Wörter gleich der kleinsten Zahl von Einsen in einem Wort außer 0, und die können wir einfach nachzählen.

```python
G = 0
for exponent in (11, 10, 6, 5, 4, 2, 0):
    G = G | 1 << exponent                       # g(x) = x^11 + x^10 + x^6 + x^5 + x^4 + x^2 + 1


def golay_wort(m):
    wort = 0
    for i in range(12):
        if m >> i & 1:
            wort = wort ^ G << i                # Multiplikation ohne Übertrag
    if bin(wort).count('1') % 2 == 1:
        wort = wort | 1 << 23                   # Paritätsbit: Zahl der Einsen wird gerade
    return wort


golay = [golay_wort(m) for m in range(4096)]
kleinste_einsen = min(bin(w).count('1') for w in golay if w != 0)
print(len(set(golay)), 'Wörter, kleinste Zahl von Einsen:', kleinste_einsen)
```

Der kleinste Abstand ist also 8. Jetzt nehmen wir die ersten 1000 Wörter, die wendesicher sind (nicht symmetrisch, kein Spiegelbild eines schon gewählten Wortes), und rechnen alle 499 500 Paare nach:

```python
S1000 = []
for w in golay:
    if sicher(S1000, w):
        S1000.append(w)
    if len(S1000) == 1000:
        break

haeufigkeit = {}
for i in range(1000):
    for j in range(i):
        d = abstand(S1000[i], S1000[j])
        haeufigkeit[d] = haeufigkeit.get(d, 0) + 1

print('Schlüssel:', len(S1000), ', wendesicher:', wendesicher(S1000))
print('Abstand: Zahl der Paare')
for d in sorted(haeufigkeit):
    print(d, ':', haeufigkeit[d])
mittel1000, abw1000 = statistik(S1000)
print('Mittelwert', round(mittel1000, 3), ', Abweichung', round(abw1000, 3))
zeige(S1000[1:11], 5)
```

Bei 1000 Schlüsseln gilt also: **kleinster Abstand 8**, kein Paar liegt darunter. Es kommen nur die Abstände 8, 12 und 16 vor, der Mittelwert liegt bei 11,0 (Abweichung 1,9). Ein Fehler an bis zu 3 Punkten eines Schlüssels lässt ihn also noch eindeutig erkennen.

**Zusammenfassung**

| N | kleinster Abstand | Verfahren | größter möglicher Mindestabstand (Tabelle 3) |
|---|---|---|---|
| 7 | 14 | lokale Suche | 14 |
| 20 | 12 | lokale Suche | 12 |
| 1000 | 8 | Golay-Code | 8 |

Dabei bedeutet die Schranke: Mit Abstand 14 gibt es höchstens 8 Schlüssel (7 passt), mit Abstand 13 höchstens 14 (20 passt nicht), mit Abstand 9 höchstens 836 (1000 passt nicht).

## Zum Schluss: MiniZinc?

Man kann das Problem in MiniZinc beschreiben: eine 0/1-Matrix `x[1..N, 1..25]`, für jedes Paar von Zeilen eine Summe `sum(j)(x[i,j] != x[k,j]) >= d`, dazu die Wendesicherheit und `lex_less` zwischen den Zeilen gegen Symmetrien. Wir haben das für `N = 7` mit `d = 14` und `N = 20` mit `d = 12` ausprobiert: Gecode und Chuffed fanden in einer halben Minute nichts, nur CP-SAT löste beide Fälle (nach einigen Sekunden bis etwa einer Minute). Für `N = 1000` wäre das Modell mit 25 000 Variablen und fast 500 000 Abstandsbedingungen viel zu groß. Die lokale Suche und der Golay-Code sind hier klar besser, deshalb gibt es in diesem Notebook keinen MiniZinc-Abschnitt.

## Für Neugierige

- **Mehr als der Mindestabstand:** Die Hinweise bewerten eine Menge mit `B(S) = A(S) / U(S)` (Abweichung durch Mittelwert). Das kann man als Zielfunktion der Suche in `bewertung` einsetzen; der Mindestabstand ist aber das Maß, das für die Verfälschung eines einzelnen Schlüssels wirklich zählt.
- **Warum klappt die Suche bei 7 und 20?** Es gibt sehr viele gleich gute Lösungen, und jeder Schritt verändert nur einen Schlüssel an ein bis zwei Stellen. Bei 1000 Schlüsseln bräuchte man ein Vielfaches der Schritte und einen Merker für die Abstände, statt alle Paare neu zu bewerten.
- **Ohne Wendesicherheit** ist das Problem das aus der Kodierungstheorie bekannte `A2(25, d)`: Für viele `d` kennt man den genauen Wert noch nicht (Tabelle 3).
- **Anderer Code für N = 20:** Eine Hadamard-Matrix der Ordnung 24 liefert 48 Bitfolgen mit Abstand 12 (Paley-Konstruktion mit Quadratresten modulo 23).
