### Schwimmbad – Lösung Schritt für Schritt

*36. Bundeswettbewerb Informatik (2017), 1. Runde, Aufgabe 2*

Antonia (13 Jahre) möchte mit ihren Freundinnen ins Schwimmbad gehen. Sie versucht, sich über die Eintrittspreise zu informieren. Leider sind die Regeln etwas verwirrend:

- Der Eintritt kostet im Einzelpreis für Personen über 16 Jahre 3,50 €, für Jugendliche bis 16 Jahre 2,50 €.
- Wochentags (Mo–Fr) sind die Einzelpreise um 20 % reduziert.
- Kinder unter vier Jahren haben freien Eintritt, müssen aber in Begleitung einer Person über 16 Jahre sein.
- Eine Tageskarte für bis zu sechs Personen (gültig für Erwachsene und Jugendliche gleichermaßen) kostet 11 €, gilt aber nicht am Wochenende.
- Eine Familienkarte (2 Erwachsene und 2 Jugendliche bis 16 Jahre bzw. 1 Erwachsener und 3 Jugendliche) kostet 8 €.
- Ein Gutschein berechtigt zu einem freien Eintritt für eine Person oder 10 % Ermäßigung für eine beliebig große Gruppe. Es kann höchstens ein Gutschein für die Ermäßigung von 10 % eingesetzt werden. Gutscheine gelten nicht in den Ferien.

Antonia hat drei Freundinnen. Ihre Mutter und ihre zweijährige Schwester möchten auch mitkommen. Sie wollen an einem Wochenende in den Ferien ins Schwimmbad gehen und besitzen einen Gutschein. Wie viel müssen sie bezahlen?

#### Aufgabe 2

Gegeben sind die Datumsmerkmale „Wochentag oder Wochenende“ und „Schulzeit oder Ferien“, eine Anzahl von Gutscheinen und eine Liste von Personen mit Altersangaben.

Schreibe ein Programm, das berechnet, wie viel Geld mindestens gezahlt werden muss, damit die ganze Gruppe, falls möglich, das Schwimmbad betreten kann. Etwaige Vorteile für künftige Besuche sollen nicht berücksichtigt werden. Das Programm soll die zu kaufenden Karten und den Gesamtpreis ausgeben.

Wende dein Programm auf das obige Beispiel an und auf alle weiteren Beispiele, die du auf den BwInf-Webseiten findest.

#### Dateien (`beispieldaten/`)

`schwimmbad.txt`: Die Beispiele sind nur **in Worten** beschrieben (eine Abiturklasse, die Großfamilie Stutzenberg, zwei Kegelvereine). Wir übertragen sie unten selbst in Listen von Altersangaben. Wo das Alter nicht genannt ist, nehmen wir ein passendes an; das spielt keine Rolle, denn es zählt nur die Altersstufe.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Eingabe vereinfachen:** Das genaue Alter ist egal. Es gibt nur drei **Altersstufen** (Kind, Jugendlicher, Erwachsener). Wir zählen also nur, wie viele Personen jeder Stufe da sind.
- **Rekursion mit Minimum:** Den billigsten Preis für `j` Jugendliche und `e` Erwachsene erhält man, indem man **eine** Karte kauft und für die übrigen Personen denselben Preis wieder berechnet. Von allen Möglichkeiten für die erste Karte nimmt man die billigste (`min`).
- **Merkliste (Memoisierung):** Dieselben Teilprobleme kommen immer wieder vor. Ein Wörterbuch (`dict`) merkt sich Ergebnisse, die schon berechnet wurden. Das ist die **dynamische Programmierung** aus den Lösungshinweisen.
- **Alle Möglichkeiten durchprobieren (Brute Force) bei den Gutscheinen:** Es gibt nur wenige Gutscheine. Wir probieren alle Verteilungen der Gutscheine auf Jugendliche und Erwachsene und merken uns die billigste.
- **Minimum von Tupeln:** `min` vergleicht Tupel Stelle für Stelle. Steht der Preis vorne, wird das Tupel mit dem kleinsten Preis gewählt, und die restlichen Angaben (die Karten) kommen mit.

**Nützliche Funktionen**
- `min(liste)` und `max(a, b)`; `max(0, x)` verhindert negative Anzahlen.
- `//` ist die ganzzahlige Division (zum Runden auf ganze Cent).
- `liste.count(x)` zählt, wie oft `x` in einer Liste oder einem Tupel vorkommt.
- `dict` mit Tupeln als Schlüssel, z. B. `gemerkt[(j, e, wochenende)]`.

## Die Regeln in Zahlen

Wir rechnen in **Cent**, damit es keine Rundungsfehler mit Kommazahlen gibt.

| Karte | Wochenende | Wochentag |
|---|---|---|
| Einzelkarte Jugendlicher | 250 | 200 (20 % weniger) |
| Einzelkarte Erwachsener | 350 | 280 (20 % weniger) |
| Tageskarte, bis 6 Personen | gibt es nicht | 1100 |
| Familienkarte (2 E + 2 J oder 1 E + 3 J) | 800 | 800 |

Dazu die Gutscheine (nur außerhalb der Ferien): jeder Gutschein ist ein **freier Eintritt**; nur **einer** darf stattdessen **10 %** auf den Gesamtpreis geben.

**Wie wir die Regeln auslegen** (so machen es auch die Lösungshinweise):
- Die Aufgabe sagt „über 16 Jahre“ für Erwachsene. Wer 16 ist, ist also noch Jugendlicher. (Die Hinweise ordnen 16-Jährige den Erwachsenen zu; bei den Beispielen macht das keinen Unterschied.)
- Eine Familienkarte gilt für **höchstens** 2 Erwachsene und 2 Jugendliche oder 1 Erwachsenen und 3 Jugendliche; sie braucht aber **mindestens einen Erwachsenen**. Eine Familienkarte darf also auch für weniger Personen gekauft werden, wenn das billiger ist.
- Kinder unter 4 Jahren sind kostenlos, **wenn** ein Erwachsener in der Gruppe ist. Ohne Erwachsenen zählen sie zu den Jugendlichen und müssen bezahlen.
- Wird ein Erwachsener mit einem Gutschein „freigekauft“, begleitet er die Kinder trotzdem. Die Kinder bleiben also kostenlos.
- Die 10 % werden auf den **Preis der Karten** gerechnet, nachdem die Gutscheine für freie Eintritte abgezogen sind. Wir runden auf volle Cent **auf**.
- Sind mehrere Gutscheine da, dürfen alle bis auf einen für freie Eintritte benutzt werden und der letzte für die 10 %. Man darf aber auch weniger Gutscheine nutzen.

## Die Idee

1. **Personen zählen:** Aus der Liste der Alter werden zwei Zahlen: Anzahl Jugendliche `j` und Anzahl Erwachsene `e` (Kinder kommen nur vor, wenn sie bezahlen müssen, dann zählen sie zu `j`).
2. **Beste Karten ohne Gutscheine:** Rekursiv: `beste_karten(j, e)` probiert als **erste Karte** jede mögliche Karte (Einzelkarte, Familienkarte, Tageskarte) und löst den Rest wieder mit `beste_karten`. Ist niemand mehr übrig (`j = 0` und `e = 0`), kostet es nichts. Eine Merkliste verhindert, dass derselbe Rest tausendmal gerechnet wird.
3. **Gutscheine:** Wir probieren alle Verteilungen: `fj` Gutscheine für Jugendliche und `fe` für Erwachsene (freier Eintritt), und, falls ein Gutschein übrig bleibt, die 10 % auf den Rest.
4. **Ausgeben:** Die Karten zählen, den Preis in Euro schreiben.

## Schritt 1: Altersstufen zählen

````python
def altersstufe(alter):
    """'Kind' (unter 4), 'Jugendlicher' (4 bis 16) oder 'Erwachsener' (über 16)."""
    if alter < 4:
        return 'Kind'
    elif alter <= 16:
        return 'Jugendlicher'
    else:
        return 'Erwachsener'


def zaehle(alter_liste):
    """Gibt (jugendliche, erwachsene, freie_kinder) zurück, die zum Bezahlen relevant sind."""
    kinder = 0
    jugendliche = 0
    erwachsene = 0
    for alter in alter_liste:
        stufe = altersstufe(alter)
        if stufe == 'Kind':
            kinder = kinder + 1
        elif stufe == 'Jugendlicher':
            jugendliche = jugendliche + 1
        else:
            erwachsene = erwachsene + 1
    if erwachsene == 0:                     # ohne Erwachsene sind Kinder nicht frei, sie zahlen wie Jugendliche
        jugendliche = jugendliche + kinder
        kinder = 0
    return jugendliche, erwachsene, kinder


# Aufgabenblatt: Antonia (13) und drei Freundinnen, Mutter (40), zweijährige Schwester
antonia = [13, 13, 13, 13, 40, 2]
print(zaehle(antonia))
````

Vier Jugendliche, ein Erwachsener, ein freies Kind. Das Kind taucht in der Preisberechnung nicht mehr auf.

## Schritt 2: Die Preise

````python
FAMILIENKARTE = 800
TAGESKARTE = 1100


def einzelpreise(wochenende):
    """(Preis Jugendlicher, Preis Erwachsener) in Cent."""
    if wochenende:
        return 250, 350
    else:
        return 200, 280          # 20 % weniger


def euro(cent):
    """Cent als Text, z. B. 1050 -> '10,50 €'."""
    return str(cent // 100) + ',' + str(cent % 100).zfill(2) + ' €'


print(einzelpreise(True), einzelpreise(False), euro(1050), euro(7875))
````

## Schritt 3: Die billigsten Karten ohne Gutscheine

`beste_karten(j, e, wochenende)` gibt ein Tupel `(preis, karten)` zurück; `karten` ist ein Tupel von Namen der gekauften Karten.

Die erste Karte kann sein:
- eine **Einzelkarte** für einen Jugendlichen (falls `j > 0`) oder einen Erwachsenen (falls `e > 0`),
- eine **Familienkarte** (nur wenn `e > 0`), die entweder 2 Erwachsene und 2 Jugendliche oder 1 Erwachsenen und 3 Jugendliche abdeckt. Sind weniger Personen übrig, ziehen wir mit `max(0, ...)` nicht mehr ab, als da sind.
- eine **Tageskarte** (nur wochentags) für `a` Erwachsene und `b` Jugendliche, zusammen höchstens 6. Wir lassen nur **volle** Karten zu (sechs Personen, oder alle Übrigen, wenn es weniger als sechs sind). Eine Tageskarte, die Plätze verschenkt, obwohl noch Personen übrig sind, ist nie besser.

Der Rest wird mit demselben Verfahren gelöst. Die Rekursion endet bei `j = 0` und `e = 0`. Jede Karte deckt mindestens eine Person ab, deshalb wird das Problem immer kleiner.

````python
gemerkt = {}                 # Merkliste: (j, e, wochenende) -> (preis, karten)


def mit_karte(rest, preis, name):
    """Ergebnis für den Rest, ergänzt um eine Karte."""
    rest_preis, rest_karten = rest
    return (rest_preis + preis, (name,) + rest_karten)


def beste_karten(j, e, wochenende):
    """Billigste Karten für j Jugendliche und e Erwachsene (ohne Gutscheine): (preis, karten)."""
    if j == 0 and e == 0:
        return (0, ())
    schluessel = (j, e, wochenende)
    if schluessel in gemerkt:
        return gemerkt[schluessel]

    preis_j, preis_e = einzelpreise(wochenende)
    moeglichkeiten = []
    # Einzelkarten
    if j > 0:
        moeglichkeiten.append(mit_karte(beste_karten(j - 1, e, wochenende), preis_j, 'Jugendlich'))
    if e > 0:
        moeglichkeiten.append(mit_karte(beste_karten(j, e - 1, wochenende), preis_e, 'Erwachsen'))
    # Familienkarte: höchstens 2 Erwachsene + 2 Jugendliche oder 1 Erwachsener + 3 Jugendliche, mindestens 1 Erwachsener
    if e > 0:
        for erw, jug in [(2, 2), (1, 3)]:
            rest = beste_karten(max(0, j - jug), max(0, e - erw), wochenende)
            moeglichkeiten.append(mit_karte(rest, FAMILIENKARTE, 'Familie'))
    # Tageskarte: nur wochentags, genau min(6, Personen) Personen
    if not wochenende:
        personen = min(6, j + e)
        for erw in range(0, min(e, personen) + 1):
            jug = personen - erw
            if jug <= j:
                rest = beste_karten(j - jug, e - erw, wochenende)
                moeglichkeiten.append(mit_karte(rest, TAGESKARTE, 'Tageskarte'))

    ergebnis = min(moeglichkeiten)          # Tupel (preis, karten): kleinster Preis gewinnt
    gemerkt[schluessel] = ergebnis
    return ergebnis


print(beste_karten(4, 1, True))
print(beste_karten(4, 5, False))
print(len(gemerkt), 'Teilprobleme gemerkt')
````

Der Test zum Aufgabenblatt (ohne den Gutschein, der in den Ferien ohnehin nicht gilt): am Wochenende eine Jugendkarte (250) und eine Familienkarte (800) für den Erwachsenen mit den anderen drei Jugendlichen, zusammen 1050 Cent. Wochentags für 4 Jugendliche und 5 Erwachsene: drei Jugendkarten und eine Tageskarte.

Hinweis: In `min(moeglichkeiten)` entscheidet bei gleichem Preis die Reihenfolge der Karten-Tupel. Der Preis ist in jedem Fall der kleinstmögliche.

## Schritt 4: Gutscheine

Wir probieren aus, wie viele Gutscheine für **freie Eintritte** gebraucht werden: `frei` von 0 bis zur Anzahl der Gutscheine. Dann verteilen wir diese `frei` Gutscheine auf `fj` Jugendliche und `fe = frei - fj` Erwachsene. Für die übrigen Personen suchen wir die billigsten Karten. Bleibt mindestens **ein** Gutschein übrig (`frei < gutscheine`), kann man zusätzlich 10 % abziehen.

In den Ferien gelten keine Gutscheine, dann setzen wir ihre Anzahl auf 0.

Wichtig ist, dass wir die Verteilung **nicht** vorschreiben (zum Beispiel „Gutscheine zuerst an die Erwachsenen“). Die Lösungshinweise warnen davor: Gerade bei 4 Jugendlichen und 1 Erwachsenen lohnt sich der Gutschein für einen Jugendlichen, weil dann die Familienkarte für den Rest reicht.

````python
def rabatt(preis):
    """10 % Ermäßigung, auf volle Cent aufgerundet."""
    return (preis * 9 + 9) // 10


def bester_preis(alter_liste, wochenende, ferien, gutscheine):
    """Billigster Gesamtpreis. Gibt (preis, karten, freie_jugendliche, freie_erwachsene, mit_rabatt, freie_kinder) zurück."""
    j, e, kinder = zaehle(alter_liste)
    if ferien:
        gutscheine = 0
    beste = None
    for frei in range(0, min(gutscheine, j + e) + 1):
        for fj in range(0, min(frei, j) + 1):
            fe = frei - fj
            if fe > e:
                continue
            preis, karten = beste_karten(j - fj, e - fe, wochenende)
            kandidaten = [(preis, frei, 0, karten, fj, fe)]
            if frei < gutscheine:                                  # ein Gutschein bleibt für die 10 %
                kandidaten.append((rabatt(preis), frei + 1, 1, karten, fj, fe))
            for kandidat in kandidaten:
                if beste is None or kandidat[:2] < beste[:2]:       # kleinerer Preis, bei Gleichstand weniger Gutscheine
                    beste = kandidat
    preis, genutzt, mit_rabatt, karten, fj, fe = beste
    return preis, karten, fj, fe, mit_rabatt, kinder


print(bester_preis(antonia, True, True, 1))
````

## Schritt 5: Ausgabe

Die Aufgabe verlangt, dass das Programm **die zu kaufenden Karten und den Gesamtpreis** ausgibt. Wir zählen die Karten je Sorte und schreiben auch auf, wie die Gutscheine benutzt werden.

````python
PREISE_TEXT = {'Jugendlich': 'Einzelkarte Jugendlicher', 'Erwachsen': 'Einzelkarte Erwachsener',
               'Familie': 'Familienkarte', 'Tageskarte': 'Tageskarte (bis 6 Personen)'}


def zeige(titel, alter_liste, wochenende, ferien, gutscheine):
    """Gibt Karten und Gesamtpreis aus und liefert den Preis in Cent zurück."""
    preis, karten, fj, fe, mit_rabatt, kinder = bester_preis(alter_liste, wochenende, ferien, gutscheine)
    ort = ('Wochenende' if wochenende else 'Wochentag') + ', ' + ('Ferien' if ferien else 'Schulzeit')
    print(titel + ' (' + ort + ', ' + str(gutscheine) + ' Gutschein(e), ' + str(len(alter_liste)) + ' Personen)')
    for name in PREISE_TEXT:
        if karten.count(name) > 0:
            print('  ' + str(karten.count(name)) + ' x ' + PREISE_TEXT[name])
    if fj + fe > 0:
        print('  Gutscheine: ' + str(fj) + ' x freier Eintritt für einen Jugendlichen, ' + str(fe) + ' x für einen Erwachsenen')
    if mit_rabatt == 1:
        print('  Gutschein: 10 % Ermäßigung')
    if kinder > 0:
        print('  ' + str(kinder) + ' Kind(er) unter 4 Jahren kostenlos')
    print('  Gesamtpreis: ' + str(preis) + ' Cent = ' + euro(preis))
    return preis


zeige('Aufgabenblatt: Antonia und ihre Familie', antonia, True, True, 1)
````

## Für Neugierige: Wie viel Arbeit macht die Merkliste?

Ohne Merkliste würde `beste_karten` denselben Rest immer wieder berechnen: Jede Karte führt zu mehreren Aufrufen, und diese zu weiteren. Die Zahl der Aufrufe wächst **exponentiell** mit der Gruppengröße. Die Merkliste speichert jedes Paar `(j, e, wochenende)` nur einmal. Es gibt höchstens `(j + 1) · (e + 1)` verschiedene Paare je Tagestyp. Wir zählen es für die größte Gruppe der Beispiele:

````python
gemerkt.clear()
beste_karten(0, 32, True)
beste_karten(0, 32, False)
print(len(gemerkt), 'Teilprobleme für 32 Erwachsene (Wochenende und Wochentag)')
gemerkt.clear()
for _wochenende in (True, False):
    beste_karten(4, 5, _wochenende)
print(len(gemerkt), 'Teilprobleme für 4 Jugendliche und 5 Erwachsene')
````

Mit der Merkliste wächst die Arbeit also nur noch grob mit dem **Produkt** von Jugendlichen und Erwachsenen. Eine Gruppe von 100 Personen wäre damit kein Problem. (Python erlaubt standardmäßig etwa 1000 verschachtelte Aufrufe, bei sehr großen Gruppen müsste man die Rekursion durch Schleifen ersetzen.)

## Zum Schluss: Dasselbe mit MiniZinc

Die Aufgabe ist ein **ganzzahliges Optimierungsproblem**: Wie viele Karten jeder Sorte kauft man, damit alle Personen abgedeckt sind und der Preis minimal ist? Das beschreiben wir mit **Variablen für die Anzahlen** und überlassen MiniZinc die Suche. Es gibt keine Rekursion und keine Merkliste.

- Jede Person hat **genau eine** Art von Eintritt: Einzelkarte, Familienkarte, Tageskarte oder Gutschein. Das drücken die zwei Gleichungen `einzel_j + fam_j + tag_j + frei_j = j` (und für Erwachsene) aus. Ähnlich wie bei den Arbeitsblättern zu Zuordnungen zählt man nur, wie viele Personen es sind, nicht welche.
- Für Familienkarten gibt es zwei Variablen (`fam1`, `fam2`) und Ungleichungen für die Anzahl der Erwachsenen und Jugendlichen darauf.
- Das Wochenende schaltet die Tageskarte mit einer **Implikation** `wochenende -> tages = 0` aus.
- Die Gutscheine: `frei_j + frei_e + rabatt <= g`. Die Variable `rabatt` ist 0 oder 1, der Preis wird im Fall 1 mit `* 9` und `div 10` berechnet.
- `solve minimize preis` sucht die billigste Lösung.

Die Altersstufen zählt weiter Python (`zaehle`); die Zahlen gehen als Daten an MiniZinc. Die Modelle sind **kein Python**: Kopiere sie in die MiniZinc IDE (oder in den [MiniZinc Playground](https://play.minizinc.org/)).

Das Modell (`schwimmbad.mzn`):

```minizinc
int: j;                    % Jugendliche, die bezahlen müssen
int: e;                    % Erwachsene
bool: wochenende;
int: g;                    % Gutscheine (in den Ferien 0)
int: preis_j;              % Einzelpreise in Cent
int: preis_e;

var 0..j: einzel_j;        % Einzelkarten
var 0..e: einzel_e;
var 0..e: fam1;            % Familienkarten für 2 Erwachsene + 2 Jugendliche
var 0..e: fam2;            % Familienkarten für 1 Erwachsenen + 3 Jugendliche
var 0..(j + e): tages;     % Tageskarten
var 0..e: fam_e;           % Erwachsene auf Familienkarten
var 0..j: fam_j;           % Jugendliche auf Familienkarten
var 0..e: tag_e;           % Erwachsene auf Tageskarten
var 0..j: tag_j;
var 0..j: frei_j;          % Gutscheine als freier Eintritt
var 0..e: frei_e;
var 0..1: rabatt;          % 1: ein Gutschein für 10 %

% jede Person hat genau eine Art von Eintritt
constraint einzel_j + fam_j + tag_j + frei_j = j;
constraint einzel_e + fam_e + tag_e + frei_e = e;
% Familienkarten: mindestens 1 Erwachsener je Karte, höchstens 2 bzw. 1; höchstens 2 bzw. 3 Jugendliche
constraint fam_e >= fam1 + fam2;
constraint fam_e <= 2 * fam1 + fam2;
constraint fam_j <= 2 * fam1 + 3 * fam2;
% Tageskarten: sechs Personen je Karte, nicht am Wochenende
constraint tag_e + tag_j <= 6 * tages;
constraint wochenende -> tages = 0;
% Gutscheine
constraint frei_j + frei_e + rabatt <= g;

var int: summe = einzel_j * preis_j + einzel_e * preis_e + (fam1 + fam2) * 800 + tages * 1100;
var int: preis = if rabatt = 1 then (summe * 9 + 9) div 10 else summe endif;
solve minimize preis;

output ["Preis \(preis) Cent: \(einzel_j) Jugend-Einzel, \(einzel_e) Erwachsenen-Einzel, ",
        "\(fam1 + fam2) Familie, \(tages) Tageskarte, Gutscheine frei \(frei_j + frei_e), Rabatt \(rabatt)\n"];
```

Die Daten (`.dzn`) erzeugt eine Python-Zelle aus der Altersliste:

````python
def als_dzn(alter_liste, wochenende, ferien, gutscheine):
    """Gruppe und Datum als MiniZinc-Daten (.dzn)."""
    j, e, kinder = zaehle(alter_liste)
    preis_j, preis_e = einzelpreise(wochenende)
    zeilen = ['j = ' + str(j) + ';', 'e = ' + str(e) + ';',
              'wochenende = ' + str(wochenende).lower() + ';',
              'g = ' + str(0 if ferien else gutscheine) + ';',
              'preis_j = ' + str(preis_j) + ';', 'preis_e = ' + str(preis_e) + ';']
    return '\n'.join(zeilen)


print(als_dzn(antonia, True, True, 1))
````
