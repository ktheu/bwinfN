### Buhnenrennen – Lösung Schritt für Schritt

*35. Bundeswettbewerb Informatik (2016), 1. Runde, Aufgabe 5*

An Nordseestränden gibt es **Buhnen**: Reihen von Baumstämmen, die senkrecht zur Wasserlinie in den Sand gerammt sind. Durch Lücken zwischen den Stämmen kommt man hindurch. Eine **Maxilücke** ist so groß, dass auch der große Hund Max durchpasst (und damit auch die kleine Minnie), durch eine **Minilücke** passt nur Minnie. Die Buhnen stehen 70 m voneinander entfernt. Minnie schafft 20 km/h, Max ist mit 30 km/h schneller und will sie fangen.

Anfangs stehen beide an Lücken in der ersten Buhne. Minnie versucht, alle Buhnen zu durchqueren, indem sie von Lücke zu Lücke läuft. Ein solcher Minnieweg ist **sicher**, wenn Max sie nicht fangen kann, egal wie er läuft. Hunde und Lücken sind punktförmig.

#### Aufgabe 5

Schreibe ein Programm, das feststellt, ob ein sicherer Minnieweg existiert, und einen solchen gegebenenfalls ausgibt. Wende es mindestens auf die Beispiele von bundeswettbewerb-informatik.de an und dokumentiere die Ergebnisse.

**Zu den Beispieldaten:** Es gibt keine README, der Aufbau ist aus den Dateien abgelesen (und mit den Lösungshinweisen abgeglichen). Jede Zeile ist eine Lücke: ein Buchstabe, die x-Position der Buhne (0, 70, 140, …) und die y-Position in Metern. `m` ist eine Minilücke, `x` eine Maxilücke. In der ersten Buhne stehen `M` (Minnies Startposition) und `X` (Startposition von Max); die übrigen Lücken der ersten Buhne benutzt keiner der beiden. Das erste Beispiel (`buhnenrennen1.txt`) ist das Bild der Aufgabe:

```
X 0 168.6
M 0 42.95
m 0 93.95
m 70 76.4
x 70 45.9
m 70 117.4
...
```

Im Ordner `eigene_testdaten` liegen zwei kleine Beispiele: `ein_weg.txt` (es gibt genau einen sicheren Weg) und `kein_weg.txt` (es gibt keinen).

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Graph und kürzeste Wege:** Die Lücken sind Knoten, das Laufen von einer Lücke zur nächsten Buhne sind Kanten, deren Länge die Laufzeit ist. Gesucht ist hier nicht nur „irgendein Weg“, sondern der **schnellste** Weg zu jeder Lücke.
- **Schicht für Schicht (dynamische Programmierung):** Man kann nur von Buhne zu Buhne vorwärts laufen. Wenn die Zeiten für alle Lücken einer Buhne bekannt sind, ergeben sich die der nächsten Buhne daraus: Zeit der Lücke = kleinste (Zeit der Vorgängerlücke + Laufzeit). So braucht man keine Prioritätswarteschlange wie bei Dijkstra, eine Schleife über die Buhnen reicht.
- **Wettlauf als Zeitvergleich:** Max kann Minnie nur abfangen, wenn er an einer Lücke **vor ihr** ankommt (dann wartet er dort einfach). Und wenn er irgendwo auf Minnies Weg früher ist als sie, dann auch an der nächsten Lücke, denn er ist schneller. Es genügt also, an jeder **Lücke** die Ankunftszeiten zu vergleichen. Zwischen den Buhnen müssen wir nichts prüfen, und die kürzeste Verbindung zwischen zwei Lücken ist die gerade Strecke.
- **Vorgänger merken:** Zu jedem Zwischenziel merkt man sich, von wo man gekommen ist, und liest den Weg am Ende rückwärts ab.
- **Früh ist besser:** Für Minnie ist an jeder Lücke die **früheste** Ankunftszeit die beste (je früher sie da ist, desto eher ist sie vor Max). Deshalb muss man pro Lücke nur den schnellsten Weg merken, nicht alle.

**Nützliche Funktionen und Bibliotheken**
- `math.hypot(a, b)`: die Länge `sqrt(a*a + b*b)` (Satz des Pythagoras). Hier der Abstand zweier Lücken: 70 Meter in x-Richtung, der Unterschied der y-Werte in y-Richtung.
- `min(liste)`: kleinster Wert einer Liste, auch mit einer Listenabfrage (*list comprehension*) wie `min([t + ... for ... in ...])`.
- `float('inf')`: „unendlich“, größer als jede Zahl. Praktisch als Zeit für eine Lücke, die man nie erreicht.
- `zip(liste1, liste2)`: läuft zwei Listen gleichzeitig durch.

## Die Idee in Worten

**Einheiten:** Wir rechnen in Metern und Sekunden. 30 km/h sind 30 / 3,6 = 8,33 m/s, 20 km/h sind 5,56 m/s. Die Laufzeit zwischen einer Lücke bei `y1` und einer Lücke in der nächsten Buhne bei `y2` ist `hypot(70, y2 - y1) / Tempo`.

**1. Max' Zeiten.** Max startet bei `X`. Für jede Lücke der zweiten Buhne ist seine Zeit die Laufzeit vom Start dorthin. Für die dritte Buhne gilt: Max kommt nur von **Maxilücken** der zweiten Buhne aus weiter (durch eine Minilücke passt er nicht). Seine Zeit an einer Lücke der dritten Buhne ist das Minimum über alle Maxilücken `i` der zweiten Buhne von `Zeit[i] + Laufzeit(i, Lücke)`. Und so weiter bis zur letzten Buhne. In eine Minilücke kann Max noch hineinlaufen (das ist wichtig, denn dort kann er auf Minnie warten), aber nicht weiter.

**2. Minnies Weg.** Minnie startet bei `M` zur Zeit 0 und kommt überall durch. Für jede Lücke bestimmen wir wie bei Max die kleinste Ankunftszeit, aber nur über **sichere** Vorgängerlücken. Ist ihre Zeit an der Lücke kleiner als Max' Zeit dort, ist die Lücke **sicher**, sonst darf der Weg nicht darüber führen. Wir merken uns den Vorgänger. Gibt es am Ende in der letzten Buhne eine sichere Lücke, laufen wir von dort rückwärts bis zum Start und haben einen sicheren Weg. Am besten wählt man die Lücke mit dem größten Vorsprung (Max' Zeit minus Minnies Zeit).

Warum genügt der früheste Weg? Gäbe es irgendeinen sicheren Weg zu einer Lücke, dann ist auch der früheste sicher (er ist ja noch früher da). Und von dort aus ist jede Fortsetzung mindestens so gut. Gibt es also überhaupt einen sicheren Weg, findet dieses Verfahren einen.

## Schritt 1: Die Daten einlesen

`lies(pfad)` gibt drei Dinge zurück: Minnies Startposition (y), Max' Startposition (y) und eine Liste der übrigen Buhnen. Jede Buhne ist eine Liste von Paaren `(art, y)` mit `art` gleich `'m'` oder `'x'`. Die erste Buhne (mit `M` und `X`) kommt nicht mehr in der Liste vor, die Lücken `m`/`x` der ersten Buhne werden weggelassen, weil sie keiner benutzt.

Nebenbei: In jeder Zeile steht die x-Position der Buhne (0, 70, 140, …). Die Reihenfolge der Zeilen ist schon nach Buhnen sortiert, wir gruppieren trotzdem über die x-Position, dann ist es egal.

````python
import math

V_MAX = 30 / 3.6          # Max in Metern pro Sekunde
V_MINNIE = 20 / 3.6       # Minnie in Metern pro Sekunde
ABSTAND = 70              # Abstand der Buhnen in Metern


def lies(pfad):
    """Rückgabe: (y von Minnie, y von Max, Liste der weiteren Buhnen mit Paaren (art, y))."""
    minnie_y = None
    max_y = None
    nach_x = {}                                   # x-Position der Buhne -> Lücken
    with open(pfad) as datei:
        for zeile in datei:
            teile = zeile.split()
            if len(teile) != 3:
                continue
            art, x, y = teile[0], int(teile[1]), float(teile[2])
            if art == 'M':
                minnie_y = y
            elif art == 'X':
                max_y = y
            elif x > 0:
                if x not in nach_x:
                    nach_x[x] = []
                nach_x[x].append((art, y))
    buhnen = [nach_x[x] for x in sorted(nach_x)]
    return minnie_y, max_y, buhnen


minnie_y, max_y, buhnen = lies('beispieldaten/buhnenrennen1.txt')
print('Minnie startet bei y =', minnie_y, ', Max bei y =', max_y)
for nr, buhne in enumerate(buhnen, start=2):
    print('Buhne', nr, ':', buhne)
````

Das Beispiel hat also vier Buhnen: die erste mit den beiden Startpositionen, dann drei weitere mit drei, drei und vier Lücken (`m` Minilücke, `x` Maxilücke).

## Schritt 2: Laufzeiten und Max' früheste Zeiten

`laufzeit(y1, y2, tempo)` ist die Zeit für die gerade Strecke von einer Lücke bei `y1` zu einer Lücke bei `y2` in der nächsten Buhne.

`max_zeiten` gibt für jede Buhne eine Liste mit Max' frühester Ankunftszeit an jeder Lücke zurück. `von` enthält die Lücken, von denen aus Max weiterlaufen kann, jeweils mit ihrer Zeit `(y, zeit)`. Am Anfang ist das nur sein Startpunkt. Nach jeder Buhne sind es die **Maxilücken** dieser Buhne. Gibt es in einer Buhne keine Maxilücke, kommt Max nicht weiter (`min` über eine leere Liste mit `default=float('inf')` ergibt dann „unendlich“).

````python
def laufzeit(y1, y2, tempo):
    """Zeit in Sekunden für die gerade Strecke zwischen zwei benachbarten Buhnen."""
    return math.hypot(ABSTAND, y2 - y1) / tempo


def max_zeiten(max_y, buhnen):
    """Für jede Buhne eine Liste: früheste Ankunftszeit von Max an jeder Lücke."""
    alle = []
    von = [(max_y, 0.0)]                      # Lücken, von denen Max weiterlaufen kann: (y, Zeit)
    for buhne in buhnen:
        zeiten = []
        for art, y in buhne:
            zeit = min([t + laufzeit(py, y, V_MAX) for py, t in von], default=float('inf'))
            zeiten.append(zeit)
        alle.append(zeiten)
        von = []
        for (art, y), zeit in zip(buhne, zeiten):
            if art == 'x':                    # nur durch Maxilücken kommt Max weiter
                von.append((y, zeit))
    return alle


zeiten_max = max_zeiten(max_y, buhnen)
for nr, (buhne, zeiten) in enumerate(zip(buhnen, zeiten_max), start=2):
    print('Buhne', nr, ':', [(art, y, round(t, 1)) for (art, y), t in zip(buhne, zeiten)])
````

Zur Kontrolle von Hand: In Buhne 2 ist die erste Lücke eine Minilücke bei `y = 76,4`. Max steht bei `168,6`, also ist der Weg `hypot(70, 92,2) = 115,8` m lang, und er braucht 115,8 / 8,33 = **13,9 s**.

## Schritt 3: Minnies Weg

Für jede Buhne speichern wir die **sicheren** Lücken als Liste von Einträgen `(index, y, zeit, vorgaenger)`: der Index der Lücke in der Buhne, ihre y-Position, Minnies früheste Ankunftszeit und die **Position des Vorgängers in der Liste der vorigen Buhne**. Die Liste der Startbuhne enthält nur Minnies Startpunkt `(0, minnie_y, 0.0, None)`.

Für jede Lücke der nächsten Buhne suchen wir unter den sicheren Lücken der Buhne davor die mit der kleinsten Ankunftszeit. Ist sie **kleiner als Max' Zeit**, ist die Lücke sicher und kommt in die Liste. Eine genau gleiche Zeit nennen wir unsicher (Max steht dann schon da und fängt Minnie ab).

Am Ende nehmen wir in der letzten Buhne die sichere Lücke mit dem **größten Vorsprung** und folgen den Vorgängern rückwärts. `minnie_weg` gibt `None` zurück, wenn die letzte Buhne keine sichere Lücke hat, sonst `(weg, vorsprung, gesamtzeit)`. Der Weg ist eine Liste von y-Positionen, eine je Buhne, beginnend mit Minnies Start.

````python
def minnie_weg(minnie_y, buhnen, zeiten_max):
    """Sicherer Weg mit größtem Vorsprung am Ende, als (Liste der y-Positionen, Vorsprung, Gesamtzeit), oder None."""
    sicher = [[(0, minnie_y, 0.0, None)]]         # je Buhne: Liste von (Index, y, Zeit, Vorgänger-Position)
    for buhne, zeiten in zip(buhnen, zeiten_max):
        neue = []
        for index, (art, y) in enumerate(buhne):
            beste_zeit = None
            bester_vorgaenger = None
            for position, (_, py, t, _) in enumerate(sicher[-1]):
                zeit = t + laufzeit(py, y, V_MINNIE)
                if beste_zeit is None or zeit < beste_zeit:
                    beste_zeit = zeit
                    bester_vorgaenger = position
            if beste_zeit is not None and beste_zeit < zeiten[index]:      # Minnie ist vor Max da
                neue.append((index, y, beste_zeit, bester_vorgaenger))
        if len(neue) == 0:
            return None                           # kein sicheres Loch in dieser Buhne
        sicher.append(neue)
    # letzte Buhne: Loch mit größtem Vorsprung
    bester = None
    bester_vorsprung = None
    for position, (index, y, zeit, _) in enumerate(sicher[-1]):
        vorsprung = zeiten_max[-1][index] - zeit
        if bester_vorsprung is None or vorsprung > bester_vorsprung:
            bester_vorsprung = vorsprung
            bester = position
    weg = []
    position = bester
    for buhne_sicher in reversed(sicher):
        weg.append(buhne_sicher[position][1])
        position = buhne_sicher[position][3]
    weg.reverse()
    return weg, bester_vorsprung, sicher[-1][bester][2]


ergebnis = minnie_weg(minnie_y, buhnen, zeiten_max)
weg, vorsprung, gesamtzeit = ergebnis
print('y-Positionen des Weges:', weg)
print('Vorsprung am Ende: %.1f s, Gesamtzeit: %.1f s' % (vorsprung, gesamtzeit))
````

Für das Beispiel der Aufgabe ergibt sich ein Vorsprung von 10,6 s bei einer Gesamtzeit von 37,9 s. Das sind genau die Zahlen für den „sichersten Weg“ aus den Lösungshinweisen (dort sind es 3 sichere Wege, der gefährlichste hat nur 1,6 s Vorsprung; das schauen wir uns später an).

## Schritt 4: Das Ganze als eine Funktion, und alle 14 Beispiele

`loese(pfad)` liest eine Datei und gibt das Ergebnis von `minnie_weg` zurück. Mit den 14 Beispielen von bwinf.de bauen wir eine Tabelle: Zahl der Buhnen, Zahl der Lücken, den Vorsprung des sichersten Weges und seine Gesamtzeit. Bei den großen Beispielen drucken wir nicht den ganzen Weg aus.

````python
def loese(pfad):
    minnie_y, max_y, buhnen = lies(pfad)
    zeiten_max = max_zeiten(max_y, buhnen)
    return minnie_weg(minnie_y, buhnen, zeiten_max)


print('Nr  Buhnen  Lücken  Vorsprung  Gesamtzeit')
for nummer in range(1, 15):
    pfad = 'beispieldaten/buhnenrennen' + str(nummer) + '.txt'
    minnie_y, max_y, buhnen = lies(pfad)
    luecken = sum(len(b) for b in buhnen)
    ergebnis = loese(pfad)
    if ergebnis is None:
        print('%2d  %6d  %6d  kein sicherer Weg' % (nummer, len(buhnen) + 1, luecken))
    else:
        weg, vorsprung, gesamtzeit = ergebnis
        print('%2d  %6d  %6d  %7.1f s  %8.1f s' % (nummer, len(buhnen) + 1, luecken, vorsprung, gesamtzeit))
````

In **jedem** der 14 Beispiele gibt es einen sicheren Weg, so steht es auch in den Lösungshinweisen. Die Zahlen für Vorsprung und Gesamtzeit des sichersten Weges stimmen mit den Hinweisen überein (z. B. Beispiel 3: 13,3 s Vorsprung bei 91,8 s, Beispiel 8: 80,7 s bei 309,7 s, Beispiel 12: 102,0 s bei 484,9 s, Beispiel 14: 91,2 s bei 597,1 s). Bei Beispiel 4 ist es knapp: 0,5 s Vorsprung.

Der Weg für Beispiel 1 (y-Position je Buhne) und für Beispiel 2, die kleinen zum Nachvollziehen:

````python
for nummer in [1, 2]:
    weg, vorsprung, gesamtzeit = loese('beispieldaten/buhnenrennen' + str(nummer) + '.txt')
    print('Beispiel', nummer, ': y-Positionen', weg)
````

## Für Neugierige: Alle sicheren Wege zählen

Die Hinweise nennen zu den ersten Beispielen die **Zahl aller sicheren Wege**, den sichersten und den gefährlichsten Weg. Dazu kann man die Wege einfach alle durchprobieren (Rekursion): Minnie läuft von ihrer Lücke zu jeder Lücke der nächsten Buhne, geht dort weiter, wenn sie vor Max ankommt, und zählt jedes Mal, wenn sie die letzte Buhne erreicht.

Das geht nur bei kleinen Beispielen, denn die Zahl der Wege wächst **exponentiell** (bei `n` Lücken pro Buhne und `b` Buhnen bis zu `n^(b-1)` Wege). Beispiel 7 hat schon 156 254 sichere Wege und Beispiel 8 über 24 Millionen. Das Zeitverhalten des Verfahrens oben: für jede Buhne und jede Lücke wird jede Lücke der Buhne davor angeschaut, also etwa `b · n²` Schritte (mit `b` Buhnen und durchschnittlich `n` Lücken). Das sind bei unseren Beispielen höchstens einige Tausend Schritte, nicht Millionen.

````python
def alle_wege(minnie_y, buhnen, zeiten_max):
    """Probiert alle Wege aus. Rückgabe: Liste (Gesamtzeit, Vorsprung am Ende) der sicheren Wege."""
    ergebnisse = []

    def weiter(stufe, y, zeit):
        for index, (art, ny) in enumerate(buhnen[stufe]):
            neue_zeit = zeit + laufzeit(y, ny, V_MINNIE)
            if neue_zeit < zeiten_max[stufe][index]:
                if stufe == len(buhnen) - 1:
                    ergebnisse.append((neue_zeit, zeiten_max[stufe][index] - neue_zeit))
                else:
                    weiter(stufe + 1, ny, neue_zeit)

    weiter(0, minnie_y, 0.0)
    return ergebnisse


print('Nr  Wege  größter Vorsprung  kleinster Vorsprung')
for nummer in range(1, 7):
    minnie_y, max_y, buhnen = lies('beispieldaten/buhnenrennen' + str(nummer) + '.txt')
    wege = alle_wege(minnie_y, buhnen, max_zeiten(max_y, buhnen))
    vorspruenge = [v for z, v in wege]
    print('%2d  %4d  %8.1f s  %8.1f s' % (nummer, len(wege), max(vorspruenge), min(vorspruenge)))
````

Das sind die Zahlen der Lösungshinweise: 3, 8, 192, 2, 7 und 60 sichere Wege. Auch die Vorsprünge des gefährlichsten Weges stimmen (z. B. 1,6 s bei Beispiel 1; bei Beispiel 3 kommt Minnie auf dem gefährlichsten Weg praktisch gleichzeitig mit Max an, gerundet 0,0 s). Der **Vorsprung am Ende** ist hier (wie in den Hinweisen) der Abstand an der letzten Buhne; an einer Lücke mittendrin kann der Vorsprung noch kleiner gewesen sein.

Das schnellere Verfahren von oben findet den sichersten Weg, ohne die Wege einzeln zu zählen. Das ist der Gewinn, wenn man pro Lücke nur die früheste Zeit merkt.

## Zum Schluss: Geht das mit MiniZinc?

Ja, bei den meisten Beispielen sogar schnell. Wir beschreiben nicht das Verfahren, sondern nur, was gelten soll: Max' früheste Zeiten, Minnies Weg und dass sie überall früher ankommt als Max. Wir lassen den Solver (Gecode) den **Vorsprung am Ende maximieren**.

Das Modell rechnet in **ganzen Millisekunden** (die Laufzeiten werden aus den Positionen gerundet). Ein erster Versuch mit Kommazahlen (`float`) ist bei Gecode bei einem Beispiel gescheitert, der Solver meldete „keine Lösung“, obwohl es mehrere gibt. Mit ganzen Zahlen rechnet er zuverlässig. Dabei kann der Vorsprung um ein paar Millisekunden von unserer Python-Rechnung abweichen.

`c[k]` ist das Loch von Minnie in Buhne `k` (Nummer in der Reihenfolge der Datei, in der ersten Buhne 1 = Minnies Startloch). Die Max-Zeiten `tmax` sind selbst Variablen, die über `min` aus den Zeiten der Maxilöcher davor berechnet werden.

```minizinc
int: B;                                    % Anzahl Buhnen
int: N;                                    % größte Lochzahl einer Buhne
array[1..B] of int: n;                     % Löcher je Buhne
array[1..B, 1..N] of float: y;             % y-Position der Löcher (Buhne 1: Minnies Startloch in Spalte 1)
array[1..B, 1..N] of bool: gross;          % Maxiloch?
float: xs;                                 % y-Position von Max am Start

float: VMAX = 30 / 3.6;                    % Meter pro Sekunde
float: VMINNIE = 20 / 3.6;

% Laufzeit in Millisekunden bei Höhenunterschied dy (Abstand der Buhnen: 70 m)
function int: ms(float: dy, float: v) = round(1000 * sqrt(70.0 * 70.0 + dy * dy) / v);

% dauer[v, k, i, j]: Zeit von Loch i der Buhne k-1 zu Loch j der Buhne k; v = 1 Minnie, v = 2 Max
array[1..2, 2..B, 1..N, 1..N] of int: dauer =
  array4d(1..2, 2..B, 1..N, 1..N,
          [ms(y[k, j] - y[k-1, i], if v = 1 then VMINNIE else VMAX endif)
           | v in 1..2, k in 2..B, i in 1..N, j in 1..N]);

% früheste Ankunftszeit von Max (Buhne 2: direkt vom Start, danach nur von Maxilöchern aus)
array[2..B, 1..N] of var 0..10000000: tmax;
constraint forall(j in 1..n[2])(tmax[2, j] = ms(y[2, j] - xs, VMAX));
constraint forall(k in 3..B, j in 1..n[k])(
  tmax[k, j] = min([tmax[k-1, i] + dauer[2, k, i, j] | i in 1..n[k-1] where gross[k-1, i]]));

% Minnies Weg: c[k] ist das Loch in Buhne k, t[k] ihre Ankunftszeit
array[1..B] of var 1..N: c;
array[1..B] of var 0..10000000: t;
constraint c[1] = 1 /\ t[1] = 0;
constraint forall(k in 2..B)(c[k] <= n[k]);
constraint forall(k in 2..B)(t[k] = t[k-1] + dauer[1, k, c[k-1], c[k]]);
constraint forall(k in 2..B)(t[k] < tmax[k, c[k]]);          % Minnie ist vor Max da

solve maximize tmax[B, c[B]] - t[B];                          % größter Vorsprung am Ende
output ["vorsprung = ", show(tmax[B, c[B]] - t[B]), "\n", "c = ", show(c), "\n"];
```

Die Datendatei erzeugt eine Python-Zelle aus der Beispieldatei. Für Beispiel 1:

````python
def als_dzn(pfad):
    """Beispieldatei als MiniZinc-Daten."""
    minnie_y, max_y, buhnen = lies(pfad)
    anzahl = len(buhnen) + 1                      # Buhnen einschließlich der ersten
    breite = max(len(b) for b in buhnen)
    ys = [[minnie_y] + [0.0] * (breite - 1)]      # erste Buhne: nur Minnies Startloch
    gross = [[False] * breite]
    for buhne in buhnen:
        ys.append([y for art, y in buhne] + [0.0] * (breite - len(buhne)))
        gross.append([art == 'x' for art, y in buhne] + [False] * (breite - len(buhne)))
    text = 'B = %d;\nN = %d;\n' % (anzahl, breite)
    text += 'n = [1, %s];\n' % ', '.join(str(len(b)) for b in buhnen)
    text += 'xs = %r;\n' % max_y
    text += 'y = array2d(1..%d, 1..%d, [%s]);\n' % (anzahl, breite, ', '.join(repr(v) for zeile in ys for v in zeile))
    text += 'gross = array2d(1..%d, 1..%d, [%s]);\n' % (anzahl, breite,
                                                        ', '.join('true' if v else 'false' for zeile in gross for v in zeile))
    return text


print(als_dzn('beispieldaten/buhnenrennen1.txt'))
````
