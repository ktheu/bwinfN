### ChoreoGraph – Lösung Schritt für Schritt

*44. Bundeswettbewerb Informatik (2025), 1. Runde, Aufgabe 2*

In der Sportklasse sollen neue Tänze einstudiert werden, bei denen sechzehn Tänzerinnen und Tänzer nebeneinander auf einer
langen Linie stehen. Es werden dann von allen gleichzeitig Figuren getanzt, nach denen die gleichen Positionen auf der Linie
wieder alle belegt sind, aber eventuell in veränderter Reihenfolge. Die Schülerinnen und Schüler dürfen sich eigene Figuren
ausdenken, die jeweils zwischen einem und zwölf Takten Musik beanspruchen.

Die Figuren sollen dann zu einer Choreographie umgesetzt werden, die so gestaltet ist, dass die Linie am Anfang und am Ende
die gleiche Reihenfolge von Schülerinnen und Schülern aufweist, sodass man auf den Shirts verteilt den Namen der Schule
lesen kann.

Nora hat sich eine Notation ausgedacht, um die Dauer und die Auswirkungen einer Figur zu vermerken. So gibt es etwa die
Figur „Cast Four“, in der jeweils jede vierte Person von links hinter den anderen drei Positionen nach links tanzt und sich
dort einreiht, während die anderen eine Position nach rechts rücken. Die Figur beansprucht zwei Takte.

```
Aufstellung vor Cast Four:   A B C D  E F G H  I J K L  M N O P
Aufstellung nach Cast Four:  D A B C  H E F G  L I J K  P M N O
```

In Noras Liste wird die Figur so geführt:

```
Cast Four, 2, DABCHEFGLIJKPMNO
```

Sie notiert also den Namen und die Anzahl der Takte und dann eine Zeichenkette, die die Endpositionen der mit A bis P
bezeichneten Schülerinnen und Schüler angibt, wenn sie vor der Figur in der Reihenfolge AB … OP standen.

Es stehen unterschiedliche Musikstücke zur Auswahl mit unterschiedlichen Anzahlen von Takten, die jedoch immer glatt
durch 8 teilbar sind.

#### Aufgabe 2

Schreibe ein Programm, das die verfügbaren Figuren in Noras Notation einliest sowie die Gesamtzahl von Takten, die das
Musikstück hat. Das Programm soll dann passende Choreographien bestimmen, also eine Folge von Figuren, deren Gesamtlänge
mit der Länge des Musikstücks übereinstimmt. Außerdem muss am Ende der Choreographie jede Person wieder an ihrer
Startposition stehen.

Falls es keine mögliche Choreographie gibt, sollte dein Programm dies melden. Falls es mehrere mögliche Choreographien
gibt, gib jeweils die beste bezüglich der folgenden Kriterien aus:

- Möglichst viele unterschiedliche Figuren werden eingebaut.
- Möglichst viele Figuren werden eingebaut.
- Möglichst wenige Figuren werden eingebaut.
- Die von allen Tanzenden insgesamt zurückgelegte Strecke soll möglichst groß sein.
- Die von allen Tanzenden insgesamt zurückgelegte Strecke soll möglichst klein sein.

Wende dein Programm mindestens auf alle Beispiele an, die du auf den BWINF-Webseiten findest, und dokumentiere die Ergebnisse.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `choreo01.txt` bis `choreo06.txt` und `choreoA.txt`. Die erste Zeile enthält die Anzahl der
Takte des Musikstücks, die zweite die Anzahl der Figuren, dann folgt je Figur eine Zeile mit Name (ohne Leerzeichen), Anzahl der
Takte und den 16 Buchstaben der Endpositionen:

```
32
2
Cast_Four 2 DABCHEFGLIJKPMNO
Cast_Up 4 BCDEFGHIJKLMNOPA
```

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Eine Figur ist eine Umordnung (Permutation):** Der Buchstabe an Position `i` der Zeichenkette sagt, wer nach der Figur an
  Position `i` steht. Zwei Figuren nacheinander ergeben wieder eine Umordnung.
- **Die Kriterien hängen nur von der Auswahl ab:** Wie viele Figuren, wie viele verschiedene und wie weit alle laufen, hängt nur
  davon ab, **wie oft** jede Figur vorkommt, nicht von der Reihenfolge. Darum geht man in **zwei Stufen** vor (so auch die
  Lösungshinweise): zuerst die Anzahlen festlegen, dann eine passende Reihenfolge suchen.
- **Alle Möglichkeiten mit Rekursion:** Die Anzahlen für alle Figuren durchprobieren, deren Takte zusammen genau die Länge des
  Musikstücks ergeben. Wie viele Schleifen man bräuchte, hängt von der Zahl der Figuren ab, darum Rekursion.
- **Backtracking mit Abkürzungen:** Die Reihenfolge Figur für Figur aufbauen. Dabei hilft: Eine gültige Choreographie bleibt
  gültig, wenn man sie **im Kreis dreht** (abc → bca). Man darf also die erste Figur festlegen. Außerdem merkt man sich
  **Sackgassen**, also Zustände, von denen aus es nicht weitergeht.
- **Eine schnelle Vorprüfung über gerade und ungerade Umordnungen:** Jede Umordnung braucht entweder eine gerade oder eine
  ungerade Zahl von Vertauschungen. Das addiert sich beim Hintereinandertanzen, egal in welcher Reihenfolge. Ist die Summe
  ungerade, kann am Ende nie die Startaufstellung herauskommen.

**Nützliche Funktionen und Bibliotheken**
- Texte und Listen, `ord`, `abs`, `''.join(...)`, Tupel als Schlüssel in einer Menge (`set`)
- Keine Bibliothek nötig

## Die Idee

1. Für jede Figur vorab berechnen, wie weit alle Tanzenden laufen (die Strecke) und ob sie gerade oder ungerade ist.
2. **Stufe 1:** Alle Auswahlen aufzählen, also für jede Figur eine Anzahl, sodass die Takte zusammen genau die Länge des
   Musikstücks ergeben.
3. Für jede Auswahl die fünf Kriterien ausrechnen. Nur wenn sie in **mindestens einem** Kriterium besser wäre als alles bisher
   Gefundene und die Vorprüfung besteht, wird in Stufe 2 eine Reihenfolge gesucht.
4. **Stufe 2:** Mit Backtracking eine Reihenfolge der Figuren suchen, nach der alle wieder an ihrem Platz stehen. Gibt es eine,
   ist das eine gültige Choreographie, und sie wird für die Kriterien vermerkt, die sie verbessert.
5. Am Ende für jedes Kriterium die beste Choreographie ausgeben, oder melden, dass es keine gibt.

Die Aufgabe sagt „gib jeweils die beste bezüglich der folgenden Kriterien aus“. Wir verstehen das wie die Lösungshinweise: Für
**jedes** der fünf Kriterien wird eine eigene beste Choreographie bestimmt. Haben mehrere Choreographien denselben besten Wert,
geben wir die zuerst gefundene aus.

## Schritt 1: Die Eingabe lesen

Wir speichern die Figuren in drei Listen: `namen`, `takte` und `muster`. Das Muster wandeln wir gleich in eine Liste von Zahlen
um: Aus `D` wird 3, aus `A` wird 0 und so weiter (`ord(buchstabe) - ord('A')`). Damit lässt sich besser rechnen.

````python
def lies_figuren(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = []
        for zeile in f.read().splitlines():
            if zeile.split():
                zeilen.append(zeile.split())
    laenge = int(zeilen[0][0])
    anzahl = int(zeilen[1][0])
    namen = []
    takte = []
    muster = []
    for i in range(anzahl):
        name, t, buchstaben = zeilen[2 + i]
        namen.append(name)
        takte.append(int(t))
        zahlen = []
        for b in buchstaben:
            zahlen.append(ord(b) - ord('A'))
        muster.append(zahlen)
    return laenge, namen, takte, muster


laenge, namen, takte, muster = lies_figuren('beispieldaten/choreo01.txt')
print('Musikstück:', laenge, 'Takte')
for i in range(len(namen)):
    print(namen[i], takte[i], muster[i])
````

## Schritt 2: Eine Figur tanzen

Die Aufstellung ist ein Text aus 16 Buchstaben, am Anfang `ABCDEFGHIJKLMNOP`. Das Muster einer Figur sagt für jede Position
`i`, **von welcher Position** die Person kommt, die nach der Figur dort steht: Steht im Muster an Position 0 die Zahl 3 (das `D`),
dann steht nach der Figur dort, wer vorher auf Position 3 stand.

Das gilt nicht nur für die Startaufstellung, sondern für jede: Eine Figur beschreibt, wie sich die **Plätze** verändern, egal
wer gerade wo steht.

````python
START = 'ABCDEFGHIJKLMNOP'


def tanze(aufstellung, figur_muster):
    neu = []
    for von in figur_muster:
        neu.append(aufstellung[von])
    return ''.join(neu)


cast_four = muster[0]
a = START
for mal in range(1, 5):
    a = tanze(a, cast_four)
    print('nach', mal, 'mal Cast Four:', a)
````

Nach viermal Cast Four steht wieder jede Person an ihrem Platz: In jeder Vierergruppe rückt jedes Mal jede Person um einen Platz
im Kreis weiter.

## Schritt 3: Die Strecke einer Figur

Wer nach der Figur an Position `i` steht, kam von Position `muster[i]`. Er ist also `abs(i - muster[i])` Plätze weit gelaufen.
Die Strecke der Figur ist die Summe über alle 16 Positionen. Weil sie nicht davon abhängt, wer gerade wo steht, reicht es, sie
**einmal** für jede Figur auszurechnen. Bei Cast Four läuft in jeder Vierergruppe eine Person 3 Plätze und drei Personen je
einen Platz, zusammen 4 · 6 = 24.

````python
def strecke(figur_muster):
    summe = 0
    for i in range(16):
        summe = summe + abs(i - figur_muster[i])
    return summe


for i in range(len(namen)):
    print(namen[i], 'Strecke', strecke(muster[i]))
````

## Schritt 4: Stufe 1, alle Auswahlen

Eine **Auswahl** ist eine Liste `anzahlen`: `anzahlen[i]` sagt, wie oft Figur `i` vorkommt. Gesucht sind alle Auswahlen, bei
denen die Takte zusammen genau die Länge des Musikstücks ergeben.

Die rekursive Funktion `alle_auswahlen` entscheidet nacheinander über Figur 0, Figur 1 und so weiter. Für Figur `i` probiert sie
alle Anzahlen 0, 1, 2, …, solange die Takte noch passen, und ruft sich dann für Figur `i + 1` auf. Wenn über alle Figuren
entschieden ist und die Länge genau stimmt, wird die Auswahl in die Ergebnisliste übernommen (als Kopie, denn `anzahlen` wird
danach weiter verändert).

````python
def alle_auswahlen(i, takte_bisher, anzahlen, takte, laenge, ergebnis):
    if i == len(takte):
        if takte_bisher == laenge:
            ergebnis.append(anzahlen[:])
        return
    anzahl = 0
    while takte_bisher + anzahl * takte[i] <= laenge:
        anzahlen[i] = anzahl
        alle_auswahlen(i + 1, takte_bisher + anzahl * takte[i], anzahlen, takte, laenge, ergebnis)
        anzahl = anzahl + 1
    anzahlen[i] = 0


auswahlen = []
alle_auswahlen(0, 0, [0] * len(takte), takte, laenge, auswahlen)
for a in auswahlen:
    print(a, '=', a[0], 'mal', namen[0], '+', a[1], 'mal', namen[1])
````

Für `choreo01.txt` gibt es 9 Auswahlen: Cast Four (2 Takte) und Cast Up (4 Takte) müssen zusammen 32 Takte ergeben. Ob sich eine
Auswahl auch so tanzen lässt, dass alle wieder an ihrem Platz stehen, ist damit aber noch nicht gesagt. Das klärt Stufe 2.

## Schritt 5: Stufe 2, eine Reihenfolge suchen

`reihenfolge(aufstellung, rest, folge, ...)` baut eine Choreographie Figur für Figur auf. `rest[i]` sagt, wie oft Figur `i`
noch eingebaut werden muss, `folge` ist die Liste der bisher gewählten Figuren.

- Sind keine Figuren mehr übrig, ist die Folge gültig, wenn die Aufstellung wieder `START` ist.
- Sonst wird jede Figur ausprobiert, die noch übrig ist: tanzen, sich selbst aufrufen, und wenn das nicht zum Ziel führt, die
  Figur wieder zurücklegen (Backtracking).

Zwei Abkürzungen machen das schneller:

1. **Im Kreis drehen:** Ist `a b c` eine gültige Choreographie, dann auch `b c a` und `c a b`. Darum dürfen wir die **erste Figur
   festlegen**. Wir nehmen eine, die in der Auswahl vorkommt, und zwar die seltenste. Dann wird jede Choreographie nur in einer
   ihrer gedrehten Formen gesucht.
2. **Sackgassen merken:** Kommt man auf verschiedenen Wegen zu derselben Aufstellung mit denselben übrigen Figuren, geht es von
   dort genau gleich weiter. Ist dieser Zustand schon einmal gescheitert, braucht man ihn nicht noch einmal zu durchsuchen. Die
   Zustände speichern wir in der Menge `sackgassen` als Paare `(aufstellung, tuple(rest))`. Tupel, nicht Listen, denn nur
   unveränderliche Werte dürfen in eine Menge.

Mit `schritte` zählen wir, wie oft die Funktion aufgerufen wird.

````python
schritte = [0]


def reihenfolge(aufstellung, rest, folge, muster, sackgassen):
    schritte[0] = schritte[0] + 1
    if sum(rest) == 0:
        if aufstellung == START:
            return folge[:]                         # gültig: alle wieder am Platz
        return None
    zustand = (aufstellung, tuple(rest))
    if zustand in sackgassen:
        return None
    for i in range(len(rest)):
        if rest[i] > 0:
            rest[i] = rest[i] - 1
            folge.append(i)
            ergebnis = reihenfolge(tanze(aufstellung, muster[i]), rest, folge, muster, sackgassen)
            folge.pop()
            rest[i] = rest[i] + 1
            if ergebnis is not None:
                return ergebnis
    sackgassen.add(zustand)                         # von hier aus geht es nicht
    return None


def finde_reihenfolge(anzahlen, muster):
    rest = anzahlen[:]
    erste = -1
    for i in range(len(rest)):                      # die seltenste vorkommende Figur kommt zuerst
        if rest[i] > 0 and (erste == -1 or rest[i] < rest[erste]):
            erste = i
    rest[erste] = rest[erste] - 1
    return reihenfolge(tanze(START, muster[erste]), rest, [erste], muster, set())


for a in auswahlen:
    folge = finde_reihenfolge(a, muster)
    if folge is None:
        print(a, 'keine gültige Reihenfolge')
    else:
        text = []
        for i in folge:
            text.append(namen[i])
        print(a, ' '.join(text))
````

Von den 9 Auswahlen lassen sich nur 2 gültig tanzen: 16-mal Cast Four und 8-mal Cast Four mit 4-mal Cast Up. Nur Cast Up allein (8-mal) geht zum Beispiel nicht: Danach ist jede Person 8 Plätze weitergerückt.

## Schritt 6: Gerade und ungerade Figuren

Jede Umordnung kann man durch **Vertauschen von je zwei Personen** herstellen. Wie viele Vertauschungen man braucht, hängt vom
Vorgehen ab, aber ob die Zahl **gerade oder ungerade** ist, steht für jede Umordnung fest. Beim Hintereinandertanzen addieren
sich die Vertauschungen, also:

- gerade + gerade = gerade, ungerade + ungerade = gerade, gerade + ungerade = ungerade.
- Das gilt in **jeder** Reihenfolge.
- Die Startaufstellung braucht 0 Vertauschungen, ist also gerade.

Ist die Summe der Vertauschungen einer Auswahl ungerade, kann sie in keiner Reihenfolge zur Startaufstellung führen. Das prüfen
wir, bevor wir lange nach einer Reihenfolge suchen.

Die Vertauschungen einer Figur zählen wir, indem wir ihr Muster mit Vertauschungen wieder sortieren: Solange an Position `i`
nicht die Zahl `i` steht, tauschen wir die Zahl an ihren richtigen Platz.

````python
def vertauschungen(figur_muster):
    liste = figur_muster[:]
    anzahl = 0
    for i in range(16):
        while liste[i] != i:
            j = liste[i]
            liste[i], liste[j] = liste[j], liste[i]       # Zahl j an ihren Platz bringen
            anzahl = anzahl + 1
    return anzahl


def ist_gerade(anzahlen, muster):
    summe = 0
    for i in range(len(anzahlen)):
        summe = summe + anzahlen[i] * vertauschungen(muster[i])
    return summe % 2 == 0


for i in range(len(namen)):
    print(namen[i], vertauschungen(muster[i]), 'Vertauschungen')
for a in auswahlen:
    print(a, 'gerade' if ist_gerade(a, muster) else 'ungerade')
````

Cast Four braucht 12 Vertauschungen (gerade), Cast Up 15 (ungerade). Auswahlen mit einer ungeraden Zahl von Cast Up fallen
sofort heraus, ohne dass eine Reihenfolge gesucht werden muss. Bei den größeren Beispielen halbiert das die Rechenzeit ungefähr.

## Schritt 7: Die Kriterien und die besten Choreographien

Für eine Auswahl berechnen wir die fünf Werte: Zahl der verschiedenen Figuren, Zahl der Figuren und Gesamtstrecke (die Strecke
jeder Figur mal ihrer Anzahl). In `beste` merken wir uns für jedes Kriterium den besten Wert und die zugehörige Choreographie.
`besser` sagt, ob ein neuer Wert für ein Kriterium besser ist: Bei „wenige Figuren“ und „kurze Strecke“ ist kleiner besser,
sonst größer.

Eine Reihenfolge suchen wir nur, wenn die Auswahl gerade ist **und** in mindestens einem Kriterium besser wäre als das bisher
Beste. Sonst kann sie nichts mehr ändern.

````python
KRITERIEN = ['viele verschiedene Figuren', 'viele Figuren', 'wenige Figuren', 'lange Strecke', 'kurze Strecke']


def werte(anzahlen, strecken):
    verschieden = 0
    figuren = 0
    weg = 0
    for i in range(len(anzahlen)):
        if anzahlen[i] > 0:
            verschieden = verschieden + 1
        figuren = figuren + anzahlen[i]
        weg = weg + anzahlen[i] * strecken[i]
    return [verschieden, figuren, figuren, weg, weg]


def besser(k, neu, alt):
    if KRITERIEN[k] == 'wenige Figuren' or KRITERIEN[k] == 'kurze Strecke':
        return neu < alt
    return neu > alt


def beste_choreographien(laenge, takte, muster):
    strecken = []
    for m in muster:
        strecken.append(strecke(m))
    auswahlen = []
    alle_auswahlen(0, 0, [0] * len(takte), takte, laenge, auswahlen)
    beste = [None] * len(KRITERIEN)                 # je Kriterium: [wert, folge]
    for a in auswahlen:
        w = werte(a, strecken)
        lohnt = False
        for k in range(len(KRITERIEN)):
            if beste[k] is None or besser(k, w[k], beste[k][0]):
                lohnt = True
        if not lohnt or not ist_gerade(a, muster):
            continue
        folge = finde_reihenfolge(a, muster)
        if folge is not None:
            for k in range(len(KRITERIEN)):
                if beste[k] is None or besser(k, w[k], beste[k][0]):
                    beste[k] = [w[k], folge]
    return beste


def als_text(folge, namen):
    """Gleiche Figuren hintereinander zusammenfassen: Fan Fan -> Fan(2)."""
    teile = []
    i = 0
    while i < len(folge):
        j = i
        while j < len(folge) and folge[j] == folge[i]:
            j = j + 1
        if j - i == 1:
            teile.append(namen[folge[i]])
        else:
            teile.append(namen[folge[i]] + '(' + str(j - i) + ')')
        i = j
    return ' '.join(teile)


beste = beste_choreographien(laenge, takte, muster)
for k in range(len(KRITERIEN)):
    print(KRITERIEN[k] + ':', beste[k][0], ' ', als_text(beste[k][1], namen))
````

## Schritt 8: Alle Beispiele

Für jedes Beispiel geben wir je Kriterium den Wert und eine beste Choreographie aus, oder melden, dass es keine gibt. Außerdem
zählen wir die Aufrufe der Reihenfolgesuche.

````python
def bearbeite(dateiname):
    laenge, namen, takte, muster = lies_figuren(dateiname)
    schritte[0] = 0
    beste = beste_choreographien(laenge, takte, muster)
    print(dateiname + ':', laenge, 'Takte,', len(namen), 'Figuren', ' (Suchschritte:', str(schritte[0]) + ')')
    if beste[0] is None:
        print('  Es gibt keine gültige Choreographie.')
    else:
        for k in range(len(KRITERIEN)):
            print('  ' + KRITERIEN[k] + ':', beste[k][0])
            print('     ', als_text(beste[k][1], namen))
    print()
    return beste


ergebnisse = {}
for name in ['choreo01', 'choreo02', 'choreo03', 'choreo04', 'choreo05', 'choreo06', 'choreoA']:
    ergebnisse[name] = bearbeite('beispieldaten/' + name + '.txt')
````

**Ergebnisse:** Die Werte stimmen mit der Tabelle in den offiziellen Lösungshinweisen überein:

| Beispiel | Figuren min | Figuren max | verschiedene Figuren max | Strecke min | Strecke max |
|---|---|---|---|---|---|
| choreo01 | 12 | 16 | 2 | 312 | 384 |
| choreo02 | – | – | – | – | – |
| choreo03 | – | – | – | – | – |
| choreo04 | 12 | 12 | 6 | 1154 | 1154 |
| choreo05 | 6 | 24 | 2 | 504 | 1728 |
| choreo06 | 4 | 16 | 4 | 256 | 768 |

Warum gibt es bei manchen Beispielen keine Choreographie?

- **choreo02:** Jede Figur `Cycle_k` schiebt alle um `k` Plätze nach rechts und dauert `k` Takte. Nach 24 Takten sind also alle
  um 24 Plätze weitergerückt, das sind 8 Plätze mehr als eine volle Runde. Niemand steht an seinem Platz.
- **choreo03:** Die Figuren dauern 5 und 10 Takte, 48 ist aber nicht durch 5 teilbar.
- **choreoA:** Ähnlich: Aus den 23 Figurenlängen lassen sich keine 48 Takte zusammensetzen. Darum gibt es gar keine Auswahl, und
  das Programm ist sofort fertig.

Bei `choreo04` führen alle fünf Kriterien zur selben Choreographie. Hier rechnet das Programm am längsten: Weil kaum eine Auswahl
gültig ist, muss es für viele Auswahlen alle Reihenfolgen durchsuchen, bevor feststeht, dass es keine gibt. Mit einem auf dem
Rechner installierten Python dauert das einige Sekunden, im Browser länger.

## Für Neugierige

**Warum nicht einfach alle Folgen von Figuren durchprobieren?** Bei `choreo05` gibt es eine Figur mit einem Takt, und das Stück hat
24 Takte. Folgen aus sechs Figuren mit bis zu 24 Gliedern gibt es astronomisch viele. Erst die Trennung in Auswahl und
Reihenfolge, das Drehen im Kreis und das Merken der Sackgassen machen die Suche schnell genug. Die Bewertungskriterien halten ein
Durchprobieren für ausreichend, verlangen aber, dass keine Choreographie mehrfach getestet wird.

**Mehr zum Drehen im Kreis:** Folgen, die man durch Drehen ineinander überführen kann, heißen in der Kombinatorik **Halsketten**
(englisch *necklaces*). Es gibt Verfahren, die von jeder Halskette genau eine Form erzeugen. Unser Festlegen der ersten Figur ist
eine einfachere Näherung daran, die manche Formen noch mehrfach erzeugt.

**Gerade und ungerade:** Dass die Anzahl der Vertauschungen immer gerade oder immer ungerade ist, egal wie man vorgeht, ist ein
Satz aus der Mathematik der Permutationen (das *Vorzeichen* einer Permutation). Es ist die einzige Eigenschaft dieser Art, die von
der Reihenfolge nicht abhängt. Darum muss man danach trotzdem eine Reihenfolge suchen.

**Treffen in der Mitte:** Eine schnellere Reihenfolgesuche baut die Choreographie gleichzeitig von vorn (vom Start) und von hinten
(rückwärts vom Ziel) auf und sucht eine Stelle, an der beide Hälften dieselbe Aufstellung haben. Das ist deutlich aufwendiger zu
programmieren.
