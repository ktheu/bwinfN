# 14. Zuordnen: Paare bilden mit verbessernden Wegen

Viele Aufgaben haben dieselbe Form: Es gibt **zwei Gruppen** (Schüler und Praktikumsplätze, Aufgaben und Helfer, Lücken und Wörter, Kinder und Geschenke)
und eine Liste, welche Paare **erlaubt** sind. Gesucht sind möglichst viele Paare, sodass **jeder höchstens einmal** vorkommt. Das nennt man ein **Matching**
(deutsch: Zuordnung). Als Graph (Notebook 5 und 8) sind die beiden Gruppen zwei Reihen von Knoten, und die erlaubten Paare sind Kanten zwischen den Reihen.

In [Notebook 12](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/12_greedy.ipynb) hast du gelernt, dass ein gieriges Verfahren schnell, aber nicht immer richtig ist, und dass man Gegenbeispiele findet,
indem man es mit Durchprobieren ([Notebook 1](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/01_durchprobieren.ipynb)) vergleicht. Beides brauchst du hier wieder: Das gierige Zuordnen bleibt hängen,
und die Lösung ist eine **rekursive Suche** ([Notebook 2](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/02_backtracking.ipynb) und [Notebook 8](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/08_reihenfolgen.ipynb)) mit einer
Menge der schon besuchten Knoten.

In diesem Notebook lernst du:

- eine Zuordnung als **Wörterbuch** (`dict`) darzustellen und zu prüfen, ob sie **gültig** ist,
- mit dem Vergleich zum Durchprobieren zu **zeigen**, dass das gierige Zuordnen nicht die größtmögliche Zuordnung findet,
- das Verfahren mit **verbessernden Wegen** zu schreiben: Ist der Platz besetzt, fragt man den Besitzer, ob er woanders hin kann (rekursiv, mit einer **Menge der besuchten Plätze**),
- warum die Zuordnung dabei **immer gültig** bleibt, wie viel das Verfahren **kostet** (Laufzeit **O(Schüler · Kanten)**, mit Zähler) und warum man am Ende nicht mehr verbessern kann,
- woran man erkennt, dass eine Aufgabe nach **Kosten** oder **Wünschen mit Rang** fragt und ein einfaches Matching nicht reicht.

In den Wettbewerbs-Ecken verteilst du Wichtelgeschenke (vereinfacht) und setzt Wörter in einen Lückentext.

## 1. Praktikumsplätze: Paare, die passen

Vier Schüler suchen einen Praktikumsplatz. Jeder nennt die Plätze, die er sich vorstellen kann:

| Schüler | möchte zur / zum |
|---|---|
| Anna | Bäckerei, Bank |
| Ben | Bäckerei |
| Cem | Bank, Zoo |
| Dora | Zoo, Kino |

Jeder Platz darf nur an **einen** Schüler gehen, und jeder Schüler bekommt höchstens **einen** Platz. Gesucht ist eine Zuordnung mit möglichst vielen Paaren.

In Python liegen die Wünsche in einem Wörterbuch: Der Schlüssel ist der Schüler, der Wert die Liste seiner Plätze.

```python
wuensche = {
    'Anna': ['Bäckerei', 'Bank'],
    'Ben': ['Bäckerei'],
    'Cem': ['Bank', 'Zoo'],
    'Dora': ['Zoo', 'Kino'],
}
```

Die **Zuordnung** speichern wir **umgekehrt**, nämlich als Wörterbuch `besitzer`: Der Schlüssel ist der **Platz**, der Wert der Schüler, der ihn bekommt.
Der Grund: Gleich muss man oft fragen „Wem gehört dieser Platz?“, und das ist mit `besitzer[platz]` (oder `platz in besitzer` für „ist besetzt?“) eine einzige Abfrage.
Dass jeder Platz höchstens einmal vorkommt, ist dann schon dadurch gesichert, dass ein Schlüssel nur einmal in einem Wörterbuch stehen kann.
Zum Beispiel ist `{'Bäckerei': 'Anna', 'Bank': 'Cem'}` eine Zuordnung mit zwei Paaren.

**Ein Gegenbeispiel genügt, um das Verfahren zu widerlegen.** Aber wie oft ist es wirklich schlechter, und wie sicher kannst du bei einem **anderen** Verfahren sein, dass es immer stimmt?
Wie in Notebook 12 vergleichst du mit **Durchprobieren**: Für jeden Schüler wird einer seiner Plätze **oder keiner** gewählt (`None`), alle Kombinationen mit `itertools.product`,
und man zählt die größte gültige. Das geht nur für wenige Schüler, aber mit vielen **zufälligen** kleinen Beispielen lässt sich das Verfahren gut testen.

Zuordnung **weniger** Paare hat als die größte Zuordnung.


## 3. Verbessernde Wege

Wo ging das gierige Verfahren in Frage 1.1 schief? Ben wollte die Bäckerei, die schon **Anna** gehörte. Aber Anna hätte auch die **Bank** nehmen können. Die Bank gehörte
aber schon **Cem**. Cem hätte noch den Zoo, der **Dora** gehörte, und Dora hätte das **Kino** frei gehabt. Alle Schüler hätten **nacheinander ausweichen** können:

```
Ben  -- will -->  Bäckerei (hat Anna)
Anna -- weicht aus -->  Bank (hat Cem)
Cem  -- weicht aus -->  Zoo (hat Dora)
Dora -- weicht aus -->  Kino (frei!)
```

So eine Kette nennt man einen **verbessernden Weg**: Er beginnt bei einem Schüler **ohne** Platz, geht abwechselnd über einen gewünschten, besetzten Platz zu dessen Besitzer und
endet bei einem **freien** Platz. Wenn man den Weg „umlegt“ (jeder auf der Kette nimmt den Platz weiter vorn), bekommt **ein Schüler mehr** einen Platz und alle anderen behalten einen.

**Das Verfahren (nach Kuhn):** Nimm die Schüler einer nach dem anderen vor. Für den Schüler `s` rufe `versuche(s)` auf:

1. Gehe die Plätze von `s` durch. Hat er einen, den er in **diesem Versuch noch nicht angesehen** hat:
2. Merke ihn dir als **besucht**.
3. Ist der Platz **frei**, nimm ihn. Ist er **besetzt**, frage den Besitzer: `versuche(besitzer)` (**rekursiv**): Kann der Besitzer einen **anderen** Platz bekommen?
   Wenn ja, ist der Platz jetzt frei: nimm ihn.
4. Klappt kein Platz, gib `False` zurück.

Die Menge der **besuchten Plätze** gehört zu **einem** Versuch (für einen Schüler `s` vom Anfang bis zum Ende, über alle Rekursionen hinweg). Sie sorgt dafür, dass man in der Kette nicht im Kreis läuft.

## 4. Warum das stimmt und was es kostet

**Warum bleibt die Zuordnung immer gültig?** Geändert wird `besitzer` nur dann, wenn ein Versuch **erfolgreich** endet, und zwar Schritt für Schritt auf dem Rückweg der Rekursion:
Zuerst nimmt der **letzte** Schüler der Kette den freien Platz, dann nimmt der davor den Platz, den der letzte gerade **verlassen** hat, und so weiter. Jeder Schüler der Kette gibt
also einen Platz ab (der Eintrag wird von seinem Nachfolger überschrieben) und bekommt dafür **genau einen** neuen. Kein Platz wird doppelt vergeben, kein Schüler hat zwei Plätze, und jeder bekommt
einen **Wunschplatz**. Nur der **erste** Schüler der Kette hatte vorher keinen und hat jetzt einen. Die Zahl der Paare wächst also bei jedem erfolgreichen Versuch um **genau 1**, bei einem Fehlschlag ändert sich nichts.

## Wenn Paare Kosten oder Rang haben

Das Verfahren zählt nur **Paare**: Alle erlaubten Paare sind gleich gut, und gesucht ist die **größte** Zuordnung. Viele Aufgaben wollen mehr:

- Jedes Paar hat **Kosten** oder **Punkte** (Entfernung, Zufriedenheit), und man will die Summe möglichst klein oder groß machen. Dann reicht „möglichst viele Paare“ nicht. Man sucht den besten
  verbessernden Weg (den mit dem größten Gewinn) oder benutzt die **Ungarische Methode**. Das Lösungsnotebook zu Wichteln zeigt eine Variante davon.
- Wünsche haben einen **Rang** (erster, zweiter, dritter Wunsch), und ein erster Wunsch zählt mehr als beliebig viele zweite. Man vergibt dann Punkte, die diese Rangfolge ausdrücken (zum Beispiel
  `n² + n + 1` für den ersten Wunsch, `n + 1` für den zweiten, `1` für den dritten bei `n` Personen) und sucht die Zuordnung mit den meisten Punkten.

Das ist **nicht** dasselbe wie die größtmögliche Zahl von Paaren, wie du gleich in der ersten Wettbewerbs-Ecke siehst. Schon die Frage „Geht es überhaupt?“ (Gibt es eine Zuordnung, in der **alle**
einen Platz bekommen?) ist ein einfaches Matching: Ist `len(besitzer)` gleich der Zahl der Schüler?

## Wettbewerbs-Ecke 1: Wichteln

*Angelehnt an Aufgabe 5 „Wichteln“ der 1. Runde des 39. Bundeswettbewerbs Informatik (2020/21).*

In einer Klasse bringt **jedes Kind genau ein Geschenk** mit. Die Geschenke und die Kinder sind von 1 bis `n` nummeriert (Kind `i` bringt Geschenk `i` mit). Jedes Kind schreibt auf, welche
**drei** Geschenke ihm am besten gefallen. Die Datei `daten/wichtelnN.txt` (N = 1, 3, 4, 5, 6) enthält in der ersten Zeile `n` und dann je Kind eine Zeile mit den drei Wünschen.

Die echte Aufgabe will eine Verteilung mit **möglichst vielen erfüllten ersten Wünschen**, bei Gleichstand mit möglichst vielen zweiten und so weiter. Wir lösen heute die **einfachere** Frage: Wie viele Kinder
können ein Geschenk bekommen, das **irgendwo auf ihrer Liste** steht, egal auf welchem Rang? Das ist genau das Matching aus diesem Notebook.

#### Aufgabe E.1

Das Gerüst enthält `lies_wuensche`, `zuordnen` und `greedy_zuordnung` (wie in diesem Notebook). Schreibe `untersuche(dateiname)`: Sie gibt ein Tupel zurück mit der Zahl der Kinder, der Zahl der
**verschiedenen ersten Wünsche** (das ist die größte Zahl von Kindern, die ihren **ersten** Wunsch bekommen können, denn jedes Geschenk geht nur einmal weg und Kinder mit
demselben Erstwunsch müssen sich streiten), der Paare des gierigen Verfahrens und der Paare von `zuordnen`.

````python
def lies_wuensche(dateiname):
    """Wörterbuch Kind -> Liste der drei Wunschgeschenke (alles Zahlen von 1 bis n)."""
    with open(dateiname, encoding='utf-8') as datei:
        zahlen = [int(z) for z in datei.read().split()]
    n = zahlen[0]
    wuensche = {}
    for kind in range(1, n + 1):
        wuensche[kind] = zahlen[3 * kind - 2: 3 * kind + 1]
    return wuensche


def versuche(schueler, wuensche, besitzer, besucht, zaehler):
    for platz in wuensche[schueler]:
        zaehler[0] = zaehler[0] + 1
        if platz not in besucht:
            besucht.add(platz)
            if platz not in besitzer or versuche(besitzer[platz], wuensche, besitzer, besucht, zaehler):
                besitzer[platz] = schueler
                return True
    return False

def zuordnen(wuensche):
    besitzer = {}
    zaehler = [0]
    for kind in wuensche:
        versuche(kind, wuensche, besitzer, set(), zaehler)
    return besitzer, zaehler[0]

def greedy_zuordnung(wuensche):
    besitzer = {}
    for kind in wuensche:
        for geschenk in wuensche[kind]:
            if geschenk not in besitzer:
                besitzer[geschenk] = kind
                break
    return besitzer

def untersuche(dateiname):
    wuensche = lies_wuensche(dateiname)
    # dein Code
    return (len(wuensche), 0, 0, 0)

print('Datei  Kinder  verschiedene Erstwünsche  Greedy  Matching')
for nummer in [1, 3, 4, 5, 6]:
    zeile = untersuche('daten/wichteln%d.txt' % nummer)
    print('%5d %7d %24d %7d %9d' % ((nummer,) + zeile))
````

## Wettbewerbs-Ecke 2: Wörter aufräumen

*Angelehnt an Aufgabe 1 „Wörter aufräumen“ der 1. Runde des 39. Bundeswettbewerbs Informatik (2020/21).*

Ein Lückentext hat Wörter, bei denen nur einige Buchstaben bekannt sind (`_` = fehlender Buchstabe), und eine **durcheinandergewürfelte Wortliste**. Zum Beispiel:

```
_h __, _a_ __r ___e __b___!
arbeit eine für je oh was
```

Jedes Wort passt in manche Lücken (gleiche **Länge**, und wo ein Buchstabe vorgegeben ist, muss er stimmen), und jedes Wort darf nur **einmal** benutzt werden. Das ist ein **Matching**
zwischen Lücken und Wörtern! Die Dateien `daten/raetsel0.txt` bis `raetsel4.txt` sind die Beispiele (erste Zeile: Lückentext, zweite Zeile: Wörter). Ein **Satzzeichen am Ende** einer Lücke (`,` `.` `!`)
gehört nicht zum Wort: Das Gerüst schneidet es ab (`zerlege`) und merkt es sich.

#### Aufgabe E.3

Die Lücken sind die „Schüler“, die Wörter die „Plätze“: `finde_kandidaten` liefert genau die Wunschlisten, `zuordnen` ist unverändert dieselbe Funktion wie vorher. Schreibe `loese_raetsel(dateiname)`:
Sie liest das Rätsel, bestimmt die Kandidaten, ordnet zu und gibt den fertigen Text zurück: für jede Lücke (in der Reihenfolge des Textes) das zugeordnete Wort plus das abgeschnittene Satzzeichen-Ende,
durch Leerzeichen verbunden. Die Zuordnung ist `besitzer[wortnummer] = lueckennummer`: Du musst sie also **umdrehen**: `wort_der_luecke[lueckennummer] = wortnummer`.
Die Ausgabe zeigt jeweils die ersten 70 Zeichen.

**Erwartete Ausgabe:**
```
oh je, was für eine arbeit!
Am Anfang wurde das Universum erschaffen. Das machte viele Leute sehr 
Als Gregor Samsa eines Morgens aus unruhigen Träumen erwachte, fand er
Informatik ist die Wissenschaft von der systematischen Darstellung, Sp
Opa Jürgen blättert in einer Zeitschrift aus der Apotheke und findet e
```

````python
def lies_raetsel(dateiname):
    with open(dateiname, encoding='utf-8') as datei:
        zeilen = datei.read().splitlines()
    return zeilen[0].split(), zeilen[1].split()

def zerlege(luecke):
    """Muster ohne Satzzeichen am Ende, und das Satzzeichen-Ende."""
    ende = len(luecke)
    while ende > 0 and luecke[ende - 1] != '_' and not luecke[ende - 1].isalpha():
        ende = ende - 1
    return luecke[:ende], luecke[ende:]

def passt(muster, wort):
    if len(muster) != len(wort):
        return False
    for i in range(len(wort)):
        if muster[i] != '_' and muster[i] != wort[i]:
            return False
    return True

def finde_kandidaten(luecken, woerter):
    """Wörterbuch: Nummer der Lücke -> Liste der Nummern passender Wörter."""
    kandidaten = {}
    for i in range(len(luecken)):
        muster = zerlege(luecken[i])[0]
        kandidaten[i] = []
        for j in range(len(woerter)):
            if passt(muster, woerter[j]):
                kandidaten[i].append(j)
    return kandidaten


def versuche(schueler, wuensche, besitzer, besucht, zaehler):
    for platz in wuensche[schueler]:
        zaehler[0] = zaehler[0] + 1
        if platz not in besucht:
            besucht.add(platz)
            if platz not in besitzer or versuche(besitzer[platz], wuensche, besitzer, besucht, zaehler):
                besitzer[platz] = schueler
                return True
    return False

def zuordnen(wuensche):
    besitzer = {}
    zaehler = [0]
    for schueler in wuensche:
        versuche(schueler, wuensche, besitzer, set(), zaehler)
    return besitzer, zaehler[0]

def loese_raetsel(dateiname):
    luecken, woerter = lies_raetsel(dateiname)
    kandidaten = finde_kandidaten(luecken, woerter)
    besitzer, geprueft = zuordnen(kandidaten)
    # dein Code
    return ''

for nummer in range(5):
    text = loese_raetsel('daten/raetsel%d.txt' % nummer)
    print(text[:70])
````

<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

1. Das Backtracking probiert eine Lücke nach der anderen und merkt erst viel später, dass eine frühe Wahl falsch war, und geht dann **Schritt für Schritt** zurück (es verrennt sich). Beim verbessernden Weg
   geschieht das Umlegen **gezielt**: Nur die Kette der betroffenen Lücken wird geändert, und jede Lücke wird in einem Versuch höchstens einmal gefragt. Es gibt eine **Garantie** O(Lücken · Kanten),
   Backtracking hat sie nicht.
2. Die Eindeutigkeit der Lösung garantiert nur, dass es **eine** Lösung gibt, die das Programm ausgeben darf. Gefunden werden muss sie trotzdem, und die Kandidaten sind oft mehrdeutig
   (`je` und `oh` passen beide zu `__,`). Das Matching liefert **irgendeine** vollständige Zuordnung (jede Lücke hat ein Wort, jedes Wort einmal), und bei einer einzigen Lösung ist das **die** Lösung.
   Gäbe es mehrere, bekäme man nur die erste, die das Verfahren findet.

```python
# Zuordnung als Wörterbuch: Platz -> Schüler (damit man „wem gehört der Platz?“ sofort sieht)
besitzer = {}

def versuche(schueler, wuensche, besitzer, besucht):
    for platz in wuensche[schueler]:
        if platz not in besucht:
            besucht.add(platz)                       # pro Versuch jeder Platz nur einmal
            if platz not in besitzer or versuche(besitzer[platz], wuensche, besitzer, besucht):
                besitzer[platz] = schueler           # frei, oder der Besitzer ist ausgewichen
                return True
    return False

for schueler in wuensche:                            # nacheinander, jedes Mal ein NEUES besucht
    versuche(schueler, wuensche, besitzer, set())
```

- Ein **Matching** ordnet Elemente zweier Gruppen einander zu, jedes höchstens einmal, nur über **erlaubte** Paare. Als Graph: Kanten zwischen zwei Reihen von Knoten.
- Das **gierige** Zuordnen (erster freier Platz) liefert eine **gültige**, aber oft **nicht die größte** Zuordnung. Den Beweis liefert ein kleines Gegenbeispiel (Ben und die Bäckerei), die Häufigkeit
  der Vergleich mit Durchprobieren.
- Ein **verbessernder Weg** beginnt bei einem Schüler ohne Platz, führt über besetzte Plätze zu deren Besitzern und endet bei einem freien Platz. Wird er umgelegt, gewinnt die Zuordnung **genau ein Paar**.
- Die **Rekursion** („der Besitzer soll ausweichen“) braucht eine **Menge der besuchten Plätze** pro Versuch: Ohne sie läuft die Suche endlos, zu selten neu angelegt, werden Wege gesperrt.
- Ein gescheiterter Versuch ändert nichts, und wer einmal keinen Platz findet, findet später auch keinen. Die Laufzeit ist **O(Schüler · Kanten)**, in der Praxis viel kleiner (Zähler messen!).
- Soll nicht nur die **Zahl der Paare**, sondern die **Summe von Punkten oder Kosten** optimal sein, braucht man gewichtete Verfahren. Und: **Mehr Paare** ist nicht dasselbe wie **bessere Paare** (Wichteln).
