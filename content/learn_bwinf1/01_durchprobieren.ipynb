## 1. `product`: alle Kombinationen mit Wiederholung

Ein Zahlenschloss hat drei Ringe mit den Ziffern 0 bis 9. Mit drei Schleifen ineinander gehst du alle Codes durch. Mit
`product` aus `itertools` geht es in einer Zeile:

- `product(werte, repeat=3)` liefert alle Möglichkeiten, dreimal aus `werte` zu wählen, **mit Wiederholung**.
- Jede Möglichkeit ist ein **Tupel**, zum Beispiel `(0, 4, 7)`. Ein Tupel ist wie eine Liste, die man nicht mehr ändern
  kann. Du kannst es mit `for` durchlaufen, mit `[0]` auf Stellen zugreifen und mit `sum` addieren.
- Die Reihenfolge ist wie beim Zählen: Die **letzte** Stelle ändert sich am schnellsten.

`product` liefert die Möglichkeiten nacheinander, nicht als fertige Liste. Willst du sie alle sehen, schreibst du
`list(product(...))`.

````python
from itertools import product

print(list(product('AB', repeat=2)))

codes = 0
for code in product(range(10), repeat=3):
    codes = codes + 1
print(codes, 'Codes')
````

## 2. `permutations`: alle Reihenfolgen

Fünf Freunde wollen sich in einer Reihe aufstellen. Wie viele Reihenfolgen gibt es? Hier darf sich niemand wiederholen,
jeder kommt genau einmal vor. Das ist eine **Permutation** (Umordnung).

- `permutations(liste)` liefert alle Reihenfolgen der Elemente von `liste`.
- `permutations(liste, k)` liefert alle Reihenfolgen von `k` **verschiedenen** Elementen aus `liste`. So ordnest du zum
  Beispiel drei Buchstaben drei verschiedene Ziffern zu.

````python
from itertools import permutations

for p in permutations('ABC'):
    print(p)
print(len(list(permutations('ABCDE'))), 'Reihenfolgen von 5 Personen')
print(len(list(permutations(range(10), 3))), 'Möglichkeiten, 3 Buchstaben verschiedene Ziffern zu geben')
````

## 4. Wie viele Möglichkeiten sind es?

Bevor du ein Programm alle Möglichkeiten durchprobieren lässt, solltest du ausrechnen, **wie viele** es sind. Für die drei
Werkzeuge gibt es Formeln. `n` ist die Zahl der Werte, `k` die Zahl der Wahlen:

| Werkzeug | Anzahl | in Python | Beispiel n = 10, k = 4 |
|---|---|---|---|
| `product(werte, repeat=k)` | n · n · … · n = nᵏ | `n ** k` | 10 000 |
| `permutations(werte, k)` | n · (n − 1) · … · (n − k + 1) | `math.perm(n, k)` | 5 040 |
| `permutations(werte)` | n · (n − 1) · … · 1 = n! (sprich: n Fakultät) | `math.factorial(n)` | 3 628 800 |
| `combinations(werte, k)` | n! / (k! · (n − k)!) (Binomialkoeffizient) | `math.comb(n, k)` | 210 |

Warum? Bei `permutations` hat die erste Wahl n Möglichkeiten, die zweite nur noch n − 1 (ein Wert ist schon vergeben) und so
weiter. Bei `combinations` zählt man zuerst wie bei `permutations` und teilt dann durch die k! Reihenfolgen, in denen dieselbe
Auswahl vorkommt.

**Faustregel für Python im Browser:** Einige hunderttausend Möglichkeiten sind in Sekunden geprüft (je nachdem, wie
aufwendig die Prüfung ist). Bei vielen Millionen wartest du Minuten, bei Milliarden Stunden.

## 5. Die O-Notation: Wie wächst die Rechenzeit?

Die genaue Rechenzeit hängt vom Computer ab. Wichtiger ist die Frage: **Wie viel länger dauert es, wenn die Eingabe größer
wird?** Das beschreibt die O-Notation. Man zählt die Schritte (zum Beispiel geprüfte Möglichkeiten) in Abhängigkeit von der
Größe `n` der Eingabe und schreibt nur auf, wie schnell diese Zahl **wächst**:

| Schreibweise | sprich | Beispiel | n verdoppeln … |
|---|---|---|---|
| O(n) | „O von n“, linear | jedes Element einmal ansehen | doppelt so lange |
| O(n log n) | „n log n“ | gut sortieren (`sorted`) | etwas mehr als doppelt so lange |
| O(n²) | quadratisch | alle Paare | viermal so lange |
| O(n³) | kubisch | alle Dreiergruppen | achtmal so lange |
| O(2ⁿ) | exponentiell | alle Teilmengen | n um **1** erhöhen: doppelt so lange |
| O(n!) | „n Fakultät“ | alle Reihenfolgen | n um 1 erhöhen: (n + 1)-mal so lange |

Dabei lässt man **konstante Faktoren** und **kleinere Teile** weg. Die Zahl der Paare ist n · (n − 1) / 2 = n²/2 − n/2.
Für große n zählt nur das n², also ist das O(n²). Für n = 100, 200 und 400 sind es 4 950, 19 900 und
79 800 Paare: Bei doppeltem n etwa viermal so viele.

So unterschiedlich wachsen die Zahlen:

| n | n² | 2ⁿ | n! |
|---|---|---|---|
| 10 | 100 | 1 024 | 3 628 800 |
| 20 | 400 | 1 048 576 | 2 432 902 008 176 640 000 |
| 30 | 900 | 1 073 741 824 | etwa 2,7 · 10³² |

Alle Teilmengen gehen also bis etwa n = 20, alle Reihenfolgen nur bis etwa n = 10. Mit O(n²) schafft man dagegen auch
n = 1000 noch gut.

from itertools import product

def weiss(a, b):
    return not (a and b)

def rot(a):
    return not a

def leds(quellen):
    l1 = weiss(weiss(quellen[0], quellen[1]), rot(quellen[2]))
    return (l1,)

def an_aus(wert):
    if wert:
        return 'An '
    return 'Aus'

def wahrheitstabelle(n):
    for quellen in product([False, True], repeat=n):
        zeile = ''
        for q in quellen:
            zeile = zeile + an_aus(q) + ' '
        zeile = zeile + '|'
        for l in leds(quellen):
            zeile = zeile + ' ' + an_aus(l)
        print(zeile)

## Das hast du gelernt

| Frage | Werkzeug | Anzahl |
|---|---|---|
| k-mal wählen, Wiederholung erlaubt | `product(werte, repeat=k)` | nᵏ |
| aus verschiedenen Listen je eins wählen | `product(liste1, liste2, …)` | Produkt der Längen |
| alle Reihenfolgen | `permutations(werte)` | n! |
| k verschiedene, Reihenfolge zählt | `permutations(werte, k)` | `math.perm(n, k)` |
| k verschiedene, Reihenfolge egal | `combinations(werte, k)` | `math.comb(n, k)` |
| alle Teilmengen | `product([False, True], repeat=n)` oder `combinations` für jedes k | 2ⁿ |

- Die Möglichkeiten kommen als **Tupel** und nur **einmal** nacheinander. Für mehrere Durchläufe `list(...)` oder neu aufrufen.
- **Erst rechnen, dann probieren:** Wie viele Möglichkeiten sind es? Einige hunderttausend gehen in Sekunden, viele Millionen
  dauern Minuten.
- **O-Notation:** beschreibt, wie die Zahl der Schritte mit der Eingabegröße n wächst: O(n), O(n²), O(2ⁿ), O(n!).
  Konstante Faktoren fallen weg. In die Dokumentation gehört: was gezählt wird, die O-Angabe und was das für die Beispiele
  bedeutet.
- **Doppelzählungen** vermeiden und Anzahlen an kleinen Beispielen von Hand prüfen.
