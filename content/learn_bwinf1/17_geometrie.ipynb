## 2. Kreuzprodukt und Orientierung: links, rechts oder auf der Geraden?

Das **Kreuzprodukt** (in der Ebene eine einzelne Zahl) ist

`v × w = vx · wy − vy · wx`

Es ist die Grundlage fast aller Geometrie-Aufgaben. Drei Punkte `a`, `b`, `c` führen zur **Orientierung**: Man geht von `a` nach `b` und fragt, wo `c` liegt. Dazu bildet man
`(b − a) × (c − a)`:

- **positiv**: `c` liegt **links** von der Richtung `a → b` (dreht man den Vektor `b − a` gegen den Uhrzeigersinn, kommt man zu `c − a`; die `y`-Achse zeigt nach oben),
- **negativ**: `c` liegt **rechts**,
- **0**: `a`, `b`, `c` liegen **auf einer Geraden**.

Der **Betrag** des Kreuzprodukts ist die **doppelte Fläche** des Dreiecks `a b c`.

### Kommazahlen und die Toleranz

Mit ganzen Zahlen ist das Kreuzprodukt **exakt**. Viele Aufgaben haben aber Kommazahlen als Koordinaten, und Kommazahlen (Typ `float`) sind im Rechner **gerundet**: Der Rechner speichert sie binär, und schon `0,1` lässt
sich dort nicht genau darstellen. Das zeigt sich so:

````python
import math

def minus(p, q):
    return (p[0] - q[0], p[1] - q[1])

def laenge(v):
    return math.hypot(v[0], v[1])

def skalar(v, w):
    return v[0] * w[0] + v[1] * w[1]

def kreuz(v, w):
    return v[0] * w[1] - v[1] * w[0]

print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)

a = (0.1, 0.3)
b = (0.2, 0.6)
c = (0.3, 0.9)                       # alle drei liegen auf der Geraden y = 3x
print(kreuz(minus(b, a), minus(c, a)))
````

`0.1 + 0.2` ist **nicht** genau `0.3`, und das Kreuzprodukt von drei Punkten, die **genau** auf einer Geraden liegen, ist nicht 0, sondern `2.08e-17`.
Die Schreibweise `e-17` heißt „mal 10 hoch −17“: Die Zahl ist `0,0000000000000000208`, praktisch null, aber **nicht gleich** null. Ein Test `kreuz(...) == 0` würde hier sagen: „Die Punkte liegen **nicht** auf einer Geraden.“
(Das Kreuzprodukt dieser Zahlen ist schon eine Rechnung mit Rundungsfehlern, wie in [Notebook 16](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/16_heuristiken.ipynb) beim Vergleich der Bewertungen.)

Die Lösung ist eine **Toleranz**: Man behandelt alles, was **sehr nahe an 0** liegt, als 0. Im Code schreibt man dafür `EPS = 1e-9`. Das ist die Kurzschreibweise für `1 · 10^-9 = 0,000000001`, ein Milliardstel.
Zwei Kommazahlen `x` und `y` sind dann „gleich“, wenn `abs(x - y) < EPS` ist.

## 4. Der Schnittpunkt zweier Geraden

Wenn sich zwei Strecken schneiden, will man oft **wo**. Eine Gerade durch `a` und `b` hat die Punkte `a + t · r` mit dem **Richtungsvektor** `r = b − a` (für `t` zwischen 0 und 1 liegt der Punkt auf der **Strecke**).
Die zweite Gerade hat die Punkte `c + u · s` mit `s = d − c`. Am Schnittpunkt gilt `a + t · r = c + u · s`. Bildet man auf beiden Seiten das Kreuzprodukt mit `s` (und es ist `s × s = 0`), bleibt

`t · (r × s) = (c − a) × s`, also `t = ((c − a) × s) / (r × s)`.

Der Schnittpunkt ist dann `a + t · r`. Ist `r × s = 0`, sind die Geraden **parallel** (oder gleich), und man darf **nicht** dividieren.

## 5. Abstand eines Punktes zu einer Strecke

Wie weit ist eine Hütte `p` von einem Wanderweg (der Strecke von `a` nach `b`) entfernt? Wir suchen den Punkt der Strecke, der `p` am **nächsten** liegt. Die Idee: **Lot fällen**.

Die Punkte der Geraden sind `a + t · r` mit `r = b − a`. Das `t` des **Lotfußpunkts** (Fußpunkt der senkrechten Linie von `p` auf die Gerade) ist

`t = ((p − a) · r) / (r · r)`

(Skalarprodukt statt Kreuzprodukt!). Für die **Strecke** muss `t` zwischen 0 und 1 liegen. Ist `t < 0`, ist `a` der nächste Punkt, ist `t > 1`, ist es `b`. Man **begrenzt** `t` auf das Intervall `[0, 1]`:
`t = max(0, min(1, t))`. Der Abstand ist dann die Länge des Vektors von `p` zum nächsten Punkt. Ist `a == b` (Strecke der Länge 0), ist `r · r = 0`, und man darf nicht dividieren: dann ist `a` der nächste Punkt.

Eine Zeichnung besteht aus Strecken. Wie viele **Dreiecke** sind darin zu sehen? Ein Dreieck sind **drei Strecken**, die sich **paarweise** schneiden, und zwar in **drei verschiedenen** Punkten. Fallen die drei Schnittpunkte zusammen
(drei Strecken durch **einen** Punkt, wie die Strahlen eines Sterns), gibt es kein Dreieck. Wir nehmen wie in der Aufgabe an, dass keine zwei Strecken auf einer Geraden liegen.

Die Dateien `daten/dreiecke1.txt`, `dreiecke3.txt`, `dreiecke4.txt`, `dreiecke5.txt` und `dreiecke6.txt` (die Beispiele 1 und 3 bis 6 der Aufgabe) haben in der 1. Zeile die Zahl der Strecken, dann je Zeile eine Strecke `x1 y1 x2 y2`.
Gegeben sind `lies_strecken` und `schnitt(s, t)`, das den Schnittpunkt zweier Strecken (Tupel `(x1, y1, x2, y2)`) oder `None` zurückgibt. Wegen der Kommazahlen gibt es auch `gleich(p, q)` mit Toleranz.

#### Aufgabe E.1

und zählt das Tripel, wenn alle drei existieren und **paarweise nicht `gleich`** sind. Sie gibt `(anzahl, tripel)` zurück (`tripel` zählt, wie viele Tripel angesehen wurden).

```
dreiecke6.txt: 20 Strecken, 20 Dreiecke, 1140 Tripel geprüft
```

````python
import math

def minus(p, q):
    return (p[0] - q[0], p[1] - q[1])

def laenge(v):
    return math.hypot(v[0], v[1])

def skalar(v, w):
    return v[0] * w[0] + v[1] * w[1]

def kreuz(v, w):
    return v[0] * w[1] - v[1] * w[0]

EPS = 1e-9                           # 0,000000001: kleiner als das gilt als 0

def vorzeichen(x):
    if x > EPS:
        return 1
    if x < -EPS:
        return -1
    return 0

def orientierung(a, b, c):
    return vorzeichen(kreuz(minus(b, a), minus(c, a)))

def im_rechteck(a, b, p):
    """True, wenn p im Rechteck liegt, das a und b aufspannen (für Punkte auf der Geraden ab: p liegt auf der Strecke)."""
    return min(a[0], b[0]) <= p[0] <= max(a[0], b[0]) and min(a[1], b[1]) <= p[1] <= max(a[1], b[1])

def schneiden(a, b, c, d):
    o1 = orientierung(a, b, c)
    o2 = orientierung(a, b, d)
    o3 = orientierung(c, d, a)
    o4 = orientierung(c, d, b)
    if o1 * o2 < 0 and o3 * o4 < 0:
        return True
    if o1 == 0 and im_rechteck(a, b, c):                    # c liegt auf der Strecke ab
        return True
    if o2 == 0 and im_rechteck(a, b, d):                    # d liegt auf der Strecke ab
        return True
    if o3 == 0 and im_rechteck(c, d, a):                    # a liegt auf der Strecke cd
        return True
    if o4 == 0 and im_rechteck(c, d, b):                    # b liegt auf der Strecke cd
        return True
    return False

def schnittpunkt_geraden(a, b, c, d):
    r = minus(b, a)
    s = minus(d, c)
    nenner = kreuz(r, s)
    if abs(nenner) < EPS:                         # parallel (oder dieselbe Gerade): kein einzelner Schnittpunkt
        return None
    t = kreuz(minus(c, a), s) / nenner
    return (a[0] + t * r[0], a[1] + t * r[1])

def lies_strecken(dateiname):
    with open(dateiname, encoding='utf-8') as datei:
        zeilen = datei.read().split('\n')
    anzahl = int(zeilen[0])
    return [tuple(float(z) for z in zeile.split()) for zeile in zeilen[1:anzahl + 1]]

def schnitt(s, t):
    """Schnittpunkt der Strecken s und t, oder None."""
    a, b = (s[0], s[1]), (s[2], s[3])
    c, d = (t[0], t[1]), (t[2], t[3])
    if not schneiden(a, b, c, d):
        return None
    return schnittpunkt_geraden(a, b, c, d)

def gleich(p, q):
    return abs(p[0] - q[0]) < 1e-6 and abs(p[1] - q[1]) < 1e-6

def zaehle_dreiecke(strecken):
    anzahl = 0
    tripel = 0
    # dein Code
    return anzahl, tripel

for nummer in [1, 3, 4, 5, 6]:
    strecken = lies_strecken('daten/dreiecke' + str(nummer) + '.txt')
    anzahl, tripel = zaehle_dreiecke(strecken)
    print('dreiecke' + str(nummer) + '.txt:', len(strecken), 'Strecken,', anzahl, 'Dreiecke,', tripel, 'Tripel geprüft')
````

## Für Schnelle: die Idee von Krocket

*Angelehnt an Aufgabe 4 „Krocket“ der 1. Runde des 43. Bundeswettbewerbs Informatik (2024/25).*

Laura will einen Ball mit **einem Schlag** durch mehrere Tore (Strecken) schießen. Wir vereinfachen: Der Ball ist ein **Punkt** (Radius 0), und der Schlag ist eine **Gerade** durch zwei Punkte `p` und `q`.
Die Gerade geht durch ein Tor, wenn die beiden Endpunkte des Tores **nicht auf derselben Seite** liegen, also wenn `orientierung(p, q, t1) * orientierung(p, q, t2) <= 0` ist (bei 0 liegt ein Endpunkt auf der Geraden: Tor getroffen am Pfosten).
Die **Reihenfolge** der Tore prüfen wir hier nicht.

````python
import math

def minus(p, q):
    return (p[0] - q[0], p[1] - q[1])

def laenge(v):
    return math.hypot(v[0], v[1])

def skalar(v, w):
    return v[0] * w[0] + v[1] * w[1]

def kreuz(v, w):
    return v[0] * w[1] - v[1] * w[0]

EPS = 1e-9                           # 0,000000001: kleiner als das gilt als 0

def vorzeichen(x):
    if x > EPS:
        return 1
    if x < -EPS:
        return -1
    return 0

def orientierung(a, b, c):
    return vorzeichen(kreuz(minus(b, a), minus(c, a)))

def geht_durch_alle(p, q, tore):
    for t1, t2 in tore:
        if orientierung(p, q, t1) * orientierung(p, q, t2) > 0:      # beide Pfosten auf derselben Seite
            return False
    return True

tore = [((2, 1), (2, 5)), ((5, 0), (5, 4)), ((8, 2), (8, 6)), ((11, 3), (11, 7))]
print(geht_durch_alle((0, 2), (12, 6), tore))
print(geht_durch_alle((0, 0), (12, 0), tore))
print(geht_durch_alle((0, 2), (12, 0), tore))
````

Die erste Gerade von `(0, 2)` nach `(12, 6)` geht durch alle vier Tore (bei `x = 2` ist sie bei `y ≈ 2,67`, bei `x = 5` bei `y ≈ 3,67`, bei `x = 8` bei `y ≈ 4,67`, bei `x = 11` bei `y ≈ 5,67`, jeweils zwischen den Pfosten).
Die zweite (die `x`-Achse) verfehlt schon das erste Tor bei `x = 2` (`y` von 1 bis 5). Die dritte Gerade `(0, 2) → (12, 0)` fällt nach rechts ab und verfehlt das dritte Tor: Bei `x = 8` ist sie bei `y ≈ 0,67`, das Tor
beginnt erst bei `y = 2`. Rechne die anderen Werte nach (die Gerade durch `(0, 2)` und `(12, 6)` ist `y = 2 + x / 3`).
