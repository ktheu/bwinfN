## 3. Die Idee begründen

In die Dokumentation gehört nicht nur **was** du machst, sondern **warum es stimmt**. Bei Ferienhäusern ist die Idee:

> Zähle **für jeden Tag**, wie viele Buchungen ihn belegen. Die **größte** Zahl ist die Antwort.

#### Frage 3.1

Begründe, dass die **größte Tageszahl** die richtige Antwort ist. Du brauchst **zwei** Teile: Warum reichen weniger Häuser **nicht**, und warum reicht diese Zahl **aus**?

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

**Weniger reicht nicht** (untere Schranke): Am vollsten Tag belegen `m` Buchungen je ein Haus, und kein Haus kann zwei Buchungen gleichzeitig aufnehmen. Also braucht man mindestens `m` Häuser.

**Diese Zahl reicht** (man gibt ein Verfahren an, das mit `m` Häusern auskommt): Man sortiert die Buchungen nach dem Anreisetag und gibt jeder das **erste Haus, das bei ihrer Anreise frei ist**.
Ein **neues** Haus nimmt man nur, wenn **alle** bisherigen Häuser bei der Anreise belegt sind. Dann belegen an diesem Tag alle Häuser plus die neue Buchung ein Haus, also sind dort **so viele Buchungen** aktiv, wie Häuser gebraucht werden.
Damit ist die Zahl der benutzten Häuser höchstens das Maximum der Tageszahlen.

Beides zusammen: Es braucht mindestens `m` Häuser, und das Verfahren kommt mit `m` aus. Also ist `m` die **kleinste** Zahl.

</details>

## 4. Die Laufzeit begründen und abschätzen

Zu jeder Idee gehört die Frage: **Wie lange rechnet das Programm?** Dazu brauchst du die **O-Notation** (die Tabelle steht in [Notebook 1](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/01_durchprobieren.ipynb)) und ein Gefühl für die **größte Eingabe**.
Die Laufzeit in Sekunden gehört **nicht** in die Dokumentation: Sie hängt vom Rechner ab. Besser ist eine **Zählung**: Wie viele Schritte macht das Programm? Das hat unser Zähler `pruefungen` schon getan.

So begründest du die Laufzeit:

1. **Größen benennen:** `n` Buchungen, `T` Tage zwischen erster Anreise und letzter Abreise.
2. **Verschachtelte Schleifen malen:** Die äußere Schleife läuft über `T` Tage, die innere über `n` Buchungen. Das sind `T · n` Prüfungen, also **O(T · n)**.
3. **Mit der größten Eingabe einsetzen:** Bei 1 000 Buchungen in einem Jahr (365 Tage) sind es rund 365 000 Prüfungen. Das ist unkritisch. Bei einer Million Buchungen wäre es nicht mehr.
4. **Wenn es besser geht, sagen:** Man kann die Tage **überspringen**: Nur an den Anreise- und Abreisetagen ändert sich die Zahl der belegten Häuser. Sortiert man diese `2n` **Ereignisse** (`sorted`, **O(n log n)**), genügt ein Durchgang.

### B. Modell und Idee

**Modell.** Die Schülerinnen sind **Knoten** eines Graphen (wie in [Notebook 5](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/05_breitensuche.ipynb)). Ein „+“-Wunsch ist eine **Kante** (ungerichtet, wegen Lesart 1). Ein „−“-Wunsch ist **keine Kante**, sondern eine **Prüfung**:
Er verbietet nur, dass zwei Schülerinnen im selben Zimmer sind, und erzwingt nichts.

**Idee.** Alle Schülerinnen, die über „+“-Kanten verbunden sind (die **Zusammenhangskomponente**), müssen **in einem Zimmer** sein. Das sind die **Pflichtzimmer**. Dann prüft man, ob ein „−“-Wunsch **innerhalb** eines Pflichtzimmers liegt.
Gibt es so einen Wunsch, ist die Aufgabe **unlösbar**. Gibt es keinen, sind die Pflichtzimmer eine **gültige Belegung**.

#### Frage B.1

Begründe, dass das Verfahren **richtig** ist. Zeige zwei Dinge: (a) Wenn ein „−“-Wunsch innerhalb eines Pflichtzimmers liegt, gibt es **keine** gültige Belegung. (b) Wenn keiner so liegt, sind die Pflichtzimmer eine gültige Belegung.
Welches Prinzip ist das: **zwingende Folgerung**, **vollständige Suche** oder **Austauschargument**?

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

(a) Sind Anna und Lotta im selben Pflichtzimmer, gibt es eine **Kette** von „+“-Wünschen `Anna – … – Lotta`. In **jeder** gültigen Belegung sind je zwei Nachbarn der Kette im selben Zimmer, also auch Anna und Lotta am Ende der Kette.
Aber Anna oder Lotta will nicht mit der anderen zusammen sein, ein Widerspruch. Also gibt es keine gültige Belegung.

(b) Jeder „+“-Wunsch liegt **innerhalb** eines Pflichtzimmers, ist also erfüllt. Jeder „−“-Wunsch liegt nach Voraussetzung **zwischen** zwei Pflichtzimmern, ist also auch erfüllt.

Es ist eine **zwingende Folgerung**: Jede gültige Belegung muss die Pflichtzimmer **zusammenlassen** (sie sind die kleinsten möglichen Zimmer). Findet man in den kleinsten Zimmern einen Widerspruch, dann erst recht in größeren.
Das ist ähnlich wie bei „zwingenden Folgerungen“ in [Notebook 4](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/04_logisch_schliessen.ipynb). Das Ergebnis ist **vollständig**: Das Verfahren findet eine Belegung, **wenn** es eine gibt.

</details>

````python
def lies_wuensche(dateiname):
    """Gibt (namen, will, nicht, selbst) zurück: die Namen in Dateireihenfolge, je Name zwei Mengen von Namen
    und die Zahl der weggelassenen Selbstnennungen (eine Schülerin nennt sich selbst)."""
    with open(dateiname, encoding='utf-8') as datei:
        text = datei.read()
    namen = []
    will = {}
    nicht = {}
    selbst = 0
    for block in text.strip().split('\n\n'):
        zeilen = block.split('\n')
        name = zeilen[0].strip()
        namen.append(name)
        will[name] = set(zeilen[1].split()[1:])        # das erste Wort ist das Zeichen + bzw. -
        nicht[name] = set(zeilen[2].split()[1:])
        selbst = selbst + (name in will[name]) + (name in nicht[name])
        will[name].discard(name)                       # Selbstnennungen sind sinnlos
        nicht[name].discard(name)
    return namen, will, nicht, selbst

def zimmer_bilden(namen, will):
    """Gibt die Pflichtzimmer zurück (Liste von Listen von Namen), jede Schülerin in genau einem."""
    nachbarn = {}
    for n in namen:
        nachbarn[n] = set()
    for n in namen:
        for m in will[n]:
            nachbarn[n].add(m)                 # Lesart 1: Der Wunsch gilt für beide Seiten
            nachbarn[m].add(n)
    zimmer_liste = []
    gesehen = set()
    for start in namen:
        if start in gesehen:
            continue
        zimmer = [start]
        gesehen.add(start)
        i = 0
        while i < len(zimmer):                 # zimmer dient zugleich als Warteschlange
            n = zimmer[i]
            i = i + 1
            for m in sorted(nachbarn[n]):
                if m not in gesehen:
                    gesehen.add(m)
                    zimmer.append(m)
        zimmer_liste.append(zimmer)
    return zimmer_liste

def konflikte(zimmer_liste, nicht):
    """Gibt die Paare (a, b) mit a < b zurück, die im selben Zimmer sind, obwohl eine von beiden das nicht will."""
    nummer = {}
    for i in range(len(zimmer_liste)):
        for n in zimmer_liste[i]:
            nummer[n] = i
    paare = set()
    for a in nicht:
        for b in nicht[a]:
            if nummer[a] == nummer[b]:
                paare.add((min(a, b), max(a, b)))
    return sorted(paare)

def finde_kette(a, b, will):
    """Kürzeste Kette a - ... - b über '+'-Wünsche (Breitensuche mit Vorgängern, Notebook 5)."""
    nachbarn = {}
    for n in will:
        nachbarn[n] = set()
    for n in will:
        for m in will[n]:
            nachbarn[n].add(m)
            nachbarn[m].add(n)
    vorgaenger = {a: None}
    warteschlange = [a]
    i = 0
    while i < len(warteschlange):
        n = warteschlange[i]
        i = i + 1
        for m in sorted(nachbarn[n]):
            if m not in vorgaenger:
                vorgaenger[m] = n
                warteschlange.append(m)
    kette = [b]
    while vorgaenger[kette[-1]] is not None:
        kette.append(vorgaenger[kette[-1]])
    kette.reverse()
    return kette


def zimmerbelegung(dateiname):
    """Löst die Aufgabe für eine Datei und druckt das Ergebnis kurz und lesbar."""
    namen, will, nicht, selbst = lies_wuensche(dateiname)
    zimmer_liste = zimmer_bilden(namen, will)
    liste = konflikte(zimmer_liste, nicht)
    if len(liste) > 0:
        print('Nicht alle Wünsche können erfüllt werden:')
        for a, b in liste:
            print('  ', a, 'und', b, 'müssen zusammen sein (Kette: ' + ' - '.join(finde_kette(a, b, will)) + '), wollen aber nicht.')
        return None
    print('Alle Wünsche können erfüllt werden,', len(zimmer_liste), 'Zimmer:')
    einzel = []
    for zimmer in sorted(zimmer_liste, key=lambda z: (-len(z), sorted(z)[0])):
        if len(zimmer) == 1:
            einzel.append(zimmer[0])
        else:
            print('  ', len(zimmer), 'Schülerinnen:', ', '.join(sorted(zimmer)))
    if len(einzel) > 0:
        print('  ', len(einzel), 'Einzelzimmer:', ', '.join(sorted(einzel)))
    return zimmer_liste

for nummer in [1, 2]:
    print('--- zimmer%d.txt' % nummer)
    zimmerbelegung('daten/zimmer%d.txt' % nummer)
````

## Zimmerbelegung: Dokumentation

### 1. Annahmen und Lesarten

- **Ein Wunsch gilt für beide Seiten.** Wenn Anna mit Paula zusammen sein will, müssen beide im selben Zimmer sein, auch wenn Paula nichts dazu sagt. Ebenso dürfen Anna und Paula nicht zusammen sein, wenn Anna das nicht will.
  Grund: Der Aufgabentext nennt Annas einseitigen Wunsch als Konflikt, und fast alle Wünsche in den Beispielen sind einseitig (zimmer3: 41 von 41 „+“-Wünschen).
  *Mit der Lesart „nur gegenseitige Wünsche zählen“ wäre schon zimmer1 lösbar, das Ergebnis hängt also von dieser Wahl ab.*
- **Zimmer sind Gruppen.** Wer mit A zusammen sein will und A mit B, ist im selben Zimmer wie B (Zimmer sind keine Paare).
- **Nicht genannt heißt „egal“.**
- **Selbstnennungen** („Nele will mit Nele zusammen sein“) werden weggelassen.
- **Zimmergröße und -zahl** sind beliebig. Wir geben die Belegung mit den **meisten** Zimmern an (jede Schülerin sitzt nur dort, wo sie sitzen muss).

### 2. Lösungsidee

Die Schülerinnen sind **Knoten**, jeder „+“-Wunsch ist eine **Kante**. Alle Schülerinnen, die über Kanten verbunden sind, müssen in **einem** Zimmer sein (**Pflichtzimmer**). Ein „−“-Wunsch erzwingt nichts, er ist nur eine **Prüfung**.

*Begründung.* (a) Liegen zwei Schülerinnen im selben Pflichtzimmer und will eine nicht mit der anderen zusammen sein, gibt es keine gültige Belegung: Es gibt eine Kette von „+“-Wünschen zwischen ihnen, und in jeder gültigen Belegung sind je zwei Nachbarn der Kette im selben Zimmer.
(b) Gibt es keinen solchen Wunsch, sind die Pflichtzimmer eine gültige Belegung: Jeder „+“-Wunsch liegt in einem Pflichtzimmer, jeder „−“-Wunsch zwischen zwei Pflichtzimmern.

### 3. Umsetzung

Python. `lies_wuensche` liest die Datei und merkt für jede Schülerin zwei Mengen (`will`, `nicht`). `zimmer_bilden` trägt jeden „+“-Wunsch für beide Seiten in ein Wörterbuch ein und sammelt die Pflichtzimmer mit **Breitensuche (Flutfüllen)**.
`konflikte` gibt zu jedem Zimmer die Paare aus, die nicht zusammen sein wollen. Bei einem Konflikt zeigt `finde_kette` die Kette von Wünschen, die beide zusammenführt (ein **Beleg**, den man von Hand nachprüfen kann).

### 4. Laufzeit

Mit `n` Schülerinnen und `w` Wünschen ist die Laufzeit **O(n + w)** (bis auf den Faktor `log` für das Sortieren der Nachbarn). Die größte Beispieleingabe hat n = 43 und w = 150 (zimmer6), das sind wenige hundert Schritte.
Alle Paare zu vergleichen wäre O(n²) gewesen und ist unnötig.

```python
def zimmer_bilden(namen, will):
    """Gibt die Pflichtzimmer zurück (Liste von Listen von Namen), jede Schülerin in genau einem."""
    nachbarn = {}
    for n in namen:
        nachbarn[n] = set()
    for n in namen:
        for m in will[n]:
            nachbarn[n].add(m)                 # Lesart 1: Der Wunsch gilt für beide Seiten
            nachbarn[m].add(n)
    zimmer_liste = []
    gesehen = set()
    for start in namen:
        if start in gesehen:
            continue
        zimmer = [start]
        gesehen.add(start)
        i = 0
        while i < len(zimmer):                 # zimmer dient zugleich als Warteschlange
            n = zimmer[i]
            i = i + 1
            for m in sorted(nachbarn[n]):
                if m not in gesehen:
                    gesehen.add(m)
                    zimmer.append(m)
        zimmer_liste.append(zimmer)
    return zimmer_liste


def konflikte(zimmer_liste, nicht):
    """Gibt die Paare (a, b) mit a < b zurück, die im selben Zimmer sind, obwohl eine von beiden das nicht will."""
    nummer = {}
    for i in range(len(zimmer_liste)):
        for n in zimmer_liste[i]:
            nummer[n] = i
    paare = set()
    for a in nicht:
        for b in nicht[a]:
            if nummer[a] == nummer[b]:
                paare.add((min(a, b), max(a, b)))
    return sorted(paare)
```
