# 02. Backtracking

In [Notebook 01](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/01_durchprobieren.ipynb) hast du alle Möglichkeiten **fertig** erzeugt und erst dann
geprüft. Bei den Alphametiken hieß das: Alle Buchstaben bekommen eine Ziffer, dann wird gerechnet. Dabei prüft man viele
Belegungen, die schon nach den ersten Ziffern keine Chance mehr hatten.

**Backtracking** (Zurückverfolgen) macht es geschickter. Man baut eine Lösung **Schritt für Schritt** auf. Nach jedem
Schritt prüft man, ob das bisher Gebaute noch zu einer Lösung werden kann. Wenn nicht, nimmt man den letzten Schritt
**zurück** und probiert die nächste Möglichkeit. So fallen ganze Bündel von Möglichkeiten auf einmal weg.

Du brauchst dafür Rekursion aus [Notebook 22 von learn_jwinf3](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_jwinf3/22_rekursion.ipynb): eine Funktion, die
sich selbst aufruft.

In diesem Notebook lernst du:

- das Muster **setzen, weitersuchen, zurücknehmen**,
- **früh abbrechen**, sobald ein Teilstück nicht mehr zu einer Lösung werden kann,
- die **Aufrufe zu zählen**, um zu sehen (und in der Dokumentation zu belegen), wie viel das Abbrechen spart,
- **alle Lösungen** zu finden oder bei der **ersten** aufzuhören,
- die typischen Fehler: Zurücknehmen vergessen, Lösung ohne Kopie speichern.

In der Wettbewerbs-Ecke löst du Dreieckspuzzles und suchst Kassiopeias Weg durch alle Felder.

## 1. Setzen, weitersuchen, zurücknehmen

In learn_jwinf3 hast du alle Wörter so gebildet: `alle_woerter(anfang + b, ...)`. Jeder Aufruf bekommt einen **neuen** Text.
Beim Backtracking arbeitet man meist mit **einer einzigen Liste**, die wächst und schrumpft:

- **setzen:** `reihe.append(b)` hängt `b` hinten an,
- **weitersuchen:** die Funktion ruft sich selbst auf und baut auf der längeren Liste weiter,
- **zurücknehmen:** `reihe.pop()` entfernt das letzte Element wieder, danach ist die Liste wie vorher.

Das Zurücknehmen ist wichtig: Nach dem Aufruf muss die Liste wieder so aussehen wie vorher, damit die nächste Möglichkeit
auf demselben Stand aufbauen kann. Warum nicht einfach Kopien? Bei einem großen Spielbrett oder einer Menge besuchter
Felder wäre es teuer, bei jedem Schritt alles zu kopieren. Setzen und Zurücknehmen kostet dagegen nur einen Schritt.

Das Beispiel legt alle Reihenfolgen der Buchstaben A, B, C, also dasselbe wie `permutations('ABC')`, aber selbst
programmiert. `''.join(reihe)` klebt die Buchstaben der Liste zu einem Text zusammen.

````python
def lege(reihe, buchstaben):
    if len(reihe) == len(buchstaben):    # fertig: alle Buchstaben liegen
        print(''.join(reihe))
        return
    for b in buchstaben:
        if b not in reihe:               # b ist noch frei
            reihe.append(b)              # setzen
            lege(reihe, buchstaben)      # weitersuchen
            reihe.pop()                  # zurücknehmen

lege([], 'ABC')
````

#### Aufgabe 1.1: Was wird ausgegeben?

Überlege zuerst, was der Code ausgibt, und schreibe es auf. Führe die Zelle erst danach aus.

````python
def zeige_schritte(reihe, buchstaben):
    einrueckung = '  ' * len(reihe)
    if len(reihe) == len(buchstaben):
        print(einrueckung + 'fertig: ' + ''.join(reihe))
        return
    for b in buchstaben:
        if b not in reihe:
            print(einrueckung + 'setze ' + b)
            reihe.append(b)
            zeige_schritte(reihe, buchstaben)
            reihe.pop()
            print(einrueckung + 'nimm ' + b + ' zurück')

zeige_schritte([], 'AB')
````

**Ausgabe:**

```
  setze B

  nimm B zurück

```

Die Einrückung zeigt, wie tief die Rekursion gerade ist. Nach „fertig: AB“ wird B zurückgenommen, und in der Ebene darüber
gibt es keinen freien Buchstaben mehr. Also wird auch A zurückgenommen, und die oberste Ebene probiert B als ersten
Buchstaben. Man kann sich das als **Baum** vorstellen: Die Wurzel ist die leere Reihe, jede Abzweigung ist eine Wahl, die
Blätter sind die fertigen Reihen. Backtracking läuft diesen Baum ab, zuerst ganz in die Tiefe und dann zurück.

</details>

## 2. Früh abbrechen

Liegen am Anfang schon 3 und 4 nebeneinander, wird die Reihe nie erlaubt, egal wie es weitergeht. Trotzdem legt
`lege_ende` noch alle 24 Fortsetzungen. Die Idee des Backtrackings: **Prüfe schon beim Setzen.** Eine Zahl wird nur
gelegt, wenn sie zur letzten Zahl der Reihe passt. Dann ist jede volle Reihe automatisch erlaubt.

Damit das stimmt, muss die Prüfung zwei Bedingungen erfüllen:

- Sie darf **keine Lösung verlieren**: Abbrechen nur, wenn **sicher** ist, dass keine Fortsetzung mehr zum Ziel führt.
- Am Ende muss alles geprüft sein: Hier reicht es, jedes neue Paar von Nachbarn zu prüfen, denn jedes Paar entsteht
  genau einmal beim Setzen.

```
n = 8 früh abbrechen: 5242 erlaubt, 19185 Aufrufe
```

## 3. Damen auf dem Schachbrett

Ein klassisches Rätsel: Stelle `n` Damen so auf ein `n`×`n`-Schachbrett, dass keine eine andere schlagen kann. Eine Dame
schlägt waagerecht, senkrecht und diagonal.

In jeder Zeile steht also genau eine Dame. Darum genügt eine Liste `damen`: `damen[zeile]` ist die Spalte der Dame in dieser
Zeile. Wir setzen die Damen Zeile für Zeile, von oben nach unten. Eine neue Dame in Zeile `len(damen)` und Spalte `spalte`
passt, wenn keine schon stehende Dame

- in derselben Spalte steht (`damen[r] == spalte`) oder
- auf derselben Diagonale: Der Unterschied der Spalten ist genauso groß wie der Unterschied der Zeilen.

`spalte` von keiner Dame in `damen` bedroht wird.

**Erwartete Ausgabe:**
```
False False True True
```

````python
def passt(damen, spalte):
    zeile = len(damen)
    # dein Code
    return True

print(passt([1, 3], 3), passt([1, 3], 2), passt([1, 3], 0), passt([1, 3], 5))
````

<details>
<summary>Tipp anzeigen</summary>

Für jede Zeile `r` von 0 bis `zeile - 1`: Steht `damen[r]` in Spalte `spalte`, oder ist `abs(damen[r] - spalte)` gleich
`zeile - r`, dann gib `False` zurück.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def passt(damen, spalte):
    zeile = len(damen)
    for r in range(zeile):
        if damen[r] == spalte or abs(damen[r] - spalte) == zeile - r:
            return False
    return True

print(passt([1, 3], 3), passt([1, 3], 2), passt([1, 3], 0), passt([1, 3], 5))
```

Die Damen stehen in Zeile 0, Spalte 1 und in Zeile 1, Spalte 3. In Zeile 2 ist Spalte 3 belegt, Spalte 2 liegt diagonal neben
der Dame aus Zeile 1. Spalte 0 und Spalte 5 sind frei.

</details>

#### Aufgabe 4.1: Was wird ausgegeben?

````python
def sammle(reihe, buchstaben, ergebnisse):
    if len(reihe) == len(buchstaben):
        ergebnisse.append(reihe)
        return
    for b in buchstaben:
        if b not in reihe:
            reihe.append(b)
            sammle(reihe, buchstaben, ergebnisse)
            reihe.pop()

ergebnisse = []
sammle([], 'ABC', ergebnisse)
print(ergebnisse)
````

## Wettbewerbs-Ecke 2: Kassiopeias Weg (für Schnelle)

*Angelehnt an Aufgabe 1 „Kassiopeias Weg“ der 1. Runde des 34. Bundeswettbewerbs Informatik (2015/16).*

Das ist Backtracking auf einem Raster: **setzen** heißt, auf ein Nachbarfeld gehen, es in `besucht` eintragen und den
Buchstaben an `schritte` anhängen. **Zurücknehmen** heißt, beides wieder zu entfernen.


Begründe **ohne Programm**, warum es im freien 3×3-Feld von `kassiopeia_3x3.txt` keinen Weg gibt. Kassiopeia steht in der

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

Färbt man die 3×3 Felder wie ein Schachbrett, sind die vier Ecken und die Mitte schwarz (5 Felder), die vier Randmitten weiß
(4 Felder). Jeder Schritt wechselt die Farbe. Ein Weg durch alle 9 Felder wechselt also ab: Startfarbe, andere Farbe,
Startfarbe, …, und die Startfarbe kommt 5-mal vor. Kassiopeia startet aber auf einem **weißen** Feld, und davon gibt es nur 4.

Solche Färbungsargumente sind in der 1. Runde sehr nützlich: Sie beweisen „unmöglich“, ohne etwas durchzuprobieren, und sie
liefern einen Abbruch für das Backtracking (zu wenige Felder einer Farbe übrig → aufgeben). Das Lösungsnotebook zu
Kassiopeias Weg baut diesen Farbentest ein, `kassiopeia2.txt` ist das Beispiel der Lösungshinweise dafür.

</details>

## Das hast du gelernt

Das Muster des Backtrackings:

```python
def suche(teilloesung):
    if teilloesung ist fertig:
        Kopie speichern (alle Lösungen) oder True zurückgeben (erste Lösung)
        return
    for wahl in möglichkeiten:
        if wahl passt zur teilloesung:      # früh abbrechen
            setzen
            suche(teilloesung)               # bei „erste Lösung“: if suche(...): return True
            zurücknehmen
```

- **Setzen und zurücknehmen** gehören immer zusammen (`append`/`pop`, `add`/`remove`, `True`/`False`).
- **Früh abbrechen**, aber nur, wenn sicher ist, dass keine Fortsetzung zum Ziel führt. Diese Begründung gehört in die
  Dokumentation.
- **Lösungen als Kopie** speichern: `list(teilloesung)`.
- **Alle Lösungen** (Anzahl, beste) oder **erste Lösung** (gibt es eine?): `True` nach oben weiterreichen.
- **Aufrufe zählen** zeigt, wie viel das Abbrechen spart. Die obere Grenze in O-Notation ist oft O(n!) oder O(2ⁿ), die
  gemessenen Zähler sind viel kleiner und gehören mit in die Dokumentation.
