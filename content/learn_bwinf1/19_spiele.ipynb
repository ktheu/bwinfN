## 1. Das Streichholzspiel

Auf dem Tisch liegen `n` Streichhölzer. Zwei Spieler nehmen abwechselnd **1, 2 oder 3** Hölzer. **Wer das letzte Holz nimmt, gewinnt.** Beide spielen so gut wie möglich.

Der **Zustand** des Spiels ist hier nur die Zahl `n` der Hölzer, die noch auf dem Tisch liegen (und wer am Zug ist, das ist immer der, der gerade dran ist). Wir fragen:

> Wenn ich bei `n` Hölzern **am Zug** bin: Kann ich den Sieg **erzwingen**?

Eine Stellung, in der der Spieler am Zug den Sieg erzwingen kann, heißt **Gewinnstellung**. Kann er es nicht, ist es eine **Verluststellung**. Die Regel, die alles klärt:

> Eine Stellung ist eine **Gewinnstellung**, wenn es **einen** Zug gibt, der in eine **Verluststellung** des Gegners führt.
> Sonst (alle Züge führen in Gewinnstellungen des Gegners, oder es gibt gar keinen Zug) ist sie eine **Verluststellung**.

Bei `n = 0` liegt nichts mehr auf dem Tisch. Der Spieler, der jetzt am Zug wäre, hat **verloren**, denn sein Gegner hat gerade das letzte Holz genommen.

## 2. Der Spielbaum und das Merken

Man kann ein Spiel als **Baum** aufzeichnen: Die Wurzel ist die Anfangsstellung, jeder Zug ist eine Kante, und die Stellungen, die du erreichst, sind die Knoten darunter. Bei den Streichhölzern mit `n = 5`:

```
            5
       /    |    \
      4     3     2
    / | \  / | \  | \
   3  2  1 2  1 0  1  0
   ...
```

(Der Baum ist hier nach der zweiten Reihe abgeschnitten.) Dieselbe Stellung kommt **mehrfach** vor, zum Beispiel steht in der dritten Reihe die 2 zweimal und die 1 sogar dreimal. Die Rekursion aus Aufgabe 1.3 rechnet sie jedes Mal **neu** aus.
Das ist die gleiche Falle wie bei den Treppenstufen in [Notebook 10](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/10_merken.ipynb).

    ergebnis = []
    for nehmen in [1, 2, 3]:
        if nehmen <= n:

def gewinnt_merk(n, merk):
    """Wie gewinnt(n), aber mit einem Wörterbuch merk, das jedes Ergebnis nur einmal berechnet."""
    if n in merk:

    for neu in zuege(n):

Python kann dir das Wörterbuch auch **abnehmen**: `@lru_cache(maxsize=None)` aus [Notebook 10](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/10_merken.ipynb) merkt sich die Ergebnisse einer Funktion mit **einfachen Argumenten** (Zahlen, Tupel, Texte)
selbst. Das ist eine bequeme Abkürzung für genau diese Aufgabe. Du darfst sie benutzen, solltest aber wissen, was dahinter steckt.

````python
def zuege(n):
    """Alle Stellungen (Zahl der Hölzer), die man mit einem Zug von n aus erreicht."""
    ergebnis = []
    for nehmen in [1, 2, 3]:
        if nehmen <= n:
            ergebnis.append(n - nehmen)
    return ergebnis

from functools import lru_cache

@lru_cache(maxsize=None)
def gewinnt_lru(n):
    for neu in zuege(n):
        if not gewinnt_lru(neu):
            return True
    return False

print(gewinnt_lru(150), gewinnt_lru(151))
print(gewinnt_lru.cache_info().currsize, 'Stellungen gemerkt')
````

## 3. Von klein nach groß: die Tabelle und das Muster

Wie in [Notebook 11](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/11_tabellen_dp.ipynb) kann man die Rekursion auch **umdrehen**: Man füllt eine Liste `gewinn` von `n = 0` an aufwärts. Wenn man bei `n` ankommt, sind alle kleineren Stellungen schon ausgerechnet.
`gewinn[n]` ist `True`, wenn bei `n` Hölzern der Spieler am Zug gewinnt. Das braucht **keine Rekursion** und keine Grenze für die Tiefe.

#### Aufgabe 3.1

Schreibe `gewinn_tabelle(maximal)`. Die Liste `gewinn` fängt mit lauter `False` an (`gewinn[0]` bleibt `False`: Verlust). Für `n` von 1 bis `maximal` ist `gewinn[n]` `True`, wenn es **einen** Zug `neu` in `zuege(n)` mit `gewinn[neu]` gleich `False` gibt.
Die Ausgabe zeigt die Verluststellungen bis 60.

**Erwartete Ausgabe:**
```
Verluststellungen bis 60: [0, 4, 8, 12, 16, 20, 24, 28, 32, 36, 40, 44, 48, 52, 56, 60]
Gewinnstellungen bis 60: 45 von 61
```

````python
def zuege(n):
    """Alle Stellungen (Zahl der Hölzer), die man mit einem Zug von n aus erreicht."""
    ergebnis = []
    for nehmen in [1, 2, 3]:
        if nehmen <= n:
            ergebnis.append(n - nehmen)
    return ergebnis

def gewinn_tabelle(maximal):
    gewinn = [False] * (maximal + 1)
    # dein Code
    return gewinn

tabelle = gewinn_tabelle(60)
verluste = []
for n in range(61):
    if not tabelle[n]:
        verluste.append(n)
print('Verluststellungen bis 60:', verluste)
print('Gewinnstellungen bis 60:', tabelle.count(True), 'von', len(tabelle))
````

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

- **Bei `a ≠ b` gewinnt, wer am Zug ist:** Er nimmt vom größeren Haufen so viele Hölzer, dass beide Haufen **gleich groß** sind.
- **Bei `a = b` verliert, wer am Zug ist:** Egal, was er nimmt (von einem der beiden gleich großen Haufen), danach sind die Haufen verschieden groß. Der Gegner **spiegelt**: Er nimmt vom anderen Haufen dieselbe Zahl und stellt `a = b` wieder her. Zum Schluss ist `(0, 0)` erreicht, und der Spiegler hat das letzte Holz genommen.

Die **Invariante** ist „beide Haufen gleich groß“: Sie ist nach dem Zug des Spieglers immer wahr, und der Gegner kann sie nicht erhalten, weil jeder Zug **einen** Haufen ändert. (Bei drei und mehr Haufen ist das Muster viel schwieriger, es steckt das binäre XOR dahinter.)

</details>

## 7. Strategien gegeneinander antreten lassen

Zurück zu den Streichhölzern. Eine **Strategie** sagt für jede Stellung, welchen Zug man macht. Wir vergleichen drei:

- **`zufall`**: ein zufälliger erlaubter Zug,
- **`gierig`**: immer so viel wie möglich nehmen,
- **`klug`**: den Gewinnzug nehmen, wenn es einen gibt (aus einer **Tabelle**), sonst nur das Kleinste (so ist die Chance am größten, dass der Gegner einen Fehler macht).

Die Tabelle `zug[n]` der klugen Strategie ist die **beste Strategie als Tabelle**: Man berechnet sie einmal von klein nach groß (wie in Abschnitt 3) und schlägt dann nur noch nach. Das ist schneller als bei jedem Zug neu zu rechnen,
und es geht auch für **andere Regeln** (`erlaubt`).

## 9. Wenn sich Stellungen wiederholen können

Bei den Streichhölzern wird `n` immer kleiner, das Spiel **muss** enden, und von klein nach groß rechnen geht. Es gibt aber Spiele, in denen man **im Kreis laufen** kann, zum Beispiel, wenn Figuren vor und zurück ziehen dürfen.
Dann gibt es **drei** Ausgänge: Gewinn, Verlust oder **Unentschieden** (die Partie endet nie). Und die Rekursion „bewerte die Folgestellungen zuerst“ gerät in eine **Endlosschleife**.

Ein kleines Beispiel: Ein Stein steht auf einem **Ring** mit `n` Feldern `0, 1, …, n − 1` (nach `n − 1` kommt wieder 0). Die Spieler ziehen den Stein abwechselnd um **ein Feld vor oder zurück**.
**Wer den Stein auf das Feld 0 zieht, gewinnt.** Der Spieler am Zug, wenn der Stein schon auf 0 steht, hat also **verloren**.

Wir finden die Gewinn- und Verluststellungen **in Runden** (wie die Wellen bei der Breitensuche in [Notebook 5](https://ktheu.github.io/bwinfN/notebooks/index.html?path=learn_bwinf1/05_breitensuche.ipynb)):

1. Feld 0 ist eine Verluststellung.
2. Ein Feld ist eine **Gewinnstellung**, wenn **ein** Zug in eine bekannte Verluststellung führt.
3. Ein Feld ist eine **Verluststellung**, wenn **alle** Züge in bekannte Gewinnstellungen führen.
4. Das wiederholen wir, **bis sich nichts mehr ändert**. Was dann **übrig bleibt**, ist unentschieden: Beide Seiten können verhindern, dass der Gegner gewinnt.

```python
def ring_zuege(feld, n):
    return [(feld + 1) % n, (feld - 1) % n]

def gewinnmengen(n):
    """Gibt (gewonnen, verloren, unentschieden) zurück: Listen der Felder, auf denen der Stein steht, wenn der Spieler am Zug ... ist."""
    verloren = {0}
    gewonnen = set()
    aenderung = True
    while aenderung:
        aenderung = False
        for feld in range(n):
            if feld in verloren or feld in gewonnen:
                continue
            ziele = ring_zuege(feld, n)
            alle_gewonnen = True
            for ziel in ziele:
                if ziel in verloren:
                    gewonnen.add(feld)
                    aenderung = True
                    break
                if ziel not in gewonnen:
                    alle_gewonnen = False
            if feld not in gewonnen and alle_gewonnen:
                verloren.add(feld)
                aenderung = True
    unentschieden = [feld for feld in range(n) if feld not in gewonnen and feld not in verloren]
    return sorted(gewonnen), sorted(verloren), unentschieden

for n in [3, 5, 8]:
    gewonnen, verloren, unentschieden = gewinnmengen(n)
    print('Ring mit', n, 'Feldern: gewonnen', gewonnen, ', verloren', verloren, ', unentschieden', unentschieden)
```

Beim Ring mit 3 Feldern ist alles entschieden: Von 1 und 2 aus zieht man auf 0. Beim Ring mit 5 Feldern sind 1 und 4 gewonnen, und 2 und 3 sind **unentschieden**: Von 2 geht man nach 1 (dann gewinnt der Gegner) oder nach 3, und von 3 nach 4 (der Gegner gewinnt) oder zurück nach 2.
Beide ziehen also immer weiter hin und her, und niemand kommt auf 0. Beim Ring mit 8 Feldern sind es die Felder 2 bis 6. Wer nicht gewonnen hat und nicht verloren, bleibt für immer im Spiel, **das ist das Unentschieden**.

Das Spiel Bauernopfer wird zwischen weißen **Bauern** und einem schwarzen **Turm** auf einem Schachbrett gespielt. Zuerst stellen sich die Bauern auf, jeder auf ein anderes Feld. Dann stellt sich der Turm auf ein freies Feld.
Dann ziehen Weiß und Schwarz **abwechselnd**, Weiß beginnt. Ein weißer Zug bewegt **einen** Bauern um **ein** Feld nach oben, unten, links oder rechts. Der Turm zieht beliebig weit (auch **gar nicht**) nach oben, unten, links oder rechts,
darf aber **weder durch einen Bauern hindurch noch auf ein Feld mit einem Bauern**. Weiß **fängt** den Turm, wenn ein Bauer auf das Feld des Turms zieht. Weiß möchte den Turm fangen, der Turm möchte nie gefangen werden.
Die echte Aufgabe fragt, ob 8 Bauern (oder 7) den Turm auf dem 8 × 8-Brett fangen können, und untersucht mehrere Varianten.

**Wir vereinfachen:** Das Brett ist ein `n × n`-Brett mit `n = 3` oder `4`, es gibt `k` Bauern, und wir fragen nur, **wer gewinnt**, indem wir **alle Stellungen** durchrechnen. Auf dem 8 × 8-Brett geht das nicht (Frage 8.1).
Weggelassen haben wir: die Strategien der Bauern auf dem 8 × 8-Brett, mehrere Bauernzüge pro Runde und die Dame statt des Turms.

Dateien: `daten/bauernopfer_faelle.txt` hat in jeder Zeile `n k` (Brettgröße und Zahl der Bauern). `daten/bauernopfer_stellungen.txt` enthält Stellungen mit Weiß am Zug als Textbild (`B` Bauer, `T` Turm, `.` leer, oben ist die erste Zeile),
getrennt durch Leerzeilen. Fertig sind `lies_faelle`, `lies_stellungen` und `weisse_zuege`.
Der Zustand ist das Paar `(bauern, turm)` mit **Weiß am Zug**: `bauern` ist ein `frozenset` aus Feldern `(x, y)` (ein `frozenset` ist eine Menge, die man als Wörterbuchschlüssel verwenden darf) und `turm` ein Feld.

RICHTUNGEN = [(1, 0), (-1, 0), (0, 1), (0, -1)]

def weisse_zuege(n, bauern, turm):
    """Alle weißen Züge: Liste der neuen Bauernmengen. None bedeutet: Ein Bauer zieht auf den Turm (Fang)."""
    ergebnis = []
    for bauer in bauern:
        for dx, dy in RICHTUNGEN:
            ziel = (bauer[0] + dx, bauer[1] + dy)
            if not (0 <= ziel[0] < n and 0 <= ziel[1] < n):
                continue
            if ziel == turm:
                ergebnis.append(None)
            elif ziel not in bauern:
                ergebnis.append((bauern - {bauer}) | {ziel})
    return ergebnis

def turm_ziele(n, bauern, turm):
    """Alle Felder, auf die der Turm ziehen darf, einschließlich seines eigenen Feldes (null Schritte)."""
    ziele = [turm]
    for dx, dy in RICHTUNGEN:
        x = turm[0] + dx
        y = turm[1] + dy
        while 0 <= x < n and 0 <= y < n and (x, y) not in bauern:
            ziele.append((x, y))
            x = x + dx
            y = y + dy
    return ziele

    """Dict (bauern, turm) -> Zahl der weißen Züge bis zum Fang bei bestem Spiel (Weiß am Zug). Fehlt eine Stellung: Der Turm gewinnt."""
    felder = [(x, y) for y in range(n) for x in range(n)]
    alle = []
    for auswahl in combinations(felder, k):
        bauern = frozenset(auswahl)
        for turm in felder:
            if turm not in bauern:
                alle.append((bauern, turm))
    runde = {}

    geaendert = True
    while geaendert:
        geaendert = False
        zug_nr = zug_nr + 1
        neu = []                              # in dieser Runde neu gewonnene Stellungen
        for bauern, turm in alle:
            if (bauern, turm) in runde:
                continue
            for neue_bauern in weisse_zuege(n, bauern, turm):
                if neue_bauern is None:
                    gut = True                # Fang
                else:
                    gut = True                # jede Antwort des Turms führt in eine schon gewonnene Stellung
                    for ziel in turm_ziele(n, neue_bauern, turm):
                        if (neue_bauern, ziel) not in runde:
                            gut = False
                            break
                if gut:
                    neu.append((bauern, turm))
                    break
        for stellung in neu:
            runde[stellung] = zug_nr
            geaendert = True
    return runde

Ein Feld ist **sicher**, wenn in seiner **Reihe und in seiner Spalte kein Bauer** steht. Ein Bauer fängt den Turm nur, indem er **auf sein Feld zieht**, und dazu muss er vorher **direkt daneben** stehen, also in der Reihe oder Spalte des Turms. Auf einem sicheren Feld ist der Turm nie fangbar.

- **Zu Beginn** gibt es ein sicheres Feld: 7 Bauern belegen höchstens 7 von 8 Spalten und höchstens 7 von 8 Reihen (**Schubfachprinzip**), also bleiben mindestens eine freie Spalte und eine freie Reihe, und deren Kreuzungsfeld ist sicher.
- **Invariante:** Der Turm steht nach seinem Zug immer auf einem sicheren Feld. Zieht ein Bauer um ein Feld, ist er danach in höchstens einer der beiden Linien des Turms. Ist das Feld noch sicher, bleibt der Turm stehen. Sonst steht der Bauer in seiner Spalte (oder Reihe), aber die Reihe (oder Spalte) ist noch frei.
  Der Turm zieht entlang der freien Linie auf eine andere freie Spalte (es gibt wieder eine, denn es sind nur 7 Bauern auf 8 Spalten) und steht wieder sicher.
- Wo hilft die **7**? Bei **8** Bauern können alle 8 Spalten (oder alle 8 Reihen) besetzt sein, dann gibt es kein sicheres Feld mehr.

- **Rückwärtsanalyse:** Bei Spielen mit Wiederholungen arbeitet man von den Endstellungen **rückwärts** und merkt sich pro Stellung die Zahl der noch ungeprüften Züge. Dann ist jede Stellung nur einmal dran, und die Laufzeit ist wieder `O(Zustände · Züge)`.
