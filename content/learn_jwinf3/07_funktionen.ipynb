## 1. Eine Funktion definieren und aufrufen

In Blockly ziehst du den Block „etwas tun“ auf die Fläche, gibst ihm einen Namen und steckst Blöcke hinein. Danach gibt es
einen neuen Block mit diesem Namen, den du benutzen kannst. In Python:

```python
def begruessung():
    print('Hallo!')
    print('Willkommen beim Jugendwettbewerb.')
```

- `def` (von *define*, festlegen) leitet die Funktion ein, dann kommen der Name, runde Klammern und ein Doppelpunkt.
- Was die Funktion tun soll, ist eingerückt, wie bei `if` und `for`.
- Die Definition allein tut noch nichts. Erst der **Aufruf** `begruessung()` führt die Funktion aus. Die Klammern gehören
  dazu.

````python
def begruessung():
    print('Hallo!')
    print('Willkommen beim Jugendwettbewerb.')

begruessung()
begruessung()
````

**Wichtig im Notebook:** Eine Funktion gibt es erst, wenn die Zelle mit ihrem `def` ausgeführt wurde. Wenn du später eine
Fehlermeldung `NameError: name '…' is not defined` bekommst, hast du wahrscheinlich die Zelle mit der Definition noch nicht
ausgeführt (zum Beispiel nach einem Neustart).

## 2. Parameter: der Funktion etwas mitgeben

Über das Zahnrad ⚙ kannst du dem Blockly-Block „etwas tun“ Eingaben geben. In Python heißen sie **Parameter** und stehen in
den Klammern. Beim Aufruf schreibst du die Werte in die Klammern. Mehrere Parameter trennst du durch Kommas.

````python
def begruesse(name):
    print('Hallo', name + '!')

def zeige_ergebnis(name, punkte):
    print(name, 'hat', punkte, 'Punkte.')

begruesse('Bela')
begruesse('Amira')
zeige_ergebnis('Mara', 23)
````

## 3. Ergebnisse zurückgeben: `return`

Viele Blockly-Blöcke **liefern einen Wert**, zum Beispiel „Länge von“ oder „Rest von“. Solche Funktionen baust du in
Blockly mit „etwas tun … gib zurück“. In Python steht dafür `return`:

```python
def quadrat(x):
    return x * x
```

`return` beendet die Funktion und gibt den Wert an die Stelle zurück, an der sie aufgerufen wurde. Dort kannst du mit dem
Wert weiterrechnen, ihn in einer Variablen speichern oder ausgeben.

**`print` und `return` sind verschieden:** `print` zeigt etwas auf dem Bildschirm an. `return` gibt das Ergebnis an das
Programm zurück. Funktionen, die etwas ausrechnen, sollen fast immer `return` benutzen, damit man mit dem Ergebnis
weiterarbeiten kann.

````python
def quadrat(x):
    return x * x

print(quadrat(7))
flaeche = quadrat(3) + quadrat(4)
print(flaeche)
````

## 4. Früh zurückgeben und testen

`return` beendet die Funktion **sofort**. Das ist praktisch, wenn die Antwort schon feststeht, bevor alles durchgerechnet
ist. In Blockly gibt es dafür den Block **falls … gib zurück**.

Beispiel: Eine Zahl ist eine **Primzahl**, wenn sie größer als 1 ist und keinen Teiler außer 1 und sich selbst hat.
Sobald ein Teiler gefunden ist, steht fest: keine Primzahl.

````python
def ist_primzahl(n):
    if n < 2:
        return False
    for t in range(2, n):
        if n % t == 0:
            return False            # Teiler gefunden: sofort fertig
    return True                     # kein Teiler gefunden

print(ist_primzahl(13), ist_primzahl(15), ist_primzahl(1))
````

**Testen:** Rufe jede Funktion gleich nach dem Schreiben mit ein paar Werten auf, deren Ergebnis du kennst. Denk dabei an
**Sonderfälle** wie 0, 1, einen leeren Text oder eine leere Liste. Im Wettbewerb ist das besonders wichtig: Eine kleine,
getestete Funktion macht viel seltener Fehler als ein langes Programm am Stück.

## 5. Mehrere Ergebnisse und Listen

Eine Funktion kann **mehrere** Werte zurückgeben, durch Kommas getrennt. Beim Aufruf verteilst du sie auf mehrere Variablen:

````python
def verteile(spiele, tueten):
    jede = spiele // tueten
    mehr = spiele % tueten
    return jede, mehr

jede, mehr = verteile(10, 3)
print(jede, 'Spiele je Tüte,', mehr, 'Tüten bekommen eins mehr')
````

Eine Funktion kann auch eine **Liste** zurückgeben oder eine Liste als Parameter bekommen. Variablen, die **in** einer
Funktion entstehen, gibt es nur dort. Sie stören also keine gleichnamigen Variablen draußen. Deshalb gibst du Ergebnisse
mit `return` zurück, statt sie in Variablen „liegen zu lassen“.

#### Aufgabe E.1

````python
def zug(feld, wurf):
    return feld + wurf      # noch ohne Leitern und Zurücklaufen

print(zug(0, 2))      # erwartet 2
print(zug(0, 4))      # erwartet 12 (Leiter hinauf)
print(zug(10, 2))     # erwartet 4 (Leiter hinunter)
print(zug(18, 5))     # erwartet 17 -> 13 (3 zurück, dann Leiter)
print(zug(16, 4))     # erwartet 20
````

<details>
<summary>Tipp anzeigen</summary>

Der Code aus Notebook 4 kommt in die Funktion. Statt `feld = 12` kannst du gleich `return 12` schreiben. Dann brauchst du

</details>

```python

    ziel = 20
    feld = feld + wurf
    if feld > ziel:

    if feld == 12:
```

**Für Schnelle:** Verpacke das Wundertüten-Packen aus Notebook 6 in eine Funktion `packe(sorten, anzahlen, n)`, die eine

<details>
<summary>Tipp anzeigen</summary>

Der Code aus Notebook 6, E.3 kommt in die Funktion, mit `n` statt `len(tueten)`. Am Ende `return tueten`.

```python
def packe(sorten, anzahlen, n):
    tueten = [''] * n
    tuete = 0
    for s in range(len(sorten)):
        for i in range(anzahlen[s]):
            tueten[tuete] = tueten[tuete] + sorten[s]
            tuete = (tuete + 1) % n
    return tueten

print(packe(['W', 'K', 'G'], [4, 4, 2], 3))
print(packe(['W', 'K', 'G'], [4, 4, 2], 2))
```

## Das hast du gelernt

| Blockly | Python |
|---|---|
| etwas tun | `def etwas_tun():` und eingerückter Teil |
| etwas tun (mit Eingaben über ⚙) | `def etwas_tun(a, b):` |
| etwas tun … gib zurück | `return wert` |
| falls … gib zurück | `if bedingung:` `return wert` |
| (Aufruf) | `etwas_tun()`, `ergebnis = etwas_tun(3, 4)` |

- Erst definieren (Zelle ausführen), dann aufrufen.
- `print` zeigt an, `return` gibt zurück. Ohne `return` liefert eine Funktion `None`.
- `return a, b` gibt zwei Werte zurück: `x, y = funktion()`.
- Jede Funktion gleich mit ein paar Testaufrufen prüfen, auch mit Sonderfällen.
