# 25. Die Dokumentation

Im Wettbewerb gibst du nicht nur ein Programm ab, sondern vor allem eine **Dokumentation**: einen Text, der erklärt, wie du die
Aufgabe gelöst hast, und der die Ergebnisse zeigt. Das wird oft unterschätzt. Dabei ist die Dokumentation **die Grundlage der
Bewertung**: Die Bewertenden haben nicht die Zeit, jedes eingesandte Programm gründlich zu testen. Sie lesen, was du schreibst.

In diesem Notebook lernst du:

- aus welchen Teilen eine Dokumentation besteht und was jeweils hineingehört,
- wie bewertet wird und welche typischen Fehler Punkte kosten,
- wie eine gute Dokumentation aussieht (am Beispiel Texthopsen aus Notebook 24),
- was du in welcher Form abgibst.

Hier programmierst du nicht, sondern schreibst. Benutze dafür die Felder *Deine Antwort* (Doppelklick zum Bearbeiten).

## 1. Wie bewertet wird

In der 3. Runde des Jugendwettbewerbs (zugleich die 1. Runde des Bundeswettbewerbs) gilt laut den Lösungshinweisen der
letzten Jahre:

- Jede Aufgabe startet mit **5 Punkten**. Für Mängel gibt es **Abzüge**, meist einen Punkt, manchmal zwei.
- Zu jeder Aufgabe gibt es eine Liste von **Bewertungskriterien**, die nach der Runde in den Lösungshinweisen veröffentlicht
  wird.
- Einige Abzüge gelten **für alle Aufgaben**: Die Dokumentation ist sehr unverständlich oder unvollständig, das Verfahren ist
  schlecht begründet, Beispiele fehlen.

So sahen die Kriterien zum Beispiel für **Texthopsen** aus (sinngemäß, Abzug jeweils 1 Punkt):

| Kriterium | worum es geht |
|---|---|
| Gewinnkriterium nicht erkennbar | Es muss klar sein, ob „wer als Erster draußen ist“ oder „wer weniger Sprünge braucht“ gewinnt. |
| Modellierung ungeeignet | Umlaute und ß müssen richtig eingelesen werden, die Sprungweite muss sich leicht ermitteln lassen. |
| Lösungsverfahren fehlerhaft | Die Sprünge müssen stimmen, Nicht-Buchstaben dürfen nicht mitzählen, Großbuchstaben müssen richtig behandelt werden. |
| Ergebnisse schlecht nachvollziehbar | Mindestens muss ausgegeben werden, wer gewinnt oder ob es unentschieden ist. |
| Beispiele fehlerhaft oder zu wenige | Ergebnisse zu mindestens vier der fünf Beispieldateien müssen in der Dokumentation stehen. |

Fällt dir etwas auf? Fast alle Kriterien prüfen Dinge, die du in Notebook 24 Schritt für Schritt bedacht hast: Unklarheiten
festlegen (Notebook 23), Sonderfälle (Notebook 14), alle Beispiele.

## 2. Die Teile einer Dokumentation

Laut Aufgabenblatt gehört zu jeder bearbeiteten Aufgabe:

1. **Lösungsidee:** Welches Problem steckt hinter der Aufgabe, und wie wird es grundsätzlich gelöst?
2. **Umsetzung:** Wie wurde die Idee in ein Programm übertragen? Welche Datenstrukturen, welche Besonderheiten?
3. **Beispiele:** An genügend Beispielen zeigen, dass und wie die Lösung funktioniert. Seit dem 44. Wettbewerb gilt: **alle**
   vorgegebenen Beispiele, dazu eigene, besonders Sonderfälle.
4. **Quellcode:** die wichtigsten Teile des Programms.

Seit dem 45. Wettbewerb kommen dazu: Die genutzten **Werkzeuge** sollen erläutert werden (zum Beispiel Programmierumgebung und
Bibliotheken), und manche Aufgaben schreiben ein **Ausgabeformat** vor. Lies auf dem aktuellen Aufgabenblatt nach, was genau
verlangt ist. Die Regeln können sich von Jahr zu Jahr ändern.

Eine gute Dokumentation muss **nicht lang** sein, für eine Junioraufgabe genügen oft wenige Seiten. Sie muss aber
**vollständig** und **verständlich** sein.

## 3. Die Lösungsidee

Die Lösungsidee ist der wichtigste Teil. Sie erklärt das **Verfahren**, nicht das Programm. Drei Faustregeln aus den
Lösungshinweisen:

- **Keine Bedienungsanleitung** („Man startet das Programm und gibt den Dateinamen ein …“).
- **Keine Wiederholung der Aufgabe.** Die Bewertenden kennen die Aufgabe.
- **Keine Namen aus dem Programm** (Variablen, Funktionen). Die Idee muss auch verständlich sein, wenn man das Programm nicht kennt.

Dafür gehört in die Lösungsidee: das Verfahren in Worten, **Festlegungen** zu Unklarheiten mit Begründung, und warum das Verfahren
richtig ist (zum Beispiel, warum man Bela und Amira getrennt simulieren darf).

## 4. Die Umsetzung

Hier darfst (und sollst) du über das Programm sprechen: in welcher Sprache es geschrieben ist, wie die Daten gespeichert werden,
welche Funktionen es gibt und was sie tun, und was du beachten musstest. Gut ist eine kurze Übersicht über die Funktionen.
Ein Beispiel für Texthopsen:

> Das Programm ist in Python geschrieben (Version 3.x, ausgeführt in JupyterLite). Die Sprungweiten stehen in einem Wörterbuch,
> das jedem der 30 Buchstaben seine Weite zuordnet. Die Dateien werden mit der Kodierung UTF-8 gelesen, damit Umlaute und ß
> richtig ankommen.
>
> - `nur_buchstaben(text)` macht alles klein und behält nur die Buchstaben, die im Wörterbuch stehen.
> - `anzahl_spruenge(buchstaben, start)` spielt das Hopsen ab einer Startposition nach und zählt die Sprünge.
> - `ergebnis(text)` ruft beides für Bela (Start 0) und Amira (Start 1) auf und vergleicht.

Die genaue Python-Version gibt `import sys` und `print(sys.version)` aus. Trage sie statt „3.x“ ein. Die Umsetzung ist

## 5. Die Beispiele

Die Beispiele sind **besonders wichtig**. Aus den Lösungshinweisen:

- Es reicht **nicht**, die Beispiele nur als Dateien abzugeben, sie fest ins Programm einzubauen oder den Bewertenden das
  Ausprobieren zu überlassen. Die Ergebnisse müssen **in der Dokumentation** stehen.
- Die Beispiele sollen zeigen, dass die Lösung **richtig** ist, und wie das Programm mit **Sonderfällen** umgeht.

Praktische Tipps:

- Alle vorgegebenen Beispiele mit Eingabe (oder Dateiname) und vollständiger Ausgabe.
- Eine **Tabelle** ist oft am übersichtlichsten.
- Ist eine Ausgabe sehr lang, zeigst du einen aussagekräftigen Ausschnitt und sagst, wo der Rest steht. Manche Aufgaben legen
  genau fest, was in der Dokumentation stehen muss, zum Beispiel bei sehr langen Nachrichten nur Anfang und Ende.
- Eigene Beispiele für Sonderfälle mit einem Satz dazu, **was** sie zeigen sollen.

## 6. Der Quellcode

In die Dokumentation gehören die **wichtigen** Teile des Programms, also die, in denen die Lösungsidee steckt. Nicht nötig sind
Einlesen, Ausgabe und andere Routine, außer sie enthalten etwas Besonderes. Das vollständige Programm gibst du zusätzlich als
Datei ab.

- Der Code soll **kommentiert** sein: kurze Kommentare, **warum** etwas passiert, nicht nur was.
- Benutze eine **Schrift mit fester Breite** (zum Beispiel Courier oder Consolas), damit Einrückungen erhalten bleiben. In
  Python ist das besonders wichtig.
- Nimm Reste von Versuchen und auskommentierten Code heraus.

## 7. Eine vollständige Musterdokumentation

So könnte die Dokumentation zu Texthopsen aussehen. Sie ist kurz, aber alles Wichtige steht drin.

---

### Texthopsen

#### Lösungsidee

Zuerst werden aus dem Text alle Zeichen entfernt, die nicht in der Sprungweiten-Tabelle stehen (Leerzeichen, Satzzeichen, Ziffern,
aber auch Buchstaben wie é). Großbuchstaben werden wie Kleinbuchstaben behandelt, denn im Beispiel des Aufgabenblatts hat das große
B die Sprungweite 2. Übrig bleibt eine Folge von Buchstaben.

Für Bela (Start beim ersten Buchstaben) und Amira (Start beim zweiten) wird das Hopsen getrennt nachgespielt: Von der aktuellen
Position aus wird um die Sprungweite des Buchstabens dort weitergezählt, bis die Position hinter dem letzten Buchstaben liegt. Die
Sprünge werden gezählt. Getrennt nachspielen ist erlaubt, weil sich die beiden nicht gegenseitig beeinflussen.

**Gewinnregel:** „Wer dies als Erstes schafft“ verstehen wir so: Es gewinnt, wer **weniger Sprünge** braucht. Bei gleich vielen
Sprüngen ist es unentschieden. Weil die beiden abwechselnd hopsen und Bela beginnt, ist Bela in diesem Fall zeitlich zuerst
draußen. Das Programm gibt das zusätzlich aus, sodass das Ergebnis auch für diese Lesart ablesbar ist.

**Sonderfälle:** Enthält der Text weniger als zwei Buchstaben, kann Amira nicht starten. Das Programm wertet das als 0 Sprünge.
Bei einem Text ganz ohne Buchstaben brauchen beide 0 Sprünge (unentschieden).

#### Umsetzung

Python 3.x in JupyterLite (genaue Version mit `sys.version` ermitteln), ohne zusätzliche Bibliotheken. Die Sprungweiten stehen in einem Wörterbuch. Die Texte werden mit der
Kodierung UTF-8 gelesen, damit Umlaute und ß stimmen. Funktionen: `nur_buchstaben` (Text filtern und klein schreiben),
`anzahl_spruenge` (Hopsen ab einer Startposition nachspielen), `ergebnis` (beide vergleichen und das Ergebnis formulieren).

#### Beispiele

| Datei | Buchstaben | Sprünge Bela | Sprünge Amira | Ergebnis |
|---|---|---|---|---|
| Beispiel des Aufgabenblatts | 21 | 3 | 3 | unentschieden (Bela zuerst draußen) |
| hopsen1.txt | 836 | 68 | 70 | Bela gewinnt |
| hopsen2.txt | 221 | 25 | 25 | unentschieden (Bela zuerst draußen) |
| hopsen3.txt | 218 | 18 | 18 | unentschieden (Bela zuerst draußen) |
| hopsen4.txt | 378 | 35 | 32 | Amira gewinnt |
| hopsen5.txt | 10748 | 923 | 930 | Bela gewinnt |

Eigene Beispiele:

| Text | Bela | Amira | Ergebnis | zeigt |
|---|---|---|---|---|
| `cab` | 1 | 2 | Bela gewinnt | ein Fall, den man von Hand nachprüfen kann |
| `ZEBRA zebra` | 1 | 2 | Bela gewinnt | Großbuchstaben und Leerzeichen |
| `ÄÖÜß` | 1 | 1 | unentschieden | Umlaute und ß mit Sprungweiten 27 bis 30 |
| `a` | 1 | 0 | Amira gewinnt | Sonderfall: Amira kann nicht starten |
| `123 !?` | 0 | 0 | unentschieden | Sonderfall: keine Buchstaben |

#### Quellcode (Auszug)

```python
def nur_buchstaben(text):
    # Groß- wie Kleinbuchstaben behandeln, alles außer den 30 Tabellen-Buchstaben weglassen
    ergebnis = ''
    for zeichen in text.lower():
        if zeichen in WEITE:
            ergebnis = ergebnis + zeichen
    return ergebnis


def anzahl_spruenge(buchstaben, start):
    # so lange springen, bis die Position hinter dem letzten Buchstaben liegt
    position = start
    spruenge = 0
    while position < len(buchstaben):
        position = position + WEITE[buchstaben[position]]
        spruenge = spruenge + 1
    return spruenge
```
