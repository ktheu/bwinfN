## 1. Listen anlegen und lesen

Eine Liste schreibst du in eckigen Klammern, die Werte durch Kommas getrennt:

| Blockly | Python |
|---|---|
| erzeuge eine leere Liste | `[]` |
| erzeuge Liste mit 32, 27, 28 | `[32, 27, 28]` |
| erzeuge Liste mit 5 mal dem Element 0 | `[0] * 5` |
| Länge von | `len(liste)` |
| in der Liste nimm #tes | `liste[i - 1]` |
| ist leer | `liste == []` |

Wie bei Texten zählt Python die Positionen **ab 0**. Das erste Element ist `liste[0]`, das letzte `liste[-1]`.
Die Position heißt auch **Index**.

````python
kinder = [32, 27, 28, 26]
print(kinder)
print(len(kinder))
print(kinder[0], kinder[-1])
namen = ['Bela', 'Amira', 'Mara']
print(namen[1])
nullen = [0] * 5
print(nullen)
````

## 2. Ändern und anhängen

Anders als ein Text lässt sich eine Liste **verändern**:

- `liste[i] = wert` ersetzt das Element an Position `i` (Blockly: „in der Liste setze für #tes ein“).
- `liste.append(wert)` hängt einen Wert **hinten an**. Die Liste wird dadurch um eins länger.

Oft beginnst du mit einer leeren Liste und hängst in einer Schleife nach und nach Werte an. Das Muster kennst du von der
Summe und vom neuen Text: vorher leer, in der Schleife anhängen, danach fertig.

````python
punkte = [10, 20, 30]
punkte[1] = 25
punkte.append(40)
print(punkte)

quadrate = []
for i in range(1, 6):
    quadrate.append(i * i)
print(quadrate)
````

## 3. Listen durchgehen

`for wert in liste:` geht die Liste Element für Element durch. Das ist der Blockly-Block „für jeden Wert i aus der Liste“.
Brauchst du die Position, nimmst du wie bei Texten `for i in range(len(liste)):` und `liste[i]`.

Für häufige Aufgaben hat Python fertige Funktionen:

| Python | Bedeutung |
|---|---|
| `sum(liste)` | Summe (Blockly: „Summe über die Liste“) |
| `max(liste)`, `min(liste)` | größter und kleinster Wert |
| `x in liste` | kommt `x` in der Liste vor? |
| `liste.index(x)` | Position, an der `x` zuerst vorkommt (Blockly: „suche erstes Auftreten“, aber ab 0) |
| `sorted(liste)` | neue, aufsteigend sortierte Liste (Blockly: „sortieren“) |

`liste.index(x)` meldet einen Fehler, wenn `x` nicht vorkommt. Prüfe deshalb vorher mit `if x in liste:`.

````python
kinder = [32, 27, 28, 26]
summe = 0
for k in kinder:
    summe = summe + k
print('Von Hand:', summe, ' mit sum:', sum(kinder))
print('Größte Klasse:', max(kinder), 'an Position', kinder.index(max(kinder)))
print('Sortiert:', sorted(kinder))
print(27 in kinder, 30 in kinder)
````

## 5. Vom Text zur Liste und zurück

Daten kommen im Wettbewerb fast immer als Text an, zum Beispiel eine Zeile `4 4 2` aus einer Datei.

- `text.split()` zerlegt einen Text an den Leerzeichen in eine **Liste von Texten** (Blockly: „Liste aus Text erstellen
  mit Trennzeichen“). Mit `text.split(',')` wird an den Kommas zerlegt.
- `' '.join(liste)` macht aus einer Liste von **Texten** wieder einen Text, hier mit Leerzeichen dazwischen (Blockly:
  „Text aus Liste erstellen“).

Achtung: Nach `split` sind auch Zahlen noch **Texte**. Mit ihnen rechnen kannst du erst nach `int(...)`.

````python
zeile = '4 4 2'
teile = zeile.split()
print(teile)
zahlen = []
for t in teile:
    zahlen.append(int(t))
print(zahlen, sum(zahlen))
print('-'.join(['W', 'K', 'G']))
````

Äußere Schleife: `for s in range(len(sorten)):`. Innere Schleife: `for i in range(anzahlen[s]):`. Darin hängst du
