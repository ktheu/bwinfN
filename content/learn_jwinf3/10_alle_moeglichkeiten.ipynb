# 10. Schleifen ineinander: alle Möglichkeiten

Viele Wettbewerbsaufgaben kann man lösen, indem man **alle Möglichkeiten durchprobiert** und die passenden oder die beste
heraussucht. Das heißt **Brute Force** (englisch für rohe Gewalt). Der Computer ist schnell, er probiert in einer Sekunde
Hunderttausende Möglichkeiten. Oft ist Brute Force die einfachste und sicherste Lösung.

Dafür brauchst du **Schleifen ineinander**: für jede erste Wahl jede zweite Wahl, für jede zweite jede dritte …

In diesem Notebook lernst du:

- wie Schleifen ineinander ablaufen,
- wie du **alle Paare** aus einer Liste bildest, mit und ohne Wiederholung,
- wie du mit zwei oder drei Schleifen alle Möglichkeiten durchprobierst,
- wie du abschätzt, ob das schnell genug geht, und die Schleifen **klug begrenzt**.

In der Wettbewerbs-Ecke teilst du Gärten auf, suchst Reimpaare und erzeugst Bruch-Aufgaben.

## 1. Schleifen ineinander

Eine Schleife in einer anderen Schleife läuft **für jeden** Durchgang der äußeren Schleife **vollständig** durch. Bei 3
Durchgängen außen und 4 innen wird der innere Teil also 3 · 4 = 12-mal ausgeführt.

In Blockly steckst du dafür einen „zähle“-Block in einen anderen. In Python rückst du die innere Schleife eine Stufe weiter
ein.

````python
for i in range(1, 4):
    for j in range(1, 5):
        print(i, j, end='   ')
    print()
````

## 2. Alle Paare

Oft musst du jedes Element mit jedem anderen vergleichen: Welche Wörter reimen sich? Welche zwei Zahlen ergeben zusammen
100? Wer spielt gegen wen? Dafür gehst du mit zwei Schleifen alle **Paare von Positionen** `i` und `j` durch. Es gibt zwei
Varianten:

| Variante | Schleifen | Für 4 Elemente |
|---|---|---|
| jedes Paar in **beiden** Reihenfolgen (A gegen B **und** B gegen A) | `for i in range(n):` `for j in range(n):` `if i != j:` | 4 · 3 = 12 Paare |
| jedes Paar **nur einmal** (A gegen B) | `for i in range(n):` `for j in range(i + 1, n):` | 4 · 3 / 2 = 6 Paare |

In der zweiten Variante beginnt die innere Schleife **hinter** `i`. So kommt jedes Paar genau einmal vor, und kein Element
wird mit sich selbst gepaart. Das ist die Variante, die du meistens brauchst.

````python
spieler = ['Ada', 'Bela', 'Cem', 'Dana']
n = len(spieler)
spiele = 0
for i in range(n):
    for j in range(i + 1, n):
        print(spieler[i], 'gegen', spieler[j])
        spiele = spiele + 1
print(spiele, 'Spiele')
````

## 4. Ist das schnell genug?

Python im Browser schafft grob eine bis einige **Millionen** einfache Schritte pro Sekunde. Bevor du Schleifen ineinander
schreibst, schätze ab, **wie oft** der innerste Teil ausgeführt wird:

| Schleifen | Durchläufe bei n = 100 | bei n = 1000 | bei n = 10000 |
|---|---|---|---|
| eine | 100 | 1000 | 10 000 |
| zwei (alle Paare) | 10 000 | 1 000 000 | 100 000 000 |
| drei | 1 000 000 | 1 000 000 000 | … |

Eine Million geht in einer Sekunde, hundert Millionen dauern Minuten, eine Milliarde dauert zu lange. Ist es zu viel, gibt es
zwei Auswege:

1. **Früh aufhören:** Wenn du nur **eine** Lösung brauchst, beende die Suche, sobald du sie gefunden hast. Am einfachsten geht
   das in einer Funktion mit `return`. **Achtung:** `break` beendet nur die **innerste** Schleife, die äußere läuft weiter!
2. **Schleifengrenzen klug wählen:** Überlege, welche Werte gar nicht in Frage kommen, und lass die Schleifen dort gar nicht
   erst hinlaufen.

````python
def erstes_paar(zahlen, ziel):
    for i in range(len(zahlen)):
        for j in range(i + 1, len(zahlen)):
            if zahlen[i] + zahlen[j] == ziel:
                return zahlen[i], zahlen[j]     # beendet beide Schleifen auf einmal
    return None                                 # kein Paar gefunden


print(erstes_paar([35, 60, 50, 65, 40, 50, 25], 100))
print(erstes_paar([1, 2, 3], 100))
````

<details>
<summary>Tipp anzeigen</summary>

Zwei Schleifen: `for z in range(1, hoechstens + 1):` und `for s in range(1, hoechstens // z + 1):`. Wegen der Grenze für `s`
ist `z * s` nie zu groß, du musst nur noch `z * s >= k` prüfen. Dann rechnest du das Verhältnis aus und merkst dir das Beste

<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def beste_aufteilung(k, hoehe, breite):
    hoechstens = k * 11 // 10
    beste_z = 0
    beste_s = 0
    bestes_verhaeltnis = 0
    for z in range(1, hoechstens + 1):
        for s in range(1, hoechstens // z + 1):
            if z * s >= k:
                a = hoehe / z
                b = breite / s
                verhaeltnis = max(a, b) / min(a, b)
                if beste_z == 0 or verhaeltnis < bestes_verhaeltnis:
                    beste_z = z
                    beste_s = s
                    bestes_verhaeltnis = verhaeltnis
    return beste_z, beste_s

print(beste_aufteilung(23, 42, 66))
print(beste_aufteilung(35, 10, 20))
```

                    print(f'Kürze {a}/{b} = {p}/{q}')
print(anzahl, 'leichte Aufgaben')
