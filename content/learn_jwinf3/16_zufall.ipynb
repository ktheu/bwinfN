# 16. Zufall

Manche Aufgaben verlangen ausdrücklich Zufall: „Schreibe ein Programm, das zufällige Passwörter erzeugt“, „das immer neue
Songtexte schreibt“, „das zufällige Übungsaufgaben ausgibt“. Und manchmal ist Zufall ein gutes **Werkzeug**: Wenn man nicht
alle Möglichkeiten durchprobieren kann, probiert man eben sehr viele zufällige und nimmt die beste.

In diesem Notebook lernst du:

- wie du Zufallszahlen und zufällige Auswahlen erzeugst,
- wie du dafür sorgst, dass beim Testen trotzdem immer dasselbe herauskommt (`seed`),
- wie du etwas nach **Regeln** zufällig erzeugst und das Ergebnis prüfst,
- wie du mit vielen zufälligen Versuchen eine gute Lösung findest.

In der Wettbewerbs-Ecke erzeugst du aussprechbare Passwörter und Silben-Songs.

## 1. Zufallszahlen

Die Zufallsfunktionen stehen im Modul `random`. Du lädst es einmal mit `import random`.

| Blockly | Python | Ergebnis |
|---|---|---|
| ganzzahlige Zufallszahl zwischen 1 und 6 | `random.randint(1, 6)` | 1, 2, 3, 4, 5 oder 6 (**beide** Grenzen möglich) |
| Zufallszahl (0.0 - 1.0) | `random.random()` | Kommazahl von 0 bis unter 1 |
| vom Text nehme zufälligen Buchstaben | `random.choice(text)` | ein zufälliges Zeichen |
| (zufälliges Element einer Liste) | `random.choice(liste)` | ein zufälliges Element |
| (mehrere verschiedene Elemente) | `random.sample(liste, k)` | Liste mit `k` verschiedenen Elementen |
| (Liste mischen) | `random.shuffle(liste)` | mischt die Liste selbst |

Achtung: Bei `randint` ist die obere Grenze **dabei**, anders als bei `range`.

Führe die Zelle mehrmals aus. Jedes Mal kommt etwas anderes heraus.

````python
import random

print(random.randint(1, 6), random.randint(1, 6), random.randint(1, 6))
print(random.random())
print(random.choice(['Schere', 'Stein', 'Papier']))
print(random.choice('aeiou'))
print(random.sample(range(1, 50), 6))
````

**Zufall beim Testen:** Wenn jedes Mal etwas anderes herauskommt, ist Fehlersuche schwer. Darum gibt es `random.seed(zahl)`
(*seed* heißt Samen oder Startwert). Nach `random.seed(1)` erzeugt Python **immer dieselbe** Folge von „Zufallszahlen“, solange
sich die Python-Version nicht ändert. In diesem Notebook beginnt jede Lösung mit `random.seed(...)`. Darum siehst du bei dir
dieselbe Ausgabe wie in der erwarteten Ausgabe. Im fertigen Programm lässt du `seed` weg, dann ist es wirklich zufällig.

## 4. Erzeugen und prüfen

Nicht jede Regel lässt sich in die Bausteine einbauen. Dann gibt es ein zweites Muster: **Erzeugen und prüfen**. Du erzeugst
zufällig, prüfst die Regel und erzeugst neu, solange sie nicht erfüllt ist.

```python
kandidat = erzeuge()
while not ist_gut(kandidat):
    kandidat = erzeuge()
```

Wichtig: Die Regel darf nicht zu streng sein, sonst probiert das Programm sehr lange.

## 5. Viele zufällige Versuche, das Beste merken

Manchmal gibt es so viele Möglichkeiten, dass man nicht alle durchprobieren kann (Notebook 10). Dann probiert man **sehr viele
zufällige** und merkt sich die beste. Das findet nicht immer die beste Lösung, aber oft eine sehr gute.

Ein Beispiel: Sechs Pakete sollen so auf zwei Wagen verteilt werden, dass beide möglichst gleich schwer sind. Ein Versuch:
Jedes Paket kommt zufällig auf Wagen A oder B. Bewertet wird der Unterschied der Gewichte, je kleiner, desto besser.

#### Aufgabe 5.1

Probiere 1000 zufällige Verteilungen aus und merke dir die mit dem kleinsten Unterschied. Gib diesen Unterschied und die
beiden Wagen aus.

**Erwartete Ausgabe:**
```
Unterschied: 0
Wagen A: [23, 21, 12] 56
Wagen B: [17, 9, 30] 56
```

````python
import random
random.seed(6)
pakete = [17, 23, 9, 30, 21, 12]
bester_unterschied = sum(pakete)
beste_a = []
beste_b = []
# dein Code
````

<details>
<summary>Tipp anzeigen</summary>

In jedem Versuch zwei leere Listen `a` und `b`. Jedes Paket kommt mit `random.randint(0, 1)` in eine davon. Dann den
Unterschied `abs(sum(a) - sum(b))` mit dem besten bisher vergleichen (Notebook 9).

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
import random
random.seed(6)
pakete = [17, 23, 9, 30, 21, 12]
bester_unterschied = sum(pakete)
beste_a = []
beste_b = []
for versuch in range(1000):
    a = []
    b = []
    for p in pakete:
        if random.randint(0, 1) == 0:
            a.append(p)
        else:
            b.append(p)
    unterschied = abs(sum(a) - sum(b))
    if unterschied < bester_unterschied:
        bester_unterschied = unterschied
        beste_a = a
        beste_b = b
print('Unterschied:', bester_unterschied)
print('Wagen A:', beste_a, sum(beste_a))
print('Wagen B:', beste_b, sum(beste_b))
```

Die Pakete wiegen zusammen 112, das Beste wäre also 56 auf jedem Wagen, und genau das wird gefunden. Bei sechs Paketen gibt
es nur 2 · 2 · 2 · 2 · 2 · 2 = 64 Verteilungen, hier hätte man auch alle durchprobieren können. Bei 60 Paketen wären es aber
mehr, als ein Computer je schaffen könnte. Dann hilft der Zufall. Wie man Strategien vergleicht, lernst du in Notebook 18.

</details>
