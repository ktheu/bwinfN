## 1. Tabellen als Listen von Listen

Eine Tabelle speicherst du als Liste ihrer **Zeilen**. Jede Zeile ist selbst eine Liste. Bei der Treffsicherheit in
Notebook 9 hast du das schon gesehen.

```python
punkte = [[3, 5, 0, 4],      # Zeile 0: Bela
          [5, 5, 2, 1],      # Zeile 1: Amira
          [4, 0, 5, 5]]      # Zeile 2: Mara
```

- `punkte[1]` ist die ganze Zeile 1: `[5, 5, 2, 1]`.
- `punkte[1][2]` ist in Zeile 1 das Element an Position 2: `2`. Erst die Zeile, dann die Spalte.
- `len(punkte)` ist die Anzahl der Zeilen, `len(punkte[0])` die Anzahl der Spalten.

````python
punkte = [[3, 5, 0, 4],
          [5, 5, 2, 1],
          [4, 0, 5, 5]]
namen = ['Bela', 'Amira', 'Mara']
print(punkte[1])
print(punkte[1][2])
print(len(punkte), 'Zeilen,', len(punkte[0]), 'Spalten')
for z in range(len(punkte)):
    print(f'{namen[z]:6}', punkte[z])
````

## 4. Rechnen im Kreis

Nach 23 Uhr kommt 0 Uhr, nach Sonntag kommt Montag, nach der letzten Landzunge einer Insel kommt wieder die erste. Solche
Zählungen, die **im Kreis** laufen, rechnest du mit dem Rest `%` (Notebook 1). Gibt es `n` Plätze mit den Nummern 0 bis
`n - 1`, dann ist `(platz + schritte) % n` der Platz, auf dem man nach `schritte` Schritten landet.

Das klappt auch **rückwärts**: In Python ist `-1 % 7` gleich `6`. Ein Schritt zurück von Platz 0 führt also auf den letzten
Platz, genau wie es sein soll.

````python
print('5 Stunden nach 22 Uhr ist es', (22 + 5) % 24, 'Uhr')
tage = ['Mo', 'Di', 'Mi', 'Do', 'Fr', 'Sa', 'So']
heute = 4                                # Freitag
print('In 10 Tagen ist', tage[(heute + 10) % 7])
print('Gestern war', tage[(heute - 1) % 7])
print('Vor 5 Tagen von Montag aus:', tage[(0 - 5) % 7])
````

<details>
<summary>Tipp anzeigen</summary>

Das neue Element an Position `i` ist das alte an Position `(i + k) % n`.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def drehe(liste, k):
    n = len(liste)
    neu = []
    for i in range(n):
        neu.append(liste[(i + k) % n])
    return neu

print(drehe(['A', 'B', 'C', 'D', 'E'], 2))
print(drehe(['A', 'B', 'C', 'D', 'E'], 7))
print(drehe(['A', 'B', 'C', 'D', 'E'], -1))
```

Dank `%` funktioniert auch `k = 7` (wie 2, einmal ganz herum und noch 2) und `k = -1` (ein Schritt zurück).

## 5. Koordinate und Position

Manchmal ist eine Tabelle als **eine lange Liste** gespeichert, Zeile für Zeile hintereinander. Bilddateien sind oft so
aufgebaut. Bei einer Breite von `breite` Spalten gilt:

| von | nach | Rechnung |
|---|---|---|
| Zeile `y`, Spalte `x` | Position in der langen Liste | `pos = y * breite + x` |
| Position `pos` | Zeile und Spalte | `y = pos // breite` und `x = pos % breite` |

Vor Zeile `y` liegen ja `y` vollständige Zeilen mit je `breite` Elementen.

Juliane versteckt Nachrichten in Bildern. Jedes Pixel hat drei Farbwerte `r`, `g`, `b`. Man beginnt beim Pixel oben links.
`r` ist der Zeichencode des nächsten Buchstabens (`chr(r)`, Notebook 5). `g` sagt, wie viele Pixel man nach **rechts**
gehen muss, `b`, wie viele nach **unten**. Wer über den rechten Rand hinauskommt, macht in derselben Zeile links weiter, wer
über den unteren Rand hinauskommt, in derselben Spalte oben: Rechnen im Kreis! Sind `g` und `b` beide 0, ist die Nachricht
zu Ende.

Das Bild ist eine Tabelle `bild` mit `hoehe` Zeilen und `breite` Spalten. Jedes Pixel `bild[y][x]` ist eine Liste `[r, g, b]`.

#### Aufgabe E.3

Entschlüssele die Nachricht im Bild. Gib außerdem jedes besuchte Pixel mit seiner Spalte `x` und Zeile `y` aus.


````python
breite = 8
hoehe = 6
bild = []
for y in range(hoehe):                     # ein Bild mit zufällig aussehenden Werten ...
    zeile = []
    for x in range(breite):
        zeile.append([(37 * x + 11 * y) % 256, (5 * x + y) % 7 + 1, (3 * x + 2 * y) % 5 + 1])
    bild.append(zeile)
bild[0][0] = [72, 4, 2]                    # ... in dem fünf Pixel die Nachricht tragen
bild[2][4] = [97, 6, 3]
bild[5][2] = [108, 2, 6]
bild[5][4] = [108, 5, 4]
bild[3][1] = [111, 0, 0]

x = 0
y = 0
nachricht = ''
# dein Code
print(nachricht)
````

## Das hast du gelernt

| Aufgabe | Python |
|---|---|
| Element in Zeile `z`, Spalte `s` | `tabelle[z][s]` |
| Anzahl Zeilen / Spalten | `len(tabelle)` / `len(tabelle[0])` |
| Tabelle anlegen (richtig) | in einer Schleife `tabelle.append([0] * spalten)` |
| Tabelle anlegen (**falsch**) | `[[0] * spalten] * zeilen`: alle Zeilen sind dieselbe |
| jede Zelle besuchen | `for z in range(...):` `for s in range(...):` |
| im Kreis weiterzählen | `(platz + schritte) % n`, auch rückwärts |
| Koordinate → Position | `pos = y * breite + x` |
