## 1. Eine Datei lesen

Die Datei `daten/zahlen.txt` enthält:


So liest du sie:

```python
with open('daten/zahlen.txt', encoding='utf-8') as f:

```

- `open(...)` öffnet die Datei. Der Pfad `daten/zahlen.txt` bedeutet: die Datei `zahlen.txt` im Ordner `daten`, der neben

## 2. Zeile für Zeile

Oft hat jede Zeile eine eigene Bedeutung. Die Datei `daten/punkte.txt` ist typisch für den Wettbewerb: In der ersten Zeile
steht, wie viele Zeilen folgen, danach kommt in jeder Zeile ein Name und eine Punktzahl.

```
5
Bela 17
Amira 23
Mara 9
Leo 30
Tim 21
```

`text.splitlines()` zerlegt den Text in eine Liste von **Zeilen**. Jede Zeile kannst du dann mit `split()` in ihre Teile
zerlegen:

| Blockly | Python |
|---|---|
| lies Zeile als ganze Zahl | `n = int(zeilen[0])` |
| lies Zeile als Text | `zeile = zeilen[i]` |
| (Zeile in Teile zerlegen) | `teile = zeile.split()` |

````python
with open('daten/punkte.txt', encoding='utf-8') as f:
    zeilen = f.read().splitlines()
print(zeilen)
n = int(zeilen[0])
for i in range(1, n + 1):
    teile = zeilen[i].split()
    print('Name:', teile[0], ' Punkte:', int(teile[1]))
````

**Unsichtbare Zeichen:** Dateien, die unter Windows erstellt wurden, haben am Zeilenende oft noch ein `\r` vor dem `\n`.
Das siehst du in `repr` (probiere `print(repr(open('daten/baellebad.txt', encoding='utf-8').read()[:30]))`). `splitlines()`,
`split()` und `strip()` entfernen solche Zeichen. Trenne deshalb nie selbst mit `split('\n')`, sonst klebt an jeder Zeile
noch ein `\r`, und Vergleiche wie `tag == 'Montag'` gehen schief.

## 4. Ergebnisse ordentlich ausgeben

Mit den **f-Strings** aus Notebook 5 kannst du auch die Breite und die Nachkommastellen festlegen. Dazu kommt hinter den
Wert in den geschweiften Klammern ein Doppelpunkt und eine Angabe:

| f-String | Bedeutung | Ergebnis für `x = 3.14159`, `name = 'Leo'` |
|---|---|---|
| `f'{x:.2f}'` | 2 Nachkommastellen | `3.14` |
| `f'{name:6}'` | mindestens 6 Zeichen breit (Text linksbündig) | `Leo   ` |
| `f'{42:5}'` | mindestens 5 Zeichen breit (Zahl rechtsbündig) | `   42` |
| `f'{x:7.1f}'` | 7 breit, 1 Nachkommastelle | `    3.1` |

So stehen die Werte in einer Tabelle sauber untereinander.

````python
namen = ['Bela', 'Amira', 'Leo']
punkte = [17, 23, 30]
for i in range(len(namen)):
    print(f'{namen[i]:6} {punkte[i]:3} Punkte')
print(f'Durchschnitt: {sum(punkte) / len(punkte):.2f}')
````

#### Aufgabe E.2

Lies `daten/wundertuete.txt`, packe die Tüten mit `packe` aus Notebook 7 und schreibe das Ergebnis im festgelegten Format in

**Erwartete Ausgabe:**
```

2 1 2 0
3 1 1 1
```

````python
def packe(sorten, anzahlen, n):           # aus Notebook 7
    tueten = [''] * n
    tuete = 0
    for s in range(len(sorten)):
        for i in range(anzahlen[s]):
            tueten[tuete] = tueten[tuete] + sorten[s]
            tuete = (tuete + 1) % n
    return tueten

n, anzahlen = lies_wundertuete('daten/wundertuete.txt')
sorten = []
for s in range(len(anzahlen)):
    sorten.append(chr(ord('A') + s))
tueten = packe(sorten, anzahlen, n)
print(tueten)
# dein Code
````
