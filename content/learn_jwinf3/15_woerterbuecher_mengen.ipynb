# 15. Wörterbücher und Mengen

Listen sind praktisch, wenn die Dinge eine **Reihenfolge** haben und über ihre **Position** angesprochen werden. Oft will
man aber etwas **nachschlagen**: Wohin führt die Leiter auf Feld 27? Wie viele Punkte hat Amira? Oder man will nur wissen, ob
etwas **dabei** ist: Habe ich dieses Feld schon besucht? Dafür hat Python zwei weitere Datentypen, die es in Blockly nicht gibt:

- das **Wörterbuch** (`dict`, von *dictionary*): Zu jedem **Schlüssel** gehört ein **Wert**, wie in einem Wörterbuch zu jedem
  deutschen Wort das englische.
- die **Menge** (`set`): eine Sammlung ohne Reihenfolge und ohne doppelte Elemente.

In der Wettbewerbs-Ecke löst du die Aufgabe „Auf und Ab“ mit dem echten Spielbrett und findest den schwersten Container.

## 1. Nachschlagen im Wörterbuch

Ein Wörterbuch schreibst du in geschweiften Klammern: `{schlüssel: wert, schlüssel: wert, ...}`.

| Python | Bedeutung |
|---|---|
| `punkte = {'Bela': 17, 'Amira': 23}` | Wörterbuch anlegen |
| `punkte['Amira']` | Wert zum Schlüssel nachschlagen: `23` |
| `punkte['Mara'] = 9` | neuen Eintrag anlegen oder einen vorhandenen ändern |
| `'Mara' in punkte` | gibt es diesen Schlüssel? |
| `punkte.get('Leo', 0)` | Wert nachschlagen, `0`, falls es den Schlüssel nicht gibt |
| `len(punkte)` | Anzahl der Einträge |
| `for name in punkte:` | alle Schlüssel durchgehen |

Schlüssel können Texte oder Zahlen sein, Werte alles Mögliche (auch Listen). Jeder Schlüssel kommt nur **einmal** vor.

````python
punkte = {'Bela': 17, 'Amira': 23}
print(punkte['Amira'])
punkte['Mara'] = 9
punkte['Bela'] = punkte['Bela'] + 5
print(punkte)
print('Leo' in punkte, punkte.get('Leo', 0))
for name in punkte:
    print(name, 'hat', punkte[name], 'Punkte')
````

## 2. Leitern als Wörterbuch

Im Leiterspiel brauchtest du für jede Leiter zwei `if`-Zeilen. Beim echten Spielbrett mit 12 Leitern wären das 24. Mit einem
Wörterbuch `leiter`, das jedem Leiterende das andere Ende zuordnet, wird daraus **eine** Zeile:

## 3. Zählen mit einem Wörterbuch

In Notebook 9 hast du mit einer Zählliste gezählt. Das klappt nur, wenn die Werte kleine Zahlen sind, die als Position taugen.
Für **Wörter** oder große Zahlen nimmst du ein Wörterbuch als Zähler. Das Muster:

```python
anzahl = {}
for w in woerter:
    anzahl[w] = anzahl.get(w, 0) + 1
```

Beim ersten Vorkommen gibt `get` die 0 zurück, danach den bisherigen Stand.

## 4. Gruppieren

Der Wert in einem Wörterbuch darf auch eine **Liste** sein. Damit kannst du Dinge **gruppieren**: Alle Wörter mit derselben
Endung kommen in dieselbe Liste. Das Muster:

```python
gruppen = {}
for w in woerter:
    schluessel = ...                    # wonach gruppiert wird
    if schluessel not in gruppen:
        gruppen[schluessel] = []        # neue, leere Gruppe
    gruppen[schluessel].append(w)
```

<details>
<summary>Tipp anzeigen</summary>

Der Schlüssel ist `w.lower()[-3:]`. Danach das Gruppier-Muster von oben und eine Schleife über alle Schlüssel.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

## 5. Mengen

Eine **Menge** enthält jedes Element höchstens einmal, und es gibt keine Reihenfolge (also auch keine Positionen).

| Python | Bedeutung |
|---|---|
| `s = set()` | leere Menge (Achtung: `{}` ist ein leeres **Wörterbuch**!) |
| `s = {3, 1, 4}` | Menge mit Elementen |
| `s.add(x)` | Element hinzufügen (ist es schon da, passiert nichts) |
| `x in s` | ist `x` in der Menge? Das geht auch bei riesigen Mengen blitzschnell. |
| `set(liste)` | Menge aus einer Liste machen, doppelte fallen weg |
| `a - b` | alle Elemente von `a`, die **nicht** in `b` sind |
| `len(s)`, `for x in s:` | Anzahl, alle durchgehen |

Mengen sind ideal für „**schon gesehen**“ (Notebook 13): `x in gesehen` ist bei einer Menge viel schneller als bei einer
Liste, weil Python die Liste von vorn bis hinten durchsuchen muss.

````python
farben = ['rot', 'grün', 'rot', 'blau', 'grün', 'rot']
verschieden = set(farben)
print(verschieden, len(verschieden))
verschieden.add('gelb')
verschieden.add('rot')
print('gelb' in verschieden, len(verschieden))
print({1, 2, 3, 4, 5} - {2, 4})
````

#### Aufgabe E.1

Schreibe `zug(feld, wurf)` mit dem Wörterbuch `leiter` und `spiele(wurf)` mit einer **Menge** `gesehen` (Notebook 13). Gib das
Ergebnis für alle Würfe von 1 bis 6 aus.

**Erwartete Ausgabe:**
```
Mit 1 im Ziel nach 26 Würfen
Mit 2 im Ziel nach 17 Würfen
Mit 3 nie im Ziel
Mit 4 nie im Ziel
Mit 5 im Ziel nach 16 Würfen
Mit 6 im Ziel nach 14 Würfen
```

````python
def zug(feld, wurf):
    return feld + wurf      # ersetze diese Zeile


def spiele(wurf):
    return -1               # ersetze diese Zeile


for wurf in range(1, 7):
    print(wurf, spiele(wurf))
````

<details>
<summary>Tipp anzeigen</summary>

`zug`: weiterziehen, zu viel zurücklaufen, dann `return leiter.get(feld, feld)`. `spiele`: wie in Notebook 13, Aufgabe 3.2,
aber mit `gesehen = set()` und `gesehen.add(feld)`, und die Würfe zählst du extra mit.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def zug(feld, wurf):
    feld = feld + wurf
    if feld > 100:
        feld = 100 - (feld - 100)
    return leiter.get(feld, feld)


def spiele(wurf):
    feld = 0
    wuerfe = 0
    gesehen = set()
    while feld != 100:
        if feld in gesehen:
            return -1
        gesehen.add(feld)
        feld = zug(feld, wurf)
        wuerfe = wuerfe + 1
    return wuerfe


for wurf in range(1, 7):
    w = spiele(wurf)
    if w == -1:
        print('Mit', wurf, 'nie im Ziel')
    else:
        print('Mit', wurf, 'im Ziel nach', w, 'Würfen')
```

Das sind genau die Ergebnisse der offiziellen Lösungshinweise: Mit der 3 und der 4 kommt Frederick nie an, mit der 6 am
schnellsten. Vergleiche die Länge des Programms mit Notebook 4: Das Wörterbuch macht aus einer langen `elif`-Kette eine

## Wettbewerbs-Ecke 2: Container

*Angelehnt an die Junioraufgabe „Container“ (Jugendwettbewerb 2022).*

Container wurden paarweise gewogen. Für jedes Paar weiß man nur, welcher schwerer ist. In jeder Zeile der Datei steht ein
Paar `a b`: Container `a` ist schwerer als Container `b`. Kann man daraus sicher sagen, welcher Container der **schwerste** ist?

Die Idee: Ein Container, der bei **irgendeiner** Wägung der leichtere war, kann nicht der schwerste sein. Er scheidet aus.
Übrig bleiben die **Kandidaten**. Ist es genau einer, ist er sicher der schwerste (der schwerste verliert ja nie). Sind es
mehrere, könnte jeder von ihnen der schwerste sein.

#### Aufgabe E.2

Schreibe `kandidaten(dateiname)`, die die Menge aller Container und die Menge aller leichteren Container bildet und die
Differenz zurückgibt. Wende sie auf `daten/container0.txt` und `daten/container1.txt` an und gib aus, ob der schwerste
eindeutig ist.

**Erwartete Ausgabe:**
```
daten/container0.txt: nicht eindeutig, in Frage kommen [3, 4, 5]
daten/container1.txt: Der schwerste ist sicher Container 4
```

````python
def kandidaten(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = f.read().splitlines()
    alle = set()
    leichter = set()
    # dein Code
    return alle - leichter


for name in ['daten/container0.txt', 'daten/container1.txt']:
    print(name, kandidaten(name))
````

<details>
<summary>Tipp anzeigen</summary>

Für jede Zeile mit zwei Zahlen kommen beide in `alle`, die zweite außerdem in `leichter`. `len(teile) == 2` überspringt leere
Zeilen. Eine Menge kannst du mit `sorted(k)` als sortierte Liste ausgeben und mit `list(k)[0]` das einzige Element holen.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def kandidaten(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zeilen = f.read().splitlines()
    alle = set()
    leichter = set()
    for zeile in zeilen:
        teile = zeile.split()
        if len(teile) == 2:
            alle.add(int(teile[0]))
            alle.add(int(teile[1]))
            leichter.add(int(teile[1]))
    return alle - leichter


for name in ['daten/container0.txt', 'daten/container1.txt']:
    k = kandidaten(name)
    if len(k) == 1:
        print(name + ': Der schwerste ist sicher Container', list(k)[0])
    else:
        print(name + ': nicht eindeutig, in Frage kommen', sorted(k))
```

In `container0.txt` bleiben die Container 3, 4 und 5 übrig: Keiner von ihnen wurde mit einem anderen der drei gewogen. In
`container1.txt` ist es eindeutig Container 4. Was wäre, wenn **gar kein** Kandidat übrig bleibt? Dann widersprechen sich
die Wägungen (etwa 1 schwerer als 2, 2 schwerer als 3, 3 schwerer als 1). Auch das ist ein Sonderfall für eine eigene
Testdatei (Notebook 14). Siehe `jwinf3/41_Container/container_loesung.ipynb`.

</details>

## Das hast du gelernt

| Wörterbuch (`dict`) | Menge (`set`) |
|---|---|
| `d = {}` oder `{'a': 1}` | `s = set()` oder `{1, 2}` |
| `d[schluessel]`, `d.get(schluessel, ersatz)` | `x in s` |
| `d[schluessel] = wert` | `s.add(x)` |
| `for schluessel in d:` | `for x in s:` |
| zählen: `d[w] = d.get(w, 0) + 1` | doppelte entfernen: `set(liste)` |
| gruppieren: Werte sind Listen | ausscheiden lassen: `alle - raus` |

- Ein Wörterbuch für **Nachschlagen**, **Zählen** und **Gruppieren**.
- Eine Menge für **schon gesehen**, **verschiedene** und **Ausscheiden**.
- `{}` ist ein leeres Wörterbuch, eine leere Menge ist `set()`.
