## 1. Länge und einzelne Zeichen

`len(text)` ist die Länge des Textes, also die Anzahl der Zeichen (Blockly: „Länge von“). Leerzeichen und Satzzeichen
zählen mit.

Ein einzelnes Zeichen bekommst du mit eckigen Klammern: `text[0]`. **Achtung, das ist der wichtigste Unterschied zu
Blockly:** Python zählt die Positionen **ab 0**. Das erste Zeichen ist `text[0]`, das zweite `text[1]` und so weiter.

| Zeichen | J | u | g | e | n | d |
|---|---|---|---|---|---|---|
| Blockly: „vom Text nehme #ten Buchstaben“ | 1 | 2 | 3 | 4 | 5 | 6 |
| Python: `text[…]` | 0 | 1 | 2 | 3 | 4 | 5 |
| Python von hinten: `text[…]` | -6 | -5 | -4 | -3 | -2 | -1 |

Das letzte Zeichen hat also die Position `len(text) - 1`. Bequemer ist `text[-1]`: Negative Zahlen zählen von hinten
(Blockly: „#ten Buchstaben von hinten“).

````python
text = 'Jugend'
print(len(text))
print(text[0], text[1], text[5])
print(text[-1], text[-2])
````

## 2. Ausschnitte

`text[a:b]` ist der Ausschnitt von Position `a` **bis vor** Position `b`. Wie bei `range` ist das Ende nicht dabei. Fehlt
eine der Zahlen, geht der Ausschnitt bis zum Anfang oder bis zum Ende.

| Python | Bedeutung | für `'Jugendwettbewerb'` |
|---|---|---|
| `text[0:6]` | Positionen 0 bis 5 | `'Jugend'` |
| `text[:6]` | die ersten 6 Zeichen | `'Jugend'` |
| `text[6:]` | alles ab Position 6 | `'wettbewerb'` |
| `text[-4:]` | die letzten 4 Zeichen | `'werb'` |

Der Blockly-Block „im Text nehme Teil ab #tem Buchstaben bis zum #ten Buchstaben“ zählt ab 1 und mit Ende. Aus „ab dem
3. bis zum 5. Buchstaben“ wird deshalb `text[2:5]`.

````python
text = 'Jugendwettbewerb'
print(text[:6])
print(text[6:])
print(text[-4:])
print(text[2:5])
````

## 3. Zeichen für Zeichen

Mit `for` kannst du einen Text Zeichen für Zeichen durchgehen. Die Schleifenvariable ist dann in jedem Durchgang das
nächste **Zeichen**:

```python
for zeichen in text:
```

Das ist wie der Blockly-Block „für jeden Wert i aus der Liste“, nur dass der Text die Rolle der Liste spielt.

Brauchst du auch die **Position**, gehst du die Positionen mit `range(len(text))` durch und holst dir das Zeichen mit `text[i]`.

````python
text = 'Hopsen'
for zeichen in text:
    print(zeichen, end=' ')
print()
for i in range(len(text)):
    print(i, text[i])
````

Zum Prüfen von Zeichen gibt es nützliche Werkzeuge:

| Python | Bedeutung | Beispiel |
|---|---|---|
| `x in text` | kommt `x` in `text` vor? | `'e' in 'aeiou'` ist `True` |
| `text.lower()` | alles klein (Blockly: „wandel um in kleinbuchstaben“) | `'Ball'.lower()` ist `'ball'` |
| `text.upper()` | alles groß (Blockly: „GROSSBUCHSTABEN“) | `'Ball'.upper()` ist `'BALL'` |
| `zeichen.isalpha()` | ist es ein Buchstabe (auch ä, ö, ü, ß)? | `'7'.isalpha()` ist `False` |
| `text.find(x)` | Position, an der `x` zuerst vorkommt | `'Ball'.find('l')` ist `2` |

`find` liefert `-1`, wenn `x` nicht vorkommt. Blockly zählt bei „suche erstes Auftreten“ ab 1 und liefert 0, wenn nichts
gefunden wird. Auch hier musst du also umdenken.

`text.lower()` verändert `text` übrigens nicht, sondern liefert einen **neuen** Text. Willst du ihn behalten, schreibst du
`text = text.lower()`.

## 4. Einen neuen Text aufbauen

Einen Text kann man nicht verändern, man kann aber einen **neuen** bauen. Das Muster kennst du von der Summe: Vor der
Schleife ist der neue Text leer (`neu = ''`), in der Schleife wird angehängt (`neu = neu + zeichen`, Blockly: „an neu
Text anhängen“), nach der Schleife ist er fertig.

````python
text = 'Beispieltexte sind toll'
neu = ''
for zeichen in text:
    if zeichen != ' ':
        neu = neu + zeichen
print(neu)
````

## 5. Buchstaben sind Zahlen

Im Computer hat jedes Zeichen eine Nummer. `ord(zeichen)` liefert die Nummer (Blockly: „Zeichen zu Zahl“), `chr(nummer)`
das Zeichen dazu (Blockly: „Zahl zu Zeichen“). Die Kleinbuchstaben a bis z haben die Nummern 97 bis 122, **ohne Lücke
und in der Reihenfolge des Alphabets**. Deshalb kannst du damit rechnen:

- Position im Alphabet: `ord(b) - ord('a') + 1` ist 1 für `a`, 2 für `b`, …, 26 für `z`.
- Ein Buchstabe weiter: `chr(ord(b) + 1)`.

Umlaute und ß haben ganz andere Nummern (`ord('ä')` ist 228). Sie musst du deshalb extra behandeln.

````python
print(ord('a'), ord('b'), ord('z'))
print(chr(97), chr(98))
b = 'e'
print(b, 'ist der', ord(b) - ord('a') + 1, '. Buchstabe')
````

## Für Profis: Texte zusammensetzen mit f-Strings

Mit `+` und `str()` lassen sich Texte zusammensetzen. Kürzer geht es mit einem **f-String**: Vor das Anführungszeichen kommt
ein `f`, und alles in geschweiften Klammern `{…}` wird ausgerechnet und eingesetzt (Blockly: „erstelle Text aus“).

```python

```

## Das hast du gelernt

| Blockly | Python |
|---|---|
| Länge von | `len(text)` |
| vom Text nehme #ten Buchstaben | `text[i - 1]` (Python zählt **ab 0**!) |
| letzten Buchstaben / #ten von hinten | `text[-1]` / `text[-i]` |
| im Text nehme Teil ab #tem bis zum #ten Buchstaben | `text[a - 1:b]` (Ende **nicht** dabei) |
| für jeden Wert … (Text statt Liste) | `for zeichen in text:` |
| im Text suche erstes Auftreten | `text.find(x)` (ab 0, `-1` = nicht gefunden) |
| wandel um in kleinbuchstaben / GROSSBUCHSTABEN | `text.lower()` / `text.upper()` |
| an … Text anhängen | `neu = neu + zeichen` |
| Zeichen zu Zahl / Zahl zu Zeichen | `ord(zeichen)` / `chr(nummer)` |
| (kein Block) | `x in text`, `zeichen.isalpha()` |
