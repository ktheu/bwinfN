# 2. Entscheidungen mit `if`

Bisher hat dein Programm jede Zeile ausgeführt, eine nach der anderen. Oft soll ein Programm aber **entscheiden**:
Nur wenn etwas zutrifft, soll etwas passieren. In Blockly machst du das mit dem grünen Block **falls … mache**.
In Python heißt er `if`.

In diesem Notebook lernst du:

- wie Python **vergleicht** (`==`, `<`, `>=`, …),
- wie du mit `if`, `else` und `elif` entscheidest,

## 1. Vergleichen

Ein Vergleich ergibt **wahr** oder **unwahr**. In Python heißen diese beiden Werte `True` und `False` (groß geschrieben).

| Blockly | Python | Bedeutung |
|---|---|---|
| … **=** … | `==` | gleich |
| … **≠** … | `!=` | ungleich |
| … **<** … | `<` | kleiner |
| … **≤** … | `<=` | kleiner oder gleich |
| … **>** … | `>` | größer |
| … **≥** … | `>=` | größer oder gleich |
| wahr / unwahr | `True` / `False` | |

**Achtung:** Zum Vergleichen schreibst du **zwei** Gleichheitszeichen `==`. Ein einzelnes `=` kennst du schon:
Damit bekommt eine Variable einen Wert (Blockly: „setze … auf“).

````python
punkte = 85
print(punkte > 50)
print(punkte == 100)
print(punkte != 100)
````

## 2. falls … mache: `if`

So sieht eine Entscheidung in Blockly aus:

```
falls  punkte ≥ 100
mache  schreibe „Gewonnen!“
```

Und so in Python:

```python
if punkte >= 100:
    print('Gewonnen!')
```

Drei Dinge sind wichtig:

1. Hinter der Bedingung steht ein **Doppelpunkt** `:`.
2. Was im „mache“-Teil steht, wird **eingerückt**, und zwar mit 4 Leerzeichen. Die **Tab-Taste** ⇥ macht das für dich.
   Nach einem Doppelpunkt rückt das Notebook die nächste Zeile beim Drücken von Enter sogar von selbst ein.
3. Alle eingerückten Zeilen gehören zum „mache“-Teil. Die erste Zeile, die **nicht** mehr eingerückt ist, gehört nicht
   mehr dazu und wird immer ausgeführt.

Die Einrückung ersetzt also die „Klammer“ des Blockly-Blocks, in die du die Blöcke hineinsteckst.

````python
punkte = 120
if punkte >= 100:
    print('Gewonnen!')
    print('Du bekommst einen Pokal.')
print('Das Spiel ist vorbei.')
````

## 3. sonst: `else`

Über das Zahnrad ⚙ kannst du in Blockly an „falls“ ein **sonst** anbauen. In Python heißt es `else`. Der `else`-Teil
wird ausgeführt, wenn die Bedingung **nicht** zutrifft. Auch hinter `else` steht ein Doppelpunkt, und auch hier wird eingerückt.

```
falls  alter < 12
mache  setze preis auf 3
sonst  setze preis auf 6
```

```python
if alter < 12:
    preis = 3
else:
    preis = 6
```

`else` steht genau unter dem `if`, nicht eingerückt.

````python
alter = 13
if alter < 12:
    preis = 3
else:
    preis = 6
print('Eintritt:', preis, 'Euro')
````

## 4. sonst falls: `elif`

Mit dem Zahnrad ⚙ kannst du auch **sonst falls** anbauen, so oft du willst. In Python heißt das `elif`
(kurz für *else if*). Python prüft die Bedingungen **von oben nach unten**. Bei der **ersten**, die zutrifft, wird der
eingerückte Teil ausgeführt, und alle weiteren werden übersprungen. Trifft keine zu, kommt der `else`-Teil dran
(falls es einen gibt).

````python
platz = 2
if platz == 1:
    print('Gold')
elif platz == 2:
    print('Silber')
elif platz == 3:
    print('Bronze')
else:
    print('Leider keine Medaille')
````

## 5. und, oder, nicht

Mehrere Bedingungen verknüpfst du wie in Blockly:

| Blockly | Python | wahr, wenn … |
|---|---|---|
| … **und** … | `and` | **beide** Bedingungen wahr sind |
| … **oder** … | `or` | **mindestens eine** wahr ist |
| **nicht** … | `not` | die Bedingung unwahr ist |

Python kann außerdem etwas, das Blockly nicht kann: `1 <= wurf <= 6` prüft wie in Mathe, ob `wurf` zwischen 1 und 6 liegt.
Das ist dasselbe wie `1 <= wurf and wurf <= 6`.

````python
wurf = 7
if 1 <= wurf <= 6:
    print('Gültiger Wurf')
else:
    print('Das kann kein Würfel sein!')

alter = 15
schueler = True
if alter < 18 or schueler:
    print('Ermäßigung')
````

## Für Profis: prüfe … falls wahr … falls unwahr

In Blockly gibt es den Block **prüfe … falls wahr … falls unwahr**, der einen von zwei Werten liefert. In Python schreibt man
das in einer Zeile:

```python
preis = 3 if alter < 12 else 6
```

Das ist dasselbe wie das `if`/`else` aus Abschnitt 3. Für den Anfang ist die lange Form übersichtlicher.

## Das hast du gelernt

| Blockly | Python |
|---|---|
| = ≠ < ≤ > ≥ | `==` `!=` `<` `<=` `>` `>=` |
| wahr / unwahr | `True` / `False` |
| falls … mache | `if bedingung:` und eingerückter Teil |
| sonst falls | `elif bedingung:` |
| sonst | `else:` |
| und / oder / nicht | `and` / `or` / `not` |
| (kein Block) | `1 <= x <= 6` |

- Hinter `if`, `elif` und `else` steht ein **Doppelpunkt**.
- Was dazugehört, wird mit 4 Leerzeichen **eingerückt** (Tab-Taste).
- Bei `elif` gewinnt die **erste** Bedingung, die zutrifft.

Im nächsten Notebook lernst du Schleifen: aus „wiederhole 10 mal“ und „zähle i von 1 bis 10“ wird `for`.
