## Schritt 1: Die Aufgabe lesen

> **Texthopsen**
>
> Bela und Amira ist im Deutschunterricht oft langweilig. Daher haben sie sich ein neues Spiel ausgedacht: Texthopsen.
> Sie suchen sich einen Text im Schulbuch aus und hopsen darin um die Wette. Das geht so: Jeder Buchstabe hat eine
> Sprungweite zugeordnet (siehe Tabelle). Bela fängt beim ersten, Amira beim zweiten Buchstaben des Textes an.
> Von dieser Startposition springen sie jeweils so viele Positionen weiter, wie die Sprungweite des Buchstabens vorgibt.
> Alles was kein Buchstabe ist, überspringen sie dabei einfach. An der neuen Position lesen sie dann jeweils den nächsten
> Buchstaben mit einer neuen Sprungweite. Das Hopsen wiederholen sie abwechselnd so lange, bis sie aus dem Text
> herausspringen. Wer dies als Erstes schafft, hat gewonnen.
>
> | Buchstabe | a | b | c | ... | z | ä | ö | ü | ß |
> |---|---|---|---|---|---|---|---|---|---|
> | Sprungweite | 1 | 2 | 3 | ... | 26 | 27 | 28 | 29 | 30 |
>
> **Junioraufgabe 2:** Schreibe ein Programm, das für einen gegebenen Text bestimmt, ob Bela oder Amira beim Texthopsen
> gewinnt. Wende dein Programm mindestens auf alle Beispiele an, die du auf den BWINF-Webseiten findest, und dokumentiere die
> Ergebnisse.

*(Jugendwettbewerb Informatik 2024, 3. Runde, Junioraufgabe 2)*

## Schritt 2: Unklarheiten klären und Festlegungen treffen

#### Frage 2.1

Welche Stellen der Aufgabe kann man verschieden verstehen? (Tipp: Notebook 23, Fall 2.) Was entscheidet das Beispiel, was
nicht?

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

1. **Groß- und Kleinbuchstaben:** Die Tabelle nennt nur Kleinbuchstaben. Im Beispiel hat das große B die Sprungweite 2. Also
   zählen Groß- und Kleinbuchstaben gleich. *Das Beispiel entscheidet.*
2. **Was ist ein Buchstabe?** Nur die 30 Zeichen der Tabelle. Andere Buchstaben wie é zählen wie Satzzeichen nicht mit.
   *Die README entscheidet.*
3. **Wer gewinnt?** „Wer dies als Erstes schafft“ heißt entweder „wer weniger Sprünge braucht“ (bei Gleichstand
   unentschieden) oder „wer zeitlich zuerst draußen ist“. Weil Bela beginnt, ist Bela bei Gleichstand zuerst draußen.
   *Das Beispiel entscheidet nicht*, denn das Aufgabenblatt sagt nicht, wer dort gewinnt.

</details>

**Unsere Festlegungen** (die gehören später wörtlich in die Dokumentation):

1. Groß- und Kleinbuchstaben haben dieselbe Sprungweite.
2. Es zählen nur die 30 Buchstaben der Tabelle, alle anderen Zeichen werden vorher entfernt.
3. Es gewinnt, wer **weniger Sprünge** braucht. Bei gleich vielen Sprüngen ist es **unentschieden**. Das Programm gibt dann
   zusätzlich aus, dass Bela zuerst draußen ist, weil Bela beginnt. So ist das Ergebnis für **beide** Lesarten ablesbar.

Die offiziellen Lösungshinweise akzeptieren beide Lesarten, ziehen aber einen Punkt ab, wenn nicht erkennbar ist, welche man
gewählt hat. Eine klare Festlegung kostet also nichts und sichert einen Punkt.

## Schritt 3: Die Lösungsidee

#### Frage 3.1

Bela und Amira hopsen abwechselnd. Muss man das im Programm auch abwechselnd simulieren, oder geht es einfacher?

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

Es geht einfacher: Die beiden **beeinflussen sich nicht**. Belas Sprünge hängen nur vom Text und von Belas Startposition ab,
nicht davon, wo Amira gerade steht. Man kann also zuerst Bela allein durch den Text hopsen lassen, dann Amira allein, und am Ende
die Anzahl der Sprünge vergleichen. Das Abwechseln spielt nur bei Gleichstand eine Rolle, und das regelt Festlegung 3.

Solche Vereinfachungen sind wertvoll: Das Programm wird kürzer und weniger fehleranfällig. In der Dokumentation begründest du sie
in einem Satz.

</details>

**Die Lösungsidee in Worten** (so ähnlich steht sie später in der Dokumentation):

> Zuerst werden aus dem Text alle Zeichen entfernt, die keine Buchstaben der Tabelle sind, und alle Buchstaben werden klein
> geschrieben. Übrig bleibt eine Folge von Buchstaben. Für jede Spielerin und jeden Spieler wird das Hopsen getrennt
> nachgespielt: Beginnend bei der Startposition wird immer die Sprungweite des Buchstabens an der aktuellen Position
> dazugezählt, bis die Position hinter dem letzten Buchstaben liegt. Dabei werden die Sprünge gezählt. Weil sich die beiden
> nicht gegenseitig beeinflussen, genügt es, am Ende die beiden Anzahlen zu vergleichen.

<details>
<summary>Tipp anzeigen</summary>

Wie in Notebook 15, Aufgabe 1.2. In `nur_buchstaben` prüfst du mit `if zeichen in WEITE:`. Ein Wörterbuch kann man mit `in` nach

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
ALPHABET = 'abcdefghijklmnopqrstuvwxyzäöüß'
WEITE = {}
for i in range(len(ALPHABET)):
    WEITE[ALPHABET[i]] = i + 1


def nur_buchstaben(text):
    ergebnis = ''
    for zeichen in text.lower():
        if zeichen in WEITE:
            ergebnis = ergebnis + zeichen
    return ergebnis


print(WEITE.get('a'), WEITE.get('z'), WEITE.get('ß'))
print(nur_buchstaben('Beispieltexte sind toll'))
print(nur_buchstaben('  "Ärger!", rief 1 Bär.'))
```

Achtung: `'ß'.lower()` ist `'ß'`, aber `'ẞ'` (das große ß) wird klein zu `'ß'`. Das kommt in normalen Texten kaum vor. Wichtig
ist, dass ä, ö, ü und ß überhaupt richtig ankommen. Dafür sorgt `encoding='utf-8'` beim Einlesen (Notebook 8). Die
Lösungshinweise nennen die Kodierung ausdrücklich als Bewertungskriterium.

```python
def anzahl_spruenge(buchstaben, start):
    position = start
    spruenge = 0
    while position < len(buchstaben):
        position = position + WEITE[buchstaben[position]]
        spruenge = spruenge + 1
    return spruenge
```
