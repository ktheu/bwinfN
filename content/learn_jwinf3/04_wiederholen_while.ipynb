# 4. Wiederholen mit `while`

Bei `for` steht vorher fest, wie oft wiederholt wird. Oft weißt du das aber nicht: „Spare, **bis** du 100 Euro hast“ oder
„Würfle, **bis** du im Ziel bist“. Dafür gibt es in Blockly den Block **wiederhole solange … mache**. In Python heißt er
`while`.

In diesem Notebook lernst du:

- wie eine `while`-Schleife funktioniert,
- wie du die Ziffern einer Zahl einzeln bearbeitest,
- was eine **Endlosschleife** ist und wie du dich davor schützt,
- wie du eine Schleife mit `break` abbrichst.

In der Wettbewerbs-Ecke spielst du ein Leiterspiel, bis du im Ziel bist, oder merkst, dass du nie ankommst.

## 1. wiederhole solange: `while`

In Blockly:

```
wiederhole solange  geld < 100
mache  erhöhe geld um 7
```

In Python:

```python
while geld < 100:
    geld = geld + 7
```

Python prüft die Bedingung **vor jedem Durchgang**. Ist sie wahr, wird der eingerückte Teil ausgeführt, danach wird
wieder geprüft. Ist sie unwahr, geht es nach der Schleife weiter. Doppelpunkt und Einrückung kennst du schon.

````python
zahl = 1
while zahl < 1000:
    print(zahl, end=' ')
    zahl = zahl * 2
print()
print('Die erste Zweierpotenz ab 1000 ist', zahl)
````

## 2. Die Ziffern einer Zahl

Mit `//` und `%` aus Notebook 1 kannst du eine Zahl in ihre Ziffern zerlegen:

| Rechnung | Ergebnis | Bedeutung |
|---|---|---|
| `4728 % 10` | `8` | die **letzte** Ziffer (Rest beim Teilen durch 10) |
| `4728 // 10` | `472` | die Zahl **ohne** die letzte Ziffer |

Das wiederholst du, **solange** noch Ziffern übrig sind, also solange die Zahl größer als 0 ist. Wie oft das ist, weißt du
vorher nicht. Ein Fall für `while`!

````python
zahl = 4728
while zahl > 0:
    ziffer = zahl % 10
    print('Ziffer', ziffer, 'abgetrennt, übrig bleibt', zahl // 10)
    zahl = zahl // 10
````

## 3. Endlosschleifen und `break`

Wenn die Bedingung **nie** unwahr wird, hört eine `while`-Schleife nie auf. Das ist eine **Endlosschleife**. Ein Beispiel
(nicht ausführen!):

```python
x = 1
while x != 10:
    x = x + 2
```

`x` wird 3, 5, 7, 9, 11, 13, … und ist nie genau 10. Das Notebook rechnet und rechnet. Am Rand der Zelle steht dann
`[*]`. Klicke in diesem Fall oben auf ■ (Stopp) oder wähle im Menü **Kernel → Restart Kernel**.

Im Wettbewerb kommen solche Schleifen vor, auch ohne Fehler im Programm. In der Wettbewerbs-Ecke siehst du ein Spiel,
das wirklich nie endet. Darum gibt es einen einfachen Schutz: Zähle die Durchgänge mit und **brich ab**, wenn es zu viele
werden. Der Blockly-Block **die Schleife abbrechen** heißt in Python `break`. Er beendet die Schleife sofort, und es geht
nach der Schleife weiter.

````python
x = 1
schritte = 0
while x != 10:
    x = x + 2
    schritte = schritte + 1
    if schritte == 100:
        print('Abbruch nach 100 Schritten, x ist', x)
        break
print('Nach der Schleife')
````

#### Frage 3.1

Dieses Programm soll 1 + 2 + … + 10 ausrechnen, hört aber nie auf. Warum? (Nicht ausführen!)

```python

```

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

`i` wird in der Schleife nie verändert. Es bleibt 1, und `1 <= 10` ist immer wahr. In die Schleife gehört noch

## 4. Suchen mit `while`

Oft suchst du die **erste** Zahl, die eine Bedingung erfüllt. Dann probierst du eine Zahl nach der anderen, **solange**
die Bedingung noch nicht erfüllt ist. Gesucht: die kleinste Zahl ab 100, die durch 7 **und** durch 11 teilbar ist.

````python
zahl = 100
while not (zahl % 7 == 0 and zahl % 11 == 0):
    zahl = zahl + 1
print('Gefunden:', zahl)
````

Die Bedingung „solange es **nicht** passt“ ist etwas verdreht. Viele schreiben es deshalb mit `while True` („wiederhole
für immer“) und brechen ab, sobald es passt:

```python
zahl = 100
while True:
    if zahl % 7 == 0 and zahl % 11 == 0:
        break
    zahl = zahl + 1
print('Gefunden:', zahl)
```

Beides ist richtig. Nimm die Form, die du leichter verstehst.

Das Ziel 20 muss man **genau** treffen. Würfelt man zu viel, läuft man den Rest zurück (wie in Notebook 2, E.3).

## Das hast du gelernt

| Blockly | Python |
|---|---|
| wiederhole solange … mache | `while bedingung:` und eingerückter Teil |
| die Schleife abbrechen | `break` |
| (wiederhole für immer) | `while True:` mit `break` darin |

- `while` prüft die Bedingung **vor** jedem Durchgang. Ist sie gleich am Anfang unwahr, läuft die Schleife gar nicht.
- In der Schleife muss sich etwas ändern, sonst entsteht eine **Endlosschleife**. Schutz: Durchgänge zählen und mit `break` abbrechen.
- Ziffern: `zahl % 10` ist die letzte Ziffer, `zahl // 10` die Zahl ohne sie.
- `for`, wenn die Anzahl vorher feststeht, `while`, wenn nicht.
