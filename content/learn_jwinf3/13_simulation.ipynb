# 13. Simulation

Viele Wettbewerbsaufgaben beschreiben einen **Ablauf**: Vögel fliegen über einen Wald, eine Spielfigur zieht über ein Brett,
Lämpchen schalten an und aus. Die einfachste Lösung ist oft, den Ablauf im Computer **nachzuspielen**, Schritt für Schritt.
Das heißt **Simulation**.

Jede Simulation besteht aus drei Teilen:

1. Der **Zustand**: alles, was man wissen muss, um weiterzumachen (Position, Richtung, Uhrzeit, welche Lampen an sind …).
2. Ein **Schritt**: eine Funktion, die aus dem jetzigen Zustand den nächsten berechnet.
3. Eine **Schleife**, die Schritte ausführt, bis die Simulation zu Ende ist, und dabei mitschreibt, was passiert
   (ein **Protokoll**).

In diesem Notebook lernst du diese drei Teile kennen, dazu zwei typische Fallen: einen Zustand versehentlich zu früh
überschreiben, und eine Simulation, die nie endet.

## 1. Zustand, Schritt, Schleife

Ein Vogel fliegt über einen Waldweg mit den Feldern 1 bis `n` hin und her. Jede Minute schaut er auf das Feld unter sich und
fliegt dann ein Feld weiter. Am Ende des Weges kehrt er um: Auf Feld `n` fliegt er als Nächstes zurück nach `n - 1`.

- **Zustand:** Position `pos` und Richtung `richtung` (+1 nach rechts, -1 nach links).
- **Schritt:** Würde der Vogel den Weg verlassen, dreht er die Richtung um. Dann fliegt er ein Feld weiter.

#### Aufgabe 1.1

Schreibe `schritt(pos, richtung, n)`, die die neue Position und die neue Richtung zurückgibt. Lass den Vogel dann 12 Minuten
fliegen (Start auf Feld 1, Richtung +1, `n = 5`) und gib in jeder Minute das Feld aus.

**Erwartete Ausgabe:**
```
1 2 3 4 5 4 3 2 1 2 3 4
```

````python
def schritt(pos, richtung, n):
    return pos, richtung        # ersetze diese Zeile

pos = 1
richtung = 1
for minute in range(12):
    print(pos, end=' ')
    pos, richtung = schritt(pos, richtung, 5)
print()
````

<details>
<summary>Tipp anzeigen</summary>

Prüfe zuerst, ob `pos + richtung` außerhalb von 1 bis `n` läge. Wenn ja, `richtung = -richtung`. Danach `pos = pos + richtung`
und `return pos, richtung` (zwei Rückgabewerte, Notebook 7).

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

def schritt(pos, richtung, n):
    if pos + richtung < 1 or pos + richtung > n:
        richtung = -richtung            # am Rand umkehren

## 2. Der Zustand ist eine Liste: Vorsicht beim Überschreiben

Eine Lichterkette hat 15 Lämpchen, nur das mittlere leuchtet. In jedem Schritt gilt für jedes Lämpchen: Es leuchtet im
nächsten Schritt genau dann, wenn **genau eines** seiner beiden Nachbarlämpchen jetzt leuchtet. Die Lämpchen außerhalb der
Kette zählen als aus.

Der Zustand ist hier eine Liste aus 0 (aus) und 1 (an). Die Regel benutzt den **jetzigen** Zustand der Nachbarn. Darum musst
du den neuen Zustand in einer **neuen** Liste aufbauen und darfst die alte erst danach ersetzen.

#### Aufgabe 2.1

Schreibe `naechster(lampen)`, die die Liste für den nächsten Schritt zurückgibt. Gib dann 8 Schritte aus, ein `#` für an, einen
Punkt für aus.

**Erwartete Ausgabe:**
```
.......#.......
......#.#......
.....#...#.....
....#.#.#.#....
...#.......#...
..#.#.....#.#..
.#...#...#...#.
#.#.#.#.#.#.#.#
```

````python
def naechster(lampen):
    return lampen       # ersetze diese Zeile

lampen = [0] * 15
lampen[7] = 1
for t in range(8):
    zeile = ''
    for x in lampen:
        if x == 1:
            zeile = zeile + '#'
        else:
            zeile = zeile + '.'
    print(zeile)
    lampen = naechster(lampen)
````

<details>
<summary>Tipp anzeigen</summary>

`neu = [0] * n`, dann für jedes `i` die beiden Nachbarn aus `lampen` lesen (am Rand 0) und `neu[i] = 1` setzen, wenn genau
einer leuchtet. Am Ende `return neu`.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def naechster(lampen):
    n = len(lampen)
    neu = [0] * n
    for i in range(n):
        links = 0
        rechts = 0
        if i > 0:
            links = lampen[i - 1]
        if i < n - 1:
            rechts = lampen[i + 1]
        if links + rechts == 1:
            neu[i] = 1
    return neu

lampen = [0] * 15
lampen[7] = 1
for t in range(8):
    zeile = ''
    for x in lampen:
        if x == 1:
            zeile = zeile + '#'
        else:
            zeile = zeile + '.'
    print(zeile)
    lampen = naechster(lampen)
```

Aus der einfachen Regel entsteht ein überraschend regelmäßiges Muster, das **Sierpinski-Dreieck**.

</details>

## 3. Schon gesehen? Endlosschleifen sicher erkennen

In Notebook 4 hast du das Leiterspiel nach 100 Würfen abgebrochen. Das ist unsicher: Vielleicht wäre die Figur im 101. Wurf
angekommen. Es geht besser: Wenn eine Simulation **wieder in einem Zustand ankommt, in dem sie schon einmal war**, geht ab
dort alles genauso weiter wie beim ersten Mal. Sie dreht sich also für immer im Kreis.

Darum merkst du dir alle Zustände in einer Liste `gesehen`. Ist der neue Zustand schon darin, steckst du in einem **Kreis**.
Mit `gesehen.index(zustand)` findest du sogar heraus, wo der Kreis beginnt.

Ein Beispiel: Die Zahlenfolge, bei der man immer `x * x + 1` rechnet und nur die letzten zwei Ziffern behält (`% 100`).

````python
x = 3
gesehen = []
while x not in gesehen:
    gesehen.append(x)
    x = (x * x + 1) % 100
print('Folge:', gesehen)
start = gesehen.index(x)
print('Der Kreis beginnt bei', x, 'nach', start, 'Schritten und ist', len(gesehen) - start, 'Zahlen lang:', gesehen[start:])
````

def spiele(wurf):

## Das hast du gelernt

| Teil einer Simulation | So geht es |
|---|---|
| Zustand | Variablen oder Listen mit allem, was die Zukunft bestimmt |
| Schritt | eine Funktion: alter Zustand rein, neuer Zustand raus |
| Schleife | `for` bei fester Dauer, `while` bis zum Ende, mit Protokoll-Ausgabe |
| neuer Zustand aus altem | in eine **neue** Liste schreiben, nicht die alte überschreiben, solange sie noch gelesen wird |
| Endlosschleife erkennen | alle Zustände in `gesehen` merken; ist der neue schon drin, läuft die Simulation im Kreis |
