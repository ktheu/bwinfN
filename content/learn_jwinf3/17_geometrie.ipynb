## 1. Abstand zweier Punkte

Zwischen den Punkten P = (x1, y1) und Q = (x2, y2) liegt waagerecht der Unterschied `dx = x2 - x1` und senkrecht `dy = y2 - y1`.
Die Strecke PQ ist die längste Seite eines rechtwinkligen Dreiecks mit den Seiten `dx` und `dy`. Nach dem **Satz des
Pythagoras** gilt:

    abstand² = dx² + dy²,  also  abstand = Wurzel aus (dx² + dy²)

Beispiel: Von (0, 0) nach (3, 4) ist `dx = 3`, `dy = 4`, und 3² + 4² = 9 + 16 = 25. Der Abstand ist also 5.

In Python: `x ** 2` ist x hoch 2, und `math.sqrt(z)` ist die Wurzel aus z (Blockly: „Quadratwurzel“). Davor muss `import math`
stehen.

````python
import math


def abstand(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    return math.sqrt(dx ** 2 + dy ** 2)


print(abstand(0, 0, 3, 4))
print(abstand(1, 1, 4, 5))
print(abstand(-2, 3, 1, -1))
````

deren Quadrat höchstens `z` ist. Auch das rechnet exakt.

## 3. Abstand zwischen Punkt und Rechteck

Wie weit ist ein Punkt P von einem Rechteck entfernt? Gesucht ist der **nächste Punkt** des Rechtecks. Für die x-Richtung gibt
es drei Fälle:

- P liegt links vom Rechteck (`px < x1`): Der Unterschied ist `x1 - px`.
- P liegt rechts davon (`px > x2`): Der Unterschied ist `px - x2`.
- P liegt dazwischen: Der Unterschied ist 0.

Kurz: `dx = max(x1 - px, 0, px - x2)`. Genauso für y. Danach geht es mit Pythagoras weiter.

## 4. Links oder rechts?

Liegt ein Punkt P links oder rechts von der Geraden durch A und B (wenn man von A nach B schaut)? Das beantwortet eine Rechnung,
die nur Malnehmen und Abziehen braucht, das sogenannte **Kreuzprodukt**:

    wert = (bx - ax) · (py - ay) - (by - ay) · (px - ax)

- `wert > 0`: P liegt **links**.
- `wert < 0`: P liegt **rechts**.
- `wert == 0`: P liegt **genau auf** der Geraden.

Warum das so ist, lernst du später im Mathematikunterricht. Hier reicht es, die Formel zu benutzen und an Beispielen zu prüfen.
Bei ganzzahligen Koordinaten ist das Ergebnis exakt, auch der wichtige Fall „genau auf der Geraden“.

## 5. In welcher Richtung liegt ein Punkt?

Manchmal braucht man die **Richtung** zu einem Punkt als Winkel, zum Beispiel um Punkte rund um eine Mitte zu sortieren.
`math.atan2(dy, dx)` liefert den Winkel zwischen der Richtung nach rechts (Osten) und der Richtung zum Punkt, gegen den
Uhrzeigersinn gemessen. Das Ergebnis ist im **Bogenmaß**, `math.degrees(...)` macht Grad daraus. Achtung: Erst `dy`, dann `dx`!

`atan2` liefert Winkel von -180° bis 180°. Für Winkel von 0° bis 360° rechnest du im Kreis (Notebook 11): `winkel % 360`.

In Blockly gibt es nur `atan` (über die Auswahl im Block „sin“). `atan2` ist besser, weil es alle vier Richtungen (rechts oben,
links oben, links unten, rechts unten) richtig unterscheidet.
