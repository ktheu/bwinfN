Computer sind gut darin, dasselbe sehr oft zu tun, ohne müde zu werden. In Blockly benutzt du dafür die türkisen

## 1. wiederhole … mal

In Blockly:

```
wiederhole 3 mal:
mache  schreibe „Hallo“
```

In Python:

```python
for i in range(3):
    print('Hallo')
```

Das kennst du schon von `if`: Hinter der ersten Zeile steht ein **Doppelpunkt**, und alles, was wiederholt werden soll,
ist **eingerückt**. Die erste Zeile, die nicht mehr eingerückt ist, kommt erst **nach** der Schleife dran.

Das `i` ist eine Variable, die die Durchgänge mitzählt. Wofür sie gut ist, siehst du in Abschnitt 2.

````python
for i in range(3):
    print('Hallo')
print('Fertig')
````

## 2. zähle i von … bis …

Der Blockly-Block **zähle i von 1 bis 10 in Schritten von 1** heißt in Python:

```python
for i in range(1, 11):
```

**Achtung, das ist der wichtigste Unterschied zu Blockly:** Bei `range` ist die **Endzahl nicht mehr dabei**.
`range(1, 11)` zählt 1, 2, …, 10. Wenn du bis 10 zählen willst, schreibst du also 11.

| Blockly | Python | Werte von `i` |
|---|---|---|
| zähle i von 1 bis 10 in Schritten von 1 | `range(1, 11)` | 1, 2, …, 10 |
| zähle i von 0 bis 4 in Schritten von 1 | `range(5)` | 0, 1, 2, 3, 4 |
| zähle i von 2 bis 10 in Schritten von 2 | `range(2, 11, 2)` | 2, 4, 6, 8, 10 |
| (rückwärts) | `range(10, 0, -1)` | 10, 9, …, 1 |

- Mit **einer** Zahl beginnt `range` bei 0: `range(5)` zählt fünfmal, nämlich 0 bis 4. Deshalb wiederholt
  `for i in range(3):` genau dreimal.
- Die dritte Zahl ist die **Schrittweite**. Mit `-1` zählt Python rückwärts.

Noch ein Trick: Normalerweise beginnt jedes `print` eine neue Zeile. Mit `print(i, end=' ')` hängt Python statt des
Zeilenwechsels ein Leerzeichen an. So stehen alle Zahlen in einer Zeile. Ein leeres `print()` beendet die Zeile.

````python
for i in range(1, 6):
    print(i, end=' ')
print()
print('Und jetzt jede Zahl in einer eigenen Zeile:')
for i in range(1, 4):
    print('Durchgang', i)
````

## 3. Summen bilden

Der kleine Carl Friedrich Gauß sollte in der Schule alle Zahlen von 1 bis 100 zusammenzählen. Für Python ist das
eine Kleinigkeit. Das Muster dafür brauchst du sehr oft:

1. **Vor** der Schleife: eine Variable auf 0 setzen (Blockly: „setze summe auf 0“).
2. **In** der Schleife: die Variable erhöhen (Blockly: „erhöhe summe um i“).
3. **Nach** der Schleife: das Ergebnis ausgeben.

````python
summe = 0
for i in range(1, 101):
    summe = summe + i
print('Summe:', summe)
````

## 4. Schleife mit `if`: zählen und auswählen

In eine Schleife kannst du alles hineinstecken, auch ein `if`. Dann wird in **jedem** Durchgang neu entschieden.
Achte auf die Einrückung: Was zum `if` gehört, wird noch einmal eingerückt, also mit 8 Leerzeichen.

Ein **Zähler** funktioniert wie eine Summe, nur dass du jedes Mal um 1 erhöhst:

````python
anzahl = 0
for i in range(1, 21):
    if i % 2 == 0:
        anzahl = anzahl + 1
print('Gerade Zahlen von 1 bis 20:', anzahl)
````

<details>

Probiere alle Zahlen `t` von 1 bis `zahl` durch (Ende: `zahl + 1`). `t` ist ein Teiler, wenn `zahl % t == 0`.


13 hat nur die Teiler 1 und 13: Es ist eine **Primzahl**. Das Muster „alle Möglichkeiten durchprobieren und die

#### Aufgabe E.1

Gib für 7 Spiele und 3 Tüten aus, in welche Tüte jedes Spiel kommt.

So kannst du vorgehen: Merke dir in der Variablen `tuete`, welche Tüte gerade dran ist. Nach jedem Spiel ist die nächste
Tüte dran. Wird `tuete` dabei größer als die Anzahl der Tüten, geht es wieder bei Tüte 1 los.

**Erwartete Ausgabe:**
```
Spiel 1 -> Tüte 1
Spiel 2 -> Tüte 2
Spiel 3 -> Tüte 3
Spiel 4 -> Tüte 1
Spiel 5 -> Tüte 2
Spiel 6 -> Tüte 3
Spiel 7 -> Tüte 1
```

````python
spiele = 7
tueten = 3
tuete = 1
# dein Code
````

<details>
<summary>Tipp anzeigen</summary>

In der Schleife: erst ausgeben, dann `tuete = tuete + 1`, dann mit `if tuete > tueten:` wieder auf 1 setzen.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
spiele = 7

tuete = 1

    print('Spiel', spiel, '-> Tüte', tuete)
    tuete = tuete + 1
    if tuete > tueten:
        tuete = 1
```

Mit dem Rest beim Teilen geht es auch ohne `if`: `tuete = (spiel - 1) % tueten + 1`. Der Rest `% tueten` läuft immer
0, 1, 2, 0, 1, 2, …, und `+ 1` macht daraus die Tütennummern 1, 2, 3. Dieses **Rechnen im Kreis** begegnet dir im
Wettbewerb oft.

</details>

## Das hast du gelernt

| Blockly | Python |
|---|---|
| wiederhole 3 mal | `for i in range(3):` |
| zähle i von 1 bis 10 in Schritten von 1 | `for i in range(1, 11):` (Ende **nicht** dabei!) |
| zähle i von 2 bis 10 in Schritten von 2 | `for i in range(2, 11, 2):` |
| (rückwärts) | `for i in range(10, 0, -1):` |
| setze summe auf 0, erhöhe summe um i | `summe = 0` vor der Schleife, `summe = summe + i` darin |

- Was wiederholt werden soll, ist **eingerückt**. Ein `if` in der Schleife wird noch einmal eingerückt.
- `print(x, end=' ')` schreibt ohne Zeilenwechsel, `print()` beendet die Zeile.
- Summen und Zähler: **vor** der Schleife auf 0 setzen, **in** der Schleife erhöhen, **nach** der Schleife ausgeben.
