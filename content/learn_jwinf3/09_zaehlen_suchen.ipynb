# 9. Zählen, Suchen, das Beste merken

Ab jetzt geht es um **Werkzeuge**, die in vielen Wettbewerbsaufgaben vorkommen. Drei davon sind so häufig, dass du sie im
Schlaf können solltest:

- **Suchen:** Gibt es ein Element mit einer bestimmten Eigenschaft, und wo ist das erste?
- **Zählen:** Wie viele Elemente haben eine Eigenschaft, und wie oft kommt jeder Wert vor?
- **Das Beste merken:** Welches Element ist das größte, kleinste, nächste, längste … und was ist bei **Gleichstand**?

Einiges kennst du schon aus Notebook 6. Hier lernst du die Muster genauer kennen und verpackst sie in Funktionen. In der
Wettbewerbs-Ecke suchst du den beliebtesten Termin für ein Treffen und die höchste erlaubte Windkraftanlage.

## 1. Suchen

Die einfachste Art zu suchen: **der Reihe nach** alle Elemente ansehen, bis man das Gesuchte findet. Das heißt **lineare
Suche**. Sobald das Element gefunden ist, kann die Funktion mit `return` sofort aufhören (Notebook 7). Erst wenn die Schleife
ganz durchgelaufen ist, steht fest: nicht vorhanden.

````python
def kommt_vor(liste, x):
    for wert in liste:
        if wert == x:
            return True             # gefunden: sofort fertig
    return False                    # alle angesehen, nichts gefunden


zahlen = [4, 8, 15, 16, 23, 42]
print(kommt_vor(zahlen, 15), kommt_vor(zahlen, 7))
````

Für „kommt vor“ gibt es in Python schon `x in liste`. Das Muster brauchst du trotzdem, denn meistens suchst du nicht einen
bestimmten Wert, sondern ein Element mit einer **Eigenschaft**: das erste Wort mit einem z, die erste Zahl über 100, den
ersten freien Platz. Dafür gibt es kein fertiges Werkzeug.

Willst du wissen, **wo** das Element steht, gehst du die Positionen durch und gibst die Position zurück. Wenn nichts
gefunden wird, gibst du `-1` zurück, wie `find` bei Texten.

## 2. Zählen

Ein **Zähler** beginnt bei 0 und wird erhöht, wenn eine Bedingung zutrifft. Das kennst du. Oft willst du aber nicht nur eine
Sache zählen, sondern **für jeden Wert, wie oft er vorkommt**: wie oft jede Note, wie oft jeder Buchstabe, wie oft jede
Augenzahl. Dafür nimmst du eine **Zählliste** (Notebook 6): Die Position ist der Wert, der Inhalt die Anzahl.

Damit das klappt, muss jeder Wert eine Position bekommen:

- Noten 1 bis 6: Liste mit 7 Plätzen, Position = Note (Platz 0 bleibt leer).
- Buchstaben a bis z: Liste mit 26 Plätzen, Position = `ord(b) - ord('a')` (Notebook 5).

````python
wuerfe = [3, 6, 1, 3, 3, 5, 6, 2, 3, 1]
anzahl = [0] * 7                    # Platz 1 bis 6 für die Augenzahlen
for w in wuerfe:
    anzahl[w] = anzahl[w] + 1
for augen in range(1, 7):
    print(augen, '*' * anzahl[augen])
````

`'*' * anzahl[augen]` malt so viele Sternchen, wie die Zahl angibt. So entsteht ein kleines **Balkendiagramm** aus Text.

## 3. Das Beste merken, mit eigenem Maß

`max` und `min` finden die größte und kleinste **Zahl**. Oft ist „das Beste“ aber nach einem anderen **Maß** gemeint: das
Wort mit den meisten Vokalen, die Zahl, die am nächsten an 100 liegt, das Haus mit dem kleinsten Abstand. Dann benutzt du das
Muster aus Notebook 6, Aufgabe 3.2, und vergleichst nach deinem Maß:

```python
bestes = liste[0]
for x in liste:
    if mass(x) > mass(bestes):     # bzw. <, wenn das Kleinste gesucht ist

```

Das Maß schreibst du am besten als eigene kleine Funktion. Für Abstände hilft `abs(x)`, der **Betrag**: `abs(-7)` ist `7`,
`abs(7)` auch. Der Abstand zweier Zahlen `a` und `b` ist `abs(a - b)`.

## 4. Gleichstand: alle Besten finden

Haben mehrere Elemente denselben besten Wert, liefert das Muster aus Abschnitt 3 nur **eines** davon: mit `>` das erste,
mit `>=` das letzte. Im Wettbewerb ist das eine typische Stelle, an der man genau hinsehen muss (Notebook 23): Will die
Aufgabe einen Besten oder alle? Wenn sie „einen“ verlangt, sag in der Dokumentation, welchen du nimmst.

Alle Besten findest du in **zwei Durchgängen**: zuerst den besten Wert bestimmen, dann alle Positionen mit genau diesem
Wert sammeln.

````python
punkte = [17, 30, 9, 30, 21, 30]
bester_wert = max(punkte)
alle_besten = []
for i in range(len(punkte)):
    if punkte[i] == bester_wert:
        alle_besten.append(i)
print('Bester Wert', bester_wert, 'an den Positionen', alle_besten)
````

## Wettbewerbs-Ecke 1: Treffsicherheit

*Angelehnt an die Junioraufgabe „Treffsicherheit“ (Jugendwettbewerb 2021).*

Ada schlägt für das Treffen ihrer Clique 7 Termine vor. Jedes der 6 Mitglieder bewertet jeden Termin mit 0 (grün, sehr gut),
1 (gelb, mäßig) oder 2 (rot, gar nicht). Kleiner ist also besser. In der **Tabelle** ist jede Zeile ein Mitglied und jede
Spalte ein Termin. In Python ist das eine **Liste von Listen**: `tabelle[i]` ist die Zeile von Mitglied `i`, und
`tabelle[i][t]` ist seine Bewertung von Termin `t` (beides ab 0 gezählt). Mehr zu Tabellen in Notebook 11.

Ein Termin ist **allseits beliebt**, wenn **keinem** Mitglied ein anderer Termin besser gefällt. Ist kein Termin allseits
beliebt, will Ada wissen, wie viele Einträge man mindestens ändern müsste, damit einer es wird.

Die Idee: Gefällt einem Mitglied ein anderer Termin besser als Termin `t`, ist das ein **Konflikt**. Jeder Konflikt lässt sich
mit **einer** Änderung beheben: Man setzt die Bewertung dieses Mitglieds für Termin `t` auf seine beste Bewertung. Ein
Mitglied hat einen Konflikt, wenn seine Bewertung von `t` schlechter ist als seine beste Bewertung, also
`min(tabelle[i]) < tabelle[i][t]`.

## Wettbewerbs-Ecke 2: Zum Winde verweht

*Angelehnt an die Junioraufgabe „Zum Winde verweht“ (Jugendwettbewerb 2021).*

Eine Windkraftanlage mit der Höhe H Metern muss mindestens 10 · H Meter von **jedem** Wohnhaus entfernt sein. Wie hoch darf
eine Anlage an einem bestimmten Standort höchstens sein? Entscheidend ist das **nächste** Haus: Ist es `d` Meter entfernt,
darf die Anlage höchstens `d / 10` Meter hoch sein.

Den Abstand zweier Punkte berechnet man mit dem Satz des Pythagoras: Wurzel aus (Unterschied in x)² + (Unterschied in y)².
Die Wurzel ist dasselbe wie „hoch 0,5“, in Python `** 0.5`. Die Funktion dafür steht schon im Gerüst. (Genaueres zur
Geometrie in Notebook 17.)

#### Aufgabe E.3

Die Häuser stehen in den Listen `hx` und `hy` (x- und y-Koordinaten in Metern). Schreibe `naechstes_haus(wx, wy)`, die den
**kleinsten** Abstand vom Standort `(wx, wy)` zu einem Haus zurückgibt. Gib damit für jeden geplanten Standort die
höchste erlaubte Höhe aus.

**Erwartete Ausgabe:**
```
5.0
Am Standort 100 200 höchstens 22.36 m
Am Standort 400 0 höchstens 26.92 m
Am Standort -100 -300 höchstens 31.62 m
```

````python
hx = [0, 300, 150, -200, 500]
hy = [0, 400, -100, 50, 250]
standorte_x = [100, 400, -100]
standorte_y = [200, 0, -300]


def abstand(x1, y1, x2, y2):
    return ((x1 - x2) ** 2 + (y1 - y2) ** 2) ** 0.5


def naechstes_haus(wx, wy):
    return 0            # ersetze diese Zeile


print(abstand(0, 0, 3, 4))      # erwartet 5.0
````

<details>
<summary>Tipp anzeigen</summary>

Das Muster „das Kleinste merken“: Beginne mit dem Abstand zum ersten Haus und gehe alle Häuser durch. Die Höhe ist der
kleinste Abstand geteilt durch 10.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
hx = [0, 300, 150, -200, 500]
hy = [0, 400, -100, 50, 250]
standorte_x = [100, 400, -100]
standorte_y = [200, 0, -300]


def abstand(x1, y1, x2, y2):
    return ((x1 - x2) ** 2 + (y1 - y2) ** 2) ** 0.5


def naechstes_haus(wx, wy):
    kleinster = abstand(wx, wy, hx[0], hy[0])
    for i in range(len(hx)):
        d = abstand(wx, wy, hx[i], hy[i])
        if d < kleinster:
            kleinster = d
    return kleinster


print(abstand(0, 0, 3, 4))      # erwartet 5.0
for s in range(len(standorte_x)):
    d = naechstes_haus(standorte_x[s], standorte_y[s])
    hoehe = d / 10
    print('Am Standort', standorte_x[s], standorte_y[s], 'höchstens', int(hoehe * 100) / 100, 'm')
```

Achtung beim Runden: Am zweiten Standort ist die erlaubte Höhe 26,926… m. Normales Runden auf zwei Nachkommastellen ergäbe
26,93 m, und das wäre schon **zu hoch**! Deshalb wird hier abgeschnitten: `int(hoehe * 100) / 100` ergibt 26,92. Bei
„höchstens“ muss man immer **abrunden**. Auch das ist eine kleine Entscheidung für die Dokumentation.
