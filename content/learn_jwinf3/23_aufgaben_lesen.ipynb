## Fünf Schritte zum Lesen einer Aufgabe

1. **Eingabe und Ausgabe:** Was bekommt das Programm (aus den Beispieldateien), und was soll es ausgeben?
2. **Regeln herausschreiben:** Schreibe jede Regel in eigenen Worten als eigenen Punkt auf. Die Geschichte drumherum lässt
   du weg.
3. **Beispiel von Hand nachrechnen:** Fast jedes Aufgabenblatt enthält ein Beispiel mit Ergebnis. Rechne es mit Bleistift
   nach. Dann weißt du, ob du die Regeln richtig verstanden hast. Später ist es dein erster Test für das Programm.
4. **Unklarheiten suchen:** Achte besonders auf
   - **Grenzen:** „bis“, „von … bis“, „höchstens“, „mindestens“: mit oder ohne den Grenzwert?
   - **Zählen:** „n weiter“, „der erste“: ab 0 oder ab 1, mitgezählt oder übersprungen?
   - **Gleichstand:** Was passiert, wenn zwei gleich gut sind?
   - **„möglichst“, „am besten“:** Woran misst man das?
   - **Sonderfälle:** 0, leer, keine Lösung, Rand, Rundung, Groß- und Kleinschreibung, Zeichen, die nicht zählen.
5. **Entscheiden und aufschreiben:** Wähle eine Lesart, begründe sie und schreibe sie in die Dokumentation. Wenn es nicht
   viel Arbeit macht, rechne mit **beiden** Lesarten und vergleiche die Ergebnisse.

Für die Dokumentation hilft ein fester Satzbau:

> *Die Aufgabe lässt offen, ob … . Wir gehen davon aus, dass …, weil … . Mit der anderen Lesart ergäbe sich … .*

Jetzt prüfst du mit einem Programm, wie viel die Lesart von „13 – 15“ ausmacht. Die Funktion `braucht(i, tag, stunde)`
entscheidet, ob Klasse Nummer `i` zur angegebenen Zeit Bälle braucht. Die Funktion `maximum()` probiert alle Tage und Stunden
durch und benutzt dabei `braucht`. Zuerst gilt **Lesart A: Ende ausgeschlossen**.

````python
klassen = ['6a', '6b', '7a', '7a', '8a', '8b']
tage = ['Montag', 'Dienstag', 'Montag', 'Dienstag', 'Montag', 'Freitag']
beginn = [14, 14, 13, 13, 15, 13]
ende = [16, 16, 15, 15, 17, 19]
kinder = [32, 27, 28, 28, 26, 15]
wochentage = ['Montag', 'Dienstag', 'Mittwoch', 'Donnerstag', 'Freitag']


def braucht(i, tag, stunde):
    """Lesart A: Eine Stunde von 13 bis 15 Uhr braucht Bälle um 13 und 14 Uhr."""
    return tage[i] == tag and beginn[i] <= stunde < ende[i]


def maximum():
    bester = 0
    bester_tag = ''
    beste_stunde = 0
    for tag in wochentage:
        for stunde in range(24):
            baelle = 0
            for i in range(len(kinder)):
                if braucht(i, tag, stunde):
                    baelle = baelle + kinder[i]
            if baelle > bester:
                bester = baelle
                bester_tag = tag
                beste_stunde = stunde
    return bester, bester_tag, beste_stunde


print(maximum())
````

#### Frage 3.1

Was kann „springe n Wörter weiter“ bedeuten? Probiere es an diesem Satz aus: Du stehst auf dem Wort **„Es“** (2 Buchstaben).

`Es gingen zwei Parallelen ins Endlose hinaus`

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

- **Lesart A: n Positionen weiter.** Von „Es“ (Position 0) geht es 2 Positionen weiter auf Position 2: **„zwei“**. Man zählt
  „gingen, zwei“ und landet auf dem zweiten gezählten Wort.
- **Lesart B: n Wörter überspringen.** Man überspringt „gingen“ und „zwei“ und landet auf Position 3: **„Parallelen“**.

Beides kann man aus dem Satz herauslesen. Die offiziellen Lösungshinweise erlauben beide Lesarten, wenn man sagt, welche
man benutzt. Weitere Fragen: Zählen Satzzeichen wie das Komma in „hinaus,“ als Buchstaben? (Nein, nur Buchstaben.) Was ist

Jetzt prüfst du, ob die Lesart das Ergebnis verändert. Dazu nimmst du die ersten beiden Strophen des Gedichts (24 Wörter).
Die Funktion `sprung(wort)` legt die Lesart fest, zuerst Lesart A.

````python
text = """Es gingen zwei Parallelen
ins Endlose hinaus,
zwei kerzengerade Seelen
und aus solidem Haus.

Sie wollten sich nicht schneiden
bis an ihr seliges Grab:"""
woerter = text.split()


def laenge(wort):
    """Anzahl der Buchstaben, Satzzeichen zählen nicht mit."""
    anzahl = 0
    for zeichen in wort:
        if zeichen.isalpha():
            anzahl = anzahl + 1
    return anzahl


def sprung(wort):
    return laenge(wort)                 # Lesart A: n Positionen weiter


print(len(woerter), 'Wörter')
````

#### Frage 4.1

Wie könnte man messen, wie „quadratisch“ ein Rechteck mit den Seiten `a` und `b` ist? Finde mindestens zwei Möglichkeiten.
Gibt es eine, die gar nicht gut funktioniert?

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

<details>
<summary>Lösung anzeigen (erst selbst überlegen!)</summary>

- **Verhältnis** der längeren zur kürzeren Seite: `max(a, b) / min(a, b)`. Ein Quadrat hat den Wert 1, je größer, desto
  schiefer. Das ist ein gutes Maß.
- **Summe beider Verhältnisse:** `a / b + b / a`. Ein Quadrat hat den Wert 2. So machen es die offiziellen Lösungshinweise.
- **Unterschied** der Seiten: `abs(a - b)`. Das hat einen Nachteil: Ein Garten mit 2 m × 3 m ist genauso schief wie einer mit
  20 m × 30 m, aber der Unterschied ist einmal 1 und einmal 10. Das Maß hängt von der Größe ab.

Welches Maß du nimmst, ist deine Entscheidung, aber du musst sie begründen.

</details>

## Das hast du gelernt

- **Erst lesen, dann programmieren:** Eingabe und Ausgabe, Regeln in eigenen Worten, Beispiel von Hand nachrechnen.
- **Typische Unklarheiten:** Grenzen (mit oder ohne Ende), Zählen (n weiter oder n überspringen), Gleichstand,
  „möglichst“ (Maß festlegen), Sonderfälle (Start, leer, 0, Rundung, Endlosschleife).
- **Das Beispiel vom Aufgabenblatt entscheidet oft**, welche Lesart gemeint ist (Bällebad: 60 statt 86).
- Lässt sich eine Frage nicht klären: **entscheiden, begründen, dokumentieren**, wenn möglich beide Lesarten rechnen.
- Lesarten lassen sich im Programm gut umschalten, wenn sie in **einer** kleinen Funktion stecken, die man neu definiert.
