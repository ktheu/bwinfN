## 1. Sortieren

| Python | Bedeutung |
|---|---|
| `sorted(liste)` | **neue** Liste, aufsteigend sortiert (Blockly: „numerisch aufsteigend sortieren“) |
| `sorted(liste, reverse=True)` | absteigend |
| `liste.sort()` | sortiert die Liste **selbst** (die alte Reihenfolge ist weg) |

Texte werden alphabetisch sortiert, aber nach den Nummern aus `ord` (Notebook 5): Großbuchstaben kommen vor Kleinbuchstaben, und
Umlaute ganz am Ende.

**Sortieren nach einem bestimmten Wert:** Eine Liste von Listen wird nach dem **ersten** Element sortiert, bei Gleichstand nach
dem zweiten und so weiter. Willst du Namen nach Punkten sortieren, baust du also Paare `[punkte, name]` und sortierst die.

#### Aufgabe 1.1: Was wird ausgegeben?

Überlege zuerst, was der Code ausgibt, und schreibe es auf. Führe die Zelle erst danach aus.

## 2. Gierige Strategien

Eine **gierige** Strategie trifft jede Entscheidung so, dass sie **im Moment** am besten aussieht, und denkt nicht an später.
Beispiel Wechselgeld: Um 88 Cent mit möglichst wenigen Münzen herauszugeben, nimmt man immer die **größte** Münze, die noch passt:
50, 20, 10, 5, 2, 1. Das sind 6 Münzen.

Gierige Strategien sind einfach zu programmieren und schnell. Die große Frage ist: **Liefern sie die beste Lösung?**

<summary>Tipp anzeigen</summary>

Gehe die Münzen von groß nach klein durch (`sorted(muenzen, reverse=True)`). Solange die Münze noch in den Restbetrag passt,

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def wechselgeld(betrag, muenzen):
    ergebnis = []
    for m in sorted(muenzen, reverse=True):
        while betrag >= m:
            ergebnis.append(m)
            betrag = betrag - m
    return ergebnis


euro = [1, 2, 5, 10, 20, 50]
print(wechselgeld(88, euro))
print(wechselgeld(43, euro))
```

<details>
<summary>Tipp anzeigen</summary>

Zwei Schleifen für die Anzahl `c` der 3er und `d` der 4er. Was übrig bleibt (`rest`), wird mit 1ern bezahlt, wenn es nicht
negativ ist. Die Anzahl der Münzen ist dann `rest + c + d`.

</details>

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def beste_anzahl(betrag):
    beste = betrag          # nur 1er-Münzen geht immer
    for c in range(betrag // 3 + 1):
        for d in range(betrag // 4 + 1):
            rest = betrag - 3 * c - 4 * d
            if rest >= 0 and rest + c + d < beste:
                beste = rest + c + d
    return beste


print(beste_anzahl(6))
for b in range(1, 21):
    gierig = len(wechselgeld(b, [1, 3, 4]))
    if gierig > beste_anzahl(b):
        print(b, 'Taler: gierig', wechselgeld(b, [1, 3, 4]), 'aber nur', beste_anzahl(b), 'Münzen nötig')
```

Bei 6 Talern nimmt die gierige Strategie 4 + 1 + 1, besser ist 3 + 3. Die gierige Wahl der 4 sah im Moment gut aus, war aber
ein Fehler. Bei den Euro-Münzen passiert das nie (das kann man beweisen). Ob gierig genügt, hängt also von der Aufgabe ab, und

## 3. Strategien gegeneinander testen

Wenn man nicht beweisen kann, welche Strategie die beste ist, **vergleicht** man sie an vielen Beispielen: an den Beispielen der
Aufgabe und an eigenen, ausgedachten Beispielen, die eine Strategie besonders gut oder schlecht aussehen lassen.

Damit du mehrere Strategien mit **einer** Funktion ausprobieren kannst, gibst du der Funktion einen Parameter mit dem **Namen**
der Strategie, zum Beispiel `'A'` oder `'B'`, und unterscheidest darin mit `if`. Die Ergebnisse gibst du als **Tabelle** aus
(f-Strings mit Breite, Notebook 8). So sieht man auf einen Blick, wann welche Strategie besser ist.

## Wettbewerbs-Ecke 1: Fähre füllen

*Angelehnt an die Junioraufgabe „Fähre füllen“ (Bundeswettbewerb Informatik 2014).*

Eine Autofähre hat drei Parkbahnen, jede 20 m lang. Der Fährbegleiter weist die wartenden Fahrzeuge der Reihe nach einer Bahn
zu. Er kennt dabei nur das Fahrzeug, das gerade vorne steht, nicht die späteren. Zwischen zwei Fahrzeugen auf derselben Bahn
bleiben 30 cm frei. Sobald das vorderste Fahrzeug nicht mehr passt, legt die Fähre ab. Gesucht sind zwei Strategien, von denen
mal die eine, mal die andere mehr Fahrzeuge unterbringt.

- **Strategie A, „der Reihe nach“:** Nimm die erste Bahn (1, 2, 3), in die das Fahrzeug passt.
- **Strategie B, „balanciert“:** Nimm die Bahn, auf der am wenigsten belegt ist (bei Gleichstand die erste). Passt es dort nicht,
  passt es nirgends.

Die Beispieldateien enthalten die Längen in Metern, zum Beispiel `daten/faehre2.txt`:

```
4.14 3.63 3.92 7.95 5.23 3.30 4.86 15.06
```

Rechne in **Zentimetern**, dann sind alle Längen ganze Zahlen (Notebook 17): `round(float(x) * 100)`.

<summary>Tipp anzeigen</summary>

Für jedes Fahrzeug suchst du eine Bahn (`bahn = -1` heißt: noch keine gefunden). Strategie A: die erste Bahn, bei der
`belegt[b] + bedarf(belegt[b], l) <= 2000` ist. Strategie B: `b = belegt.index(min(belegt))` und nur diese prüfen. Ist keine
Bahn gefunden, gib die bisherige Anzahl zurück. In den f-Strings stehen die Texte in doppelten Anführungszeichen, weil der

```python
def lies_laengen(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        teile = f.read().split()
    laengen = []
    for t in teile:
        laengen.append(round(float(t) * 100))
    return laengen


def bedarf(belegt_bahn, l):
    if belegt_bahn == 0:
        return l
    return l + 30


def lade(laengen, strategie):
    belegt = [0, 0, 0]
    anzahl = 0
    for l in laengen:
        bahn = -1
        if strategie == 'A':
            for b in range(3):
                if bahn == -1 and belegt[b] + bedarf(belegt[b], l) <= 2000:
                    bahn = b
        else:                                   # Strategie B
            b = belegt.index(min(belegt))
            if belegt[b] + bedarf(belegt[b], l) <= 2000:
                bahn = b
        if bahn == -1:
            return anzahl                       # passt nicht: Fähre legt ab
        belegt[bahn] = belegt[bahn] + bedarf(belegt[bahn], l)
        anzahl = anzahl + 1
    return anzahl


print(lies_laengen('daten/faehre2.txt'))
beispiele = ['daten/faehre1.txt', 'daten/faehre2.txt', 'daten/faehre3.txt']
print(f'{"Beispiel":20} {"A":>3} {"B":>3}')
for name in beispiele:
    laengen = lies_laengen(name)
    print(f'{name:20} {lade(laengen, "A"):3} {lade(laengen, "B"):3}')
eigenes = [300, 1000, 1100, 1100, 900]
print(f'{"eigenes Beispiel":20} {lade(eigenes, "A"):3} {lade(eigenes, "B"):3}')
```

In Beispiel 2 und 3 ist A besser: Ganz am Ende kommen lange Fahrzeuge (15,06 m in Beispiel 2), und A hat eine Bahn lange frei
gehalten, während B alle Bahnen gleichmäßig gefüllt hat. Im eigenen Beispiel ist es umgekehrt: A packt das 3-m-Auto und das
10-m-Fahrzeug zusammen in Bahn 1 und die beiden 11-m-Fahrzeuge in die Bahnen 2 und 3. Dann ist für das 9-m-Fahrzeug nirgends
mehr Platz. B verteilt die Fahrzeuge gleichmäßiger, und alle fünf passen. Genau so belegt man in der Dokumentation, dass
mal die eine, mal die andere Strategie gewinnt. Siehe `jwinf3/33_FaehreFuellen/faehrefuellen_loesung.ipynb`.

## Wettbewerbs-Ecke 2: Bücherregal

*Angelehnt an die Junioraufgabe „Bücherregal“ (Bundeswettbewerb Informatik 2017).*

Laura teilt ein Regalbrett mit Deko-Figuren in Abschnitte: Mit `f` Figuren entstehen `f + 1` Abschnitte. In jedem Abschnitt
dürfen sich die Bücher in der Höhe um **höchstens 30 mm** unterscheiden (die Dateien geben die Höhen in Millimetern an). Kein
Buch darf übrig bleiben. Ist das möglich?

Eine gierige Strategie: Sortiere die Bücher nach der Höhe. Das kleinste Buch eröffnet den ersten Abschnitt, und alle Bücher, die
höchstens 30 mm höher sind, kommen dazu. Das nächste Buch, das nicht mehr passt, eröffnet den nächsten Abschnitt, und so weiter.

Diesmal findet die gierige Strategie wirklich die **wenigsten** Abschnitte. Der Grund: Das kleinste Buch muss in irgendeinem
Abschnitt stehen. In seinem Abschnitt können höchstens Bücher bis 30 mm darüber stehen. Nimmt man **alle** diese Bücher dazu,
hat man für den Rest so wenige Bücher wie möglich übrig, und es kann nicht schaden.

`daten/buecherregal1.txt` (erste Zeile Figuren, zweite Zeile Bücher, dann die Höhen):

```python
def lies_regal(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        werte = []
        for t in f.read().split():
            werte.append(int(t))
    return werte[0], werte[2:2 + werte[1]]


def abschnitte(hoehen):
    ergebnis = []
    for h in sorted(hoehen):
        if len(ergebnis) > 0 and h - ergebnis[-1][0] <= 30:
            ergebnis[-1].append(h)          # passt zum kleinsten Buch des letzten Abschnitts
        else:
            ergebnis.append([h])            # neuer Abschnitt
    return ergebnis


figuren, hoehen = lies_regal('daten/buecherregal1.txt')
print(figuren, hoehen)
for name in ['daten/buecherregal1.txt', 'daten/buecherregal3.txt']:
    figuren, hoehen = lies_regal(name)
    a = abschnitte(hoehen)
    print(name + ':', figuren, 'Figuren,', len(a), 'Abschnitte nötig, möglich:', len(a) <= figuren + 1)
    for abschnitt in a:
        print('  ', abschnitt)
```

Für Datei 1 reichen 5 Abschnitte und damit genau die 4 Figuren. Für Datei 3 bräuchte Laura 4 Abschnitte, hat aber nur 2 Figuren,
also 3 Abschnitte. Weil die gierige Strategie die wenigsten Abschnitte findet, ist damit **bewiesen**, dass es nicht geht.
Das ist der Unterschied zur Fähre: Dort ist keine Strategie immer die beste, hier schon. Siehe
`jwinf3/36_Buecherregal/buecherregal_loesung.ipynb`.

## Das hast du gelernt

| Werkzeug | So geht es |
|---|---|
| sortieren | `sorted(liste)`, `sorted(liste, reverse=True)`, `liste.sort()` |
| nach einem Wert sortieren | Paare `[wert, ding]` bauen und die sortieren |
| gierige Strategie | immer die Wahl, die im Moment am besten aussieht |
| Strategie als Parameter | `lade(laengen, 'A')`, in der Funktion mit `if` unterscheiden |
| Vergleichstabelle | eine Zeile pro Beispiel, eine Spalte pro Strategie |

- Gierig ist **manchmal** optimal (Euro-Münzen, Bücherregal) und **manchmal nicht** (1-3-4-Münzen, Fähre).
- Optimal: begründen. Nicht optimal: ein Gegenbeispiel zeigen.
- Strategien an den Beispielen der Aufgabe **und** an eigenen Beispielen vergleichen.
