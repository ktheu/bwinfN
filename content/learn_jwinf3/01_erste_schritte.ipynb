## 2. Rechnen

Python rechnet wie ein Taschenrechner. Die Rechenzeichen sind:

| Blockly | Python | Bedeutung | Beispiel | Ergebnis |
|---|---|---|---|---|
| 1 **+** 1 | `+` | plus | `7 + 2` | `9` |
| 1 **-** 1 | `-` | minus | `7 - 2` | `5` |
| 1 **×** 1 | `*` | mal | `7 * 2` | `14` |
| 1 **÷** 1 | `/` | geteilt (mit Komma) | `7 / 2` | `3.5` |
| (kein Block) | `//` | geteilt **ohne Rest** (ganzzahlig) | `7 // 2` | `3` |
| **Rest von** 7 ÷ 2 | `%` | **Rest** beim Teilen | `7 % 2` | `1` |

In Blockly wählst du das Rechenzeichen im Block „1 + 1“ aus; in Python tippst du es einfach. Statt der Blöcke ineinander zu stecken, schreibst du **Klammern**.
Die funktionieren wie in Mathe: Punkt vor Strich, Klammern zuerst. Das Komma heißt in Python **Punkt**: `3.5` statt 3,5.

`//` und `%` sind besonders nützlich. Du kennst das aus der Grundschule: 17 : 5 = 3 Rest 2. In Python ist `17 // 5` gleich `3` und `17 % 5` gleich `2`.

````python
print(17 // 5, 17 % 5)
print(2 + 3 * 4, (2 + 3) * 4)
print(10 / 4)
````

## 3. Variablen

Eine **Variable** ist ein Name für einen Wert, wie in Blockly. In Python legst du sie einfach an, indem du ihr einen Wert gibst:

| Blockly | Python |
|---|---|
| setze `punkte` auf 10 | `punkte = 10` |
| erhöhe `punkte` um 5 | `punkte = punkte + 5` |
| `punkte` (Wert benutzen) | `punkte` |

Das `=` heißt in Python **„bekommt den Wert“**, nicht „ist gleich“. `punkte = punkte + 5` bedeutet also: Rechne `punkte + 5` aus und speichere das Ergebnis wieder in `punkte`.

Regeln für Namen: Buchstaben, Ziffern und `_`, nicht mit einer Ziffer beginnen, keine Leerzeichen. Groß- und Kleinschreibung ist wichtig: `Punkte` und `punkte` sind
**verschiedene** Variablen. Wir schreiben Namen klein und ohne Umlaute (`tueten` statt `tüten`).

````python
punkte = 10
print(punkte)
punkte = punkte + 5
print(punkte)
bonus = 3
gesamt = punkte + bonus
print('Gesamt:', gesamt)
````

## 4. Zahlen und Texte sind verschieden

Python unterscheidet ganze Zahlen (`7`), Kommazahlen (`3.5`) und Texte (`'7'`). Ein Text, der wie eine Zahl aussieht, ist trotzdem ein Text. Das ist später wichtig, denn alles, was du
aus einer Datei liest, ist zunächst **Text**.

- `int('7')` macht aus dem Text `'7'` die Zahl `7` (in Blockly: „Text zu Zahl“).
- `str(7)` macht aus der Zahl `7` den Text `'7'`.
- `+` rechnet bei Zahlen und **hängt** Texte aneinander.

#### Aufgabe 4.1: Was wird ausgegeben?

Überlege zuerst, was der Code ausgibt, und schreibe es auf. Führe die Zelle erst danach aus.

````python
print(5 + 3)
print('5' + '3')
print(int('5') + 3)
print('Ergebnis: ' + str(5 + 3))
````

</details>

## 6. Fehlermeldungen lesen

In Blockly passen falsche Blöcke gar nicht zusammen. In Python kannst du alles hinschreiben, und Python meldet einen **Fehler**, wenn etwas nicht stimmt. Das ist normal und passiert
allen, auch Profis. Wichtig ist, die Meldung zu **lesen**: Die **letzte Zeile** sagt, was los ist. Darüber steht, in welcher Zeile es passiert ist.

Die häufigsten Fehler am Anfang (die genaue Formulierung kann je nach Python-Version leicht anders sein):

**Code:**
```python
summe = 5
print(Summe)
```
**Meldung:** `NameError: name 'Summe' is not defined`  
Den Namen gibt es nicht. Hier: `Summe` mit großem S statt `summe`. Auch Tippfehler führen dazu.

**Code:**
```python
print('Alter: ' + 12)
```
**Meldung:** `TypeError: can only concatenate str (not "int") to str`  
Text und Zahl kann man nicht mit `+` verbinden. Lösung: `str(12)` oder ein Komma statt `+`.

**Code:**
```python
print('Hallo)
```
**Meldung:** `SyntaxError: unterminated string literal (detected at line 1)`  
Ein Anführungszeichen fehlt: Der Text hört nie auf.

**Code:**
```python
print(10 / 0)
```
**Meldung:** `ZeroDivisionError: division by zero`  
Durch 0 kann man nicht teilen.

**Code:**
```python
  print(1)
```
**Meldung:** `IndentationError: unexpected indent`  
Am Zeilenanfang stehen Leerzeichen, die dort nicht hingehören. In Python haben Einrückungen eine Bedeutung (das lernst du bei Schleifen).

✏️ *Deine Antwort (Doppelklick zum Bearbeiten):*

- mit `str`: `print('Jede Tüte bekommt ' + str(12 // tueten) + ' Spiele')`
