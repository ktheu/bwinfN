## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zufall:** Ein Programm kann mit dem Modul `random` zufällige Entscheidungen treffen. Hier wählen wir zufällig Silben aus.
- **Baukasten-Prinzip (Grammatik):** Ein Passwort besteht aus Silben, jede Silbe aus Konsonanten und Vokalen in einer festen Reihenfolge
  (zum Beispiel Konsonant + Vokal). Wir bauen das Passwort aus diesen Teilen zusammen. So entstehen Wörter, die wie Deutsch klingen.
- **Erzeugen und prüfen (Rejection Sampling):** Wir erzeugen ein Passwort und prüfen, ob es unsere Regeln erfüllt. Wenn nicht,
  erzeugen wir einfach ein neues. Das ist oft viel einfacher, als den Zufall so zu steuern, dass die Regeln nie verletzt werden.
- **Zeichen durchgehen und vergleichen:** Für die Regel „kein Zeichen doppelt“ vergleicht man jedes Zeichen mit seinem Vorgänger.

**Nützliche Funktionen und Bibliotheken**
- `import random` mit `random.choice(liste)` (zufälliges Element), `random.randint(a, b)` (zufällige ganze Zahl von a bis b)
  und `random.seed(zahl)` (damit bei jedem Lauf dasselbe herauskommt, praktisch zum Testen)
- `str(zahl)` wandelt eine Zahl in Text um, `zeichen.isdigit()` prüft, ob ein Zeichen eine Ziffer ist
- Für die Sicherheitsbetrachtung am Ende: `math.log2`

**Wichtig zur Sicherheit:** Für echte Passwörter sollte man statt `random` das Modul `secrets` benutzen, denn `random` ist nicht
für Sicherheitszwecke gedacht (seine Zahlen lassen sich vorhersagen). Für diese Aufgabe genügt `random`.

## Die Idee: Silben als Bausteine

Deutsche Wörter bestehen aus Silben, und die haben meist einen ähnlichen Aufbau, zum Beispiel in „Som-mer-fe-ri-en“:

| Aufbau | Bedeutung | Beispiele |
|:-:|---|---|
| `CV` | Konsonant + Vokal | fe, ri |
| `VC` | Vokal + Konsonant | en |
| `CVC` | Konsonant + Vokal + Konsonant | Som, mer |

Dazu kommt zur Abwechslung eine Zahl `N` mit einer oder zwei Ziffern. (`C` steht für *consonant*, `V` für *vowel*, `N` für *number*.)

Das Programm wählt zuerst zufällig, aus wie vielen Silben das Passwort bestehen soll, dann für jede Silbe einen Aufbau
und füllt zum Schluss die Aufbauten mit zufälligen Buchstaben. So kann man jedes Passwort immer in Silben sprechen:
„nus-siem-fus-di-mien“.

**Unsere Regeln** (jeweils mit Begründung):

1. **Silbenaufbau:** Es gibt nur die Aufbauten CV, VC und CVC, also immer abwechselnd Konsonanten und Vokale.
   *Warum?* Das entspricht dem Klang der deutschen Sprache. Wo Vokale und Konsonanten sich abwechseln, kann man
   die Buchstabenfolge fast immer gut aussprechen. Folgen wie „kpftz“ gibt es dagegen nicht.
2. **Lautgruppen als Einheit:** Als Vokale gibt es auch Doppellaute (`ie`, `ei`, `au`, `eu`, `ey`), als Konsonanten `sch`, `ch` und `qu`.
   *Warum?* Diese Buchstabengruppen werden im Deutschen als ein Laut gesprochen. Einzelne Konsonanten wie `c` lassen wir weg,
   weil sie im Deutschen fast nur in Gruppen (`ch`, `ck`) vorkommen.
3. **Zahlen:** Zahlen haben höchstens zwei Ziffern, und zwei Zahlen stehen nie nebeneinander.
   *Warum?* Lange Zahlenreihen kann man sich schlecht merken, kurze Zahlen zwischen Silben gehen gut.
4. **Keine doppelten Zeichen hintereinander:** Zwei gleiche Zeichen nebeneinander (auch über Silbengrenzen: „ma-an“, „fe-ell“)
   lässt das Programm nicht zu.
   *Warum?* Doppelte Buchstaben sind schwer auszusprechen und verleiten zu Tippfehlern.
5. **Länge und Zahl:** Das Passwort hat 5 bis 7 Silben, mindestens 12 Zeichen und mindestens eine Zahl. Es gibt nur
   Kleinbuchstaben und Ziffern.
   *Warum?* Kurze Passwörter sind leicht zu erraten (Länge ist wichtiger als Sonderzeichen). Bei mehr als 7 Silben wird das Passwort
   schwer zu merken. Eine Zahl ist bei vielen Webseiten gefordert. Auf Großbuchstaben und Sonderzeichen verzichten wir, weil man sie außerhalb
   bekannter Wörter schlecht im Kopf behält.

## Schritt 1: Die Bausteine

In `VOKALE` und `KONSONANTEN` stehen die Laute, aus denen die Silben gebaut werden. Am Anfang einer Silbe oder am Ende kann
ein beliebiger Laut aus der Liste stehen.

```python
import random

VOKALE = ['a', 'e', 'i', 'o', 'u', 'ie', 'ei', 'ey', 'au', 'eu']
KONSONANTEN = ['b', 'd', 'f', 'g', 'h', 'j', 'k', 'l', 'm', 'n', 'p', 'r', 's', 't', 'v', 'w', 'z', 'sch', 'ch', 'qu']

print(len(VOKALE), 'Vokale,', len(KONSONANTEN), 'Konsonanten')
random.seed(0)          # fester Startwert, damit bei jedem Lauf dasselbe herauskommt
print(random.choice(VOKALE), random.choice(KONSONANTEN))
```

## Schritt 2: Eine Silbe erzeugen

`erzeuge_silbe(aufbau)` baut je nach Aufbau (`'CV'`, `'VC'`, `'CVC'` oder `'N'`) eine Silbe aus zufälligen Bausteinen.
Der Text der Silbe wird mit `+` zusammengesetzt. Eine Zahl von 1 bis 99 wandeln wir mit `str(...)` in Text um.

```python
def erzeuge_silbe(aufbau):
    if aufbau == 'CV':
        return random.choice(KONSONANTEN) + random.choice(VOKALE)
    if aufbau == 'VC':
        return random.choice(VOKALE) + random.choice(KONSONANTEN)
    if aufbau == 'CVC':
        return random.choice(KONSONANTEN) + random.choice(VOKALE) + random.choice(KONSONANTEN)
    return str(random.randint(1, 99))        # Aufbau 'N': eine ein- oder zweistellige Zahl


random.seed(1)          # damit bei jedem Lauf dasselbe herauskommt (zum Ausprobieren)
for aufbau in ['CV', 'VC', 'CVC', 'N']:
    print(aufbau, erzeuge_silbe(aufbau), erzeuge_silbe(aufbau), erzeuge_silbe(aufbau))
```

## Schritt 3: Ein Passwort aus Silben bauen

Zuerst würfeln wir die Anzahl der Silben (5 bis 7) und für jede Silbe einen Aufbau. Zwei Zahlen hintereinander verhindern wir
sofort, indem wir statt der zweiten Zahl einen `'CV'`-Aufbau nehmen (Regel 3). Danach setzen wir die Silben zusammen.

Die Funktion gibt auch die Liste der Aufbauten zurück, so sehen wir, aus welchen Silben das Passwort entstanden ist.

```python
def erzeuge_rohpasswort():
    anzahl = random.randint(5, 7)
    aufbauten = []
    for i in range(anzahl):
        aufbau = random.choice(['CV', 'VC', 'CVC', 'N'])
        if aufbau == 'N' and i > 0 and aufbauten[i - 1] == 'N':
            aufbau = 'CV'                    # Regel 3: keine zwei Zahlen hintereinander
        aufbauten.append(aufbau)

    passwort = ''
    silben = []
    for aufbau in aufbauten:
        silbe = erzeuge_silbe(aufbau)
        silben.append(silbe)
        passwort = passwort + silbe
    return passwort, silben, aufbauten


random.seed(2)
for i in range(3):
    passwort, silben, aufbauten = erzeuge_rohpasswort()
    print(passwort)
    print('   Silben:  ', '-'.join(silben))
    print('   Aufbauten:', '-'.join(aufbauten))
```

## Schritt 4: Die Regeln prüfen

Nicht alle Wünsche lassen sich direkt beim Bauen erfüllen. Zum Beispiel können zwei Silben, die jede für sich in Ordnung sind,
zusammen einen doppelten Buchstaben ergeben („ma“ + „an“ = „maan“). Statt kompliziert vorzubeugen, **prüfen** wir das fertige Passwort:

- `hat_doppelte(passwort)` ist wahr, wenn irgendwo zwei gleiche Zeichen nebeneinander stehen (Regel 4).
  Wir vergleichen jedes Zeichen mit dem davor.
- `hat_zahl(passwort)` ist wahr, wenn mindestens eine Ziffer vorkommt (Regel 5).
- `ist_gut(passwort)` fasst die Regeln zusammen: mindestens 12 Zeichen, mindestens eine Zahl, keine doppelten Zeichen.

```python
def hat_doppelte(passwort):
    for i in range(1, len(passwort)):
        if passwort[i] == passwort[i - 1]:
            return True
    return False


def hat_zahl(passwort):
    for zeichen in passwort:
        if zeichen.isdigit():
            return True
    return False


def ist_gut(passwort):
    return len(passwort) >= 12 and hat_zahl(passwort) and not hat_doppelte(passwort)


for test in ['maan', 'schriopmeta9bunzquafke', 'kenschu28teyllalof', 'bofi', 'tefa7ruwo', 'dumeluzorabeko']:
    print(f'{test:24} doppelt: {hat_doppelte(test)!s:6} Zahl: {hat_zahl(test)!s:6} gut: {ist_gut(test)}')
```

Das Beispiel aus der Aufgabe (`schriopmeta9bunzquafke`) erfüllt alle Regeln. `kenschu28teyllalof` hat ein doppeltes `l`, `dumeluzorabeko` keine Zahl und `bofi`
ist viel zu kurz.

## Schritt 5: Der Generator

Jetzt kommt alles zusammen: Wir erzeugen so lange neue Passwörter, bis eines alle Regeln erfüllt. Die `while`-Schleife läuft,
solange das Passwort nicht gut ist. Etwa zwei von fünf zufällig erzeugten Passwörtern werden verworfen (siehe unten), im Mittel braucht man also weniger als zwei Versuche.

```python
def neues_passwort():
    passwort, silben, aufbauten = erzeuge_rohpasswort()
    while not ist_gut(passwort):
        passwort, silben, aufbauten = erzeuge_rohpasswort()
    return passwort, silben, aufbauten


random.seed(2020)
for i in range(8):
    passwort, silben, aufbauten = neues_passwort()
    print(f'{passwort:26} {"-".join(silben)}  ({"-".join(aufbauten)})')
```

Die Passwörter sind meist 12 bis 25 Zeichen lang, enthalten Zahlen und lassen sich gut in Silben aussprechen.
Mit `random.seed(2020)` kommt bei jedem Lauf dasselbe heraus. Für ein echtes Passwort lässt man die Zeile weg
(oder benutzt, wie oben erwähnt, das Modul `secrets`).

Wie oft wird ein Passwort verworfen? Wir zählen es für 10 000 Versuche:

```python
random.seed(7)
verworfen = 0
for i in range(10000):
    passwort, silben, aufbauten = erzeuge_rohpasswort()
    if not ist_gut(passwort):
        verworfen = verworfen + 1
print(verworfen, 'von 10000 Passwörtern verworfen')
```

## Für Neugierige: Wie sicher sind die Passwörter?

Ein Maß für die Sicherheit ist die Anzahl der Möglichkeiten (und in Bits `log2` davon). Ein Passwort aus 11 Zeichen, die zufällig aus 72
verschiedenen gewählt werden (wie `Dfq4T&o%%kM`), hat 72^11 ≈ 2,7 · 10^20 Möglichkeiten, das sind etwa **68 Bit**.

Bei unserem Generator wählt jede Silbe zuerst einen von vier Aufbauten (2 Bit) und dann Laute aus den Listen. Die Zahl der Möglichkeiten
je Aufbau ist `len(KONSONANTEN) * len(VOKALE)` für CV und VC, `len(KONSONANTEN) * len(VOKALE) * len(KONSONANTEN)` für CVC und 99 für N.
Im Mittel hat eine Silbe

`2 + (log2(M_CV) + log2(M_VC) + log2(M_CVC) + log2(M_N)) / 4`

Bit. (Die Regeln, die Passwörter verwerfen, kosten ein wenig Sicherheit, hier vernachlässigt.)

```python
import math

m_cv = len(KONSONANTEN) * len(VOKALE)
m_vc = len(VOKALE) * len(KONSONANTEN)
m_cvc = len(KONSONANTEN) * len(VOKALE) * len(KONSONANTEN)
m_n = 99

bit_je_silbe = 2 + (math.log2(m_cv) + math.log2(m_vc) + math.log2(m_cvc) + math.log2(m_n)) / 4
print(f'Bit je Silbe: {bit_je_silbe:.2f}')
for silben in [5, 6, 7]:
    print(f'{silben} Silben: etwa {silben * bit_je_silbe:.1f} Bit')
print(f'Passwort mit 11 Zeichen aus 72 Zeichen: {11 * math.log2(72):.1f} Bit')
```

Mit 6 bis 7 Silben sind unsere Passwörter etwa so sicher wie ein zufälliges Passwort aus 11 Zeichen mit Sonderzeichen,
aber viel leichter zu merken. Heute empfiehlt man oft noch einfachere Passwörter aus mehreren zufällig gewählten **ganzen Wörtern**
(„Passphrase“), weil man sich aus Wörtern gut eine Geschichte merken kann.
