Bela und Amira ist im Deutschunterricht oft langweilig. Daher haben sie sich ein neues Spiel ausgedacht: Texthopsen.
Sie suchen sich einen Text im Schulbuch aus und hopsen darin um die Wette. Das geht so: Jeder Buchstabe hat eine
Sprungweite zugeordnet (siehe Tabelle). Bela fängt beim ersten, Amira beim zweiten Buchstaben des Textes an.
Von dieser Startposition springen sie jeweils so viele Positionen weiter, wie die Sprungweite des Buchstabens vorgibt.
Alles was kein Buchstabe ist, überspringen sie dabei einfach. An der neuen Position lesen sie dann jeweils den nächsten
Buchstaben mit einer neuen Sprungweite. Das Hopsen wiederholen sie abwechselnd so lange, bis sie aus dem Text
herausspringen. Wer dies als Erstes schafft, hat gewonnen.

| Buchstabe   | a | b | c | ... | z  | ä  | ö  | ü  | ß  |
|-------------|---|---|---|-----|----|----|----|----|----|
| Sprungweite | 1 | 2 | 3 | ... | 26 | 27 | 28 | 29 | 30 |

#### Junioraufgabe 2

Schreibe ein Programm, das für einen gegebenen Text bestimmt, ob Bela oder Amira beim Texthopsen gewinnt.

Wende dein Programm mindestens auf alle Beispiele an, die du auf den BWINF-Webseiten findest,
und dokumentiere die Ergebnisse.

#### Das Beispiel auf dem Aufgabenblatt

Im Text **„Beispieltexte sind toll“** (das Leerzeichen wird übersprungen) springt

- Bela: `B` (Sprungweite 2) → `i` (9) → `t` (20) → aus dem Text heraus,
- Amira: `e` (5) → `e` (5) → `t` (20) → aus dem Text heraus.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `hopsen1.txt` bis `hopsen5.txt`, jeweils ein Text. Zeichen, die nicht in der Tabelle
stehen, werden übersprungen. Das gilt auch am Anfang des Textes. Dass Groß- und Kleinbuchstaben dieselbe Sprungweite
haben, steht nicht ausdrücklich da, folgt aber aus dem Beispiel (das große `B` hat die Sprungweite 2).

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Simulation:** Wir spielen das Spiel einfach nach, Sprung für Sprung. Eine `while`-Schleife wiederholt den Sprung,
  bis man aus dem Text heraus ist. Wie oft das passiert, weiß man vorher nicht.
- **Filtern:** Zuerst werfen wir alles weg, was kein Buchstabe der Tabelle ist. Danach sind „die nächsten Positionen“
  einfach die nächsten Zeichen im gefilterten Text. Das macht das Hopsen sehr einfach.
- **Unabhängige Abläufe:** Bela und Amira stören sich nicht gegenseitig. Man kann zuerst Belas ganzen Weg und dann Amiras
  ganzen Weg berechnen und nur die **Anzahl der Sprünge** vergleichen.
- **Lineare Laufzeit:** Jeder Sprung geht mindestens ein Zeichen weiter. Die Anzahl der Sprünge ist also höchstens so groß
  wie die Länge des Textes.

**Nützliche Funktionen** (Bibliotheken braucht man keine)
- `text.lower()` macht aus großen Buchstaben kleine, `zeichen in alphabet` prüft, ob ein Zeichen in einem Text vorkommt,
  `alphabet.index(zeichen)` liefert die Position eines Zeichens im Alphabet (`'a'` hat Position 0).
- Beim Öffnen der Datei `encoding='utf-8'` angeben, sonst werden `ä`, `ö`, `ü` und `ß` falsch gelesen.

## Die Idee

1. Aus dem Text nehmen wir nur die Buchstaben der Tabelle (in Kleinbuchstaben). Aus „Beispieltexte sind toll“ wird
   `beispieltextesindtoll`. Die Sprungweiten für diese Buchstaben berechnen wir aus ihrer Position im Alphabet.
2. Wer an Position `p` steht, springt zu Position `p + sprungweite`. Das zählen wir als einen Sprung.
   Ist die neue Position nicht mehr im Text, ist man draußen.
3. Bela startet bei Position 0 (erster Buchstabe), Amira bei Position 1 (zweiter Buchstabe).
4. Wir zählen die Sprünge von beiden und vergleichen.

**Wer gewinnt?** Die Aufgabe sagt nur „wer als Erstes herausspringt“. Es gibt zwei Lesarten:

- **Lesart 1:** Wer weniger Sprünge braucht, gewinnt. Bei gleich vielen Sprüngen ist es unentschieden.
- **Lesart 2:** Die beiden springen abwechselnd, und das Spiel endet sofort, wenn jemand draußen ist.
  Bela beginnt (er steht ja auch im Text zuerst). Bei gleich vielen Sprüngen ist Bela deshalb schon vor Amira draußen
  und gewinnt.

Beide Lesarten sind in Ordnung, man muss nur sagen, welche man benutzt. Wir berechnen beide.

## Schritt 1: Sprungweite eines Zeichens

Die Tabelle enthält die Buchstaben `a` bis `z`, danach `ä`, `ö`, `ü`, `ß`. Wir schreiben sie als Text in genau dieser
Reihenfolge auf. Die Sprungweite eines Buchstabens ist dann seine Position in diesem Text plus 1,
denn `index` zählt bei 0 an, `a` soll aber die Weite 1 haben.

```python
ALPHABET = 'abcdefghijklmnopqrstuvwxyzäöüß'


def sprungweite(buchstabe):
    # Aufruf nur mit einem Buchstaben der Tabelle, kleine Buchstaben
    return ALPHABET.index(buchstabe) + 1


for b in ['a', 'b', 'e', 'z', 'ä', 'ö', 'ü', 'ß']:
    print(b, sprungweite(b))
```

## Schritt 2: Nur die Buchstaben der Tabelle behalten

Wir gehen den Text Zeichen für Zeichen durch. Zuerst machen wir aus dem Zeichen mit `lower()` einen Kleinbuchstaben.
Steht dieser im `ALPHABET`, hängen wir ihn an das Ergebnis an. Alles andere (Leerzeichen, Satzzeichen, Ziffern, ...)
fällt weg. Das gilt auch am Textanfang, wie es die Aufgabe verlangt.

```python
def nur_buchstaben(text):
    ergebnis = ''
    for zeichen in text:
        klein = zeichen.lower()
        if klein in ALPHABET:
            ergebnis = ergebnis + klein
    return ergebnis


print(nur_buchstaben('Beispieltexte sind toll'))
print(nur_buchstaben('  "Ärger!", rief 1 Bär.'))
```

## Schritt 3: Einen Spieler hopsen lassen

Die Funktion bekommt den gefilterten Text und die Startposition und zählt die Sprünge. Die Schleife läuft,
solange die Position noch im Text liegt (`position < len(buchstaben)`). In jedem Durchgang:
Sprungweite des Buchstabens an der aktuellen Position nachschlagen, `position` um diese Weite erhöhen,
einen Sprung mehr zählen.

Liegt schon die **Startposition** nicht im Text (zum Beispiel Amira bei einem Text mit nur einem Buchstaben),
läuft die Schleife gar nicht, und der Spieler hat 0 Sprünge.

```python
def zaehle_spruenge(buchstaben, start):
    position = start
    spruenge = 0
    while position < len(buchstaben):
        position = position + sprungweite(buchstaben[position])
        spruenge = spruenge + 1
    return spruenge
```

Wir testen das mit dem Beispiel vom Aufgabenblatt. Zur Kontrolle geben wir bei jedem Sprung Position, Buchstabe und
Sprungweite aus. Das Ergebnis muss mit dem Bild übereinstimmen (Bela: 2, 9, 20, Amira: 5, 5, 20).

```python
buchstaben = nur_buchstaben('Beispieltexte sind toll')
print('gefilterter Text:', buchstaben, '(', len(buchstaben), 'Buchstaben )')

for name, start in [('Bela', 0), ('Amira', 1)]:
    position = start
    while position < len(buchstaben):
        weite = sprungweite(buchstaben[position])
        print(f'{name}: Position {position} ist {buchstaben[position]!r}, Sprungweite {weite}')
        position = position + weite
    print(f'{name} springt aus dem Text heraus nach Position {position}.')

print('Sprünge Bela:', zaehle_spruenge(buchstaben, 0), ' Sprünge Amira:', zaehle_spruenge(buchstaben, 1))
```

Beide brauchen drei Sprünge und landen auf demselben `t`. Der Ausgang hängt hier also von der Lesart ab,
und das passiert bei einigen Beispieldateien auch (siehe unten).

## Schritt 4: Wer gewinnt?

Zuerst werden die Sprünge gezählt (`hb` für Bela, `ha` für Amira), dann entscheiden die beiden Lesarten:

```python
def gewinner_lesart1(hb, ha):
    # Lesart 1: wer weniger Sprünge braucht, gewinnt; gleich viele = unentschieden
    if hb < ha:
        return 'Bela'
    if ha < hb:
        return 'Amira'
    return 'unentschieden'


def gewinner_lesart2(hb, ha):
    # Lesart 2: Bela hopst in jeder Runde zuerst, bei gleich vielen Sprüngen ist er also zuerst draußen
    if hb <= ha:
        return 'Bela'
    return 'Amira'


def texthopsen(text):
    buchstaben = nur_buchstaben(text)
    hb = zaehle_spruenge(buchstaben, 0)     # Bela beginnt beim ersten Buchstaben
    ha = zaehle_spruenge(buchstaben, 1)     # Amira beim zweiten
    return hb, ha


hb, ha = texthopsen('Beispieltexte sind toll')
print(hb, ha, gewinner_lesart1(hb, ha), gewinner_lesart2(hb, ha))
```

## Schritt 5: Alle Beispiele

Jetzt lesen wir die Textdateien ein und lassen das Programm laufen. Die Anzahl der Zeichen und der Buchstaben
geben wir mit aus, das hilft beim Nachvollziehen.

```python
def lies_text(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        return f.read()


print('Datei        Zeichen  Buchstaben  Bela  Amira  Lesart 1       Lesart 2')
ergebnisse = []
for nr in range(1, 6):
    text = lies_text(f'beispieldaten/hopsen{nr}.txt')
    hb, ha = texthopsen(text)
    ergebnisse.append((hb, ha))
    print(f'hopsen{nr}.txt {len(text):7} {len(nur_buchstaben(text)):11} {hb:5} {ha:6}  '
          f'{gewinner_lesart1(hb, ha):14} {gewinner_lesart2(hb, ha)}')
```

Bei `hopsen1`, `hopsen4` und `hopsen5` ist das Ergebnis eindeutig. Bei `hopsen2` und `hopsen3` brauchen beide gleich viele Sprünge.
Dann entscheidet die Lesart: unentschieden (Lesart 1) oder Bela (Lesart 2). In der Dokumentation gehört deshalb dazu,
welche Lesart man gewählt hat.

Die Anzahlen der Sprünge stimmen mit den Werten in den offiziellen Lösungshinweisen überein
(zum Beispiel `hopsen1`: Amira 70, Bela 68).

## Für Neugierige: Das Spiel wirklich abwechselnd spielen (Lesart 2)

Wenn man Lesart 2 wörtlich nimmt, kann man die beiden auch Zug um Zug abwechselnd hopsen lassen und aufhören,
sobald einer draußen ist. Das Ergebnis ist dasselbe wie oben.

```python
def spiele_abwechselnd(text):
    buchstaben = nur_buchstaben(text)
    positionen = [0, 1]                     # Bela und Amira
    namen = ['Bela', 'Amira']
    am_zug = 0
    while positionen[0] < len(buchstaben) and positionen[1] < len(buchstaben):
        p = positionen[am_zug]
        positionen[am_zug] = p + sprungweite(buchstaben[p])
        am_zug = 1 - am_zug                 # 0 wird zu 1, 1 wird zu 0
    # Wer zuerst draußen ist, gewinnt
    if positionen[0] >= len(buchstaben):
        return 'Bela'
    return 'Amira'


for nr in range(1, 6):
    text = lies_text(f'beispieldaten/hopsen{nr}.txt')
    hb, ha = ergebnisse[nr - 1]
    print(f'hopsen{nr}.txt:', spiele_abwechselnd(text), '(Lesart 2 oben:', gewinner_lesart2(hb, ha) + ')')
```
