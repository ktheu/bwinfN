### Zahlenspiel – Lösung Schritt für Schritt

*33. Bundeswettbewerb Informatik (2014), 1. Runde, Junioraufgabe 2*

Ulla gibt regelmäßig Nachhilfe. Immer wieder wollen die Schüler und Schülerinnen das Kürzen von Brüchen üben. Deshalb möchte
Ulla gute Übungsaufgaben zum Kürzen automatisch erzeugen lassen: Ein Programm soll Brüche a/b wie 144/56 erzeugen, die dann
von den Schülern zu p/q möglichst weit gekürzt werden, also 144/56 = 18/7.

Sie benötigt Aufgaben von drei Schwierigkeitsstufen: leicht, mittel und schwer. Die Schwierigkeit der Aufgabe „Kürze a/b“
(mit Ergebnis p/q) hängt von der Größe der Zahlen a, b, p und q ab.

Ulla nennt die Gesamtanzahl der Ziffern in a und b die Länge des Bruchs a/b; z. B. hat 144/56 die Länge 5. Sie legt die
Schwierigkeitsstufen wie folgt fest:

| Stufe | Länge von a/b | Bedingung für p/q |
|---|---|---|
| leicht | 4 | p + q ≤ 10 |
| mittel | 5 | 10 < p + q ≤ 20 |
| schwer | 5 | 20 < p + q ≤ 30 |

Die für eine Stufe angegebenen Bedingungen müssen beide erfüllt sein. Außerdem müssen die Aufgaben weitere Bedingungen erfüllen:

- a/b soll tatsächlich gekürzt werden können, d. h. es soll q < b sein.
- Es soll immer a ≠ b sein.

Wie man sieht, ist die Aufgabe „Kürze 144/56“ schwer, und „Kürze 22/77“ ist leicht.

#### Junioraufgabe 2

Hilf Ulla und schreibe ein Programm, das für eine gegebene Schwierigkeitsstufe eine gewünschte Anzahl von zufällig erzeugten
Aufgaben der angegebenen Stufe mit ihren Lösungen ausgibt.

Beispieldaten gibt es zu dieser Aufgabe nicht.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Größter gemeinsamer Teiler (ggT):** Ein Bruch `a/b` wird vollständig gekürzt, indem man Zähler und Nenner durch `ggT(a, b)` teilt. Das Ergebnis `p/q` ist dann **teilerfremd**. In Python gibt es dafür `math.gcd`.
- **Ganzzahldivision `//`:** `a // g` teilt ohne Nachkommastellen.
- **Ziffern zählen:** Die Länge eines Bruchs ist die Anzahl der Ziffern von `a` plus die von `b`, zum Beispiel `len(str(a)) + len(str(b))`.
- **Alle Möglichkeiten aufzählen und filtern (Brute Force):** Es gibt nur wenige hunderttausend Brüche der Länge 4 oder 5. Wir prüfen alle und behalten die, die die Bedingungen erfüllen.
- **Zufällige Auswahl ohne Wiederholung:** `random.sample(liste, n)` zieht `n` verschiedene Elemente zufällig aus einer Liste. So kommt keine Aufgabe doppelt vor.
- **Erzeugen oder aufzählen?** Man kann auch zuerst einen Lösungsbruch `p/q` zufällig wählen und ihn mit einer kleinen Zahl erweitern (das ist der Weg der offiziellen Lösungshinweise). Wir zählen stattdessen alle gültigen Aufgaben auf. Das ist einfacher
  zu verstehen und erlaubt es, die Bedingungen direkt zu prüfen.

**Nützliche Funktionen und Bibliotheken**
- `import math` mit `math.gcd(a, b)`, `import random` mit `random.sample` und `random.seed`, `str`, `len`, `range`, f-Strings

## Die Idee

Eine Aufgabe ist ein Paar `(a, b)` aus positiven ganzen Zahlen. Es ist eine Aufgabe der Stufe `s`, wenn gilt:

1. Die Länge stimmt: `len(str(a)) + len(str(b))` ist 4 (leicht) oder 5 (mittel, schwer).
2. `a ≠ b`.
3. Wir kürzen mit dem ggT: `g = ggT(a, b)`, `p = a / g`, `q = b / g`. Die Aufgabe ist kürzbar, wenn `q < b`, das heißt `g > 1`.
4. Die Summe `p + q` liegt im Bereich der Stufe: leicht höchstens 10, mittel größer als 10 und höchstens 20, schwer größer als 20 und höchstens 30.

Wir gehen **alle** Paare `(a, b)` der richtigen Länge durch, behalten die gültigen in einer Liste (zusammen mit der Lösung `p/q`) und ziehen daraus zufällig die gewünschte Anzahl.

## Schritt 1: Die Stufen

Wir legen die Regeln der drei Stufen in einer Liste fest: Name, Länge, kleinste und größte erlaubte Summe `p + q` (die Summe muss größer als der erste und höchstens so groß wie der zweite Wert sein).
Für „leicht“ steht als untere Grenze 0 (jede Summe ist größer als 0).

```python
STUFEN = [['leicht', 4, 0, 10],
          ['mittel', 5, 10, 20],
          ['schwer', 5, 20, 30]]

for name, laenge, untere, obere in STUFEN:
    print(f'{name}: Länge {laenge}, p + q größer als {untere} und höchstens {obere}')
```

## Schritt 2: Ist (a, b) eine gültige Aufgabe?

`pruefe_aufgabe(a, b, stufe)` wendet die Regeln der Idee an und gibt die Lösung `(p, q)` zurück, wenn alles stimmt, sonst `None`. `math.gcd(a, b)` ist der größte gemeinsame Teiler.

```python
import math


def pruefe_aufgabe(a, b, stufe):
    name, laenge, untere, obere = STUFEN[stufe]
    if len(str(a)) + len(str(b)) != laenge:
        return None
    if a == b:
        return None
    g = math.gcd(a, b)
    p = a // g
    q = b // g
    if q >= b:                              # nicht kürzbar (g = 1)
        return None
    if not (untere < p + q <= obere):
        return None
    return (p, q)


print(pruefe_aufgabe(144, 56, 2))           # schwer: 18/7
print(pruefe_aufgabe(22, 77, 0))            # leicht: 2/7
print(pruefe_aufgabe(22, 77, 1))            # falsche Stufe: Länge 4 passt nicht zu 'mittel'
print(pruefe_aufgabe(13, 17, 0))            # nicht kürzbar
print(pruefe_aufgabe(15, 15, 0))            # a = b
```

## Schritt 3: Alle gültigen Aufgaben einer Stufe

Wir probieren alle Paare durch. Für die Länge `L` hat `a` zwischen 1 und `L − 1` Ziffern, `b` den Rest. Wir laufen über alle `a` mit höchstens `L − 1` Ziffern (`range(1, 10 ** (L - 1))`)
und für jedes `a` über alle `b` mit der passenden Ziffernzahl `k = L − len(str(a))`, also `range(10 ** (k - 1), 10 ** k)`. Die gültigen Aufgaben sammeln wir als `(a, b, p, q)` in einer Liste.

Das sind für Länge 5 etwa 324 000 Paare, für Python kein Problem (es dauert nur Bruchteile einer Sekunde).

```python
def alle_aufgaben(stufe):
    laenge = STUFEN[stufe][1]
    liste = []
    for a in range(1, 10 ** (laenge - 1)):
        k = laenge - len(str(a))                    # so viele Ziffern hat b
        for b in range(10 ** (k - 1), 10 ** k):
            loesung = pruefe_aufgabe(a, b, stufe)
            if loesung is not None:
                liste.append((a, b, loesung[0], loesung[1]))
    return liste


mengen = []
for stufe in range(3):
    mengen.append(alle_aufgaben(stufe))
    print(f'{STUFEN[stufe][0]}: {len(mengen[stufe])} mögliche Aufgaben')
```

## Schritt 4: Aufgaben zufällig auswählen und ausgeben

`erzeuge_aufgaben(anzahl, stufe)` zieht mit `random.sample` die gewünschte Anzahl **verschiedener** Aufgaben zufällig aus der Liste der gültigen Aufgaben. Wir berechnen die Liste einmal pro Stufe und merken sie uns (`mengen`),
damit das Ziehen schnell geht.

Die Ausgabe zeigt die Aufgabe und in eckigen Klammern die Lösung, so wie in der Aufgabe. Mit `random.seed(1)` kommt bei jedem Lauf dasselbe heraus (für die Dokumentation praktisch).
Möchte man jedes Mal andere Aufgaben, lässt man die Zeile weg.

```python
import random


def erzeuge_aufgaben(anzahl, stufe):
    verfuegbar = mengen[stufe]
    if anzahl > len(verfuegbar):
        print(f'Es gibt nur {len(verfuegbar)} verschiedene Aufgaben dieser Stufe.')
        anzahl = len(verfuegbar)
    auswahl = random.sample(verfuegbar, anzahl)
    print(f'{anzahl} Aufgaben der Schwierigkeit {STUFEN[stufe][0]}')
    for a, b, p, q in auswahl:
        print(f'Aufgabe: Kürze {a} / {b} maximal.  [Lösung: {p} / {q}]')
    return auswahl


random.seed(1)
for stufe in range(3):
    erzeuge_aufgaben(6, stufe)
    print()
```

Jede Aufgabe lässt sich nachrechnen: Bei „Kürze 144 / 56“ wäre der ggT 8, und 144 : 8 = 18 und 56 : 8 = 7 ergibt die Lösung 18 / 7 (schwer, denn 18 + 7 = 25).
Wenn man die Aufgaben dem Programm vertraut, sollte man es trotzdem kontrollieren. Das tun wir im nächsten Schritt.

## Schritt 5: Die Ergebnisse überprüfen

Wir prüfen **alle** Aufgaben, die das Programm erzeugen könnte, noch einmal unabhängig. Dazu schreiben wir die Bedingungen ein zweites Mal ausführlich auf (ohne `math.gcd`: wir suchen den größten Teiler mit einer Schleife) und kontrollieren
für bis zu 1000 zufällig gezogene Aufgaben jeder Stufe:

- Der Bruch hat die richtige Länge.
- `a ≠ b` und `q < b`.
- Die Lösung `p / q` ist **wirklich vollständig gekürzt** (kein Teiler größer als 1 teilt beide) und gleich `a / b`.
- Die Summe `p + q` liegt im Bereich der Stufe.
- Es gibt keine Doppelten in der gezogenen Liste.

```python
def groesster_teiler(a, b):
    teiler = 1
    for t in range(2, min(a, b) + 1):
        if a % t == 0 and b % t == 0:
            teiler = t
    return teiler


random.seed(7)
alles_gut = True
for stufe in range(3):
    name, laenge, untere, obere = STUFEN[stufe]
    auswahl = random.sample(mengen[stufe], min(1000, len(mengen[stufe])))
    for a, b, p, q in auswahl:
        g = groesster_teiler(a, b)
        if len(str(a)) + len(str(b)) != laenge or a == b or q >= b:
            alles_gut = False
        if p * g != a or q * g != b or groesster_teiler(p, q) != 1:
            alles_gut = False
        if not (untere < p + q <= obere):
            alles_gut = False
    if len(set(auswahl)) != len(auswahl):
        alles_gut = False
print('Alle gezogenen Aufgaben sind in Ordnung:', alles_gut)
```

## Für Neugierige: Schönere Aufgaben

Eine Aufgabe, bei der die Lösung eine ganze Zahl ist (`q = 1`, zum Beispiel „Kürze 84 / 42“, Lösung 2 / 1), ist nach den Regeln der Aufgabe erlaubt, wirkt aber nicht wie eine typische Kürzaufgabe.
Man kann solche Aufgaben aussortieren. Wir sehen nach, wie viele es sind, und welche Verteilung die Lösungen haben.

```python
for stufe in range(3):
    mit_ganzer_zahl = 0
    for a, b, p, q in mengen[stufe]:
        if q == 1:
            mit_ganzer_zahl = mit_ganzer_zahl + 1
    print(f'{STUFEN[stufe][0]}: {mit_ganzer_zahl} von {len(mengen[stufe])} Aufgaben haben q = 1 '
          f'({100 * mit_ganzer_zahl / len(mengen[stufe]):.1f} %)')


def erzeuge_aufgaben_ohne_ganze(anzahl, stufe):
    liste = []
    for aufgabe in mengen[stufe]:
        if aufgabe[3] != 1:
            liste.append(aufgabe)
    return random.sample(liste, anzahl)


random.seed(3)
for a, b, p, q in erzeuge_aufgaben_ohne_ganze(5, 0):
    print(f'Kürze {a} / {b}  [Lösung: {p} / {q}]')
```

**Ein anderes Verfahren** (aus den offiziellen Lösungshinweisen): Man wählt zuerst einen teilerfremden Lösungsbruch `p / q` zufällig und **erweitert** ihn dann mit kleinen Zahlen (zum Beispiel mit 2, 3, 5 oder 7), bis die Länge stimmt.
Dabei muss man gut aufpassen, dass `p / q` wirklich teilerfremd ist, sonst ergäbe maximales Kürzen einen anderen Bruch als den erwarteten. Dieses Verfahren erzeugt viel schneller eine einzelne Aufgabe, aber bestimmte Aufgaben
(zum Beispiel solche mit großen Primfaktoren im ggT) kommen nie vor. Unser Aufzählen findet dagegen **alle** gültigen Aufgaben.
