#### Beispieldaten

Im Ordner `beispieldaten` stehen `garten0.txt` bis `garten5.txt`. Jede Datei hat drei Zeilen:

- 1. Zeile: Anzahl der Interessenten
- 2. Zeile: Höhe des Grundstücks in Metern
- 3. Zeile: Breite des Grundstücks in Metern

Beispiel: Bei den Zahlen 35, 10 und 20 muss das 10 m × 20 m große Grundstück in mindestens 35, aber höchstens 10 % mehr
(also höchstens 38) rechteckige Kleingärten aufgeteilt werden. Eine Aufteilung in deckungsgleiche Rechtecke reicht aus.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Brute Force (Ausprobieren aller Möglichkeiten):** Es gibt nur wenige Aufteilungen, nämlich Gitter aus `z` Reihen und
  `s` Spalten. Wir probieren alle durch und merken uns die beste. Dafür braucht man zwei Schleifen ineinander.
- **Das Beste merken (Minimum suchen):** Wir behalten die bisher beste Aufteilung und ihren Wert und ersetzen sie,
  wenn wir eine bessere finden.
- **Ein Maß festlegen:** Damit der Computer „möglichst quadratisch“ vergleichen kann, braucht es eine Zahl (eine
  „Kostenfunktion“), die angibt, wie weit ein Rechteck von einem Quadrat entfernt ist.
- **Schleifengrenzen klug wählen:** Die Schleifen müssen nicht bis zu jeder beliebigen Zahl laufen, sonst wird das Programm
  bei großen Beispielen zu langsam.

**Nützliche Funktionen** (Bibliotheken braucht man keine)
- `//` ist die Division ohne Nachkommastellen, `/` die Division mit Nachkommastellen.
- `max(a, b)` und `min(a, b)` liefern die größere und die kleinere von zwei Zahlen.
- In einem f-String formatiert `{zahl:.4f}` eine Kommazahl mit vier Nachkommastellen.

## Die Idee

Ein Gitter aus `z` Reihen und `s` Spalten ergibt `z * s` Gärten. Das Grundstück ist `H` Meter hoch und `B` Meter breit,
also ist jeder Garten `H / z` Meter hoch und `B / s` Meter breit.

**Erlaubt** ist eine Aufteilung, wenn es mindestens `k` Gärten gibt (so viele wie Interessenten), aber höchstens 10 % mehr:

`k ≤ z * s ≤ 1,1 * k`

**Wie quadratisch ist ein Garten?** Wir brauchen eine einzige Zahl, die man vergleichen kann. Wir nehmen das Verhältnis der
längeren zur kürzeren Seite:

`unquadratisch(a, b) = längere Seite / kürzere Seite`

- Ein Quadrat hat den Wert genau **1**. Je größer der Wert, desto weniger quadratisch ist der Garten.
- Es ist egal, ob der Garten hoch und schmal oder niedrig und breit ist (2 m × 3 m und 3 m × 2 m sind gleich gut).
- Es ist egal, wie groß der Garten ist (2 m × 3 m und 4 m × 6 m sind gleich gut).

Das Programm probiert alle erlaubten Gitter aus und nimmt das mit dem **kleinsten** Wert.

## Schritt 1: Die Datei einlesen

`read()` liest die ganze Datei, `split()` zerlegt sie an den Zeilenumbrüchen in einzelne Wörter, und `int(...)` macht
daraus Zahlen.

```python
def lies_garten(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zahlen = f.read().split()
    interessenten = int(zahlen[0])
    hoehe = int(zahlen[1])
    breite = int(zahlen[2])
    return interessenten, hoehe, breite


k, hoehe, breite = lies_garten('beispieldaten/garten0.txt')
print('Interessenten:', k, ' Grundstück:', hoehe, 'm hoch,', breite, 'm breit')
```

## Schritt 2: Wie quadratisch ist ein Garten?

Die Funktion `unquadratisch(a, b)` teilt die längere Seite durch die kürzere. Wir probieren sie mit Beispielen aus der
Aufgabe aus: 4 Reihen und 6 Spalten ergeben bei 42 m × 66 m Gärten von 10,5 m × 11 m, bei 5 × 5 Gärten sind es 8,4 m × 13,2 m.

```python
def unquadratisch(a, b):
    if a > b:
        return a / b
    return b / a


print(unquadratisch(3, 3))              # ein Quadrat
print(unquadratisch(2, 3), unquadratisch(3, 2), unquadratisch(4, 6))     # drei gleich gute Rechtecke
print(unquadratisch(42 / 4, 66 / 6))    # 4 Reihen, 6 Spalten
print(unquadratisch(42 / 5, 66 / 5))    # 5 Reihen, 5 Spalten
print(unquadratisch(42 / 1, 66 / 23))   # 23 schmale Streifen
```

Die Aufteilung in 4 × 6 Gärten ist also deutlich quadratischer (1,048) als 5 × 5 (1,571) oder 23 Streifen (14,6).

## Schritt 3: Alle erlaubten Aufteilungen ausprobieren

Zuerst die größte erlaubte Gartenzahl: `maximal = k * 11 // 10`. Das ist 1,1 · `k`, abgerundet. Wir rechnen mit
ganzen Zahlen (`k * 11 // 10` statt `1.1 * k`), dann gibt es keine Rundungsfehler.

Wegen `z * s ≥ k` kann es höchstens `maximal` Reihen und höchstens `maximal` Spalten geben (die jeweils andere Zahl
ist mindestens 1). Wir probieren deshalb alle `z` von 1 bis `maximal` und alle `s` von 1 bis `maximal` aus.
Nur wenn `k ≤ z * s ≤ maximal` gilt, ist die Aufteilung erlaubt, und dann berechnen wir ihren Wert.

Die Variable `beste_kosten` beginnt mit `None` („noch nichts gefunden“). Die erste erlaubte Aufteilung wird damit
automatisch die beste, danach ersetzt nur eine **kleinere** Zahl die bisher beste. Bei Gleichstand gewinnt deshalb die
zuerst gefundene Aufteilung.

```python
def beste_aufteilung_einfach(k, hoehe, breite):
    maximal = k * 11 // 10
    beste_z = 0
    beste_s = 0
    beste_kosten = None
    for z in range(1, maximal + 1):             # Anzahl der Reihen
        for s in range(1, maximal + 1):         # Anzahl der Spalten
            gaerten = z * s
            if gaerten >= k and gaerten <= maximal:
                kosten = unquadratisch(hoehe / z, breite / s)
                if beste_kosten is None or kosten < beste_kosten:
                    beste_z = z
                    beste_s = s
                    beste_kosten = kosten
    return beste_z, beste_s


print(beste_aufteilung_einfach(23, 42, 66))
print(beste_aufteilung_einfach(35, 10, 20))
```

Für 23 Interessenten und 42 m × 66 m ergibt sich `(4, 6)`, genau wie in der Aufgabenstellung beschrieben.
Beim zweiten Beispiel (35 Interessenten, 10 m × 20 m, erlaubt sind 35 bis 38 Gärten) rechnen wir nach, was das
Ergebnis bedeutet:

```python
z, s = beste_aufteilung_einfach(35, 10, 20)
print('Reihen:', z, 'Spalten:', s, 'Gärten:', z * s)
print('Garten:', 10 / z, 'm hoch,', 20 / s, 'm breit, Wert:', unquadratisch(10 / z, 20 / s))
```

Es sind 4 × 9 = 36 Gärten, also im erlaubten Bereich von 35 bis 38. Die Gärten sind 2,5 m × 2,22 m groß.

## Schritt 4: Ein Problem mit großen Beispielen

Bei `garten5.txt` sind es 35 000 Interessenten, also `maximal = 38500`. Die beiden Schleifen müssten
38 500 · 38 500 = 1 482 250 000 Kombinationen durchprobieren, etwa anderthalb Milliarden. Das dauert in Python viele Minuten.

Fast alle diese Kombinationen sind sinnlos: Ist `z` zum Beispiel 1000, dann darf `s` höchstens `38500 // 1000 = 38`
sein, sonst wären es mehr als 38 500 Gärten. Allgemein gilt `z * s ≤ maximal`, also `s ≤ maximal // z`.
Wir lassen die innere Schleife deshalb nur bis `maximal // z` laufen. Der Rest der Funktion bleibt gleich.

So sind es nur noch etwa 410 000 Kombinationen, und `garten5.txt` ist in einem Wimpernschlag fertig.

```python
def beste_aufteilung(k, hoehe, breite):
    maximal = k * 11 // 10
    beste_z = 0
    beste_s = 0
    beste_kosten = None
    for z in range(1, maximal + 1):
        for s in range(1, maximal // z + 1):    # hier: nur so viele Spalten, dass z * s <= maximal bleibt
            gaerten = z * s
            if gaerten >= k:
                kosten = unquadratisch(hoehe / z, breite / s)
                if beste_kosten is None or kosten < beste_kosten:
                    beste_z = z
                    beste_s = s
                    beste_kosten = kosten
    return beste_z, beste_s


# Kontrolle: beide Versionen finden dieselbe Aufteilung
print(beste_aufteilung(23, 42, 66), beste_aufteilung_einfach(23, 42, 66))
print(beste_aufteilung(35, 10, 20), beste_aufteilung_einfach(35, 10, 20))

# Wie viele Kombinationen sind es bei garten5.txt (maximal = 38500)?
durchlaeufe = 0
for z in range(1, 38500 + 1):
    durchlaeufe = durchlaeufe + 38500 // z
print('Kombinationen bei garten5.txt:', durchlaeufe)
```

## Schritt 5: Übersichtliche Ausgabe

Für jedes Beispiel geben wir Reihen, Spalten, Anzahl der Gärten, Größe eines Gartens und das Verhältnis aus. Die Größe
runden wir auf vier Nachkommastellen (0,1 Millimeter). Das reicht für ein Grundstück völlig aus. Mit `.replace('.', ',')`
schreiben wir deutsche Kommazahlen.

```python
def komma(zahl):
    return f'{zahl:.4f}'.replace('.', ',')


def zeige_aufteilung(dateiname):
    k, hoehe, breite = lies_garten(dateiname)
    z, s = beste_aufteilung(k, hoehe, breite)
    garten_hoehe = hoehe / z
    garten_breite = breite / s
    print(f'{dateiname}: {k} Interessenten, Grundstück {hoehe} m x {breite} m')
    print(f'  {z} Reihen x {s} Spalten = {z * s} Gärten (erlaubt: {k} bis {k * 11 // 10})')
    print(f'  ein Garten ist {komma(garten_hoehe)} m hoch und {komma(garten_breite)} m breit')
    print(f'  Verhältnis längere zu kürzere Seite: {komma(unquadratisch(garten_hoehe, garten_breite))}')


for nr in range(6):
    zeige_aufteilung(f'beispieldaten/garten{nr}.txt')
```

Die Ergebnisse stimmen mit denen der offiziellen Lösungshinweise überein, zum Beispiel für `garten0.txt` 4 × 6 Gärten
(10,5 m × 11 m, Verhältnis 1,048) und für `garten5.txt` 120 × 308 Gärten mit dem Verhältnis 1,0002.
Bei `garten2.txt` bleibt es bei 1,4: Es ist das beste Verhältnis, das bei 36 bis 39 Gärten auf 55 m × 77 m möglich ist,
denn das Programm hat alle erlaubten Aufteilungen ausprobiert.

**Hinweis zu Kommazahlen:** Der Computer rechnet mit Kommazahlen auf etwa 16 Stellen genau. Für diese Aufgabe spielt das
keine Rolle, weil wir nur vergleichen, welches Verhältnis kleiner ist.

## Für Neugierige: Andere Maße für „quadratisch“

Statt „längere durch kürzere Seite“ kann man auch die Summe `a / b + b / a` nehmen (so machen es die offiziellen
Lösungshinweise), oder den Unterschied der Seitenlängen `abs(a - b)`. Der Unterschied hat allerdings einen Nachteil:
Er hängt von der Gartengröße ab. Ein 2 m × 3 m großer Garten ist genauso „schief“ wie einer mit 20 m × 30 m, aber
der Unterschied ist 1 bzw. 10.

Das Verhältnis und die Summe wählen fast immer dieselbe Aufteilung. Probier es aus:

```python
def unquadratisch2(a, b):
    return a / b + b / a


def beste_aufteilung2(k, hoehe, breite):
    maximal = k * 11 // 10
    beste_z = 0
    beste_s = 0
    beste_kosten = None
    for z in range(1, maximal + 1):
        for s in range(1, maximal // z + 1):
            if z * s >= k:
                kosten = unquadratisch2(hoehe / z, breite / s)
                if beste_kosten is None or kosten < beste_kosten:
                    beste_z = z
                    beste_s = s
                    beste_kosten = kosten
    return beste_z, beste_s


for nr in range(6):
    k, hoehe, breite = lies_garten(f'beispieldaten/garten{nr}.txt')
    print(f'garten{nr}.txt:', beste_aufteilung(k, hoehe, breite), beste_aufteilung2(k, hoehe, breite))
```
