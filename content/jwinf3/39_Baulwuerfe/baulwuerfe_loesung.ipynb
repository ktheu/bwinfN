Ein Forschungsteam will das Verhalten der Maulwürfe untersuchen, die offenkundig, aber unterirdisch ein großes Feld
bevölkern und dort ihre Hügel aufwerfen. Das Team hat das Feld gesichtet und dabei eine sogenannte Hügelkarte erstellt.
Eine Hügelkarte ist in kleine Planquadrate eingeteilt; jedes Planquadrat, auf dem sich ein Hügel befindet, ist mit X
markiert.

Beim Betrachten der Karte macht das Forschungsteam eine Entdeckung: Es kann nicht sein, dass dort nur gewöhnliche Maulwürfe
leben. Ein Planquadrat mit einem Maulwurfshügel ist nämlich immer rundum von Quadraten ohne Hügel umgeben. Die Markierungen
bilden aber auch genau dieses Muster (3 Spalten, 4 Zeilen):

```
XXX
X X
X X
XXX
```

Diese Rechtecke grenzen gelegentlich aneinander, überlappen sich aber nicht.

Im Forschungsteam herrscht große Aufregung: Es muss sich um eine neue Maulwurfsart handeln, die ihre Hügel zu solch
rechteckigen Bauen formiert. Die Forscher taufen sie „Baulwürfe“. Nun wollen sie anhand der Hügelkarte die auf dem Feld
befindlichen Baulwurfsbaue zählen.

#### Junioraufgabe 2

Hilf dem Forschungsteam und schreibe ein Programm. Es soll die Hügelkarte einlesen und die Anzahl der Baulwurfsbaue ausgeben.

Wende dein Programm auf die Beispiele an, die du auf den BWINF-Webseiten findest, und dokumentiere die Ergebnisse.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `karte0.txt` bis `karte6.txt`. Jede Datei ist so aufgebaut:

- 1. Zeile: Breite `b`, 2. Zeile: Höhe `h`
- folgende `h` Zeilen: je `b` Zeichen, ein Leerzeichen (kein Hügel) oder ein `X` (Hügel)

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Ein Raster als Liste von Zeilen:** Die Karte ist eine Liste von Texten. `karte[z][s]` ist das Zeichen in Zeile `z` und Spalte `s`.
- **Schablone über ein Raster schieben (Mustersuche):** Für jede Position prüfen wir, ob dort das Muster beginnt. Dafür braucht
  man zwei Schleifen ineinander und einen Vergleich Feld für Feld.
- **Rand behandeln:** Greift man über den Rand der Karte hinaus, darf das Programm nicht abstürzen. Eine kleine Hilfsfunktion,
  die außerhalb der Karte „kein Hügel“ meldet, erspart viele Fehler.
- **Doppelzählung vermeiden:** Zwei Baue dürfen nicht dieselben Hügel benutzen. Wir markieren Hügel, die schon zu einem Bau gehören.
- **Zählen statt suchen:** Eine zweite, ganz einfache Lösung nutzt, dass jeder Bau genau 10 Hügel hat und alle Hügel eines Baus
  aneinandergrenzen. Dann kann man Hügel zählen und durch 10 teilen.

**Nützliche Funktionen** (Bibliotheken braucht man keine)
- `str.ljust(breite)` füllt einen Text rechts mit Leerzeichen auf, `rstrip('\r\n')` entfernt den Zeilenumbruch am Ende
- `range`, geschachtelte `for`-Schleifen, `and`/`or`, `in`

## Die Idee

**Lösung 1: Das Muster suchen.** Das Muster ist 3 Spalten breit und 4 Zeilen hoch. Wir schieben es über die Karte und prüfen an jeder Stelle,
ob alle 12 Felder passen: Die zehn Randfelder müssen Hügel sein, die zwei Felder in der Mitte müssen leer sein.
Die Umgebung des Musters ist egal. Die Stelle ist die linke obere Ecke des Baus.

Das genügt noch nicht: Stoßen mehrere Baue aneinander, kann dort ein Muster erkannt werden, das gar kein echter Bau ist, weil es Hügel
von mehreren Bauen benutzt (die Aufgabenstellung sagt: Baue überlappen sich nicht). Deshalb merken wir uns, welche Hügel schon
zu einem Bau gehören, und zählen einen Bau nur, wenn kein Hügel davon schon benutzt ist. Wir suchen von oben nach unten und von links nach rechts.

**Lösung 2: Hügel zählen.** Ein einzelner Maulwurf baut seinen Hügel ganz allein, er hat keine Nachbarhügel (oben, unten, links, rechts).
Jeder Hügel eines Baus hat dagegen mindestens einen Nachbarhügel. Wir zählen alle Hügel, die mindestens einen Nachbarhügel haben, und
teilen die Anzahl durch 10, denn ein Bau besteht aus genau 10 Hügeln.

Wir werden sehen: Die einfache Mustersuche zählt bei zwei Karten **zu viele** Baue. Mit dem Schutz vor Überlappung und mit der
Hügelzählung kommt dagegen immer dasselbe heraus.

## Schritt 1: Die Karte einlesen

Die erste Zeile ist die Breite `b`, die zweite die Höhe `h`. Danach folgen `h` Zeilen der Karte. Die Karte speichern wir als Liste von Texten,
jede Zeile ein Text.

Zwei Dinge machen das Einlesen robust:

- `rstrip('\r\n')` entfernt nur den Zeilenumbruch (nicht die Leerzeichen am Ende, denn die gehören zur Karte).
- Manche Editoren löschen Leerzeichen am Zeilenende. `ljust(b)` füllt jede Zeile wieder bis zur Breite `b` mit Leerzeichen auf. Fehlt eine Zeile
  ganz (weil sie nur aus Leerzeichen bestand), nehmen wir eine leere Zeile.

````python
def lies_karte(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        breite = int(f.readline())
        hoehe = int(f.readline())
        karte = []
        for z in range(hoehe):
            zeile = f.readline().rstrip('\r\n')
            karte.append(zeile.ljust(breite))
    return karte


karte = lies_karte('beispieldaten/karte0.txt')
print(len(karte), 'Zeilen,', len(karte[0]), 'Spalten')
for zeile in karte:
    print('|' + zeile + '|')
````

In `karte0.txt` sieht man links oben einen Bau (`XXX`, `X X`, `X X`, `XXX`). Es gibt dort noch zwei weitere Baue und
einzelne Maulwurfshügel (einzelne `X`).

## Schritt 2: Ist an einer Stelle ein Hügel?

`ist_huegel(karte, z, s)` gibt `True` zurück, wenn in Zeile `z` und Spalte `s` ein `X` steht, und `False` sonst. Liegt die Stelle
**außerhalb der Karte**, gibt es dort keinen Hügel. Damit müssen wir später nie mehr an den Rand denken.

````python
def ist_huegel(karte, z, s):
    if z < 0 or z >= len(karte) or s < 0 or s >= len(karte[0]):
        return False                        # außerhalb der Karte
    return karte[z][s] == 'X'


print(ist_huegel(karte, 0, 0), ist_huegel(karte, 1, 1), ist_huegel(karte, -1, 0), ist_huegel(karte, 0, 100))
````

## Schritt 3: Beginnt hier ein Bau?

Das Muster schreiben wir als Liste von vier Texten auf. An der Stelle `(z, s)` (der linken oberen Ecke) vergleichen wir jedes der 12 Felder
mit dem Muster: Steht im Muster ein `X`, muss auf der Karte ein Hügel sein, sonst muss das Feld leer sein.
Das erledigt der Vergleich `ist_huegel(...) != (MUSTER[dz][ds] == 'X')`: Er ist wahr, wenn Karte und Muster **nicht** übereinstimmen.
Sobald ein Feld nicht stimmt, geben wir sofort `False` zurück. Stimmen alle 12, ist es ein Bau.

````python
MUSTER = ['XXX',
          'X X',
          'X X',
          'XXX']


def bau_bei(karte, z, s):
    for dz in range(4):
        for ds in range(3):
            soll_huegel = MUSTER[dz][ds] == 'X'
            if ist_huegel(karte, z + dz, s + ds) != soll_huegel:
                return False
    return True


print(bau_bei(karte, 0, 0), bau_bei(karte, 0, 1), bau_bei(karte, 1, 0))
````

## Schritt 4: Alle Baue zählen (mit Schutz vor Überlappung)

Wir gehen die Karte Zeile für Zeile, von links nach rechts, durch. An jeder Stelle, an der ein Bau beginnt, prüfen wir noch, ob einer
seiner 10 Hügel schon zu einem früher gefundenen Bau gehört. Dazu gibt es eine zweite Karte `benutzt` mit lauter `False`. Nur wenn kein Hügel
des neuen Baus schon benutzt ist, zählen wir ihn und setzen seine 10 Hügel auf `benutzt`.

`benutzt` bauen wir mit Schleifen auf (nicht mit `[[False] * b] * h`, denn das würde dieselbe Zeile mehrfach benutzen).
Geprüft werden nur die 10 Hügel des Musters (die zwei Felder in der Mitte sind leer).

````python
def zaehle_baue(karte):
    hoehe = len(karte)
    breite = len(karte[0])

    benutzt = []
    for z in range(hoehe):
        zeile = []
        for s in range(breite):
            zeile.append(False)
        benutzt.append(zeile)

    anzahl = 0
    for z in range(hoehe):
        for s in range(breite):
            if bau_bei(karte, z, s):
                # Gehört schon ein Hügel dieses Baus zu einem anderen Bau?
                schon_benutzt = False
                for dz in range(4):
                    for ds in range(3):
                        if MUSTER[dz][ds] == 'X' and z + dz < hoehe and s + ds < breite and benutzt[z + dz][s + ds]:
                            schon_benutzt = True
                if not schon_benutzt:
                    anzahl = anzahl + 1
                    for dz in range(4):
                        for ds in range(3):
                            if MUSTER[dz][ds] == 'X':
                                benutzt[z + dz][s + ds] = True
    return anzahl


print(zaehle_baue(lies_karte('beispieldaten/karte0.txt')))
````

In `karte0.txt` gibt es 3 Baue. Wie der Name schon sagt, sind die einzelnen Hügel dazwischen gewöhnliche Maulwürfe.

## Schritt 5: Die zweite, einfache Lösung

Jeder Hügel eines Baus hat mindestens einen Nachbarhügel oben, unten, links oder rechts. Ein einzelner Maulwurfshügel hat keinen.
Wir zählen also alle Hügel mit mindestens einem Nachbarhügel und teilen durch 10 (`//` ist die Division ohne Nachkommastellen).
Die Hilfsfunktion `ist_huegel` erlaubt auch Stellen außerhalb der Karte, deshalb braucht man am Rand keine Sonderbehandlung.

````python
def zaehle_baue_einfach(karte):
    huegel_in_bauen = 0
    for z in range(len(karte)):
        for s in range(len(karte[0])):
            if ist_huegel(karte, z, s):
                if (ist_huegel(karte, z - 1, s) or ist_huegel(karte, z + 1, s)
                        or ist_huegel(karte, z, s - 1) or ist_huegel(karte, z, s + 1)):
                    huegel_in_bauen = huegel_in_bauen + 1
    return huegel_in_bauen // 10


print(zaehle_baue_einfach(lies_karte('beispieldaten/karte0.txt')))
````

## Schritt 6: Alle Beispiele

Wir zählen mit beiden Lösungen. Als Kontrolle zählen wir außerdem **alle** Stellen, an denen das Muster passt (ohne Schutz vor Überlappung).

````python
def zaehle_muster_alle(karte):
    anzahl = 0
    for z in range(len(karte)):
        for s in range(len(karte[0])):
            if bau_bei(karte, z, s):
                anzahl = anzahl + 1
    return anzahl


print('Karte     Muster (alle)  Baue (ohne Überlappung)  einfach (Hügel / 10)')
for nr in range(7):
    karte = lies_karte(f'beispieldaten/karte{nr}.txt')
    print(f'karte{nr}.txt {zaehle_muster_alle(karte):10} {zaehle_baue(karte):18} {zaehle_baue_einfach(karte):22}')
````

Die Spalten „Baue (ohne Überlappung)“ und „einfach“ sind bei allen sieben Karten gleich. Die reine Mustersuche zählt bei `karte4.txt`
(3193 statt 3189) und `karte5.txt` (20 statt 16) **mehr** Baue. Das sind scheinbare Baue: Stoßen echte Baue so aneinander, kann das Muster
auch an einer Stelle passen, die aus Teilen mehrerer Baue besteht. Ein solcher Bau würde Hügel mit anderen Bauen teilen. Baue überlappen sich aber nicht.

Die offiziellen Lösungshinweise nennen als Ergebnisse 3, 37, 32, 318, 3189, 16 und 559. Das sind genau unsere Zahlen aus den Spalten
„ohne Überlappung“ und „einfach“. Die Hinweise erlauben ausdrücklich auch die einfache Mustersuche, falls man die Aufgabenstellung so liest,
dass sich Baue auch nicht scheinbar überlappen. Dann wären 3193 und 20 richtig. In die Dokumentation gehört, welche Lesart man gewählt hat.

Die beiden Lösungen sind sehr schnell: Selbst bei `karte4.txt` (140 Spalten, 500 Zeilen) braucht das Programm nur Sekundenbruchteile.

## Für Neugierige: Wo sind die scheinbaren Baue?

`karte5.txt` ist klein (14 Spalten, 20 Zeilen) und besteht aus 16 Bauen. Wir zeigen die Karte und markieren die vier Stellen, an
denen das Muster passt, obwohl es kein eigener Bau ist. Jede Stelle ist die linke obere Ecke eines scheinbaren Baus.

````python
karte = lies_karte('beispieldaten/karte5.txt')

# Alle Stellen, an denen das Muster passt, aber ein Hügel schon zu einem früheren Bau gehört
benutzt = set()
scheinbar = []
for z in range(len(karte)):
    for s in range(len(karte[0])):
        if bau_bei(karte, z, s):
            felder = set()
            for dz in range(4):
                for ds in range(3):
                    if MUSTER[dz][ds] == 'X':
                        felder.add((z + dz, s + ds))
            if felder & benutzt:                 # gemeinsame Hügel mit einem früheren Bau?
                scheinbar.append((z, s))
            else:
                benutzt = benutzt | felder

print('Scheinbare Baue (Zeile, Spalte):', scheinbar)
print()
for z in range(len(karte)):
    print(f'{z:2} |{karte[z]}|')
````

An diesen vier Stellen passt das Muster, aber mindestens einer seiner Hügel gehört schon zu einem anderen, früher gefundenen Bau.
Deshalb ist es kein eigener Bau. Die vier scheinbaren Baue erklären den Unterschied zwischen 20 und 16 bei dieser Karte.
