Das Duo „Fake that“ ist der aktuelle Stern am Pop-Himmel. Sie bringen einen Hit nach dem anderen heraus, aber nach einiger Zeit
dämmert es selbst ihren größten Fans, dass die Texte ihrer Silben-Songs (so nennt das Duo seinen Stil) sich alle sehr ähneln.
Das letzte Album hieß wohl nicht umsonst „It's Always The Same“.

Nellie, ein mittlerweile etwas gelangweilter Fan der beiden, hat festgestellt, wie die Texte der Silben-Songs funktionieren:

- Eine Silbe wird aus einem Konsonanten und einem Vokal gebildet. Beispiele: 'do', 'nu', 'la'.
- Eine Zeile besteht aus einer ungeraden Anzahl von Wiederholungen einer Grundsilbe, wobei derjenigen in der Mitte ein 'p di'
  angehängt wird. Beispiele: 'sup di', 'da dap di da', 'ne ne nep di ne ne'.
- Eine Strophe besteht aus mindestens zwei Zeilen. Die Menge der Konsonanten und Vokale, aus denen die Grundsilben einer
  Strophe gebildet werden, ist immer sehr klein, z. B. {s, u, a}. Die Anzahl der Silbenwiederholungen ist für alle Zeilen
  einer Strophe gleich (Nellie spricht deshalb von der Silbenzahl einer Strophe). Am Ende einer Strophe steht gelegentlich ein
  markiger Call wie 'yeah!', 'yo man', 'fake that!' oder ähnlich pseudo-cooles Zeugs.
- Ein Song besteht aus mindestens zwei Strophen. Die Zeilenzahlen der Strophen eines Songs folgen einem Muster, z. B.
  „immer 3 Zeilen“, „abwechselnd 4 und 6 Zeilen“ usw. Auch die Silbenzahlen der Strophen folgen ähnlichen Mustern.

Nellie zweifelt nun an der Kreativität von „Fake that“. Solche Songtexte kann man sich bestimmt vom Computer schreiben lassen!

#### Junioraufgabe 1

Schreibe einen „Songwriter“, also ein Programm, das einen nach Nellies Regeln aufgebauten Text eines Silben-Songs erzeugen
kann. Stelle dabei so weit wie möglich sicher, dass die von deinem Programm erzeugten Texte sich voneinander unterscheiden.
Gib in der Dokumentation mindestens drei unterschiedliche Songtexte an, die dein Programm erzeugt hat.

Beispieldaten gibt es zu dieser Aufgabe nicht.

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Zerlegen in Teilaufgaben:** Ein Song besteht aus Strophen, eine Strophe aus Zeilen, eine Zeile aus Silben. Für jede Ebene schreiben wir eine Funktion (`silbe`, `zeile`, `strophe`, `song`), die die darunter liegenden benutzt.
- **Zufall:** Mit dem Modul `random` wählen wir Buchstaben, Silbenzahlen und Calls zufällig aus, damit jeder Song anders wird.
- **Teilmengen auswählen:** Für jede Strophe wählen wir zufällig wenige Konsonanten und Vokale aus (`random.sample`). Nur aus diesen werden die Grundsilben der Strophe gebaut.
- **Muster als zyklische Liste:** Ein Muster wie „abwechselnd 4 und 6 Zeilen“ ist die Liste `[4, 6]`. Die Strophe Nummer `i` bekommt den Eintrag `liste[i % len(liste)]`. Das `%` (Rest) wiederholt die Liste immer wieder von vorn.
- **Prüfen, ob Regeln eingehalten wurden:** Eine eigene Funktion `pruefe_song` untersucht den erzeugten Text und kontrolliert die Regeln der Aufgabe.
- **Doppelte vermeiden:** Wir merken uns die schon erzeugten Songs in einer Liste oder Menge und erzeugen einen neuen Song, wenn er schon einmal vorkam.

**Nützliche Funktionen und Bibliotheken**
- `import random` mit `random.choice`, `random.sample`, `random.randint`, `random.seed` (für wiederholbare Ergebnisse), `str.join`, `str.split`, `%`

## Die Idee

- **Silbe:** ein Konsonant plus ein Vokal, zum Beispiel `'ka'`.
- **Zeile:** eine Grundsilbe, `anzahl` mal wiederholt (eine ungerade Zahl). Die mittlere Silbe bekommt `'p di'` angehängt. Bei 7 Wiederholungen der Silbe `ka`: `ka ka ka kap di ka ka ka`.
- **Strophe:** `zeilen` Zeilen mit derselben Silbenzahl. Zuerst wählen wir für die Strophe wenige Konsonanten und Vokale (1 bis 3). Jede Zeile bekommt eine zufällige Grundsilbe aus diesen Buchstaben.
  Mit der Wahrscheinlichkeit 1/2 steht am Ende ein „Call“ (`yeah!`, `yo man`, ...).
- **Song:** mindestens 2 Strophen. Für die Zeilenzahlen und die Silbenzahlen wählen wir je ein Muster (eine Liste, die sich wiederholt). Die Silbenzahlen müssen ungerade sein.

## Schritt 1: Silbe und Zeile

`random.choice(liste)` wählt ein Element aus einer Liste (oder einem Text), `random.sample(liste, n)` wählt `n` verschiedene.

**Ein Randfall:** Die Silbe `di` lassen wir nicht zu. Sonst entstünde zum Beispiel die Zeile `di dip di di`, und man könnte nicht mehr erkennen, welches `di` das Anhängsel der mittleren Silbe ist.
Das fiel uns bei der Prüffunktion in Schritt 4 auf.

In `zeile` bauen wir eine Liste von Silben und hängen an die mittlere Silbe `'p di'` an. Bei `anzahl = 7` ist der Index der mittleren Silbe `anzahl // 2 = 3`
(Indizes 0 bis 6). Dann fügen wir die Silben mit `' '.join(...)` zu einem Text zusammen.

```python
import random

KONSONANTEN = 'bdfghjklmnprstvwz'
VOKALE = 'aeiou'


def silbe(konsonanten, vokale):
    s = random.choice(konsonanten) + random.choice(vokale)
    while s == 'di':                                       # 'di' ist verboten (siehe unten)
        s = random.choice(konsonanten) + random.choice(vokale)
    return s


def zeile(anzahl, konsonanten, vokale):
    grundsilbe = silbe(konsonanten, vokale)
    silben = []
    for i in range(anzahl):
        silben.append(grundsilbe)
    silben[anzahl // 2] = grundsilbe + 'p di'              # mittlere Silbe
    return ' '.join(silben)


random.seed(1)
print(silbe('sd', 'ua'))
print(zeile(3, 'sd', 'ua'))
print(zeile(7, 'k', 'a'))
```

## Schritt 2: Strophe

Für jede Strophe schränken wir die Buchstaben ein: Wir wählen mit `random.sample` zufällig 1 bis 3 Konsonanten und 1 bis 3 Vokale aus. (Die Anzahl 1 kommt häufiger vor als 3, weil `[1, 1, 2, 3]` die 1 doppelt enthält. Damit wird die Menge „sehr klein“.)
Jede Zeile bekommt eine eigene Grundsilbe aus diesen Buchstaben, aber alle haben dieselbe Silbenzahl.

Mit der Wahrscheinlichkeit 1/2 hängen wir einen Call an: `random.random()` liefert eine Zahl zwischen 0 und 1, und `< 0.5` ist in der Hälfte der Fälle wahr.

```python
CALLS = ['yeah!', 'yo man', 'fake that!', 'oh baby!', 'hey hey!', 'come on!']


def strophe(zeilenzahl, silbenzahl):
    anzahl_konsonanten = random.choice([1, 1, 2, 3])
    anzahl_vokale = random.choice([1, 1, 2, 3])
    konsonanten = random.sample(KONSONANTEN, anzahl_konsonanten)
    vokale = random.sample(VOKALE, anzahl_vokale)
    while konsonanten == ['d'] and vokale == ['i']:         # daraus ließe sich nur die verbotene Silbe 'di' bilden
        konsonanten = random.sample(KONSONANTEN, anzahl_konsonanten)
        vokale = random.sample(VOKALE, anzahl_vokale)
    zeilen = []
    for i in range(zeilenzahl):
        zeilen.append(zeile(silbenzahl, konsonanten, vokale))
    if random.random() < 0.5:
        zeilen.append(random.choice(CALLS))
    return zeilen


random.seed(2)
for text in strophe(4, 5):
    print(text)
```

## Schritt 3: Der Song

Für die Muster verwenden wir Listen, die sich wiederholen. Die Strophe Nummer `i` (von 0 an gezählt) bekommt `zeilen_muster[i % len(zeilen_muster)]` Zeilen und `silben_muster[i % len(silben_muster)]` Silben.
Beispiele: `[3]` heißt „immer 3“, `[4, 6]` „abwechselnd 4 und 6“, `[2, 3, 4]` „2, 3, 4, 2, 3, 4, ...“.

Der Song soll mindestens 2 Strophen haben, wir nehmen zufällig 2 bis 5. Für die Muster nehmen wir eine zufällige Auswahl aus fertigen Mustern. Die Silbenzahlen müssen **ungerade** sein
(die Zeile hat eine ungerade Anzahl von Silben, damit es eine Mitte gibt).

```python
ZEILEN_MUSTER = [[3], [4], [4, 6], [2, 3, 4], [3, 5], [2, 4, 2, 6]]
SILBEN_MUSTER = [[3], [5], [3, 5], [3, 5, 7], [7, 3], [5, 3, 3]]


def song():
    zeilen_muster = random.choice(ZEILEN_MUSTER)
    silben_muster = random.choice(SILBEN_MUSTER)
    strophen = []
    for i in range(random.randint(2, 5)):
        zeilenzahl = zeilen_muster[i % len(zeilen_muster)]
        silbenzahl = silben_muster[i % len(silben_muster)]
        strophen.append(strophe(zeilenzahl, silbenzahl))
    return strophen, zeilen_muster, silben_muster


def song_als_text(strophen):
    teile = []
    for st in strophen:
        teile.append('\n'.join(st))
    return '\n\n'.join(teile)


random.seed(3)
strophen, zm, sm = song()
print('Muster Zeilen:', zm, ' Muster Silben:', sm)
print()
print(song_als_text(strophen))
```

## Schritt 4: Die Regeln prüfen

Damit wir sicher sind, dass das Programm keine Regel verletzt, schreiben wir eine **Prüffunktion**, die einen Song untersucht. Sie kennt nur den Text und die Zeilen, nicht den Erzeugungsvorgang.

Für jede Zeile (außer einem Call) prüft sie:

1. Die Zeile besteht aus Wörtern. Das Wort `di` gehört zur Mitte. Alle anderen Wörter sind Silben: Eine Grundsilbe `xy` (Konsonant + Vokal), wiederholt, und in der Mitte `xyp` direkt vor `di`.
2. Die Zahl der Silben ist ungerade, und `xyp di` steht genau in der Mitte.

Für jede Strophe:

3. Mindestens zwei Zeilen (ohne Call), alle haben dieselbe Silbenzahl.
4. Die Menge der Konsonanten und der Vokale ist klein (höchstens 3 jeweils).

Für den Song: mindestens zwei Strophen. Als Calls erkennt die Prüfung die Texte aus der Liste `CALLS`.

```python
def pruefe_zeile(text):
    woerter = text.split(' ')
    if woerter.count('di') != 1:
        return None
    pos = woerter.index('di')
    silben = woerter[:pos] + woerter[pos + 1:]               # alle Silben ohne das 'di'
    anzahl = len(silben)
    if anzahl % 2 == 0 or pos != anzahl // 2 + 1 or len(silben[anzahl // 2]) != 3:
        return None
    grundsilbe = silben[0][:2]                               # die ersten zwei Buchstaben (bei nur einer Silbe ist das auch die mittlere)
    if len(grundsilbe) != 2 or grundsilbe[0] not in KONSONANTEN or grundsilbe[1] not in VOKALE:
        return None
    for i in range(anzahl):
        erwartet = grundsilbe + 'p' if i == anzahl // 2 else grundsilbe
        if silben[i] != erwartet:
            return None
    return anzahl, grundsilbe


def pruefe_song(strophen):
    if len(strophen) < 2:
        return False
    for st in strophen:
        zeilen = []
        for text in st:
            if text not in CALLS:
                zeilen.append(text)
        if len(zeilen) < 2:
            return False
        ergebnisse = []
        for text in zeilen:
            erg = pruefe_zeile(text)
            if erg is None:
                return False
            ergebnisse.append(erg)
        anzahlen = set()
        konsonanten = set()
        vokale = set()
        for anzahl, grundsilbe in ergebnisse:
            anzahlen.add(anzahl)
            konsonanten.add(grundsilbe[0])
            vokale.add(grundsilbe[1])
        if len(anzahlen) != 1 or len(konsonanten) > 3 or len(vokale) > 3:
            return False
    return True


print(pruefe_zeile('da dap di da'), pruefe_zeile('sup di'), pruefe_zeile('ne ne nep di ne ne'))
print(pruefe_zeile('da da dap da'), pruefe_zeile('da dap di da da'))
print('Der Song von oben ist korrekt:', pruefe_song(strophen))
```

Die ersten drei Zeilen sind Beispiele aus der Aufgabe und werden akzeptiert (das Ergebnis ist die Silbenzahl und die Grundsilbe). Zeilen, bei denen das `p di` fehlt oder nicht in der Mitte steht, werden abgelehnt (`None`).

## Schritt 5: Viele verschiedene Songs

Die Aufgabe verlangt, dass die Songs sich voneinander unterscheiden und dass man mindestens drei davon in der Dokumentation zeigt. Wir merken uns den Text jedes erzeugten Songs in einer Liste `bisher`.
Erzeugt der Zufall einen schon vorhandenen Song, ziehen wir einen neuen. (Bei so vielen Möglichkeiten passiert das praktisch nie, aber so ist es **sichergestellt**.)
Dazu prüfen wir noch jeden Song mit `pruefe_song`.

```python
bisher = []


def neuer_song():
    while True:
        strophen, zeilen_muster, silben_muster = song()
        text = song_als_text(strophen)
        if text not in bisher and pruefe_song(strophen):
            bisher.append(text)
            return text, zeilen_muster, silben_muster


random.seed(2013)
for nr in range(1, 4):
    text, zm, sm = neuer_song()
    print(f'===== Song {nr}  (Zeilen pro Strophe: {zm}, Silben pro Zeile: {sm}) =====')
    print(text)
    print()
```

Drei verschiedene Songs, alle nach Nellies Regeln. In der ersten Zeile jedes Songs steht, welches Muster die Zeilen- und Silbenzahlen der Strophen hatten.
Zum Beispiel heißt `Zeilen pro Strophe: [4, 6]`, dass die Strophen abwechselnd 4 und 6 Zeilen haben (ein Call am Ende zählt dabei nicht mit).

## Für Neugierige: Wie sicher sind „keine Doppelten“?

Wie viele verschiedene Songs kann das Programm überhaupt erzeugen? Schon eine einzige Strophe mit 4 Zeilen und 5 Silben hat sehr viele Möglichkeiten (jede Zeile hat eine zufällige Grundsilbe aus den gewählten Buchstaben).
Wir erzeugen 2000 Songs und zählen, ob es Doppelte gab, und prüfen jeden Song.

```python
random.seed(99)
texte = set()
alle_korrekt = True
for versuch in range(2000):
    strophen, zm, sm = song()
    if not pruefe_song(strophen):
        alle_korrekt = False
    texte.add(song_als_text(strophen))
print('Verschiedene Songs:', len(texte), 'von 2000')
print('Alle Songs entsprechen den Regeln:', alle_korrekt)
```

Fast alle der 2000 Songs sind verschieden. (Wenige können doppelt vorkommen, wenn die Strophen sehr klein sind, zum Beispiel zwei Strophen mit je 2 Zeilen und nur einem Konsonanten und einem Vokal.)
Das Programm lässt solche Doppelten mit `neuer_song()` gar nicht erst zu.
