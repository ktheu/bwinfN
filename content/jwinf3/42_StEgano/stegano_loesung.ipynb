## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Einer Kette von Verweisen folgen:** Jedes versteckte Pixel „zeigt“ auf das nächste (wie bei einer Schatzsuche, bei der
  jeder Zettel verrät, wo der nächste liegt). Man startet am Anfang und folgt den Verweisen, bis ein Pixel das Ende anzeigt.
  Die Anzahl der Schritte kennt man vorher nicht, deshalb nimmt man eine `while`-Schleife.
- **Rechnen mit Rest (Modulo):** Geht man rechts über den Bildrand hinaus, kommt man links wieder herein. Das rechnet man mit
  `x = (x + g) % breite`. Genauso nach unten mit `y = (y + b) % hoehe`.
- **Koordinaten in eine Position umrechnen:** Die Pixel eines Bildes werden zeilenweise hintereinander gespeichert. Das Pixel
  in Spalte `x` und Zeile `y` ist das Pixel Nummer `y * breite + x`.
- **Zahl in Zeichen umwandeln:** `chr(zahl)` liefert das Zeichen zu einem Code, zum Beispiel `chr(72)` = `'H'`
  (Gegenstück: `ord('H')` = 72).

**Nützliche Bibliotheken und Funktionen**
- Für das Textformat **PPM** braucht man nur `open`, `read` und `split`.
- Für **PNG**-Bilder braucht man die Bibliothek **Pillow** (Modul `PIL`). Installation: `pip install pillow`.
  Wir brauchen nur vier Befehle: `Image.open(dateiname)` öffnet das Bild, `.convert('RGB')` stellt sicher, dass jedes Pixel
  aus genau drei Werten besteht, `.size` liefert Breite und Höhe, `.tobytes()` liefert alle Farbwerte hintereinander.

## Die Idee

Wir halten zwei Zahlen fest: die Spalte `x` und die Zeile `y` des aktuellen Pixels. Am Anfang sind beide 0 (oben links).
Dann wiederholen wir:

1. Farbwerte `r`, `g`, `b` des Pixels an der Stelle `(x, y)` nachsehen.
2. Das Zeichen `chr(r)` an die Nachricht anhängen.
3. Sind `g` und `b` beide 0, ist die Nachricht zu Ende.
4. Sonst `x` um `g` und `y` um `b` weiterschieben, und zwar im Kreis (Modulo), und bei 1 weitermachen.

Jedes Pixel muss dabei höchstens einmal angesehen werden. Das Programm ist also schnell, auch bei Bildern mit
Millionen Pixeln. Langsam ist nur das **Einlesen** großer Textdateien.

## Schritt 1: Wie speichern wir ein Bild?

Wir speichern ein Bild in drei Teilen:

- `breite`: die Anzahl der Spalten,
- `hoehe`: die Anzahl der Zeilen,
- `daten`: **eine** lange Liste mit allen Farbwerten, Pixel für Pixel, Zeile für Zeile von oben links:
  `[r0, g0, b0, r1, g1, b1, r2, g2, b2, ...]`

Das Pixel Nummer `i` hat dann seine Farbwerte an den Positionen `3 * i`, `3 * i + 1` und `3 * i + 2`.
Und weil die Pixel zeilenweise gespeichert sind, hat das Pixel in Spalte `x` und Zeile `y` die Nummer `y * breite + x`.
Die Funktion `pixel_an` rechnet das für uns aus.

````python
def pixel_an(daten, breite, x, y):
    nummer = y * breite + x          # das wievielte Pixel (von 0 an gezählt) ist es?
    anfang = 3 * nummer              # an dieser Stelle der Liste stehen seine drei Farbwerte
    r = daten[anfang]
    g = daten[anfang + 1]
    b = daten[anfang + 2]
    return r, g, b


# Kleiner Test mit einem Bild aus 3 Spalten und 2 Zeilen (6 Pixel, also 18 Zahlen)
mini = [10, 11, 12,   20, 21, 22,   30, 31, 32,
        40, 41, 42,   50, 51, 52,   60, 61, 62]
print(pixel_an(mini, 3, 0, 0))     # oben links
print(pixel_an(mini, 3, 2, 0))     # oben rechts
print(pixel_an(mini, 3, 1, 1))     # unten in der Mitte
````

## Schritt 2: Die Nachricht auslesen

Das ist der Kern der Aufgabe, direkt nach der Idee oben. Die Variable `weiter` sagt, ob die Schleife noch
ein weiteres Pixel besuchen soll. Sie wird `False`, sobald ein Pixel `g = 0` und `b = 0` hat.

Das `%` sorgt dafür, dass `x` immer zwischen 0 und `breite - 1` bleibt (und `y` zwischen 0 und `hoehe - 1`).
Ein Beispiel: In einem Bild mit 9 Spalten ergibt `(4 + 6) % 9` die Spalte 1. Man ist rechts über den Rand gegangen
und links wieder hereingekommen.

*Tipp:* Falls dein Programm einmal nicht mehr aufhört (zum Beispiel wegen eines Tippfehlers),
kannst du es in Jupyter mit dem Stopp-Knopf in der Werkzeugleiste (Kernel → Interrupt) anhalten.

````python
def entschluessele(breite, hoehe, daten):
    nachricht = ''
    x = 0                                   # wir beginnen oben links
    y = 0
    weiter = True
    while weiter:
        r, g, b = pixel_an(daten, breite, x, y)
        nachricht = nachricht + chr(r)      # r ist der Code des nächsten Zeichens
        if g == 0 and b == 0:
            weiter = False                  # Ende der Nachricht
        else:
            x = (x + g) % breite            # g Pixel nach rechts, am Rand wieder links hinein
            y = (y + b) % hoehe             # b Pixel nach unten, am Rand wieder oben hinein
    return nachricht
````

## Schritt 3: Ein erster Test mit dem Beispiel vom Aufgabenblatt

Bevor wir echte Bilder einlesen, bauen wir das Bild vom Aufgabenblatt nach. Seine Größe steht nicht in der Aufgabe, wir
wählen 9 mal 9 Pixel. Alle Pixel haben zunächst eine beliebige „Hintergrundfarbe“, nur die fünf Pixel der Nachricht
setzen wir mit den Werten aus der Tabelle. Ihre Positionen ergeben sich aus den Sprüngen:

| Pixel | Position | r | g | b | nächste Position |
|:-:|:-:|:-:|:-:|:-:|:-:|
| 1 | (0, 0) | „H“ | 4 | 2 | (4, 2) |
| 2 | (4, 2) | „a“ | 6 | 3 | (10 % 9, 5) = **(1, 5)**, rechts über den Rand |
| 3 | (1, 5) | „l“ | 2 | 6 | (3, 11 % 9) = **(3, 2)**, unten über den Rand |
| 4 | (3, 2) | „l“ | 5 | 4 | (8, 6) |
| 5 | (8, 6) | „o“ | 0 | 0 | Ende |

`ord('H')` liefert den Code 72 des Buchstabens `H`, so müssen wir die Codes nicht auswendig kennen.

````python
breite = 9
hoehe = 9
testdaten = [200, 100, 50] * (breite * hoehe)       # alle 81 Pixel bekommen dieselbe Hintergrundfarbe


def setze_pixel(daten, breite, x, y, r, g, b):
    anfang = 3 * (y * breite + x)
    daten[anfang] = r
    daten[anfang + 1] = g
    daten[anfang + 2] = b


setze_pixel(testdaten, breite, 0, 0, ord('H'), 4, 2)
setze_pixel(testdaten, breite, 4, 2, ord('a'), 6, 3)
setze_pixel(testdaten, breite, 1, 5, ord('l'), 2, 6)
setze_pixel(testdaten, breite, 3, 2, ord('l'), 5, 4)
setze_pixel(testdaten, breite, 8, 6, ord('o'), 0, 0)

print(entschluessele(breite, hoehe, testdaten))
````

## Schritt 4: Bilder im PPM-Format einlesen

Das PPM-Format (Variante „P3“) ist ein reines **Textformat**. Man kann die Datei sogar im Editor öffnen:

```
P3
16 12
255
72 231 237 178 192 211 160 176 202 ...
```

Erst steht `P3`, dann Breite und Höhe, dann der größte Farbwert (255) und danach für jedes Pixel drei Zahlen,
zeilenweise von oben links. Mit `split()` zerlegen wir die ganze Datei in einzelne Wörter:
`woerter[0]` ist `'P3'`, `woerter[1]` die Breite, `woerter[2]` die Höhe, `woerter[3]` der Maximalwert,
und ab `woerter[4]` kommen die Farbwerte. Das passt genau zu unserem Format `daten`.

````python
def lade_ppm(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        woerter = f.read().split()
    breite = int(woerter[1])
    hoehe = int(woerter[2])
    daten = []
    for i in range(4, len(woerter)):        # ab Position 4 stehen die Farbwerte
        daten.append(int(woerter[i]))
    return breite, hoehe, daten


breite, hoehe, daten = lade_ppm('beispieldaten/bild01.ppm')
print('Breite:', breite, 'Höhe:', hoehe)
print('Anzahl Farbwerte:', len(daten), '= 3 *', breite * hoehe)
print('Oben links:', pixel_an(daten, breite, 0, 0))
````

Das erste Pixel hat `r = 72`, das ist `chr(72)` = `'H'`. Mit `g = 231` und `b = 237` geht es weiter zu
Spalte `231 % 16 = 7` und Zeile `237 % 12 = 9`. Schauen wir uns die ersten Schritte genau an:

````python
x = 0
y = 0
for schritt in range(1, 5):
    r, g, b = pixel_an(daten, breite, x, y)
    print(f'Schritt {schritt}: Pixel ({x}, {y}) hat r={r} g={g} b={b} -> Zeichen {chr(r)!r}')
    x = (x + g) % breite
    y = (y + b) % hoehe
````

Das stimmt mit der Beschreibung überein. Jetzt lassen wir das ganze Programm laufen, zuerst für `bild01`:

````python
print(entschluessele(breite, hoehe, daten))
````

## Schritt 5: Die Bilder 1 bis 4 (PPM)

Für `bild01` bis `bild04` liegen PPM-Dateien vor. Wir fassen das Einlesen und Entschlüsseln in einer Funktion zusammen
und geben die Nachrichten aus. Das Einlesen der größeren Dateien kann einen Moment dauern (`bild04` hat 360 000 Pixel
und damit über eine Million Zahlen).

````python
def lies_nachricht_ppm(dateiname):
    breite, hoehe, daten = lade_ppm(dateiname)
    return entschluessele(breite, hoehe, daten)


nachrichten_ppm = []
for nr in range(1, 5):
    nachricht = lies_nachricht_ppm(f'beispieldaten/bild0{nr}.ppm')
    nachrichten_ppm.append(nachricht)
    print(f'--- bild0{nr} ({len(nachricht)} Zeichen)')
    print(nachricht)
    print()
````

## Schritt 6: PNG-Bilder mit Pillow einlesen

Für `bild05` bis `bild07` gibt es hier keine PPM-Dateien. Dafür lesen wir die PNG-Dateien mit der Bibliothek **Pillow**.
Das Ergebnis ist genau im selben Format wie bei `lade_ppm`: Breite, Höhe und eine lange Liste mit den Farbwerten.
So können wir `entschluessele` unverändert weiterbenutzen.

- `Image.open(dateiname)` öffnet das Bild.
- `.convert('RGB')` sorgt dafür, dass jedes Pixel genau drei Werte hat. Manche PNG-Dateien haben noch einen vierten Wert
  (Durchsichtigkeit, *Alpha*), den wir nicht brauchen.
- `.size` ist ein Paar aus Breite und Höhe.
- `.tobytes()` liefert alle Farbwerte hintereinander. Jeder Wert dieser Folge ist eine Zahl von 0 bis 255,
  man kann sie wie eine Liste mit `daten[i]` ansprechen.

````python
from PIL import Image


def lade_png(dateiname):
    bild = Image.open(dateiname).convert('RGB')
    breite, hoehe = bild.size
    daten = bild.tobytes()
    return breite, hoehe, daten


def lies_nachricht_png(dateiname):
    breite, hoehe, daten = lade_png(dateiname)
    return entschluessele(breite, hoehe, daten)


# Probe: Bild 1 bis 4 als PNG müssen dieselben Nachrichten liefern wie als PPM
for nr in range(1, 5):
    nachricht = lies_nachricht_png(f'beispieldaten/bild0{nr}.png')
    print(f'bild0{nr}: PNG und PPM liefern dieselbe Nachricht:', nachricht == nachrichten_ppm[nr - 1])
````

````python
nachricht7 = lies_nachricht_png('beispieldaten/bild07.png')
print(f'--- bild07 ({len(nachricht7)} Zeichen), die ersten 300 Zeichen:')
print(nachricht7[:300])
print()
print('die letzten 300 Zeichen:')
print(nachricht7[-300:])
````

`nachricht[:300]` sind die ersten 300 Zeichen eines Textes, `nachricht[-300:]` die letzten 300.

**Hinweis zu den Sonderzeichen:** Das `ß` ist kein ASCII-Zeichen (ASCII geht nur bis 127). In den Bildern ist es mit dem
Code 223 versteckt, und `chr(223)` ergibt genau dieses `ß`. Die Umlaute funktionieren genauso. Deshalb ist `chr` hier
die richtige Wahl, auch wenn die Aufgabe von ASCII spricht.
