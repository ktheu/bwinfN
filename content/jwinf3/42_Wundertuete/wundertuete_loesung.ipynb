Maras Kindergarten hat eine große Spende von Spielen erhalten. Darunter sind Würfelspiele, Kartenspiele und
Geschicklichkeitsspiele. Der Kindergarten beschließt, jedem Kind, das dieses Jahr den Kindergarten verlässt,
zum Abschied eine Wundertüte mit Spielen zu spendieren.

Nach Möglichkeit soll in jeder Wundertüte die gleiche Gesamtzahl von Spielen sein; außerdem soll jede der drei
Spielesorten möglichst gleichmäßig auf die Wundertüten verteilt sein. Die Unterschiede sollen möglichst gering sein
und es dürfen keine Spiele übrig bleiben.

Leider wird es nicht immer gelingen, die Spiele vollkommen gleichmäßig zu verteilen. Ein Beispiel: Es gibt vier
Würfelspiele (W), vier Kartenspiele (K) und zwei Geschicklichkeitsspiele (G), die auf drei Wundertüten verteilt werden
sollen. Wenn die Spiele so verteilt werden wie im Bild, dann unterscheiden sich die Gesamtzahlen der Spiele zwischen
je zwei Tüten um höchstens eins. Ebenso gilt auch für jede Spielesorte, dass sich ihre Anzahlen zwischen je zwei Tüten
um höchstens eins unterscheiden.

#### Junioraufgabe 1

Schreibe ein Programm, das Maras Kindergarten hilft, die Wundertüten zu packen. Als Eingabe erhält es die Anzahl jeder
Spielesorte sowie die Anzahl der gewünschten Wundertüten. Anschließend gibt es aus, wie die einzelnen Wundertüten
zusammengestellt werden sollen.

Wende dein Programm mindestens auf alle Beispiele an, die du auf den BWINF-Webseiten findest,
und dokumentiere die Ergebnisse.

#### Beispieldaten

Im Ordner `beispieldaten` stehen `wundertuete0.txt` bis `wundertuete5.txt` (`wundertuete0.txt` ist das Beispiel aus der
Aufgabenstellung). Jede Datei ist so aufgebaut:

- 1. Zeile: Anzahl `n` der Wundertüten
- 2. Zeile: Anzahl `k` der Spielesorten (es können auch mehr als drei Sorten sein)
- folgende `k` Zeilen: Anzahl der Spiele jeder Sorte

## Was dir bei dieser Aufgabe hilft

**Algorithmische Kenntnisse**
- **Reihum verteilen** (englisch *Round Robin*): Genau wie beim Austeilen von Spielkarten bekommt jede Tüte der Reihe nach
  ein Spiel. Nach der letzten Tüte geht es wieder bei der ersten weiter. Das ist die ganze Lösungsidee.
- **Rechnen mit Rest:** `//` ist die Division ohne Nachkommastellen, `%` (Modulo) ist der Rest. Mit `%` kann man
  „im Kreis zählen“: `(tuete + 1) % n` springt nach der letzten Tüte wieder auf die erste.
- **Eine Tabelle als Liste von Listen:** Eine Zeile pro Tüte, eine Spalte pro Spielesorte.

**Nützliche Funktionen** (Bibliotheken braucht man hier keine)
- `sum(liste)`, `max(liste)`, `min(liste)` für die Summe, den größten und den kleinsten Wert einer Liste
- `divmod(a, b)` liefert `a // b` und `a % b` auf einmal (nur für die schnellere Variante am Ende)

## Die Idee

Wir teilen die Spiele aus wie Spielkarten: Immer ein Spiel in die nächste Tüte, und nach der letzten Tüte wieder
von vorn. Das machen wir zuerst mit allen Spielen der ersten Sorte, dann mit der zweiten Sorte, und so weiter.

Wichtig ist nur eine Kleinigkeit: Bei einer **neuen Sorte** fangen wir nicht wieder bei Tüte 1 an, sondern bei der Tüte,
die als nächstes dran wäre. Sonst bekäme Tüte 1 von jeder Sorte einen Rest-Bonus, und die letzte Tüte ginge immer leer aus.
Dann wäre Tüte 1 am Ende deutlich voller.

Warum ist das gleichmäßig? In jeder Runde bekommt jede Tüte höchstens ein Spiel. Deshalb unterscheiden sich zwei Tüten
nie um mehr als ein Spiel, weder bei den Gesamtzahlen noch bei einer einzelnen Sorte.

## Schritt 1: Die Datei einlesen

Die Datei enthält nur ganze Zahlen, getrennt durch Zeilenumbrüche. `read()` liest den ganzen Inhalt als einen Text,
`split()` zerlegt ihn an den Leerzeichen und Zeilenumbrüchen in eine Liste von Wörtern (hier: Zahlen als Text).
Mit `int(...)` machen wir echte Zahlen daraus. Leere Zeilen am Ende der Datei stören dabei nicht.

```python
def lies_wundertuete(dateiname):
    with open(dateiname, encoding='utf-8') as f:
        zahlen = f.read().split()
    anzahl_tueten = int(zahlen[0])      # 1. Zeile
    anzahl_sorten = int(zahlen[1])      # 2. Zeile
    sorten = []                         # ab der 3. Zeile: Anzahl der Spiele jeder Sorte
    for i in range(anzahl_sorten):
        sorten.append(int(zahlen[2 + i]))
    return anzahl_tueten, sorten


n, sorten = lies_wundertuete('beispieldaten/wundertuete0.txt')
print('Tüten:', n)
print('Spiele je Sorte:', sorten)
```

## Schritt 2: Eine Sorte reihum verteilen

Wir fangen klein an: Wir verteilen **vier** Spiele einer Sorte auf **drei** Tüten und schauen, was passiert.
Die Tüten sind die Positionen 0, 1 und 2 einer Liste (Tüte 1, 2 und 3). Die Variable `tuete` merkt sich,
welche Tüte als nächstes ein Spiel bekommt.

Die Zeile `tuete = (tuete + 1) % n` rechnet im Kreis: Nach 0 kommt 1, nach 1 kommt 2, und nach 2 kommt wieder 0,
weil `3 % 3` den Rest 0 ergibt.

```python
n = 3
spiele_in_tuete = [0, 0, 0]     # am Anfang ist in jeder Tüte nichts
tuete = 0                       # wir beginnen bei Tüte 1 (Position 0)

for spiel in range(4):
    spiele_in_tuete[tuete] = spiele_in_tuete[tuete] + 1
    print('Spiel', spiel + 1, 'kommt in Tüte', tuete + 1, '->', spiele_in_tuete)
    tuete = (tuete + 1) % n     # weiter zur nächsten Tüte, nach der letzten wieder zur ersten
```

## Schritt 3: Alle Sorten verteilen

Jetzt machen wir es für alle Sorten. Die Tabelle `tueten` hat eine Zeile pro Tüte und eine Spalte pro Sorte:
`tueten[t][s]` ist die Anzahl der Spiele der Sorte `s` in der Tüte `t`.

Zwei Dinge sind wichtig:
- Die Tabelle bauen wir mit Schleifen auf. Die kürzere Schreibweise `[[0] * k] * n` wäre ein Fehler: Sie legt nur **eine**
  Zeile an und benutzt sie `n`-mal, eine Änderung würde dann in allen Tüten gleichzeitig auftauchen.
- Die Variable `tuete` setzen wir **nicht** bei jeder neuen Sorte zurück. Sie läuft über alle Sorten hinweg weiter
  (siehe „Die Idee“).

```python
def verteile_reihum(n, sorten):
    k = len(sorten)

    # Tabelle: n Zeilen (Tüten), k Spalten (Sorten), am Anfang überall 0
    tueten = []
    for t in range(n):
        zeile = []
        for s in range(k):
            zeile.append(0)
        tueten.append(zeile)

    tuete = 0                                   # die Tüte, die als nächstes ein Spiel bekommt
    for s in range(k):                          # Sorte für Sorte ...
        for spiel in range(sorten[s]):          # ... jedes Spiel dieser Sorte ...
            tueten[tuete][s] = tueten[tuete][s] + 1     # ... in die aktuelle Tüte legen
            tuete = (tuete + 1) % n                     # und zur nächsten Tüte weitergehen
    return tueten


n, sorten = lies_wundertuete('beispieldaten/wundertuete0.txt')
tueten = verteile_reihum(n, sorten)
print(tueten)
```

## Schritt 4: Übersichtliche Ausgabe

Die Liste von Listen ist schwer zu lesen. Wir geben eine Tabelle aus: eine Zeile pro Tüte mit der Anzahl jeder Sorte
(`S1`, `S2`, ...) und der Gesamtzahl der Spiele in der Tüte. Die Gesamtzahl berechnet `sum(zeile)`.

In einem f-String legt `:>5` fest, dass die Ausgabe mindestens 5 Zeichen breit und rechtsbündig sein soll.
So stehen die Zahlen schön untereinander.

Bei sehr vielen Tüten (`wundertuete5.txt` hat 97) soll man die Ausgabe kürzen können. Mit dem Parameter `anzahl`
zeigt die Funktion nur die ersten `anzahl` Tüten und die letzte Tüte, dazwischen steht `...`.

```python
def zeige(tueten, anzahl=None):
    k = len(tueten[0])
    letzte = len(tueten) - 1

    kopf = f'{"Tüte":>6}'
    for s in range(k):
        kopf = kopf + f'{"S" + str(s + 1):>5}'
    kopf = kopf + f'{"Gesamt":>8}'
    print(kopf)

    for t in range(len(tueten)):
        if anzahl is None or t < anzahl or t == letzte:
            zeile = f'{t + 1:>6}'
            for s in range(k):
                zeile = zeile + f'{tueten[t][s]:>5}'
            zeile = zeile + f'{sum(tueten[t]):>8}'
            print(zeile)
        elif t == anzahl:
            print('   ...')


zeige(tueten)
```

Das ist genau die Verteilung, die auch die offiziellen Lösungshinweise für das Beispiel zeigen:
Tüte 1 hat vier Spiele, die Tüten 2 und 3 haben je drei. Die Tüten unterscheiden sich also um höchstens ein Spiel.
Auch bei jeder einzelnen Sorte (Spalte) unterscheiden sich die Zahlen um höchstens eins.

## Schritt 5: Selbst prüfen

Statt auf Verdacht zu glauben, dass alles stimmt, lassen wir das Programm nachprüfen:

1. Wurde von jeder Sorte **genau** die richtige Anzahl verteilt (also nichts übrig oder verloren)?
2. Unterscheiden sich die Anzahlen jeder Sorte zwischen den Tüten um **höchstens 1**?
3. Unterscheiden sich die Gesamtzahlen der Tüten um **höchstens 1**?

`max(liste) - min(liste)` ist der Unterschied zwischen dem größten und dem kleinsten Wert.

```python
def pruefe(tueten, sorten):
    n = len(tueten)
    ok = True

    # Sorte für Sorte: die Spalte s der Tabelle prüfen
    for s in range(len(sorten)):
        spalte = []
        for t in range(n):
            spalte.append(tueten[t][s])
        if sum(spalte) != sorten[s]:            # 1. alle Spiele verteilt?
            ok = False
        if max(spalte) - min(spalte) > 1:       # 2. gleichmäßig innerhalb der Sorte?
            ok = False

    # Gesamtzahl je Tüte
    gesamt = []
    for t in range(n):
        gesamt.append(sum(tueten[t]))
    if max(gesamt) - min(gesamt) > 1:           # 3. gleichmäßig insgesamt?
        ok = False

    return ok


print(pruefe(tueten, sorten))
```

## Schritt 6: Alle Beispiele

Jetzt wenden wir das Programm auf alle sechs Beispieldateien an. Für die Dokumentation reicht bei `wundertuete5.txt`
ein Auszug, deshalb zeigen wir dort nur die ersten sechs Tüten und die letzte.

```python
for nr in range(6):
    dateiname = f'beispieldaten/wundertuete{nr}.txt'
    n, sorten = lies_wundertuete(dateiname)
    tueten = verteile_reihum(n, sorten)

    print(f'{dateiname}: {n} Tüten, Spiele je Sorte: {sorten}')
    if nr < 5:
        zeige(tueten)
    else:
        zeige(tueten, 6)
    print('Gleichmäßig verteilt:', pruefe(tueten, sorten))
    print()
```

In allen sechs Beispielen ist die Verteilung gleichmäßig. Bei `wundertuete1.txt` geht es sogar ohne Rest auf:
Alle Tüten sind genau gleich gefüllt (jede Sorte lässt sich ohne Rest durch 6 teilen).
Mehr als drei Spielesorten sind kein Problem, das Programm hat keine feste Zahl von Sorten eingebaut.

## Für Neugierige: Schneller mit `//` und `%`

Unser Programm legt jedes Spiel einzeln in eine Tüte. Bei 100 Spielen ist das egal, bei 100 Millionen Spielen dauert es
spürbar. Es geht auch schneller, ohne jedes Spiel einzeln anzufassen (das ist Lösungsidee 2 in den offiziellen Lösungshinweisen):

- Von einer Sorte mit `s` Spielen bekommt **jede** Tüte zunächst `s // n` Spiele.
- Es bleibt der Rest `s % n`. Diese wenigen Spiele verteilen wir einzeln, **ab der Starttüte** `start`.
- Die Starttüte für die nächste Sorte ist die erste Tüte, die in diesem Durchgang keinen Rest bekommen hat.

`divmod(s, n)` liefert beides auf einmal: `grundmenge, rest = divmod(s, n)` entspricht
`grundmenge = s // n` und `rest = s % n`.

Die Funktion hat dieselben Eingaben und Ausgaben wie `verteile_reihum`. Zum Vergleich lassen wir beide auf allen
Beispielen laufen und prüfen, ob sie dieselbe Tabelle liefern.

```python
def verteile_rechnen(n, sorten):
    k = len(sorten)
    tueten = []
    for t in range(n):
        zeile = []
        for s in range(k):
            zeile.append(0)
        tueten.append(zeile)

    start = 0                                   # erste Tüte, die vom Rest bekommt
    for s in range(k):
        grundmenge, rest = divmod(sorten[s], n)
        for t in range(n):
            tueten[t][s] = grundmenge           # jede Tüte bekommt gleich viele ...
        for i in range(rest):
            t = (start + i) % n
            tueten[t][s] = tueten[t][s] + 1     # ... und der Rest wird ab `start` einzeln verteilt
        start = (start + rest) % n
    return tueten


for nr in range(6):
    n, sorten = lies_wundertuete(f'beispieldaten/wundertuete{nr}.txt')
    gleich = verteile_reihum(n, sorten) == verteile_rechnen(n, sorten)
    print(f'wundertuete{nr}.txt: beide Verfahren liefern dasselbe: {gleich}')

# Das Beispiel aus den Lösungshinweisen: 7, 12 und 19 Spiele auf 4 Tüten
zeige(verteile_rechnen(4, [7, 12, 19]))
```
